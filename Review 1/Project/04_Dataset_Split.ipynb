```python
# ==========================================================
# STEP 1 : Import Required Libraries
# Project : Tuberculosis Detection using CoAtNet
# Purpose :
# Import all libraries required for dataset splitting.
# ==========================================================

import pandas as pd

from sklearn.model_selection import train_test_split

print("Libraries Imported Successfully!")
```

```python
# ==========================================================
# STEP 2 : Load Labeled Metadata
# ==========================================================

metadata = pd.read_csv("labeled_metadata.csv")

print("="*50)
print("Metadata Loaded Successfully!")
print("="*50)

print("Total Images :", len(metadata))
```

```python
# ==========================================================
# STEP 3 : Dataset Verification
# ==========================================================

print(metadata.head())

print("\nDataset Shape :", metadata.shape)
```

```python
# ==========================================================
# STEP 4 : Check Class Distribution
# ==========================================================

print(metadata["label"].value_counts())
```

```python
# ==========================================================
# STEP 5 : Split Dataset into Train and Temporary Dataset
# Purpose :
# 70% -> Train
# 30% -> Temporary (Validation + Test)
# ==========================================================

train_data, temp_data = train_test_split(
    metadata,
    test_size=0.30,
    stratify=metadata["label"],
    random_state=42
)

print("=" * 50)
print("Train Dataset Created Successfully!")
print("=" * 50)

print("Train Images :", len(train_data))
print("Temporary Images :", len(temp_data))
```

```python
# ==========================================================
# STEP 6 : Split Temporary Dataset
# Purpose :
# 15% -> Validation
# 15% -> Test
# ==========================================================

validation_data, test_data = train_test_split(
    temp_data,
    test_size=0.50,
    stratify=temp_data["label"],
    random_state=42
)

print("=" * 50)
print("Validation and Test Dataset Created Successfully!")
print("=" * 50)

print("Validation Images :", len(validation_data))
print("Test Images :", len(test_data))
```

```python
# ==========================================================
# STEP 7 : Verify Dataset Split
# ==========================================================

print("=" * 60)
print("TRAIN DATASET")
print("=" * 60)
print(train_data["label"].value_counts())

print("\n" + "=" * 60)
print("VALIDATION DATASET")
print("=" * 60)
print(validation_data["label"].value_counts())

print("\n" + "=" * 60)
print("TEST DATASET")
print("=" * 60)
print(test_data["label"].value_counts())
```

```python
# ==========================================================
# STEP 8 : Save Train, Validation and Test Metadata
# ==========================================================

train_data.to_csv("train_metadata.csv", index=False)
validation_data.to_csv("validation_metadata.csv", index=False)
test_data.to_csv("test_metadata.csv", index=False)

print("=" * 50)
print("Dataset Split Completed Successfully!")
print("=" * 50)

print("Train Metadata      : train_metadata.csv")
print("Validation Metadata : validation_metadata.csv")
print("Test Metadata       : test_metadata.csv")
```
