```python
# ==========================================================
# STEP 1 : Import Required Libraries
# ==========================================================

import pandas as pd
import os

print("Libraries Imported Successfully!")
```

```python
# ==========================================================
# STEP 2 : Load Combined Metadata
# ==========================================================

metadata = pd.read_csv("combined_metadata.csv")

print("Metadata Loaded Successfully!")

print("\nTotal Records :", len(metadata))
```

```python
# ==========================================================
# STEP 3 : Display First Five Records
# ==========================================================

metadata.head()
```

```python
# ==========================================================
# STEP 4 : Display Unique Findings
# Purpose :
# Check all disease categories available in the dataset.
# ==========================================================

print("Unique Findings:\n")

for finding in sorted(metadata["findings"].unique()):
    print(finding)
```

```python
# ==========================================================
# STEP 5 : Count Images in Each Finding
# ==========================================================

print(metadata["findings"].value_counts())
```

```python
# ==========================================================
# STEP 6 : Create Binary Labels
# ==========================================================

# Normal  -> 0
# TB      -> 1

metadata["label"] = metadata["findings"].apply(
    lambda x: 0 if str(x).strip().lower() == "normal" else 1
)

print("Binary Labels Created Successfully!")
```

```python
# ==========================================================
# STEP 7 : Check Label Distribution
# ==========================================================

print(metadata["label"].value_counts())
```

```python
# ==========================================================
# STEP 8 : Verify Labels
# ==========================================================

metadata.head(10)
```

```python
# ==========================================================
# STEP 9 : Save Final Labeled Metadata
# ==========================================================

metadata.to_csv("labeled_metadata.csv", index=False)

print("="*50)
print("Labeled Metadata Saved Successfully!")
print("File Name : labeled_metadata.csv")
print("="*50)
```
