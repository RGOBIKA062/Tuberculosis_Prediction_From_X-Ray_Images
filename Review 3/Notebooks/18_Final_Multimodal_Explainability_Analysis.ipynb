```python
# =============================================================================
# NOTEBOOK 18 — CELL 2A
# EXACT MULTIMODAL CHECKPOINT ARCHITECTURE AUDIT — FINAL CORRECT VERSION
# =============================================================================

from pathlib import Path
import json
import torch
import torch.nn as nn


# =============================================================================
# 1. PATHS
# =============================================================================

FINAL_ROOT = Path(
    r"C:\TBP\Metadata\Step_3B_8_CXR_CoAtNet_Preparation\QC\Clean_PSPNet_CXR_Cohort\FINAL_FROZEN_CLEAN_COHORT"
)

CHECKPOINT = (
    FINAL_ROOT
    / "Multimodal_15"
    / "Cell4_Fusion_Training"
    / "checkpoints"
    / "Notebook15_Cell4_Best_Multimodal_Validation_ROC_AUC.pth"
)

OUTPUT_ROOT = (
    FINAL_ROOT
    / "Notebook18_Final_Multimodal_Explainability_Analysis"
    / "Cell2A_Exact_Checkpoint_Architecture_Audit"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# =============================================================================
# 2. LOAD CHECKPOINT
# =============================================================================

print("=" * 90)
print("NOTEBOOK 18 — CELL 2A")
print("EXACT MULTIMODAL CHECKPOINT ARCHITECTURE AUDIT")
print("=" * 90)

if not CHECKPOINT.is_file():
    raise FileNotFoundError(
        f"Checkpoint not found:\n{CHECKPOINT}"
    )

checkpoint = torch.load(
    CHECKPOINT,
    map_location="cpu"
)

if not isinstance(checkpoint, dict):
    raise RuntimeError(
        "Unsupported checkpoint format."
    )

if "model_state_dict" in checkpoint:
    state_dict = checkpoint["model_state_dict"]
    state_source = "model_state_dict"

elif "state_dict" in checkpoint:
    state_dict = checkpoint["state_dict"]
    state_source = "state_dict"

else:
    state_dict = checkpoint
    state_source = "direct_state_dict"


# =============================================================================
# 3. CLEAN OPTIONAL MODULE PREFIX
# =============================================================================

clean_state_dict = {}

for key, value in state_dict.items():

    if key.startswith("module."):
        key = key[7:]

    clean_state_dict[key] = value


print("\nCheckpoint path audit : PASS")
print("State-dict source     :", state_source)
print("State-dict loading    : PASS")


# =============================================================================
# 4. EXACT KEYS PRESENT IN THE ACTUAL CHECKPOINT
# =============================================================================

expected_keys = {

    "cxr_projection.0.weight",
    "cxr_projection.0.bias",
    "cxr_projection.1.weight",
    "cxr_projection.1.bias",

    "genomic_projection.0.weight",
    "genomic_projection.0.bias",
    "genomic_projection.1.weight",
    "genomic_projection.1.bias",

    "fusion.0.weight",
    "fusion.0.bias",
    "fusion.3.weight",
    "fusion.3.bias",
}

actual_keys = set(
    clean_state_dict.keys()
)

missing = expected_keys - actual_keys
unexpected = actual_keys - expected_keys

if missing:
    raise RuntimeError(
        "Expected checkpoint keys are missing:\n"
        + "\n".join(sorted(missing))
    )

if unexpected:
    raise RuntimeError(
        "Unexpected checkpoint keys detected:\n"
        + "\n".join(sorted(unexpected))
    )


print("\nExact checkpoint key audit : PASS")


# =============================================================================
# 5. EXACT ARCHITECTURE
#
# IMPORTANT:
# The saved checkpoint contains ONLY:
#
# BatchNorm weight + bias
#
# It does NOT contain:
#
# running_mean
# running_var
#
# Therefore the architecture must use:
#
# BatchNorm1d(..., track_running_stats=False)
#
# so that the module has exactly the same saved state-dict structure.
# =============================================================================

class FrozenMultimodalFusionExact(nn.Module):

    def __init__(self):

        super().__init__()

        self.cxr_projection = nn.Sequential(

            nn.Linear(
                768,
                256
            ),

            nn.BatchNorm1d(
                256,
                track_running_stats=False
            )
        )

        self.genomic_projection = nn.Sequential(

            nn.Linear(
                128,
                128
            ),

            nn.BatchNorm1d(
                128,
                track_running_stats=False
            )
        )

        self.fusion = nn.Sequential(

            nn.Linear(
                384,
                128
            ),

            nn.ReLU(),

            nn.Dropout(
                p=0.2
            ),

            nn.Linear(
                128,
                2
            )
        )

    def forward(
        self,
        cxr,
        genomic
    ):

        cxr_features = self.cxr_projection(
            cxr
        )

        genomic_features = self.genomic_projection(
            genomic
        )

        fused_features = torch.cat(
            [
                cxr_features,
                genomic_features
            ],
            dim=1
        )

        logits = self.fusion(
            fused_features
        )

        return logits


# =============================================================================
# 6. CREATE EXACT MODEL
# =============================================================================

model = FrozenMultimodalFusionExact()


# =============================================================================
# 7. STRICT LOAD
# =============================================================================

load_result = model.load_state_dict(
    clean_state_dict,
    strict=True
)

if load_result.missing_keys:
    raise RuntimeError(
        "Missing keys after strict loading:\n"
        + "\n".join(load_result.missing_keys)
    )

if load_result.unexpected_keys:
    raise RuntimeError(
        "Unexpected keys after strict loading:\n"
        + "\n".join(load_result.unexpected_keys)
    )


# =============================================================================
# 8. FREEZE
# =============================================================================

model.eval()

for parameter in model.parameters():
    parameter.requires_grad = False


# =============================================================================
# 9. PARAMETER AUDIT
# =============================================================================

total_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
)

trainable_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
    if parameter.requires_grad
)

if trainable_parameters != 0:
    raise RuntimeError(
        "Frozen model still contains trainable parameters."
    )


# =============================================================================
# 10. STRUCTURAL FORWARD TEST
# =============================================================================

dummy_cxr = torch.zeros(
    4,
    768,
    dtype=torch.float32
)

dummy_genomic = torch.zeros(
    4,
    128,
    dtype=torch.float32
)

with torch.no_grad():

    dummy_logits = model(
        dummy_cxr,
        dummy_genomic
    )


if dummy_logits.shape != (4, 2):
    raise RuntimeError(
        f"Unexpected output shape: "
        f"{tuple(dummy_logits.shape)}"
    )


# =============================================================================
# 11. SAVE AUDIT
# =============================================================================

summary = {

    "status":
        "PASS",

    "checkpoint":
        str(CHECKPOINT),

    "state_dict_source":
        state_source,

    "architecture": {

        "cxr_projection":
            "Linear(768,256) + BatchNorm1d(256, track_running_stats=False)",

        "genomic_projection":
            "Linear(128,128) + BatchNorm1d(128, track_running_stats=False)",

        "fusion":
            "Linear(384,128) + ReLU + Dropout(0.2) + Linear(128,2)"
    },

    "checkpoint_keys":
        sorted(actual_keys),

    "total_parameters":
        int(total_parameters),

    "trainable_parameters":
        int(trainable_parameters),

    "strict_checkpoint_load":
        True,

    "forward_output_shape":
        list(dummy_logits.shape),

    "training_performed":
        False,

    "fine_tuning_performed":
        False,

    "checkpoint_selection":
        False,

    "threshold_tuning":
        False,

    "test_predictions_modified":
        False
}


SUMMARY_FILE = (
    OUTPUT_ROOT
    / "Notebook18_Cell2A_Exact_Checkpoint_Architecture_Audit.json"
)

with open(
    SUMMARY_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=4
    )


# =============================================================================
# 12. VERIFY OUTPUT
# =============================================================================

if not SUMMARY_FILE.is_file():
    raise RuntimeError(
        f"Audit summary was not created:\n{SUMMARY_FILE}"
    )


# =============================================================================
# 13. FINAL STATUS
# =============================================================================

print("\n" + "=" * 90)
print("NOTEBOOK 18 — CELL 2A STATUS: PASS")
print("=" * 90)

print(
    "CXR projection       : Linear(768,256) + BatchNorm1d(256)"
)

print(
    "CXR BatchNorm stats  : track_running_stats=False"
)

print(
    "Genomic projection   : Linear(128,128) + BatchNorm1d(128)"
)

print(
    "Genomic BatchNorm    : track_running_stats=False"
)

print(
    "Fusion               : Linear(384,128) + ReLU + Dropout(0.2) + Linear(128,2)"
)

print(
    "Strict checkpoint load : PASS"
)

print(
    "Total parameters       :",
    total_parameters
)

print(
    "Trainable parameters   :",
    trainable_parameters
)

print(
    "Forward pass           : PASS"
)

print(
    "Output shape           :",
    tuple(dummy_logits.shape)
)

print(
    "\nTraining performed     : NO"
)

print(
    "Fine-tuning            : NO"
)

print(
    "Checkpoint selection   : NO"
)

print(
    "Threshold tuning       : NO"
)

print(
    "Test predictions modified : NO"
)

print("\nAudit summary:")
print(SUMMARY_FILE)

print("\nCELL 2A COMPLETE.")
```

```python
# =============================================================================
# NOTEBOOK 18 — CELL 2B
# FROZEN MULTIMODAL MODALITY-CONTRIBUTION ANALYSIS
#
# IMPORTANT:
# This cell uses the EXACT Notebook 15 Cell 6 architecture.
#
# Verified from the uploaded Notebook 15:
#
# CXR:
#   Linear(768,256)
#   LayerNorm(256)
#   ReLU
#   Dropout(0.20)
#
# Genomic:
#   Linear(128,128)
#   LayerNorm(128)
#   ReLU
#   Dropout(0.20)
#
# Fusion:
#   Linear(384,128)
#   ReLU
#   Dropout(0.20)
#   Linear(128,2)
#
# Test inference:
#   model.eval()
#   batch size = 32
#   softmax(logits, dim=1)[:,1]
#
# NO training
# NO fine-tuning
# NO checkpoint selection
# NO threshold tuning
# NO modification of official predictions
# =============================================================================

from pathlib import Path
import json
import numpy as np
import pandas as pd
import torch
import torch.nn as nn


# =============================================================================
# 1. PATHS
# =============================================================================

FINAL_ROOT = Path(
    r"C:\TBP\Metadata\Step_3B_8_CXR_CoAtNet_Preparation\QC\Clean_PSPNet_CXR_Cohort\FINAL_FROZEN_CLEAN_COHORT"
)

CELL2B_ROOT = (
    FINAL_ROOT
    / "Notebook18_Final_Multimodal_Explainability_Analysis"
    / "Cell2B_Modality_Contribution"
)

CELL2B_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

CHECKPOINT = (
    FINAL_ROOT
    / "Multimodal_15"
    / "Cell4_Fusion_Training"
    / "checkpoints"
    / "Notebook15_Cell4_Best_Multimodal_Validation_ROC_AUC.pth"
)

CXR_EMBEDDINGS = (
    FINAL_ROOT
    / "Multimodal_15"
    / "Cell2_Frozen_Embedding_Extraction"
    / "Notebook15_Cell2_FINAL_1976_CXR_Embeddings_768D.npy"
)

GENOMIC_EMBEDDINGS = (
    FINAL_ROOT
    / "Multimodal_15"
    / "Cell2_Frozen_Embedding_Extraction"
    / "Notebook15_Cell2_FINAL_1976_Genomic_Embeddings_128D.npy"
)

CONDITION_IDS = (
    FINAL_ROOT
    / "Multimodal_15"
    / "Cell2_Frozen_Embedding_Extraction"
    / "Notebook15_Cell2_FINAL_1976_Condition_IDs.npy"
)

TARGETS = (
    FINAL_ROOT
    / "Multimodal_15"
    / "Cell2_Frozen_Embedding_Extraction"
    / "Notebook15_Cell2_FINAL_1976_Targets.npy"
)

SPLITS = (
    FINAL_ROOT
    / "Multimodal_15"
    / "Cell2_Frozen_Embedding_Extraction"
    / "Notebook15_Cell2_FINAL_1976_Splits.npy"
)

OFFICIAL_TEST_PREDICTIONS = (
    FINAL_ROOT
    / "Multimodal_15"
    / "Cell6_Frozen_Test_Evaluation"
    / "Notebook15_Cell6_FINAL_Frozen_Test_Predictions.csv"
)

CELL2A_AUDIT = (
    FINAL_ROOT
    / "Notebook18_Final_Multimodal_Explainability_Analysis"
    / "Cell2A_Exact_Checkpoint_Architecture_Audit"
    / "Notebook18_Cell2A_Exact_Checkpoint_Architecture_Audit.json"
)


# =============================================================================
# 2. INPUT AUDIT
# =============================================================================

print("=" * 90)
print("NOTEBOOK 18 — CELL 2B")
print("FROZEN MULTIMODAL MODALITY-CONTRIBUTION ANALYSIS")
print("=" * 90)

required_files = {
    "Cell 2A audit": CELL2A_AUDIT,
    "Multimodal checkpoint": CHECKPOINT,
    "CXR embeddings": CXR_EMBEDDINGS,
    "Genomic embeddings": GENOMIC_EMBEDDINGS,
    "Condition IDs": CONDITION_IDS,
    "Targets": TARGETS,
    "Splits": SPLITS,
    "Official test predictions": OFFICIAL_TEST_PREDICTIONS,
}

for name, path in required_files.items():

    if not path.is_file():
        raise FileNotFoundError(
            f"{name} not found:\n{path}"
        )

    print(f"{name:<36}: PASS")


# =============================================================================
# 3. LOAD CELL 2A AUDIT
# =============================================================================

with open(
    CELL2A_AUDIT,
    "r",
    encoding="utf-8"
) as f:

    cell2a_summary = json.load(f)

if cell2a_summary.get("status") != "PASS":
    raise RuntimeError(
        "Notebook 18 Cell 2A is not PASS."
    )

print(
    "Cell 2A frozen checkpoint audit       : PASS"
)


# =============================================================================
# 4. LOAD FROZEN DATA
# =============================================================================

cxr_embeddings = np.load(
    CXR_EMBEDDINGS
).astype(
    np.float32
)

genomic_embeddings = np.load(
    GENOMIC_EMBEDDINGS
).astype(
    np.float32
)

condition_ids = np.load(
    CONDITION_IDS,
    allow_pickle=True
).astype(str)

targets = np.load(
    TARGETS
).astype(np.int64)

splits = np.load(
    SPLITS,
    allow_pickle=True
).astype(str)


# =============================================================================
# 5. FULL DATASET AUDIT
# =============================================================================

if cxr_embeddings.shape != (1976, 768):
    raise RuntimeError(
        f"Unexpected CXR shape: {cxr_embeddings.shape}"
    )

if genomic_embeddings.shape != (1976, 128):
    raise RuntimeError(
        f"Unexpected genomic shape: {genomic_embeddings.shape}"
    )

if condition_ids.shape != (1976,):
    raise RuntimeError(
        f"Unexpected condition ID shape: {condition_ids.shape}"
    )

if targets.shape != (1976,):
    raise RuntimeError(
        f"Unexpected target shape: {targets.shape}"
    )

if splits.shape != (1976,):
    raise RuntimeError(
        f"Unexpected split shape: {splits.shape}"
    )

if len(set(condition_ids)) != 1976:
    raise RuntimeError(
        "Condition IDs are not unique."
    )

if not np.isfinite(cxr_embeddings).all():
    raise RuntimeError(
        "CXR embeddings contain non-finite values."
    )

if not np.isfinite(genomic_embeddings).all():
    raise RuntimeError(
        "Genomic embeddings contain non-finite values."
    )

print(
    "Frozen embedding dataset               : PASS"
)


# =============================================================================
# 6. AUTHORITATIVE TEST ORDER
# =============================================================================

test_indices = np.where(
    splits == "test"
)[0]

if len(test_indices) != 396:
    raise RuntimeError(
        f"Expected 396 test cases, found {len(test_indices)}."
    )

test_ids = condition_ids[
    test_indices
]

test_targets = targets[
    test_indices
]

test_cxr = cxr_embeddings[
    test_indices
]

test_genomic = genomic_embeddings[
    test_indices
]

if test_cxr.shape != (396, 768):
    raise RuntimeError(
        f"Unexpected test CXR shape: {test_cxr.shape}"
    )

if test_genomic.shape != (396, 128):
    raise RuntimeError(
        f"Unexpected test genomic shape: {test_genomic.shape}"
    )

print(
    "Authoritative test cohort              : 396/396 PASS"
)


# =============================================================================
# 7. OFFICIAL TEST PREDICTIONS
# =============================================================================

official_df = pd.read_csv(
    OFFICIAL_TEST_PREDICTIONS
)

if len(official_df) != 396:
    raise RuntimeError(
        f"Expected 396 official predictions, "
        f"found {len(official_df)}."
    )

official_df["condition_id"] = (
    official_df["condition_id"].astype(str)
)

official_ids = official_df[
    "condition_id"
].to_numpy()

if not np.array_equal(
    official_ids,
    test_ids
):

    raise RuntimeError(
        "Official test prediction order does not exactly "
        "match the authoritative test order."
    )

print(
    "Official test condition order           : EXACT 396/396 MATCH"
)


# =============================================================================
# 8. EXACT NOTEBOOK 15 CELL 6 MODEL
# =============================================================================

class MultimodalFusionModel(nn.Module):

    def __init__(
        self,
        cxr_input_dim=768,
        cxr_projection_dim=256,
        genomic_input_dim=128,
        genomic_projection_dim=128,
        fusion_hidden_dim=128,
        num_classes=2,
        dropout=0.20
    ):

        super().__init__()

        self.cxr_projection = nn.Sequential(

            nn.Linear(
                cxr_input_dim,
                cxr_projection_dim
            ),

            nn.LayerNorm(
                cxr_projection_dim
            ),

            nn.ReLU(),

            nn.Dropout(
                dropout
            )
        )

        self.genomic_projection = nn.Sequential(

            nn.Linear(
                genomic_input_dim,
                genomic_projection_dim
            ),

            nn.LayerNorm(
                genomic_projection_dim
            ),

            nn.ReLU(),

            nn.Dropout(
                dropout
            )
        )

        self.fusion = nn.Sequential(

            nn.Linear(
                cxr_projection_dim
                +
                genomic_projection_dim,
                fusion_hidden_dim
            ),

            nn.ReLU(),

            nn.Dropout(
                dropout
            ),

            nn.Linear(
                fusion_hidden_dim,
                num_classes
            )
        )

    def forward(
        self,
        cxr_embedding,
        genomic_embedding
    ):

        cxr_representation = (
            self.cxr_projection(
                cxr_embedding
            )
        )

        genomic_representation = (
            self.genomic_projection(
                genomic_embedding
            )
        )

        fused_representation = torch.cat(
            [
                cxr_representation,
                genomic_representation
            ],
            dim=1
        )

        logits = self.fusion(
            fused_representation
        )

        return logits


# =============================================================================
# 9. LOAD FROZEN CHECKPOINT
# =============================================================================

device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

model = MultimodalFusionModel(
    cxr_input_dim=768,
    cxr_projection_dim=256,
    genomic_input_dim=128,
    genomic_projection_dim=128,
    fusion_hidden_dim=128,
    num_classes=2,
    dropout=0.20
).to(device)


loaded_checkpoint = torch.load(
    CHECKPOINT,
    map_location=device,
    weights_only=False
)

if "model_state_dict" not in loaded_checkpoint:
    raise RuntimeError(
        "Checkpoint does not contain model_state_dict."
    )

missing_keys, unexpected_keys = (
    model.load_state_dict(
        loaded_checkpoint[
            "model_state_dict"
        ],
        strict=False
    )
)

if missing_keys:
    raise RuntimeError(
        "Missing checkpoint keys:\n"
        + "\n".join(missing_keys)
    )

if unexpected_keys:
    raise RuntimeError(
        "Unexpected checkpoint keys:\n"
        + "\n".join(unexpected_keys)
    )

model.eval()

for parameter in model.parameters():
    parameter.requires_grad = False

print(
    "Exact Notebook 15 model loaded       : PASS"
)

print(
    "Checkpoint epoch                      :",
    loaded_checkpoint["epoch"]
)


# =============================================================================
# 10. FROZEN BATCHED INFERENCE
# =============================================================================
#
# EXACTLY MATCHES NOTEBOOK 15 CELL 6:
#
#   model.eval()
#   BATCH_SIZE = 32
#   torch.no_grad()
#   softmax(logits, dim=1)[:, 1]
# =============================================================================

BATCH_SIZE = 32

def frozen_predict(
    cxr_array,
    genomic_array
):

    if len(cxr_array) != len(genomic_array):
        raise RuntimeError(
            "CXR and genomic sample counts differ."
        )

    probabilities = []

    with torch.no_grad():

        for start in range(
            0,
            len(cxr_array),
            BATCH_SIZE
        ):

            end = min(
                start + BATCH_SIZE,
                len(cxr_array)
            )

            cxr_batch = torch.from_numpy(
                cxr_array[
                    start:end
                ]
            ).to(device)

            genomic_batch = torch.from_numpy(
                genomic_array[
                    start:end
                ]
            ).to(device)

            logits = model(
                cxr_batch,
                genomic_batch
            )

            if not torch.isfinite(
                logits
            ).all():

                raise RuntimeError(
                    "Non-finite logits detected."
                )

            batch_probability = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            probabilities.extend(
                batch_probability
                .detach()
                .cpu()
                .numpy()
                .tolist()
            )

    return np.asarray(
        probabilities,
        dtype=np.float64
    )


# =============================================================================
# 11. REPRODUCE OFFICIAL MULTIMODAL TEST PREDICTIONS
# =============================================================================

print("\nReproducing official frozen multimodal predictions...")

full_probability = frozen_predict(
    test_cxr,
    test_genomic
)

if len(full_probability) != 396:
    raise RuntimeError(
        "Reproduced prediction count is not 396."
    )


# Resolve official probability column.
if "probability_DR" in official_df.columns:

    official_probability = (
        official_df[
            "probability_DR"
        ]
        .astype(float)
        .to_numpy()
    )

elif "predicted_probability_DR" in official_df.columns:

    official_probability = (
        official_df[
            "predicted_probability_DR"
        ]
        .astype(float)
        .to_numpy()
    )

else:

    raise RuntimeError(
        "Official multimodal prediction probability column "
        "could not be identified."
    )


maximum_probability_difference = float(
    np.max(
        np.abs(
            full_probability
            -
            official_probability
        )
    )
)

mean_probability_difference = float(
    np.mean(
        np.abs(
            full_probability
            -
            official_probability
        )
    )
)


# =============================================================================
# 12. CRITICAL CONSISTENCY GATE
# =============================================================================

print("\n")
print("=" * 90)
print("OFFICIAL MULTIMODAL PREDICTION REPRODUCTION")
print("=" * 90)

print(
    f"Maximum probability difference : "
    f"{maximum_probability_difference:.12f}"
)

print(
    f"Mean probability difference    : "
    f"{mean_probability_difference:.12f}"
)


if maximum_probability_difference > 1e-7:

    raise RuntimeError(
        "FROZEN MODEL REPRODUCTION FAILED.\n"
        "The exact Notebook 15 Cell 6 architecture/checkpoint/input order "
        "did not reproduce the official predictions.\n"
        f"Maximum probability difference: "
        f"{maximum_probability_difference:.12f}"
    )

print(
    "Prediction reproduction        : PASS"
)


# =============================================================================
# 13. MODALITY ABLATIONS
# =============================================================================
#
# These are descriptive perturbations of the EXACT FROZEN model.
#
# Full:
#     CXR + genomic
#
# CXR-only:
#     CXR + zero genomic
#
# Genomic-only:
#     zero CXR + genomic
#
# Zero-modality:
#     zero CXR + zero genomic
#
# No parameters are changed.
# =============================================================================

zero_cxr = np.zeros_like(
    test_cxr,
    dtype=np.float32
)

zero_genomic = np.zeros_like(
    test_genomic,
    dtype=np.float32
)


print("\nRunning frozen modality ablations...")

cxr_only_probability = frozen_predict(
    test_cxr,
    zero_genomic
)

genomic_only_probability = frozen_predict(
    zero_cxr,
    test_genomic
)

zero_modality_probability = frozen_predict(
    zero_cxr,
    zero_genomic
)


# =============================================================================
# 14. ABLATION AUDIT
# =============================================================================

for name, values in {

    "full_multimodal":
        full_probability,

    "cxr_only":
        cxr_only_probability,

    "genomic_only":
        genomic_only_probability,

    "zero_modality":
        zero_modality_probability,

}.items():

    if len(values) != 396:
        raise RuntimeError(
            f"{name}: expected 396 predictions, "
            f"found {len(values)}."
        )

    if not np.isfinite(values).all():
        raise RuntimeError(
            f"{name}: non-finite probabilities detected."
        )

    if (
        values.min() < 0
        or
        values.max() > 1
    ):
        raise RuntimeError(
            f"{name}: probability outside [0,1]."
        )


# =============================================================================
# 15. BUILD CASE-LEVEL CONTRIBUTION TABLE
# =============================================================================

analysis_df = pd.DataFrame({

    "condition_id":
        test_ids,

    "target_binary":
        test_targets,

    "true_class":
        np.where(
            test_targets == 1,
            "DR",
            "DS"
        ),

    "official_probability_DR":
        official_probability,

    "recomputed_full_multimodal_probability_DR":
        full_probability,

    "cxr_only_ablation_probability_DR":
        cxr_only_probability,

    "genomic_only_ablation_probability_DR":
        genomic_only_probability,

    "zero_modality_probability_DR":
        zero_modality_probability,

})

analysis_df[
    "full_minus_cxr_only"
] = (
    analysis_df[
        "recomputed_full_multimodal_probability_DR"
    ]
    -
    analysis_df[
        "cxr_only_ablation_probability_DR"
    ]
)

analysis_df[
    "full_minus_genomic_only"
] = (
    analysis_df[
        "recomputed_full_multimodal_probability_DR"
    ]
    -
    analysis_df[
        "genomic_only_ablation_probability_DR"
    ]
)

analysis_df[
    "full_minus_zero_modality"
] = (
    analysis_df[
        "recomputed_full_multimodal_probability_DR"
    ]
    -
    analysis_df[
        "zero_modality_probability_DR"
    ]
)

# Contribution of CXR:
# Compare full model against genomic-only ablation.
analysis_df[
    "absolute_CXR_contribution"
] = np.abs(
    analysis_df[
        "full_minus_genomic_only"
    ]
)

# Contribution of genomic:
# Compare full model against CXR-only ablation.
analysis_df[
    "absolute_genomic_contribution"
] = np.abs(
    analysis_df[
        "full_minus_cxr_only"
    ]
)


# =============================================================================
# 16. FINAL TABLE AUDIT
# =============================================================================

if len(analysis_df) != 396:
    raise RuntimeError(
        "Contribution table does not contain 396 cases."
    )

if analysis_df[
    "condition_id"
].nunique() != 396:
    raise RuntimeError(
        "Contribution table contains duplicate conditions."
    )

if not np.array_equal(
    analysis_df[
        "condition_id"
    ].to_numpy(),
    test_ids
):

    raise RuntimeError(
        "Contribution table condition order changed."
    )


# =============================================================================
# 17. CONTRIBUTION SUMMARY
# =============================================================================

mean_cxr_contribution = float(
    analysis_df[
        "absolute_CXR_contribution"
    ].mean()
)

median_cxr_contribution = float(
    analysis_df[
        "absolute_CXR_contribution"
    ].median()
)

mean_genomic_contribution = float(
    analysis_df[
        "absolute_genomic_contribution"
    ].mean()
)

median_genomic_contribution = float(
    analysis_df[
        "absolute_genomic_contribution"
    ].median()
)

cxr_larger_cases = int(
    (
        analysis_df[
            "absolute_CXR_contribution"
        ]
        >
        analysis_df[
            "absolute_genomic_contribution"
        ]
    ).sum()
)

genomic_larger_cases = int(
    (
        analysis_df[
            "absolute_genomic_contribution"
        ]
        >
        analysis_df[
            "absolute_CXR_contribution"
        ]
    ).sum()
)

equal_cases = int(
    np.isclose(
        analysis_df[
            "absolute_CXR_contribution"
        ],
        analysis_df[
            "absolute_genomic_contribution"
        ],
        atol=1e-12
    ).sum()
)


# =============================================================================
# 18. TARGET-SPECIFIC SUMMARY
# =============================================================================

target_summary = (
    analysis_df
    .groupby(
        "true_class",
        sort=True
    )
    .agg(

        n=(
            "condition_id",
            "count"
        ),

        mean_CXR_contribution=(
            "absolute_CXR_contribution",
            "mean"
        ),

        median_CXR_contribution=(
            "absolute_CXR_contribution",
            "median"
        ),

        mean_genomic_contribution=(
            "absolute_genomic_contribution",
            "mean"
        ),

        median_genomic_contribution=(
            "absolute_genomic_contribution",
            "median"
        ),

        mean_full_probability=(
            "recomputed_full_multimodal_probability_DR",
            "mean"
        ),

        mean_CXR_only_probability=(
            "cxr_only_ablation_probability_DR",
            "mean"
        ),

        mean_genomic_only_probability=(
            "genomic_only_ablation_probability_DR",
            "mean"
        ),

    )
    .reset_index()
)


# =============================================================================
# 19. SAVE RESULTS
# =============================================================================

RESULTS_CSV = (
    CELL2B_ROOT
    / "Notebook18_Cell2B_FINAL_Frozen_Multimodal_Modality_Contribution.csv"
)

TARGET_SUMMARY_CSV = (
    CELL2B_ROOT
    / "Notebook18_Cell2B_Target_Specific_Modality_Contribution.csv"
)

SUMMARY_JSON = (
    CELL2B_ROOT
    / "Notebook18_Cell2B_Modality_Contribution_Summary.json"
)

analysis_df.to_csv(
    RESULTS_CSV,
    index=False
)

target_summary.to_csv(
    TARGET_SUMMARY_CSV,
    index=False
)


# =============================================================================
# 20. SAVE SUMMARY
# =============================================================================

summary = {

    "status":
        "PASS",

    "test_cases":
        396,

    "checkpoint_epoch":
        int(
            loaded_checkpoint[
                "epoch"
            ]
        ),

    "batch_size":
        BATCH_SIZE,

    "official_prediction_reproduction":
        "PASS",

    "maximum_probability_difference":
        maximum_probability_difference,

    "mean_probability_difference":
        mean_probability_difference,

    "mean_absolute_CXR_contribution":
        mean_cxr_contribution,

    "median_absolute_CXR_contribution":
        median_cxr_contribution,

    "mean_absolute_genomic_contribution":
        mean_genomic_contribution,

    "median_absolute_genomic_contribution":
        median_genomic_contribution,

    "CXR_contribution_larger_cases":
        cxr_larger_cases,

    "genomic_contribution_larger_cases":
        genomic_larger_cases,

    "equal_contribution_cases":
        equal_cases,

    "training_performed":
        False,

    "fine_tuning_performed":
        False,

    "checkpoint_selection_performed":
        False,

    "threshold_tuning_performed":
        False,

    "official_test_predictions_modified":
        False,

}


with open(
    SUMMARY_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=4
    )


# =============================================================================
# 21. OUTPUT VERIFICATION
# =============================================================================

for output_file in [
    RESULTS_CSV,
    TARGET_SUMMARY_CSV,
    SUMMARY_JSON
]:

    if not output_file.is_file():
        raise RuntimeError(
            f"Output file was not created:\n{output_file}"
        )


# =============================================================================
# 22. FINAL STATUS
# =============================================================================

print("\n")
print("=" * 90)
print("NOTEBOOK 18 — CELL 2B STATUS: PASS")
print("=" * 90)

print(
    "Frozen test cases                   : 396"
)

print(
    "Exact Notebook 15 architecture      : PASS"
)

print(
    "Official prediction reproduction    : PASS"
)

print(
    f"Maximum probability difference      : "
    f"{maximum_probability_difference:.12f}"
)

print(
    f"Mean |CXR contribution|             : "
    f"{mean_cxr_contribution:.6f}"
)

print(
    f"Median |CXR contribution|           : "
    f"{median_cxr_contribution:.6f}"
)

print(
    f"Mean |Genomic contribution|         : "
    f"{mean_genomic_contribution:.6f}"
)

print(
    f"Median |Genomic contribution|       : "
    f"{median_genomic_contribution:.6f}"
)

print(
    "CXR contribution larger cases       :",
    cxr_larger_cases
)

print(
    "Genomic contribution larger cases   :",
    genomic_larger_cases
)

print(
    "Equal contribution cases            :",
    equal_cases
)

print(
    "\nTraining performed                  : NO"
)

print(
    "Fine-tuning performed               : NO"
)

print(
    "Checkpoint selection                : NO"
)

print(
    "Threshold tuning                    : NO"
)

print(
    "Official test predictions modified  : NO"
)

print("\nResults:")
print(RESULTS_CSV)

print("\nTarget-specific summary:")
print(TARGET_SUMMARY_CSV)

print("\nSummary:")
print(SUMMARY_JSON)

print("\n" + "=" * 90)
print("CELL 2B COMPLETE")
print("=" * 90)
```

```python
# =============================================================================
# NOTEBOOK 18 — CELL 4
# PUBLICATION-QUALITY MULTIMODAL MODALITY-CONTRIBUTION FIGURES
#
# FROZEN ANALYSIS ONLY
# NO TRAINING
# NO FINE-TUNING
# NO MODEL INFERENCE
# NO CHECKPOINT SELECTION
# NO THRESHOLD TUNING
# NO TEST PREDICTION MODIFICATION
# =============================================================================

from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# =============================================================================
# 1. EXACT PATHS
# =============================================================================

FINAL_ROOT = Path(
    r"C:\TBP\Metadata\Step_3B_8_CXR_CoAtNet_Preparation\QC"
    r"\Clean_PSPNet_CXR_Cohort\FINAL_FROZEN_CLEAN_COHORT"
)

CELL2B_ROOT = (
    FINAL_ROOT
    / "Notebook18_Final_Multimodal_Explainability_Analysis"
    / "Cell2B_Modality_Contribution"
)

CELL4_ROOT = (
    FINAL_ROOT
    / "Notebook18_Final_Multimodal_Explainability_Analysis"
    / "Cell4_Modality_Contribution_Figures"
)

CELL4_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

CELL2B_RESULTS = (
    CELL2B_ROOT
    / "Notebook18_Cell2B_FINAL_Frozen_Multimodal_Modality_Contribution.csv"
)


# =============================================================================
# 2. HEADER
# =============================================================================

print("=" * 90)
print("NOTEBOOK 18 — CELL 4")
print("PUBLICATION-QUALITY MULTIMODAL MODALITY-CONTRIBUTION FIGURES")
print("=" * 90)

print(
    f"Figure directory:\n{CELL4_ROOT}"
)

print(
    f"Directory exists: {CELL4_ROOT.is_dir()}"
)


# =============================================================================
# 3. INPUT AUDIT
# =============================================================================

if not CELL2B_RESULTS.is_file():
    raise FileNotFoundError(
        f"Cell 2B results not found:\n{CELL2B_RESULTS}"
    )

df = pd.read_csv(
    CELL2B_RESULTS
)

required_columns = [
    "condition_id",
    "true_class",
    "official_probability_DR",
    "recomputed_full_multimodal_probability_DR",
    "cxr_only_ablation_probability_DR",
    "genomic_only_ablation_probability_DR",
    "zero_modality_probability_DR",
    "absolute_CXR_contribution",
    "absolute_genomic_contribution"
]

missing = [
    c for c in required_columns
    if c not in df.columns
]

if missing:
    raise RuntimeError(
        "Required Cell 2B columns are missing:\n"
        + "\n".join(missing)
    )

if len(df) != 396:
    raise RuntimeError(
        f"Expected 396 frozen test cases, found {len(df)}."
    )

if df["condition_id"].nunique() != 396:
    raise RuntimeError(
        "Condition IDs are not unique."
    )

print(
    "Cell 2B results              : PASS"
)

print(
    "Test cases                   : 396"
)


# =============================================================================
# 4. NUMERIC AUDIT
# =============================================================================

numeric_columns = [
    "official_probability_DR",
    "recomputed_full_multimodal_probability_DR",
    "cxr_only_ablation_probability_DR",
    "genomic_only_ablation_probability_DR",
    "zero_modality_probability_DR",
    "absolute_CXR_contribution",
    "absolute_genomic_contribution"
]

if not np.isfinite(
    df[numeric_columns]
    .to_numpy(dtype=float)
).all():

    raise RuntimeError(
        "Non-finite values detected."
    )

print(
    "Finite-value audit           : PASS"
)


# =============================================================================
# 5. GROUP DATA
# =============================================================================

ds = df[
    df["true_class"] == "DS"
]

dr = df[
    df["true_class"] == "DR"
]

if len(ds) != 141:
    raise RuntimeError(
        f"Expected DS=141, found {len(ds)}."
    )

if len(dr) != 255:
    raise RuntimeError(
        f"Expected DR=255, found {len(dr)}."
    )


# =============================================================================
# 6. FIGURE SETTINGS
# =============================================================================

DPI = 600

plt.rcParams.update({
    "font.size": 10,
    "axes.titlesize": 12,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10
})


# =============================================================================
# FIGURE 1
# CXR VS GENOMIC CONTRIBUTION DISTRIBUTION
# =============================================================================

fig1 = CELL4_ROOT / "Fig1_CXR_vs_Genomic_Contribution.png"

fig = plt.figure(
    figsize=(8, 6)
)

ax = fig.add_subplot(111)

ax.boxplot(
    [
        df["absolute_CXR_contribution"].to_numpy(),
        df["absolute_genomic_contribution"].to_numpy()
    ],
    tick_labels=[
        "CXR",
        "Genomic"
    ],
    showfliers=False
)

ax.set_ylabel(
    "Absolute modality contribution"
)

ax.set_title(
    "CXR and Genomic Modality Contributions"
)

ax.grid(
    axis="y",
    alpha=0.25
)

fig.tight_layout()

fig.savefig(
    str(fig1),
    format="png",
    dpi=DPI,
    bbox_inches="tight"
)

plt.close(fig)

if not fig1.is_file():
    raise RuntimeError(
        "Figure 1 was not created."
    )

print(
    "Figure 1: PASS"
)


# =============================================================================
# FIGURE 2
# PAIRED CONTRIBUTION COMPARISON
# =============================================================================

fig2 = CELL4_ROOT / "Fig2_Paired_Modality_Contribution.png"

fig = plt.figure(
    figsize=(8, 6)
)

ax = fig.add_subplot(111)

sample_indices = np.arange(
    len(df)
)

for i in sample_indices:

    ax.plot(
        [
            0,
            1
        ],
        [
            df.iloc[i]["absolute_CXR_contribution"],
            df.iloc[i]["absolute_genomic_contribution"]
        ],
        linewidth=0.35,
        alpha=0.15
    )

ax.plot(
    [0, 1],
    [
        df["absolute_CXR_contribution"].mean(),
        df["absolute_genomic_contribution"].mean()
    ],
    marker="o",
    linewidth=2
)

ax.set_xticks(
    [0, 1]
)

ax.set_xticklabels(
    [
        "CXR",
        "Genomic"
    ]
)

ax.set_ylabel(
    "Absolute modality contribution"
)

ax.set_title(
    "Paired Modality Contribution Across the Frozen Test Set"
)

ax.grid(
    axis="y",
    alpha=0.25
)

fig.tight_layout()

fig.savefig(
    str(fig2),
    format="png",
    dpi=DPI,
    bbox_inches="tight"
)

plt.close(fig)

if not fig2.is_file():
    raise RuntimeError(
        "Figure 2 was not created."
    )

print(
    "Figure 2: PASS"
)


# =============================================================================
# FIGURE 3
# TARGET-SPECIFIC CONTRIBUTION
# =============================================================================

fig3 = CELL4_ROOT / "Fig3_DS_DR_Modality_Contribution.png"

means = pd.DataFrame({
    "DS": [
        ds["absolute_CXR_contribution"].mean(),
        ds["absolute_genomic_contribution"].mean()
    ],
    "DR": [
        dr["absolute_CXR_contribution"].mean(),
        dr["absolute_genomic_contribution"].mean()
    ]
})

fig = plt.figure(
    figsize=(8, 6)
)

ax = fig.add_subplot(111)

x = np.arange(2)
width = 0.35

ax.bar(
    x - width / 2,
    means["DS"].to_numpy(),
    width,
    label="DS"
)

ax.bar(
    x + width / 2,
    means["DR"].to_numpy(),
    width,
    label="DR"
)

ax.set_xticks(
    x
)

ax.set_xticklabels(
    [
        "CXR",
        "Genomic"
    ]
)

ax.set_ylabel(
    "Mean absolute contribution"
)

ax.set_title(
    "Modality Contributions by Resistance Class"
)

ax.legend()

ax.grid(
    axis="y",
    alpha=0.25
)

fig.tight_layout()

fig.savefig(
    str(fig3),
    format="png",
    dpi=DPI,
    bbox_inches="tight"
)

plt.close(fig)

if not fig3.is_file():
    raise RuntimeError(
        "Figure 3 was not created."
    )

print(
    "Figure 3: PASS"
)


# =============================================================================
# FIGURE 4
# FULL VS ABLATED PROBABILITY
# =============================================================================

fig4 = CELL4_ROOT / "Fig4_Multimodal_Ablation_Probabilities.png"

probability_data = [
    df[
        "recomputed_full_multimodal_probability_DR"
    ].to_numpy(),

    df[
        "cxr_only_ablation_probability_DR"
    ].to_numpy(),

    df[
        "genomic_only_ablation_probability_DR"
    ].to_numpy(),

    df[
        "zero_modality_probability_DR"
    ].to_numpy()
]

fig = plt.figure(
    figsize=(10, 6)
)

ax = fig.add_subplot(111)

ax.boxplot(
    probability_data,
    tick_labels=[
        "Full\nMultimodal",
        "CXR-only\nablation",
        "Genomic-only\nablation",
        "Zero-modality\nablation"
    ],
    showfliers=False
)

ax.set_ylabel(
    "Predicted probability of DR"
)

ax.set_title(
    "Multimodal Prediction Under Modality Ablation"
)

ax.grid(
    axis="y",
    alpha=0.25
)

fig.tight_layout()

fig.savefig(
    str(fig4),
    format="png",
    dpi=DPI,
    bbox_inches="tight"
)

plt.close(fig)

if not fig4.is_file():
    raise RuntimeError(
        "Figure 4 was not created."
    )

print(
    "Figure 4: PASS"
)


# =============================================================================
# FIGURE 5
# CONTRIBUTION SCATTER
# =============================================================================

fig5 = CELL4_ROOT / "Fig5_CXR_vs_Genomic_Contribution_Scatter.png"

fig = plt.figure(
    figsize=(8, 7)
)

ax = fig.add_subplot(111)

ax.scatter(
    ds["absolute_CXR_contribution"],
    ds["absolute_genomic_contribution"],
    alpha=0.65,
    label="DS"
)

ax.scatter(
    dr["absolute_CXR_contribution"],
    dr["absolute_genomic_contribution"],
    alpha=0.65,
    label="DR"
)

max_value = max(
    df["absolute_CXR_contribution"].max(),
    df["absolute_genomic_contribution"].max()
)

ax.plot(
    [0, max_value],
    [0, max_value],
    linestyle="--",
    linewidth=1
)

ax.set_xlabel(
    "Absolute CXR contribution"
)

ax.set_ylabel(
    "Absolute genomic contribution"
)

ax.set_title(
    "CXR vs Genomic Modality Contribution"
)

ax.legend()

ax.grid(
    alpha=0.25
)

fig.tight_layout()

fig.savefig(
    str(fig5),
    format="png",
    dpi=DPI,
    bbox_inches="tight"
)

plt.close(fig)

if not fig5.is_file():
    raise RuntimeError(
        "Figure 5 was not created."
    )

print(
    "Figure 5: PASS"
)


# =============================================================================
# 7. FIGURE MANIFEST
# =============================================================================

manifest = pd.DataFrame({

    "figure_number": [
        1,
        2,
        3,
        4,
        5
    ],

    "figure_name": [
        "CXR vs Genomic Contribution",
        "Paired Modality Contribution",
        "DS vs DR Modality Contribution",
        "Multimodal Ablation Probabilities",
        "CXR vs Genomic Contribution Scatter"
    ],

    "filename": [
        fig1.name,
        fig2.name,
        fig3.name,
        fig4.name,
        fig5.name
    ],

    "resolution_dpi": [
        DPI
    ] * 5
})

MANIFEST = (
    CELL4_ROOT
    / "Notebook18_Cell4_Figure_Manifest.csv"
)

manifest.to_csv(
    MANIFEST,
    index=False
)


# =============================================================================
# 8. SUMMARY
# =============================================================================

SUMMARY = (
    CELL4_ROOT
    / "Notebook18_Cell4_Publication_Figures_Summary.json"
)

summary = {

    "status":
        "PASS",

    "test_cases":
        396,

    "DS_cases":
        141,

    "DR_cases":
        255,

    "figures_generated":
        5,

    "resolution_dpi":
        600,

    "figure_directory":
        str(CELL4_ROOT),

    "training_performed":
        False,

    "fine_tuning_performed":
        False,

    "model_inference_performed":
        False,

    "checkpoint_selection":
        False,

    "threshold_tuning":
        False,

    "official_test_predictions_modified":
        False
}

with SUMMARY.open(
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=4
    )


# =============================================================================
# 9. FINAL VERIFICATION
# =============================================================================

figure_files = [
    fig1,
    fig2,
    fig3,
    fig4,
    fig5
]

for file_path in figure_files:

    if not file_path.is_file():

        raise RuntimeError(
            f"Figure verification failed:\n{file_path}"
        )

    if file_path.stat().st_size == 0:

        raise RuntimeError(
            f"Figure file is empty:\n{file_path}"
        )

if not MANIFEST.is_file():
    raise RuntimeError(
        "Figure manifest was not created."
    )

if not SUMMARY.is_file():
    raise RuntimeError(
        "Figure summary was not created."
    )


# =============================================================================
# 10. FINAL STATUS
# =============================================================================

print("\n")
print("=" * 90)
print("NOTEBOOK 18 — CELL 4 STATUS: PASS")
print("=" * 90)

print(
    "Figures generated              : 5"
)

print(
    "Resolution                     : 600 DPI"
)

print(
    "Training performed             : NO"
)

print(
    "Fine-tuning performed          : NO"
)

print(
    "Model inference performed      : NO"
)

print(
    "Checkpoint selection           : NO"
)

print(
    "Threshold tuning               : NO"
)

print(
    "Official test predictions modified : NO"
)

print("\nSaved to:")
print(CELL4_ROOT)

for file_path in figure_files:
    print(f"\n{file_path}")

print(f"\nManifest:\n{MANIFEST}")

print(f"\nSummary:\n{SUMMARY}")

print("\nCELL 4 COMPLETE.")
```
