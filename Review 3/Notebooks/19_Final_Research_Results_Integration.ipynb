```python
# =============================================================================
# NOTEBOOK 19 — CELL 2
# FINAL MODEL COMPARISON + PUBLICATION-READY ANALYSIS
#
# ANALYSIS ONLY
# NO TRAINING
# NO FINE-TUNING
# NO MODEL INFERENCE
# NO CHECKPOINT SELECTION
# NO THRESHOLD TUNING
# NO TEST PREDICTIONS MODIFIED
# =============================================================================

from pathlib import Path
import json
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

FINAL_ROOT = (
    Path(r"C:\TBP\Metadata\Step_3B_8_CXR_CoAtNet_Preparation")
    / "QC"
    / "Clean_PSPNet_CXR_Cohort"
    / "FINAL_FROZEN_CLEAN_COHORT"
)

NB19_ROOT = (
    FINAL_ROOT
    / "Notebook19_Final_Research_Results_Integration"
)

CELL1_ROOT = (
    NB19_ROOT
    / "Cell1_Final_Frozen_Artifact_Audit"
)

CELL2_ROOT = (
    NB19_ROOT
    / "Cell2_Final_Model_Comparison_Analysis"
)

CELL2_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# =============================================================================
# 2. CELL 1 INPUT
# =============================================================================

PERFORMANCE_INPUT = (
    CELL1_ROOT
    / "Cell1_Final_Model_Performance_Table.csv"
)

AUDIT_INPUT = (
    CELL1_ROOT
    / "Cell1_Final_Frozen_Artifact_Audit.json"
)


if not PERFORMANCE_INPUT.is_file():
    raise FileNotFoundError(
        f"Cell 1 performance table not found:\n"
        f"{PERFORMANCE_INPUT}"
    )

if not AUDIT_INPUT.is_file():
    raise FileNotFoundError(
        f"Cell 1 audit JSON not found:\n"
        f"{AUDIT_INPUT}"
    )


# =============================================================================
# 3. LOAD CELL 1 RESULTS
# =============================================================================

performance_df = pd.read_csv(
    PERFORMANCE_INPUT
)

with open(
    AUDIT_INPUT,
    "r",
    encoding="utf-8"
) as f:
    cell1_audit = json.load(f)


# =============================================================================
# 4. EXPECTED MODELS
# =============================================================================

EXPECTED_MODELS = [
    "CXR-only",
    "Genomic-only",
    "CXR + Genomic"
]

if set(performance_df["model"]) != set(EXPECTED_MODELS):
    raise RuntimeError(
        "Unexpected model names in Cell 1 table.\n"
        f"Found: {performance_df['model'].tolist()}\n"
        f"Expected: {EXPECTED_MODELS}"
    )

performance_df = (
    performance_df
    .set_index("model")
    .loc[EXPECTED_MODELS]
    .reset_index()
)


# =============================================================================
# 5. REQUIRED METRICS
# =============================================================================

METRICS = [
    "accuracy",
    "balanced_accuracy",
    "precision",
    "sensitivity",
    "specificity",
    "f1",
    "roc_auc",
    "pr_auc"
]

missing_metrics = [
    metric
    for metric in METRICS
    if metric not in performance_df.columns
]

if missing_metrics:
    raise RuntimeError(
        "Missing required metrics:\n"
        + "\n".join(missing_metrics)
    )

if performance_df[METRICS].isna().any().any():
    raise RuntimeError(
        "NaN values detected in final performance table."
    )


# =============================================================================
# 6. FINAL FROZEN VALUES
# =============================================================================

EXPECTED_VALUES = {

    "CXR-only": {
        "accuracy": 0.691919,
        "balanced_accuracy": 0.629203,
        "precision": 0.722408,
        "sensitivity": 0.847059,
        "specificity": 0.411348,
        "f1": 0.779783,
        "roc_auc": 0.715589,
        "pr_auc": 0.797732
    },

    "Genomic-only": {
        "accuracy": 0.936869,
        "balanced_accuracy": 0.943054,
        "precision": 0.979167,
        "sensitivity": 0.921569,
        "specificity": 0.964539,
        "f1": 0.949495,
        "roc_auc": 0.960298,
        "pr_auc": 0.981941
    },

    "CXR + Genomic": {
        "accuracy": 0.936869,
        "balanced_accuracy": 0.943054,
        "precision": 0.979167,
        "sensitivity": 0.921569,
        "specificity": 0.964539,
        "f1": 0.949495,
        "roc_auc": 0.971437,
        "pr_auc": 0.986393
    }
}


# =============================================================================
# 7. VERIFY ALL FINAL VALUES
# =============================================================================

for model_name in EXPECTED_MODELS:

    row = performance_df[
        performance_df["model"] == model_name
    ].iloc[0]

    for metric in METRICS:

        actual = float(
            row[metric]
        )

        expected = float(
            EXPECTED_VALUES[model_name][metric]
        )

        if not np.isclose(
            actual,
            expected,
            atol=1e-5
        ):
            raise RuntimeError(
                f"Metric mismatch:\n"
                f"Model   : {model_name}\n"
                f"Metric  : {metric}\n"
                f"Expected: {expected}\n"
                f"Found   : {actual}"
            )


# =============================================================================
# 8. FINAL PUBLICATION PERFORMANCE TABLE
# =============================================================================

publication_table = performance_df[
    [
        "model",
        "accuracy",
        "balanced_accuracy",
        "precision",
        "sensitivity",
        "specificity",
        "f1",
        "roc_auc",
        "pr_auc"
    ]
].copy()


# Percentage versions

publication_table["accuracy_percent"] = (
    publication_table["accuracy"] * 100
)

publication_table["balanced_accuracy_percent"] = (
    publication_table["balanced_accuracy"] * 100
)

publication_table["precision_percent"] = (
    publication_table["precision"] * 100
)

publication_table["sensitivity_percent"] = (
    publication_table["sensitivity"] * 100
)

publication_table["specificity_percent"] = (
    publication_table["specificity"] * 100
)

publication_table["f1_percent"] = (
    publication_table["f1"] * 100
)

publication_table["roc_auc_percent"] = (
    publication_table["roc_auc"] * 100
)

publication_table["pr_auc_percent"] = (
    publication_table["pr_auc"] * 100
)


# =============================================================================
# 9. GET INDIVIDUAL MODEL ROWS
# =============================================================================

cxr = performance_df[
    performance_df["model"] == "CXR-only"
].iloc[0]

genomic = performance_df[
    performance_df["model"] == "Genomic-only"
].iloc[0]

multimodal = performance_df[
    performance_df["model"] == "CXR + Genomic"
].iloc[0]


# =============================================================================
# 10. CXR → MULTIMODAL COMPARISON
# =============================================================================

cxr_comparison_rows = []

for metric in METRICS:

    old_value = float(
        cxr[metric]
    )

    new_value = float(
        multimodal[metric]
    )

    difference = (
        new_value - old_value
    )

    cxr_comparison_rows.append({

        "comparison":
            "CXR-only vs CXR + Genomic",

        "metric":
            metric,

        "CXR_only":
            old_value,

        "CXR_plus_Genomic":
            new_value,

        "absolute_difference":
            difference,

        "percentage_point_difference":
            difference * 100

    })


cxr_comparison_df = pd.DataFrame(
    cxr_comparison_rows
)


# =============================================================================
# 11. GENOMIC → MULTIMODAL COMPARISON
# =============================================================================

genomic_comparison_rows = []

for metric in METRICS:

    old_value = float(
        genomic[metric]
    )

    new_value = float(
        multimodal[metric]
    )

    difference = (
        new_value - old_value
    )

    genomic_comparison_rows.append({

        "comparison":
            "Genomic-only vs CXR + Genomic",

        "metric":
            metric,

        "Genomic_only":
            old_value,

        "CXR_plus_Genomic":
            new_value,

        "absolute_difference":
            difference,

        "percentage_point_difference":
            difference * 100

    })


genomic_comparison_df = pd.DataFrame(
    genomic_comparison_rows
)


# =============================================================================
# 12. COMBINED IMPROVEMENT TABLE
# =============================================================================

improvement_table = pd.concat(
    [
        cxr_comparison_df,
        genomic_comparison_df
    ],
    ignore_index=True
)


# =============================================================================
# 13. KEY RESULTS SUMMARY
# =============================================================================

summary_table = pd.DataFrame([

    {
        "comparison":
            "CXR-only vs CXR + Genomic",

        "accuracy_difference_pp":
            (
                multimodal["accuracy"]
                - cxr["accuracy"]
            ) * 100,

        "balanced_accuracy_difference_pp":
            (
                multimodal["balanced_accuracy"]
                - cxr["balanced_accuracy"]
            ) * 100,

        "precision_difference_pp":
            (
                multimodal["precision"]
                - cxr["precision"]
            ) * 100,

        "sensitivity_difference_pp":
            (
                multimodal["sensitivity"]
                - cxr["sensitivity"]
            ) * 100,

        "specificity_difference_pp":
            (
                multimodal["specificity"]
                - cxr["specificity"]
            ) * 100,

        "f1_difference_pp":
            (
                multimodal["f1"]
                - cxr["f1"]
            ) * 100,

        "roc_auc_difference_pp":
            (
                multimodal["roc_auc"]
                - cxr["roc_auc"]
            ) * 100,

        "pr_auc_difference_pp":
            (
                multimodal["pr_auc"]
                - cxr["pr_auc"]
            ) * 100
    },

    {
        "comparison":
            "Genomic-only vs CXR + Genomic",

        "accuracy_difference_pp":
            (
                multimodal["accuracy"]
                - genomic["accuracy"]
            ) * 100,

        "balanced_accuracy_difference_pp":
            (
                multimodal["balanced_accuracy"]
                - genomic["balanced_accuracy"]
            ) * 100,

        "precision_difference_pp":
            (
                multimodal["precision"]
                - genomic["precision"]
            ) * 100,

        "sensitivity_difference_pp":
            (
                multimodal["sensitivity"]
                - genomic["sensitivity"]
            ) * 100,

        "specificity_difference_pp":
            (
                multimodal["specificity"]
                - genomic["specificity"]
            ) * 100,

        "f1_difference_pp":
            (
                multimodal["f1"]
                - genomic["f1"]
            ) * 100,

        "roc_auc_difference_pp":
            (
                multimodal["roc_auc"]
                - genomic["roc_auc"]
            ) * 100,

        "pr_auc_difference_pp":
            (
                multimodal["pr_auc"]
                - genomic["pr_auc"]
            ) * 100
    }

])


# =============================================================================
# 14. DESCRIPTIVE METRIC ORDER
# =============================================================================
# This is descriptive only and is NOT a recommendation or model ranking.

descriptive_rows = []

for metric in METRICS:

    metric_df = performance_df[
        [
            "model",
            metric
        ]
    ].copy()

    metric_df = metric_df.sort_values(
        by=metric,
        ascending=False
    ).reset_index(
        drop=True
    )

    for position in range(
        len(metric_df)
    ):

        descriptive_rows.append({

            "metric":
                metric,

            "model":
                metric_df.loc[
                    position,
                    "model"
                ],

            "value":
                float(
                    metric_df.loc[
                        position,
                        metric
                    ]
                ),

            "descriptive_position":
                position + 1

        })


descriptive_order_df = pd.DataFrame(
    descriptive_rows
)


# =============================================================================
# 15. IMPORTANT FINAL FINDINGS TABLE
# =============================================================================

final_findings = pd.DataFrame([

    {
        "finding":
            "CXR-only accuracy",
        "value":
            float(cxr["accuracy"]),
        "percentage":
            float(cxr["accuracy"]) * 100
    },

    {
        "finding":
            "Genomic-only accuracy",
        "value":
            float(genomic["accuracy"]),
        "percentage":
            float(genomic["accuracy"]) * 100
    },

    {
        "finding":
            "CXR + Genomic accuracy",
        "value":
            float(multimodal["accuracy"]),
        "percentage":
            float(multimodal["accuracy"]) * 100
    },

    {
        "finding":
            "CXR → Multimodal accuracy improvement",
        "value":
            float(
                multimodal["accuracy"]
                - cxr["accuracy"]
            ),
        "percentage":
            float(
                multimodal["accuracy"]
                - cxr["accuracy"]
            ) * 100
    },

    {
        "finding":
            "Genomic → Multimodal ROC-AUC difference",
        "value":
            float(
                multimodal["roc_auc"]
                - genomic["roc_auc"]
            ),
        "percentage":
            float(
                multimodal["roc_auc"]
                - genomic["roc_auc"]
            ) * 100
    },

    {
        "finding":
            "Genomic → Multimodal PR-AUC difference",
        "value":
            float(
                multimodal["pr_auc"]
                - genomic["pr_auc"]
            ),
        "percentage":
            float(
                multimodal["pr_auc"]
                - genomic["pr_auc"]
            ) * 100
    }

])


# =============================================================================
# 16. OUTPUT PATHS
# =============================================================================

PERFORMANCE_OUTPUT = (
    CELL2_ROOT /
    "Cell2_Final_Publication_Model_Performance.csv"
)

IMPROVEMENT_OUTPUT = (
    CELL2_ROOT /
    "Cell2_Final_Model_Improvement_Analysis.csv"
)

SUMMARY_OUTPUT = (
    CELL2_ROOT /
    "Cell2_Final_Comparison_Summary.csv"
)

DESCRIPTIVE_OUTPUT = (
    CELL2_ROOT /
    "Cell2_Descriptive_Metric_Order.csv"
)

FINDINGS_OUTPUT = (
    CELL2_ROOT /
    "Cell2_Final_Result_Interpretation_Table.csv"
)

JSON_OUTPUT = (
    CELL2_ROOT /
    "Cell2_Final_Model_Comparison_Summary.json"
)


# =============================================================================
# 17. SAVE CSV FILES
# =============================================================================

publication_table.to_csv(
    PERFORMANCE_OUTPUT,
    index=False
)

improvement_table.to_csv(
    IMPROVEMENT_OUTPUT,
    index=False
)

summary_table.to_csv(
    SUMMARY_OUTPUT,
    index=False
)

descriptive_order_df.to_csv(
    DESCRIPTIVE_OUTPUT,
    index=False
)

final_findings.to_csv(
    FINDINGS_OUTPUT,
    index=False
)


# =============================================================================
# 18. SAVE JSON
# =============================================================================

json_summary = {

    "status":
        "PASS",

    "analysis":
        "Final frozen model comparison",

    "test_cases":
        396,

    "models":
        EXPECTED_MODELS,

    "metrics":
        METRICS,

    "performance": {

        "CXR-only":
            {
                metric:
                    float(cxr[metric])
                for metric in METRICS
            },

        "Genomic-only":
            {
                metric:
                    float(genomic[metric])
                for metric in METRICS
            },

        "CXR + Genomic":
            {
                metric:
                    float(multimodal[metric])
                for metric in METRICS
            }

    },

    "CXR_to_multimodal_percentage_point_difference":
        {
            metric:
                float(
                    (
                        multimodal[metric]
                        - cxr[metric]
                    ) * 100
                )
            for metric in METRICS
        },

    "genomic_to_multimodal_percentage_point_difference":
        {
            metric:
                float(
                    (
                        multimodal[metric]
                        - genomic[metric]
                    ) * 100
                )
            for metric in METRICS
        },

    "training_performed":
        False,

    "fine_tuning_performed":
        False,

    "model_inference_performed":
        False,

    "checkpoint_selection":
        False,

    "threshold_tuning":
        False,

    "test_predictions_modified":
        False

}


with open(
    JSON_OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        json_summary,
        f,
        indent=4
    )


# =============================================================================
# 19. VERIFY OUTPUTS
# =============================================================================

outputs = [
    PERFORMANCE_OUTPUT,
    IMPROVEMENT_OUTPUT,
    SUMMARY_OUTPUT,
    DESCRIPTIVE_OUTPUT,
    FINDINGS_OUTPUT,
    JSON_OUTPUT
]

for output in outputs:

    if not output.is_file():

        raise RuntimeError(
            f"Output was not created:\n{output}"
        )

    if output.stat().st_size == 0:

        raise RuntimeError(
            f"Output file is empty:\n{output}"
        )


# =============================================================================
# 20. DISPLAY RESULTS
# =============================================================================

print("\n")
print("=" * 90)
print("NOTEBOOK 19 — CELL 2")
print("FINAL MODEL PERFORMANCE")
print("=" * 90)

display(
    performance_df[
        [
            "model",
            "accuracy",
            "balanced_accuracy",
            "precision",
            "sensitivity",
            "specificity",
            "f1",
            "roc_auc",
            "pr_auc"
        ]
    ]
)


print("\n")
print("=" * 90)
print("CXR-ONLY → CXR + GENOMIC")
print("=" * 90)

display(
    cxr_comparison_df[
        [
            "metric",
            "CXR_only",
            "CXR_plus_Genomic",
            "percentage_point_difference"
        ]
    ]
)


print("\n")
print("=" * 90)
print("GENOMIC-ONLY → CXR + GENOMIC")
print("=" * 90)

display(
    genomic_comparison_df[
        [
            "metric",
            "Genomic_only",
            "CXR_plus_Genomic",
            "percentage_point_difference"
        ]
    ]
)


# =============================================================================
# 21. FINAL STATUS
# =============================================================================

print("\n")
print("=" * 90)
print("NOTEBOOK 19 — CELL 2 STATUS: PASS")
print("=" * 90)

print("Test cases                  : 396")
print("Models compared             : 3")
print("Metrics compared            : 8")

print("\nCXR-only → Multimodal       : COMPLETE")
print("Genomic-only → Multimodal   : COMPLETE")

print("\nTraining performed          : NO")
print("Fine-tuning performed       : NO")
print("Model inference performed   : NO")
print("Checkpoint selection        : NO")
print("Threshold tuning            : NO")
print("Test predictions modified   : NO")

print("\nSaved outputs:")

for output in outputs:
    print(output)

print("\n")
print("=" * 90)
print("CELL 2 COMPLETE")
print("=" * 90)
```

```python
# =============================================================================
# NOTEBOOK 19 — CELL 3
# FINAL PAIRED STATISTICAL + ERROR ANALYSIS
#
# IMPORTANT:
# - Uses EXISTING frozen test predictions only
# - NO training
# - NO fine-tuning
# - NO model inference
# - NO checkpoint selection
# - NO threshold tuning
# - NO modification of official predictions
# =============================================================================

from pathlib import Path
import json
import numpy as np
import pandas as pd

from scipy.stats import binomtest


# =============================================================================
# 1. PATHS
# =============================================================================

FINAL_ROOT = (
    Path(r"C:\TBP\Metadata\Step_3B_8_CXR_CoAtNet_Preparation")
    / "QC"
    / "Clean_PSPNet_CXR_Cohort"
    / "FINAL_FROZEN_CLEAN_COHORT"
)

NB19_ROOT = (
    FINAL_ROOT
    / "Notebook19_Final_Research_Results_Integration"
)

CELL3_ROOT = (
    NB19_ROOT
    / "Cell3_Paired_Statistical_Error_Analysis"
)

CELL3_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


print("=" * 90)
print("NOTEBOOK 19 — CELL 3")
print("FINAL PAIRED STATISTICAL + ERROR ANALYSIS")
print("=" * 90)


# =============================================================================
# 2. EXACT FROZEN PREDICTION FILES
# =============================================================================

CXR_PREDICTIONS = (
    FINAL_ROOT
    / "CoAtNet_Test_Evaluation_Run2_10G"
    / "Cell2_Final_Test_Predictions"
    / "Notebook10G_Cell2_Run2_FINAL_396_Test_Predictions.csv"
)

GENOMIC_PREDICTIONS = (
    FINAL_ROOT
    / "Genomic_Preparation_13"
    / "Cell8_Frozen_Test_Evaluation"
    / "Notebook13_Cell8_FINAL_Frozen_Test_Predictions.csv"
)

MULTIMODAL_PREDICTIONS = (
    FINAL_ROOT
    / "Multimodal_15"
    / "Cell6_Frozen_Test_Evaluation"
    / "Notebook15_Cell6_FINAL_Frozen_Test_Predictions.csv"
)


# =============================================================================
# 3. INPUT FILE AUDIT
# =============================================================================

input_files = {
    "CXR": CXR_PREDICTIONS,
    "Genomic": GENOMIC_PREDICTIONS,
    "CXR + Genomic": MULTIMODAL_PREDICTIONS
}

print("\n")
print("=" * 90)
print("INPUT FILE AUDIT")
print("=" * 90)

for name, path in input_files.items():

    if not path.is_file():

        raise FileNotFoundError(
            f"{name} prediction file not found:\n{path}"
        )

    print(
        f"{name:<25}: PASS"
    )


# =============================================================================
# 4. LOAD FROZEN PREDICTIONS
# =============================================================================

cxr = pd.read_csv(
    CXR_PREDICTIONS
)

genomic = pd.read_csv(
    GENOMIC_PREDICTIONS
)

multimodal = pd.read_csv(
    MULTIMODAL_PREDICTIONS
)


# =============================================================================
# 5. BASIC AUDIT
# =============================================================================

for name, df in {
    "CXR": cxr,
    "Genomic": genomic,
    "CXR + Genomic": multimodal
}.items():

    if len(df) != 396:

        raise RuntimeError(
            f"{name}: expected 396 rows, found {len(df)}."
        )

    if "condition_id" not in df.columns:

        raise RuntimeError(
            f"{name}: condition_id column missing."
        )

    if df["condition_id"].nunique() != 396:

        raise RuntimeError(
            f"{name}: condition_id values are not unique."
        )


# =============================================================================
# 6. SORT ALL FILES BY CONDITION ID
# =============================================================================

cxr = (
    cxr
    .sort_values("condition_id")
    .reset_index(drop=True)
)

genomic = (
    genomic
    .sort_values("condition_id")
    .reset_index(drop=True)
)

multimodal = (
    multimodal
    .sort_values("condition_id")
    .reset_index(drop=True)
)


# =============================================================================
# 7. CONDITION ALIGNMENT
# =============================================================================

cxr_ids = (
    cxr["condition_id"]
    .astype(str)
    .to_numpy()
)

genomic_ids = (
    genomic["condition_id"]
    .astype(str)
    .to_numpy()
)

multimodal_ids = (
    multimodal["condition_id"]
    .astype(str)
    .to_numpy()
)


if not np.array_equal(
    cxr_ids,
    genomic_ids
):

    raise RuntimeError(
        "CXR and Genomic condition IDs are not aligned."
    )


if not np.array_equal(
    cxr_ids,
    multimodal_ids
):

    raise RuntimeError(
        "CXR and Multimodal condition IDs are not aligned."
    )


print("\n")
print("=" * 90)
print("CONDITION ALIGNMENT")
print("=" * 90)

print(
    "CXR vs Genomic          : 396/396 PASS"
)

print(
    "CXR vs Multimodal       : 396/396 PASS"
)


# =============================================================================
# 8. FIND TARGET COLUMNS
# =============================================================================

def find_target_column(
    df,
    model_name
):

    candidates = [
        "target_binary",
        "target",
        "true_target",
        "target_class"
    ]

    for column in candidates:

        if column in df.columns:

            return column

    raise RuntimeError(
        f"{model_name}: target column not found.\n"
        f"Available columns:\n{df.columns.tolist()}"
    )


cxr_target_col = find_target_column(
    cxr,
    "CXR"
)

genomic_target_col = find_target_column(
    genomic,
    "Genomic"
)

multimodal_target_col = find_target_column(
    multimodal,
    "CXR + Genomic"
)


# =============================================================================
# 9. FIND PREDICTION COLUMNS
# =============================================================================

def find_prediction_column(
    df,
    candidates,
    model_name
):

    for column in candidates:

        if column in df.columns:

            return column

    raise RuntimeError(
        f"{model_name}: prediction column not found.\n"
        f"Available columns:\n{df.columns.tolist()}"
    )


cxr_prediction_col = find_prediction_column(
    cxr,
    [
        "predicted_class",
        "predicted_binary"
    ],
    "CXR"
)

genomic_prediction_col = find_prediction_column(
    genomic,
    [
        "predicted_class",
        "predicted_binary"
    ],
    "Genomic"
)

multimodal_prediction_col = find_prediction_column(
    multimodal,
    [
        "predicted_binary",
        "predicted_class"
    ],
    "CXR + Genomic"
)


print("\n")
print("=" * 90)
print("PREDICTION SCHEMA")
print("=" * 90)

print(
    f"CXR prediction column          : {cxr_prediction_col}"
)

print(
    f"Genomic prediction column      : {genomic_prediction_col}"
)

print(
    f"Multimodal prediction column   : {multimodal_prediction_col}"
)


# =============================================================================
# 10. ROBUST LABEL CONVERSION
#
# Target definition:
#   DS = 0
#   DR = 1
#
# Handles both:
#   numeric labels: 0 / 1
#   text labels: DS-TB / DR-TB
# =============================================================================

def normalize_binary_labels(
    series,
    column_name
):

    original = series.copy()

    # -------------------------------------------------------------------------
    # First: numeric values
    # -------------------------------------------------------------------------

    numeric = pd.to_numeric(
        original,
        errors="coerce"
    )

    non_missing = original.notna()

    numeric_valid = (
        numeric[non_missing]
        .dropna()
    )

    if len(numeric_valid) == len(
        original[non_missing]
    ):

        unique_numeric = set(
            numeric_valid.astype(float).unique()
        )

        if unique_numeric.issubset(
            {0.0, 1.0}
        ):

            return (
                numeric
                .astype(int)
                .to_numpy()
            )

    # -------------------------------------------------------------------------
    # Text conversion
    # -------------------------------------------------------------------------

    text = (
        original
        .astype(str)
        .str.strip()
        .str.upper()
    )

    # Remove harmless whitespace
    text = (
        text
        .str.replace(
            " ",
            "",
            regex=False
        )
        .str.replace(
            "_",
            "-",
            regex=False
        )
    )

    mapping = {

        # DS
        "DS": 0,
        "DS-TB": 0,
        "DSTB": 0,
        "SENSITIVE": 0,
        "SENSITIVE-TB": 0,
        "DRUG-SENSITIVE": 0,
        "DRUG-SENSITIVE-TB": 0,

        # DR
        "DR": 1,
        "DR-TB": 1,
        "DRTB": 1,
        "RESISTANT": 1,
        "DRUG-RESISTANT": 1,
        "DRUG-RESISTANT-TB": 1

    }

    unknown = sorted(
        set(
            text[
                ~text.isin(mapping.keys())
            ].tolist()
        )
    )

    if unknown:

        raise ValueError(
            f"Unknown labels found in '{column_name}':\n"
            f"{unknown}\n\n"
            f"Observed unique values:\n"
            f"{sorted(text.unique().tolist())}"
        )

    converted = text.map(
        mapping
    )

    if converted.isna().any():

        raise ValueError(
            f"NaN generated while converting "
            f"'{column_name}'."
        )

    return converted.astype(int).to_numpy()


# =============================================================================
# 11. CONVERT TARGETS
# =============================================================================

cxr_target = normalize_binary_labels(
    cxr[cxr_target_col],
    f"CXR target ({cxr_target_col})"
)

genomic_target = normalize_binary_labels(
    genomic[genomic_target_col],
    f"Genomic target ({genomic_target_col})"
)

multimodal_target = normalize_binary_labels(
    multimodal[multimodal_target_col],
    f"Multimodal target ({multimodal_target_col})"
)


# =============================================================================
# 12. CONVERT PREDICTIONS
# =============================================================================

cxr_prediction = normalize_binary_labels(
    cxr[cxr_prediction_col],
    f"CXR prediction ({cxr_prediction_col})"
)

genomic_prediction = normalize_binary_labels(
    genomic[genomic_prediction_col],
    f"Genomic prediction ({genomic_prediction_col})"
)

multimodal_prediction = normalize_binary_labels(
    multimodal[multimodal_prediction_col],
    f"Multimodal prediction ({multimodal_prediction_col})"
)


# =============================================================================
# 13. DISPLAY DETECTED LABELS
# =============================================================================

print("\n")
print("=" * 90)
print("LABEL CONVERSION AUDIT")
print("=" * 90)

print(
    "CXR target unique values:"
)

print(
    sorted(
        cxr[cxr_target_col]
        .astype(str)
        .unique()
        .tolist()
    )
)

print(
    "\nCXR prediction unique values:"
)

print(
    sorted(
        cxr[cxr_prediction_col]
        .astype(str)
        .unique()
        .tolist()
    )
)

print(
    "\nGenomic target unique values:"
)

print(
    sorted(
        genomic[genomic_target_col]
        .astype(str)
        .unique()
        .tolist()
    )
)

print(
    "\nGenomic prediction unique values:"
)

print(
    sorted(
        genomic[genomic_prediction_col]
        .astype(str)
        .unique()
        .tolist()
    )
)

print(
    "\nMultimodal target unique values:"
)

print(
    sorted(
        multimodal[multimodal_target_col]
        .astype(str)
        .unique()
        .tolist()
    )
)

print(
    "\nMultimodal prediction unique values:"
)

print(
    sorted(
        multimodal[multimodal_prediction_col]
        .astype(str)
        .unique()
        .tolist()
    )
)


# =============================================================================
# 14. TARGET CONSISTENCY
# =============================================================================

if not np.array_equal(
    cxr_target,
    genomic_target
):

    raise RuntimeError(
        "CXR and Genomic target arrays differ."
    )


if not np.array_equal(
    cxr_target,
    multimodal_target
):

    raise RuntimeError(
        "CXR and Multimodal target arrays differ."
    )


if int(
    (cxr_target == 0).sum()
) != 141:

    raise RuntimeError(
        "Expected 141 DS test cases."
    )


if int(
    (cxr_target == 1).sum()
) != 255:

    raise RuntimeError(
        "Expected 255 DR test cases."
    )


print("\n")
print(
    "Target alignment                  : PASS"
)

print(
    "DS test cases                     : 141"
)

print(
    "DR test cases                     : 255"
)


# =============================================================================
# 15. CREATE ANALYSIS DATAFRAME
# =============================================================================

analysis_df = pd.DataFrame({

    "condition_id":
        cxr["condition_id"].astype(str),

    "target_binary":
        cxr_target,

    "cxr_prediction":
        cxr_prediction,

    "genomic_prediction":
        genomic_prediction,

    "multimodal_prediction":
        multimodal_prediction

})


# =============================================================================
# 16. CORRECTNESS
# =============================================================================

analysis_df["cxr_correct"] = (
    analysis_df["cxr_prediction"]
    ==
    analysis_df["target_binary"]
)

analysis_df["genomic_correct"] = (
    analysis_df["genomic_prediction"]
    ==
    analysis_df["target_binary"]
)

analysis_df["multimodal_correct"] = (
    analysis_df["multimodal_prediction"]
    ==
    analysis_df["target_binary"]
)


# =============================================================================
# 17. ERRORS
# =============================================================================

analysis_df["cxr_error"] = (
    ~analysis_df["cxr_correct"]
)

analysis_df["genomic_error"] = (
    ~analysis_df["genomic_correct"]
)

analysis_df["multimodal_error"] = (
    ~analysis_df["multimodal_correct"]
)


# =============================================================================
# 18. BASIC ERROR COUNTS
# =============================================================================

cxr_correct_count = int(
    analysis_df["cxr_correct"].sum()
)

genomic_correct_count = int(
    analysis_df["genomic_correct"].sum()
)

multimodal_correct_count = int(
    analysis_df["multimodal_correct"].sum()
)

cxr_error_count = int(
    analysis_df["cxr_error"].sum()
)

genomic_error_count = int(
    analysis_df["genomic_error"].sum()
)

multimodal_error_count = int(
    analysis_df["multimodal_error"].sum()
)


# =============================================================================
# 19. PAIRED COMPARISON FUNCTION
# =============================================================================

def paired_comparison(
    correct_a,
    correct_b,
    name_a,
    name_b
):

    a = np.asarray(
        correct_a,
        dtype=bool
    )

    b = np.asarray(
        correct_b,
        dtype=bool
    )

    n = len(a)

    both_correct = int(
        np.sum(
            a & b
        )
    )

    both_wrong = int(
        np.sum(
            ~a & ~b
        )
    )

    a_correct_b_wrong = int(
        np.sum(
            a & ~b
        )
    )

    a_wrong_b_correct = int(
        np.sum(
            ~a & b
        )
    )

    accuracy_a = float(
        a.mean()
    )

    accuracy_b = float(
        b.mean()
    )

    difference = (
        accuracy_b
        - accuracy_a
    )

    discordant = (
        a_correct_b_wrong
        +
        a_wrong_b_correct
    )

    if discordant == 0:

        mcnemar_p = 1.0

    else:

        mcnemar_p = float(
            binomtest(
                a_wrong_b_correct,
                n=discordant,
                p=0.5,
                alternative="two-sided"
            ).pvalue
        )

    return {

        "comparison":
            f"{name_a} vs {name_b}",

        "n":
            n,

        "accuracy_model_A":
            accuracy_a,

        "accuracy_model_B":
            accuracy_b,

        "accuracy_difference":
            difference,

        "accuracy_difference_pp":
            difference * 100,

        "both_correct":
            both_correct,

        "both_wrong":
            both_wrong,

        "A_correct_B_wrong":
            a_correct_b_wrong,

        "A_wrong_B_correct":
            a_wrong_b_correct,

        "discordant_pairs":
            discordant,

        "mcnemar_exact_p":
            mcnemar_p

    }


# =============================================================================
# 20. PAIRED COMPARISONS
# =============================================================================

paired_table = pd.DataFrame([

    paired_comparison(
        analysis_df["cxr_correct"],
        analysis_df["genomic_correct"],
        "CXR-only",
        "Genomic-only"
    ),

    paired_comparison(
        analysis_df["cxr_correct"],
        analysis_df["multimodal_correct"],
        "CXR-only",
        "CXR + Genomic"
    ),

    paired_comparison(
        analysis_df["genomic_correct"],
        analysis_df["multimodal_correct"],
        "Genomic-only",
        "CXR + Genomic"
    )

])


# =============================================================================
# 21. PAIRED BOOTSTRAP
# =============================================================================

def paired_bootstrap(
    correct_a,
    correct_b,
    n_bootstrap=2000,
    seed=42
):

    a = np.asarray(
        correct_a,
        dtype=float
    )

    b = np.asarray(
        correct_b,
        dtype=float
    )

    n = len(a)

    rng = np.random.default_rng(
        seed
    )

    observed = (
        b.mean()
        -
        a.mean()
    )

    bootstrap_values = np.empty(
        n_bootstrap,
        dtype=float
    )

    for i in range(
        n_bootstrap
    ):

        indices = rng.integers(
            0,
            n,
            size=n
        )

        bootstrap_values[i] = (
            b[indices].mean()
            -
            a[indices].mean()
        )

    low = float(
        np.percentile(
            bootstrap_values,
            2.5
        )
    )

    high = float(
        np.percentile(
            bootstrap_values,
            97.5
        )
    )

    return {

        "observed_difference":
            float(observed),

        "observed_difference_pp":
            float(observed * 100),

        "bootstrap_mean":
            float(
                bootstrap_values.mean()
            ),

        "ci_95_low":
            low,

        "ci_95_high":
            high,

        "ci_95_low_pp":
            low * 100,

        "ci_95_high_pp":
            high * 100,

        "n_bootstrap":
            n_bootstrap,

        "seed":
            seed

    }


bootstrap_table = pd.DataFrame([

    {
        "comparison":
            "CXR-only vs Genomic-only",

        **paired_bootstrap(
            analysis_df["cxr_correct"],
            analysis_df["genomic_correct"]
        )

    },

    {
        "comparison":
            "CXR-only vs CXR + Genomic",

        **paired_bootstrap(
            analysis_df["cxr_correct"],
            analysis_df["multimodal_correct"]
        )

    },

    {
        "comparison":
            "Genomic-only vs CXR + Genomic",

        **paired_bootstrap(
            analysis_df["genomic_correct"],
            analysis_df["multimodal_correct"]
        )

    }

])


# =============================================================================
# 22. ERROR OVERLAP
# =============================================================================

error_overlap_table = pd.DataFrame([

    {
        "comparison":
            "CXR vs Genomic",

        "both_error":
            int(
                (
                    analysis_df["cxr_error"]
                    &
                    analysis_df["genomic_error"]
                ).sum()
            ),

        "only_CXR_error":
            int(
                (
                    analysis_df["cxr_error"]
                    &
                    ~analysis_df["genomic_error"]
                ).sum()
            ),

        "only_Genomic_error":
            int(
                (
                    ~analysis_df["cxr_error"]
                    &
                    analysis_df["genomic_error"]
                ).sum()
            ),

        "neither_error":
            int(
                (
                    ~analysis_df["cxr_error"]
                    &
                    ~analysis_df["genomic_error"]
                ).sum()
            )
    },

    {
        "comparison":
            "CXR vs CXR + Genomic",

        "both_error":
            int(
                (
                    analysis_df["cxr_error"]
                    &
                    analysis_df["multimodal_error"]
                ).sum()
            ),

        "only_CXR_error":
            int(
                (
                    analysis_df["cxr_error"]
                    &
                    ~analysis_df["multimodal_error"]
                ).sum()
            ),

        "only_Genomic_error":
            int(
                (
                    ~analysis_df["cxr_error"]
                    &
                    analysis_df["multimodal_error"]
                ).sum()
            ),

        "neither_error":
            int(
                (
                    ~analysis_df["cxr_error"]
                    &
                    ~analysis_df["multimodal_error"]
                ).sum()
            )
    },

    {
        "comparison":
            "Genomic vs CXR + Genomic",

        "both_error":
            int(
                (
                    analysis_df["genomic_error"]
                    &
                    analysis_df["multimodal_error"]
                ).sum()
            ),

        "only_CXR_error":
            int(
                (
                    analysis_df["genomic_error"]
                    &
                    ~analysis_df["multimodal_error"]
                ).sum()
            ),

        "only_Genomic_error":
            int(
                (
                    ~analysis_df["genomic_error"]
                    &
                    analysis_df["multimodal_error"]
                ).sum()
            ),

        "neither_error":
            int(
                (
                    ~analysis_df["genomic_error"]
                    &
                    ~analysis_df["multimodal_error"]
                ).sum()
            )
    }

])


# =============================================================================
# 23. CXR ERROR CORRECTION
# =============================================================================

cxr_corrected_by_genomic = int(
    (
        analysis_df["cxr_error"]
        &
        analysis_df["genomic_correct"]
    ).sum()
)

cxr_corrected_by_multimodal = int(
    (
        analysis_df["cxr_error"]
        &
        analysis_df["multimodal_correct"]
    ).sum()
)

cxr_remaining_genomic = int(
    (
        analysis_df["cxr_error"]
        &
        analysis_df["genomic_error"]
    ).sum()
)

cxr_remaining_multimodal = int(
    (
        analysis_df["cxr_error"]
        &
        analysis_df["multimodal_error"]
    ).sum()
)

error_correction_table = pd.DataFrame([

    {
        "baseline":
            "CXR-only",

        "baseline_errors":
            cxr_error_count,

        "corrected_by_genomic":
            cxr_corrected_by_genomic,

        "remaining_with_genomic":
            cxr_remaining_genomic,

        "corrected_by_multimodal":
            cxr_corrected_by_multimodal,

        "remaining_with_multimodal":
            cxr_remaining_multimodal
    }

])


# =============================================================================
# 24. THREE-MODEL CORRECTNESS PATTERNS
# =============================================================================

all_three_correct = int(
    (
        analysis_df["cxr_correct"]
        &
        analysis_df["genomic_correct"]
        &
        analysis_df["multimodal_correct"]
    ).sum()
)

cxr_only_correct = int(
    (
        analysis_df["cxr_correct"]
        &
        ~analysis_df["genomic_correct"]
        &
        ~analysis_df["multimodal_correct"]
    ).sum()
)

genomic_only_correct = int(
    (
        ~analysis_df["cxr_correct"]
        &
        analysis_df["genomic_correct"]
        &
        ~analysis_df["multimodal_correct"]
    ).sum()
)

multimodal_only_correct = int(
    (
        ~analysis_df["cxr_correct"]
        &
        ~analysis_df["genomic_correct"]
        &
        analysis_df["multimodal_correct"]
    ).sum()
)

genomic_and_multimodal_correct = int(
    (
        ~analysis_df["cxr_correct"]
        &
        analysis_df["genomic_correct"]
        &
        analysis_df["multimodal_correct"]
    ).sum()
)

all_three_wrong = int(
    (
        ~analysis_df["cxr_correct"]
        &
        ~analysis_df["genomic_correct"]
        &
        ~analysis_df["multimodal_correct"]
    ).sum()
)


three_model_table = pd.DataFrame([

    {
        "pattern":
            "All three correct",

        "count":
            all_three_correct,

        "percentage":
            all_three_correct / 396 * 100
    },

    {
        "pattern":
            "CXR-only correct",

        "count":
            cxr_only_correct,

        "percentage":
            cxr_only_correct / 396 * 100
    },

    {
        "pattern":
            "Genomic-only correct",

        "count":
            genomic_only_correct,

        "percentage":
            genomic_only_correct / 396 * 100
    },

    {
        "pattern":
            "Multimodal-only correct",

        "count":
            multimodal_only_correct,

        "percentage":
            multimodal_only_correct / 396 * 100
    },

    {
        "pattern":
            "Genomic + Multimodal correct",

        "count":
            genomic_and_multimodal_correct,

        "percentage":
            genomic_and_multimodal_correct / 396 * 100
    },

    {
        "pattern":
            "All three wrong",

        "count":
            all_three_wrong,

        "percentage":
            all_three_wrong / 396 * 100
    }

])


# =============================================================================
# 25. HARD PREDICTION AGREEMENT
# =============================================================================

agreement_table = pd.DataFrame([

    {
        "comparison":
            "CXR vs Genomic",

        "agreement_count":
            int(
                (
                    analysis_df["cxr_prediction"]
                    ==
                    analysis_df["genomic_prediction"]
                ).sum()
            )
    },

    {
        "comparison":
            "CXR vs CXR + Genomic",

        "agreement_count":
            int(
                (
                    analysis_df["cxr_prediction"]
                    ==
                    analysis_df["multimodal_prediction"]
                ).sum()
            )
    },

    {
        "comparison":
            "Genomic vs CXR + Genomic",

        "agreement_count":
            int(
                (
                    analysis_df["genomic_prediction"]
                    ==
                    analysis_df["multimodal_prediction"]
                ).sum()
            )
    }

])

agreement_table["disagreement_count"] = (
    396
    -
    agreement_table["agreement_count"]
)

agreement_table["agreement_percent"] = (
    agreement_table["agreement_count"]
    /
    396
    *
    100
)


# =============================================================================
# 26. VALIDATE KNOWN FROZEN RESULTS
# =============================================================================

expected_patterns = {

    "CXR-only vs Genomic-only": {
        "both_correct": 254,
        "both_wrong": 5,
        "A_correct_B_wrong": 20,
        "A_wrong_B_correct": 117
    },

    "CXR-only vs CXR + Genomic": {
        "both_correct": 254,
        "both_wrong": 5,
        "A_correct_B_wrong": 20,
        "A_wrong_B_correct": 117
    },

    "Genomic-only vs CXR + Genomic": {
        "both_correct": 371,
        "both_wrong": 25,
        "A_correct_B_wrong": 0,
        "A_wrong_B_correct": 0
    }

}


for _, row in paired_table.iterrows():

    comparison = row["comparison"]

    expected = expected_patterns[
        comparison
    ]

    for field in [
        "both_correct",
        "both_wrong",
        "A_correct_B_wrong",
        "A_wrong_B_correct"
    ]:

        actual = int(
            row[field]
        )

        if actual != expected[field]:

            raise RuntimeError(
                f"Frozen paired-result mismatch.\n"
                f"Comparison: {comparison}\n"
                f"Field: {field}\n"
                f"Expected: {expected[field]}\n"
                f"Found: {actual}"
            )


# =============================================================================
# 27. VALIDATE FINAL ERROR COUNTS
# =============================================================================

if cxr_error_count != 122:
    raise RuntimeError(
        f"Expected 122 CXR errors, found {cxr_error_count}."
    )

if genomic_error_count != 25:
    raise RuntimeError(
        f"Expected 25 genomic errors, found {genomic_error_count}."
    )

if multimodal_error_count != 25:
    raise RuntimeError(
        f"Expected 25 multimodal errors, found {multimodal_error_count}."
    )

if cxr_corrected_by_genomic != 117:
    raise RuntimeError(
        f"Expected 117 CXR errors corrected by genomic, "
        f"found {cxr_corrected_by_genomic}."
    )

if cxr_corrected_by_multimodal != 117:
    raise RuntimeError(
        f"Expected 117 CXR errors corrected by multimodal, "
        f"found {cxr_corrected_by_multimodal}."
    )


# =============================================================================
# 28. OUTPUT FILES
# =============================================================================

PAIRED_OUTPUT = (
    CELL3_ROOT
    / "Cell3_Final_Paired_Statistical_Comparison.csv"
)

BOOTSTRAP_OUTPUT = (
    CELL3_ROOT
    / "Cell3_Final_Paired_Bootstrap_2000.csv"
)

ERROR_OVERLAP_OUTPUT = (
    CELL3_ROOT
    / "Cell3_Final_Error_Overlap.csv"
)

ERROR_CORRECTION_OUTPUT = (
    CELL3_ROOT
    / "Cell3_Final_CXR_Error_Correction.csv"
)

THREE_MODEL_OUTPUT = (
    CELL3_ROOT
    / "Cell3_Final_Three_Model_Correctness_Patterns.csv"
)

AGREEMENT_OUTPUT = (
    CELL3_ROOT
    / "Cell3_Final_Hard_Prediction_Agreement.csv"
)

ANALYSIS_DATA_OUTPUT = (
    CELL3_ROOT
    / "Cell3_Final_Paired_Test_Analysis_Data.csv"
)

SUMMARY_JSON_OUTPUT = (
    CELL3_ROOT
    / "Cell3_Final_Paired_Statistical_Error_Analysis_Summary.json"
)


# =============================================================================
# 29. SAVE
# =============================================================================

paired_table.to_csv(
    PAIRED_OUTPUT,
    index=False
)

bootstrap_table.to_csv(
    BOOTSTRAP_OUTPUT,
    index=False
)

error_overlap_table.to_csv(
    ERROR_OVERLAP_OUTPUT,
    index=False
)

error_correction_table.to_csv(
    ERROR_CORRECTION_OUTPUT,
    index=False
)

three_model_table.to_csv(
    THREE_MODEL_OUTPUT,
    index=False
)

agreement_table.to_csv(
    AGREEMENT_OUTPUT,
    index=False
)

analysis_df.to_csv(
    ANALYSIS_DATA_OUTPUT,
    index=False
)


# =============================================================================
# 30. SAVE JSON
# =============================================================================

final_summary = {

    "status":
        "PASS",

    "test_cases":
        396,

    "test_DS":
        141,

    "test_DR":
        255,

    "error_counts": {

        "CXR":
            cxr_error_count,

        "Genomic":
            genomic_error_count,

        "CXR + Genomic":
            multimodal_error_count
    },

    "CXR_error_correction": {

        "total":
            cxr_error_count,

        "corrected_by_genomic":
            cxr_corrected_by_genomic,

        "corrected_by_multimodal":
            cxr_corrected_by_multimodal,

        "remaining_with_genomic":
            cxr_remaining_genomic,

        "remaining_with_multimodal":
            cxr_remaining_multimodal
    },

    "hard_prediction_agreement":
        agreement_table.to_dict(
            orient="records"
        ),

    "three_model_correctness":
        three_model_table.to_dict(
            orient="records"
        ),

    "paired_comparisons":
        paired_table.to_dict(
            orient="records"
        ),

    "bootstrap":
        bootstrap_table.to_dict(
            orient="records"
        ),

    "constraints": {

        "training_performed":
            False,

        "fine_tuning_performed":
            False,

        "model_inference_performed":
            False,

        "checkpoint_selection":
            False,

        "threshold_tuning":
            False,

        "test_predictions_modified":
            False
    }

}


with open(
    SUMMARY_JSON_OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        final_summary,
        f,
        indent=4
    )


# =============================================================================
# 31. VERIFY OUTPUTS
# =============================================================================

output_files = [

    PAIRED_OUTPUT,
    BOOTSTRAP_OUTPUT,
    ERROR_OVERLAP_OUTPUT,
    ERROR_CORRECTION_OUTPUT,
    THREE_MODEL_OUTPUT,
    AGREEMENT_OUTPUT,
    ANALYSIS_DATA_OUTPUT,
    SUMMARY_JSON_OUTPUT

]


for output in output_files:

    if not output.is_file():

        raise RuntimeError(
            f"Output was not created:\n{output}"
        )

    if output.stat().st_size == 0:

        raise RuntimeError(
            f"Output is empty:\n{output}"
        )


# =============================================================================
# 32. DISPLAY
# =============================================================================

print("\n")
print("=" * 90)
print("PAIRED STATISTICAL COMPARISON")
print("=" * 90)

display(
    paired_table[
        [
            "comparison",
            "accuracy_model_A",
            "accuracy_model_B",
            "accuracy_difference_pp",
            "both_correct",
            "both_wrong",
            "A_correct_B_wrong",
            "A_wrong_B_correct",
            "mcnemar_exact_p"
        ]
    ]
)


print("\n")
print("=" * 90)
print("PAIRED BOOTSTRAP — 2000 RESAMPLES")
print("=" * 90)

display(
    bootstrap_table[
        [
            "comparison",
            "observed_difference_pp",
            "ci_95_low_pp",
            "ci_95_high_pp",
            "n_bootstrap"
        ]
    ]
)


print("\n")
print("=" * 90)
print("ERROR OVERLAP")
print("=" * 90)

display(
    error_overlap_table
)


print("\n")
print("=" * 90)
print("CXR ERROR CORRECTION")
print("=" * 90)

display(
    error_correction_table
)


print("\n")
print("=" * 90)
print("THREE-MODEL CORRECTNESS")
print("=" * 90)

display(
    three_model_table
)


print("\n")
print("=" * 90)
print("HARD-PREDICTION AGREEMENT")
print("=" * 90)

display(
    agreement_table
)


# =============================================================================
# 33. FINAL STATUS
# =============================================================================

print("\n")
print("=" * 90)
print("NOTEBOOK 19 — CELL 3 STATUS: PASS")
print("=" * 90)

print(
    "Test cases                         : 396"
)

print(
    "CXR errors                         : 122"
)

print(
    "Genomic errors                     : 25"
)

print(
    "Multimodal errors                  : 25"
)

print(
    "CXR errors corrected by Genomic    : 117"
)

print(
    "CXR errors corrected by Multimodal : 117"
)

print(
    "Genomic/Multimodal hard agreement  : 396/396"
)

print(
    "Paired bootstrap resamples         : 2000"
)

print(
    "McNemar analysis                   : COMPLETE"
)

print(
    "Error-overlap analysis             : COMPLETE"
)

print(
    "Three-model analysis               : COMPLETE"
)

print("\n")
print(
    "Training performed                 : NO"
)

print(
    "Fine-tuning performed              : NO"
)

print(
    "Model inference performed          : NO"
)

print(
    "Checkpoint selection               : NO"
)

print(
    "Threshold tuning                   : NO"
)

print(
    "Test predictions modified          : NO"
)

print("\n")
print("Saved outputs:")

for output in output_files:

    print(output)

print("\n")
print("=" * 90)
print("NOTEBOOK 19 — CELL 3 COMPLETE")
print("=" * 90)
```
