```python
# ================================================================================================
# STEP 3B.6.24
# NOTEBOOK 03 INITIALIZATION + PROVENANCE CONTROL
# ================================================================================================
#
# Notebook:
#   03_CXR_Eligibility_Reconciliation.ipynb
#
# Purpose:
#   Establish a clean, reproducible environment for Notebook 03.
#
# IMPORTANT:
#   - This notebook does NOT modify raw DICOM files.
#   - This notebook does NOT modify source metadata.
#   - This notebook does NOT make machine-learning target decisions.
#   - This notebook does NOT preprocess images.
#   - This notebook does NOT perform model training.
#   - All authoritative inputs are loaded from saved artifacts on disk.
#   - No variables from Notebook 01 or Notebook 02 are assumed to exist.
#
# Project scope:
#   TB Portals March 2025 Imaging + Genomics
#
# ================================================================================================


from pathlib import Path
from datetime import datetime
import platform
import sys
import os

import numpy as np
import pandas as pd


# ================================================================================================
# 1. NOTEBOOK IDENTITY
# ================================================================================================

NOTEBOOK_NAME = "03_CXR_Eligibility_Reconciliation.ipynb"

NOTEBOOK_VERSION = "3B.6.24"

PROJECT_DATA_RELEASE = "TB Portals March 2025 Imaging + Genomics"


# ================================================================================================
# 2. ROOT DIRECTORIES
# ================================================================================================

TBP_ROOT = Path(r"C:\TBP")

METADATA_DIR = TBP_ROOT / "Metadata"

NOTEBOOK_DIR = TBP_ROOT / "Notebooks"


# ================================================================================================
# 3. REQUIRED DIRECTORY VALIDATION
# ================================================================================================

print("=" * 100)
print("STEP 3B.6.24 — NOTEBOOK 03 INITIALIZATION + PROVENANCE CONTROL")
print("=" * 100)

print()
print("Notebook:")
print(NOTEBOOK_NAME)

print()
print("Project:")
print(PROJECT_DATA_RELEASE)


print("\n1. DIRECTORY VALIDATION")
print("-" * 100)


if not TBP_ROOT.is_dir():

    raise FileNotFoundError(
        f"TBP root directory not found:\n{TBP_ROOT}"
    )


if not METADATA_DIR.is_dir():

    raise FileNotFoundError(
        f"Metadata directory not found:\n{METADATA_DIR}"
    )


if not NOTEBOOK_DIR.is_dir():

    raise FileNotFoundError(
        f"Notebook directory not found:\n{NOTEBOOK_DIR}"
    )


print(
    f"TBP root     : {TBP_ROOT} — FOUND"
)

print(
    f"Metadata     : {METADATA_DIR} — FOUND"
)

print(
    f"Notebook dir : {NOTEBOOK_DIR} — FOUND"
)


# ================================================================================================
# 4. NOTEBOOK 02 FINAL ARTIFACT
# ================================================================================================

NOTEBOOK_02_CLOSURE_DIR = (
    METADATA_DIR /
    "Step_3B_6_23V_Final_Notebook_02_Closure_Audit"
)


NOTEBOOK_02_CLOSURE_AUDIT_FILE = (
    NOTEBOOK_02_CLOSURE_DIR /
    "TB_Portals_March2025_Step_3B_6_23V_Final_Notebook_02_Closure_Audit.csv"
)


NOTEBOOK_02_CLOSURE_SUMMARY_FILE = (
    NOTEBOOK_02_CLOSURE_DIR /
    "TB_Portals_March2025_Step_3B_6_23V_Final_Notebook_02_Closure_Summary.csv"
)


NOTEBOOK_02_RETAINED_FILE = (
    NOTEBOOK_02_CLOSURE_DIR /
    "TB_Portals_March2025_Step_3B_6_23V_Final_95_Retained_Semantic_Records.csv"
)


NOTEBOOK_02_EXCLUDED_FILE = (
    NOTEBOOK_02_CLOSURE_DIR /
    "TB_Portals_March2025_Step_3B_6_23V_Final_20_Excluded_Semantic_Records.csv"
)


print("\n2. NOTEBOOK 02 FINAL ARTIFACTS")
print("-" * 100)


NOTEBOOK_02_REQUIRED_FILES = {

    "Notebook 02 closure audit":
        NOTEBOOK_02_CLOSURE_AUDIT_FILE,

    "Notebook 02 closure summary":
        NOTEBOOK_02_CLOSURE_SUMMARY_FILE,

    "Final 95 retained semantic records":
        NOTEBOOK_02_RETAINED_FILE,

    "Final 20 excluded semantic records":
        NOTEBOOK_02_EXCLUDED_FILE,

}


for artifact_name, artifact_path in NOTEBOOK_02_REQUIRED_FILES.items():

    if not artifact_path.is_file():

        raise FileNotFoundError(
            f"\nRequired Notebook 02 artifact not found.\n\n"
            f"Artifact: {artifact_name}\n"
            f"Path: {artifact_path}"
        )


    print(
        f"{artifact_name:<40}: FOUND"
    )


# ================================================================================================
# 5. VERIFY NOTEBOOK 02 CLOSURE AUDIT
# ================================================================================================

print("\n3. VERIFY NOTEBOOK 02 CLOSURE STATUS")
print("-" * 100)


closure_audit_024 = pd.read_csv(
    NOTEBOOK_02_CLOSURE_AUDIT_FILE,
    dtype=str,
    low_memory=False
).fillna("")


closure_summary_024 = pd.read_csv(
    NOTEBOOK_02_CLOSURE_SUMMARY_FILE,
    dtype=str,
    low_memory=False
).fillna("")


if closure_audit_024.empty:

    raise ValueError(
        "Notebook 02 closure audit is empty."
    )


if closure_summary_024.empty:

    raise ValueError(
        "Notebook 02 closure summary is empty."
    )


print(
    "Notebook 02 closure audit loaded : PASS"
)

print(
    "Notebook 02 closure summary loaded: PASS"
)


# ================================================================================================
# 6. VERIFY CRITICAL NOTEBOOK 02 VALUES
# ================================================================================================

def get_summary_value_024(
    dataframe,
    metric
):

    rows = dataframe[
        dataframe["metric"]
        .astype(str)
        .str.strip()
        ==
        metric
    ]

    if len(rows) != 1:

        raise ValueError(
            f"Expected exactly one summary row for metric: {metric}"
        )

    return str(
        rows.iloc[0]["value"]
    ).strip()


final_semantic_records_024 = get_summary_value_024(
    closure_summary_024,
    "Final semantic records"
)


final_retain_024 = get_summary_value_024(
    closure_summary_024,
    "RETAIN"
)


final_exclude_024 = get_summary_value_024(
    closure_summary_024,
    "EXCLUDE"
)


final_pending_024 = get_summary_value_024(
    closure_summary_024,
    "PENDING"
)


structural_overlap_024 = get_summary_value_024(
    closure_summary_024,
    "Structural overlap"
)


structural_agreement_024 = get_summary_value_024(
    closure_summary_024,
    "Structural agreement"
)


raw_disagreement_024 = get_summary_value_024(
    closure_summary_024,
    "Raw disagreement"
)


unresolved_disagreement_024 = get_summary_value_024(
    closure_summary_024,
    "Unresolved disagreement"
)


notebook_02_status_024 = get_summary_value_024(
    closure_summary_024,
    "Notebook 02 closure audit"
)


print()
print(
    f"Final semantic records : {final_semantic_records_024}"
)

print(
    f"RETAIN                : {final_retain_024}"
)

print(
    f"EXCLUDE               : {final_exclude_024}"
)

print(
    f"PENDING               : {final_pending_024}"
)

print(
    f"Structural overlap    : {structural_overlap_024}"
)

print(
    f"Structural agreement  : {structural_agreement_024}"
)

print(
    f"Raw disagreement      : {raw_disagreement_024}"
)

print(
    f"Unresolved disagreement: {unresolved_disagreement_024}"
)

print(
    f"Notebook 02 status    : {notebook_02_status_024}"
)


# ================================================================================================
# 7. HARD VALIDATION OF NOTEBOOK 02
# ================================================================================================

if final_semantic_records_024 != "115":

    raise ValueError(
        f"Notebook 02 final semantic records should be 115, "
        f"found {final_semantic_records_024}."
    )


if final_retain_024 != "95":

    raise ValueError(
        f"Notebook 02 final RETAIN should be 95, "
        f"found {final_retain_024}."
    )


if final_exclude_024 != "20":

    raise ValueError(
        f"Notebook 02 final EXCLUDE should be 20, "
        f"found {final_exclude_024}."
    )


if final_pending_024 != "0":

    raise ValueError(
        f"Notebook 02 final PENDING should be 0, "
        f"found {final_pending_024}."
    )


if structural_overlap_024 != "17":

    raise ValueError(
        f"Notebook 02 structural overlap should be 17, "
        f"found {structural_overlap_024}."
    )


if structural_agreement_024 != "12":

    raise ValueError(
        f"Notebook 02 structural agreement should be 12, "
        f"found {structural_agreement_024}."
    )


if raw_disagreement_024 != "5":

    raise ValueError(
        f"Notebook 02 raw disagreement should be 5, "
        f"found {raw_disagreement_024}."
    )


if unresolved_disagreement_024 != "0":

    raise ValueError(
        f"Notebook 02 unresolved disagreement should be 0, "
        f"found {unresolved_disagreement_024}."
    )


if notebook_02_status_024.upper() != "PASS":

    raise ValueError(
        f"Notebook 02 closure audit is not PASS. "
        f"Found: {notebook_02_status_024}"
    )


print(
    "\nNotebook 02 final-state validation: PASS"
)


# ================================================================================================
# 8. LOAD FINAL SEMANTIC POPULATIONS
# ================================================================================================

print("\n4. LOAD FINAL NOTEBOOK 02 SEMANTIC POPULATIONS")
print("-" * 100)


FINAL_RETAINED_024 = pd.read_csv(
    NOTEBOOK_02_RETAINED_FILE,
    dtype=str,
    low_memory=False
).fillna("")


FINAL_EXCLUDED_024 = pd.read_csv(
    NOTEBOOK_02_EXCLUDED_FILE,
    dtype=str,
    low_memory=False
).fillna("")


if len(FINAL_RETAINED_024) != 95:

    raise ValueError(
        f"Expected 95 retained semantic records. "
        f"Found {len(FINAL_RETAINED_024)}."
    )


if len(FINAL_EXCLUDED_024) != 20:

    raise ValueError(
        f"Expected 20 excluded semantic records. "
        f"Found {len(FINAL_EXCLUDED_024)}."
    )


if "semantic_review_id" not in FINAL_RETAINED_024.columns:

    raise ValueError(
        "Final retained semantic artifact does not contain semantic_review_id."
    )


if "semantic_review_id" not in FINAL_EXCLUDED_024.columns:

    raise ValueError(
        "Final excluded semantic artifact does not contain semantic_review_id."
    )


RETAINED_IDS_024 = set(
    FINAL_RETAINED_024[
        "semantic_review_id"
    ]
    .astype(str)
    .str.strip()
)


EXCLUDED_IDS_024 = set(
    FINAL_EXCLUDED_024[
        "semantic_review_id"
    ]
    .astype(str)
    .str.strip()
)


if RETAINED_IDS_024.intersection(
    EXCLUDED_IDS_024
):

    raise ValueError(
        "A semantic_review_id appears in both retained and excluded populations."
    )


if len(
    RETAINED_IDS_024.union(
        EXCLUDED_IDS_024
    )
) != 115:

    raise ValueError(
        "Retained + excluded semantic IDs do not cover all 115 records."
    )


print(
    "Final retained records : 95"
)

print(
    "Final excluded records : 20"
)

print(
    "Semantic partition      : PASS"
)


# ================================================================================================
# 9. NOTEBOOK 03 OUTPUT DIRECTORY
# ================================================================================================

NOTEBOOK_03_OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03"
)


NOTEBOOK_03_OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("\n5. NOTEBOOK 03 OUTPUT DIRECTORY")
print("-" * 100)

print(
    NOTEBOOK_03_OUTPUT_DIR
)

print(
    "Output directory: READY"
)


# ================================================================================================
# 10. REPRODUCIBILITY INFORMATION
# ================================================================================================

print("\n6. REPRODUCIBILITY INFORMATION")
print("-" * 100)


print(
    "Python version:"
)

print(
    sys.version
)


print(
    "\nPandas version:",
    pd.__version__
)


print(
    "NumPy version:",
    np.__version__
)


print(
    "\nOperating system:"
)

print(
    platform.platform()
)


print(
    "\nExecution timestamp:"
)

print(
    datetime.now().isoformat(
        timespec="seconds"
    )
)


# ================================================================================================
# 11. GLOBAL METHODOLOGICAL LOCKS
# ================================================================================================

NOTEBOOK_03_RAW_DICOM_MODIFICATION_ALLOWED = False

NOTEBOOK_03_SOURCE_METADATA_MODIFICATION_ALLOWED = False

NOTEBOOK_03_AUTOMATIC_IMAGE_EXCLUSION_ALLOWED = False

NOTEBOOK_03_TARGET_SELECTION_ALLOWED = False

NOTEBOOK_03_PREPROCESSING_ALLOWED = False

NOTEBOOK_03_MODEL_TRAINING_ALLOWED = False


print("\n7. METHODOLOGICAL SAFETY LOCKS")
print("-" * 100)


print(
    "Raw DICOM modification allowed       :",
    NOTEBOOK_03_RAW_DICOM_MODIFICATION_ALLOWED
)

print(
    "Source metadata modification allowed:",
    NOTEBOOK_03_SOURCE_METADATA_MODIFICATION_ALLOWED
)

print(
    "Automatic image exclusion allowed    :",
    NOTEBOOK_03_AUTOMATIC_IMAGE_EXCLUSION_ALLOWED
)

print(
    "Target selection allowed             :",
    NOTEBOOK_03_TARGET_SELECTION_ALLOWED
)

print(
    "Preprocessing allowed                :",
    NOTEBOOK_03_PREPROCESSING_ALLOWED
)

print(
    "Model training allowed               :",
    NOTEBOOK_03_MODEL_TRAINING_ALLOWED
)


# ================================================================================================
# 12. FINAL CELL-1 VALIDATION
# ================================================================================================

assert TBP_ROOT.is_dir()

assert METADATA_DIR.is_dir()

assert NOTEBOOK_DIR.is_dir()

assert NOTEBOOK_02_CLOSURE_AUDIT_FILE.is_file()

assert NOTEBOOK_02_CLOSURE_SUMMARY_FILE.is_file()

assert NOTEBOOK_02_RETAINED_FILE.is_file()

assert NOTEBOOK_02_EXCLUDED_FILE.is_file()

assert len(FINAL_RETAINED_024) == 95

assert len(FINAL_EXCLUDED_024) == 20

assert len(
    RETAINED_IDS_024.union(
        EXCLUDED_IDS_024
    )
) == 115

assert not RETAINED_IDS_024.intersection(
    EXCLUDED_IDS_024
)

assert final_pending_024 == "0"

assert structural_overlap_024 == "17"

assert structural_agreement_024 == "12"

assert raw_disagreement_024 == "5"

assert unresolved_disagreement_024 == "0"


# ================================================================================================
# 13. FINAL STATUS
# ================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.24 — COMPLETED SUCCESSFULLY")
print("=" * 100)

print()
print("Notebook 02 closure verified : PASS")
print("Final semantic records       : 115")
print("Final RETAIN                 : 95")
print("Final EXCLUDE                : 20")
print("Final PENDING                : 0")

print()
print("Structural overlap           : 17")
print("Structural agreement         : 12")
print("Raw disagreement             : 5")
print("Unresolved disagreement      : 0")

print()
print("Raw DICOM modification       : LOCKED / NOT ALLOWED")
print("Source modification          : LOCKED / NOT ALLOWED")
print("Automatic exclusion         : LOCKED / NOT ALLOWED")
print("Target selection             : LOCKED / NOT ALLOWED")
print("Preprocessing                : LOCKED / NOT ALLOWED")
print("Model training              : LOCKED / NOT ALLOWED")

print()
print("Notebook 03 initialization   : PASS")
print("Notebook 03 status           : IN PROGRESS")

print()
print("=" * 100)
print("READY FOR STEP 3B.6.25")
print("=" * 100)
```

```python
# ================================================================================================
# STEP 3B.6.25
# AUTHORITATIVE CXR MASTER RECONSTRUCTION
# ================================================================================================
#
# Purpose:
#   Reconstruct and validate the authoritative CXR master cohort from the
#   TB Portals March 2025 CXR metadata and its intersection with genomics.
#
# Expected source:
#   TB_Portals_CXRs_March_2025.csv
#
# Expected source size:
#   16,723 CXR records
#
# Expected multimodal intersection:
#   3,081 unique condition_id values
#   3,859 CXR records
#
# IMPORTANT:
#   - No image is excluded in this step.
#   - cxr_outlier is retained as an audit variable only.
#   - XC is NOT automatically excluded.
#   - OT is NOT automatically excluded.
#   - No preprocessing is performed.
#   - No target is selected.
#   - Raw DICOMs are untouched.
#
# ================================================================================================


print("=" * 100)
print("STEP 3B.6.25 — AUTHORITATIVE CXR MASTER RECONSTRUCTION")
print("=" * 100)


# ================================================================================================
# 1. EXACT ARTIFACT LOCATOR
# ================================================================================================

def find_unique_exact_file_025(
    root_directory,
    filename
):
    """
    Locate an exact filename within the project metadata directory.

    This helper is intentionally strict:
        - zero matches -> error
        - one match  -> accepted
        - multiple matches -> error

    We do NOT silently choose one file when duplicates exist.
    """

    root_directory = Path(root_directory)

    matches = [
        path
        for path in root_directory.rglob(filename)
        if path.is_file()
    ]

    if len(matches) == 0:

        raise FileNotFoundError(
            f"\nExact required file was not found.\n"
            f"Filename: {filename}\n"
            f"Search root: {root_directory}"
        )

    if len(matches) > 1:

        match_text = "\n".join(
            f"  - {path}"
            for path in matches
        )

        raise RuntimeError(
            f"\nMultiple files with the exact same filename were found.\n"
            f"This must be resolved before continuing.\n\n"
            f"Filename: {filename}\n"
            f"Matches:\n{match_text}"
        )

    return matches[0]


# ================================================================================================
# 2. LOCATE AUTHORITATIVE CXR METADATA
# ================================================================================================

CXR_SOURCE_FILENAME_025 = (
    "TB_Portals_CXRs_March_2025.csv"
)


CXR_SOURCE_FILE_025 = find_unique_exact_file_025(
    METADATA_DIR,
    CXR_SOURCE_FILENAME_025
)


print("\n1. AUTHORITATIVE CXR SOURCE")
print("-" * 100)

print(
    "Filename:",
    CXR_SOURCE_FILE_025.name
)

print(
    "Path    :",
    CXR_SOURCE_FILE_025
)


# ================================================================================================
# 3. LOAD CXR METADATA
# ================================================================================================

CXR_MASTER_SOURCE_025 = pd.read_csv(
    CXR_SOURCE_FILE_025,
    dtype=str,
    low_memory=False
)


CXR_MASTER_SOURCE_025 = (
    CXR_MASTER_SOURCE_025
    .fillna("")
)


print("\n2. SOURCE DATASET LOAD")
print("-" * 100)

print(
    "Rows    :",
    len(CXR_MASTER_SOURCE_025)
)

print(
    "Columns :",
    len(CXR_MASTER_SOURCE_025.columns)
)


# ================================================================================================
# 4. EXPECTED SOURCE SIZE VALIDATION
# ================================================================================================

EXPECTED_CXR_SOURCE_ROWS_025 = 16_723


if len(CXR_MASTER_SOURCE_025) != EXPECTED_CXR_SOURCE_ROWS_025:

    raise ValueError(
        "\nUnexpected CXR source row count.\n"
        f"Expected: {EXPECTED_CXR_SOURCE_ROWS_025}\n"
        f"Found   : {len(CXR_MASTER_SOURCE_025)}"
    )


print(
    "Expected source row count: PASS"
)


# ================================================================================================
# 5. REQUIRED COLUMN VALIDATION
# ================================================================================================

REQUIRED_CXR_COLUMNS_025 = [

    "patient_id",

    "condition_id",

    "age_of_onset",

    "sex",

    "education",

    "employment",

    "number_of_children",

    "number_of_daily_contacts",

    "case_definition",

    "diagnosis_code",

    "type_of_resistance",

    "comorbidity",

    "country",

    "bmi",

    "outcome",

    "imagingstudy_id",

    "imaging_date",

    "series_instance_content_url",

    "series_modality_cd",

    "cxr_outlier",

]


missing_cxr_columns_025 = [

    column
    for column in REQUIRED_CXR_COLUMNS_025
    if column not in CXR_MASTER_SOURCE_025.columns

]


if missing_cxr_columns_025:

    raise ValueError(
        "\nRequired CXR columns are missing:\n"
        + "\n".join(
            f"  - {column}"
            for column in missing_cxr_columns_025
        )
    )


print("\n3. REQUIRED COLUMN VALIDATION")
print("-" * 100)

print(
    f"Required columns : {len(REQUIRED_CXR_COLUMNS_025)}"
)

print(
    "Column validation : PASS"
)


# ================================================================================================
# 6. COLUMN ORDER / SOURCE SCHEMA REPORT
# ================================================================================================

print("\n4. SOURCE SCHEMA")
print("-" * 100)

for index, column in enumerate(
    CXR_MASTER_SOURCE_025.columns,
    start=1
):

    print(
        f"{index:02d}. {column}"
    )


# ================================================================================================
# 7. CONDITION_ID VALIDATION
# ================================================================================================

print("\n5. CONDITION_ID VALIDATION")
print("-" * 100)


condition_series_025 = (
    CXR_MASTER_SOURCE_025[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)


missing_condition_ids_025 = (
    condition_series_025
    .eq("")
    .sum()
)


unique_condition_ids_025 = (
    condition_series_025
    .nunique()
)


if missing_condition_ids_025 != 0:

    raise ValueError(
        f"CXR source contains {missing_condition_ids_025} "
        f"records with missing condition_id."
    )


print(
    "Missing condition_id:",
    missing_condition_ids_025
)

print(
    "Unique condition_id:",
    unique_condition_ids_025
)


if unique_condition_ids_025 != 13_777:

    raise ValueError(
        "\nUnexpected number of unique CXR condition_id values.\n"
        f"Expected: 13,777\n"
        f"Found   : {unique_condition_ids_025}"
    )


print(
    "Expected unique condition_id count: PASS"
)


# ================================================================================================
# 8. DICOM LINK FIELD VALIDATION
# ================================================================================================

print("\n6. DICOM LINK FIELD VALIDATION")
print("-" * 100)


url_series_025 = (
    CXR_MASTER_SOURCE_025[
        "series_instance_content_url"
    ]
    .astype(str)
    .str.strip()
)


missing_urls_025 = (
    url_series_025
    .eq("")
    .sum()
)


unique_urls_025 = (
    url_series_025
    .replace("", np.nan)
    .nunique()
)


if missing_urls_025 != 0:

    raise ValueError(
        f"CXR source contains {missing_urls_025} "
        f"records with missing series_instance_content_url."
    )


print(
    "Missing series_instance_content_url:",
    missing_urls_025
)

print(
    "Unique series_instance_content_url:",
    unique_urls_025
)


# At source level we expect every CXR record to represent
# a unique linked image record.

if unique_urls_025 != len(CXR_MASTER_SOURCE_025):

    raise ValueError(
        "\nThe source CXR metadata does not have a one-to-one "
        "mapping between rows and series_instance_content_url.\n"
        f"Rows: {len(CXR_MASTER_SOURCE_025)}\n"
        f"Unique URLs: {unique_urls_025}"
    )


print(
    "Source-level URL uniqueness: PASS"
)


# ================================================================================================
# 9. LOAD GENOMICS SOURCE
# ================================================================================================

print("\n7. LOAD GENOMICS SOURCE FOR MULTIMODAL INTERSECTION")
print("-" * 100)


GENOMICS_SOURCE_FILENAME_025 = (
    "TB_Portals_Genomics_March_2025.csv"
)


GENOMICS_SOURCE_FILE_025 = find_unique_exact_file_025(
    METADATA_DIR,
    GENOMICS_SOURCE_FILENAME_025
)


print(
    "Genomics file:",
    GENOMICS_SOURCE_FILE_025
)


GENOMICS_SOURCE_025 = pd.read_csv(
    GENOMICS_SOURCE_FILE_025,
    dtype=str,
    low_memory=False
)


GENOMICS_SOURCE_025 = (
    GENOMICS_SOURCE_025
    .fillna("")
)


print(
    "Genomics rows:",
    len(GENOMICS_SOURCE_025)
)


if len(GENOMICS_SOURCE_025) != 4_467:

    raise ValueError(
        "\nUnexpected genomics source row count.\n"
        f"Expected: 4,467\n"
        f"Found   : {len(GENOMICS_SOURCE_025)}"
    )


if "condition_id" not in GENOMICS_SOURCE_025.columns:

    raise ValueError(
        "Genomics source does not contain condition_id."
    )


genomics_condition_ids_025 = set(
    GENOMICS_SOURCE_025[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)


if "" in genomics_condition_ids_025:

    raise ValueError(
        "Genomics source contains missing condition_id values."
    )


print(
    "Genomics source validation: PASS"
)


# ================================================================================================
# 10. COMPUTE UNIQUE CASE INTERSECTION
# ================================================================================================

print("\n8. MULTIMODAL CASE INTERSECTION")
print("-" * 100)


cxr_condition_ids_025 = set(
    condition_series_025
)


common_condition_ids_025 = (
    cxr_condition_ids_025
    .intersection(
        genomics_condition_ids_025
    )
)


print(
    "Unique CXR cases       :",
    len(cxr_condition_ids_025)
)

print(
    "Unique genomic cases   :",
    len(genomics_condition_ids_025)
)

print(
    "Common multimodal cases:",
    len(common_condition_ids_025)
)


EXPECTED_COMMON_CASES_025 = 3_081


if len(common_condition_ids_025) != EXPECTED_COMMON_CASES_025:

    raise ValueError(
        "\nUnexpected multimodal case intersection.\n"
        f"Expected: {EXPECTED_COMMON_CASES_025}\n"
        f"Found   : {len(common_condition_ids_025)}"
    )


print(
    "Expected common case count: PASS"
)


# ================================================================================================
# 11. CREATE THE 3,859-RECORD MULTIMODAL CXR MASTER
# ================================================================================================

print("\n9. BUILD 3,859-RECORD MULTIMODAL CXR MASTER")
print("-" * 100)


CXR_MULTIMODAL_MASTER_025 = (
    CXR_MASTER_SOURCE_025[
        condition_series_025.isin(
            common_condition_ids_025
        )
    ]
    .copy()
)


EXPECTED_MULTIMODAL_CXR_ROWS_025 = 3_859


if len(CXR_MULTIMODAL_MASTER_025) != EXPECTED_MULTIMODAL_CXR_ROWS_025:

    raise ValueError(
        "\nUnexpected multimodal CXR row count.\n"
        f"Expected: {EXPECTED_MULTIMODAL_CXR_ROWS_025}\n"
        f"Found   : {len(CXR_MULTIMODAL_MASTER_025)}"
    )


multimodal_unique_conditions_025 = (
    CXR_MULTIMODAL_MASTER_025[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
    .nunique()
)


if multimodal_unique_conditions_025 != EXPECTED_COMMON_CASES_025:

    raise ValueError(
        "\nUnexpected unique condition count in multimodal CXR master.\n"
        f"Expected: {EXPECTED_COMMON_CASES_025}\n"
        f"Found   : {multimodal_unique_conditions_025}"
    )


print(
    "Multimodal CXR records :",
    len(CXR_MULTIMODAL_MASTER_025)
)

print(
    "Unique condition_id    :",
    multimodal_unique_conditions_025
)

print(
    "3,859-record master    : PASS"
)


# ================================================================================================
# 12. MULTIMODAL URL UNIQUENESS
# ================================================================================================

multimodal_url_count_025 = (
    CXR_MULTIMODAL_MASTER_025[
        "series_instance_content_url"
    ]
    .astype(str)
    .str.strip()
    .nunique()
)


if multimodal_url_count_025 != EXPECTED_MULTIMODAL_CXR_ROWS_025:

    raise ValueError(
        "\nMultimodal CXR master contains duplicate "
        "series_instance_content_url values.\n"
        f"Rows: {EXPECTED_MULTIMODAL_CXR_ROWS_025}\n"
        f"Unique URLs: {multimodal_url_count_025}"
    )


print(
    "Multimodal URL uniqueness: PASS"
)


# ================================================================================================
# 13. MODALITY DISTRIBUTION
# ================================================================================================

print("\n10. CXR MODALITY DISTRIBUTION")
print("-" * 100)


modality_distribution_025 = (
    CXR_MULTIMODAL_MASTER_025[
        "series_modality_cd"
    ]
    .astype(str)
    .str.strip()
    .replace("", "MISSING")
    .value_counts()
    .sort_index()
)


print(
    modality_distribution_025.to_string()
)


EXPECTED_MODALITY_COUNTS_025 = {

    "CR": 1537,

    "XC": 2270,

    "XA": 32,

    "RG": 16,

    "DX": 4,

}


observed_modalities_025 = (
    modality_distribution_025
    .to_dict()
)


if observed_modalities_025 != EXPECTED_MODALITY_COUNTS_025:

    raise ValueError(
        "\nCXR modality distribution does not match the "
        "validated multimodal master distribution.\n\n"
        f"Expected:\n{EXPECTED_MODALITY_COUNTS_025}\n\n"
        f"Observed:\n{observed_modalities_025}"
    )


print(
    "\nModality distribution: PASS"
)


# ================================================================================================
# 14. CXR OUTLIER DISTRIBUTION
# ================================================================================================

print("\n11. CXR OUTLIER DISTRIBUTION")
print("-" * 100)


cxr_outlier_distribution_025 = (
    CXR_MULTIMODAL_MASTER_025[
        "cxr_outlier"
    ]
    .astype(str)
    .str.strip()
    .replace("", "MISSING")
    .value_counts()
    .sort_index()
)


print(
    cxr_outlier_distribution_025.to_string()
)


expected_outlier_distribution_025 = {

    "normal": 2915,

    "outlier": 932,

    "error": 12,

}


observed_outlier_distribution_025 = (
    cxr_outlier_distribution_025
    .to_dict()
)


if observed_outlier_distribution_025 != expected_outlier_distribution_025:

    raise ValueError(
        "\nUnexpected cxr_outlier distribution.\n\n"
        f"Expected:\n{expected_outlier_distribution_025}\n\n"
        f"Observed:\n{observed_outlier_distribution_025}"
    )


print(
    "\nCXR outlier distribution: PASS"
)


# ================================================================================================
# 15. IMAGING DATE VALIDATION
# ================================================================================================

print("\n12. IMAGING DATE VALIDATION")
print("-" * 100)


imaging_date_numeric_025 = pd.to_numeric(
    CXR_MULTIMODAL_MASTER_025[
        "imaging_date"
    ],
    errors="coerce"
)


missing_imaging_dates_025 = (
    imaging_date_numeric_025
    .isna()
    .sum()
)


if missing_imaging_dates_025 != 0:

    raise ValueError(
        f"Multimodal CXR master contains "
        f"{missing_imaging_dates_025} invalid/missing imaging_date values."
    )


print(
    "Missing/invalid imaging_date:",
    missing_imaging_dates_025
)

print(
    "Minimum imaging_date:",
    imaging_date_numeric_025.min()
)

print(
    "Maximum imaging_date:",
    imaging_date_numeric_025.max()
)

print(
    "Median imaging_date:",
    imaging_date_numeric_025.median()
)


if imaging_date_numeric_025.min() != -3652:

    raise ValueError(
        "Unexpected minimum imaging_date."
    )


if imaging_date_numeric_025.max() != 2623:

    raise ValueError(
        "Unexpected maximum imaging_date."
    )


print(
    "Imaging date validation: PASS"
)


# ================================================================================================
# 16. MULTIPLE IMAGES PER CONDITION ANALYSIS
# ================================================================================================

print("\n13. IMAGES PER CONDITION")
print("-" * 100)


images_per_condition_025 = (
    CXR_MULTIMODAL_MASTER_025
    .groupby(
        "condition_id",
        dropna=False
    )
    .size()
    .sort_index()
)


images_per_condition_distribution_025 = (
    images_per_condition_025
    .value_counts()
    .sort_index()
)


print(
    images_per_condition_distribution_025.to_string()
)


multiple_image_cases_025 = (
    images_per_condition_025
    .gt(1)
    .sum()
)


print()
print(
    "Cases with >1 CXR:",
    multiple_image_cases_025
)


if multiple_image_cases_025 != 485:

    raise ValueError(
        "\nUnexpected number of cases with multiple CXRs.\n"
        f"Expected: 485\n"
        f"Found   : {multiple_image_cases_025}"
    )


print(
    "Multiple-image case validation: PASS"
)


# ================================================================================================
# 17. CREATE EXPLICIT MASTER IDENTIFIER
# ================================================================================================

CXR_MULTIMODAL_MASTER_025 = (
    CXR_MULTIMODAL_MASTER_025
    .reset_index(drop=True)
)


CXR_MULTIMODAL_MASTER_025.insert(
    0,
    "cxr_master_row_id",
    np.arange(
        1,
        len(CXR_MULTIMODAL_MASTER_025) + 1
    )
)


# ================================================================================================
# 18. ADD DERIVED AUDIT FIELDS
# ================================================================================================

CXR_MULTIMODAL_MASTER_025[
    "imaging_date_num"
] = imaging_date_numeric_025.reset_index(
    drop=True
)


CXR_MULTIMODAL_MASTER_025[
    "images_per_condition"
] = (
    CXR_MULTIMODAL_MASTER_025[
        "condition_id"
    ]
    .map(
        images_per_condition_025
    )
    .astype(int)
)


CXR_MULTIMODAL_MASTER_025[
    "has_multiple_cxr"
] = (
    CXR_MULTIMODAL_MASTER_025[
        "images_per_condition"
    ]
    > 1
)


CXR_MULTIMODAL_MASTER_025[
    "has_genomics"
] = True


# ================================================================================================
# 19. FINAL MASTER SCHEMA VALIDATION
# ================================================================================================

print("\n14. FINAL MASTER VALIDATION")
print("-" * 100)


if len(CXR_MULTIMODAL_MASTER_025) != 3_859:

    raise ValueError(
        "Final CXR master does not contain 3,859 records."
    )


if (
    CXR_MULTIMODAL_MASTER_025[
        "condition_id"
    ]
    .nunique()
    != 3_081
):

    raise ValueError(
        "Final CXR master does not contain 3,081 unique conditions."
    )


if (
    CXR_MULTIMODAL_MASTER_025[
        "series_instance_content_url"
    ]
    .nunique()
    != 3_859
):

    raise ValueError(
        "Final CXR master does not contain 3,859 unique DICOM link values."
    )


if (
    CXR_MULTIMODAL_MASTER_025[
        "condition_id"
    ]
    .isna()
    .any()
):

    raise ValueError(
        "Final CXR master contains NaN condition_id values."
    )


# ================================================================================================
# 20. SAVE AUTHORITATIVE MASTER
# ================================================================================================

CXR_MASTER_OUTPUT_025 = (
    NOTEBOOK_03_OUTPUT_DIR /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)


CXR_MULTIMODAL_MASTER_025.to_csv(
    CXR_MASTER_OUTPUT_025,
    index=False
)


if not CXR_MASTER_OUTPUT_025.is_file():

    raise IOError(
        "CXR master output file was not created."
    )


# ================================================================================================
# 21. SAVE MODALITY SUMMARY
# ================================================================================================

CXR_MODALITY_SUMMARY_OUTPUT_025 = (
    NOTEBOOK_03_OUTPUT_DIR /
    "TB_Portals_March2025_3859_CXR_Modality_Summary.csv"
)


modality_summary_output_025 = (
    modality_distribution_025
    .rename_axis("series_modality_cd")
    .reset_index(
        name="record_count"
    )
)


modality_summary_output_025.to_csv(
    CXR_MODALITY_SUMMARY_OUTPUT_025,
    index=False
)


# ================================================================================================
# 22. SAVE CONDITION IMAGE-COUNT SUMMARY
# ================================================================================================

CXR_CONDITION_IMAGE_COUNT_OUTPUT_025 = (
    NOTEBOOK_03_OUTPUT_DIR /
    "TB_Portals_March2025_3859_CXR_Condition_Image_Count.csv"
)


condition_image_count_output_025 = (
    images_per_condition_025
    .rename("cxr_count")
    .reset_index()
)


condition_image_count_output_025.to_csv(
    CXR_CONDITION_IMAGE_COUNT_OUTPUT_025,
    index=False
)


# ================================================================================================
# 23. FINAL HARD ASSERTIONS
# ================================================================================================

assert len(CXR_MASTER_SOURCE_025) == 16_723

assert len(GENOMICS_SOURCE_025) == 4_467

assert len(common_condition_ids_025) == 3_081

assert len(CXR_MULTIMODAL_MASTER_025) == 3_859

assert (
    CXR_MULTIMODAL_MASTER_025[
        "condition_id"
    ].nunique()
    == 3_081
)

assert (
    CXR_MULTIMODAL_MASTER_025[
        "series_instance_content_url"
    ].nunique()
    == 3_859
)

assert multiple_image_cases_025 == 485

assert missing_imaging_dates_025 == 0


# ================================================================================================
# 24. FINAL STATUS
# ================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.25 — COMPLETED SUCCESSFULLY")
print("=" * 100)

print()
print("CXR source records             : 16,723")
print("CXR unique conditions          : 13,777")

print()
print("Genomics source records        : 4,467")
print("Common multimodal conditions   : 3,081")

print()
print("Multimodal CXR records         : 3,859")
print("Unique CXR DICOM links         : 3,859")

print()
print("CXR modalities:")
print(modality_distribution_025.to_string())

print()
print("CXR outlier values:")
print(cxr_outlier_distribution_025.to_string())

print()
print("Cases with multiple CXRs       :", multiple_image_cases_025)

print()
print("Imaging date range             :",
      imaging_date_numeric_025.min(),
      "to",
      imaging_date_numeric_025.max())

print()
print("Authoritative master saved to:")
print(CXR_MASTER_OUTPUT_025)

print()
print("Modality summary saved to:")
print(CXR_MODALITY_SUMMARY_OUTPUT_025)

print()
print("Condition image-count summary saved to:")
print(CXR_CONDITION_IMAGE_COUNT_OUTPUT_025)

print()
print("NO IMAGE EXCLUSIONS MADE")
print("NO DICOM FILES MODIFIED")
print("NO PREPROCESSING PERFORMED")
print("NO TARGET SELECTED")
print("NO MODELING PERFORMED")

print()
print("STEP 3B.6.25 STATUS: PASS")

print("=" * 100)
```

```python
# ================================================================================================
# STEP 3B.6.26
# DICOM / HEADER / STRUCTURAL PROVENANCE VALIDATION
# ================================================================================================
#
# DESIGN PRINCIPLE
# ----------------
# This step deliberately avoids broad dataframe merges.
#
# The authoritative CXR master owns:
#
#     condition_id
#     imagingstudy_id
#     imaging_date
#     series_instance_content_url
#     series_modality_cd
#     cxr_outlier
#
# The Header QC artifact owns DICOM technical evidence.
#
# The Structural Screening artifact owns structural screening evidence.
#
# The common immutable provenance key is:
#
#     series_instance_content_url
#
# We validate each evidence layer independently and then construct a
# controlled technical provenance table with explicitly selected columns.
#
# NO clinical exclusion is made here.
# NO image is removed.
# NO DICOM is modified.
# NO preprocessing is performed.
# NO target is selected.
#
# ================================================================================================


print("=" * 100)
print("STEP 3B.6.26 — DICOM / HEADER / STRUCTURAL PROVENANCE VALIDATION")
print("=" * 100)


# ================================================================================================
# 1. STRICT FILE LOCATOR
# ================================================================================================

def find_unique_exact_file_026(
    root_directory,
    filename
):

    root_directory = Path(root_directory)

    matches = [
        path
        for path in root_directory.rglob(filename)
        if path.is_file()
    ]

    if len(matches) == 0:

        raise FileNotFoundError(
            f"\nRequired artifact not found.\n"
            f"Filename: {filename}\n"
            f"Search root: {root_directory}"
        )

    if len(matches) > 1:

        match_text = "\n".join(
            f"  - {path}"
            for path in matches
        )

        raise RuntimeError(
            "\nMultiple exact copies found.\n"
            "Execution stopped to prevent provenance ambiguity.\n\n"
            f"Filename: {filename}\n\n"
            f"Matches:\n{match_text}"
        )

    return matches[0]


# ================================================================================================
# 2. LOCATE ARTIFACTS
# ================================================================================================

HEADER_QC_FILENAME_026 = (
    "TB_Portals_March2025_DICOM_Header_QC_All_3859.csv"
)

STRUCTURAL_SCREENING_FILENAME_026 = (
    "TB_Portals_March2025_Full_3859_Structural_Screening.csv"
)


HEADER_QC_FILE_026 = find_unique_exact_file_026(
    METADATA_DIR,
    HEADER_QC_FILENAME_026
)


STRUCTURAL_SCREENING_FILE_026 = find_unique_exact_file_026(
    METADATA_DIR,
    STRUCTURAL_SCREENING_FILENAME_026
)


print("\n1. AUTHORITATIVE QC ARTIFACTS")
print("-" * 100)

print(
    "DICOM Header QC:"
)

print(
    HEADER_QC_FILE_026
)

print()

print(
    "Structural Screening:"
)

print(
    STRUCTURAL_SCREENING_FILE_026
)


# ================================================================================================
# 3. LOAD ARTIFACTS
# ================================================================================================

HEADER_QC_026 = pd.read_csv(
    HEADER_QC_FILE_026,
    dtype=str,
    low_memory=False
).fillna("")


STRUCTURAL_SCREENING_026 = pd.read_csv(
    STRUCTURAL_SCREENING_FILE_026,
    dtype=str,
    low_memory=False
).fillna("")


print("\n2. ARTIFACT LOAD")
print("-" * 100)

print(
    "CXR master records:",
    len(CXR_MULTIMODAL_MASTER_025)
)

print(
    "Header QC records:",
    len(HEADER_QC_026)
)

print(
    "Structural records:",
    len(STRUCTURAL_SCREENING_026)
)


# ================================================================================================
# 4. ROW COUNT VALIDATION
# ================================================================================================

EXPECTED_CXR_ROWS_026 = 3_859

EXPECTED_CONDITIONS_026 = 3_081


if len(CXR_MULTIMODAL_MASTER_025) != EXPECTED_CXR_ROWS_026:

    raise ValueError(
        "CXR master does not contain 3,859 records."
    )


if len(HEADER_QC_026) != EXPECTED_CXR_ROWS_026:

    raise ValueError(
        "Header QC does not contain 3,859 records."
    )


if len(STRUCTURAL_SCREENING_026) != EXPECTED_CXR_ROWS_026:

    raise ValueError(
        "Structural screening does not contain 3,859 records."
    )


print(
    "CXR master row count: PASS"
)

print(
    "Header QC row count: PASS"
)

print(
    "Structural screening row count: PASS"
)


# ================================================================================================
# 5. DEFINE PROVENANCE KEY
# ================================================================================================

PROVENANCE_KEY_026 = (
    "series_instance_content_url"
)


print("\n3. PROVENANCE KEY")
print("-" * 100)

print(
    "Primary provenance key:",
    PROVENANCE_KEY_026
)


# ================================================================================================
# 6. VERIFY PROVENANCE KEY EXISTS
# ================================================================================================

for name_026, dataframe_026 in [

    (
        "CXR master",
        CXR_MULTIMODAL_MASTER_025
    ),

    (
        "DICOM Header QC",
        HEADER_QC_026
    ),

    (
        "Structural screening",
        STRUCTURAL_SCREENING_026
    ),

]:

    if PROVENANCE_KEY_026 not in dataframe_026.columns:

        raise ValueError(
            f"{name_026} does not contain "
            f"{PROVENANCE_KEY_026}."
        )


print(
    "Provenance key availability: PASS"
)


# ================================================================================================
# 7. NORMALIZE PROVENANCE KEYS
# ================================================================================================

master_key_series_026 = (
    CXR_MULTIMODAL_MASTER_025[
        PROVENANCE_KEY_026
    ]
    .astype(str)
    .str.strip()
)


header_key_series_026 = (
    HEADER_QC_026[
        PROVENANCE_KEY_026
    ]
    .astype(str)
    .str.strip()
)


structural_key_series_026 = (
    STRUCTURAL_SCREENING_026[
        PROVENANCE_KEY_026
    ]
    .astype(str)
    .str.strip()
)


# ================================================================================================
# 8. MISSING KEY VALIDATION
# ================================================================================================

missing_master_keys_026 = int(
    master_key_series_026.eq("").sum()
)

missing_header_keys_026 = int(
    header_key_series_026.eq("").sum()
)

missing_structural_keys_026 = int(
    structural_key_series_026.eq("").sum()
)


print("\n4. MISSING PROVENANCE KEYS")
print("-" * 100)

print(
    "CXR master:",
    missing_master_keys_026
)

print(
    "Header QC:",
    missing_header_keys_026
)

print(
    "Structural screening:",
    missing_structural_keys_026
)


if missing_master_keys_026 != 0:
    raise ValueError(
        "CXR master contains missing provenance keys."
    )


if missing_header_keys_026 != 0:
    raise ValueError(
        "Header QC contains missing provenance keys."
    )


if missing_structural_keys_026 != 0:
    raise ValueError(
        "Structural screening contains missing provenance keys."
    )


print(
    "Missing-key validation: PASS"
)


# ================================================================================================
# 9. DUPLICATE KEY VALIDATION
# ================================================================================================

duplicate_master_keys_026 = int(
    master_key_series_026.duplicated().sum()
)

duplicate_header_keys_026 = int(
    header_key_series_026.duplicated().sum()
)

duplicate_structural_keys_026 = int(
    structural_key_series_026.duplicated().sum()
)


print("\n5. DUPLICATE PROVENANCE KEYS")
print("-" * 100)

print(
    "CXR master:",
    duplicate_master_keys_026
)

print(
    "Header QC:",
    duplicate_header_keys_026
)

print(
    "Structural screening:",
    duplicate_structural_keys_026
)


if duplicate_master_keys_026 != 0:
    raise ValueError(
        "CXR master contains duplicate provenance keys."
    )


if duplicate_header_keys_026 != 0:
    raise ValueError(
        "Header QC contains duplicate provenance keys."
    )


if duplicate_structural_keys_026 != 0:
    raise ValueError(
        "Structural screening contains duplicate provenance keys."
    )


print(
    "Duplicate-key validation: PASS"
)


# ================================================================================================
# 10. PROVENANCE SET COMPARISON
# ================================================================================================

master_key_set_026 = set(
    master_key_series_026
)

header_key_set_026 = set(
    header_key_series_026
)

structural_key_set_026 = set(
    structural_key_series_026
)


master_to_header_missing_026 = (
    master_key_set_026
    -
    header_key_set_026
)

header_to_master_extra_026 = (
    header_key_set_026
    -
    master_key_set_026
)

master_to_structural_missing_026 = (
    master_key_set_026
    -
    structural_key_set_026
)

structural_to_master_extra_026 = (
    structural_key_set_026
    -
    master_key_set_026
)


print("\n6. PROVENANCE SET RECONCILIATION")
print("-" * 100)

print(
    "Master → Header QC missing:",
    len(master_to_header_missing_026)
)

print(
    "Header QC → Master extra:",
    len(header_to_master_extra_026)
)

print(
    "Master → Structural missing:",
    len(master_to_structural_missing_026)
)

print(
    "Structural → Master extra:",
    len(structural_to_master_extra_026)
)


if master_to_header_missing_026:
    raise ValueError(
        "CXR master contains links missing from Header QC."
    )


if header_to_master_extra_026:
    raise ValueError(
        "Header QC contains links not present in CXR master."
    )


if master_to_structural_missing_026:
    raise ValueError(
        "CXR master contains links missing from structural screening."
    )


if structural_to_master_extra_026:
    raise ValueError(
        "Structural screening contains links not present in CXR master."
    )


print(
    "Provenance set reconciliation: PASS"
)


# ================================================================================================
# 11. AUTHORITATIVE CONDITION_ID MAP
# ================================================================================================
#
# DO NOT merge broad tables here.
#
# Create a dedicated one-to-one mapping from the authoritative CXR master.
#
# ================================================================================================

print("\n7. AUTHORITATIVE CONDITION_ID MAP")
print("-" * 100)


condition_map_026 = {

    str(row[PROVENANCE_KEY_026]).strip():
        str(row["condition_id"]).strip()

    for _, row
    in CXR_MULTIMODAL_MASTER_025[
        [
            PROVENANCE_KEY_026,
            "condition_id"
        ]
    ].iterrows()

}


if len(condition_map_026) != 3_859:

    raise ValueError(
        "Authoritative provenance → condition_id map "
        "does not contain 3,859 entries."
    )


missing_condition_mapping_026 = [

    key

    for key
    in structural_key_series_026

    if key not in condition_map_026

]


if missing_condition_mapping_026:

    raise ValueError(
        "\nStructural provenance keys missing from "
        "authoritative condition map.\n"
        f"Count: {len(missing_condition_mapping_026)}"
    )


structural_condition_ids_026 = [

    condition_map_026[key]

    for key
    in structural_key_series_026

]


structural_mapped_condition_series_026 = pd.Series(
    structural_condition_ids_026,
    index=STRUCTURAL_SCREENING_026.index,
    dtype="string"
)


print(
    "Structural records mapped:",
    len(structural_mapped_condition_series_026)
)

print(
    "Unique mapped conditions:",
    structural_mapped_condition_series_026.nunique()
)

print(
    "Unmapped records:",
    len(missing_condition_mapping_026)
)


if structural_mapped_condition_series_026.nunique() != 3_081:

    raise ValueError(
        "Mapped structural records do not represent "
        "the expected 3,081 conditions."
    )


print(
    "condition_id provenance mapping: PASS"
)


# ================================================================================================
# 12. HEADER QC ACTUAL DICOM MODALITY
# ================================================================================================

print("\n8. ACTUAL DICOM MODALITY")
print("-" * 100)


if "dicom_modality" not in HEADER_QC_026.columns:

    raise ValueError(
        "Header QC does not contain dicom_modality."
    )


actual_modality_distribution_026 = (
    HEADER_QC_026[
        "dicom_modality"
    ]
    .astype(str)
    .str.strip()
    .replace("", "MISSING")
    .value_counts()
    .sort_index()
)


print(
    actual_modality_distribution_026.to_string()
)


EXPECTED_ACTUAL_MODALITY_026 = {

    "CR": 1163,

    "DX": 4,

    "OT": 394,

    "RG": 15,

    "XA": 32,

    "XC": 2251,

}


if (
    actual_modality_distribution_026.to_dict()
    !=
    EXPECTED_ACTUAL_MODALITY_026
):

    raise ValueError(
        "\nActual DICOM modality distribution mismatch.\n\n"
        f"Expected:\n{EXPECTED_ACTUAL_MODALITY_026}\n\n"
        f"Observed:\n{actual_modality_distribution_026.to_dict()}"
    )


print(
    "\nActual DICOM modality validation: PASS"
)


# ================================================================================================
# 13. STRUCTURAL SCHEMA VALIDATION
# ================================================================================================

print("\n9. STRUCTURAL SCREENING SCHEMA")
print("-" * 100)


REQUIRED_STRUCTURAL_FIELDS_026 = [

    "screening_id",

    "series_instance_content_url",

    "manifest_file",

    "zip_file",

    "dicom_modality",

    "series_modality_cd",

    "photometric_interpretation",

    "rows",

    "columns",

    "aspect_ratio",

    "number_of_frames",

    "processing_status",

    "processing_error",

    "screen_bright_fraction",

    "screen_dark_fraction",

    "largest_margin_fraction",

    "possible_composite_structure",

    "aspect_percentile_full_cohort",

    "review_flags",

    "review_flag_count",

    "review_priority",

]


missing_structural_fields_026 = [

    column

    for column
    in REQUIRED_STRUCTURAL_FIELDS_026

    if column
    not in STRUCTURAL_SCREENING_026.columns

]


if missing_structural_fields_026:

    raise ValueError(
        "\nRequired structural fields missing:\n"
        +
        "\n".join(
            f"  - {column}"
            for column
            in missing_structural_fields_026
        )
    )


print(
    "Required structural fields:",
    len(REQUIRED_STRUCTURAL_FIELDS_026)
)

print(
    "Structural schema validation: PASS"
)


# ================================================================================================
# 14. STRUCTURAL PROCESSING STATUS
# ================================================================================================

print("\n10. STRUCTURAL PROCESSING STATUS")
print("-" * 100)


processing_status_026 = (
    STRUCTURAL_SCREENING_026[
        "processing_status"
    ]
    .astype(str)
    .str.strip()
)


processing_status_distribution_026 = (
    processing_status_026
    .replace("", "MISSING")
    .value_counts()
    .sort_index()
)


print(
    processing_status_distribution_026.to_string()
)


missing_processing_status_026 = int(
    processing_status_026.eq("").sum()
)


processing_error_count_026 = int(
    STRUCTURAL_SCREENING_026[
        "processing_error"
    ]
    .astype(str)
    .str.strip()
    .ne("")
    .sum()
)


print()
print(
    "Missing processing_status:",
    missing_processing_status_026
)

print(
    "Records with processing_error:",
    processing_error_count_026
)


if missing_processing_status_026 != 0:

    raise ValueError(
        "Structural screening contains missing processing_status."
    )


if processing_error_count_026 != 0:

    raise ValueError(
        "Structural screening contains processing errors."
    )


if processing_status_distribution_026.to_dict() != {
    "PASS": 3859
}:

    raise ValueError(
        "Unexpected structural processing-status distribution."
    )


print(
    "Structural processing status: PASS"
)


# ================================================================================================
# 15. MULTI-FRAME VALIDATION
# ================================================================================================

print("\n11. MULTI-FRAME VALIDATION")
print("-" * 100)


number_of_frames_026 = pd.to_numeric(
    STRUCTURAL_SCREENING_026[
        "number_of_frames"
    ],
    errors="coerce"
)


invalid_frame_values_026 = int(
    number_of_frames_026.isna().sum()
)


multi_frame_count_026 = int(
    number_of_frames_026.gt(1).sum()
)


print(
    "Invalid number_of_frames:",
    invalid_frame_values_026
)

print(
    "Multi-frame records:",
    multi_frame_count_026
)


if invalid_frame_values_026 != 0:

    raise ValueError(
        "Invalid number_of_frames values detected."
    )


if multi_frame_count_026 != 0:

    raise ValueError(
        "Multi-frame records detected."
    )


print(
    "Multi-frame validation: PASS"
)


# ================================================================================================
# 16. STRUCTURAL REVIEW FLAG REPRESENTATION
# ================================================================================================

print("\n12. STRUCTURAL REVIEW FLAGS")
print("-" * 100)


review_flags_026 = (
    STRUCTURAL_SCREENING_026[
        "review_flags"
    ]
    .astype(str)
    .str.strip()
)


review_flag_count_026 = pd.to_numeric(
    STRUCTURAL_SCREENING_026[
        "review_flag_count"
    ],
    errors="coerce"
)


if review_flag_count_026.isna().any():

    raise ValueError(
        "review_flag_count contains invalid values."
    )


review_flag_distribution_026 = (
    review_flags_026
    .replace("", "NO_FLAG")
    .value_counts()
)


print(
    review_flag_distribution_026.to_string()
)

print()

print(
    "Maximum review_flag_count:",
    int(review_flag_count_026.max())
)


# ================================================================================================
# 17. RECONSTRUCT 639 STRUCTURAL CANDIDATES
# ================================================================================================

structural_review_candidate_026 = (
    review_flag_count_026
    .gt(0)
)


structural_review_candidate_count_026 = int(
    structural_review_candidate_026.sum()
)


print("\n13. STRUCTURAL REVIEW CANDIDATES")
print("-" * 100)

print(
    "Records with review_flag_count > 0:",
    structural_review_candidate_count_026
)

print(
    "Percentage of 3,859:",
    round(
        structural_review_candidate_count_026
        /
        3_859
        *
        100,
        2
    ),
    "%"
)


if structural_review_candidate_count_026 != 639:

    raise ValueError(
        "\nExpected 639 structural review candidates.\n"
        f"Found: {structural_review_candidate_count_026}"
    )


print(
    "Structural candidate count: PASS"
)


# ================================================================================================
# 18. STRUCTURAL CANDIDATE ACTUAL MODALITY
# ================================================================================================

print("\n14. STRUCTURAL CANDIDATE MODALITY")
print("-" * 100)


candidate_provenance_keys_026 = (
    structural_key_series_026[
        structural_review_candidate_026
    ]
)


candidate_actual_modality_026 = (
    HEADER_QC_026[
        PROVENANCE_KEY_026
    ]
    .astype(str)
    .str.strip()
    .map(
        dict(
            zip(
                header_key_series_026,
                HEADER_QC_026[
                    "dicom_modality"
                ]
                .astype(str)
                .str.strip()
            )
        )
    )
)


candidate_modality_distribution_026 = (
    candidate_actual_modality_026[
        structural_review_candidate_026
    ]
    .value_counts()
    .sort_index()
)


print(
    candidate_modality_distribution_026.to_string()
)


EXPECTED_CANDIDATE_MODALITY_026 = {

    "CR": 274,

    "OT": 10,

    "XA": 9,

    "XC": 346,

}


if (
    candidate_modality_distribution_026.to_dict()
    !=
    EXPECTED_CANDIDATE_MODALITY_026
):

    raise ValueError(
        "\nUnexpected structural candidate modality distribution.\n\n"
        f"Expected:\n{EXPECTED_CANDIDATE_MODALITY_026}\n\n"
        f"Observed:\n{candidate_modality_distribution_026.to_dict()}"
    )


print(
    "Structural candidate modality: PASS"
)


# ================================================================================================
# 19. BUILD CONTROLLED TECHNICAL PROVENANCE TABLE
# ================================================================================================
#
# This table is constructed using explicit arrays.
#
# We deliberately DO NOT merge entire Header QC and Structural Screening
# tables because overlapping column names can create ambiguous suffixes.
#
# ================================================================================================

print("\n15. BUILD CONTROLLED TECHNICAL PROVENANCE TABLE")
print("-" * 100)


master_selected_026 = (
    CXR_MULTIMODAL_MASTER_025[
        [
            "cxr_master_row_id",
            "condition_id",
            "patient_id",
            "imagingstudy_id",
            "imaging_date",
            PROVENANCE_KEY_026,
            "series_modality_cd",
            "cxr_outlier"
        ]
    ]
    .copy()
)


# --------------------------------------------------------------------------------
# Build direct lookup dictionaries from Header QC.
# --------------------------------------------------------------------------------

header_index_map_026 = {

    key:
        index

    for index, key
    in enumerate(
        header_key_series_026
    )

}


structural_index_map_026 = {

    key:
        index

    for index, key
    in enumerate(
        structural_key_series_026
    )

}


master_keys_ordered_026 = (
    master_selected_026[
        PROVENANCE_KEY_026
    ]
    .astype(str)
    .str.strip()
    .tolist()
)


# --------------------------------------------------------------------------------
# Verify every master key exists in both evidence layers.
# --------------------------------------------------------------------------------

missing_header_for_master_026 = [

    key

    for key
    in master_keys_ordered_026

    if key not in header_index_map_026

]


missing_structural_for_master_026 = [

    key

    for key
    in master_keys_ordered_026

    if key not in structural_index_map_026

]


if missing_header_for_master_026:

    raise ValueError(
        "Master contains keys missing from Header QC."
    )


if missing_structural_for_master_026:

    raise ValueError(
        "Master contains keys missing from Structural Screening."
    )


# --------------------------------------------------------------------------------
# Build Header QC evidence in exactly the master order.
# --------------------------------------------------------------------------------

header_rows_026 = [

    HEADER_QC_026.iloc[
        header_index_map_026[key]
    ]

    for key
    in master_keys_ordered_026

]


header_aligned_026 = pd.DataFrame(
    header_rows_026
).reset_index(
    drop=True
)


# --------------------------------------------------------------------------------
# Build Structural evidence in exactly the master order.
# --------------------------------------------------------------------------------

structural_rows_026 = [

    STRUCTURAL_SCREENING_026.iloc[
        structural_index_map_026[key]
    ]

    for key
    in master_keys_ordered_026

]


structural_aligned_026 = pd.DataFrame(
    structural_rows_026
).reset_index(
    drop=True
)


# ================================================================================================
# 20. ALIGNMENT VALIDATION BEFORE COMBINING
# ================================================================================================

if not (
    header_aligned_026[
        PROVENANCE_KEY_026
    ]
    .astype(str)
    .str.strip()
    .tolist()
    ==
    master_keys_ordered_026
):

    raise ValueError(
        "Header QC alignment with CXR master failed."
    )


if not (
    structural_aligned_026[
        PROVENANCE_KEY_026
    ]
    .astype(str)
    .str.strip()
    .tolist()
    ==
    master_keys_ordered_026
):

    raise ValueError(
        "Structural screening alignment with CXR master failed."
    )


print(
    "Header QC alignment: PASS"
)

print(
    "Structural screening alignment: PASS"
)


# ================================================================================================
# 21. CREATE EXPLICIT TECHNICAL TABLE
# ================================================================================================

technical_provenance_026 = (
    master_selected_026
    .copy()
)


# Header evidence with explicit prefix.
technical_provenance_026[
    "header_dicom_modality"
] = (
    header_aligned_026[
        "dicom_modality"
    ]
    .astype(str)
    .str.strip()
    .values
)


technical_provenance_026[
    "header_photometric_interpretation"
] = (
    header_aligned_026[
        "photometric_interpretation"
    ]
    .astype(str)
    .str.strip()
    .values
)


# Structural evidence with explicit prefix.
technical_provenance_026[
    "structural_dicom_modality"
] = (
    structural_aligned_026[
        "dicom_modality"
    ]
    .astype(str)
    .str.strip()
    .values
)


technical_provenance_026[
    "structural_series_modality_cd"
] = (
    structural_aligned_026[
        "series_modality_cd"
    ]
    .astype(str)
    .str.strip()
    .values
)


technical_provenance_026[
    "structural_photometric_interpretation"
] = (
    structural_aligned_026[
        "photometric_interpretation"
    ]
    .astype(str)
    .str.strip()
    .values
)


technical_provenance_026[
    "structural_rows"
] = (
    pd.to_numeric(
        structural_aligned_026[
            "rows"
        ],
        errors="coerce"
    )
    .values
)


technical_provenance_026[
    "structural_columns"
] = (
    pd.to_numeric(
        structural_aligned_026[
            "columns"
        ],
        errors="coerce"
    )
    .values
)


technical_provenance_026[
    "structural_aspect_ratio"
] = (
    pd.to_numeric(
        structural_aligned_026[
            "aspect_ratio"
        ],
        errors="coerce"
    )
    .values
)


technical_provenance_026[
    "structural_number_of_frames"
] = (
    pd.to_numeric(
        structural_aligned_026[
            "number_of_frames"
        ],
        errors="coerce"
    )
    .values
)


technical_provenance_026[
    "structural_processing_status"
] = (
    structural_aligned_026[
        "processing_status"
    ]
    .astype(str)
    .str.strip()
    .values
)


technical_provenance_026[
    "structural_processing_error"
] = (
    structural_aligned_026[
        "processing_error"
    ]
    .astype(str)
    .str.strip()
    .values
)


technical_provenance_026[
    "structural_possible_composite"
] = (
    structural_aligned_026[
        "possible_composite_structure"
    ]
    .astype(str)
    .str.strip()
    .values
)


technical_provenance_026[
    "structural_review_flags"
] = (
    structural_aligned_026[
        "review_flags"
    ]
    .astype(str)
    .str.strip()
    .values
)


technical_provenance_026[
    "structural_review_flag_count"
] = (
    pd.to_numeric(
        structural_aligned_026[
            "review_flag_count"
        ],
        errors="coerce"
    )
    .values
)


technical_provenance_026[
    "structural_review_priority"
] = (
    structural_aligned_026[
        "review_priority"
    ]
    .astype(str)
    .str.strip()
    .values
)


technical_provenance_026[
    "structural_review_candidate"
] = (
    structural_review_candidate_026
    .values
)


# ================================================================================================
# 22. HEADER ↔ STRUCTURAL MODALITY CONSISTENCY
# ================================================================================================

technical_provenance_026[
    "header_structural_modality_match"
] = (
    technical_provenance_026[
        "header_dicom_modality"
    ]
    ==
    technical_provenance_026[
        "structural_dicom_modality"
    ]
)


modality_mismatch_count_026 = int(
    (
        ~technical_provenance_026[
            "header_structural_modality_match"
        ]
    )
    .sum()
)


print("\n16. HEADER ↔ STRUCTURAL MODALITY CONSISTENCY")
print("-" * 100)

print(
    "Modality mismatches:",
    modality_mismatch_count_026
)


if modality_mismatch_count_026 != 0:

    raise ValueError(
        "\nHeader QC and structural screening disagree "
        "on actual DICOM modality."
    )


print(
    "Header ↔ Structural modality consistency: PASS"
)


# ================================================================================================
# 23. FINAL TABLE VALIDATION
# ================================================================================================

print("\n17. FINAL TECHNICAL PROVENANCE VALIDATION")
print("-" * 100)


assert len(
    technical_provenance_026
) == 3_859


assert (
    technical_provenance_026[
        PROVENANCE_KEY_026
    ]
    .nunique()
    ==
    3_859
)


assert (
    technical_provenance_026[
        "condition_id"
    ]
    .nunique()
    ==
    3_081
)


assert (
    technical_provenance_026[
        "structural_review_candidate"
    ]
    .sum()
    ==
    639
)


assert (
    technical_provenance_026[
        "header_structural_modality_match"
    ]
    .all()
)


assert (
    technical_provenance_026[
        "structural_processing_status"
    ]
    .eq("PASS")
    .all()
)


assert (
    technical_provenance_026[
        "structural_number_of_frames"
    ]
    .eq(1)
    .all()
)


print(
    "Rows                         : 3,859 — PASS"
)

print(
    "Unique condition_id         : 3,081 — PASS"
)

print(
    "Unique provenance links     : 3,859 — PASS"
)

print(
    "Structural candidates       : 639 — PASS"
)

print(
    "Header/structural modality : 0 mismatch — PASS"
)

print(
    "Structural processing       : 3,859 PASS — PASS"
)

print(
    "Multi-frame records         : 0 — PASS"
)


# ================================================================================================
# 24. SAVE TECHNICAL PROVENANCE
# ================================================================================================

TECHNICAL_PROVENANCE_OUTPUT_026 = (
    NOTEBOOK_03_OUTPUT_DIR
    /
    "TB_Portals_March2025_3859_Technical_DICOM_Structural_Provenance.csv"
)


technical_provenance_026.to_csv(
    TECHNICAL_PROVENANCE_OUTPUT_026,
    index=False
)


if not TECHNICAL_PROVENANCE_OUTPUT_026.is_file():

    raise IOError(
        "Technical provenance output was not created."
    )


# ================================================================================================
# 25. SAVE STRUCTURAL REVIEW SUMMARY
# ================================================================================================

STRUCTURAL_FLAG_SUMMARY_OUTPUT_026 = (
    NOTEBOOK_03_OUTPUT_DIR
    /
    "TB_Portals_March2025_3859_Structural_Review_Flag_Summary.csv"
)


structural_flag_summary_026 = (
    review_flag_distribution_026
    .rename_axis(
        "review_flags"
    )
    .reset_index(
        name="record_count"
    )
)


structural_flag_summary_026[
    "percentage_of_3859"
] = (
    structural_flag_summary_026[
        "record_count"
    ]
    /
    3_859
    *
    100
).round(4)


structural_flag_summary_026.to_csv(
    STRUCTURAL_FLAG_SUMMARY_OUTPUT_026,
    index=False
)


# ================================================================================================
# 26. SAVE CANDIDATE MODALITY SUMMARY
# ================================================================================================

STRUCTURAL_CANDIDATE_MODALITY_OUTPUT_026 = (
    NOTEBOOK_03_OUTPUT_DIR
    /
    "TB_Portals_March2025_639_Structural_Candidate_Modality_Summary.csv"
)


candidate_modality_summary_026 = (
    candidate_modality_distribution_026
    .rename_axis(
        "dicom_modality"
    )
    .reset_index(
        name="record_count"
    )
)


candidate_modality_summary_026[
    "percentage_of_639"
] = (
    candidate_modality_summary_026[
        "record_count"
    ]
    /
    639
    *
    100
).round(4)


candidate_modality_summary_026.to_csv(
    STRUCTURAL_CANDIDATE_MODALITY_OUTPUT_026,
    index=False
)


# ================================================================================================
# 27. FINAL HARD ASSERTIONS
# ================================================================================================

assert len(CXR_MULTIMODAL_MASTER_025) == 3_859

assert len(HEADER_QC_026) == 3_859

assert len(STRUCTURAL_SCREENING_026) == 3_859

assert (
    master_key_set_026
    ==
    header_key_set_026
)

assert (
    master_key_set_026
    ==
    structural_key_set_026
)

assert missing_condition_mapping_026 == []

assert mapped_unique_conditions_026 == 3_081 if "mapped_unique_conditions_026" in globals() else True

assert actual_modality_distribution_026.to_dict() == (
    EXPECTED_ACTUAL_MODALITY_026
)

assert structural_review_candidate_count_026 == 639

assert multi_frame_count_026 == 0

assert modality_mismatch_count_026 == 0

assert len(technical_provenance_026) == 3_859

assert (
    technical_provenance_026[
        "condition_id"
    ].nunique()
    ==
    3_081
)


# ================================================================================================
# 28. FINAL STATUS
# ================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.26 — COMPLETED SUCCESSFULLY")
print("=" * 100)

print()
print("CXR master records             : 3,859")
print("DICOM Header QC records        : 3,859")
print("Structural screening records   : 3,859")

print()
print("Unique provenance links        : 3,859")
print("Unique multimodal conditions   : 3,081")

print()
print("Actual DICOM modality:")
print(
    actual_modality_distribution_026.to_string()
)

print()
print(
    "Structural review candidates  : 639"
)

print(
    "Header/structural modality mismatches:",
    modality_mismatch_count_026
)

print()
print("Technical provenance saved to:")
print(
    TECHNICAL_PROVENANCE_OUTPUT_026
)

print()
print("Structural review summary saved to:")
print(
    STRUCTURAL_FLAG_SUMMARY_OUTPUT_026
)

print()
print("Candidate modality summary saved to:")
print(
    STRUCTURAL_CANDIDATE_MODALITY_OUTPUT_026
)

print()
print("NO IMAGE EXCLUSIONS MADE")
print("NO DICOM FILES MODIFIED")
print("NO CLINICAL ELIGIBILITY DECISION MADE")
print("NO PREPROCESSING PERFORMED")
print("NO TARGET SELECTED")
print("NO MODELING PERFORMED")

print()
print("STEP 3B.6.26 STATUS: PASS")

print("=" * 100)
```

```python
# ================================================================================================
# STEP 3B.6.27
# STRUCTURAL HUMAN-REVIEW RECONSTRUCTION AND PROVENANCE VALIDATION
# ================================================================================================
#
# AUTHORITATIVE STRUCTURAL REVIEW ARTIFACTS
#
#   1. 2-flag review
#   2. 3-flag review
#   3. Large-margin adjudication
#   4. Extreme-aspect adjudication
#   5. Composite adjudication
#
# IMPORTANT:
# - The first two artifacts use visual_review_decision.
# - The last three artifacts use human_review_decision.
# - PENDING is not a completed human decision.
# - condition_id is taken from the authoritative CXR master.
# - series_instance_content_url is the immutable provenance key.
#
# NO automatic exclusion
# NO DICOM modification
# NO preprocessing
# NO target selection
# NO modelling
#
# ================================================================================================


print("=" * 100)
print("STEP 3B.6.27 — STRUCTURAL HUMAN-REVIEW RECONSTRUCTION")
print("=" * 100)


# ================================================================================================
# 1. VERIFY REQUIRED PREVIOUS STEP OBJECTS
# ================================================================================================

required_objects_027 = [
    "CXR_MULTIMODAL_MASTER_025",
    "technical_provenance_026",
]


missing_objects_027 = [
    name
    for name in required_objects_027
    if name not in globals()
]


if missing_objects_027:

    raise RuntimeError(
        "\nRequired objects from previous Notebook 03 steps are missing:\n"
        +
        "\n".join(
            f"  - {name}"
            for name in missing_objects_027
        )
        +
        "\n\nRun Steps 3B.6.24–3B.6.26 first."
    )


# ================================================================================================
# 2. AUTHORITATIVE REVIEW ARTIFACTS
# ================================================================================================

STRUCTURAL_REVIEW_FILENAMES_027 = [

    "TB_Portals_March2025_Step_3B_6_13_2Flag_Human_Review_FINAL.csv",

    "TB_Portals_March2025_Step_3B_6_13_3Flag_Human_Review_FINAL.csv",

    "TB_Portals_March2025_Step_3B_6_18B_LargeMargin_Human_Adjudication_FINAL.csv",

    "TB_Portals_March2025_Step_3B_6_19B_ExtremeAspect_Human_Adjudication_FINAL.csv",

    "TB_Portals_March2025_Step_3B_6_20B_Composite_Human_Adjudication_FINAL.csv",

]


def locate_unique_artifact_027(root_directory, filename):

    root_directory = Path(root_directory)

    matches = [
        path
        for path in root_directory.rglob(filename)
        if path.is_file()
    ]

    if len(matches) == 0:

        raise FileNotFoundError(
            "\nRequired structural review artifact not found.\n"
            f"Filename: {filename}\n"
            f"Search root: {root_directory}"
        )

    if len(matches) > 1:

        raise RuntimeError(
            "\nMultiple copies of the same artifact were found.\n"
            "Execution stopped to prevent provenance ambiguity.\n\n"
            f"Filename: {filename}\n\n"
            +
            "\n".join(
                f"  - {path}"
                for path in matches
            )
        )

    return matches[0]


STRUCTURAL_REVIEW_FILES_027 = {

    filename:
        locate_unique_artifact_027(
            METADATA_DIR,
            filename
        )

    for filename
    in STRUCTURAL_REVIEW_FILENAMES_027

}


# ================================================================================================
# 3. LOAD REVIEW ARTIFACTS
# ================================================================================================

STRUCTURAL_REVIEW_TABLES_027 = {}


for filename, path in STRUCTURAL_REVIEW_FILES_027.items():

    df = pd.read_csv(
        path,
        dtype=str,
        low_memory=False
    ).fillna("")

    STRUCTURAL_REVIEW_TABLES_027[
        filename
    ] = df


print("\n1. AUTHORITATIVE REVIEW ARTIFACTS")
print("-" * 100)

for filename, path in STRUCTURAL_REVIEW_FILES_027.items():

    print(
        f"{filename}  |  {len(STRUCTURAL_REVIEW_TABLES_027[filename])} rows"
    )


# ================================================================================================
# 4. EXACT SCHEMA-BASED CONFIGURATION
# ================================================================================================
#
# We now know the actual schemas from your output.
#
# First two:
#     visual_review_decision
#
# Last three:
#     human_review_decision
#
# ================================================================================================

REVIEW_CONFIGURATION_027 = {

    STRUCTURAL_REVIEW_FILENAMES_027[0]: {
        "decision_column": "visual_review_decision",
        "status_column": None,
    },

    STRUCTURAL_REVIEW_FILENAMES_027[1]: {
        "decision_column": "visual_review_decision",
        "status_column": None,
    },

    STRUCTURAL_REVIEW_FILENAMES_027[2]: {
        "decision_column": "human_review_decision",
        "status_column": "human_review_status",
    },

    STRUCTURAL_REVIEW_FILENAMES_027[3]: {
        "decision_column": "human_review_decision",
        "status_column": "human_review_status",
    },

    STRUCTURAL_REVIEW_FILENAMES_027[4]: {
        "decision_column": "human_review_decision",
        "status_column": "human_review_status",
    },

}


# ================================================================================================
# 5. VALIDATE CONFIGURATION AGAINST ACTUAL SCHEMAS
# ================================================================================================

print("\n2. SCHEMA CONFIGURATION VALIDATION")
print("-" * 100)


for filename, configuration in REVIEW_CONFIGURATION_027.items():

    df = STRUCTURAL_REVIEW_TABLES_027[
        filename
    ]

    decision_column = configuration[
        "decision_column"
    ]

    status_column = configuration[
        "status_column"
    ]

    if decision_column not in df.columns:

        raise ValueError(
            "\nConfigured decision column does not exist.\n"
            f"Artifact: {filename}\n"
            f"Expected column: {decision_column}\n"
            f"Actual columns: {list(df.columns)}"
        )

    if (
        status_column is not None
        and
        status_column not in df.columns
    ):

        raise ValueError(
            "\nConfigured status column does not exist.\n"
            f"Artifact: {filename}\n"
            f"Expected column: {status_column}\n"
            f"Actual columns: {list(df.columns)}"
        )

    if "series_instance_content_url" not in df.columns:

        raise ValueError(
            "\nProvenance key missing.\n"
            f"Artifact: {filename}"
        )


    print(
        f"{filename}"
    )

    print(
        f"  Decision column: {decision_column}"
    )

    print(
        f"  Status column  : {status_column}"
    )


print(
    "\nSchema configuration validation: PASS"
)


# ================================================================================================
# 6. RAW DECISION / STATUS AUDIT
# ================================================================================================

print("\n3. RAW DECISION / STATUS VALUES")
print("-" * 100)


for filename, configuration in REVIEW_CONFIGURATION_027.items():

    df = STRUCTURAL_REVIEW_TABLES_027[
        filename
    ]

    decision_column = configuration[
        "decision_column"
    ]

    status_column = configuration[
        "status_column"
    ]

    print()
    print(filename)

    print(
        "\nDecision values:"
    )

    print(
        df[
            decision_column
        ]
        .astype(str)
        .str.strip()
        .replace("", "<BLANK>")
        .value_counts(dropna=False)
        .sort_index()
        .to_string()
    )

    if status_column is not None:

        print(
            "\nHuman-review status values:"
        )

        print(
            df[
                status_column
            ]
            .astype(str)
            .str.strip()
            .replace("", "<BLANK>")
            .value_counts(dropna=False)
            .sort_index()
            .to_string()
        )


# ================================================================================================
# 7. NORMALIZE DECISION VALUES
# ================================================================================================

ALLOWED_FINAL_DECISIONS_027 = {
    "RETAIN",
    "EXCLUDE",
}


def normalize_text_027(value):

    value = str(value).strip()

    if value == "":
        return ""

    return value.upper()


# ================================================================================================
# 8. RECONSTRUCT COMPLETED REVIEWS
# ================================================================================================

completed_review_tables_027 = []


for filename, configuration in REVIEW_CONFIGURATION_027.items():

    df = STRUCTURAL_REVIEW_TABLES_027[
        filename
    ].copy()

    decision_column = configuration[
        "decision_column"
    ]

    status_column = configuration[
        "status_column"
    ]


    df[
        "series_instance_content_url"
    ] = (
        df[
            "series_instance_content_url"
        ]
        .astype(str)
        .str.strip()
    )


    df[
        "_decision_normalized_027"
    ] = (
        df[
            decision_column
        ]
        .map(
            normalize_text_027
        )
    )


    # --------------------------------------------------------------------------------------------
    # First two artifacts:
    #
    # visual_review_decision is itself the completed decision field.
    #
    # --------------------------------------------------------------------------------------------

    if status_column is None:

        completed_mask = (
            df[
                "_decision_normalized_027"
            ]
            .isin(
                ALLOWED_FINAL_DECISIONS_027
            )
        )


    # --------------------------------------------------------------------------------------------
    # Last three artifacts:
    #
    # human_review_status must indicate completion.
    #
    # The decision must also be RETAIN/EXCLUDE.
    #
    # --------------------------------------------------------------------------------------------

    else:

        status_normalized = (
            df[
                status_column
            ]
            .map(
                normalize_text_027
            )
        )


        completed_mask = (

            status_normalized.isin(
                {
                    "COMPLETED",
                    "COMPLETE",
                    "REVIEWED",
                    "FINAL",
                }
            )

            &

            df[
                "_decision_normalized_027"
            ]
            .isin(
                ALLOWED_FINAL_DECISIONS_027
            )

        )


    completed = df.loc[
        completed_mask
    ].copy()


    completed[
        "human_review_decision"
    ] = (
        completed[
            "_decision_normalized_027"
        ]
    )


    completed[
        "review_artifact"
    ] = filename


    completed[
        "source_decision_column"
    ] = decision_column


    completed_review_tables_027.append(
        completed
    )


STRUCTURAL_COMPLETED_REVIEWS_RAW_027 = pd.concat(
    completed_review_tables_027,
    ignore_index=True
)


print("\n4. COMPLETED REVIEW RECONSTRUCTION")
print("-" * 100)

print(
    "Completed review rows:",
    len(STRUCTURAL_COMPLETED_REVIEWS_RAW_027)
)


print(
    "\nDecision distribution:"
)

print(
    STRUCTURAL_COMPLETED_REVIEWS_RAW_027[
        "human_review_decision"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)


# ================================================================================================
# 9. PROVENANCE VALIDATION
# ================================================================================================

completed_urls_027 = (
    STRUCTURAL_COMPLETED_REVIEWS_RAW_027[
        "series_instance_content_url"
    ]
    .astype(str)
    .str.strip()
)


print("\n5. COMPLETED REVIEW PROVENANCE")
print("-" * 100)


missing_completed_urls_027 = int(
    completed_urls_027.eq("").sum()
)


if missing_completed_urls_027 != 0:

    raise ValueError(
        "Completed review rows contain missing provenance URLs."
    )


master_urls_027 = set(
    CXR_MULTIMODAL_MASTER_025[
        "series_instance_content_url"
    ]
    .astype(str)
    .str.strip()
)


completed_url_set_027 = set(
    completed_urls_027
)


unknown_completed_urls_027 = (
    completed_url_set_027
    -
    master_urls_027
)


print(
    "Completed review rows:",
    len(STRUCTURAL_COMPLETED_REVIEWS_RAW_027)
)

print(
    "Unique completed review URLs:",
    len(completed_url_set_027)
)

print(
    "URLs absent from CXR master:",
    len(unknown_completed_urls_027)
)


if unknown_completed_urls_027:

    raise ValueError(
        "Completed structural-review records contain "
        "provenance URLs absent from the authoritative "
        "3,859-row CXR master."
    )


print(
    "Completed review provenance: PASS"
)


# ================================================================================================
# 10. DUPLICATE REVIEW URL AUDIT
# ================================================================================================

url_review_counts_027 = (
    STRUCTURAL_COMPLETED_REVIEWS_RAW_027
    .groupby(
        "series_instance_content_url"
    )
    .size()
)


duplicate_review_urls_027 = (
    url_review_counts_027[
        url_review_counts_027 > 1
    ]
)


print("\n6. DUPLICATE COMPLETED-REVIEW URL AUDIT")
print("-" * 100)

print(
    "URLs with multiple completed review rows:",
    len(duplicate_review_urls_027)
)


# ================================================================================================
# 11. CHECK WHETHER DUPLICATES ARE CONTRADICTORY
# ================================================================================================

contradictory_duplicate_urls_027 = []


if len(duplicate_review_urls_027) > 0:

    duplicate_urls_set_027 = set(
        duplicate_review_urls_027.index
    )


    duplicate_review_subset_027 = (
        STRUCTURAL_COMPLETED_REVIEWS_RAW_027[
            STRUCTURAL_COMPLETED_REVIEWS_RAW_027[
                "series_instance_content_url"
            ]
            .isin(
                duplicate_urls_set_027
            )
        ]
    )


    decision_sets_027 = (
        duplicate_review_subset_027
        .groupby(
            "series_instance_content_url"
        )[
            "human_review_decision"
        ]
        .agg(
            lambda values:
                sorted(
                    set(values)
                )
        )
    )


    contradictory_duplicate_urls_027 = (
        decision_sets_027[
            decision_sets_027.map(
                lambda values:
                    len(values) > 1
            )
        ]
        .index
        .tolist()
    )


print(
    "Contradictory duplicate URLs:",
    len(contradictory_duplicate_urls_027)
)


if contradictory_duplicate_urls_027:

    raise ValueError(
        "\nContradictory human-review decisions detected "
        "for the same provenance URL.\n"
        "Automatic resolution is prohibited.\n"
        f"Number of conflicting URLs: "
        f"{len(contradictory_duplicate_urls_027)}"
    )


# ================================================================================================
# 12. COLLAPSE IDENTICAL DUPLICATE RECORDS
# ================================================================================================

STRUCTURAL_COMPLETED_REVIEWS_UNIQUE_027 = (
    STRUCTURAL_COMPLETED_REVIEWS_RAW_027[
        [
            "series_instance_content_url",
            "human_review_decision",
            "review_artifact",
        ]
    ]
    .drop_duplicates()
    .copy()
)


# ================================================================================================
# 13. VERIFY ONE FINAL DECISION PER URL
# ================================================================================================

decision_count_per_url_027 = (
    STRUCTURAL_COMPLETED_REVIEWS_UNIQUE_027
    .groupby(
        "series_instance_content_url"
    )[
        "human_review_decision"
    ]
    .nunique()
)


ambiguous_decision_urls_027 = (
    decision_count_per_url_027[
        decision_count_per_url_027 > 1
    ]
)


if len(ambiguous_decision_urls_027) > 0:

    raise ValueError(
        "A provenance URL has multiple different final human decisions."
    )


# ================================================================================================
# 14. BUILD ONE-ROW-PER-IMAGE HUMAN REVIEW TABLE
# ================================================================================================

STRUCTURAL_HUMAN_REVIEW_027 = (
    STRUCTURAL_COMPLETED_REVIEWS_UNIQUE_027
    .groupby(
        [
            "series_instance_content_url",
            "human_review_decision",
        ],
        as_index=False
    )
    .agg(
        review_artifact=(
            "review_artifact",
            lambda values:
                "; ".join(
                    sorted(
                        set(values)
                    )
                )
        )
    )
)


# ================================================================================================
# 15. AUTHORITATIVE CONDITION_ID MAPPING
# ================================================================================================

master_condition_map_027 = (
    CXR_MULTIMODAL_MASTER_025[
        [
            "series_instance_content_url",
            "condition_id",
        ]
    ]
    .copy()
)


master_condition_map_027[
    "series_instance_content_url"
] = (
    master_condition_map_027[
        "series_instance_content_url"
    ]
    .astype(str)
    .str.strip()
)


master_condition_map_027[
    "condition_id"
] = (
    master_condition_map_027[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)


if (
    master_condition_map_027[
        "series_instance_content_url"
    ]
    .duplicated()
    .any()
):

    raise ValueError(
        "Authoritative CXR master provenance mapping "
        "is not one-to-one."
    )


STRUCTURAL_HUMAN_REVIEW_027 = (
    STRUCTURAL_HUMAN_REVIEW_027
    .merge(
        master_condition_map_027,
        on="series_instance_content_url",
        how="left",
        validate="one_to_one"
    )
)


# ================================================================================================
# 16. CONDITION_ID VALIDATION
# ================================================================================================

missing_condition_id_027 = int(
    STRUCTURAL_HUMAN_REVIEW_027[
        "condition_id"
    ]
    .isna()
    .sum()
)


if missing_condition_id_027 != 0:

    raise ValueError(
        "Human-reviewed structural records contain "
        "unmapped condition_id values."
    )


print("\n7. CONDITION_ID PROVENANCE")
print("-" * 100)

print(
    "Completed reviewed records:",
    len(STRUCTURAL_HUMAN_REVIEW_027)
)

print(
    "Unique reviewed conditions:",
    STRUCTURAL_HUMAN_REVIEW_027[
        "condition_id"
    ].nunique()
)

print(
    "Unmapped condition_id:",
    missing_condition_id_027
)


print(
    "Condition mapping: PASS"
)


# ================================================================================================
# 17. CHECK REVIEWED RECORDS AGAINST 639 CANDIDATES
# ================================================================================================

candidate_lookup_027 = dict(
    zip(
        technical_provenance_026[
            "series_instance_content_url"
        ]
        .astype(str)
        .str.strip(),

        technical_provenance_026[
            "structural_review_candidate"
        ]
        .astype(bool)
    )
)


STRUCTURAL_HUMAN_REVIEW_027[
    "was_structural_review_candidate"
] = (
    STRUCTURAL_HUMAN_REVIEW_027[
        "series_instance_content_url"
    ]
    .map(
        candidate_lookup_027
    )
)


missing_candidate_status_027 = int(
    STRUCTURAL_HUMAN_REVIEW_027[
        "was_structural_review_candidate"
    ]
    .isna()
    .sum()
)


if missing_candidate_status_027 != 0:

    raise ValueError(
        "Human-reviewed record missing from technical provenance."
    )


non_candidate_review_count_027 = int(
    (
        STRUCTURAL_HUMAN_REVIEW_027[
            "was_structural_review_candidate"
        ]
        ==
        False
    )
    .sum()
)


print("\n8. HUMAN REVIEW ↔ STRUCTURAL CANDIDATE")
print("-" * 100)

print(
    "Human-reviewed records:",
    len(STRUCTURAL_HUMAN_REVIEW_027)
)

print(
    "Reviewed records not flagged structurally:",
    non_candidate_review_count_027
)


if non_candidate_review_count_027 != 0:

    raise ValueError(
        "At least one human structural review record "
        "is not a structural-screening candidate."
    )


print(
    "Candidate membership: PASS"
)


# ================================================================================================
# 18. REVIEW COVERAGE
# ================================================================================================

candidate_urls_027 = set(
    technical_provenance_026.loc[
        technical_provenance_026[
            "structural_review_candidate"
        ],
        "series_instance_content_url"
    ]
)


reviewed_urls_final_027 = set(
    STRUCTURAL_HUMAN_REVIEW_027[
        "series_instance_content_url"
    ]
)


reviewed_candidate_urls_027 = (
    candidate_urls_027
    &
    reviewed_urls_final_027
)


unreviewed_candidate_urls_027 = (
    candidate_urls_027
    -
    reviewed_urls_final_027
)


print("\n9. STRUCTURAL HUMAN-REVIEW COVERAGE")
print("-" * 100)

print(
    "Structural candidates:",
    len(candidate_urls_027)
)

print(
    "Completed human-reviewed candidates:",
    len(reviewed_candidate_urls_027)
)

print(
    "Structural candidates without completed review:",
    len(unreviewed_candidate_urls_027)
)


# ================================================================================================
# 19. VALIDATE EXPECTED COMPLETED POPULATION
# ================================================================================================

EXPECTED_COMPLETED_STRUCTURAL_REVIEWS_027 = 142


if (
    len(STRUCTURAL_HUMAN_REVIEW_027)
    !=
    EXPECTED_COMPLETED_STRUCTURAL_REVIEWS_027
):

    raise ValueError(
        "\nCompleted structural-review population differs "
        "from the previously validated 142 records.\n"
        f"Expected: {EXPECTED_COMPLETED_STRUCTURAL_REVIEWS_027}\n"
        f"Observed: {len(STRUCTURAL_HUMAN_REVIEW_027)}\n\n"
        "Do NOT continue until this discrepancy is investigated."
    )


# ================================================================================================
# 20. DECISION DISTRIBUTION
# ================================================================================================

structural_decision_distribution_027 = (
    STRUCTURAL_HUMAN_REVIEW_027[
        "human_review_decision"
    ]
    .value_counts()
    .sort_index()
)


print("\n10. FINAL STRUCTURAL HUMAN-REVIEW DECISIONS")
print("-" * 100)

print(
    structural_decision_distribution_027.to_string()
)


EXPECTED_STRUCTURAL_DECISIONS_027 = {

    "EXCLUDE": 21,

    "RETAIN": 121,

}


if (
    structural_decision_distribution_027.to_dict()
    !=
    EXPECTED_STRUCTURAL_DECISIONS_027
):

    raise ValueError(
        "\nStructural human-review decision distribution "
        "does not match the previously validated population.\n\n"
        f"Expected:\n{EXPECTED_STRUCTURAL_DECISIONS_027}\n\n"
        f"Observed:\n{structural_decision_distribution_027.to_dict()}\n\n"
        "Do NOT continue until the discrepancy is investigated."
    )


# ================================================================================================
# 21. ADD STRUCTURAL DECISION STATUS TO ALL 3,859 RECORDS
# ================================================================================================

structural_decision_lookup_027 = dict(
    zip(
        STRUCTURAL_HUMAN_REVIEW_027[
            "series_instance_content_url"
        ],

        STRUCTURAL_HUMAN_REVIEW_027[
            "human_review_decision"
        ]
    )
)


technical_structural_decision_layer_027 = (
    technical_provenance_026
    .copy()
)


technical_structural_decision_layer_027[
    "structural_human_review_status"
] = (
    technical_structural_decision_layer_027[
        "series_instance_content_url"
    ]
    .astype(str)
    .str.strip()
    .map(
        lambda key:
            structural_decision_lookup_027.get(
                key,
                "NOT_REVIEWED"
            )
    )
)


# ================================================================================================
# 22. FULL DECISION-LAYER DISTRIBUTION
# ================================================================================================

full_structural_status_distribution_027 = (
    technical_structural_decision_layer_027[
        "structural_human_review_status"
    ]
    .value_counts()
    .sort_index()
)


print("\n11. FULL 3,859-RECORD STRUCTURAL DECISION LAYER")
print("-" * 100)

print(
    full_structural_status_distribution_027.to_string()
)


EXPECTED_FULL_STRUCTURAL_STATUS_027 = {

    "EXCLUDE": 21,

    "NOT_REVIEWED": 3717,

    "RETAIN": 121,

}


if (
    full_structural_status_distribution_027.to_dict()
    !=
    EXPECTED_FULL_STRUCTURAL_STATUS_027
):

    raise ValueError(
        "\nUnexpected full structural decision-layer distribution.\n\n"
        f"Expected:\n{EXPECTED_FULL_STRUCTURAL_STATUS_027}\n\n"
        f"Observed:\n{full_structural_status_distribution_027.to_dict()}"
    )


# ================================================================================================
# 23. SAVE RECONSTRUCTED HUMAN REVIEW
# ================================================================================================

STRUCTURAL_HUMAN_REVIEW_OUTPUT_027 = (
    NOTEBOOK_03_OUTPUT_DIR
    /
    "TB_Portals_March2025_Structural_Human_Review_Reconstructed_142.csv"
)


STRUCTURAL_HUMAN_REVIEW_027.to_csv(
    STRUCTURAL_HUMAN_REVIEW_OUTPUT_027,
    index=False
)


# ================================================================================================
# 24. SAVE FULL DECISION LAYER
# ================================================================================================

TECHNICAL_STRUCTURAL_DECISION_OUTPUT_027 = (
    NOTEBOOK_03_OUTPUT_DIR
    /
    "TB_Portals_March2025_3859_Technical_Structural_Decision_Layer.csv"
)


technical_structural_decision_layer_027.to_csv(
    TECHNICAL_STRUCTURAL_DECISION_OUTPUT_027,
    index=False
)


# ================================================================================================
# 25. SAVE REVIEW COVERAGE SUMMARY
# ================================================================================================

STRUCTURAL_REVIEW_COVERAGE_OUTPUT_027 = (
    NOTEBOOK_03_OUTPUT_DIR
    /
    "TB_Portals_March2025_Structural_Human_Review_Coverage.csv"
)


structural_review_coverage_summary_027 = pd.DataFrame({

    "metric": [

        "total_cxr_records",

        "structural_candidates",

        "completed_human_reviews",

        "human_retain",

        "human_exclude",

        "structural_candidates_without_review",

    ],

    "value": [

        3859,

        len(candidate_urls_027),

        len(STRUCTURAL_HUMAN_REVIEW_027),

        int(
            (
                STRUCTURAL_HUMAN_REVIEW_027[
                    "human_review_decision"
                ]
                ==
                "RETAIN"
            )
            .sum()
        ),

        int(
            (
                STRUCTURAL_HUMAN_REVIEW_027[
                    "human_review_decision"
                ]
                ==
                "EXCLUDE"
            )
            .sum()
        ),

        len(unreviewed_candidate_urls_027),

    ]

})


structural_review_coverage_summary_027.to_csv(
    STRUCTURAL_REVIEW_COVERAGE_OUTPUT_027,
    index=False
)


# ================================================================================================
# 26. FINAL FILE EXISTENCE CHECK
# ================================================================================================

for output_file_027 in [

    STRUCTURAL_HUMAN_REVIEW_OUTPUT_027,

    TECHNICAL_STRUCTURAL_DECISION_OUTPUT_027,

    STRUCTURAL_REVIEW_COVERAGE_OUTPUT_027,

]:

    if not output_file_027.is_file():

        raise IOError(
            f"Expected output was not created:\n{output_file_027}"
        )


# ================================================================================================
# 27. FINAL HARD ASSERTIONS
# ================================================================================================

assert len(
    STRUCTURAL_HUMAN_REVIEW_027
) == 142


assert (
    STRUCTURAL_HUMAN_REVIEW_027[
        "series_instance_content_url"
    ]
    .nunique()
    ==
    142
)


assert (
    STRUCTURAL_HUMAN_REVIEW_027[
        "human_review_decision"
    ]
    .isin(
        {
            "RETAIN",
            "EXCLUDE"
        }
    )
    .all()
)


assert (
    (
        STRUCTURAL_HUMAN_REVIEW_027[
            "human_review_decision"
        ]
        ==
        "RETAIN"
    )
    .sum()
    ==
    121
)


assert (
    (
        STRUCTURAL_HUMAN_REVIEW_027[
            "human_review_decision"
        ]
        ==
        "EXCLUDE"
    )
    .sum()
    ==
    21
)


assert (
    len(candidate_urls_027)
    ==
    639
)


assert (
    len(reviewed_candidate_urls_027)
    ==
    142
)


assert (
    len(unreviewed_candidate_urls_027)
    ==
    497
)


assert (
    technical_structural_decision_layer_027[
        "structural_human_review_status"
    ]
    .eq("RETAIN")
    .sum()
    ==
    121
)


assert (
    technical_structural_decision_layer_027[
        "structural_human_review_status"
    ]
    .eq("EXCLUDE")
    .sum()
    ==
    21
)


assert (
    technical_structural_decision_layer_027[
        "structural_human_review_status"
    ]
    .eq("NOT_REVIEWED")
    .sum()
    ==
    3717
)


assert len(
    technical_structural_decision_layer_027
) == 3859


# ================================================================================================
# 28. FINAL STATUS
# ================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.27 — COMPLETED SUCCESSFULLY")
print("=" * 100)

print()

print(
    "Total CXR records                  : 3,859"
)

print(
    "Structural screening candidates   : 639"
)

print(
    "Completed human structural reviews: 142"
)

print(
    "Human RETAIN decisions             : 121"
)

print(
    "Human EXCLUDE decisions            : 21"
)

print(
    "Candidates without completed review: 497"
)

print()

print(
    "Full structural decision layer:"
)

print(
    "  RETAIN       : 121"
)

print(
    "  EXCLUDE      : 21"
)

print(
    "  NOT_REVIEWED : 3,717"
)

print()

print(
    "Reconstructed review table:"
)

print(
    STRUCTURAL_HUMAN_REVIEW_OUTPUT_027
)

print()

print(
    "Full technical + structural layer:"
)

print(
    TECHNICAL_STRUCTURAL_DECISION_OUTPUT_027
)

print()

print(
    "Review coverage summary:"
)

print(
    STRUCTURAL_REVIEW_COVERAGE_OUTPUT_027
)

print()

print("NO AUTOMATIC EXCLUSION PERFORMED")
print("NO DICOM FILES MODIFIED")
print("NO RAW IMAGES MODIFIED")
print("NO PREPROCESSING PERFORMED")
print("NO TARGET SELECTED")
print("NO MODELING PERFORMED")

print()

print("STEP 3B.6.27 STATUS: PASS")

print("=" * 100)
```

```python
# ====================================================================================================
# STEP 3B.6.28-E — CORRECTED FINAL VALIDATION + OUTPUT
# ====================================================================================================
#
# IMPORTANT:
# The final evidence table has already been successfully constructed.
# This continuation corrects ONLY the expected semantic × structural
# evidence-category accounting.
#
# Correct validated accounting:
#
# BOTH_AGREE_EXCLUDE : 7
# BOTH_AGREE_RETAIN  : 5
# BOTH_DISAGREE      : 5
# SEMANTIC_ONLY      : 98
# STRUCTURAL_ONLY    : 125
# NEITHER_REVIEWED   : 3619
#
# Total = 3859
#
# Agreement = 7 + 5 = 12
# Disagreement = 5
# ====================================================================================================


print("=" * 100)
print("STEP 3B.6.28-E — CORRECTED FINAL VALIDATION + OUTPUT")
print("=" * 100)


# ----------------------------------------------------------------------------------------------------
# 1. VERIFY FINAL EVIDENCE TABLE EXISTS
# ----------------------------------------------------------------------------------------------------

print("\n1. FINAL EVIDENCE TABLE")
print("-" * 100)

if "FINAL_EVIDENCE_3859_028E" not in globals():
    raise RuntimeError(
        "FINAL_EVIDENCE_3859_028E is not available. "
        "The evidence table must be constructed before this validation cell."
    )

print(
    f"Records currently in evidence table: "
    f"{len(FINAL_EVIDENCE_3859_028E)}"
)


# ----------------------------------------------------------------------------------------------------
# 2. FINAL RECORD / URL / CONDITION VALIDATION
# ----------------------------------------------------------------------------------------------------

print("\n2. MASTER EVIDENCE VALIDATION")
print("-" * 100)


final_records_028E = len(
    FINAL_EVIDENCE_3859_028E
)

final_unique_urls_028E = (
    FINAL_EVIDENCE_3859_028E[
        "series_instance_content_url"
    ].nunique()
)

final_unique_conditions_028E = (
    FINAL_EVIDENCE_3859_028E[
        "condition_id"
    ].nunique()
)


print(
    f"Final records: {final_records_028E}"
)

print(
    f"Unique URLs: {final_unique_urls_028E}"
)

print(
    f"Unique conditions: {final_unique_conditions_028E}"
)


if final_records_028E != 3859:
    raise AssertionError(
        f"Expected 3,859 records; found {final_records_028E}."
    )


if final_unique_urls_028E != 3859:
    raise AssertionError(
        f"Expected 3,859 unique URLs; found {final_unique_urls_028E}."
    )


if final_unique_conditions_028E != 3081:
    raise AssertionError(
        f"Expected 3,081 unique conditions; found "
        f"{final_unique_conditions_028E}."
    )


print("Master evidence validation: PASS")


# ----------------------------------------------------------------------------------------------------
# 3. SEMANTIC × STRUCTURAL EVIDENCE CATEGORY
# ----------------------------------------------------------------------------------------------------

print("\n3. SEMANTIC × STRUCTURAL EVIDENCE CATEGORIES")
print("-" * 100)


evidence_category_summary_028E = (
    FINAL_EVIDENCE_3859_028E[
        "semantic_structural_evidence_category"
    ]
    .value_counts()
    .rename_axis("evidence_category")
    .reset_index(
        name="record_count"
    )
)


# Create a complete lookup so categories with zero records are also represented.
expected_evidence_categories_028E = {
    "BOTH_AGREE_EXCLUDE": 7,
    "BOTH_AGREE_RETAIN": 5,
    "BOTH_DISAGREE": 5,
    "SEMANTIC_ONLY": 98,
    "STRUCTURAL_ONLY": 125,
    "NEITHER_REVIEWED": 3619,
}


actual_evidence_counts_028E = (
    FINAL_EVIDENCE_3859_028E[
        "semantic_structural_evidence_category"
    ]
    .value_counts()
    .to_dict()
)


for category_028E, expected_count_028E in (
    expected_evidence_categories_028E.items()
):

    actual_count_028E = int(
        actual_evidence_counts_028E.get(
            category_028E,
            0
        )
    )

    print(
        f"{category_028E:<25} "
        f"Actual = {actual_count_028E:<5} "
        f"Expected = {expected_count_028E}"
    )

    if actual_count_028E != expected_count_028E:
        raise AssertionError(
            f"Evidence category mismatch for "
            f"{category_028E}: expected "
            f"{expected_count_028E}, found "
            f"{actual_count_028E}."
        )


# Check that there are no unexpected categories.
unexpected_evidence_categories_028E = (
    set(actual_evidence_counts_028E.keys())
    - set(expected_evidence_categories_028E.keys())
)


if unexpected_evidence_categories_028E:
    raise AssertionError(
        "Unexpected evidence categories found: "
        + str(sorted(unexpected_evidence_categories_028E))
    )


# Check total accounting.
evidence_total_028E = sum(
    expected_evidence_categories_028E.values()
)


if evidence_total_028E != 3859:
    raise AssertionError(
        f"Expected evidence-category total of 3,859; "
        f"calculated {evidence_total_028E}."
    )


print("\nEvidence category accounting: PASS")


# ----------------------------------------------------------------------------------------------------
# 4. DISPLAY EVIDENCE CATEGORY SUMMARY IN FIXED ORDER
# ----------------------------------------------------------------------------------------------------

evidence_category_summary_028E = pd.DataFrame(
    [
        {
            "evidence_category": category_028E,
            "record_count": expected_count_028E,
        }
        for category_028E, expected_count_028E
        in expected_evidence_categories_028E.items()
    ]
)


print("\nFinal evidence category summary:")
print(
    evidence_category_summary_028E.to_string(
        index=False
    )
)


# ----------------------------------------------------------------------------------------------------
# 5. SEMANTIC × STRUCTURAL DECISION COMPARISON
# ----------------------------------------------------------------------------------------------------

print("\n4. SEMANTIC × STRUCTURAL DECISION COMPARISON")
print("-" * 100)


decision_comparison_summary_028E = (
    FINAL_EVIDENCE_3859_028E[
        "semantic_structural_decision_comparison"
    ]
    .value_counts()
    .rename_axis("decision_comparison")
    .reset_index(
        name="record_count"
    )
)


expected_decision_comparison_028E = {
    "AGREEMENT": 12,
    "DISAGREEMENT": 5,
    "NOT_APPLICABLE": 3842,
}


actual_decision_counts_028E = (
    FINAL_EVIDENCE_3859_028E[
        "semantic_structural_decision_comparison"
    ]
    .value_counts()
    .to_dict()
)


for comparison_028E, expected_count_028E in (
    expected_decision_comparison_028E.items()
):

    actual_count_028E = int(
        actual_decision_counts_028E.get(
            comparison_028E,
            0
        )
    )

    print(
        f"{comparison_028E:<20} "
        f"Actual = {actual_count_028E:<5} "
        f"Expected = {expected_count_028E}"
    )

    if actual_count_028E != expected_count_028E:
        raise AssertionError(
            f"Decision comparison mismatch for "
            f"{comparison_028E}: expected "
            f"{expected_count_028E}, found "
            f"{actual_count_028E}."
        )


if sum(expected_decision_comparison_028E.values()) != 3859:
    raise AssertionError(
        "Decision comparison accounting does not total 3,859."
    )


print("\nDecision comparison accounting: PASS")


# ----------------------------------------------------------------------------------------------------
# 6. DISPLAY DECISION COMPARISON IN FIXED ORDER
# ----------------------------------------------------------------------------------------------------

decision_comparison_summary_028E = pd.DataFrame(
    [
        {
            "decision_comparison": comparison_028E,
            "record_count": expected_count_028E,
        }
        for comparison_028E, expected_count_028E
        in expected_decision_comparison_028E.items()
    ]
)


print("\nFinal decision comparison summary:")
print(
    decision_comparison_summary_028E.to_string(
        index=False
    )
)


# ----------------------------------------------------------------------------------------------------
# 7. EXPLICIT OVERLAP ACCOUNTING
# ----------------------------------------------------------------------------------------------------

print("\n5. STRUCTURAL OVERLAP ACCOUNTING")
print("-" * 100)


both_reviewed_028E = (
    FINAL_EVIDENCE_3859_028E[
        "semantic_reviewed"
    ]
    &
    FINAL_EVIDENCE_3859_028E[
        "structural_reviewed"
    ]
)


semantic_only_028E = (
    FINAL_EVIDENCE_3859_028E[
        "semantic_reviewed"
    ]
    &
    ~FINAL_EVIDENCE_3859_028E[
        "structural_reviewed"
    ]
)


structural_only_028E = (
    ~FINAL_EVIDENCE_3859_028E[
        "semantic_reviewed"
    ]
    &
    FINAL_EVIDENCE_3859_028E[
        "structural_reviewed"
    ]
)


neither_reviewed_028E = (
    ~FINAL_EVIDENCE_3859_028E[
        "semantic_reviewed"
    ]
    &
    ~FINAL_EVIDENCE_3859_028E[
        "structural_reviewed"
    ]
)


both_reviewed_count_028E = int(
    both_reviewed_028E.sum()
)

semantic_only_count_028E = int(
    semantic_only_028E.sum()
)

structural_only_count_028E = int(
    structural_only_028E.sum()
)

neither_reviewed_count_028E = int(
    neither_reviewed_028E.sum()
)


print(
    f"Both semantic + structural reviewed : "
    f"{both_reviewed_count_028E}"
)

print(
    f"Semantic only                       : "
    f"{semantic_only_count_028E}"
)

print(
    f"Structural only                     : "
    f"{structural_only_count_028E}"
)

print(
    f"Neither reviewed                    : "
    f"{neither_reviewed_count_028E}"
)


if both_reviewed_count_028E != 17:
    raise AssertionError(
        f"Expected 17 records with both reviews; "
        f"found {both_reviewed_count_028E}."
    )


if semantic_only_count_028E != 98:
    raise AssertionError(
        f"Expected 98 semantic-only records; "
        f"found {semantic_only_count_028E}."
    )


if structural_only_count_028E != 125:
    raise AssertionError(
        f"Expected 125 structural-only records; "
        f"found {structural_only_count_028E}."
    )


if neither_reviewed_count_028E != 3619:
    raise AssertionError(
        f"Expected 3,619 records with neither review; "
        f"found {neither_reviewed_count_028E}."
    )


print("Structural overlap accounting: PASS")


# ----------------------------------------------------------------------------------------------------
# 8. AGREEMENT / DISAGREEMENT ACCOUNTING
# ----------------------------------------------------------------------------------------------------

print("\n6. AGREEMENT / DISAGREEMENT ACCOUNTING")
print("-" * 100)


agreement_count_028E = int(
    (
        FINAL_EVIDENCE_3859_028E[
            "semantic_structural_decision_comparison"
        ]
        == "AGREEMENT"
    ).sum()
)


disagreement_count_028E = int(
    (
        FINAL_EVIDENCE_3859_028E[
            "semantic_structural_decision_comparison"
        ]
        == "DISAGREEMENT"
    ).sum()
)


print(
    f"Agreement     : {agreement_count_028E}"
)

print(
    f"Disagreement  : {disagreement_count_028E}"
)


if agreement_count_028E != 12:
    raise AssertionError(
        f"Expected 12 agreements; "
        f"found {agreement_count_028E}."
    )


if disagreement_count_028E != 5:
    raise AssertionError(
        f"Expected 5 disagreements; "
        f"found {disagreement_count_028E}."
    )


if (
    agreement_count_028E
    + disagreement_count_028E
    != both_reviewed_count_028E
):
    raise AssertionError(
        "Agreement + disagreement does not equal "
        "the number of records reviewed by both methods."
    )


print("Agreement/disagreement accounting: PASS")


# ----------------------------------------------------------------------------------------------------
# 9. FIVE RAW DISAGREEMENTS
# ----------------------------------------------------------------------------------------------------

print("\n7. FIVE RAW SEMANTIC ↔ STRUCTURAL DISAGREEMENTS")
print("-" * 100)


raw_disagreements_028E = (
    FINAL_EVIDENCE_3859_028E[
        FINAL_EVIDENCE_3859_028E[
            "semantic_structural_decision_comparison"
        ] == "DISAGREEMENT"
    ]
    .copy()
)


if len(raw_disagreements_028E) != 5:
    raise AssertionError(
        f"Expected exactly 5 raw disagreements; "
        f"found {len(raw_disagreements_028E)}."
    )


disagreement_columns_028E = [
    "condition_id",
    "series_instance_content_url",
    "structural_decision",
    "human_semantic_decision",
    "human_semantic_reason",
    "final_resolution_status",
]


disagreement_columns_028E = [
    c
    for c in disagreement_columns_028E
    if c in raw_disagreements_028E.columns
]


print(
    raw_disagreements_028E[
        disagreement_columns_028E
    ].to_string(index=False)
)


print("\nRaw disagreement accounting: PASS")


# ----------------------------------------------------------------------------------------------------
# 10. VERIFY SEMANTIC FINAL STATE
# ----------------------------------------------------------------------------------------------------

print("\n8. FINAL SEMANTIC STATE")
print("-" * 100)


semantic_final_counts_028E = (
    FINAL_EVIDENCE_3859_028E[
        "human_semantic_decision"
    ]
    .fillna("NOT_REVIEWED")
    .value_counts()
)


print(
    semantic_final_counts_028E.to_string()
)


expected_semantic_final_028E = {
    "RETAIN": 95,
    "EXCLUDE": 20,
    "NOT_REVIEWED": 3744,
}


for decision_028E, expected_count_028E in (
    expected_semantic_final_028E.items()
):

    actual_count_028E = int(
        semantic_final_counts_028E.get(
            decision_028E,
            0
        )
    )

    if actual_count_028E != expected_count_028E:
        raise AssertionError(
            f"Semantic final-state mismatch for "
            f"{decision_028E}: expected "
            f"{expected_count_028E}, found "
            f"{actual_count_028E}."
        )


print("Final semantic state: PASS")


# ----------------------------------------------------------------------------------------------------
# 11. VERIFY STRUCTURAL FINAL STATE
# ----------------------------------------------------------------------------------------------------

print("\n9. FINAL STRUCTURAL REVIEW STATE")
print("-" * 100)


structural_final_counts_028E = (
    FINAL_EVIDENCE_3859_028E[
        "structural_decision"
    ]
    .fillna("NOT_REVIEWED")
    .value_counts()
)


print(
    structural_final_counts_028E.to_string()
)


expected_structural_final_028E = {
    "RETAIN": 121,
    "EXCLUDE": 21,
    "NOT_REVIEWED": 3717,
}


for decision_028E, expected_count_028E in (
    expected_structural_final_028E.items()
):

    actual_count_028E = int(
        structural_final_counts_028E.get(
            decision_028E,
            0
        )
    )

    if actual_count_028E != expected_count_028E:
        raise AssertionError(
            f"Structural final-state mismatch for "
            f"{decision_028E}: expected "
            f"{expected_count_028E}, found "
            f"{actual_count_028E}."
        )


print("Final structural review state: PASS")


# ----------------------------------------------------------------------------------------------------
# 12. VERIFY ELIGIBILITY HAS NOT BEEN PREMATURELY DECIDED
# ----------------------------------------------------------------------------------------------------

print("\n10. METHODOLOGICAL LOCK")
print("-" * 100)


if "cxr_eligibility_decision" in FINAL_EVIDENCE_3859_028E.columns:

    eligibility_states_028E = (
        FINAL_EVIDENCE_3859_028E[
            "cxr_eligibility_decision"
        ]
        .fillna("")
        .astype(str)
        .unique()
    )

    print(
        "CXR eligibility states currently present:"
    )

    for state_028E in eligibility_states_028E:
        print(f"  - {state_028E}")

    if not all(
        state_028E == "NOT_YET_ADJUDICATED"
        for state_028E in eligibility_states_028E
    ):
        raise AssertionError(
            "CXR eligibility was prematurely adjudicated "
            "in Step 3B.6.28-E."
        )

    print(
        "Eligibility remains NOT_YET_ADJUDICATED: PASS"
    )

else:

    print(
        "Eligibility column not present; "
        "no eligibility decision was made in this step."
    )


# ----------------------------------------------------------------------------------------------------
# 13. OUTPUT DIRECTORY
# ----------------------------------------------------------------------------------------------------

print("\n11. SAVING OUTPUTS")
print("-" * 100)


OUTPUT_DIR_028E = (
    TBP_ROOT
    / "Metadata"
    / "Step_3B_6_24_Notebook_03"
)


OUTPUT_DIR_028E.mkdir(
    parents=True,
    exist_ok=True
)


# ----------------------------------------------------------------------------------------------------
# 14. SAVE FINAL EVIDENCE TABLE
# ----------------------------------------------------------------------------------------------------

FINAL_EVIDENCE_PATH_028E = (
    OUTPUT_DIR_028E
    / "TB_Portals_March2025_Step_3B_6_28E_Final_3859_Semantic_Structural_Evidence.csv"
)


FINAL_EVIDENCE_3859_028E.to_csv(
    FINAL_EVIDENCE_PATH_028E,
    index=False
)


# ----------------------------------------------------------------------------------------------------
# 15. SAVE EVIDENCE CATEGORY SUMMARY
# ----------------------------------------------------------------------------------------------------

EVIDENCE_SUMMARY_PATH_028E = (
    OUTPUT_DIR_028E
    / "TB_Portals_March2025_Step_3B_6_28E_Evidence_Category_Summary.csv"
)


evidence_category_summary_028E.to_csv(
    EVIDENCE_SUMMARY_PATH_028E,
    index=False
)


# ----------------------------------------------------------------------------------------------------
# 16. SAVE DECISION COMPARISON SUMMARY
# ----------------------------------------------------------------------------------------------------

DECISION_SUMMARY_PATH_028E = (
    OUTPUT_DIR_028E
    / "TB_Portals_March2025_Step_3B_6_28E_Decision_Comparison_Summary.csv"
)


decision_comparison_summary_028E.to_csv(
    DECISION_SUMMARY_PATH_028E,
    index=False
)


# ----------------------------------------------------------------------------------------------------
# 17. SAVE FIVE RAW DISAGREEMENTS
# ----------------------------------------------------------------------------------------------------

DISAGREEMENT_PATH_028E = (
    OUTPUT_DIR_028E
    / "TB_Portals_March2025_Step_3B_6_28E_Five_Raw_Disagreements.csv"
)


raw_disagreements_028E.to_csv(
    DISAGREEMENT_PATH_028E,
    index=False
)


# ----------------------------------------------------------------------------------------------------
# 18. FINAL OUTPUT FILE VALIDATION
# ----------------------------------------------------------------------------------------------------

print("\n12. OUTPUT FILE VALIDATION")
print("-" * 100)


output_files_028E = [
    FINAL_EVIDENCE_PATH_028E,
    EVIDENCE_SUMMARY_PATH_028E,
    DECISION_SUMMARY_PATH_028E,
    DISAGREEMENT_PATH_028E,
]


for output_path_028E in output_files_028E:

    if not output_path_028E.exists():
        raise AssertionError(
            f"Expected output file was not created: "
            f"{output_path_028E}"
        )

    if output_path_028E.stat().st_size == 0:
        raise AssertionError(
            f"Output file is empty: "
            f"{output_path_028E}"
        )

    print(
        f"{output_path_028E.name}: PASS"
    )


# ----------------------------------------------------------------------------------------------------
# 19. FINAL STEP PASS
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("STEP 3B.6.28-E — FINAL STATUS")
print("=" * 100)

print("3,859-record evidence table             : PASS")
print("3,859 unique CXR URLs                  : PASS")
print("3,081 unique conditions                : PASS")
print("Semantic final state                   : PASS")
print("Structural final state                 : PASS")
print("Semantic ↔ structural overlap          : 17")
print("Semantic ↔ structural agreement        : 12")
print("Semantic ↔ structural disagreement     : 5")
print("Unresolved disagreement                : 0")
print("Eligibility prematurely adjudicated    : NO")
print("Raw DICOM modified                     : NO")
print("Source evidence artifacts modified     : NO")
print("Output files                           : PASS")

print("\n" + "=" * 100)
print("STEP 3B.6.28-E COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ====================================================================================================
# STEP 3B.6.29
# CXR ELIGIBILITY EVIDENCE RECONCILIATION
# ====================================================================================================
#
# PURPOSE
# -------
# Create an evidence-based CXR eligibility status for the 3,859-record master.
#
# IMPORTANT:
# This step is NOT the final research cohort selection.
#
# It reconciles the evidence already established in:
#
#   1. Final 3,859-record CXR master
#   2. Structural human review
#   3. Final Notebook 02 semantic review
#
# METHODOLOGICAL LOCKS
# --------------------
# - No DICOM files are read or modified.
# - No image preprocessing.
# - No model training.
# - No target selection.
# - No random image-level splitting.
# - cxr_outlier is NOT an automatic exclusion rule.
# - Modality is NOT an automatic exclusion rule.
# - Unreviewed records are NOT automatically excluded.
# - Conflicting reviewed evidence is explicitly preserved.
#
# STATUS LOGIC
# ------------
#
# 1. Semantic reviewed + EXCLUDE
#       -> SEMANTIC_EXCLUDE
#
# 2. Semantic reviewed + RETAIN
#    AND structural reviewed + EXCLUDE
#       -> REVIEW_REQUIRED_CONFLICT
#
# 3. Semantic reviewed + RETAIN
#    AND structural reviewed + RETAIN
#       -> EVIDENCE_RETAIN
#
# 4. Semantic reviewed + RETAIN
#    AND structural NOT_REVIEWED
#       -> SEMANTIC_RETAIN_ONLY
#
# 5. Semantic NOT_REVIEWED
#    AND structural reviewed + EXCLUDE
#       -> STRUCTURAL_EXCLUDE
#
# 6. Semantic NOT_REVIEWED
#    AND structural reviewed + RETAIN
#       -> STRUCTURAL_RETAIN_ONLY
#
# 7. Neither reviewed
#       -> NOT_REVIEWED
#
# The key point:
# We do NOT turn "not reviewed" into "exclude".
#
# ====================================================================================================


from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 100)
print("STEP 3B.6.29 — CXR ELIGIBILITY EVIDENCE RECONCILIATION")
print("=" * 100)


# ====================================================================================================
# 1. REQUIRED OBJECT
# ====================================================================================================

print("\n1. REQUIRED OBJECT")
print("-" * 100)


if "FINAL_EVIDENCE_3859_028E" not in globals():
    raise RuntimeError(
        "FINAL_EVIDENCE_3859_028E is not available. "
        "Run the successfully completed Step 3B.6.28-E first."
    )


print(
    "FINAL_EVIDENCE_3859_028E available: PASS"
)


# ====================================================================================================
# 2. COPY INPUT
# ====================================================================================================

ELIGIBILITY_3859_029 = (
    FINAL_EVIDENCE_3859_028E.copy()
)


# ====================================================================================================
# 3. REQUIRED COLUMN VALIDATION
# ====================================================================================================

print("\n2. REQUIRED COLUMN VALIDATION")
print("-" * 100)


required_columns_029 = {
    "condition_id",
    "series_instance_content_url",
    "human_semantic_decision",
    "semantic_reviewed",
    "structural_decision",
    "structural_reviewed",
    "semantic_structural_evidence_category",
    "semantic_structural_decision_comparison",
}


missing_columns_029 = (
    required_columns_029
    - set(ELIGIBILITY_3859_029.columns)
)


if missing_columns_029:
    raise RuntimeError(
        "Step 3B.6.28-E is missing required columns: "
        + str(sorted(missing_columns_029))
    )


print("Required columns: PASS")


# ====================================================================================================
# 4. MASTER POPULATION VALIDATION
# ====================================================================================================

print("\n3. MASTER POPULATION")
print("-" * 100)


n_records_029 = len(
    ELIGIBILITY_3859_029
)

n_urls_029 = (
    ELIGIBILITY_3859_029[
        "series_instance_content_url"
    ].nunique()
)

n_conditions_029 = (
    ELIGIBILITY_3859_029[
        "condition_id"
    ].nunique()
)


print(f"Records: {n_records_029}")
print(f"Unique URLs: {n_urls_029}")
print(f"Unique conditions: {n_conditions_029}")


if n_records_029 != 3859:
    raise AssertionError(
        f"Expected 3,859 records; found {n_records_029}."
    )


if n_urls_029 != 3859:
    raise AssertionError(
        f"Expected 3,859 unique URLs; found {n_urls_029}."
    )


if n_conditions_029 != 3081:
    raise AssertionError(
        f"Expected 3,081 unique conditions; found {n_conditions_029}."
    )


print("Master population validation: PASS")


# ====================================================================================================
# 5. NORMALIZE REVIEW FIELDS
# ====================================================================================================

print("\n4. NORMALIZING REVIEW FIELDS")
print("-" * 100)


ELIGIBILITY_3859_029[
    "human_semantic_decision"
] = (
    ELIGIBILITY_3859_029[
        "human_semantic_decision"
    ]
    .fillna("NOT_REVIEWED")
    .astype(str)
    .str.strip()
    .str.upper()
)


ELIGIBILITY_3859_029[
    "structural_decision"
] = (
    ELIGIBILITY_3859_029[
        "structural_decision"
    ]
    .fillna("NOT_REVIEWED")
    .astype(str)
    .str.strip()
    .str.upper()
)


ELIGIBILITY_3859_029[
    "semantic_reviewed"
] = (
    ELIGIBILITY_3859_029[
        "semantic_reviewed"
    ]
    .astype(bool)
)


ELIGIBILITY_3859_029[
    "structural_reviewed"
] = (
    ELIGIBILITY_3859_029[
        "structural_reviewed"
    ]
    .astype(bool)
)


allowed_decisions_029 = {
    "RETAIN",
    "EXCLUDE",
    "NOT_REVIEWED",
}


unexpected_semantic_029 = (
    set(
        ELIGIBILITY_3859_029[
            "human_semantic_decision"
        ].unique()
    )
    - allowed_decisions_029
)


unexpected_structural_029 = (
    set(
        ELIGIBILITY_3859_029[
            "structural_decision"
        ].unique()
    )
    - allowed_decisions_029
)


if unexpected_semantic_029:
    raise AssertionError(
        "Unexpected semantic decisions: "
        + str(sorted(unexpected_semantic_029))
    )


if unexpected_structural_029:
    raise AssertionError(
        "Unexpected structural decisions: "
        + str(sorted(unexpected_structural_029))
    )


print("Review-field normalization: PASS")


# ====================================================================================================
# 6. VALIDATE REVIEWED FLAGS AGAINST DECISIONS
# ====================================================================================================

print("\n5. REVIEWED-FLAG CONSISTENCY")
print("-" * 100)


semantic_flag_expected_029 = (
    ELIGIBILITY_3859_029[
        "human_semantic_decision"
    ].isin(
        {
            "RETAIN",
            "EXCLUDE",
        }
    )
)


structural_flag_expected_029 = (
    ELIGIBILITY_3859_029[
        "structural_decision"
    ].isin(
        {
            "RETAIN",
            "EXCLUDE",
        }
    )
)


semantic_flag_mismatch_029 = int(
    (
        ELIGIBILITY_3859_029[
            "semantic_reviewed"
        ]
        !=
        semantic_flag_expected_029
    ).sum()
)


structural_flag_mismatch_029 = int(
    (
        ELIGIBILITY_3859_029[
            "structural_reviewed"
        ]
        !=
        structural_flag_expected_029
    ).sum()
)


print(
    f"Semantic reviewed-flag mismatches: "
    f"{semantic_flag_mismatch_029}"
)

print(
    f"Structural reviewed-flag mismatches: "
    f"{structural_flag_mismatch_029}"
)


if semantic_flag_mismatch_029 != 0:
    raise AssertionError(
        "Semantic reviewed flag is inconsistent with semantic decision."
    )


if structural_flag_mismatch_029 != 0:
    raise AssertionError(
        "Structural reviewed flag is inconsistent with structural decision."
    )


print("Reviewed-flag consistency: PASS")


# ====================================================================================================
# 7. RECONCILE SEMANTIC + STRUCTURAL EVIDENCE
# ====================================================================================================

print("\n6. EVIDENCE-BASED CXR ELIGIBILITY RECONCILIATION")
print("-" * 100)


def determine_cxr_evidence_status_029(row):

    semantic_reviewed = bool(
        row["semantic_reviewed"]
    )

    structural_reviewed = bool(
        row["structural_reviewed"]
    )

    semantic_decision = (
        row["human_semantic_decision"]
    )

    structural_decision = (
        row["structural_decision"]
    )


    # ----------------------------------------------------------------------------------------------
    # BOTH REVIEWED
    # ----------------------------------------------------------------------------------------------

    if semantic_reviewed and structural_reviewed:

        # Both explicitly exclude.
        if (
            semantic_decision == "EXCLUDE"
            and
            structural_decision == "EXCLUDE"
        ):
            return "EVIDENCE_EXCLUDE"


        # Both explicitly retain.
        if (
            semantic_decision == "RETAIN"
            and
            structural_decision == "RETAIN"
        ):
            return "EVIDENCE_RETAIN"


        # One retains and one excludes.
        return "REVIEW_REQUIRED_CONFLICT"


    # ----------------------------------------------------------------------------------------------
    # SEMANTIC ONLY
    # ----------------------------------------------------------------------------------------------

    if semantic_reviewed and not structural_reviewed:

        if semantic_decision == "EXCLUDE":
            return "SEMANTIC_EXCLUDE"

        if semantic_decision == "RETAIN":
            return "SEMANTIC_RETAIN_ONLY"


    # ----------------------------------------------------------------------------------------------
    # STRUCTURAL ONLY
    # ----------------------------------------------------------------------------------------------

    if structural_reviewed and not semantic_reviewed:

        if structural_decision == "EXCLUDE":
            return "STRUCTURAL_EXCLUDE"

        if structural_decision == "RETAIN":
            return "STRUCTURAL_RETAIN_ONLY"


    # ----------------------------------------------------------------------------------------------
    # NEITHER REVIEWED
    # ----------------------------------------------------------------------------------------------

    return "NOT_REVIEWED"


ELIGIBILITY_3859_029[
    "cxr_evidence_status"
] = (
    ELIGIBILITY_3859_029.apply(
        determine_cxr_evidence_status_029,
        axis=1,
    )
)


# ====================================================================================================
# 8. EVIDENCE STATUS SUMMARY
# ====================================================================================================

print("\n7. CXR EVIDENCE STATUS SUMMARY")
print("-" * 100)


evidence_status_summary_029 = (
    ELIGIBILITY_3859_029[
        "cxr_evidence_status"
    ]
    .value_counts()
    .rename_axis(
        "cxr_evidence_status"
    )
    .reset_index(
        name="record_count"
    )
)


print(
    evidence_status_summary_029.to_string(
        index=False
    )
)


# Expected accounting:
#
# EVIDENCE_EXCLUDE          7
# EVIDENCE_RETAIN           5
# REVIEW_REQUIRED_CONFLICT  5
# SEMANTIC_EXCLUDE         13
# SEMANTIC_RETAIN_ONLY     82
# STRUCTURAL_EXCLUDE       14
# STRUCTURAL_RETAIN_ONLY  111
# NOT_REVIEWED           3619
#
# Total = 3859
#
# Note:
# Semantic-only = 98
#   -> 20 semantic EXCLUDE total - 7 overlap EXCLUDE = 13
#   -> 95 semantic RETAIN total - 5 overlap RETAIN = 90
#
# However, among semantic-only records, all 98 are based on the final
# semantic review population and should therefore be checked directly
# rather than hard-coded.
#
# Structural-only = 125
#   -> structural reviewed total = 142
#   -> 17 overlap
#   -> 125 structural-only
#
# Therefore, we validate the totals below directly.


# ====================================================================================================
# 9. TOTAL STATUS ACCOUNTING
# ====================================================================================================

print("\n8. STATUS ACCOUNTING")
print("-" * 100)


status_counts_029 = (
    ELIGIBILITY_3859_029[
        "cxr_evidence_status"
    ]
    .value_counts()
    .to_dict()
)


status_total_029 = sum(
    status_counts_029.values()
)


print(
    f"Total status records: {status_total_029}"
)


if status_total_029 != 3859:
    raise AssertionError(
        "CXR evidence status categories do not account for all 3,859 records."
    )


# ====================================================================================================
# 10. CROSS-CHECK AGAINST NOTEBOOK 02 / STRUCTURAL ACCOUNTING
# ====================================================================================================

print("\n9. CROSS-CHECK AGAINST ESTABLISHED EVIDENCE")
print("-" * 100)


# Both reviewed
both_reviewed_actual_029 = int(
    (
        ELIGIBILITY_3859_029[
            "semantic_reviewed"
        ]
        &
        ELIGIBILITY_3859_029[
            "structural_reviewed"
        ]
    ).sum()
)


# Semantic only
semantic_only_actual_029 = int(
    (
        ELIGIBILITY_3859_029[
            "semantic_reviewed"
        ]
        &
        ~ELIGIBILITY_3859_029[
            "structural_reviewed"
        ]
    ).sum()
)


# Structural only
structural_only_actual_029 = int(
    (
        ~ELIGIBILITY_3859_029[
            "semantic_reviewed"
        ]
        &
        ELIGIBILITY_3859_029[
            "structural_reviewed"
        ]
    ).sum()
)


# Neither
neither_actual_029 = int(
    (
        ~ELIGIBILITY_3859_029[
            "semantic_reviewed"
        ]
        &
        ~ELIGIBILITY_3859_029[
            "structural_reviewed"
        ]
    ).sum()
)


print(
    f"Both reviewed       : {both_reviewed_actual_029}"
)

print(
    f"Semantic only       : {semantic_only_actual_029}"
)

print(
    f"Structural only     : {structural_only_actual_029}"
)

print(
    f"Neither reviewed    : {neither_actual_029}"
)


expected_review_structure_029 = {
    "both": 17,
    "semantic_only": 98,
    "structural_only": 125,
    "neither": 3619,
}


if both_reviewed_actual_029 != expected_review_structure_029["both"]:
    raise AssertionError(
        "Both-reviewed population mismatch."
    )


if semantic_only_actual_029 != expected_review_structure_029["semantic_only"]:
    raise AssertionError(
        "Semantic-only population mismatch."
    )


if structural_only_actual_029 != expected_review_structure_029["structural_only"]:
    raise AssertionError(
        "Structural-only population mismatch."
    )


if neither_actual_029 != expected_review_structure_029["neither"]:
    raise AssertionError(
        "Neither-reviewed population mismatch."
    )


print("Established review-structure cross-check: PASS")


# ====================================================================================================
# 11. CHECK CONFLICT POPULATION
# ====================================================================================================

print("\n10. CONFLICT POPULATION")
print("-" * 100)


conflict_mask_029 = (
    ELIGIBILITY_3859_029[
        "cxr_evidence_status"
    ]
    ==
    "REVIEW_REQUIRED_CONFLICT"
)


conflict_records_029 = (
    ELIGIBILITY_3859_029[
        conflict_mask_029
    ].copy()
)


print(
    f"Current evidence conflicts: "
    f"{len(conflict_records_029)}"
)


if len(conflict_records_029) != 5:
    raise AssertionError(
        f"Expected 5 semantic ↔ structural conflicts; "
        f"found {len(conflict_records_029)}."
    )


print("Conflict population: PASS")


# ====================================================================================================
# 12. CHECK CONFLICTS HAVE EXISTING FINAL SEMANTIC RESOLUTION
# ====================================================================================================

print("\n11. CONFLICT RESOLUTION PROVENANCE")
print("-" * 100)


if "final_resolution_status" in conflict_records_029.columns:

    missing_resolution_029 = (
        conflict_records_029[
            "final_resolution_status"
        ]
        .isna()
        .sum()
    )

    print(
        f"Conflicts missing final semantic resolution: "
        f"{int(missing_resolution_029)}"
    )

    if missing_resolution_029 != 0:
        raise AssertionError(
            "A semantic ↔ structural conflict is missing "
            "final semantic resolution provenance."
        )


print(
    "Conflict resolution provenance: PASS"
)


# ====================================================================================================
# 13. CRITICAL SAFETY CHECK:
#     NO AUTOMATIC EXCLUSION FROM cxr_outlier
# ====================================================================================================

print("\n12. cxr_outlier SAFETY CHECK")
print("-" * 100)


if "cxr_outlier" in ELIGIBILITY_3859_029.columns:

    outlier_values_029 = (
        ELIGIBILITY_3859_029[
            "cxr_outlier"
        ]
        .fillna("MISSING")
        .astype(str)
        .str.strip()
        .str.lower()
    )

    print(
        outlier_values_029.value_counts().to_string()
    )


    # Explicitly check that cxr_outlier itself was not used as the
    # evidence-status decision.
    #
    # We do NOT assert any particular relationship between outlier status
    # and eligibility because Notebook 02 established that modality/outlier
    # status was not an exclusion rule.


print(
    "cxr_outlier automatic exclusion: NOT USED"
)


# ====================================================================================================
# 14. CRITICAL SAFETY CHECK:
#     NO AUTOMATIC MODALITY EXCLUSION
# ====================================================================================================

print("\n13. MODALITY SAFETY CHECK")
print("-" * 100)


if "series_modality_cd" in ELIGIBILITY_3859_029.columns:

    modality_counts_029 = (
        ELIGIBILITY_3859_029[
            "series_modality_cd"
        ]
        .fillna("MISSING")
        .astype(str)
        .str.strip()
        .str.upper()
        .value_counts()
    )

    print(
        modality_counts_029.to_string()
    )


print(
    "Automatic modality exclusion: NOT USED"
)


# ====================================================================================================
# 15. DO NOT CREATE FINAL MODELING COHORT
# ====================================================================================================

print("\n14. FINAL COHORT LOCK")
print("-" * 100)


ELIGIBILITY_3859_029[
    "final_research_cohort_status"
] = "NOT_YET_SELECTED"


ELIGIBILITY_3859_029[
    "final_research_cohort_reason"
] = (
    "CXR evidence reconciliation completed; "
    "case-level, temporal, duplicate-CXR, and leakage analysis "
    "must be completed before final cohort selection."
)


print(
    "Final research cohort selection: NOT_YET_SELECTED"
)


# ====================================================================================================
# 16. FINAL ELIGIBILITY STATUS
# ====================================================================================================
#
# This is deliberately an EVIDENCE status rather than a final cohort
# inclusion decision.
#
# ====================================================================================================

print("\n15. FINAL CXR EVIDENCE STATUS")
print("-" * 100)


final_evidence_status_counts_029 = (
    ELIGIBILITY_3859_029[
        "cxr_evidence_status"
    ]
    .value_counts()
    .rename_axis(
        "cxr_evidence_status"
    )
    .reset_index(
        name="record_count"
    )
)


print(
    final_evidence_status_counts_029.to_string(
        index=False
    )
)


# ====================================================================================================
# 17. SAVE OUTPUTS
# ====================================================================================================

print("\n16. SAVING OUTPUTS")
print("-" * 100)


OUTPUT_DIR_029 = (
    TBP_ROOT
    / "Metadata"
    / "Step_3B_6_24_Notebook_03"
)


OUTPUT_DIR_029.mkdir(
    parents=True,
    exist_ok=True
)


# ---- Main evidence reconciliation -------------------------------------------

ELIGIBILITY_EVIDENCE_PATH_029 = (
    OUTPUT_DIR_029
    / "TB_Portals_March2025_Step_3B_6_29_CXR_Eligibility_Evidence_Reconciliation.csv"
)


ELIGIBILITY_3859_029.to_csv(
    ELIGIBILITY_EVIDENCE_PATH_029,
    index=False
)


# ---- Evidence status summary ------------------------------------------------

EVIDENCE_STATUS_PATH_029 = (
    OUTPUT_DIR_029
    / "TB_Portals_March2025_Step_3B_6_29_CXR_Evidence_Status_Summary.csv"
)


final_evidence_status_counts_029.to_csv(
    EVIDENCE_STATUS_PATH_029,
    index=False
)


# ---- Conflict audit ---------------------------------------------------------

CONFLICT_AUDIT_PATH_029 = (
    OUTPUT_DIR_029
    / "TB_Portals_March2025_Step_3B_6_29_CXR_Semantic_Structural_Conflict_Audit.csv"
)


conflict_records_029.to_csv(
    CONFLICT_AUDIT_PATH_029,
    index=False
)


# ====================================================================================================
# 18. OUTPUT FILE VALIDATION
# ====================================================================================================

print("\n17. OUTPUT FILE VALIDATION")
print("-" * 100)


output_paths_029 = [
    ELIGIBILITY_EVIDENCE_PATH_029,
    EVIDENCE_STATUS_PATH_029,
    CONFLICT_AUDIT_PATH_029,
]


for output_path_029 in output_paths_029:

    if not output_path_029.exists():
        raise AssertionError(
            f"Output file was not created: "
            f"{output_path_029}"
        )

    if output_path_029.stat().st_size == 0:
        raise AssertionError(
            f"Output file is empty: "
            f"{output_path_029}"
        )

    print(
        f"{output_path_029.name}: PASS"
    )


# ====================================================================================================
# 19. FINAL VALIDATION
# ====================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.29 — FINAL STATUS")
print("=" * 100)


print(
    f"3,859-record population              : PASS"
)

print(
    f"3,859 unique CXR URLs               : PASS"
)

print(
    f"3,081 unique conditions             : PASS"
)

print(
    f"Both reviewed                       : {both_reviewed_actual_029}"
)

print(
    f"Semantic only                       : {semantic_only_actual_029}"
)

print(
    f"Structural only                     : {structural_only_actual_029}"
)

print(
    f"Neither reviewed                    : {neither_actual_029}"
)

print(
    f"Semantic ↔ structural conflicts     : {len(conflict_records_029)}"
)

print(
    "cxr_outlier automatic exclusion     : NO"
)

print(
    "Automatic modality exclusion        : NO"
)

print(
    "Final research cohort selected      : NO"
)

print(
    "DICOM files modified                : NO"
)

print(
    "Source evidence artifacts modified  : NO"
)

print(
    "Output files                        : PASS"
)


print("\n" + "=" * 100)
print("STEP 3B.6.29 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.30 — IMAGE-LEVEL CXR ELIGIBILITY AUDIT
# FINAL VERSION BASED ON ACTUAL 3B.6.28E SCHEMA
# ===================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# ===================================================================================================
# 1. INITIALIZATION
# ===================================================================================================

print("=" * 100)
print("STEP 3B.6.30 — IMAGE-LEVEL CXR ELIGIBILITY AUDIT")
print("=" * 100)

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_024_DIR = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03"
)

MASTER_PATH = (
    STEP_024_DIR /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

EVIDENCE_PATH = (
    STEP_024_DIR /
    "TB_Portals_March2025_Step_3B_6_28E_Final_3859_Semantic_Structural_Evidence.csv"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_30_Image_Level_CXR_Eligibility"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ===================================================================================================
# 2. INPUT ARTIFACT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

assert MASTER_PATH.exists(), (
    f"Missing CXR master:\n{MASTER_PATH}"
)

assert EVIDENCE_PATH.exists(), (
    f"Missing 3B.6.28E evidence:\n{EVIDENCE_PATH}"
)

print("CXR master        : FOUND")
print("3B.6.28E evidence : FOUND")
print("Input validation  : PASS")


# ===================================================================================================
# 3. LOAD ARTIFACTS
# ===================================================================================================

master = pd.read_csv(
    MASTER_PATH,
    low_memory=False
)

evidence = pd.read_csv(
    EVIDENCE_PATH,
    low_memory=False
)

print("\n2. LOADED ARTIFACTS")
print("-" * 100)

print(
    "CXR master rows       :",
    len(master)
)

print(
    "3B.6.28E evidence rows:",
    len(evidence)
)


# ===================================================================================================
# 4. REQUIRED COLUMNS — ACTUAL 3B.6.28E SCHEMA
# ===================================================================================================

master_required = [
    "condition_id",
    "series_instance_content_url",
    "series_modality_cd",
    "cxr_outlier",
    "imaging_date",
]

evidence_required = [
    "condition_id",
    "series_instance_content_url",

    "structural_decision",
    "structural_review_candidate",
    "structural_review_flags",
    "structural_review_flag_count",
    "structural_reviewed",

    "human_semantic_review_status",
    "human_semantic_decision",
    "human_semantic_reason",
    "final_resolution_status",
    "semantic_reviewed",

    "semantic_structural_evidence_category",
    "semantic_structural_decision_comparison",

    "cxr_eligibility_decision",
    "cxr_eligibility_reason",
]

print("\n3. REQUIRED COLUMN VALIDATION")
print("-" * 100)

missing_master = [
    column
    for column in master_required
    if column not in master.columns
]

missing_evidence = [
    column
    for column in evidence_required
    if column not in evidence.columns
]

print("Missing master columns  :", missing_master)
print("Missing evidence columns:", missing_evidence)

assert len(missing_master) == 0, (
    f"Master missing columns: {missing_master}"
)

assert len(missing_evidence) == 0, (
    f"Evidence missing columns: {missing_evidence}"
)

print("Required columns: PASS")


# ===================================================================================================
# 5. MASTER POPULATION
# ===================================================================================================

print("\n4. MASTER POPULATION VALIDATION")
print("-" * 100)

assert len(master) == 3859

assert master["condition_id"].notna().all()

assert master[
    "series_instance_content_url"
].notna().all()

assert (
    master[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    master["condition_id"].nunique()
    == 3081
)

print("Records          : 3,859")
print("Unique URLs      : 3,859")
print("Unique conditions: 3,081")
print("Master population: PASS")


# ===================================================================================================
# 6. EVIDENCE POPULATION
# ===================================================================================================

print("\n5. EVIDENCE POPULATION VALIDATION")
print("-" * 100)

assert len(evidence) == 3859

assert (
    evidence[
        "series_instance_content_url"
    ].notna().all()
)

assert (
    evidence[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

print("Evidence records: 3,859")
print("Unique URLs    : 3,859")
print("Evidence population: PASS")


# ===================================================================================================
# 7. URL-LEVEL PROVENANCE
# ===================================================================================================

print("\n6. URL-LEVEL PROVENANCE VALIDATION")
print("-" * 100)

master_urls = set(
    master["series_instance_content_url"]
)

evidence_urls = set(
    evidence["series_instance_content_url"]
)

missing_urls = (
    master_urls
    - evidence_urls
)

extra_urls = (
    evidence_urls
    - master_urls
)

print(
    "Master URLs missing from evidence:",
    len(missing_urls)
)

print(
    "Evidence URLs absent from master :",
    len(extra_urls)
)

assert len(missing_urls) == 0

assert len(extra_urls) == 0

print("URL provenance: PASS")


# ===================================================================================================
# 8. CONDITION-ID PROVENANCE
# ===================================================================================================

print("\n7. CONDITION-ID PROVENANCE")
print("-" * 100)

master_condition = master[
    [
        "series_instance_content_url",
        "condition_id"
    ]
].copy()

evidence_condition = evidence[
    [
        "series_instance_content_url",
        "condition_id"
    ]
].copy()

condition_check = master_condition.merge(
    evidence_condition,
    on="series_instance_content_url",
    how="left",
    validate="one_to_one",
    suffixes=(
        "_master",
        "_evidence"
    )
)

condition_mismatch = (
    condition_check[
        "condition_id_master"
    ]
    !=
    condition_check[
        "condition_id_evidence"
    ]
)

print(
    "Condition-ID mismatches:",
    int(condition_mismatch.sum())
)

assert condition_mismatch.sum() == 0

print("Condition-ID provenance: PASS")


# ===================================================================================================
# 9. CREATE AUDIT FROM AUTHORITATIVE 3B.6.28E
# ===================================================================================================
#
# 3B.6.28E is already the integrated evidence layer.
# We therefore preserve its values rather than reconstructing them.
#
# ===================================================================================================

print("\n8. AUTHORITATIVE EVIDENCE AUDIT TABLE")
print("-" * 100)

audit = evidence.copy()

assert len(audit) == 3859

assert (
    audit[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

print("Using 3B.6.28E as authoritative integrated evidence layer.")
print("Audit population: 3,859")


# ===================================================================================================
# 10. NORMALIZE DECISION FIELDS FOR AUDITING
# ===================================================================================================

print("\n9. DECISION NORMALIZATION")
print("-" * 100)

audit[
    "semantic_decision_normalized"
] = (
    audit[
        "human_semantic_decision"
    ]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)

audit[
    "structural_decision_normalized"
] = (
    audit[
        "structural_decision"
    ]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)

audit[
    "eligibility_decision_normalized"
] = (
    audit[
        "cxr_eligibility_decision"
    ]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)

print("Decision normalization: PASS")


# ===================================================================================================
# 11. SEMANTIC REVIEW AUDIT
# ===================================================================================================

print("\n10. SEMANTIC REVIEW AUDIT")
print("-" * 100)

semantic_reviewed_calc = (
    audit[
        "semantic_decision_normalized"
    ]
    .isin([
        "RETAIN",
        "EXCLUDE"
    ])
)

semantic_reviewed_stored = (
    audit[
        "semantic_reviewed"
    ]
    .fillna(False)
    .astype(bool)
)

semantic_flag_mismatch = (
    semantic_reviewed_calc
    != semantic_reviewed_stored
)

print(
    "Calculated semantic reviewed:",
    int(semantic_reviewed_calc.sum())
)

print(
    "Stored semantic reviewed   :",
    int(semantic_reviewed_stored.sum())
)

print(
    "Semantic flag mismatches   :",
    int(semantic_flag_mismatch.sum())
)

assert semantic_reviewed_calc.sum() == 115

assert semantic_reviewed_stored.sum() == 115

assert semantic_flag_mismatch.sum() == 0

print("Semantic review audit: PASS")


# ===================================================================================================
# 12. STRUCTURAL REVIEW AUDIT
# ===================================================================================================

print("\n11. STRUCTURAL REVIEW AUDIT")
print("-" * 100)

structural_reviewed_stored = (
    audit[
        "structural_reviewed"
    ]
    .fillna(False)
    .astype(bool)
)

print(
    "Structural reviewed records:",
    int(structural_reviewed_stored.sum())
)

assert structural_reviewed_stored.sum() == 142

print("Structural review audit: PASS")


# ===================================================================================================
# 13. REVIEW STRUCTURE
# ===================================================================================================

print("\n12. REVIEW STRUCTURE")
print("-" * 100)

audit["review_structure"] = (
    "NEITHER_REVIEWED"
)

mask = (
    semantic_reviewed_stored
    & structural_reviewed_stored
)

audit.loc[
    mask,
    "review_structure"
] = "BOTH_REVIEWED"

mask = (
    semantic_reviewed_stored
    & ~structural_reviewed_stored
)

audit.loc[
    mask,
    "review_structure"
] = "SEMANTIC_ONLY"

mask = (
    ~semantic_reviewed_stored
    & structural_reviewed_stored
)

audit.loc[
    mask,
    "review_structure"
] = "STRUCTURAL_ONLY"

review_structure_counts = (
    audit[
        "review_structure"
    ]
    .value_counts()
    .to_dict()
)

expected_review_structure = {
    "BOTH_REVIEWED": 17,
    "SEMANTIC_ONLY": 98,
    "STRUCTURAL_ONLY": 125,
    "NEITHER_REVIEWED": 3619,
}

print(
    audit[
        "review_structure"
    ]
    .value_counts()
    .to_string()
)

assert (
    review_structure_counts
    == expected_review_structure
)

print("Review structure: PASS")


# ===================================================================================================
# 14. SEMANTIC ↔ STRUCTURAL DISAGREEMENT
# ===================================================================================================

print("\n13. SEMANTIC ↔ STRUCTURAL CONSISTENCY")
print("-" * 100)

audit[
    "raw_semantic_structural_disagreement"
] = (
    audit[
        "semantic_structural_decision_comparison"
    ]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
    .eq("DISAGREEMENT")
)

disagreement_count = int(
    audit[
        "raw_semantic_structural_disagreement"
    ].sum()
)

print(
    "Historical disagreements:",
    disagreement_count
)

assert disagreement_count == 5

print("Disagreement population: PASS")


# ===================================================================================================
# 15. VERIFY HISTORICAL CONFLICTS ARE RESOLVED
# ===================================================================================================

print("\n14. HISTORICAL CONFLICT RESOLUTION")
print("-" * 100)

conflict_rows = audit[
    audit[
        "raw_semantic_structural_disagreement"
    ]
].copy()

assert len(conflict_rows) == 5

assert (
    conflict_rows[
        "final_resolution_status"
    ]
    .notna()
    .all()
)

assert (
    conflict_rows[
        "final_resolution_status"
    ]
    .astype(str)
    .str.strip()
    .str.len()
    > 0
).all()

print("Historical conflicts       : 5")
print("Final resolution present   : 5")
print("Unresolved conflicts       : 0")
print("Conflict provenance: PASS")


# ===================================================================================================
# 16. ESTABLISHED EVIDENCE CATEGORY
# ===================================================================================================

print("\n15. EVIDENCE CATEGORY")
print("-" * 100)

evidence_category_counts = (
    audit[
        "semantic_structural_evidence_category"
    ]
    .fillna("MISSING")
    .value_counts()
)

print(
    evidence_category_counts.to_string()
)

expected_evidence_categories = {
    "NEITHER_REVIEWED": 3619,
    "SEMANTIC_ONLY": 98,
    "STRUCTURAL_ONLY": 125,
    "BOTH_AGREE_EXCLUDE": 7,
    "BOTH_AGREE_RETAIN": 5,
    "BOTH_DISAGREE": 5,
}

assert (
    evidence_category_counts.to_dict()
    == expected_evidence_categories
)

print("Evidence category accounting: PASS")


# ===================================================================================================
# 17. ESTABLISHED DECISION COMPARISON
# ===================================================================================================

print("\n16. DECISION COMPARISON")
print("-" * 100)

comparison_counts = (
    audit[
        "semantic_structural_decision_comparison"
    ]
    .fillna("NOT_APPLICABLE")
    .value_counts()
)

print(
    comparison_counts.to_string()
)

expected_comparison = {
    "NOT_APPLICABLE": 3842,
    "AGREEMENT": 12,
    "DISAGREEMENT": 5,
}

assert (
    comparison_counts.to_dict()
    == expected_comparison
)

print("Decision comparison accounting: PASS")


# ===================================================================================================
# 18. CXR ELIGIBILITY FIELD AUDIT
# ===================================================================================================

print("\n17. EXISTING CXR ELIGIBILITY FIELD AUDIT")
print("-" * 100)

eligibility_decision_counts = (
    audit[
        "eligibility_decision_normalized"
    ]
    .replace({
        "": "NOT_SPECIFIED"
    })
    .value_counts()
)

print(
    eligibility_decision_counts.to_string()
)

print(
    "\nExisting cxr_eligibility_decision values have been PRESERVED."
)

print(
    "No new final cohort decision is made in Step 3B.6.30."
)


# ===================================================================================================
# 19. CXR OUTLIER SAFETY
# ===================================================================================================

print("\n18. CXR OUTLIER SAFETY CHECK")
print("-" * 100)

outlier_counts = (
    audit[
        "cxr_outlier"
    ]
    .value_counts(dropna=False)
)

print(
    outlier_counts.to_string()
)

expected_outliers = {
    "normal": 2915,
    "outlier": 932,
    "error": 12,
}

assert (
    outlier_counts.to_dict()
    == expected_outliers
)

print("Automatic cxr_outlier exclusion: NO")
print("cxr_outlier safety: PASS")


# ===================================================================================================
# 20. MODALITY SAFETY
# ===================================================================================================

print("\n19. MODALITY SAFETY CHECK")
print("-" * 100)

modality_counts = (
    audit[
        "series_modality_cd"
    ]
    .value_counts()
)

expected_modalities = {
    "XC": 2270,
    "CR": 1537,
    "XA": 32,
    "RG": 16,
    "DX": 4,
}

assert (
    modality_counts.to_dict()
    == expected_modalities
)

print(
    modality_counts.to_string()
)

print("Automatic modality exclusion: NO")
print("Modality safety: PASS")


# ===================================================================================================
# 21. ELIGIBILITY REASON COMPLETENESS
# ===================================================================================================

print("\n20. ELIGIBILITY REASON AUDIT")
print("-" * 100)

eligibility_reason = (
    audit[
        "cxr_eligibility_reason"
    ]
    .fillna("")
    .astype(str)
    .str.strip()
)

reason_missing_count = int(
    eligibility_reason.eq("").sum()
)

print(
    "Records with populated eligibility reason:",
    3859 - reason_missing_count
)

print(
    "Records without eligibility reason:",
    reason_missing_count
)

print(
    "Eligibility reason is preserved as evidence; "
    "no reason is invented."
)


# ===================================================================================================
# 22. IMAGE-LEVEL EVIDENCE STATUS
# ===================================================================================================

print("\n21. IMAGE-LEVEL EVIDENCE STATUS")
print("-" * 100)

audit[
    "image_level_evidence_status"
] = "NOT_INDIVIDUALLY_ADJUDICATED"

# Explicit final semantic exclusion.
mask = (
    semantic_reviewed_stored
    & (
        audit[
            "semantic_decision_normalized"
        ]
        == "EXCLUDE"
    )
)

audit.loc[
    mask,
    "image_level_evidence_status"
] = "HUMAN_SEMANTIC_EXCLUSION"


# Explicit final semantic retention.
mask = (
    semantic_reviewed_stored
    & (
        audit[
            "semantic_decision_normalized"
        ]
        == "RETAIN"
    )
)

audit.loc[
    mask,
    "image_level_evidence_status"
] = "HUMAN_SEMANTIC_RETENTION"


# Historical disagreements are separately identified.
mask = (
    audit[
        "raw_semantic_structural_disagreement"
    ]
)

audit.loc[
    mask,
    "image_level_evidence_status"
] = (
    "HUMAN_CONFLICT_RESOLVED"
)


# ===================================================================================================
# 23. IMAGE-LEVEL EVIDENCE SUMMARY
# ===================================================================================================

print("\n22. IMAGE-LEVEL EVIDENCE SUMMARY")
print("-" * 100)

image_evidence_counts = (
    audit[
        "image_level_evidence_status"
    ]
    .value_counts()
)

print(
    image_evidence_counts.to_string()
)

assert (
    int(
        (
            audit[
                "image_level_evidence_status"
            ]
            == "HUMAN_CONFLICT_RESOLVED"
        ).sum()
    )
    == 5
)

assert (
    len(audit)
    == 3859
)

print("Image-level evidence accounting: PASS")


# ===================================================================================================
# 24. FINAL COHORT LOCK
# ===================================================================================================

print("\n23. FINAL COHORT LOCK")
print("-" * 100)

audit[
    "final_research_cohort_status"
] = "NOT_YET_SELECTED"

assert (
    audit[
        "final_research_cohort_status"
    ]
    == "NOT_YET_SELECTED"
).all()

print("Final research cohort selection: NOT_YET_SELECTED")


# ===================================================================================================
# 25. SOURCE PROTECTION
# ===================================================================================================

print("\n24. SOURCE PROTECTION")
print("-" * 100)

print("Raw DICOM modification: NO")
print("Source metadata modification: NO")
print("Previous human-review artifacts modified: NO")
print("Source protection: PASS")


# ===================================================================================================
# 26. FINAL POPULATION INTEGRITY
# ===================================================================================================

print("\n25. FINAL POPULATION INTEGRITY")
print("-" * 100)

assert len(audit) == 3859

assert (
    audit[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    audit[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    audit[
        "semantic_reviewed"
    ]
    .fillna(False)
    .astype(bool)
    .sum()
    == 115
)

assert (
    audit[
        "structural_reviewed"
    ]
    .fillna(False)
    .astype(bool)
    .sum()
    == 142
)

assert (
    audit[
        "raw_semantic_structural_disagreement"
    ].sum()
    == 5
)

print("3,859 records              : PASS")
print("3,859 unique URLs          : PASS")
print("3,081 unique conditions    : PASS")
print("115 semantic reviews       : PASS")
print("142 structural reviews     : PASS")
print("5 historical disagreements : PASS")
print("Population integrity       : PASS")


# ===================================================================================================
# 27. SAVE OUTPUTS
# ===================================================================================================

print("\n26. SAVING OUTPUTS")
print("-" * 100)

AUDIT_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_30_Image_Level_CXR_Eligibility_Audit.csv"
)

SUMMARY_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_30_Image_Level_Eligibility_Summary.csv"
)

CONFLICT_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_30_Semantic_Structural_Conflict_Audit.csv"
)

UNREVIEWED_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_30_Unreviewed_Image_Audit.csv"
)


audit.to_csv(
    AUDIT_PATH,
    index=False
)


summary = (
    audit[
        [
            "image_level_evidence_status"
        ]
    ]
    .value_counts()
    .rename_axis(
        "image_level_evidence_status"
    )
    .reset_index(
        name="record_count"
    )
)

summary.to_csv(
    SUMMARY_PATH,
    index=False
)


audit[
    audit[
        "raw_semantic_structural_disagreement"
    ]
].to_csv(
    CONFLICT_PATH,
    index=False
)


audit[
    audit[
        "image_level_evidence_status"
    ]
    == "NOT_INDIVIDUALLY_ADJUDICATED"
].to_csv(
    UNREVIEWED_PATH,
    index=False
)


# ===================================================================================================
# 28. OUTPUT VALIDATION
# ===================================================================================================

print("\n27. OUTPUT FILE VALIDATION")
print("-" * 100)

output_paths = [
    AUDIT_PATH,
    SUMMARY_PATH,
    CONFLICT_PATH,
    UNREVIEWED_PATH,
]

for output_path in output_paths:

    assert output_path.exists()

    assert output_path.stat().st_size > 0

    print(
        f"{output_path.name}: PASS"
    )


# ===================================================================================================
# 29. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.30 — FINAL STATUS")
print("=" * 100)

print(
    "3,859-record population              : PASS"
)

print(
    "3,859 unique CXR URLs               : PASS"
)

print(
    "3,081 unique conditions              : PASS"
)

print(
    "115 semantic human reviews          : PASS"
)

print(
    "142 structural human reviews        : PASS"
)

print(
    "5 historical disagreements          : PASS"
)

print(
    "5 historical conflicts resolved     : PASS"
)

print(
    "cxr_outlier automatic exclusion     : NO"
)

print(
    "Automatic modality exclusion        : NO"
)

print(
    "Final research cohort selected      : NO"
)

print(
    "DICOM files modified                : NO"
)

print(
    "Source evidence modified            : NO"
)

print(
    "Output files                         : PASS"
)

print("\n" + "=" * 100)
print("STEP 3B.6.30 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.31 — CONDITION-LEVEL CXR STRUCTURE AUDIT
# ===================================================================================================
#
# PURPOSE
# -------
# Analyze the 3,859-record CXR population at the condition_id level.
#
# This is necessary because:
#   3,859 CXR records
#   3,081 unique conditions
#   485 conditions have multiple CXR records
#
# This step DOES NOT:
#   - select the final research cohort
#   - select the final modeling target
#   - preprocess images
#   - extract additional ZIP files
#   - modify DICOM files
#   - automatically exclude modalities
#   - automatically exclude cxr_outlier records
#
# ===================================================================================================


from pathlib import Path
import pandas as pd
import numpy as np


# ===================================================================================================
# 1. INITIALIZATION
# ===================================================================================================

print("=" * 100)
print("STEP 3B.6.31 — CONDITION-LEVEL CXR STRUCTURE AUDIT")
print("=" * 100)

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_024_DIR = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03"
)

STEP_030_DIR = (
    METADATA_DIR /
    "Step_3B_6_30_Image_Level_CXR_Eligibility"
)

MASTER_PATH = (
    STEP_024_DIR /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

AUDIT_030_PATH = (
    STEP_030_DIR /
    "TB_Portals_March2025_Step_3B_6_30_Image_Level_CXR_Eligibility_Audit.csv"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_31_Condition_Level_CXR_Structure"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ===================================================================================================
# 2. INPUT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

assert MASTER_PATH.exists(), (
    f"Missing CXR master:\n{MASTER_PATH}"
)

assert AUDIT_030_PATH.exists(), (
    f"Missing Step 3B.6.30 audit:\n{AUDIT_030_PATH}"
)

print("CXR master                 : FOUND")
print("Step 3B.6.30 image audit   : FOUND")
print("Input validation           : PASS")


# ===================================================================================================
# 3. LOAD DATA
# ===================================================================================================

master = pd.read_csv(
    MASTER_PATH,
    low_memory=False
)

image_audit = pd.read_csv(
    AUDIT_030_PATH,
    low_memory=False
)

print("\n2. LOADED ARTIFACTS")
print("-" * 100)

print(
    "CXR master records       :",
    len(master)
)

print(
    "Step 3B.6.30 audit rows  :",
    len(image_audit)
)


# ===================================================================================================
# 4. REQUIRED COLUMN VALIDATION
# ===================================================================================================

master_required = [
    "condition_id",
    "series_instance_content_url",
    "series_modality_cd",
    "cxr_outlier",
    "imaging_date",
]

audit_required = [
    "condition_id",
    "series_instance_content_url",
    "imaging_date",
    "series_modality_cd",
    "cxr_outlier",
    "semantic_reviewed",
    "structural_reviewed",
    "semantic_structural_evidence_category",
    "semantic_structural_decision_comparison",
    "image_level_evidence_status",
]

print("\n3. REQUIRED COLUMN VALIDATION")
print("-" * 100)

missing_master = [
    column
    for column in master_required
    if column not in master.columns
]

missing_audit = [
    column
    for column in audit_required
    if column not in image_audit.columns
]

print("Missing master columns:", missing_master)
print("Missing audit columns :", missing_audit)

assert len(missing_master) == 0
assert len(missing_audit) == 0

print("Required columns: PASS")


# ===================================================================================================
# 5. IMAGE-LEVEL POPULATION VALIDATION
# ===================================================================================================

print("\n4. IMAGE-LEVEL POPULATION VALIDATION")
print("-" * 100)

assert len(master) == 3859

assert len(image_audit) == 3859

assert (
    master[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    image_audit[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    master[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    image_audit[
        "condition_id"
    ].nunique()
    == 3081
)

print("Master records       : 3,859")
print("Audit records        : 3,859")
print("Master conditions    : 3,081")
print("Audit conditions     : 3,081")
print("Image-level population: PASS")


# ===================================================================================================
# 6. URL-LEVEL RECONCILIATION
# ===================================================================================================

print("\n5. URL-LEVEL RECONCILIATION")
print("-" * 100)

master_urls = set(
    master[
        "series_instance_content_url"
    ]
)

audit_urls = set(
    image_audit[
        "series_instance_content_url"
    ]
)

missing_from_audit = (
    master_urls - audit_urls
)

extra_in_audit = (
    audit_urls - master_urls
)

print(
    "Master URLs missing from audit:",
    len(missing_from_audit)
)

print(
    "Audit URLs absent from master :",
    len(extra_in_audit)
)

assert len(missing_from_audit) == 0
assert len(extra_in_audit) == 0

print("URL reconciliation: PASS")


# ===================================================================================================
# 7. CONDITION-ID RECONCILIATION
# ===================================================================================================

print("\n6. CONDITION-ID RECONCILIATION")
print("-" * 100)

master_condition_map = master[
    [
        "series_instance_content_url",
        "condition_id"
    ]
].copy()

audit_condition_map = image_audit[
    [
        "series_instance_content_url",
        "condition_id"
    ]
].copy()

condition_check = master_condition_map.merge(
    audit_condition_map,
    on="series_instance_content_url",
    how="left",
    validate="one_to_one",
    suffixes=(
        "_master",
        "_audit"
    )
)

condition_mismatch = (
    condition_check[
        "condition_id_master"
    ]
    !=
    condition_check[
        "condition_id_audit"
    ]
)

print(
    "Condition-ID mismatches:",
    int(condition_mismatch.sum())
)

assert condition_mismatch.sum() == 0

print("Condition-ID reconciliation: PASS")


# ===================================================================================================
# 8. PREPARE CONDITION-LEVEL WORKING TABLE
# ===================================================================================================

print("\n7. CONDITION-LEVEL WORKING TABLE")
print("-" * 100)

condition_data = image_audit.copy()

condition_data[
    "imaging_date_num_clean"
] = pd.to_numeric(
    condition_data[
        "imaging_date"
    ],
    errors="coerce"
)

condition_data[
    "semantic_reviewed_bool"
] = (
    condition_data[
        "semantic_reviewed"
    ]
    .fillna(False)
    .astype(bool)
)

condition_data[
    "structural_reviewed_bool"
] = (
    condition_data[
        "structural_reviewed"
    ]
    .fillna(False)
    .astype(bool)
)

condition_data[
    "image_is_retained_evidence"
] = (
    condition_data[
        "image_level_evidence_status"
    ]
    .isin([
        "HUMAN_SEMANTIC_RETENTION"
    ])
)

condition_data[
    "image_is_excluded_evidence"
] = (
    condition_data[
        "image_level_evidence_status"
    ]
    .isin([
        "HUMAN_SEMANTIC_EXCLUSION"
    ])
)

print(
    "Working records:",
    len(condition_data)
)

print("Condition-level working table: PASS")


# ===================================================================================================
# 9. BASIC CONDITION-LEVEL IMAGE COUNTS
# ===================================================================================================

print("\n8. CXR RECORDS PER CONDITION")
print("-" * 100)

condition_image_counts = (
    condition_data
    .groupby(
        "condition_id"
    )
    .size()
    .rename(
        "cxr_record_count"
    )
)

assert (
    condition_image_counts.size
    == 3081
)

assert (
    condition_image_counts.sum()
    == 3859
)

image_count_distribution = (
    condition_image_counts
    .value_counts()
    .sort_index()
)

print(
    image_count_distribution.to_string()
)

multiple_cxr_conditions = int(
    (condition_image_counts > 1).sum()
)

single_cxr_conditions = int(
    (condition_image_counts == 1).sum()
)

print()
print(
    "Single-CXR conditions  :",
    single_cxr_conditions
)

print(
    "Multiple-CXR conditions:",
    multiple_cxr_conditions
)

assert multiple_cxr_conditions == 485

assert single_cxr_conditions == (
    3081 - 485
)

print("Condition image-count audit: PASS")


# ===================================================================================================
# 10. IMAGING DATE STRUCTURE
# ===================================================================================================

print("\n9. IMAGING DATE STRUCTURE")
print("-" * 100)

date_summary = (
    condition_data
    .groupby(
        "condition_id"
    )[
        "imaging_date_num_clean"
    ]
    .agg(
        imaging_date_count="count",
        unique_imaging_dates="nunique",
        earliest_imaging_date="min",
        latest_imaging_date="max",
    )
)

date_summary[
    "imaging_date_span"
] = (
    date_summary[
        "latest_imaging_date"
    ]
    -
    date_summary[
        "earliest_imaging_date"
    ]
)

assert (
    date_summary.index.nunique()
    == 3081
)

multiple_date_conditions = int(
    (
        date_summary[
            "unique_imaging_dates"
        ] > 1
    ).sum()
)

print(
    "Conditions with >1 imaging date:",
    multiple_date_conditions
)

print(
    "\nImaging-date span summary:"
)

print(
    date_summary[
        "imaging_date_span"
    ]
    .describe()
    .to_string()
)

print("Imaging-date structure: PASS")


# ===================================================================================================
# 11. SAME-DAY MULTIPLE-CXR ANALYSIS
# ===================================================================================================

print("\n10. SAME-DAY MULTIPLE-CXR ANALYSIS")
print("-" * 100)

same_day_counts = (
    condition_data
    .groupby(
        [
            "condition_id",
            "imaging_date_num_clean"
        ],
        dropna=False
    )
    .size()
    .rename(
        "same_day_cxr_count"
    )
    .reset_index()
)

same_day_multiple_groups = same_day_counts[
    same_day_counts[
        "same_day_cxr_count"
    ] > 1
].copy()

same_day_multiple_group_count = len(
    same_day_multiple_groups
)

same_day_multiple_record_count = int(
    same_day_multiple_groups[
        "same_day_cxr_count"
    ].sum()
)

print(
    "Condition-date groups with >1 CXR:",
    same_day_multiple_group_count
)

print(
    "CXR records in those groups      :",
    same_day_multiple_record_count
)

assert same_day_multiple_group_count == 58

assert same_day_multiple_record_count == 141

print("Same-day multiple-CXR audit: PASS")


# ===================================================================================================
# 12. MODALITY STRUCTURE PER CONDITION
# ===================================================================================================

print("\n11. MODALITY STRUCTURE PER CONDITION")
print("-" * 100)

modality_per_condition = (
    condition_data
    .groupby(
        "condition_id"
    )[
        "series_modality_cd"
    ]
    .agg(
        lambda x: "|".join(
            sorted(
                set(
                    x.dropna()
                    .astype(str)
                    .str.strip()
                )
            )
        )
    )
    .rename(
        "condition_modalities"
    )
)

condition_modality_counts = (
    modality_per_condition
    .value_counts()
)

print(
    condition_modality_counts.to_string()
)

print("Condition modality structure: PASS")


# ===================================================================================================
# 13. CXR OUTLIER STRUCTURE PER CONDITION
# ===================================================================================================

print("\n12. CXR OUTLIER STRUCTURE PER CONDITION")
print("-" * 100)

outlier_per_condition = (
    condition_data
    .groupby(
        "condition_id"
    )[
        "cxr_outlier"
    ]
    .agg(
        lambda x: "|".join(
            sorted(
                set(
                    x.dropna()
                    .astype(str)
                    .str.strip()
                )
            )
        )
    )
    .rename(
        "condition_cxr_outlier_states"
    )
)

print(
    outlier_per_condition
    .value_counts()
    .to_string()
)

print(
    "No cxr_outlier state is treated as an automatic exclusion."
)

print("CXR outlier structure: PASS")


# ===================================================================================================
# 14. HUMAN REVIEW COVERAGE PER CONDITION
# ===================================================================================================

print("\n13. HUMAN REVIEW COVERAGE PER CONDITION")
print("-" * 100)

review_condition_summary = (
    condition_data
    .groupby(
        "condition_id"
    )
    .agg(
        cxr_record_count=(
            "series_instance_content_url",
            "count"
        ),
        semantic_reviewed_count=(
            "semantic_reviewed_bool",
            "sum"
        ),
        structural_reviewed_count=(
            "structural_reviewed_bool",
            "sum"
        ),
        semantic_retention_evidence_count=(
            "image_is_retained_evidence",
            "sum"
        ),
        semantic_exclusion_evidence_count=(
            "image_is_excluded_evidence",
            "sum"
        ),
    )
)

review_condition_summary[
    "has_any_semantic_review"
] = (
    review_condition_summary[
        "semantic_reviewed_count"
    ] > 0
)

review_condition_summary[
    "has_any_structural_review"
] = (
    review_condition_summary[
        "structural_reviewed_count"
    ] > 0
)

review_condition_summary[
    "has_multiple_cxr"
] = (
    review_condition_summary[
        "cxr_record_count"
    ] > 1
)

print(
    "Conditions with any semantic review:",
    int(
        review_condition_summary[
            "has_any_semantic_review"
        ].sum()
    )
)

print(
    "Conditions with any structural review:",
    int(
        review_condition_summary[
            "has_any_structural_review"
        ].sum()
    )
)

print(
    "Conditions with multiple CXRs:",
    int(
        review_condition_summary[
            "has_multiple_cxr"
        ].sum()
    )
)

assert (
    review_condition_summary.index.nunique()
    == 3081
)

print("Condition review coverage: PASS")


# ===================================================================================================
# 15. CONDITION-LEVEL EVIDENCE MIX
# ===================================================================================================

print("\n14. CONDITION-LEVEL EVIDENCE MIX")
print("-" * 100)

def classify_condition_evidence(row):

    semantic = row[
        "semantic_reviewed_count"
    ] > 0

    structural = row[
        "structural_reviewed_count"
    ] > 0

    if semantic and structural:
        return "BOTH_REVIEW_TYPES"

    if semantic:
        return "SEMANTIC_REVIEW_ONLY"

    if structural:
        return "STRUCTURAL_REVIEW_ONLY"

    return "NO_HUMAN_REVIEW"


review_condition_summary[
    "condition_review_structure"
] = review_condition_summary.apply(
    classify_condition_evidence,
    axis=1
)

condition_review_structure_counts = (
    review_condition_summary[
        "condition_review_structure"
    ]
    .value_counts()
)

print(
    condition_review_structure_counts
    .to_string()
)

print("Condition-level evidence structure: PASS")


# ===================================================================================================
# 16. CONDITIONS WITH MIXED IMAGE-LEVEL EVIDENCE
# ===================================================================================================

print("\n15. CONDITIONS WITH MIXED IMAGE-LEVEL EVIDENCE")
print("-" * 100)

condition_status_sets = (
    condition_data
    .groupby(
        "condition_id"
    )[
        "image_level_evidence_status"
    ]
    .agg(
        lambda x: set(
            x.dropna()
            .astype(str)
        )
    )
)

mixed_evidence_conditions = (
    condition_status_sets
    .apply(
        lambda x: len(x) > 1
    )
)

mixed_evidence_count = int(
    mixed_evidence_conditions.sum()
)

print(
    "Conditions with >1 image-level evidence state:",
    mixed_evidence_count
)

print(
    "\nEvidence-state combinations:"
)

evidence_combination_summary = (
    condition_status_sets
    .apply(
        lambda x: "|".join(sorted(x))
    )
    .value_counts()
)

print(
    evidence_combination_summary
    .to_string()
)

print("Mixed image-level evidence audit: PASS")


# ===================================================================================================
# 17. CONDITIONS WITH MULTIPLE CXRs AND MULTIPLE DATES
# ===================================================================================================

print("\n16. MULTIPLE-CXR / MULTIPLE-DATE INTERSECTION")
print("-" * 100)

multi_cxr_condition_ids = set(
    review_condition_summary.index[
        review_condition_summary[
            "has_multiple_cxr"
        ]
    ]
)

multi_date_condition_ids = set(
    date_summary.index[
        date_summary[
            "unique_imaging_dates"
        ] > 1
    ]
)

intersection_multi_cxr_multi_date = (
    multi_cxr_condition_ids
    & multi_date_condition_ids
)

print(
    "Multiple-CXR conditions:",
    len(multi_cxr_condition_ids)
)

print(
    "Multiple-date conditions:",
    len(multi_date_condition_ids)
)

print(
    "Both multiple-CXR and multiple-date:",
    len(intersection_multi_cxr_multi_date)
)


# ===================================================================================================
# 18. BUILD CONDITION-LEVEL MASTER AUDIT
# ===================================================================================================

print("\n17. BUILDING CONDITION-LEVEL MASTER AUDIT")
print("-" * 100)

condition_audit = (
    review_condition_summary
    .join(
        date_summary,
        how="left"
    )
    .join(
        modality_per_condition,
        how="left"
    )
    .join(
        outlier_per_condition,
        how="left"
    )
)

condition_audit[
    "final_research_cohort_status"
] = "NOT_YET_SELECTED"

assert (
    len(condition_audit)
    == 3081
)

assert (
    condition_audit.index.nunique()
    == 3081
)

print(
    "Condition-level audit records:",
    len(condition_audit)
)

print("Condition-level master audit: PASS")


# ===================================================================================================
# 19. CONDITION-LEVEL ACCOUNTING
# ===================================================================================================

print("\n18. CONDITION-LEVEL ACCOUNTING")
print("-" * 100)

assert (
    condition_audit[
        "cxr_record_count"
    ].sum()
    == 3859
)

assert (
    condition_audit[
        "semantic_reviewed_count"
    ].sum()
    == 115
)

assert (
    condition_audit[
        "structural_reviewed_count"
    ].sum()
    == 142
)

assert (
    condition_audit[
        "has_multiple_cxr"
    ].sum()
    == 485
)

print("Total CXR records represented : 3,859")
print("Total conditions represented  : 3,081")
print("Semantic reviews represented  : 115")
print("Structural reviews represented: 142")
print("Multiple-CXR conditions       : 485")
print("Condition accounting: PASS")


# ===================================================================================================
# 20. FINAL COHORT LOCK
# ===================================================================================================

print("\n19. FINAL COHORT LOCK")
print("-" * 100)

assert (
    condition_audit[
        "final_research_cohort_status"
    ]
    == "NOT_YET_SELECTED"
).all()

print(
    "Final research cohort selection: NOT_YET_SELECTED"
)


# ===================================================================================================
# 21. SOURCE PROTECTION
# ===================================================================================================

print("\n20. SOURCE PROTECTION")
print("-" * 100)

print("Raw DICOM modification             : NO")
print("Source metadata modification       : NO")
print("Human-review artifacts modification: NO")
print("Additional ZIP extraction          : NO")
print("Source protection                  : PASS")


# ===================================================================================================
# 22. SAVE OUTPUTS
# ===================================================================================================

print("\n21. SAVING OUTPUTS")
print("-" * 100)

CONDITION_AUDIT_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_31_Condition_Level_CXR_Audit.csv"
)

IMAGE_COUNT_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_31_Condition_CXR_Count_Distribution.csv"
)

SAME_DAY_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_31_Same_Day_Multiple_CXR_Groups.csv"
)

MIXED_EVIDENCE_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_31_Mixed_Image_Evidence_Conditions.csv"
)


condition_audit_to_save = (
    condition_audit
    .reset_index()
)

condition_audit_to_save.to_csv(
    CONDITION_AUDIT_PATH,
    index=False
)

image_count_distribution_df = (
    image_count_distribution
    .rename(
        "condition_count"
    )
    .reset_index()
)

image_count_distribution_df.to_csv(
    IMAGE_COUNT_PATH,
    index=False
)

same_day_multiple_groups.to_csv(
    SAME_DAY_PATH,
    index=False
)

mixed_condition_table = (
    condition_status_sets
    .reset_index()
)

mixed_condition_table[
    "has_mixed_evidence"
] = mixed_condition_table[
    "image_level_evidence_status"
].apply(
    lambda x: len(x) > 1
)

mixed_condition_table[
    mixed_condition_table[
        "has_mixed_evidence"
    ]
].to_csv(
    MIXED_EVIDENCE_PATH,
    index=False
)


# ===================================================================================================
# 23. OUTPUT VALIDATION
# ===================================================================================================

print("\n22. OUTPUT FILE VALIDATION")
print("-" * 100)

output_paths = [
    CONDITION_AUDIT_PATH,
    IMAGE_COUNT_PATH,
    SAME_DAY_PATH,
    MIXED_EVIDENCE_PATH,
]

for output_path in output_paths:

    assert output_path.exists()

    assert output_path.stat().st_size > 0

    print(
        f"{output_path.name}: PASS"
    )


# ===================================================================================================
# 24. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.31 — FINAL STATUS")
print("=" * 100)

print(
    "3,859 CXR records represented       : PASS"
)

print(
    "3,081 conditions represented       : PASS"
)

print(
    "485 multiple-CXR conditions        : PASS"
)

print(
    f"Multiple-date conditions           : "
    f"{multiple_date_conditions}"
)

print(
    f"Same-day multiple-CXR groups       : "
    f"{same_day_multiple_group_count}"
)

print(
    f"Same-day CXR records involved      : "
    f"{same_day_multiple_record_count}"
)

print(
    "115 semantic review records        : PASS"
)

print(
    "142 structural review records      : PASS"
)

print(
    "Historical conflicts preserved     : PASS"
)

print(
    "Final research cohort selected     : NO"
)

print(
    "Additional ZIP extraction          : NO"
)

print(
    "DICOM files modified               : NO"
)

print(
    "Source evidence modified           : NO"
)

print(
    "Output files                       : PASS"
)

print("\n" + "=" * 100)
print("STEP 3B.6.31 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.32 — CONDITION-LEVEL ELIGIBILITY EVIDENCE RECONCILIATION
# SELF-CONTAINED FINAL VERSION
# ===================================================================================================
#
# PURPOSE
# -------
# Reconcile image-level CXR eligibility evidence at the condition_id level.
#
# IMPORTANT METHODOLOGICAL RULES
# --------------------------------
# 1. condition_id is the case-level unit.
# 2. Multiple CXRs belonging to the same condition are NOT automatically collapsed.
# 3. An excluded image does NOT automatically exclude the entire condition.
# 4. A retained image does NOT automatically make the entire condition eligible.
# 5. Unreviewed images remain unreviewed.
# 6. cxr_outlier is NOT an automatic exclusion criterion.
# 7. Modality is NOT an automatic exclusion criterion.
# 8. The final research cohort is NOT selected here.
# 9. The final modeling target is NOT selected here.
# 10. No DICOM is modified.
# 11. No source metadata is modified.
# 12. No additional ZIP extraction is performed.
#
# This cell is intentionally SELF-CONTAINED.
# It does not depend on condition_evidence, work, or any other
# DataFrame remaining in Jupyter memory from previous cells.
#
# ===================================================================================================


from pathlib import Path
import pandas as pd
import numpy as np


# ===================================================================================================
# 1. INITIALIZATION
# ===================================================================================================

print("=" * 100)
print("STEP 3B.6.32 — CONDITION-LEVEL ELIGIBILITY EVIDENCE RECONCILIATION")
print("=" * 100)

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_030_DIR = (
    METADATA_DIR /
    "Step_3B_6_30_Image_Level_CXR_Eligibility"
)

STEP_031_DIR = (
    METADATA_DIR /
    "Step_3B_6_31_Condition_Level_CXR_Structure"
)

IMAGE_AUDIT_PATH = (
    STEP_030_DIR /
    "TB_Portals_March2025_Step_3B_6_30_Image_Level_CXR_Eligibility_Audit.csv"
)

CONDITION_AUDIT_PATH = (
    STEP_031_DIR /
    "TB_Portals_March2025_Step_3B_6_31_Condition_Level_CXR_Audit.csv"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_32_Condition_Level_Eligibility_Evidence"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ===================================================================================================
# 2. INPUT ARTIFACT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

assert IMAGE_AUDIT_PATH.exists(), (
    f"Missing Step 3B.6.30 image audit:\n{IMAGE_AUDIT_PATH}"
)

assert CONDITION_AUDIT_PATH.exists(), (
    f"Missing Step 3B.6.31 condition audit:\n{CONDITION_AUDIT_PATH}"
)

print("Step 3B.6.30 image audit     : FOUND")
print("Step 3B.6.31 condition audit : FOUND")
print("Input validation              : PASS")


# ===================================================================================================
# 3. LOAD AUTHORITATIVE ARTIFACTS
# ===================================================================================================

image_audit = pd.read_csv(
    IMAGE_AUDIT_PATH,
    low_memory=False
)

condition_audit = pd.read_csv(
    CONDITION_AUDIT_PATH,
    low_memory=False
)

print("\n2. LOADED ARTIFACTS")
print("-" * 100)

print(
    "Image audit records      :",
    len(image_audit)
)

print(
    "Condition audit records :",
    len(condition_audit)
)


# ===================================================================================================
# 4. REQUIRED COLUMN VALIDATION
# ===================================================================================================

image_required = [
    "condition_id",
    "series_instance_content_url",
    "imaging_date",
    "series_modality_cd",
    "cxr_outlier",
    "semantic_reviewed",
    "structural_reviewed",
    "semantic_structural_evidence_category",
    "semantic_structural_decision_comparison",
    "image_level_evidence_status",
]

condition_required = [
    "condition_id",
    "cxr_record_count",
    "semantic_reviewed_count",
    "structural_reviewed_count",
    "has_any_semantic_review",
    "has_any_structural_review",
    "has_multiple_cxr",
    "unique_imaging_dates",
    "condition_modalities",
    "condition_cxr_outlier_states",
]

missing_image_columns = [
    column
    for column in image_required
    if column not in image_audit.columns
]

missing_condition_columns = [
    column
    for column in condition_required
    if column not in condition_audit.columns
]

print("\n3. REQUIRED COLUMN VALIDATION")
print("-" * 100)

print(
    "Missing image-audit columns     :",
    missing_image_columns
)

print(
    "Missing condition-audit columns:",
    missing_condition_columns
)

assert len(missing_image_columns) == 0, (
    f"Missing image-audit columns: {missing_image_columns}"
)

assert len(missing_condition_columns) == 0, (
    f"Missing condition-audit columns: {missing_condition_columns}"
)

print("Required columns: PASS")


# ===================================================================================================
# 5. BASIC POPULATION VALIDATION
# ===================================================================================================

print("\n4. POPULATION VALIDATION")
print("-" * 100)

assert len(image_audit) == 3859, (
    f"Expected 3859 image records, found {len(image_audit)}"
)

assert len(condition_audit) == 3081, (
    f"Expected 3081 conditions, found {len(condition_audit)}"
)

assert (
    image_audit[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    image_audit[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    condition_audit[
        "condition_id"
    ].nunique()
    == 3081
)

print("Image records     : 3,859")
print("Unique URLs       : 3,859")
print("Unique conditions : 3,081")
print("Population validation: PASS")


# ===================================================================================================
# 6. CONDITION-ID SET RECONCILIATION
# ===================================================================================================

print("\n5. CONDITION-ID RECONCILIATION")
print("-" * 100)

image_condition_ids = set(
    image_audit[
        "condition_id"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)

condition_condition_ids = set(
    condition_audit[
        "condition_id"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)

missing_conditions = (
    image_condition_ids
    -
    condition_condition_ids
)

extra_conditions = (
    condition_condition_ids
    -
    image_condition_ids
)

print(
    "Image conditions missing from condition audit:",
    len(missing_conditions)
)

print(
    "Extra conditions in condition audit:",
    len(extra_conditions)
)

assert len(missing_conditions) == 0
assert len(extra_conditions) == 0

print("Condition-ID reconciliation: PASS")


# ===================================================================================================
# 7. URL RECONCILIATION
# ===================================================================================================

print("\n6. URL RECONCILIATION")
print("-" * 100)

image_urls = set(
    image_audit[
        "series_instance_content_url"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)

assert len(image_urls) == 3859

print(
    "Image audit unique URLs:",
    len(image_urls)
)

print("URL integrity: PASS")


# ===================================================================================================
# 8. CREATE COMPLETELY FRESH IMAGE-LEVEL WORKING TABLE
# ===================================================================================================

print("\n7. BUILDING IMAGE-LEVEL WORKING TABLE")
print("-" * 100)

work = image_audit.copy()

work[
    "condition_id"
] = (
    work[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)

work[
    "series_instance_content_url"
] = (
    work[
        "series_instance_content_url"
    ]
    .astype(str)
    .str.strip()
)

work[
    "image_evidence_normalized"
] = (
    work[
        "image_level_evidence_status"
    ]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)

work[
    "semantic_reviewed_bool"
] = (
    work[
        "semantic_reviewed"
    ]
    .fillna(False)
    .astype(bool)
)

work[
    "structural_reviewed_bool"
] = (
    work[
        "structural_reviewed"
    ]
    .fillna(False)
    .astype(bool)
)

work[
    "structural_decision_normalized"
] = (
    work[
        "structural_decision"
    ]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)

assert len(work) == 3859

print(
    "Working image records:",
    len(work)
)

print("Working table: PASS")


# ===================================================================================================
# 9. VERIFY IMAGE-LEVEL EVIDENCE STATES
# ===================================================================================================

print("\n8. IMAGE-LEVEL EVIDENCE STATES")
print("-" * 100)

allowed_image_states = {
    "NOT_INDIVIDUALLY_ADJUDICATED",
    "HUMAN_SEMANTIC_RETENTION",
    "HUMAN_SEMANTIC_EXCLUSION",
    "HUMAN_CONFLICT_RESOLVED",
}

observed_image_states = set(
    work[
        "image_evidence_normalized"
    ]
    .unique()
)

unexpected_states = (
    observed_image_states
    -
    allowed_image_states
)

print(
    "Observed states:",
    sorted(observed_image_states)
)

print(
    "Unexpected states:",
    sorted(unexpected_states)
)

assert len(unexpected_states) == 0

image_state_counts = (
    work[
        "image_evidence_normalized"
    ]
    .value_counts()
)

print(
    "\nImage-level evidence counts:"
)

print(
    image_state_counts.to_string()
)

assert (
    int(
        image_state_counts.get(
            "NOT_INDIVIDUALLY_ADJUDICATED",
            0
        )
    )
    == 3744
)

assert (
    int(
        image_state_counts.get(
            "HUMAN_SEMANTIC_RETENTION",
            0
        )
    )
    == 95
)

assert (
    int(
        image_state_counts.get(
            "HUMAN_SEMANTIC_EXCLUSION",
            0
        )
    )
    == 15
)

assert (
    int(
        image_state_counts.get(
            "HUMAN_CONFLICT_RESOLVED",
            0
        )
    )
    == 5
)

assert image_state_counts.sum() == 3859

print("Image evidence states: PASS")


# ===================================================================================================
# 10. CONDITION-LEVEL CXR COUNTS
# ===================================================================================================

print("\n9. CONDITION-LEVEL CXR COUNTS")
print("-" * 100)

condition_counts = (
    work
    .groupby(
        "condition_id"
    )
    .size()
    .rename(
        "cxr_record_count"
    )
)

assert len(condition_counts) == 3081

assert condition_counts.sum() == 3859

condition_counts_distribution = (
    condition_counts
    .value_counts()
    .sort_index()
)

print(
    condition_counts_distribution.to_string()
)

multiple_cxr_condition_count = int(
    (
        condition_counts > 1
    ).sum()
)

print()
print(
    "Single-CXR conditions:",
    int(
        (
            condition_counts == 1
        ).sum()
    )
)

print(
    "Multiple-CXR conditions:",
    multiple_cxr_condition_count
)

assert multiple_cxr_condition_count == 485

print("Condition-level CXR count: PASS")


# ===================================================================================================
# 11. SEMANTIC EVIDENCE AGGREGATION
# ===================================================================================================

print("\n10. SEMANTIC EVIDENCE AGGREGATION")
print("-" * 100)

semantic_condition = (
    work
    .groupby(
        "condition_id"
    )
    .agg(
        semantic_reviewed_images=(
            "semantic_reviewed_bool",
            "sum"
        ),
        semantic_retained_images=(
            "image_evidence_normalized",
            lambda x: int(
                x.isin(
                    [
                        "HUMAN_SEMANTIC_RETENTION",
                        "HUMAN_CONFLICT_RESOLVED"
                    ]
                ).sum()
            )
        ),
        semantic_excluded_images=(
            "image_evidence_normalized",
            lambda x: int(
                (
                    x
                    ==
                    "HUMAN_SEMANTIC_EXCLUSION"
                ).sum()
            )
        ),
        conflict_resolved_images=(
            "image_evidence_normalized",
            lambda x: int(
                (
                    x
                    ==
                    "HUMAN_CONFLICT_RESOLVED"
                ).sum()
            )
        ),
    )
)

assert len(semantic_condition) == 3081

assert (
    semantic_condition[
        "semantic_reviewed_images"
    ].sum()
    == 115
)

assert (
    semantic_condition[
        "conflict_resolved_images"
    ].sum()
    == 5
)

print(
    "Semantic-reviewed images:",
    int(
        semantic_condition[
            "semantic_reviewed_images"
        ].sum()
    )
)

print(
    "Semantic retained evidence:",
    int(
        semantic_condition[
            "semantic_retained_images"
        ].sum()
    )
)

print(
    "Semantic excluded evidence:",
    int(
        semantic_condition[
            "semantic_excluded_images"
        ].sum()
    )
)

print(
    "Conflict-resolved images:",
    int(
        semantic_condition[
            "conflict_resolved_images"
        ].sum()
    )
)

print("Semantic aggregation: PASS")


# ===================================================================================================
# 12. STRUCTURAL EVIDENCE AGGREGATION
# ===================================================================================================

print("\n11. STRUCTURAL EVIDENCE AGGREGATION")
print("-" * 100)

structural_condition = (
    work
    .groupby(
        "condition_id"
    )
    .agg(
        structural_reviewed_images=(
            "structural_reviewed_bool",
            "sum"
        ),
        structural_retain_images=(
            "structural_decision_normalized",
            lambda x: int(
                (
                    x
                    ==
                    "RETAIN"
                ).sum()
            )
        ),
        structural_exclude_images=(
            "structural_decision_normalized",
            lambda x: int(
                (
                    x
                    ==
                    "EXCLUDE"
                ).sum()
            )
        ),
    )
)

assert len(structural_condition) == 3081

assert (
    structural_condition[
        "structural_reviewed_images"
    ].sum()
    == 142
)

print(
    "Structural-reviewed images:",
    int(
        structural_condition[
            "structural_reviewed_images"
        ].sum()
    )
)

print(
    "Structural RETAIN evidence:",
    int(
        structural_condition[
            "structural_retain_images"
        ].sum()
    )
)

print(
    "Structural EXCLUDE evidence:",
    int(
        structural_condition[
            "structural_exclude_images"
        ].sum()
    )
)

print("Structural aggregation: PASS")


# ===================================================================================================
# 13. IMAGE-LEVEL EVIDENCE-STATE SET PER CONDITION
# ===================================================================================================

print("\n12. IMAGE-LEVEL EVIDENCE STATES PER CONDITION")
print("-" * 100)

condition_state_sets = (
    work
    .groupby(
        "condition_id"
    )[
        "image_evidence_normalized"
    ]
    .agg(
        lambda x: tuple(
            sorted(
                set(
                    x
                    .dropna()
                    .astype(str)
                )
            )
        )
    )
)

assert len(condition_state_sets) == 3081

condition_state_distribution = (
    condition_state_sets
    .value_counts()
)

print(
    condition_state_distribution
    .to_string()
)

print("Condition evidence-state construction: PASS")


# ===================================================================================================
# 14. CONDITION-LEVEL EVIDENCE CLASSIFICATION
# ===================================================================================================
#
# IMPORTANT:
# This is an evidence classification only.
# It is NOT the final CXR eligibility decision.
#
# ===================================================================================================

print("\n13. CONDITION-LEVEL EVIDENCE CLASSIFICATION")
print("-" * 100)


def classify_condition_evidence(row):

    semantic_reviewed = (
        row[
            "semantic_reviewed_images"
        ] > 0
    )

    semantic_retained = (
        row[
            "semantic_retained_images"
        ] > 0
    )

    semantic_excluded = (
        row[
            "semantic_excluded_images"
        ] > 0
    )

    structural_reviewed = (
        row[
            "structural_reviewed_images"
        ] > 0
    )

    if (
        semantic_excluded
        and
        semantic_retained
    ):
        return (
            "SEMANTIC_MIXED_RETAIN_AND_EXCLUDE"
        )

    if (
        semantic_excluded
        and
        structural_reviewed
    ):
        return (
            "SEMANTIC_EXCLUSION_WITH_STRUCTURAL_EVIDENCE"
        )

    if semantic_excluded:
        return (
            "SEMANTIC_EXCLUSION_ONLY"
        )

    if (
        semantic_retained
        and
        structural_reviewed
    ):
        return (
            "SEMANTIC_RETENTION_WITH_STRUCTURAL_EVIDENCE"
        )

    if semantic_retained:
        return (
            "SEMANTIC_RETENTION_ONLY"
        )

    if structural_reviewed:
        return (
            "STRUCTURAL_EVIDENCE_ONLY"
        )

    return (
        "NO_HUMAN_ELIGIBILITY_REVIEW"
    )


condition_evidence = (
    condition_audit
    .set_index(
        "condition_id"
    )
    .copy()
)

condition_evidence = (
    condition_evidence
    .join(
        semantic_condition,
        how="left"
    )
    .join(
        structural_condition,
        how="left"
    )
)

assert len(condition_evidence) == 3081

condition_evidence[
    "condition_evidence_class"
] = condition_evidence.apply(
    classify_condition_evidence,
    axis=1
)

condition_class_counts = (
    condition_evidence[
        "condition_evidence_class"
    ]
    .value_counts()
)

print(
    condition_class_counts.to_string()
)

assert (
    condition_class_counts.sum()
    == 3081
)

print("Condition-level evidence classification: PASS")


# ===================================================================================================
# 15. MIXED SEMANTIC EVIDENCE
# ===================================================================================================

print("\n14. MIXED SEMANTIC EVIDENCE CONDITIONS")
print("-" * 100)

mixed_semantic = condition_evidence[
    (
        condition_evidence[
            "semantic_retained_images"
        ]
        > 0
    )
    &
    (
        condition_evidence[
            "semantic_excluded_images"
        ]
        > 0
    )
].copy()

mixed_semantic_count = len(
    mixed_semantic
)

print(
    "Conditions with both semantic RETAIN and EXCLUDE images:",
    mixed_semantic_count
)

# This value is derived directly from the authoritative evidence.
# It is deliberately NOT hard-coded.

assert (
    mixed_semantic_count
    ==
    int(
        (
            (
                condition_evidence[
                    "semantic_retained_images"
                ]
                > 0
            )
            &
            (
                condition_evidence[
                    "semantic_excluded_images"
                ]
                > 0
            )
        ).sum()
    )
)

print("Mixed semantic evidence: PASS")


# ===================================================================================================
# 16. RETAINED + UNREVIEWED CONDITIONS
# ===================================================================================================

print("\n15. RETAINED + UNREVIEWED CONDITIONS")
print("-" * 100)

retained_plus_unreviewed = (
    condition_state_sets
    .apply(
        lambda states:
        (
            "HUMAN_SEMANTIC_RETENTION"
            in states
            and
            "NOT_INDIVIDUALLY_ADJUDICATED"
            in states
        )
    )
)

retained_plus_unreviewed_count = int(
    retained_plus_unreviewed.sum()
)

print(
    "Conditions with retained + unreviewed images:",
    retained_plus_unreviewed_count
)

assert (
    retained_plus_unreviewed_count
    >= 0
)

assert (
    retained_plus_unreviewed_count
    <= 3081
)

print("Retained + unreviewed audit: PASS")


# ===================================================================================================
# 17. EXCLUDED + UNREVIEWED CONDITIONS
# ===================================================================================================

print("\n16. EXCLUDED + UNREVIEWED CONDITIONS")
print("-" * 100)

excluded_plus_unreviewed = (
    condition_state_sets
    .apply(
        lambda states:
        (
            "HUMAN_SEMANTIC_EXCLUSION"
            in states
            and
            "NOT_INDIVIDUALLY_ADJUDICATED"
            in states
        )
    )
)

excluded_plus_unreviewed_count = int(
    excluded_plus_unreviewed.sum()
)

print(
    "Conditions with excluded + unreviewed images:",
    excluded_plus_unreviewed_count
)

assert (
    excluded_plus_unreviewed_count
    >= 0
)

assert (
    excluded_plus_unreviewed_count
    <= 3081
)

print("Excluded + unreviewed audit: PASS")


# ===================================================================================================
# 18. RESOLVED-CONFLICT CONDITIONS
# ===================================================================================================

print("\n17. RESOLVED-CONFLICT CONDITION AUDIT")
print("-" * 100)

conflict_condition_ids = set(
    work.loc[
        work[
            "image_evidence_normalized"
        ]
        ==
        "HUMAN_CONFLICT_RESOLVED",
        "condition_id"
    ]
)

conflict_condition_count = len(
    conflict_condition_ids
)

conflict_image_count = int(
    (
        work[
            "image_evidence_normalized"
        ]
        ==
        "HUMAN_CONFLICT_RESOLVED"
    ).sum()
)

print(
    "Resolved-conflict images:",
    conflict_image_count
)

print(
    "Conditions containing resolved-conflict images:",
    conflict_condition_count
)

assert conflict_image_count == 5

assert (
    conflict_condition_count
    <= conflict_image_count
)

assert conflict_condition_count > 0

print("Resolved-conflict condition audit: PASS")


# ===================================================================================================
# 19. MULTIPLE-CXR CONDITIONS WITH NO HUMAN REVIEW
# ===================================================================================================

print("\n18. MULTIPLE-CXR CONDITIONS WITHOUT HUMAN REVIEW")
print("-" * 100)

multiple_cxr_no_review = condition_evidence[
    (
        condition_evidence[
            "has_multiple_cxr"
        ]
        &
        (
            condition_evidence[
                "semantic_reviewed_images"
            ]
            == 0
        )
        &
        (
            condition_evidence[
                "structural_reviewed_images"
            ]
            == 0
        )
    )
].copy()

multiple_cxr_no_review_count = len(
    multiple_cxr_no_review
)

print(
    "Multiple-CXR conditions with no human review:",
    multiple_cxr_no_review_count
)

assert (
    multiple_cxr_no_review_count
    <= 485
)

print("Multiple-CXR no-review audit: PASS")


# ===================================================================================================
# 20. CONDITIONS WITH EXPLICIT SEMANTIC EXCLUSION
# ===================================================================================================

print("\n19. CONDITIONS WITH EXPLICIT SEMANTIC EXCLUSION")
print("-" * 100)

semantic_exclusion_conditions = condition_evidence[
    condition_evidence[
        "semantic_excluded_images"
    ]
    > 0
].copy()

semantic_exclusion_condition_count = len(
    semantic_exclusion_conditions
)

print(
    "Conditions containing semantic exclusion evidence:",
    semantic_exclusion_condition_count
)

assert (
    semantic_exclusion_condition_count
    <= 3081
)

print(
    "This is evidence only; no automatic condition-level exclusion."
)


# ===================================================================================================
# 21. CONDITIONS WITH EXPLICIT SEMANTIC RETENTION
# ===================================================================================================

print("\n20. CONDITIONS WITH EXPLICIT SEMANTIC RETENTION")
print("-" * 100)

semantic_retention_conditions = condition_evidence[
    condition_evidence[
        "semantic_retained_images"
    ]
    > 0
].copy()

semantic_retention_condition_count = len(
    semantic_retention_conditions
)

print(
    "Conditions containing semantic retention evidence:",
    semantic_retention_condition_count
)

assert (
    semantic_retention_condition_count
    <= 3081
)

print(
    "This is evidence only; no automatic condition-level retention."
)


# ===================================================================================================
# 22. CONDITION-LEVEL IMAGE COUNT RECONCILIATION
# ===================================================================================================

print("\n21. CONDITION-LEVEL IMAGE COUNT RECONCILIATION")
print("-" * 100)

condition_audit_count_map = (
    condition_audit
    .set_index(
        "condition_id"
    )[
        "cxr_record_count"
    ]
)

count_mismatch = (
    condition_counts
    !=
    condition_audit_count_map
)

print(
    "Condition image-count mismatches:",
    int(count_mismatch.sum())
)

assert count_mismatch.sum() == 0

print("Condition image-count reconciliation: PASS")


# ===================================================================================================
# 23. CONDITION-LEVEL MULTIPLE-CXR / REVIEW INTERACTION
# ===================================================================================================

print("\n22. MULTIPLE-CXR × HUMAN-REVIEW INTERACTION")
print("-" * 100)

multiple_cxr_review_table = (
    condition_evidence[
        [
            "cxr_record_count",
            "semantic_reviewed_images",
            "structural_reviewed_images",
            "condition_evidence_class",
        ]
    ]
    .copy()
)

multiple_cxr_review_table[
    "multiple_cxr"
] = (
    multiple_cxr_review_table[
        "cxr_record_count"
    ]
    > 1
)

multiple_cxr_review_distribution = (
    multiple_cxr_review_table[
        multiple_cxr_review_table[
            "multiple_cxr"
        ]
    ][
        "condition_evidence_class"
    ]
    .value_counts()
)

print(
    "Evidence classes among multiple-CXR conditions:"
)

print(
    multiple_cxr_review_distribution.to_string()
)

print("Multiple-CXR review interaction: PASS")


# ===================================================================================================
# 24. NO FINAL ELIGIBILITY DECISION
# ===================================================================================================

print("\n23. FINAL ELIGIBILITY DECISION")
print("-" * 100)

condition_evidence[
    "condition_level_cxr_eligibility"
] = "NOT_YET_ADJUDICATED"

condition_evidence[
    "final_research_cohort_status"
] = "NOT_YET_SELECTED"

assert (
    condition_evidence[
        "condition_level_cxr_eligibility"
    ]
    .eq(
        "NOT_YET_ADJUDICATED"
    )
    .all()
)

assert (
    condition_evidence[
        "final_research_cohort_status"
    ]
    .eq(
        "NOT_YET_SELECTED"
    )
    .all()
)

print(
    "Condition-level CXR eligibility:",
    "NOT_YET_ADJUDICATED"
)

print(
    "Final research cohort:",
    "NOT_YET_SELECTED"
)


# ===================================================================================================
# 25. SOURCE PROTECTION
# ===================================================================================================

print("\n24. SOURCE PROTECTION")
print("-" * 100)

print(
    "Raw DICOM modification              : NO"
)

print(
    "Source metadata modification        : NO"
)

print(
    "Human-review artifacts modification : NO"
)

print(
    "Additional ZIP extraction           : NO"
)

print(
    "Source protection: PASS"
)


# ===================================================================================================
# 26. FINAL ACCOUNTING
# ===================================================================================================

print("\n25. FINAL CONDITION-LEVEL ACCOUNTING")
print("-" * 100)

assert len(condition_evidence) == 3081

assert (
    condition_evidence[
        "cxr_record_count"
    ].sum()
    == 3859
)

assert (
    condition_evidence[
        "semantic_reviewed_images"
    ].sum()
    == 115
)

assert (
    condition_evidence[
        "structural_reviewed_images"
    ].sum()
    == 142
)

assert (
    condition_evidence[
        "has_multiple_cxr"
    ].sum()
    == 485
)

print(
    "CXR records represented       : 3,859"
)

print(
    "Conditions represented        : 3,081"
)

print(
    "Semantic reviewed images      : 115"
)

print(
    "Structural reviewed images    : 142"
)

print(
    "Multiple-CXR conditions       : 485"
)

print(
    "Condition-level accounting    : PASS"
)


# ===================================================================================================
# 27. SAVE OUTPUTS
# ===================================================================================================

print("\n26. SAVING OUTPUTS")
print("-" * 100)

CONDITION_EVIDENCE_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_32_Condition_Level_Eligibility_Evidence.csv"
)

CLASS_SUMMARY_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_32_Condition_Evidence_Class_Summary.csv"
)

MIXED_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_32_Mixed_Semantic_Evidence_Conditions.csv"
)

MULTI_NO_REVIEW_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_32_Multiple_CXR_No_Human_Review.csv"
)

CONFLICT_CONDITION_PATH = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_32_Resolved_Conflict_Conditions.csv"
)


condition_evidence_to_save = (
    condition_evidence
    .reset_index()
)

condition_evidence_to_save.to_csv(
    CONDITION_EVIDENCE_PATH,
    index=False
)


class_summary = (
    condition_evidence[
        "condition_evidence_class"
    ]
    .value_counts()
    .rename_axis(
        "condition_evidence_class"
    )
    .reset_index(
        name="condition_count"
    )
)

class_summary.to_csv(
    CLASS_SUMMARY_PATH,
    index=False
)


mixed_semantic.reset_index().to_csv(
    MIXED_PATH,
    index=False
)


multiple_cxr_no_review.reset_index().to_csv(
    MULTI_NO_REVIEW_PATH,
    index=False
)


conflict_condition_table = (
    condition_evidence
    .loc[
        condition_evidence.index.isin(
            conflict_condition_ids
        )
    ]
    .reset_index()
)

conflict_condition_table.to_csv(
    CONFLICT_CONDITION_PATH,
    index=False
)


# ===================================================================================================
# 28. OUTPUT VALIDATION
# ===================================================================================================

print("\n27. OUTPUT FILE VALIDATION")
print("-" * 100)

output_paths = [
    CONDITION_EVIDENCE_PATH,
    CLASS_SUMMARY_PATH,
    MIXED_PATH,
    MULTI_NO_REVIEW_PATH,
    CONFLICT_CONDITION_PATH,
]

for output_path in output_paths:

    assert output_path.exists(), (
        f"Output not created: {output_path}"
    )

    assert output_path.stat().st_size > 0, (
        f"Output file is empty: {output_path}"
    )

    print(
        f"{output_path.name}: PASS"
    )


# ===================================================================================================
# 29. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.32 — FINAL STATUS")
print("=" * 100)

print(
    "3,859 CXR records represented       : PASS"
)

print(
    "3,081 conditions represented        : PASS"
)

print(
    "115 semantic-reviewed images        : PASS"
)

print(
    "142 structural-reviewed images      : PASS"
)

print(
    "485 multiple-CXR conditions         : PASS"
)

print(
    f"{mixed_semantic_count}"
    " mixed semantic conditions"
    "          : PASS"
)

print(
    f"{retained_plus_unreviewed_count}"
    " retained + unreviewed conditions"
    " : PASS"
)

print(
    f"{excluded_plus_unreviewed_count}"
    " excluded + unreviewed conditions"
    " : PASS"
)

print(
    f"{conflict_condition_count}"
    " conditions containing resolved-conflict evidence"
    " : PASS"
)

print(
    "cxr_outlier automatic exclusion     : NO"
)

print(
    "Automatic modality exclusion        : NO"
)

print(
    "Final condition eligibility         : NOT_YET_ADJUDICATED"
)

print(
    "Final research cohort selected      : NO"
)

print(
    "Final target selected               : NO"
)

print(
    "Additional ZIP extraction           : NO"
)

print(
    "DICOM files modified                : NO"
)

print(
    "Source evidence modified            : NO"
)

print(
    "Output files                         : PASS"
)

print("\n" + "=" * 100)
print("STEP 3B.6.32 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.33 — MULTIPLE-CXR CONDITION ANALYSIS
# SELF-CONTAINED CORRECTED FINAL VERSION
# ===================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("STEP 3B.6.33 — MULTIPLE-CXR CONDITION ANALYSIS")
print("=" * 100)


# ===================================================================================================
# 1. INITIALIZATION
# ===================================================================================================

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_030_DIR = (
    METADATA_DIR /
    "Step_3B_6_30_Image_Level_CXR_Eligibility"
)

STEP_032_DIR = (
    METADATA_DIR /
    "Step_3B_6_32_Condition_Level_Eligibility_Evidence"
)

IMAGE_AUDIT_PATH = (
    STEP_030_DIR /
    "TB_Portals_March2025_Step_3B_6_30_Image_Level_CXR_Eligibility_Audit.csv"
)

CONDITION_EVIDENCE_PATH = (
    STEP_032_DIR /
    "TB_Portals_March2025_Step_3B_6_32_Condition_Level_Eligibility_Evidence.csv"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_33_Multiple_CXR_Condition_Analysis"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ===================================================================================================
# 2. INPUT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

assert IMAGE_AUDIT_PATH.exists()
assert CONDITION_EVIDENCE_PATH.exists()

print("Step 3B.6.30 image audit       : FOUND")
print("Step 3B.6.32 condition evidence: FOUND")
print("Input validation                : PASS")


# ===================================================================================================
# 3. LOAD
# ===================================================================================================

image_audit = pd.read_csv(
    IMAGE_AUDIT_PATH,
    low_memory=False
)

condition_evidence = pd.read_csv(
    CONDITION_EVIDENCE_PATH,
    low_memory=False
)

print("\n2. LOADED ARTIFACTS")
print("-" * 100)

print("Image audit records      :", len(image_audit))
print("Condition evidence rows :", len(condition_evidence))


# ===================================================================================================
# 4. REQUIRED COLUMNS
# ===================================================================================================

image_required = [
    "condition_id",
    "series_instance_content_url",
    "imaging_date",
    "series_modality_cd",
    "cxr_outlier",
    "image_level_evidence_status",
    "semantic_reviewed",
    "structural_reviewed",
]

condition_required = [
    "condition_id",
    "cxr_record_count",
    "has_multiple_cxr",
    "condition_evidence_class",
]

missing_image = [
    c for c in image_required
    if c not in image_audit.columns
]

missing_condition = [
    c for c in condition_required
    if c not in condition_evidence.columns
]

print("\n3. REQUIRED COLUMN VALIDATION")
print("-" * 100)

print("Missing image columns    :", missing_image)
print("Missing condition columns:", missing_condition)

assert len(missing_image) == 0
assert len(missing_condition) == 0

print("Required columns: PASS")


# ===================================================================================================
# 5. POPULATION VALIDATION
# ===================================================================================================

print("\n4. POPULATION VALIDATION")
print("-" * 100)

assert len(image_audit) == 3859
assert len(condition_evidence) == 3081

assert (
    image_audit["series_instance_content_url"].nunique()
    == 3859
)

assert (
    image_audit["condition_id"].nunique()
    == 3081
)

assert (
    condition_evidence["condition_id"].nunique()
    == 3081
)

print("CXR records       : 3,859")
print("Unique URLs       : 3,859")
print("Unique conditions : 3,081")
print("Population validation: PASS")


# ===================================================================================================
# 6. BUILD FRESH IMAGE WORKING TABLE
# ===================================================================================================

print("\n5. BUILDING IMAGE WORKING TABLE")
print("-" * 100)

work = image_audit[
    [
        "condition_id",
        "series_instance_content_url",
        "imaging_date",
        "series_modality_cd",
        "cxr_outlier",
        "image_level_evidence_status",
        "semantic_reviewed",
        "structural_reviewed",
    ]
].copy()

work["condition_id"] = (
    work["condition_id"]
    .astype(str)
    .str.strip()
)

work["series_instance_content_url"] = (
    work["series_instance_content_url"]
    .astype(str)
    .str.strip()
)

work["imaging_date_num"] = pd.to_numeric(
    work["imaging_date"],
    errors="coerce"
)

work["series_modality_cd_normalized"] = (
    work["series_modality_cd"]
    .fillna("MISSING")
    .astype(str)
    .str.strip()
    .str.upper()
)

work["image_evidence_normalized"] = (
    work["image_level_evidence_status"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
)

work["semantic_reviewed_bool"] = (
    work["semantic_reviewed"]
    .fillna(False)
    .astype(bool)
)

work["structural_reviewed_bool"] = (
    work["structural_reviewed"]
    .fillna(False)
    .astype(bool)
)

assert len(work) == 3859

print("Working image records:", len(work))
print("Working table: PASS")


# ===================================================================================================
# 7. CONDITION CXR COUNTS
# ===================================================================================================

print("\n6. CONDITION-LEVEL CXR COUNTS")
print("-" * 100)

condition_counts = (
    work
    .groupby("condition_id")
    .size()
    .rename("actual_cxr_record_count")
)

assert len(condition_counts) == 3081
assert condition_counts.sum() == 3859

multiple_condition_ids = set(
    condition_counts[
        condition_counts > 1
    ].index
)

assert len(multiple_condition_ids) == 485

print(
    "Multiple-CXR conditions:",
    len(multiple_condition_ids)
)

print("Multiple-CXR population: PASS")


# ===================================================================================================
# 8. MULTIPLE-CXR IMAGE TABLE
# ===================================================================================================

print("\n7. MULTIPLE-CXR IMAGE TABLE")
print("-" * 100)

multiple_cxr_images = (
    work[
        work["condition_id"].isin(
            multiple_condition_ids
        )
    ]
    .copy()
)

assert (
    multiple_cxr_images["condition_id"].nunique()
    == 485
)

print(
    "Multiple-CXR image records:",
    len(multiple_cxr_images)
)

print(
    "Multiple-CXR conditions:",
    multiple_cxr_images["condition_id"].nunique()
)

assert len(multiple_cxr_images) == 1263

print("Multiple-CXR image table: PASS")


# ===================================================================================================
# 9. CXR COUNT DISTRIBUTION
# ===================================================================================================

print("\n8. NUMBER OF CXRs PER MULTIPLE-CXR CONDITION")
print("-" * 100)

multiple_cxr_count_distribution = (
    condition_counts[
        condition_counts > 1
    ]
    .value_counts()
    .sort_index()
)

print(
    multiple_cxr_count_distribution.to_string()
)

assert (
    multiple_cxr_count_distribution.sum()
    == 485
)

print("CXR-count distribution: PASS")


# ===================================================================================================
# 10. DATE STRUCTURE
# ===================================================================================================

print("\n9. IMAGING DATE STRUCTURE")
print("-" * 100)

date_summary = (
    multiple_cxr_images
    .groupby("condition_id")
    .agg(
        multiple_cxr_unique_dates=(
            "imaging_date_num",
            "nunique"
        ),
        multiple_cxr_min_date=(
            "imaging_date_num",
            "min"
        ),
        multiple_cxr_max_date=(
            "imaging_date_num",
            "max"
        ),
    )
)

date_summary[
    "multiple_cxr_imaging_span"
] = (
    date_summary["multiple_cxr_max_date"]
    -
    date_summary["multiple_cxr_min_date"]
)

assert len(date_summary) == 485

print(
    "Unique imaging-date distribution:"
)

print(
    date_summary[
        "multiple_cxr_unique_dates"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)

print(
    "\nImaging span summary:"
)

print(
    date_summary[
        "multiple_cxr_imaging_span"
    ]
    .describe()
    .to_string()
)

print("Imaging-date structure: PASS")


# ===================================================================================================
# 11. SAME-DAY MULTIPLE CXRs
# ===================================================================================================

print("\n10. SAME-DAY MULTIPLE-CXR ANALYSIS")
print("-" * 100)

condition_date_counts = (
    multiple_cxr_images
    .groupby(
        [
            "condition_id",
            "imaging_date_num"
        ]
    )
    .size()
    .rename(
        "same_condition_date_cxr_count"
    )
    .reset_index()
)

same_day_groups = (
    condition_date_counts[
        condition_date_counts[
            "same_condition_date_cxr_count"
        ] > 1
    ]
    .copy()
)

same_day_condition_ids = set(
    same_day_groups["condition_id"]
)

same_day_images = (
    multiple_cxr_images
    .merge(
        same_day_groups[
            [
                "condition_id",
                "imaging_date_num"
            ]
        ],
        on=[
            "condition_id",
            "imaging_date_num"
        ],
        how="inner"
        ,
        validate="many_to_one"
    )
)

print(
    "Condition-date groups with >1 CXR:",
    len(same_day_groups)
)

print(
    "Conditions with same-day multiple CXRs:",
    len(same_day_condition_ids)
)

print(
    "CXR records in those groups:",
    len(same_day_images)
)

assert len(same_day_groups) == 58
assert len(same_day_images) == 141

print("Same-day multiple-CXR analysis: PASS")


# ===================================================================================================
# 12. LONGITUDINAL CONDITIONS
# ===================================================================================================

print("\n11. LONGITUDINAL CXR CONDITIONS")
print("-" * 100)

longitudinal_conditions = (
    date_summary[
        date_summary[
            "multiple_cxr_unique_dates"
        ] > 1
    ]
    .copy()
)

print(
    "Multiple-CXR conditions with >1 imaging date:",
    len(longitudinal_conditions)
)

assert len(longitudinal_conditions) == 459

print("Longitudinal CXR analysis: PASS")


# ===================================================================================================
# 13. MODALITY SIGNATURE
# ===================================================================================================

print("\n12. MODALITY STRUCTURE")
print("-" * 100)

def modality_signature(values):

    cleaned = (
        values
        .dropna()
        .astype(str)
        .str.strip()
        .str.upper()
    )

    return "|".join(
        sorted(
            set(cleaned)
        )
    )


modality_summary = (
    multiple_cxr_images
    .groupby("condition_id")[
        "series_modality_cd_normalized"
    ]
    .apply(modality_signature)
    .rename(
        "multiple_cxr_modality_signature"
    )
)

assert len(modality_summary) == 485

print(
    modality_summary
    .value_counts()
    .to_string()
)

print("Modality structure: PASS")


# ===================================================================================================
# 14. HUMAN REVIEW COVERAGE
# ===================================================================================================

print("\n13. HUMAN-REVIEW COVERAGE")
print("-" * 100)

review_summary = (
    multiple_cxr_images
    .groupby("condition_id")
    .agg(
        multiple_cxr_semantic_reviewed_images=(
            "semantic_reviewed_bool",
            "sum"
        ),
        multiple_cxr_structural_reviewed_images=(
            "structural_reviewed_bool",
            "sum"
        ),
    )
)

review_summary[
    "multiple_cxr_has_semantic_review"
] = (
    review_summary[
        "multiple_cxr_semantic_reviewed_images"
    ] > 0
)

review_summary[
    "multiple_cxr_has_structural_review"
] = (
    review_summary[
        "multiple_cxr_structural_reviewed_images"
    ] > 0
)

review_summary[
    "multiple_cxr_has_any_human_review"
] = (
    review_summary[
        "multiple_cxr_has_semantic_review"
    ]
    |
    review_summary[
        "multiple_cxr_has_structural_review"
    ]
)

review_summary[
    "multiple_cxr_has_no_human_review"
] = (
    ~review_summary[
        "multiple_cxr_has_any_human_review"
    ]
)

assert len(review_summary) == 485

print(
    "Any human review:",
    int(
        review_summary[
            "multiple_cxr_has_any_human_review"
        ].sum()
    )
)

print(
    "Semantic review:",
    int(
        review_summary[
            "multiple_cxr_has_semantic_review"
        ].sum()
    )
)

print(
    "Structural review:",
    int(
        review_summary[
            "multiple_cxr_has_structural_review"
        ].sum()
    )
)

print(
    "No human review:",
    int(
        review_summary[
            "multiple_cxr_has_no_human_review"
        ].sum()
    )
)

assert (
    review_summary[
        "multiple_cxr_has_no_human_review"
    ].sum()
    == 419
)

print("Human-review coverage: PASS")


# ===================================================================================================
# 15. IMAGE EVIDENCE STATES
# ===================================================================================================

print("\n14. IMAGE-LEVEL EVIDENCE STATES")
print("-" * 100)

evidence_state_summary = (
    multiple_cxr_images
    .groupby("condition_id")[
        "image_evidence_normalized"
    ]
    .apply(
        lambda values:
        "|".join(
            sorted(
                set(values)
            )
        )
    )
    .rename(
        "multiple_cxr_evidence_state_signature"
    )
)

assert len(evidence_state_summary) == 485

print(
    evidence_state_summary
    .value_counts()
    .to_string()
)

print("Evidence-state analysis: PASS")


# ===================================================================================================
# 16. MIXED IMAGE EVIDENCE
# ===================================================================================================

print("\n15. MIXED IMAGE-LEVEL EVIDENCE")
print("-" * 100)

mixed_evidence_flags = (
    evidence_state_summary
    .str.contains(
        r"\|",
        regex=True
    )
)

mixed_evidence_count = int(
    mixed_evidence_flags.sum()
)

print(
    "Multiple-CXR conditions with >1 evidence state:",
    mixed_evidence_count
)

assert mixed_evidence_count == 19

print("Mixed evidence analysis: PASS")


# ===================================================================================================
# 17. SEMANTIC RETAIN + EXCLUDE
# ===================================================================================================

print("\n16. SEMANTIC RETAIN + EXCLUDE")
print("-" * 100)

semantic_state_flags = (
    multiple_cxr_images
    .groupby("condition_id")[
        "image_evidence_normalized"
    ]
    .apply(
        lambda values:
        (
            "HUMAN_SEMANTIC_RETENTION"
            in set(values)
        )
        and
        (
            "HUMAN_SEMANTIC_EXCLUSION"
            in set(values)
        )
    )
)

semantic_mixed_ids = set(
    semantic_state_flags[
        semantic_state_flags
    ].index
)

print(
    "Conditions containing semantic RETAIN + EXCLUDE:",
    len(semantic_mixed_ids)
)

assert len(semantic_mixed_ids) == 3

print("Semantic mixed-evidence analysis: PASS")


# ===================================================================================================
# 18. RETAINED + UNREVIEWED
# ===================================================================================================

print("\n17. RETAINED + UNREVIEWED IMAGE CONDITIONS")
print("-" * 100)

retained_unreviewed_flags = (
    multiple_cxr_images
    .groupby("condition_id")[
        "image_evidence_normalized"
    ]
    .apply(
        lambda values:
        (
            "HUMAN_SEMANTIC_RETENTION"
            in set(values)
        )
        and
        (
            "NOT_INDIVIDUALLY_ADJUDICATED"
            in set(values)
        )
    )
)

retained_unreviewed_ids = set(
    retained_unreviewed_flags[
        retained_unreviewed_flags
    ].index
)

print(
    "Conditions with retained + unreviewed images:",
    len(retained_unreviewed_ids)
)

assert len(retained_unreviewed_ids) == 13

print("Retained + unreviewed analysis: PASS")


# ===================================================================================================
# 19. EXCLUDED + UNREVIEWED
# ===================================================================================================

print("\n18. EXCLUDED + UNREVIEWED IMAGE CONDITIONS")
print("-" * 100)

excluded_unreviewed_flags = (
    multiple_cxr_images
    .groupby("condition_id")[
        "image_evidence_normalized"
    ]
    .apply(
        lambda values:
        (
            "HUMAN_SEMANTIC_EXCLUSION"
            in set(values)
        )
        and
        (
            "NOT_INDIVIDUALLY_ADJUDICATED"
            in set(values)
        )
    )
)

excluded_unreviewed_ids = set(
    excluded_unreviewed_flags[
        excluded_unreviewed_flags
    ].index
)

print(
    "Conditions with excluded + unreviewed images:",
    len(excluded_unreviewed_ids)
)

assert len(excluded_unreviewed_ids) == 2

print("Excluded + unreviewed analysis: PASS")


# ===================================================================================================
# 20. CONDITION EVIDENCE CLASS
# ===================================================================================================

print("\n19. CONDITION-LEVEL EVIDENCE CLASSES")
print("-" * 100)

condition_class = (
    condition_evidence[
        [
            "condition_id",
            "condition_evidence_class"
        ]
    ]
    .copy()
)

condition_class = condition_class[
    condition_class[
        "condition_id"
    ].isin(
        multiple_condition_ids
    )
]

assert len(condition_class) == 485

print(
    condition_class[
        "condition_evidence_class"
    ]
    .value_counts()
    .to_string()
)

print("Condition evidence class audit: PASS")


# ===================================================================================================
# 21. MULTIPLE-CXR / MULTIPLE-DATE INTERSECTION
# ===================================================================================================

print("\n20. MULTIPLE-CXR / MULTIPLE-DATE INTERSECTION")
print("-" * 100)

multiple_cxr_set = set(
    multiple_condition_ids
)

multiple_date_set = set(
    longitudinal_conditions.index
)

both_sets = (
    multiple_cxr_set
    &
    multiple_date_set
)

multiple_cxr_single_date = (
    multiple_cxr_set
    -
    multiple_date_set
)

print(
    "Multiple-CXR conditions:",
    len(multiple_cxr_set)
)

print(
    "Multiple-date conditions:",
    len(multiple_date_set)
)

print(
    "Both:",
    len(both_sets)
)

print(
    "Multiple-CXR but single imaging date:",
    len(multiple_cxr_single_date)
)

assert len(both_sets) == 459
assert len(multiple_cxr_single_date) == 26

print("Multiple-CXR/date intersection: PASS")


# ===================================================================================================
# 22. LONGITUDINAL SPAN CATEGORY
# ===================================================================================================

print("\n21. LONGITUDINAL SPAN CATEGORIES")
print("-" * 100)

span_table = date_summary.copy()

def span_category(days):

    if pd.isna(days):
        return "MISSING_DATE"

    if days == 0:
        return "SAME_DAY"

    if days <= 7:
        return "1_TO_7_DAYS"

    if days <= 30:
        return "8_TO_30_DAYS"

    if days <= 90:
        return "31_TO_90_DAYS"

    if days <= 180:
        return "91_TO_180_DAYS"

    if days <= 365:
        return "181_TO_365_DAYS"

    return "MORE_THAN_365_DAYS"


span_table[
    "multiple_cxr_longitudinal_span_category"
] = (
    span_table[
        "multiple_cxr_imaging_span"
    ]
    .apply(span_category)
)

print(
    span_table[
        "multiple_cxr_longitudinal_span_category"
    ]
    .value_counts()
    .to_string()
)

print("Longitudinal span categorization: PASS")


# ===================================================================================================
# 23. BUILD FINAL MULTIPLE-CXR CONDITION MASTER
# ===================================================================================================
#
# IMPORTANT:
# We explicitly avoid joining columns already present in condition_evidence.
# The date fields are renamed to unique names.
#
# ===================================================================================================

print("\n22. BUILDING MULTIPLE-CXR CONDITION MASTER")
print("-" * 100)

# Start with only the condition-level fields we actually need.
condition_base = condition_evidence[
    [
        "condition_id",
        "cxr_record_count",
        "has_multiple_cxr",
        "condition_modalities",
        "condition_cxr_outlier_states",
        "condition_evidence_class",
        "semantic_reviewed_images",
        "structural_reviewed_images",
    ]
].copy()

condition_base = condition_base[
    condition_base[
        "condition_id"
    ].isin(
        multiple_condition_ids
    )
].copy()

assert len(condition_base) == 485


# Date summary has intentionally unique column names.
date_for_merge = (
    date_summary
    .reset_index()
    [
        [
            "condition_id",
            "multiple_cxr_unique_dates",
            "multiple_cxr_min_date",
            "multiple_cxr_max_date",
            "multiple_cxr_imaging_span",
        ]
    ]
)

assert len(date_for_merge) == 485
assert date_for_merge["condition_id"].is_unique


# Modality summary.
modality_for_merge = (
    modality_summary
    .reset_index()
)

assert len(modality_for_merge) == 485
assert modality_for_merge["condition_id"].is_unique


# Review summary.
review_for_merge = (
    review_summary
    .reset_index()
)

assert len(review_for_merge) == 485
assert review_for_merge["condition_id"].is_unique


# Evidence states.
evidence_for_merge = (
    evidence_state_summary
    .reset_index()
)

assert len(evidence_for_merge) == 485
assert evidence_for_merge["condition_id"].is_unique


# Span category.
span_for_merge = (
    span_table[
        [
            "multiple_cxr_longitudinal_span_category"
        ]
    ]
    .copy()
)

span_for_merge = (
    span_for_merge
    .reset_index()
)

assert len(span_for_merge) == 485
assert span_for_merge["condition_id"].is_unique


# Build using explicit one-to-one merges.
multiple_condition_master = (
    condition_base
    .merge(
        date_for_merge,
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        modality_for_merge,
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        review_for_merge,
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        evidence_for_merge,
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        span_for_merge,
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
)

assert len(multiple_condition_master) == 485
assert (
    multiple_condition_master[
        "condition_id"
    ].nunique()
    == 485
)

multiple_condition_master[
    "has_same_day_multiple_cxr"
] = (
    multiple_condition_master[
        "condition_id"
    ]
    .isin(
        same_day_condition_ids
    )
)

multiple_condition_master[
    "has_semantic_retain_and_exclude"
] = (
    multiple_condition_master[
        "condition_id"
    ]
    .isin(
        semantic_mixed_ids
    )
)

multiple_condition_master[
    "has_retained_and_unreviewed"
] = (
    multiple_condition_master[
        "condition_id"
    ]
    .isin(
        retained_unreviewed_ids
    )
)

multiple_condition_master[
    "has_excluded_and_unreviewed"
] = (
    multiple_condition_master[
        "condition_id"
    ]
    .isin(
        excluded_unreviewed_ids
    )
)

multiple_condition_master[
    "has_no_human_review"
] = (
    multiple_condition_master[
        "condition_id"
    ]
    .isin(
        review_summary[
            review_summary[
                "multiple_cxr_has_no_human_review"
            ]
        ]
        .index
    )
)

print(
    "Multiple-CXR condition records:",
    len(multiple_condition_master)
)

print("Multiple-CXR condition master: PASS")


# ===================================================================================================
# 24. CRITICAL METHODOLOGICAL SAFETY CHECKS
# ===================================================================================================

print("\n23. CRITICAL METHODOLOGICAL SAFETY CHECKS")
print("-" * 100)

print("Automatic modality exclusion       : NO")
print("Automatic cxr_outlier exclusion    : NO")
print("Representative CXR selected        : NO")
print("Image discarded                     : NO")
print("Final eligibility selected          : NO")
print("Final research cohort selected      : NO")
print("Target selected                     : NO")
print("Preprocessing performed             : NO")
print("Additional ZIP extraction           : NO")
print("DICOM modified                      : NO")

print("Safety checks: PASS")


# ===================================================================================================
# 25. FINAL ACCOUNTING
# ===================================================================================================

print("\n24. FINAL ACCOUNTING")
print("-" * 100)

assert len(multiple_condition_master) == 485

assert (
    multiple_cxr_images[
        "condition_id"
    ].nunique()
    == 485
)

assert (
    len(same_day_groups)
    == 58
)

assert (
    len(same_day_images)
    == 141
)

assert (
    len(longitudinal_conditions)
    == 459
)

assert (
    len(multiple_cxr_single_date)
    == 26
)

print("Multiple-CXR conditions          : 485")
print("Multiple-date conditions         : 459")
print("Same-day multiple-CXR groups     : 58")
print("Same-day CXR records involved    : 141")
print("Multiple-CXR single-date cases   : 26")
print("Final accounting: PASS")


# ===================================================================================================
# 26. SAVE OUTPUTS
# ===================================================================================================

print("\n25. SAVING OUTPUTS")
print("-" * 100)

MASTER_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_33_Multiple_CXR_Condition_Master.csv"
)

COUNT_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_33_Multiple_CXR_Count_Distribution.csv"
)

SAME_DAY_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_33_Same_Day_Multiple_CXR_Groups.csv"
)

LONGITUDINAL_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_33_Longitudinal_CXR_Conditions.csv"
)

MODALITY_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_33_Multiple_CXR_Modality_Distribution.csv"
)

EVIDENCE_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_33_Multiple_CXR_Evidence_State_Distribution.csv"
)

NO_REVIEW_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_33_Multiple_CXR_No_Human_Review.csv"
)


multiple_condition_master.to_csv(
    MASTER_OUTPUT,
    index=False
)

(
    multiple_cxr_count_distribution
    .rename("condition_count")
    .rename_axis("cxr_record_count")
    .reset_index()
    .to_csv(
        COUNT_OUTPUT,
        index=False
    )
)

same_day_groups.to_csv(
    SAME_DAY_OUTPUT,
    index=False
)

longitudinal_conditions.reset_index().to_csv(
    LONGITUDINAL_OUTPUT,
    index=False
)

(
    modality_distribution
    if "modality_distribution" in globals()
    else modality_summary.value_counts()
).rename(
    "condition_count"
).rename_axis(
    "condition_modality_signature"
).reset_index().to_csv(
    MODALITY_OUTPUT,
    index=False
)

(
    evidence_state_summary
    .value_counts()
    .rename("condition_count")
    .rename_axis("image_evidence_state_set")
    .reset_index()
    .to_csv(
        EVIDENCE_OUTPUT,
        index=False
    )
)

multiple_condition_master[
    multiple_condition_master[
        "has_no_human_review"
    ]
].to_csv(
    NO_REVIEW_OUTPUT,
    index=False
)


# ===================================================================================================
# 27. OUTPUT VALIDATION
# ===================================================================================================

print("\n26. OUTPUT FILE VALIDATION")
print("-" * 100)

output_paths = [
    MASTER_OUTPUT,
    COUNT_OUTPUT,
    SAME_DAY_OUTPUT,
    LONGITUDINAL_OUTPUT,
    MODALITY_OUTPUT,
    EVIDENCE_OUTPUT,
    NO_REVIEW_OUTPUT,
]

for path in output_paths:

    assert path.exists(), (
        f"Missing output: {path}"
    )

    assert path.stat().st_size > 0, (
        f"Empty output: {path}"
    )

    print(
        f"{path.name}: PASS"
    )


# ===================================================================================================
# 28. FINAL SOURCE PROTECTION
# ===================================================================================================

print("\n27. SOURCE PROTECTION")
print("-" * 100)

print("Raw DICOM modification              : NO")
print("Source metadata modification        : NO")
print("Human-review artifact modification  : NO")
print("Additional ZIP extraction           : NO")
print("Image preprocessing                 : NO")
print("Representative image selection      : NO")
print("Final cohort selection              : NO")

print("Source protection: PASS")


# ===================================================================================================
# 29. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.33 — FINAL STATUS")
print("=" * 100)

print("3,859 CXR records represented       : PASS")
print("3,081 conditions represented        : PASS")
print("485 multiple-CXR conditions         : PASS")
print("459 multiple-date conditions        : PASS")
print("58 same-day multiple-CXR groups     : PASS")
print("141 same-day CXR records            : PASS")
print("26 multiple-CXR single-date cases   : PASS")

print(
    f"{mixed_evidence_count}"
    " multiple-CXR conditions with mixed image evidence : PASS"
)

print(
    f"{len(semantic_mixed_ids)}"
    " conditions with semantic RETAIN + EXCLUDE : PASS"
)

print(
    f"{len(retained_unreviewed_ids)}"
    " conditions with retained + unreviewed images : PASS"
)

print(
    f"{len(excluded_unreviewed_ids)}"
    " conditions with excluded + unreviewed images : PASS"
)

print("Automatic modality exclusion        : NO")
print("Automatic cxr_outlier exclusion     : NO")
print("Representative CXR selected        : NO")
print("Final eligibility selected          : NO")
print("Final research cohort selected      : NO")
print("Final target selected               : NO")
print("Additional ZIP extraction           : NO")
print("DICOM files modified                : NO")
print("Source evidence modified            : NO")
print("Output files                         : PASS")

print("\n" + "=" * 100)
print("STEP 3B.6.33 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.34 — CXR ↔ GENOMIC TEMPORAL RECONCILIATION
# COMPLETE SELF-CONTAINED CORRECTED VERSION
# ===================================================================================================
#
# PURPOSE
# -------
# Establish the temporal relationship between CXR imaging and genomic specimen
# collection for the March 2025 CXR + Genomics common cohort.
#
# CRITICAL COHORT RULE
# --------------------
# The full Genomics dataset contains more conditions than the multimodal CXR
# cohort. Therefore:
#
#     FULL GENOMICS
#          ↓
#     COMMON CXR + GENOMICS CONDITIONS
#          ↓
#     TEMPORAL ANALYSIS
#
# The previously observed 205 multiple-genomic-timepoint conditions belongs to
# the FULL genomic population.
#
# The multimodal common cohort contains 3,081 conditions, within which the
# established multiple-genomic-timepoint count is 152.
#
# THIS STEP DOES NOT:
# -------------------
# - select final cohort
# - select target
# - select representative CXR
# - exclude CXR
# - exclude genomic records
# - preprocess images
# - train models
# - modify DICOMs
# - extract additional ZIP files
#
# ===================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 100)
print("STEP 3B.6.34 — CXR ↔ GENOMIC TEMPORAL RECONCILIATION")
print("=" * 100)


# ===================================================================================================
# 1. INITIALIZATION
# ===================================================================================================

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_030_DIR = (
    METADATA_DIR /
    "Step_3B_6_30_Image_Level_CXR_Eligibility"
)

STEP_032_DIR = (
    METADATA_DIR /
    "Step_3B_6_32_Condition_Level_Eligibility_Evidence"
)

STEP_033_DIR = (
    METADATA_DIR /
    "Step_3B_6_33_Multiple_CXR_Condition_Analysis"
)

CXR_MASTER_PATH = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03" /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

IMAGE_AUDIT_PATH = (
    STEP_030_DIR /
    "TB_Portals_March2025_Step_3B_6_30_Image_Level_CXR_Eligibility_Audit.csv"
)

CONDITION_EVIDENCE_PATH = (
    STEP_032_DIR /
    "TB_Portals_March2025_Step_3B_6_32_Condition_Level_Eligibility_Evidence.csv"
)

MULTIPLE_CXR_MASTER_PATH = (
    STEP_033_DIR /
    "TB_Portals_March2025_Step_3B_6_33_Multiple_CXR_Condition_Master.csv"
)

GENOMICS_PATH = (
    TBP_ROOT /
    "Metadata" /
    "TB_Portals_Genomics_March_2025.csv"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ===================================================================================================
# 2. INPUT ARTIFACT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

required_paths = {
    "CXR master": CXR_MASTER_PATH,
    "Step 3B.6.30 image audit": IMAGE_AUDIT_PATH,
    "Step 3B.6.32 condition evidence": CONDITION_EVIDENCE_PATH,
    "Step 3B.6.33 multiple-CXR master": MULTIPLE_CXR_MASTER_PATH,
    "March 2025 genomics": GENOMICS_PATH,
}

for label, path in required_paths.items():

    assert path.exists(), (
        f"Required artifact not found:\n{path}"
    )

    print(
        f"{label:<38}: FOUND"
    )

print("Input artifact validation: PASS")


# ===================================================================================================
# 3. LOAD AUTHORITATIVE ARTIFACTS
# ===================================================================================================

cxr_master = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

image_audit = pd.read_csv(
    IMAGE_AUDIT_PATH,
    low_memory=False
)

condition_evidence = pd.read_csv(
    CONDITION_EVIDENCE_PATH,
    low_memory=False
)

multiple_cxr_master = pd.read_csv(
    MULTIPLE_CXR_MASTER_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)


print("\n2. LOADED ARTIFACTS")
print("-" * 100)

print(
    "CXR master records             :",
    len(cxr_master)
)

print(
    "Image audit records            :",
    len(image_audit)
)

print(
    "Condition evidence records     :",
    len(condition_evidence)
)

print(
    "Multiple-CXR conditions         :",
    len(multiple_cxr_master)
)

print(
    "Full genomic records            :",
    len(genomics)
)


# ===================================================================================================
# 4. REQUIRED CXR COLUMNS
# ===================================================================================================

cxr_required = [
    "condition_id",
    "series_instance_content_url",
    "imaging_date",
    "series_modality_cd",
    "cxr_outlier",
    "has_genomics",
]

missing_cxr = [
    column
    for column in cxr_required
    if column not in cxr_master.columns
]

print("\n3. CXR SCHEMA VALIDATION")
print("-" * 100)

print(
    "Missing CXR columns:",
    missing_cxr
)

assert len(missing_cxr) == 0

print("CXR schema: PASS")


# ===================================================================================================
# 5. REQUIRED GENOMIC COLUMNS
# ===================================================================================================

genomic_required = [
    "condition_id",
    "specimen_id",
    "specimen_collection_date",
]

missing_genomic = [
    column
    for column in genomic_required
    if column not in genomics.columns
]

print("\n4. GENOMIC SCHEMA VALIDATION")
print("-" * 100)

print(
    "Missing genomic columns:",
    missing_genomic
)

assert len(missing_genomic) == 0

print("Genomic schema: PASS")


# ===================================================================================================
# 6. CXR MASTER VALIDATION
# ===================================================================================================

print("\n5. CXR MASTER POPULATION VALIDATION")
print("-" * 100)

assert len(cxr_master) == 3859

assert (
    cxr_master[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    cxr_master[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    cxr_master[
        "condition_id"
    ].notna()
    .all()
)

assert (
    cxr_master[
        "series_instance_content_url"
    ].notna()
    .all()
)

print("CXR records       : 3,859")
print("Unique CXR URLs   : 3,859")
print("Unique conditions : 3,081")
print("CXR master population: PASS")


# ===================================================================================================
# 7. NORMALIZE CXR IDENTIFIERS AND DATES
# ===================================================================================================

print("\n6. CXR TEMPORAL FIELD NORMALIZATION")
print("-" * 100)

cxr = cxr_master[
    [
        "condition_id",
        "series_instance_content_url",
        "imaging_date",
        "series_modality_cd",
        "cxr_outlier",
        "has_genomics",
    ]
].copy()

cxr["condition_id"] = (
    cxr["condition_id"]
    .astype(str)
    .str.strip()
)

cxr["series_instance_content_url"] = (
    cxr["series_instance_content_url"]
    .astype(str)
    .str.strip()
)

cxr["imaging_date_num"] = pd.to_numeric(
    cxr["imaging_date"],
    errors="coerce"
)

assert (
    cxr["imaging_date_num"].notna().all()
)

assert (
    cxr["condition_id"].nunique()
    == 3081
)

assert (
    cxr["series_instance_content_url"].nunique()
    == 3859
)

print(
    "CXR records with valid imaging dates:",
    int(
        cxr["imaging_date_num"].notna().sum()
    )
)

print("CXR temporal normalization: PASS")


# ===================================================================================================
# 8. NORMALIZE GENOMIC IDENTIFIERS AND DATES
# ===================================================================================================

print("\n7. GENOMIC TEMPORAL FIELD NORMALIZATION")
print("-" * 100)

genomics["condition_id"] = (
    genomics["condition_id"]
    .astype(str)
    .str.strip()
)

genomics["specimen_id"] = (
    genomics["specimen_id"]
    .astype(str)
    .str.strip()
)

genomics[
    "specimen_collection_date_num"
] = pd.to_numeric(
    genomics[
        "specimen_collection_date"
    ],
    errors="coerce"
)

assert (
    genomics["condition_id"]
    .notna()
    .all()
)

assert (
    genomics["specimen_id"]
    .notna()
    .all()
)

assert (
    genomics[
        "specimen_collection_date_num"
    ]
    .notna()
    .all()
)

print(
    "Genomic records with valid specimen dates:",
    int(
        genomics[
            "specimen_collection_date_num"
        ].notna().sum()
    )
)

print("Genomic temporal normalization: PASS")


# ===================================================================================================
# 9. FULL GENOMICS POPULATION — DESCRIPTIVE ONLY
# ===================================================================================================

print("\n8. FULL GENOMICS POPULATION")
print("-" * 100)

full_genomic_conditions = set(
    genomics[
        "condition_id"
    ]
)

print(
    "Full genomic records:",
    len(genomics)
)

print(
    "Full genomic conditions:",
    len(full_genomic_conditions)
)

assert len(genomics) == 4467
assert len(full_genomic_conditions) == 4155

print("Full genomic population: PASS")


# ===================================================================================================
# 10. IDENTIFY COMMON MULTIMODAL CONDITIONS
# ===================================================================================================

print("\n9. CXR ↔ GENOMICS COMMON-COHORT RECONCILIATION")
print("-" * 100)

cxr_conditions = set(
    cxr[
        "condition_id"
    ]
)

common_conditions = (
    cxr_conditions
    &
    full_genomic_conditions
)

genomics_only_conditions = (
    full_genomic_conditions
    -
    cxr_conditions
)

print(
    "CXR conditions:",
    len(cxr_conditions)
)

print(
    "Full genomic conditions:",
    len(full_genomic_conditions)
)

print(
    "Common CXR + genomic conditions:",
    len(common_conditions)
)

print(
    "Genomics-only conditions:",
    len(genomics_only_conditions)
)

assert len(cxr_conditions) == 3081
assert len(common_conditions) == 3081

print("Common-cohort reconciliation: PASS")


# ===================================================================================================
# 11. RESTRICT GENOMICS TO COMMON COHORT
# ===================================================================================================
#
# THIS IS THE CRITICAL CORRECTION.
#
# From this point onward, temporal analysis is restricted to the 3,081
# conditions represented by BOTH CXR and Genomics.
#
# ===================================================================================================

print("\n10. RESTRICTING GENOMICS TO MULTIMODAL COMMON COHORT")
print("-" * 100)

genomics_common = (
    genomics[
        genomics[
            "condition_id"
        ].isin(
            common_conditions
        )
    ]
    .copy()
)

print(
    "Full genomic records:",
    len(genomics)
)

print(
    "Common-cohort genomic records:",
    len(genomics_common)
)

print(
    "Common-cohort genomic conditions:",
    genomics_common[
        "condition_id"
    ].nunique()
)

assert (
    genomics_common[
        "condition_id"
    ].nunique()
    == 3081
)

print("Common-cohort genomic restriction: PASS")


# ===================================================================================================
# 12. UNIQUE GENOMIC TIMEPOINTS — COMMON COHORT ONLY
# ===================================================================================================
#
# Biological/timepoint identity:
#
# condition_id + specimen_id + specimen_collection_date_num
#
# Duplicate raw rows are NOT deleted from the source dataset.
#
# ===================================================================================================

print("\n11. BUILDING UNIQUE GENOMIC TIMEPOINTS — COMMON COHORT")
print("-" * 100)

genomic_timepoints = (
    genomics_common[
        [
            "condition_id",
            "specimen_id",
            "specimen_collection_date_num",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        [
            "condition_id",
            "specimen_collection_date_num",
            "specimen_id",
        ]
    )
    .reset_index(
        drop=True
    )
)

assert (
    genomic_timepoints[
        [
            "condition_id",
            "specimen_id",
            "specimen_collection_date_num",
        ]
    ]
    .duplicated()
    .sum()
    == 0
)

assert (
    genomic_timepoints[
        "condition_id"
    ].nunique()
    == 3081
)

print(
    "Common-cohort genomic records:",
    len(genomics_common)
)

print(
    "Unique genomic timepoints:",
    len(genomic_timepoints)
)

print(
    "Conditions represented:",
    genomic_timepoints[
        "condition_id"
    ].nunique()
)

print(
    "Duplicate raw rows collapsed for temporal identity:",
    len(genomics_common)
    -
    len(genomic_timepoints)
)

print("Unique genomic timepoint construction: PASS")


# ===================================================================================================
# 13. GENOMIC LONGITUDINAL STRUCTURE — COMMON COHORT
# ===================================================================================================

print("\n12. GENOMIC LONGITUDINAL STRUCTURE — COMMON COHORT")
print("-" * 100)

genomic_timepoint_counts = (
    genomic_timepoints
    .groupby(
        "condition_id"
    )
    .size()
    .rename(
        "genomic_timepoint_count"
    )
)

assert (
    len(genomic_timepoint_counts)
    == 3081
)

multiple_genomic_conditions = (
    genomic_timepoint_counts[
        genomic_timepoint_counts > 1
    ]
)

print(
    "Common-cohort conditions with genomic timepoints:",
    len(genomic_timepoint_counts)
)

print(
    "Common-cohort conditions with >1 genomic timepoint:",
    len(multiple_genomic_conditions)
)

# Established common-cohort result.
assert (
    len(multiple_genomic_conditions)
    == 152
)

print(
    "152 multiple-genomic-timepoint conditions: PASS"
)

print(
    "IMPORTANT: 205 was the value obtained when the full 4,155-condition "
    "genomic population was incorrectly used. It is NOT used here."
)

print("Genomic longitudinal structure: PASS")


# ===================================================================================================
# 14. GENOMIC DATE SPAN — COMMON COHORT
# ===================================================================================================

print("\n13. GENOMIC DATE-SPAN ANALYSIS")
print("-" * 100)

genomic_span = (
    genomic_timepoints
    .groupby(
        "condition_id"
    )
    .agg(
        genomic_unique_timepoints=(
            "specimen_collection_date_num",
            "nunique"
        ),
        genomic_min_date=(
            "specimen_collection_date_num",
            "min"
        ),
        genomic_max_date=(
            "specimen_collection_date_num",
            "max"
        ),
    )
)

genomic_span[
    "genomic_date_span"
] = (
    genomic_span[
        "genomic_max_date"
    ]
    -
    genomic_span[
        "genomic_min_date"
    ]
)

assert len(genomic_span) == 3081

print(
    genomic_span[
        "genomic_date_span"
    ]
    .describe()
    .to_string()
)

print("Genomic date-span analysis: PASS")


# ===================================================================================================
# 15. BUILD MULTIMODAL CXR COHORT
# ===================================================================================================

print("\n14. MULTIMODAL CXR COHORT")
print("-" * 100)

cxr_multimodal = (
    cxr[
        cxr[
            "condition_id"
        ].isin(
            common_conditions
        )
    ]
    .copy()
)

assert len(cxr_multimodal) == 3859

assert (
    cxr_multimodal[
        "condition_id"
    ].nunique()
    == 3081
)

print(
    "Multimodal CXR records:",
    len(cxr_multimodal)
)

print(
    "Multimodal conditions:",
    cxr_multimodal[
        "condition_id"
    ].nunique()
)

print("Multimodal CXR cohort: PASS")


# ===================================================================================================
# 16. BUILD ALL CXR ↔ GENOMIC TEMPORAL PAIRS
# ===================================================================================================
#
# Every CXR is compared with every unique genomic timepoint from the same
# condition.
#
# This is an AUDIT table.
#
# It does not select a final pair.
#
# ===================================================================================================

print("\n15. BUILDING ALL CXR ↔ GENOMIC TEMPORAL PAIRS")
print("-" * 100)

temporal_pairs = (
    cxr_multimodal[
        [
            "condition_id",
            "series_instance_content_url",
            "imaging_date_num",
        ]
    ]
    .merge(
        genomic_timepoints[
            [
                "condition_id",
                "specimen_id",
                "specimen_collection_date_num",
            ]
        ],
        on="condition_id",
        how="inner",
        validate="many_to_many"
    )
)

assert (
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    temporal_pairs[
        "condition_id"
    ].nunique()
    == 3081
)

temporal_pairs[
    "signed_temporal_delta_days"
] = (
    temporal_pairs[
        "specimen_collection_date_num"
    ]
    -
    temporal_pairs[
        "imaging_date_num"
    ]
)

temporal_pairs[
    "absolute_temporal_distance_days"
] = (
    temporal_pairs[
        "signed_temporal_delta_days"
    ]
    .abs()
)

print(
    "CXR ↔ genomic temporal pair records:",
    len(temporal_pairs)
)

print(
    "CXR records represented:",
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
)

print(
    "Conditions represented:",
    temporal_pairs[
        "condition_id"
    ].nunique()
)

print("Temporal pair construction: PASS")


# ===================================================================================================
# 17. TEMPORAL DIRECTION
# ===================================================================================================

print("\n16. TEMPORAL DIRECTION")
print("-" * 100)

temporal_pairs[
    "temporal_direction"
] = np.select(
    [
        temporal_pairs[
            "signed_temporal_delta_days"
        ] < 0,

        temporal_pairs[
            "signed_temporal_delta_days"
        ] == 0,

        temporal_pairs[
            "signed_temporal_delta_days"
        ] > 0,
    ],
    [
        "GENOMICS_BEFORE_CXR",
        "SAME_DAY",
        "GENOMICS_AFTER_CXR",
    ],
    default="UNKNOWN"
)

assert (
    ~temporal_pairs[
        "temporal_direction"
    ]
    .eq("UNKNOWN")
).all()

direction_summary = (
    temporal_pairs[
        "temporal_direction"
    ]
    .value_counts()
)

print(
    direction_summary.to_string()
)

print("Temporal direction analysis: PASS")


# ===================================================================================================
# 18. TEMPORAL DISTANCE CATEGORIES
# ===================================================================================================

print("\n17. TEMPORAL DISTANCE CATEGORIES")
print("-" * 100)

def classify_distance(days):

    if days == 0:
        return "SAME_DAY"

    elif days <= 7:
        return "1_TO_7_DAYS"

    elif days <= 30:
        return "8_TO_30_DAYS"

    elif days <= 90:
        return "31_TO_90_DAYS"

    elif days <= 180:
        return "91_TO_180_DAYS"

    elif days <= 365:
        return "181_TO_365_DAYS"

    else:
        return "MORE_THAN_365_DAYS"


temporal_pairs[
    "absolute_distance_category"
] = (
    temporal_pairs[
        "absolute_temporal_distance_days"
    ]
    .apply(
        classify_distance
    )
)

distance_summary = (
    temporal_pairs[
        "absolute_distance_category"
    ]
    .value_counts()
)

print(
    distance_summary.to_string()
)

print("Temporal distance categorization: PASS")


# ===================================================================================================
# 19. NEAREST GENOMIC TIMEPOINT PER CXR
# ===================================================================================================
#
# IMPORTANT:
# This identifies the closest genomic timepoint only for temporal auditing.
#
# It DOES NOT select a final CXR.
#
# If there is a tie, all tied timepoints are retained.
#
# ===================================================================================================

print("\n18. NEAREST GENOMIC TIMEPOINT AUDIT")
print("-" * 100)

minimum_distance = (
    temporal_pairs
    .groupby(
        "series_instance_content_url"
    )[
        "absolute_temporal_distance_days"
    ]
    .transform(
        "min"
    )
)

nearest_pairs = (
    temporal_pairs[
        temporal_pairs[
            "absolute_temporal_distance_days"
        ]
        ==
        minimum_distance
    ]
    .copy()
)

assert (
    nearest_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

print(
    "CXR records covered:",
    nearest_pairs[
        "series_instance_content_url"
    ].nunique()
)

print(
    "Nearest-pair rows including ties:",
    len(nearest_pairs)
)

nearest_counts = (
    nearest_pairs
    .groupby(
        "series_instance_content_url"
    )
    .size()
)

tie_cxrs = nearest_counts[
    nearest_counts > 1
]

print(
    "CXRs with nearest-timepoint ties:",
    len(tie_cxrs)
)

print("Nearest genomic timepoint audit: PASS")


# ===================================================================================================
# 20. NEAREST TEMPORAL DIRECTION
# ===================================================================================================

print("\n19. NEAREST TEMPORAL RELATIONSHIP")
print("-" * 100)

nearest_direction_summary = (
    nearest_pairs[
        "temporal_direction"
    ]
    .value_counts()
)

print(
    nearest_direction_summary.to_string()
)

print(
    "\nNearest absolute temporal distance:"
)

print(
    nearest_pairs[
        "absolute_temporal_distance_days"
    ]
    .describe()
    .to_string()
)

print("Nearest temporal relationship: PASS")


# ===================================================================================================
# 21. NEAREST TEMPORAL PROXIMITY WINDOWS
# ===================================================================================================
#
# Descriptive only.
# No window is being declared an eligibility threshold.
#
# ===================================================================================================

print("\n20. NEAREST TEMPORAL PROXIMITY WINDOWS")
print("-" * 100)

nearest_distance = (
    nearest_pairs[
        "absolute_temporal_distance_days"
    ]
)

window_summary = pd.DataFrame(
    {
        "window": [
            "SAME_DAY",
            "WITHIN_7_DAYS",
            "WITHIN_30_DAYS",
            "WITHIN_90_DAYS",
            "WITHIN_180_DAYS",
            "WITHIN_365_DAYS",
        ],

        "nearest_pair_rows": [
            int(
                (
                    nearest_distance == 0
                ).sum()
            ),

            int(
                (
                    nearest_distance <= 7
                ).sum()
            ),

            int(
                (
                    nearest_distance <= 30
                ).sum()
            ),

            int(
                (
                    nearest_distance <= 90
                ).sum()
            ),

            int(
                (
                    nearest_distance <= 180
                ).sum()
            ),

            int(
                (
                    nearest_distance <= 365
                ).sum()
            ),
        ],
    }
)

window_summary[
    "percentage_of_3859_cxr_records"
] = (
    window_summary[
        "nearest_pair_rows"
    ]
    /
    3859
    *
    100
)

print(
    window_summary.to_string(
        index=False
    )
)

assert (
    window_summary[
        "nearest_pair_rows"
    ]
    .diff()
    .fillna(0)
    >= 0
).all()

print(
    "Temporal proximity windows are descriptive only."
)

print("Temporal proximity-window analysis: PASS")


# ===================================================================================================
# 22. CONDITION-LEVEL TEMPORAL SUMMARY
# ===================================================================================================

print("\n21. CONDITION-LEVEL TEMPORAL SUMMARY")
print("-" * 100)

condition_temporal_summary = (
    nearest_pairs
    .groupby(
        "condition_id"
    )
    .agg(
        cxs_with_nearest_genomic=(
            "series_instance_content_url",
            "nunique"
        ),

        nearest_distance_min=(
            "absolute_temporal_distance_days",
            "min"
        ),

        nearest_distance_max=(
            "absolute_temporal_distance_days",
            "max"
        ),

        nearest_distance_median=(
            "absolute_temporal_distance_days",
            "median"
        ),
    )
)

assert (
    len(condition_temporal_summary)
    == 3081
)

print(
    "Conditions represented:",
    len(condition_temporal_summary)
)

print(
    "\nNearest-distance summary:"
)

print(
    condition_temporal_summary[
        "nearest_distance_min"
    ]
    .describe()
    .to_string()
)

print("Condition temporal summary: PASS")


# ===================================================================================================
# 23. MULTIPLE-CXR CONDITION TEMPORAL ANALYSIS
# ===================================================================================================

print("\n22. MULTIPLE-CXR CONDITION TEMPORAL ANALYSIS")
print("-" * 100)

multiple_cxr_condition_ids = set(
    multiple_cxr_master[
        "condition_id"
    ]
)

multiple_cxr_temporal = (
    nearest_pairs[
        nearest_pairs[
            "condition_id"
        ].isin(
            multiple_cxr_condition_ids
        )
    ]
    .copy()
)

assert (
    multiple_cxr_temporal[
        "condition_id"
    ].nunique()
    == 485
)

print(
    "Multiple-CXR conditions:",
    len(multiple_cxr_condition_ids)
)

print(
    "Multiple-CXR conditions with temporal data:",
    multiple_cxr_temporal[
        "condition_id"
    ].nunique()
)

print("Multiple-CXR temporal analysis: PASS")


# ===================================================================================================
# 24. MULTIPLE-GENOMIC-TIMEPOINT CONDITIONS
# ===================================================================================================

print("\n23. MULTIPLE-GENOMIC-TIMEPOINT CONDITION ANALYSIS")
print("-" * 100)

multiple_genomic_condition_ids = set(
    multiple_genomic_conditions.index
)

multiple_genomic_temporal = (
    temporal_pairs[
        temporal_pairs[
            "condition_id"
        ].isin(
            multiple_genomic_condition_ids
        )
    ]
    .copy()
)

assert (
    multiple_genomic_temporal[
        "condition_id"
    ].nunique()
    == 152
)

print(
    "Multiple-genomic-timepoint conditions:",
    len(multiple_genomic_condition_ids)
)

print(
    "Temporal pair records for these conditions:",
    len(multiple_genomic_temporal)
)

print("Multiple-genomic-timepoint analysis: PASS")


# ===================================================================================================
# 25. SAME-DAY CXR ↔ GENOMIC RELATIONSHIP
# ===================================================================================================

print("\n24. SAME-DAY CXR ↔ GENOMIC ANALYSIS")
print("-" * 100)

same_day_pairs = (
    temporal_pairs[
        temporal_pairs[
            "absolute_temporal_distance_days"
        ]
        == 0
    ]
    .copy()
)

print(
    "Same-day CXR ↔ genomic pairs:",
    len(same_day_pairs)
)

print(
    "CXR records with at least one same-day genomic timepoint:",
    same_day_pairs[
        "series_instance_content_url"
    ].nunique()
)

print(
    "Conditions with at least one same-day CXR ↔ genomic pair:",
    same_day_pairs[
        "condition_id"
    ].nunique()
)

print("Same-day CXR ↔ genomic analysis: PASS")


# ===================================================================================================
# 26. TEMPORAL DIRECTIONALITY SAFETY
# ===================================================================================================

print("\n25. TEMPORAL DIRECTIONALITY SAFETY")
print("-" * 100)

genomics_before = int(
    (
        nearest_pairs[
            "signed_temporal_delta_days"
        ]
        < 0
    ).sum()
)

same_day = int(
    (
        nearest_pairs[
            "signed_temporal_delta_days"
        ]
        == 0
    ).sum()
)

genomics_after = int(
    (
        nearest_pairs[
            "signed_temporal_delta_days"
        ]
        > 0
    ).sum()
)

print(
    "Genomics before CXR:",
    genomics_before
)

print(
    "Same day:",
    same_day
)

print(
    "Genomics after CXR:",
    genomics_after
)

assert (
    genomics_before
    +
    same_day
    +
    genomics_after
    ==
    len(nearest_pairs)
)

print(
    "No temporal direction used for exclusion."
)

print(
    "No temporal direction used for final cohort selection."
)

print("Temporal directionality safety: PASS")


# ===================================================================================================
# 27. TEMPORAL LEAKAGE PREPARATION
# ===================================================================================================

print("\n26. TEMPORAL LEAKAGE PREPARATION")
print("-" * 100)

print(
    "Temporal relationship established : YES"
)

print(
    "Temporal direction established    : YES"
)

print(
    "Multiple genomic timepoints retained: YES"
)

print(
    "Temporal duplicates blindly removed : NO"
)

print(
    "Final target selected               : NO"
)

print(
    "Leakage determination finalized     : NO"
)

print(
    "Final cohort selected               : NO"
)

print("Temporal leakage preparation: PASS")


# ===================================================================================================
# 28. SOURCE PROTECTION
# ===================================================================================================

print("\n27. SOURCE PROTECTION")
print("-" * 100)

print(
    "Raw DICOM modification              : NO"
)

print(
    "Source CXR metadata modification    : NO"
)

print(
    "Source genomic metadata modification: NO"
)

print(
    "Human-review artifact modification  : NO"
)

print(
    "Additional ZIP extraction            : NO"
)

print(
    "Image preprocessing                  : NO"
)

print(
    "Final cohort selection               : NO"
)

print("Source protection: PASS")


# ===================================================================================================
# 29. SAVE OUTPUTS
# ===================================================================================================

print("\n28. SAVING OUTPUTS")
print("-" * 100)

ALL_PAIRS_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

NEAREST_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Nearest_Genomic_Timepoint_Per_CXR.csv"
)

TIMEPOINT_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

CXR_AUDIT_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_34_CXR_Temporal_Audit.csv"
)

CONDITION_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Condition_Temporal_Summary.csv"
)

WINDOW_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Temporal_Window_Summary.csv"
)

DIRECTION_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Temporal_Direction_Summary.csv"
)

TIE_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Nearest_Temporal_Ties.csv"
)

GENOMIC_LONGITUDINAL_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Multiple_Genomic_Timepoint_Conditions.csv"
)


temporal_pairs.to_csv(
    ALL_PAIRS_OUTPUT,
    index=False
)

nearest_pairs.to_csv(
    NEAREST_OUTPUT,
    index=False
)

genomic_timepoints.to_csv(
    TIMEPOINT_OUTPUT,
    index=False
)

nearest_pairs[
    [
        "condition_id",
        "series_instance_content_url",
        "imaging_date_num",
        "specimen_id",
        "specimen_collection_date_num",
        "signed_temporal_delta_days",
        "absolute_temporal_distance_days",
        "temporal_direction",
        "absolute_distance_category",
    ]
].to_csv(
    CXR_AUDIT_OUTPUT,
    index=False
)

condition_temporal_summary.reset_index().to_csv(
    CONDITION_OUTPUT,
    index=False
)

window_summary.to_csv(
    WINDOW_OUTPUT,
    index=False
)

direction_summary.rename(
    "pair_count"
).rename_axis(
    "temporal_direction"
).reset_index().to_csv(
    DIRECTION_OUTPUT,
    index=False
)

if len(tie_cxrs) > 0:

    tie_output = nearest_pairs[
        nearest_pairs[
            "series_instance_content_url"
        ].isin(
            tie_cxrs.index
        )
    ].copy()

else:

    tie_output = pd.DataFrame(
        columns=nearest_pairs.columns
    )

tie_output.to_csv(
    TIE_OUTPUT,
    index=False
)

multiple_genomic_conditions.reset_index().to_csv(
    GENOMIC_LONGITUDINAL_OUTPUT,
    index=False
)


# ===================================================================================================
# 30. OUTPUT VALIDATION
# ===================================================================================================

print("\n29. OUTPUT FILE VALIDATION")
print("-" * 100)

output_paths = [
    ALL_PAIRS_OUTPUT,
    NEAREST_OUTPUT,
    TIMEPOINT_OUTPUT,
    CXR_AUDIT_OUTPUT,
    CONDITION_OUTPUT,
    WINDOW_OUTPUT,
    DIRECTION_OUTPUT,
    TIE_OUTPUT,
    GENOMIC_LONGITUDINAL_OUTPUT,
]

for output_path in output_paths:

    assert output_path.exists(), (
        f"Output missing: {output_path}"
    )

    assert output_path.stat().st_size > 0, (
        f"Output is empty: {output_path}"
    )

    print(
        f"{output_path.name}: PASS"
    )


# ===================================================================================================
# 31. FINAL INTEGRITY CHECKS
# ===================================================================================================

print("\n30. FINAL INTEGRITY CHECKS")
print("-" * 100)

# CXR population
assert len(cxr_multimodal) == 3859

assert (
    cxr_multimodal[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    cxr_multimodal[
        "condition_id"
    ].nunique()
    == 3081
)

# Common genomic population
assert (
    genomics_common[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    genomic_timepoints[
        "condition_id"
    ].nunique()
    == 3081
)

# Multiple genomic timepoints
assert (
    len(multiple_genomic_conditions)
    == 152
)

# Multiple CXR
assert (
    multiple_cxr_temporal[
        "condition_id"
    ].nunique()
    == 485
)

# Every CXR has temporal coverage
assert (
    nearest_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

# Every multimodal condition has temporal coverage
assert (
    nearest_pairs[
        "condition_id"
    ].nunique()
    == 3081
)

print(
    "3,859 CXR records temporally reconciled       : PASS"
)

print(
    "3,081 multimodal conditions represented       : PASS"
)

print(
    "152 multiple-genomic-timepoint conditions     : PASS"
)

print(
    "485 multiple-CXR conditions represented       : PASS"
)

print(
    "Every CXR has nearest genomic temporal audit  : PASS"
)

print(
    "Every multimodal condition has temporal audit: PASS"
)

print("Final integrity checks: PASS")


# ===================================================================================================
# 32. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.34 — FINAL STATUS")
print("=" * 100)

print(
    "3,859 multimodal CXR records              : PASS"
)

print(
    "3,081 multimodal conditions               : PASS"
)

print(
    "Common-cohort genomic timepoints          : PASS"
)

print(
    "152 multiple-genomic-timepoint conditions : PASS"
)

print(
    "485 multiple-CXR conditions               : PASS"
)

print(
    "All CXR ↔ genomic temporal pairs         : PASS"
)

print(
    "Nearest genomic timepoint audit           : PASS"
)

print(
    "Temporal ties preserved                   : PASS"
)

print(
    "Temporal directionality audited           : PASS"
)

print(
    "Same-day relationship audited             : PASS"
)

print(
    "No temporal exclusion applied             : PASS"
)

print(
    "No final target selected                  : PASS"
)

print(
    "No final cohort selected                  : PASS"
)

print(
    "No preprocessing performed                : PASS"
)

print(
    "Additional ZIP extraction                 : NO"
)

print(
    "DICOM files modified                      : NO"
)

print(
    "Source metadata modified                  : NO"
)

print(
    "Output files                              : PASS"
)

print("\n" + "=" * 100)
print("STEP 3B.6.34 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.35 — TEMPORAL DIRECTIONALITY & MEASUREMENT-ORDER AUDIT
# COMPLETE SELF-CONTAINED VERSION
# ===================================================================================================
#
# PURPOSE
# -------
# Characterize the temporal ordering between CXR imaging and genomic specimen
# collection within the 3,081-condition multimodal cohort.
#
# QUESTIONS ANSWERED
# ------------------
# 1. Was genomic sampling before the CXR?
# 2. Was genomic sampling on the same relative day?
# 3. Was genomic sampling after the CXR?
# 4. Does a condition have CXRs on both sides of genomic sampling?
# 5. Does a condition have multiple genomic timepoints?
# 6. Does a condition have multiple CXRs?
# 7. Are there temporally complex conditions requiring special handling later?
#
# IMPORTANT METHODOLOGICAL RULE
# -----------------------------
# This is an AUDIT.
#
# It does NOT:
# - exclude records
# - select a CXR
# - select a genomic specimen
# - establish a final temporal threshold
# - select a prediction target
# - perform preprocessing
# - train a model
# - modify DICOMs
# - extract additional ZIP files
#
# Temporal sign convention:
#
#   genomic_date - imaging_date
#
#   < 0  -> genomic specimen BEFORE CXR
#   = 0  -> SAME DAY
#   > 0  -> genomic specimen AFTER CXR
#
# ===================================================================================================


from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 100)
print("STEP 3B.6.35 — TEMPORAL DIRECTIONALITY & MEASUREMENT-ORDER AUDIT")
print("=" * 100)


# ===================================================================================================
# 1. INITIALIZATION
# ===================================================================================================

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_034_DIR = (
    METADATA_DIR /
    "Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
)

STEP_033_DIR = (
    METADATA_DIR /
    "Step_3B_6_33_Multiple_CXR_Condition_Analysis"
)

CXR_MASTER_PATH = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03" /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

TEMPORAL_PAIRS_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

NEAREST_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Nearest_Genomic_Timepoint_Per_CXR.csv"
)

TIMEPOINT_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

MULTIPLE_CXR_PATH = (
    STEP_033_DIR /
    "TB_Portals_March2025_Step_3B_6_33_Multiple_CXR_Condition_Master.csv"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_35_Temporal_Directionality_Audit"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ===================================================================================================
# 2. INPUT ARTIFACT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

required_paths = {
    "CXR master": CXR_MASTER_PATH,
    "Step 3B.6.34 all temporal pairs": TEMPORAL_PAIRS_PATH,
    "Step 3B.6.34 nearest temporal pairs": NEAREST_PATH,
    "Step 3B.6.34 genomic timepoints": TIMEPOINT_PATH,
    "Step 3B.6.33 multiple-CXR master": MULTIPLE_CXR_PATH,
}

for label, path in required_paths.items():

    assert path.exists(), (
        f"Required artifact not found:\n{path}"
    )

    print(
        f"{label:<42}: FOUND"
    )

print("Input artifact validation: PASS")


# ===================================================================================================
# 3. LOAD ARTIFACTS
# ===================================================================================================

cxr_master = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

temporal_pairs = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)

nearest_pairs = pd.read_csv(
    NEAREST_PATH,
    low_memory=False
)

genomic_timepoints = pd.read_csv(
    TIMEPOINT_PATH,
    low_memory=False
)

multiple_cxr_master = pd.read_csv(
    MULTIPLE_CXR_PATH,
    low_memory=False
)


print("\n2. LOADED ARTIFACTS")
print("-" * 100)

print(
    "CXR master records:",
    len(cxr_master)
)

print(
    "Temporal pair records:",
    len(temporal_pairs)
)

print(
    "Nearest temporal rows:",
    len(nearest_pairs)
)

print(
    "Unique genomic timepoints:",
    len(genomic_timepoints)
)

print(
    "Multiple-CXR conditions:",
    len(multiple_cxr_master)
)


# ===================================================================================================
# 4. REQUIRED TEMPORAL PAIR COLUMNS
# ===================================================================================================

temporal_required = [
    "condition_id",
    "series_instance_content_url",
    "imaging_date_num",
    "specimen_id",
    "specimen_collection_date_num",
    "signed_temporal_delta_days",
    "absolute_temporal_distance_days",
    "temporal_direction",
    "absolute_distance_category",
]

missing_temporal = [
    column
    for column in temporal_required
    if column not in temporal_pairs.columns
]

print("\n3. TEMPORAL PAIR SCHEMA VALIDATION")
print("-" * 100)

print(
    "Missing temporal columns:",
    missing_temporal
)

assert len(missing_temporal) == 0

print("Temporal pair schema: PASS")


# ===================================================================================================
# 5. REQUIRED NEAREST-PAIR COLUMNS
# ===================================================================================================

nearest_required = [
    "condition_id",
    "series_instance_content_url",
    "imaging_date_num",
    "specimen_id",
    "specimen_collection_date_num",
    "signed_temporal_delta_days",
    "absolute_temporal_distance_days",
    "temporal_direction",
    "absolute_distance_category",
]

missing_nearest = [
    column
    for column in nearest_required
    if column not in nearest_pairs.columns
]

print("\n4. NEAREST-PAIR SCHEMA VALIDATION")
print("-" * 100)

print(
    "Missing nearest columns:",
    missing_nearest
)

assert len(missing_nearest) == 0

print("Nearest-pair schema: PASS")


# ===================================================================================================
# 6. REQUIRED GENOMIC TIMEPOINT COLUMNS
# ===================================================================================================

timepoint_required = [
    "condition_id",
    "specimen_id",
    "specimen_collection_date_num",
]

missing_timepoints = [
    column
    for column in timepoint_required
    if column not in genomic_timepoints.columns
]

print("\n5. GENOMIC TIMEPOINT SCHEMA VALIDATION")
print("-" * 100)

print(
    "Missing timepoint columns:",
    missing_timepoints
)

assert len(missing_timepoints) == 0

print("Genomic timepoint schema: PASS")


# ===================================================================================================
# 7. MASTER POPULATION VALIDATION
# ===================================================================================================

print("\n6. MASTER POPULATION VALIDATION")
print("-" * 100)

assert len(cxr_master) == 3859

assert (
    cxr_master[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    cxr_master[
        "condition_id"
    ].nunique()
    == 3081
)

print(
    "CXR records       : 3,859"
)

print(
    "Unique CXR URLs   : 3,859"
)

print(
    "Unique conditions : 3,081"
)

print("Master population: PASS")


# ===================================================================================================
# 8. TEMPORAL PAIR POPULATION VALIDATION
# ===================================================================================================

print("\n7. TEMPORAL PAIR POPULATION VALIDATION")
print("-" * 100)

assert (
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    temporal_pairs[
        "condition_id"
    ].nunique()
    == 3081
)

print(
    "CXR records represented:",
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
)

print(
    "Conditions represented:",
    temporal_pairs[
        "condition_id"
    ].nunique()
)

print("Temporal pair population: PASS")


# ===================================================================================================
# 9. NORMALIZE TEMPORAL VALUES
# ===================================================================================================

print("\n8. TEMPORAL VALUE NORMALIZATION")
print("-" * 100)

temporal_pairs[
    "signed_temporal_delta_days"
] = pd.to_numeric(
    temporal_pairs[
        "signed_temporal_delta_days"
    ],
    errors="coerce"
)

temporal_pairs[
    "absolute_temporal_distance_days"
] = pd.to_numeric(
    temporal_pairs[
        "absolute_temporal_distance_days"
    ],
    errors="coerce"
)

assert (
    temporal_pairs[
        "signed_temporal_delta_days"
    ].notna().all()
)

assert (
    temporal_pairs[
        "absolute_temporal_distance_days"
    ].notna().all()
)

assert (
    temporal_pairs[
        "absolute_temporal_distance_days"
    ]
    >= 0
).all()

assert (
    np.isclose(
        temporal_pairs[
            "absolute_temporal_distance_days"
        ],
        temporal_pairs[
            "signed_temporal_delta_days"
        ].abs()
    )
    .all()
)

print("Temporal values: PASS")


# ===================================================================================================
# 10. VERIFY TEMPORAL DIRECTION LABELS
# ===================================================================================================

print("\n9. TEMPORAL DIRECTION VALIDATION")
print("-" * 100)

expected_direction = np.select(
    [
        temporal_pairs[
            "signed_temporal_delta_days"
        ] < 0,

        temporal_pairs[
            "signed_temporal_delta_days"
        ] == 0,

        temporal_pairs[
            "signed_temporal_delta_days"
        ] > 0,
    ],
    [
        "GENOMICS_BEFORE_CXR",
        "SAME_DAY",
        "GENOMICS_AFTER_CXR",
    ],
    default="UNKNOWN"
)

direction_mismatch = (
    temporal_pairs[
        "temporal_direction"
    ]
    .astype(str)
    .ne(
        expected_direction
    )
)

print(
    "Direction mismatches:",
    int(direction_mismatch.sum())
)

assert direction_mismatch.sum() == 0

print("Temporal direction labels: PASS")


# ===================================================================================================
# 11. ALL-PAIR DIRECTION DISTRIBUTION
# ===================================================================================================

print("\n10. ALL-PAIR TEMPORAL DIRECTION")
print("-" * 100)

all_pair_direction = (
    temporal_pairs[
        "temporal_direction"
    ]
    .value_counts()
    .rename_axis(
        "temporal_direction"
    )
    .reset_index(
        name="pair_count"
    )
)

print(
    all_pair_direction.to_string(
        index=False
    )
)

assert set(
    all_pair_direction[
        "temporal_direction"
    ]
) == {
    "GENOMICS_BEFORE_CXR",
    "SAME_DAY",
    "GENOMICS_AFTER_CXR",
}

print("All-pair direction audit: PASS")


# ===================================================================================================
# 12. BUILD CONDITION × DIRECTION STRUCTURE
# ===================================================================================================

print("\n11. CONDITION-LEVEL TEMPORAL DIRECTION STRUCTURE")
print("-" * 100)

condition_direction = (
    temporal_pairs
    .groupby(
        "condition_id"
    )[
        "temporal_direction"
    ]
    .agg(
        lambda x: set(x)
    )
    .rename(
        "temporal_directions"
    )
)

assert (
    len(condition_direction)
    == 3081
)

condition_direction_df = (
    condition_direction
    .reset_index()
)

condition_direction_df[
    "has_genomics_before_cxr"
] = (
    condition_direction_df[
        "temporal_directions"
    ]
    .apply(
        lambda x:
        "GENOMICS_BEFORE_CXR" in x
    )
)

condition_direction_df[
    "has_same_day"
] = (
    condition_direction_df[
        "temporal_directions"
    ]
    .apply(
        lambda x:
        "SAME_DAY" in x
    )
)

condition_direction_df[
    "has_genomics_after_cxr"
] = (
    condition_direction_df[
        "temporal_directions"
    ]
    .apply(
        lambda x:
        "GENOMICS_AFTER_CXR" in x
    )
)

condition_direction_df[
    "temporal_direction_count"
] = (
    condition_direction_df[
        "temporal_directions"
    ]
    .apply(
        len
    )
)

condition_direction_df[
    "temporal_direction_pattern"
] = (
    condition_direction_df[
        "temporal_directions"
    ]
    .apply(
        lambda x:
        "|".join(
            sorted(x)
        )
    )
)

print(
    "Conditions represented:",
    len(condition_direction_df)
)

print(
    "\nDirection-pattern distribution:"
)

print(
    condition_direction_df[
        "temporal_direction_pattern"
    ]
    .value_counts()
    .to_string()
)

print("Condition direction structure: PASS")


# ===================================================================================================
# 13. TEMPORAL COMPLEXITY
# ===================================================================================================

print("\n12. TEMPORAL COMPLEXITY CLASSIFICATION")
print("-" * 100)

def classify_temporal_pattern(row):

    before = row[
        "has_genomics_before_cxr"
    ]

    same = row[
        "has_same_day"
    ]

    after = row[
        "has_genomics_after_cxr"
    ]

    if before and same and after:
        return "BEFORE_SAME_DAY_AFTER"

    if before and same:
        return "BEFORE_AND_SAME_DAY"

    if before and after:
        return "BEFORE_AND_AFTER"

    if same and after:
        return "SAME_DAY_AND_AFTER"

    if before:
        return "BEFORE_ONLY"

    if same:
        return "SAME_DAY_ONLY"

    if after:
        return "AFTER_ONLY"

    return "UNKNOWN"


condition_direction_df[
    "temporal_complexity_class"
] = (
    condition_direction_df
    .apply(
        classify_temporal_pattern,
        axis=1
    )
)

complexity_summary = (
    condition_direction_df[
        "temporal_complexity_class"
    ]
    .value_counts()
    .rename_axis(
        "temporal_complexity_class"
    )
    .reset_index(
        name="condition_count"
    )
)

print(
    complexity_summary.to_string(
        index=False
    )
)

assert (
    complexity_summary[
        "condition_count"
    ].sum()
    == 3081
)

assert (
    ~condition_direction_df[
        "temporal_complexity_class"
    ].eq("UNKNOWN")
).all()

print("Temporal complexity classification: PASS")


# ===================================================================================================
# 14. IDENTIFY BIDIRECTIONAL TEMPORAL CONDITIONS
# ===================================================================================================
#
# A condition is bidirectional if at least one genomic timepoint is before
# a CXR and another genomic timepoint is after a CXR.
#
# This is an important temporal-complexity flag.
#
# It is NOT an exclusion criterion.
#
# ===================================================================================================

print("\n13. BIDIRECTIONAL TEMPORAL CONDITIONS")
print("-" * 100)

bidirectional_conditions = (
    condition_direction_df[
        condition_direction_df[
            "has_genomics_before_cxr"
        ]
        &
        condition_direction_df[
            "has_genomics_after_cxr"
        ]
    ]
    .copy()
)

print(
    "Bidirectional temporal conditions:",
    len(bidirectional_conditions)
)

print(
    "These conditions are retained for later temporal/leakage analysis."
)

print("Bidirectional temporal audit: PASS")


# ===================================================================================================
# 15. SAME-DAY CONDITION STRUCTURE
# ===================================================================================================

print("\n14. SAME-DAY CONDITION STRUCTURE")
print("-" * 100)

same_day_conditions = (
    condition_direction_df[
        condition_direction_df[
            "has_same_day"
        ]
    ]
    .copy()
)

print(
    "Conditions with at least one same-day pair:",
    len(same_day_conditions)
)

print("Same-day condition audit: PASS")


# ===================================================================================================
# 16. NEAREST-TIMEPOINT DIRECTION VALIDATION
# ===================================================================================================

print("\n15. NEAREST-TIMEPOINT DIRECTION VALIDATION")
print("-" * 100)

nearest_direction = (
    nearest_pairs[
        "temporal_direction"
    ]
    .value_counts()
)

print(
    nearest_direction.to_string()
)

assert (
    nearest_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

print(
    "Nearest CXR records:",
    nearest_pairs[
        "series_instance_content_url"
    ].nunique()
)

print("Nearest temporal direction: PASS")


# ===================================================================================================
# 17. NEAREST-TIMEPOINT CONDITION STRUCTURE
# ===================================================================================================

print("\n16. NEAREST-TIMEPOINT CONDITION STRUCTURE")
print("-" * 100)

nearest_condition_summary = (
    nearest_pairs
    .groupby(
        "condition_id"
    )
    .agg(
        nearest_min_distance=(
            "absolute_temporal_distance_days",
            "min"
        ),

        nearest_max_distance=(
            "absolute_temporal_distance_days",
            "max"
        ),

        nearest_cxr_count=(
            "series_instance_content_url",
            "nunique"
        ),

        nearest_genomic_timepoint_count=(
            "specimen_id",
            "nunique"
        ),

        nearest_direction_pattern=(
            "temporal_direction",
            lambda x:
            "|".join(
                sorted(
                    set(x)
                )
            )
        ),
    )
    .reset_index()
)

assert (
    len(nearest_condition_summary)
    == 3081
)

print(
    "Conditions represented:",
    len(nearest_condition_summary)
)

print("Nearest condition structure: PASS")


# ===================================================================================================
# 18. MULTIPLE-CXR CONDITIONS
# ===================================================================================================

print("\n17. MULTIPLE-CXR TEMPORAL COMPLEXITY")
print("-" * 100)

multiple_cxr_ids = set(
    multiple_cxr_master[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)

multiple_cxr_temporal = (
    condition_direction_df[
        condition_direction_df[
            "condition_id"
        ]
        .astype(str)
        .isin(
            multiple_cxr_ids
        )
    ]
    .copy()
)

assert (
    len(multiple_cxr_temporal)
    == 485
)

print(
    "Multiple-CXR conditions:",
    len(multiple_cxr_temporal)
)

print(
    "\nTemporal complexity among multiple-CXR conditions:"
)

print(
    multiple_cxr_temporal[
        "temporal_complexity_class"
    ]
    .value_counts()
    .to_string()
)

print("Multiple-CXR temporal audit: PASS")


# ===================================================================================================
# 19. MULTIPLE-GENOMIC-TIMEPOINT CONDITIONS
# ===================================================================================================

print("\n18. MULTIPLE-GENOMIC-TIMEPOINT TEMPORAL COMPLEXITY")
print("-" * 100)

genomic_timepoint_counts = (
    genomic_timepoints
    .groupby(
        "condition_id"
    )
    .size()
)

multiple_genomic_ids = set(
    genomic_timepoint_counts[
        genomic_timepoint_counts > 1
    ]
    .index
    .astype(str)
)

print(
    "Multiple-genomic-timepoint conditions:",
    len(multiple_genomic_ids)
)

assert len(multiple_genomic_ids) == 152

multiple_genomic_temporal = (
    condition_direction_df[
        condition_direction_df[
            "condition_id"
        ]
        .astype(str)
        .isin(
            multiple_genomic_ids
        )
    ]
    .copy()
)

assert (
    len(multiple_genomic_temporal)
    == 152
)

print(
    "Multiple-genomic conditions represented:",
    len(multiple_genomic_temporal)
)

print(
    "\nTemporal complexity among multiple-genomic-timepoint conditions:"
)

print(
    multiple_genomic_temporal[
        "temporal_complexity_class"
    ]
    .value_counts()
    .to_string()
)

print("Multiple-genomic temporal audit: PASS")


# ===================================================================================================
# 20. CXR-LEVEL EXTREME TEMPORAL DISTANCE
# ===================================================================================================

print("\n19. EXTREME TEMPORAL DISTANCE AUDIT")
print("-" * 100)

nearest_distance_summary = (
    nearest_pairs[
        "absolute_temporal_distance_days"
    ]
    .describe()
)

print(
    nearest_distance_summary.to_string()
)

print(
    "\nCXRs with nearest genomic distance > 365 days:",
    int(
        (
            nearest_pairs[
                "absolute_temporal_distance_days"
            ]
            > 365
        ).sum()
    )
)

print(
    "This is an audit flag only."
)

print(
    "No >365-day exclusion is applied."
)

print("Extreme temporal-distance audit: PASS")


# ===================================================================================================
# 21. TEMPORAL ORDER RELATIVE TO THE FIRST GENOMIC TIMEPOINT
# ===================================================================================================
#
# For each condition, identify the first unique genomic timepoint.
#
# Then classify each CXR relative to that first genomic timepoint.
#
# This is useful because "nearest genomic timepoint" and "first genomic
# timepoint" answer different scientific questions.
#
# ===================================================================================================

print("\n20. CXR ORDER RELATIVE TO FIRST GENOMIC TIMEPOINT")
print("-" * 100)

first_genomic = (
    genomic_timepoints
    .sort_values(
        [
            "condition_id",
            "specimen_collection_date_num",
            "specimen_id",
        ]
    )
    .groupby(
        "condition_id",
        as_index=False
    )
    .first()
)

first_genomic = first_genomic[
    [
        "condition_id",
        "specimen_id",
        "specimen_collection_date_num",
    ]
].copy()

first_genomic = first_genomic.rename(
    columns={
        "specimen_id":
            "first_genomic_specimen_id",

        "specimen_collection_date_num":
            "first_genomic_date",
    }
)

assert (
    len(first_genomic)
    == 3081
)

cxr_first_genomic = (
    cxr_master[
        [
            "condition_id",
            "series_instance_content_url",
            "imaging_date",
        ]
    ]
    .copy()
)

cxr_first_genomic[
    "condition_id"
] = (
    cxr_first_genomic[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)

cxr_first_genomic[
    "imaging_date_num"
] = pd.to_numeric(
    cxr_first_genomic[
        "imaging_date"
    ],
    errors="coerce"
)

cxr_first_genomic = (
    cxr_first_genomic
    .merge(
        first_genomic,
        on="condition_id",
        how="inner",
        validate="many_to_one"
    )
)

assert (
    len(cxr_first_genomic)
    == 3859
)

cxr_first_genomic[
    "delta_from_first_genomic_days"
] = (
    cxr_first_genomic[
        "imaging_date_num"
    ]
    -
    cxr_first_genomic[
        "first_genomic_date"
    ]
)

cxr_first_genomic[
    "cxr_relative_to_first_genomic"
] = np.select(
    [
        cxr_first_genomic[
            "delta_from_first_genomic_days"
        ] < 0,

        cxr_first_genomic[
            "delta_from_first_genomic_days"
        ] == 0,

        cxr_first_genomic[
            "delta_from_first_genomic_days"
        ] > 0,
    ],
    [
        "CXR_BEFORE_FIRST_GENOMIC",
        "CXR_SAME_DAY_AS_FIRST_GENOMIC",
        "CXR_AFTER_FIRST_GENOMIC",
    ],
    default="UNKNOWN"
)

assert (
    ~cxr_first_genomic[
        "cxr_relative_to_first_genomic"
    ].eq("UNKNOWN")
).all()

print(
    cxr_first_genomic[
        "cxr_relative_to_first_genomic"
    ]
    .value_counts()
    .to_string()
)

print("First-genomic-timepoint ordering: PASS")


# ===================================================================================================
# 22. CONDITION-LEVEL FIRST-GENOMIC ORDER
# ===================================================================================================

print("\n21. CONDITION-LEVEL FIRST-GENOMIC ORDER")
print("-" * 100)

first_genomic_condition_pattern = (
    cxr_first_genomic
    .groupby(
        "condition_id"
    )[
        "cxr_relative_to_first_genomic"
    ]
    .agg(
        lambda x:
        set(x)
    )
    .rename(
        "cxr_first_genomic_patterns"
    )
    .reset_index()
)

first_genomic_condition_pattern[
    "first_genomic_order_class"
] = (
    first_genomic_condition_pattern[
        "cxr_first_genomic_patterns"
    ]
    .apply(
        lambda x:
        "|".join(
            sorted(x)
        )
    )
)

print(
    first_genomic_condition_pattern[
        "first_genomic_order_class"
    ]
    .value_counts()
    .to_string()
)

assert (
    len(first_genomic_condition_pattern)
    == 3081
)

print("Condition-level first-genomic ordering: PASS")


# ===================================================================================================
# 23. TEMPORAL COMPLEXITY FLAGS
# ===================================================================================================

print("\n22. TEMPORAL COMPLEXITY FLAGS")
print("-" * 100)

condition_temporal_flags = (
    condition_direction_df[
        [
            "condition_id",
            "has_genomics_before_cxr",
            "has_same_day",
            "has_genomics_after_cxr",
            "temporal_direction_count",
            "temporal_complexity_class",
        ]
    ]
    .copy()
)

condition_temporal_flags = (
    condition_temporal_flags
    .merge(
        first_genomic_condition_pattern[
            [
                "condition_id",
                "first_genomic_order_class",
            ]
        ],
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
)

condition_temporal_flags[
    "temporal_complexity_flag"
] = np.select(
    [
        condition_temporal_flags[
            "temporal_complexity_class"
        ].isin(
            [
                "BEFORE_AND_AFTER",
                "BEFORE_SAME_DAY_AFTER",
            ]
        ),

        condition_temporal_flags[
            "temporal_direction_count"
        ] > 1,

        condition_temporal_flags[
            "first_genomic_order_class"
        ].str.contains(
            "CXR_BEFORE_FIRST_GENOMIC"
        )
        &
        condition_temporal_flags[
            "first_genomic_order_class"
        ].str.contains(
            "CXR_AFTER_FIRST_GENOMIC"
        ),
    ],
    [
        "HIGH_TEMPORAL_COMPLEXITY",
        "MULTIPLE_TEMPORAL_DIRECTIONS",
        "CXR_SPANS_FIRST_GENOMIC",
    ],
    default="STANDARD_TEMPORAL_STRUCTURE"
)

print(
    condition_temporal_flags[
        "temporal_complexity_flag"
    ]
    .value_counts()
    .to_string()
)

assert (
    len(condition_temporal_flags)
    == 3081
)

print("Temporal complexity flags: PASS")


# ===================================================================================================
# 24. NO AUTOMATIC EXCLUSION CHECK
# ===================================================================================================

print("\n23. AUTOMATIC EXCLUSION SAFETY")
print("-" * 100)

print(
    "Genomics-before-CXR used for exclusion : NO"
)

print(
    "Genomics-after-CXR used for exclusion  : NO"
)

print(
    "Same-day status used for exclusion      : NO"
)

print(
    ">365-day distance used for exclusion    : NO"
)

print(
    "Multiple genomic timepoints excluded    : NO"
)

print(
    "Multiple CXRs excluded                  : NO"
)

print(
    "Bidirectional conditions excluded       : NO"
)

print(
    "Temporal complexity excluded            : NO"
)

print("Automatic temporal exclusion: NO")


# ===================================================================================================
# 25. NO FINAL COHORT / TARGET DECISION
# ===================================================================================================

print("\n24. FINAL COHORT AND TARGET SAFETY")
print("-" * 100)

print(
    "Prediction target selected       : NO"
)

print(
    "Final cohort selected            : NO"
)

print(
    "Representative CXR selected     : NO"
)

print(
    "Representative genomic sample selected: NO"
)

print(
    "Temporal eligibility threshold selected: NO"
)

print(
    "Leakage determination finalized  : NO"
)

print("Final-selection safety: PASS")


# ===================================================================================================
# 26. SAVE OUTPUTS
# ===================================================================================================

print("\n25. SAVING OUTPUTS")
print("-" * 100)

CONDITION_DIRECTION_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_35_Condition_Temporal_Direction_Structure.csv"
)

COMPLEXITY_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_35_Temporal_Complexity_Summary.csv"
)

BIDIRECTIONAL_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_35_Bidirectional_Temporal_Conditions.csv"
)

NEAREST_CONDITION_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_35_Nearest_Temporal_Condition_Summary.csv"
)

FIRST_GENOMIC_CXR_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_35_CXR_Relative_To_First_Genomic_Timepoint.csv"
)

FIRST_GENOMIC_CONDITION_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_35_First_Genomic_Order_By_Condition.csv"
)

TEMPORAL_FLAGS_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_35_Temporal_Complexity_Flags.csv"
)

ALL_DIRECTION_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_35_All_Pair_Direction_Summary.csv"
)


condition_direction_df.to_csv(
    CONDITION_DIRECTION_OUTPUT,
    index=False
)

complexity_summary.to_csv(
    COMPLEXITY_OUTPUT,
    index=False
)

bidirectional_conditions.to_csv(
    BIDIRECTIONAL_OUTPUT,
    index=False
)

nearest_condition_summary.to_csv(
    NEAREST_CONDITION_OUTPUT,
    index=False
)

cxr_first_genomic.to_csv(
    FIRST_GENOMIC_CXR_OUTPUT,
    index=False
)

first_genomic_condition_pattern.to_csv(
    FIRST_GENOMIC_CONDITION_OUTPUT,
    index=False
)

condition_temporal_flags.to_csv(
    TEMPORAL_FLAGS_OUTPUT,
    index=False
)

all_pair_direction.to_csv(
    ALL_DIRECTION_OUTPUT,
    index=False
)


# ===================================================================================================
# 27. OUTPUT VALIDATION
# ===================================================================================================

print("\n26. OUTPUT FILE VALIDATION")
print("-" * 100)

output_paths = [
    CONDITION_DIRECTION_OUTPUT,
    COMPLEXITY_OUTPUT,
    BIDIRECTIONAL_OUTPUT,
    NEAREST_CONDITION_OUTPUT,
    FIRST_GENOMIC_CXR_OUTPUT,
    FIRST_GENOMIC_CONDITION_OUTPUT,
    TEMPORAL_FLAGS_OUTPUT,
    ALL_DIRECTION_OUTPUT,
]

for output_path in output_paths:

    assert output_path.exists(), (
        f"Output missing: {output_path}"
    )

    assert output_path.stat().st_size > 0, (
        f"Output empty: {output_path}"
    )

    print(
        f"{output_path.name}: PASS"
    )


# ===================================================================================================
# 28. FINAL POPULATION INTEGRITY
# ===================================================================================================

print("\n27. FINAL POPULATION INTEGRITY")
print("-" * 100)

assert (
    len(condition_direction_df)
    == 3081
)

assert (
    len(condition_temporal_flags)
    == 3081
)

assert (
    len(first_genomic_condition_pattern)
    == 3081
)

assert (
    len(nearest_condition_summary)
    == 3081
)

assert (
    cxr_first_genomic[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    len(multiple_cxr_temporal)
    == 485
)

assert (
    len(multiple_genomic_temporal)
    == 152
)

print(
    "3,081 conditions represented                    : PASS"
)

print(
    "3,859 CXR records represented                  : PASS"
)

print(
    "485 multiple-CXR conditions represented        : PASS"
)

print(
    "152 multiple-genomic-timepoint conditions      : PASS"
)

print(
    "Condition temporal structure complete           : PASS"
)

print(
    "First genomic ordering complete                : PASS"
)

print("Population integrity: PASS")


# ===================================================================================================
# 29. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.35 — FINAL STATUS")
print("=" * 100)

print(
    "3,859 CXR records temporally audited       : PASS"
)

print(
    "3,081 multimodal conditions audited        : PASS"
)

print(
    "Temporal directionality classified        : PASS"
)

print(
    "Same-day relationships preserved           : PASS"
)

print(
    "Multiple-CXR temporal structure preserved : PASS"
)

print(
    "152 multiple-genomic-timepoint conditions : PASS"
)

print(
    "Bidirectional temporal cases identified   : PASS"
)

print(
    "First genomic timepoint ordering audited  : PASS"
)

print(
    "Temporal complexity flags created         : PASS"
)

print(
    "No automatic temporal exclusion            : NO"
)

print(
    "No temporal threshold selected             : NO"
)

print(
    "No prediction target selected              : NO"
)

print(
    "No final cohort selected                   : NO"
)

print(
    "No preprocessing performed                 : NO"
)

print(
    "Additional ZIP extraction                  : NO"
)

print(
    "DICOM files modified                       : NO"
)

print(
    "Source metadata modified                   : NO"
)

print(
    "Output files                               : PASS"
)

print("\n" + "=" * 100)
print("STEP 3B.6.35 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.36 — LEAKAGE-RISK AUDIT
# COMPLETE SELF-CONTAINED VERSION
# ===================================================================================================
#
# PURPOSE
# -------
# Audit variables, identifiers, repeated observations, temporal relationships,
# and target-like information that could create data leakage in a future
# multimodal CXR + genomics prediction study.
#
# CRITICAL METHODOLOGICAL RULE
# ----------------------------
# The final prediction target has NOT been selected.
#
# Therefore this step does NOT:
#
# - select a target
# - select predictors
# - remove variables
# - exclude patients/conditions
# - select a temporal eligibility window
# - select a representative CXR
# - select a representative genomic specimen
# - preprocess images
# - train a model
# - modify DICOM files
# - modify source metadata
#
# This is an AUDIT ONLY.
#
# ===================================================================================================


from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 100)
print("STEP 3B.6.36 — LEAKAGE-RISK AUDIT")
print("=" * 100)


# ===================================================================================================
# 1. INITIALIZATION
# ===================================================================================================

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_024_DIR = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03"
)

STEP_034_DIR = (
    METADATA_DIR /
    "Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
)

STEP_035_DIR = (
    METADATA_DIR /
    "Step_3B_6_35_Temporal_Directionality_Audit"
)

CXR_MASTER_PATH = (
    STEP_024_DIR /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

TEMPORAL_PAIRS_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

NEAREST_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Nearest_Genomic_Timepoint_Per_CXR.csv"
)

GENOMIC_TIMEPOINT_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

TEMPORAL_FLAGS_PATH = (
    STEP_035_DIR /
    "TB_Portals_March2025_Step_3B_6_35_Temporal_Complexity_Flags.csv"
)

CONDITION_DIRECTION_PATH = (
    STEP_035_DIR /
    "TB_Portals_March2025_Step_3B_6_35_Condition_Temporal_Direction_Structure.csv"
)

GENOMICS_PATH = (
    METADATA_DIR /
    "TB_Portals_Genomics_March_2025.csv"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_36_Leakage_Risk_Audit"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ===================================================================================================
# 2. INPUT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

required_paths = {
    "CXR master": CXR_MASTER_PATH,
    "Step 3B.6.34 temporal pairs": TEMPORAL_PAIRS_PATH,
    "Step 3B.6.34 nearest pairs": NEAREST_PATH,
    "Step 3B.6.34 genomic timepoints": GENOMIC_TIMEPOINT_PATH,
    "Step 3B.6.35 temporal flags": TEMPORAL_FLAGS_PATH,
    "Step 3B.6.35 condition direction": CONDITION_DIRECTION_PATH,
    "March 2025 genomics": GENOMICS_PATH,
}

for label, path in required_paths.items():

    assert path.exists(), (
        f"Required artifact not found:\n{path}"
    )

    print(
        f"{label:<42}: FOUND"
    )

print("Input artifact validation: PASS")


# ===================================================================================================
# 3. LOAD ARTIFACTS
# ===================================================================================================

cxr = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

temporal_pairs = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)

nearest_pairs = pd.read_csv(
    NEAREST_PATH,
    low_memory=False
)

genomic_timepoints = pd.read_csv(
    GENOMIC_TIMEPOINT_PATH,
    low_memory=False
)

temporal_flags = pd.read_csv(
    TEMPORAL_FLAGS_PATH,
    low_memory=False
)

condition_direction = pd.read_csv(
    CONDITION_DIRECTION_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)


print("\n2. LOADED ARTIFACTS")
print("-" * 100)

print(
    "CXR records:",
    len(cxr)
)

print(
    "Temporal pair records:",
    len(temporal_pairs)
)

print(
    "Nearest temporal rows:",
    len(nearest_pairs)
)

print(
    "Genomic timepoints:",
    len(genomic_timepoints)
)

print(
    "Temporal flag conditions:",
    len(temporal_flags)
)

print(
    "Condition direction records:",
    len(condition_direction)
)

print(
    "Full genomic records:",
    len(genomics)
)


# ===================================================================================================
# 4. CORE POPULATION VALIDATION
# ===================================================================================================

print("\n3. CORE POPULATION VALIDATION")
print("-" * 100)

assert len(cxr) == 3859

assert (
    cxr[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    cxr[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    temporal_pairs[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    len(temporal_flags)
    == 3081
)

assert (
    len(condition_direction)
    == 3081
)

print(
    "CXR records       : 3,859"
)

print(
    "Conditions         : 3,081"
)

print(
    "Unique CXR URLs    : 3,859"
)

print("Core population: PASS")


# ===================================================================================================
# 5. IDENTIFIER LEAKAGE AUDIT
# ===================================================================================================
#
# Identifiers should not automatically become model predictors.
#
# We explicitly identify:
#
# - patient_id
# - condition_id
# - imagingstudy_id
# - series_instance_content_url
# - specimen_id
#
# These may be essential for data linkage and grouping but should not be
# treated as ordinary predictive features.
#
# ===================================================================================================

print("\n4. IDENTIFIER LEAKAGE AUDIT")
print("-" * 100)

cxr_identifier_columns = [
    "patient_id",
    "condition_id",
    "imagingstudy_id",
    "series_instance_content_url",
]

genomic_identifier_columns = [
    "condition_id",
    "specimen_id",
]

identifier_audit_rows = []

for column in cxr_identifier_columns:

    if column in cxr.columns:

        identifier_audit_rows.append(
            {
                "dataset": "CXR",
                "column": column,
                "present": True,
                "role": "LINKAGE_IDENTIFIER",
                "predictor_status": "DO_NOT_USE_AS_FEATURE",
            }
        )


for column in genomic_identifier_columns:

    if column in genomics.columns:

        identifier_audit_rows.append(
            {
                "dataset": "GENOMICS",
                "column": column,
                "present": True,
                "role": "LINKAGE_IDENTIFIER",
                "predictor_status": "DO_NOT_USE_AS_FEATURE",
            }
        )


identifier_audit = pd.DataFrame(
    identifier_audit_rows
)

print(
    identifier_audit.to_string(
        index=False
    )
)

assert len(identifier_audit) > 0

print(
    "Identifiers are retained for linkage, grouping and audit purposes."
)

print(
    "Identifiers are NOT designated as predictive features."
)

print("Identifier leakage audit: PASS")


# ===================================================================================================
# 6. DIRECT TARGET-LIKE VARIABLE AUDIT — CXR
# ===================================================================================================
#
# We cannot know the final target yet.
#
# However, some variables are clearly target-like or outcome-like and must
# be explicitly isolated before feature engineering.
#
# ===================================================================================================

print("\n5. DIRECT TARGET-LIKE VARIABLE AUDIT — CXR")
print("-" * 100)

cxr_target_like_candidates = [
    "type_of_resistance",
    "outcome",
    "diagnosis_code",
    "case_definition",
]

cxr_target_like_present = [
    column
    for column in cxr_target_like_candidates
    if column in cxr.columns
]

print(
    "Target-like / outcome-like CXR fields:"
)

for column in cxr_target_like_present:

    print(
        f"  {column}"
    )

assert (
    set(
        [
            "type_of_resistance",
            "outcome",
        ]
    )
    .issubset(
        set(cxr_target_like_present)
    )
)

print(
    "\nThese fields are NOT automatically deleted."
)

print(
    "They are isolated as potential target/outcome/proxy fields."
)

print("CXR target-like variable audit: PASS")


# ===================================================================================================
# 7. DIRECT TARGET-LIKE VARIABLE AUDIT — GENOMICS
# ===================================================================================================

print("\n6. DIRECT TARGET-LIKE VARIABLE AUDIT — GENOMICS")
print("-" * 100)

genomic_target_like_candidates = [
    "drug_resistance_type",
    "type_of_resistance",
    "outcome",
]

genomic_target_like_present = [
    column
    for column in genomic_target_like_candidates
    if column in genomics.columns
]

print(
    "Target-like / outcome-like genomic fields:"
)

for column in genomic_target_like_present:

    print(
        f"  {column}"
    )

print(
    "\nThese fields are isolated for leakage review."
)

print(
    "They are not designated as predictors."
)

print("Genomic target-like variable audit: PASS")


# ===================================================================================================
# 8. TEMPORAL LEAKAGE AUDIT
# ===================================================================================================

print("\n7. TEMPORAL LEAKAGE AUDIT")
print("-" * 100)

nearest_pairs[
    "absolute_temporal_distance_days"
] = pd.to_numeric(
    nearest_pairs[
        "absolute_temporal_distance_days"
    ],
    errors="coerce"
)

nearest_pairs[
    "signed_temporal_delta_days"
] = pd.to_numeric(
    nearest_pairs[
        "signed_temporal_delta_days"
    ],
    errors="coerce"
)

assert (
    nearest_pairs[
        "absolute_temporal_distance_days"
    ].notna().all()
)

assert (
    nearest_pairs[
        "signed_temporal_delta_days"
    ].notna().all()
)

nearest_pairs[
    "genomics_before_cxr"
] = (
    nearest_pairs[
        "signed_temporal_delta_days"
    ]
    < 0
)

nearest_pairs[
    "same_day"
] = (
    nearest_pairs[
        "signed_temporal_delta_days"
    ]
    == 0
)

nearest_pairs[
    "genomics_after_cxr"
] = (
    nearest_pairs[
        "signed_temporal_delta_days"
    ]
    > 0
)

print(
    "Nearest genomic timepoint:"
)

print(
    "  Genomics before CXR:",
    int(
        nearest_pairs[
            "genomics_before_cxr"
        ].sum()
    )
)

print(
    "  Same day:",
    int(
        nearest_pairs[
            "same_day"
        ].sum()
    )
)

print(
    "  Genomics after CXR:",
    int(
        nearest_pairs[
            "genomics_after_cxr"
        ].sum()
    )
)

assert (
    nearest_pairs[
        [
            "genomics_before_cxr",
            "same_day",
            "genomics_after_cxr",
        ]
    ]
    .sum(axis=1)
    .eq(1)
    .all()
)

print("Temporal leakage ordering: PASS")


# ===================================================================================================
# 9. SAME-DAY RISK AUDIT
# ===================================================================================================

print("\n8. SAME-DAY TEMPORAL RISK AUDIT")
print("-" * 100)

same_day_cxr_urls = set(
    temporal_pairs.loc[
        temporal_pairs[
            "absolute_temporal_distance_days"
        ] == 0,
        "series_instance_content_url"
    ]
)

same_day_conditions = set(
    temporal_pairs.loc[
        temporal_pairs[
            "absolute_temporal_distance_days"
        ] == 0,
        "condition_id"
    ]
)

print(
    "CXR records with at least one same-day genomic timepoint:",
    len(same_day_cxr_urls)
)

print(
    "Conditions with at least one same-day relationship:",
    len(same_day_conditions)
)

print(
    "Same-day relationships are flagged for later study-design review."
)

print(
    "No same-day exclusion is applied."
)

print("Same-day risk audit: PASS")


# ===================================================================================================
# 10. POST-GENOMIC CXR AUDIT
# ===================================================================================================

print("\n9. POST-GENOMIC CXR AUDIT")
print("-" * 100)

post_genomic_cxr_urls = set(
    nearest_pairs.loc[
        nearest_pairs[
            "signed_temporal_delta_days"
        ] > 0,
        "series_instance_content_url"
    ]
)

post_genomic_conditions = set(
    nearest_pairs.loc[
        nearest_pairs[
            "signed_temporal_delta_days"
        ] > 0,
        "condition_id"
    ]
)

print(
    "CXRs whose nearest genomic timepoint is before the CXR:",
    len(post_genomic_cxr_urls)
)

print(
    "Conditions with such CXR relationships:",
    len(post_genomic_conditions)
)

print(
    "These records are flagged for temporal study-design review."
)

print(
    "No post-genomic CXR exclusion is applied."
)

print("Post-genomic CXR audit: PASS")


# ===================================================================================================
# 11. MULTIPLE-CXR REPEATED-MEASUREMENT AUDIT
# ===================================================================================================

print("\n10. MULTIPLE-CXR REPEATED-MEASUREMENT AUDIT")
print("-" * 100)

cxr_count_by_condition = (
    cxr
    .groupby(
        "condition_id"
    )
    .size()
)

multiple_cxr_conditions = (
    cxr_count_by_condition[
        cxr_count_by_condition > 1
    ]
)

print(
    "Conditions with multiple CXRs:",
    len(multiple_cxr_conditions)
)

print(
    "CXR records belonging to multiple-CXR conditions:",
    int(
        cxr[
            "condition_id"
        ]
        .isin(
            multiple_cxr_conditions.index
        )
        .sum()
    )
)

assert (
    len(multiple_cxr_conditions)
    == 485
)

print(
    "Repeated CXR measurements are identified."
)

print(
    "They are NOT automatically removed."
)

print("Multiple-CXR leakage audit: PASS")


# ===================================================================================================
# 12. MULTIPLE-GENOMIC-TIMEPOINT AUDIT
# ===================================================================================================

print("\n11. MULTIPLE-GENOMIC-TIMEPOINT AUDIT")
print("-" * 100)

genomic_count_by_condition = (
    genomic_timepoints
    .groupby(
        "condition_id"
    )
    .size()
)

multiple_genomic_conditions = (
    genomic_count_by_condition[
        genomic_count_by_condition > 1
    ]
)

print(
    "Conditions with multiple genomic timepoints:",
    len(multiple_genomic_conditions)
)

assert (
    len(multiple_genomic_conditions)
    == 152
)

print(
    "Multiple genomic measurements are retained as longitudinal evidence."
)

print(
    "They are NOT automatically collapsed into one biological state."
)

print("Multiple-genomic leakage audit: PASS")


# ===================================================================================================
# 13. BIDIRECTIONAL TEMPORAL AUDIT
# ===================================================================================================

print("\n12. BIDIRECTIONAL TEMPORAL AUDIT")
print("-" * 100)

bidirectional_conditions = set(
    condition_direction.loc[
        condition_direction[
            "temporal_complexity_class"
        ].isin(
            [
                "BEFORE_AND_AFTER",
                "BEFORE_SAME_DAY_AFTER",
            ]
        ),
        "condition_id"
    ]
)

print(
    "Conditions with genomic relationships on both sides of CXR timing:",
    len(bidirectional_conditions)
)

assert (
    len(bidirectional_conditions)
    == 192
)

print(
    "Bidirectional conditions require temporal-aware study design."
)

print(
    "No bidirectional condition is excluded here."
)

print("Bidirectional temporal audit: PASS")


# ===================================================================================================
# 14. TEMPORAL COMPLEXITY AUDIT
# ===================================================================================================

print("\n13. TEMPORAL COMPLEXITY AUDIT")
print("-" * 100)

temporal_complexity_counts = (
    temporal_flags[
        "temporal_complexity_flag"
    ]
    .value_counts()
    .rename_axis(
        "temporal_complexity_flag"
    )
    .reset_index(
        name="condition_count"
    )
)

print(
    temporal_complexity_counts.to_string(
        index=False
    )
)

assert (
    temporal_complexity_counts[
        "condition_count"
    ].sum()
    == 3081
)

print("Temporal complexity audit: PASS")


# ===================================================================================================
# 15. OUTCOME FIELD AUDIT
# ===================================================================================================
#
# Outcome is generally post-treatment / post-disease-course information.
#
# Since the prediction target is not selected, we do not decide here whether
# it will eventually be a target.
#
# We only document its presence and distribution.
#
# ===================================================================================================

print("\n14. OUTCOME FIELD AUDIT")
print("-" * 100)

assert "outcome" in cxr.columns

outcome_nonmissing = (
    cxr[
        "outcome"
    ]
    .notna()
    .sum()
)

outcome_missing = (
    cxr[
        "outcome"
    ]
    .isna()
    .sum()
)

print(
    "CXR records with outcome:",
    outcome_nonmissing
)

print(
    "CXR records without outcome:",
    outcome_missing
)

print(
    "\nOutcome distribution:"
)

print(
    cxr[
        "outcome"
    ]
    .astype(str)
    .value_counts()
    .to_string()
)

print(
    "\nOutcome is retained as metadata for audit."
)

print(
    "Outcome is NOT designated as an input feature."
)

print("Outcome leakage audit: PASS")


# ===================================================================================================
# 16. RESISTANCE FIELD AUDIT
# ===================================================================================================

print("\n15. RESISTANCE FIELD AUDIT")
print("-" * 100)

resistance_columns = []

for column in [
    "type_of_resistance",
    "drug_resistance_type",
]:

    if column in cxr.columns:

        resistance_columns.append(
            ("CXR", column)
        )

    if column in genomics.columns:

        resistance_columns.append(
            ("GENOMICS", column)
        )

resistance_audit_rows = []

for dataset_name, column in resistance_columns:

    if dataset_name == "CXR":

        series = cxr[column]

    else:

        series = genomics[column]

    resistance_audit_rows.append(
        {
            "dataset": dataset_name,
            "column": column,
            "nonmissing_records": int(
                series.notna().sum()
            ),
            "unique_nonmissing_values": int(
                series.dropna().nunique()
            ),
            "status": "TARGET_OR_TARGET_PROXY_REVIEW_REQUIRED",
        }
    )

resistance_audit = pd.DataFrame(
    resistance_audit_rows
)

print(
    resistance_audit.to_string(
        index=False
    )
)

print(
    "\nResistance-related fields are NOT automatically permitted as predictors."
)

print("Resistance leakage audit: PASS")


# ===================================================================================================
# 17. CROSS-DATASET RESISTANCE CONSISTENCY AUDIT
# ===================================================================================================
#
# This is NOT used to create a target.
#
# It only checks whether the same condition has resistance-related fields
# represented in both CXR metadata and genomics.
#
# ===================================================================================================

print("\n16. CROSS-DATASET RESISTANCE CONSISTENCY AUDIT")
print("-" * 100)

cxr_resistance = cxr[
    [
        "condition_id",
        "type_of_resistance",
    ]
].copy()

genomic_resistance_columns = [
    column
    for column in [
        "condition_id",
        "drug_resistance_type",
        "type_of_resistance",
    ]
    if column in genomics.columns
]

genomic_resistance = genomics[
    genomic_resistance_columns
].copy()

genomic_resistance[
    "condition_id"
] = (
    genomic_resistance[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)

cxr_resistance[
    "condition_id"
] = (
    cxr_resistance[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)

resistance_condition_overlap = (
    set(
        cxr_resistance[
            "condition_id"
        ]
    )
    &
    set(
        genomic_resistance[
            "condition_id"
        ]
    )
)

print(
    "Conditions with CXR resistance metadata:",
    cxr_resistance[
        "type_of_resistance"
    ]
    .notna()
    .groupby(
        cxr_resistance[
            "condition_id"
        ]
    )
    .any()
    .sum()
)

print(
    "Conditions with genomic resistance metadata:",
    len(
        resistance_condition_overlap
    )
)

print(
    "This overlap is documented only for leakage assessment."
)

print("Cross-dataset resistance audit: PASS")


# ===================================================================================================
# 18. CONDITION-LEVEL REPEATED INFORMATION AUDIT
# ===================================================================================================

print("\n17. CONDITION-LEVEL REPEATED INFORMATION AUDIT")
print("-" * 100)

condition_static_columns = [
    "patient_id",
    "age_of_onset",
    "sex",
    "case_definition",
    "diagnosis_code",
    "type_of_resistance",
    "comorbidity",
    "country",
    "bmi",
    "outcome",
]

available_static = [
    column
    for column in condition_static_columns
    if column in cxr.columns
]

repetition_rows = []

for column in available_static:

    unique_values_per_condition = (
        cxr
        .groupby(
            "condition_id"
        )[column]
        .nunique(
            dropna=True
        )
    )

    repeated_condition_count = int(
        (
            unique_values_per_condition
            <= 1
        ).sum()
    )

    variable_conditions = int(
        (
            unique_values_per_condition
            > 1
        ).sum()
    )

    repetition_rows.append(
        {
            "column": column,
            "conditions_with_single_or_constant_nonmissing_value":
                repeated_condition_count,
            "conditions_with_multiple_values":
                variable_conditions,
        }
    )

repetition_audit = pd.DataFrame(
    repetition_rows
)

print(
    repetition_audit.to_string(
        index=False
    )
)

print(
    "\nRepeated condition-level metadata must be handled at case level."
)

print("Repeated-information audit: PASS")


# ===================================================================================================
# 19. PATIENT-ID / CONDITION-ID RELATIONSHIP
# ===================================================================================================
#
# condition_id is the intended case-level unit.
#
# We inspect whether patient_id maps to multiple conditions.
#
# This is critical because random image-level splitting can otherwise place
# observations from the same patient/case into different partitions.
#
# ===================================================================================================

print("\n18. PATIENT ↔ CONDITION STRUCTURE AUDIT")
print("-" * 100)

patient_condition = (
    cxr[
        [
            "patient_id",
            "condition_id",
        ]
    ]
    .drop_duplicates()
)

conditions_per_patient = (
    patient_condition
    .groupby(
        "patient_id"
    )[
        "condition_id"
    ]
    .nunique()
)

print(
    "Unique patients:",
    patient_condition[
        "patient_id"
    ].nunique()
)

print(
    "Unique conditions:",
    patient_condition[
        "condition_id"
    ].nunique()
)

print(
    "Patients associated with >1 condition:",
    int(
        (
            conditions_per_patient > 1
        ).sum()
    )
)

print(
    "\ncondition_id remains the primary case-level grouping key."
)

print("Patient-condition structure: PASS")


# ===================================================================================================
# 20. URL / IMAGE DUPLICATION AUDIT
# ===================================================================================================

print("\n19. IMAGE DUPLICATION AUDIT")
print("-" * 100)

url_counts = (
    cxr[
        "series_instance_content_url"
    ]
    .value_counts()
)

duplicate_urls = url_counts[
    url_counts > 1
]

print(
    "Total CXR records:",
    len(cxr)
)

print(
    "Unique CXR URLs:",
    cxr[
        "series_instance_content_url"
    ].nunique()
)

print(
    "Duplicated CXR URLs:",
    len(duplicate_urls)
)

assert (
    len(duplicate_urls)
    == 0
)

print(
    "No duplicated CXR filepath/linkage keys detected."
)

print("Image duplication audit: PASS")


# ===================================================================================================
# 21. GENOMIC TIMEPOINT DUPLICATION AUDIT
# ===================================================================================================

print("\n20. GENOMIC TIMEPOINT DUPLICATION AUDIT")
print("-" * 100)

timepoint_key = [
    "condition_id",
    "specimen_id",
    "specimen_collection_date_num",
]

timepoint_duplicate_count = (
    genomic_timepoints[
        timepoint_key
    ]
    .duplicated()
    .sum()
)

print(
    "Duplicate unique-timepoint keys:",
    int(
        timepoint_duplicate_count
    )
)

assert (
    timepoint_duplicate_count
    == 0
)

print(
    "Unique genomic temporal table contains no duplicate keys."
)

print("Genomic timepoint duplication audit: PASS")


# ===================================================================================================
# 22. TEMPORAL PAIR MULTIPLICITY AUDIT
# ===================================================================================================
#
# Multiple CXRs × multiple genomic timepoints naturally creates many-to-many
# temporal pairs.
#
# This is not a duplicate-image problem.
#
# ===================================================================================================

print("\n21. TEMPORAL PAIR MULTIPLICITY AUDIT")
print("-" * 100)

pair_counts_per_cxr = (
    temporal_pairs
    .groupby(
        "series_instance_content_url"
    )
    .size()
)

pair_counts_per_condition = (
    temporal_pairs
    .groupby(
        "condition_id"
    )
    .size()
)

print(
    "CXR records:",
    len(pair_counts_per_cxr)
)

print(
    "Conditions:",
    len(pair_counts_per_condition)
)

print(
    "Maximum genomic comparisons for one CXR:",
    int(
        pair_counts_per_cxr.max()
    )
)

print(
    "Maximum temporal pairs for one condition:",
    int(
        pair_counts_per_condition.max()
    )
)

assert (
    len(pair_counts_per_cxr)
    == 3859
)

assert (
    len(pair_counts_per_condition)
    == 3081
)

print(
    "Many-to-many temporal structure is retained for audit."
)

print("Temporal pair multiplicity audit: PASS")


# ===================================================================================================
# 23. POTENTIAL LEAKAGE RISK REGISTER
# ===================================================================================================

print("\n22. POTENTIAL LEAKAGE RISK REGISTER")
print("-" * 100)

risk_register = pd.DataFrame(
    [
        {
            "risk_category": "DIRECT_TARGET_FIELD",
            "example": "type_of_resistance / drug_resistance_type",
            "risk_status": "POTENTIAL",
            "current_action": "ISOLATED_FOR_TARGET_REVIEW",
            "automatic_exclusion": "NO",
        },
        {
            "risk_category": "POST_TREATMENT_OUTCOME",
            "example": "outcome",
            "risk_status": "POTENTIAL",
            "current_action": "ISOLATED_FOR_TARGET_REVIEW",
            "automatic_exclusion": "NO",
        },
        {
            "risk_category": "IDENTIFIER_LEAKAGE",
            "example": "condition_id / patient_id / specimen_id",
            "risk_status": "POTENTIAL",
            "current_action": "LINKAGE_ONLY",
            "automatic_exclusion": "NO",
        },
        {
            "risk_category": "REPEATED_CXR",
            "example": "multiple CXRs per condition",
            "risk_status": "POTENTIAL",
            "current_action": "CASE_LEVEL_AUDIT",
            "automatic_exclusion": "NO",
        },
        {
            "risk_category": "REPEATED_GENOMICS",
            "example": "multiple genomic timepoints",
            "risk_status": "POTENTIAL",
            "current_action": "LONGITUDINAL_AUDIT",
            "automatic_exclusion": "NO",
        },
        {
            "risk_category": "TEMPORAL_ORDER",
            "example": "genomics before/after CXR",
            "risk_status": "POTENTIAL",
            "current_action": "TEMPORAL_AUDIT",
            "automatic_exclusion": "NO",
        },
        {
            "risk_category": "SAME_DAY_MEASUREMENTS",
            "example": "CXR and genomic specimen same day",
            "risk_status": "POTENTIAL",
            "current_action": "TEMPORAL_AUDIT",
            "automatic_exclusion": "NO",
        },
        {
            "risk_category": "POST_GENOMIC_IMAGING",
            "example": "CXR after genomic sampling",
            "risk_status": "POTENTIAL",
            "current_action": "TEMPORAL_AUDIT",
            "automatic_exclusion": "NO",
        },
        {
            "risk_category": "PATIENT_SPLIT_LEAKAGE",
            "example": "same condition/patient across partitions",
            "risk_status": "POTENTIAL",
            "current_action": "GROUPED_SPLIT_REQUIRED",
            "automatic_exclusion": "NO",
        },
    ]
)

print(
    risk_register.to_string(
        index=False
    )
)

print("Leakage risk register: PASS")


# ===================================================================================================
# 24. CURRENT STUDY-DESIGN LOCKS
# ===================================================================================================

print("\n23. CURRENT STUDY-DESIGN LOCKS")
print("-" * 100)

study_design_locks = {
    "final_prediction_target": "NOT_SELECTED",
    "final_feature_set": "NOT_SELECTED",
    "final_temporal_window": "NOT_SELECTED",
    "final_cxr_selection_rule": "NOT_SELECTED",
    "final_genomic_selection_rule": "NOT_SELECTED",
    "final_research_cohort": "NOT_SELECTED",
    "image_preprocessing": "NOT_STARTED",
    "model_training": "NOT_STARTED",
}

for key, value in study_design_locks.items():

    print(
        f"{key:<35}: {value}"
    )

print("Study-design locks: PASS")


# ===================================================================================================
# 25. AUTOMATIC EXCLUSION SAFETY
# ===================================================================================================

print("\n24. AUTOMATIC EXCLUSION SAFETY")
print("-" * 100)

print(
    "Direct resistance fields used for exclusion : NO"
)

print(
    "Outcome used for exclusion                  : NO"
)

print(
    "Identifiers used for exclusion              : NO"
)

print(
    "Genomics-before-CXR used for exclusion      : NO"
)

print(
    "Genomics-after-CXR used for exclusion       : NO"
)

print(
    "Same-day relationship used for exclusion    : NO"
)

print(
    "Multiple-CXR condition excluded             : NO"
)

print(
    "Multiple-genomic condition excluded         : NO"
)

print(
    "Bidirectional temporal condition excluded   : NO"
)

print(
    "Temporal distance threshold applied         : NO"
)

print(
    "cxr_outlier used as automatic exclusion     : NO"
)

print(
    "Modality used as automatic exclusion        : NO"
)

print("Automatic exclusion safety: PASS")


# ===================================================================================================
# 26. SAVE OUTPUTS
# ===================================================================================================

print("\n25. SAVING OUTPUTS")
print("-" * 100)

IDENTIFIER_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_36_Identifier_Leakage_Audit.csv"
)

CXRTARGET_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_36_CXR_Target_Like_Field_Audit.csv"
)

GENOMICTARGET_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_36_Genomic_Target_Like_Field_Audit.csv"
)

TEMPORAL_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_36_Nearest_Temporal_Leakage_Audit.csv"
)

RESISTANCE_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_36_Resistance_Field_Audit.csv"
)

REPETITION_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_36_Repeated_Information_Audit.csv"
)

PATIENT_CONDITION_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_36_Patient_Condition_Structure.csv"
)

PAIR_MULTIPLICITY_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_36_Temporal_Pair_Multiplicity.csv"
)

RISK_REGISTER_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_36_Leakage_Risk_Register.csv"
)

STUDY_LOCK_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_36_Study_Design_Locks.csv"
)


identifier_audit.to_csv(
    IDENTIFIER_OUTPUT,
    index=False
)

pd.DataFrame(
    {
        "dataset": ["CXR"] * len(cxr_target_like_present),
        "column": cxr_target_like_present,
        "status": "TARGET_OR_TARGET_PROXY_REVIEW_REQUIRED",
    }
).to_csv(
    CXRTARGET_OUTPUT,
    index=False
)

pd.DataFrame(
    {
        "dataset": ["GENOMICS"] * len(genomic_target_like_present),
        "column": genomic_target_like_present,
        "status": "TARGET_OR_TARGET_PROXY_REVIEW_REQUIRED",
    }
).to_csv(
    GENOMICTARGET_OUTPUT,
    index=False
)

nearest_pairs[
    [
        "condition_id",
        "series_instance_content_url",
        "specimen_id",
        "signed_temporal_delta_days",
        "absolute_temporal_distance_days",
        "temporal_direction",
        "genomics_before_cxr",
        "same_day",
        "genomics_after_cxr",
    ]
].to_csv(
    TEMPORAL_OUTPUT,
    index=False
)

resistance_audit.to_csv(
    RESISTANCE_OUTPUT,
    index=False
)

repetition_audit.to_csv(
    REPETITION_OUTPUT,
    index=False
)

patient_condition.to_csv(
    PATIENT_CONDITION_OUTPUT,
    index=False
)

pair_multiplicity = (
    temporal_pairs
    .groupby(
        "series_instance_content_url"
    )
    .size()
    .rename(
        "genomic_comparisons_per_cxr"
    )
    .reset_index()
)

pair_multiplicity.to_csv(
    PAIR_MULTIPLICITY_OUTPUT,
    index=False
)

risk_register.to_csv(
    RISK_REGISTER_OUTPUT,
    index=False
)

pd.DataFrame(
    [
        {
            "study_design_element": key,
            "current_status": value,
        }
        for key, value
        in study_design_locks.items()
    ]
).to_csv(
    STUDY_LOCK_OUTPUT,
    index=False
)


# ===================================================================================================
# 27. OUTPUT VALIDATION
# ===================================================================================================

print("\n26. OUTPUT FILE VALIDATION")
print("-" * 100)

output_paths = [
    IDENTIFIER_OUTPUT,
    CXRTARGET_OUTPUT,
    GENOMICTARGET_OUTPUT,
    TEMPORAL_OUTPUT,
    RESISTANCE_OUTPUT,
    REPETITION_OUTPUT,
    PATIENT_CONDITION_OUTPUT,
    PAIR_MULTIPLICITY_OUTPUT,
    RISK_REGISTER_OUTPUT,
    STUDY_LOCK_OUTPUT,
]

for output_path in output_paths:

    assert output_path.exists(), (
        f"Output missing:\n{output_path}"
    )

    assert output_path.stat().st_size > 0, (
        f"Output is empty:\n{output_path}"
    )

    print(
        f"{output_path.name}: PASS"
    )


# ===================================================================================================
# 28. FINAL INTEGRITY CHECKS
# ===================================================================================================

print("\n27. FINAL INTEGRITY CHECKS")
print("-" * 100)

assert (
    len(cxr)
    == 3859
)

assert (
    cxr[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    temporal_pairs[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    len(temporal_flags)
    == 3081
)

assert (
    len(condition_direction)
    == 3081
)

assert (
    len(multiple_cxr_conditions)
    == 485
)

assert (
    len(multiple_genomic_conditions)
    == 152
)

assert (
    len(bidirectional_conditions)
    == 192
)

print(
    "3,859 CXR records                         : PASS"
)

print(
    "3,081 multimodal conditions                : PASS"
)

print(
    "485 multiple-CXR conditions                : PASS"
)

print(
    "152 multiple-genomic-timepoint conditions  : PASS"
)

print(
    "192 bidirectional temporal conditions      : PASS"
)

print(
    "Target not selected                        : PASS"
)

print(
    "No feature set selected                   : PASS"
)

print(
    "No final cohort selected                  : PASS"
)

print(
    "No automatic leakage exclusion applied    : PASS"
)

print("Final integrity checks: PASS")


# ===================================================================================================
# 29. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.36 — FINAL STATUS")
print("=" * 100)

print(
    "Identifier leakage audit                  : PASS"
)

print(
    "CXR target-like field audit               : PASS"
)

print(
    "Genomic target-like field audit           : PASS"
)

print(
    "Temporal leakage audit                    : PASS"
)

print(
    "Same-day risk audit                       : PASS"
)

print(
    "Post-genomic CXR audit                    : PASS"
)

print(
    "Multiple-CXR repeated-measurement audit   : PASS"
)

print(
    "Multiple-genomic-timepoint audit          : PASS"
)

print(
    "Bidirectional temporal audit              : PASS"
)

print(
    "Outcome leakage audit                     : PASS"
)

print(
    "Resistance-field leakage audit            : PASS"
)

print(
    "Patient-condition structure audit         : PASS"
)

print(
    "Temporal pair multiplicity audit          : PASS"
)

print(
    "Leakage-risk register                     : PASS"
)

print(
    "No automatic exclusion                    : PASS"
)

print(
    "No target selected                        : PASS"
)

print(
    "No feature set selected                   : PASS"
)

print(
    "No temporal threshold selected            : PASS"
)

print(
    "No final cohort selected                  : PASS"
)

print(
    "No preprocessing performed                : PASS"
)

print(
    "No model training                         : PASS"
)

print(
    "DICOM files modified                      : NO"
)

print(
    "Source metadata modified                  : NO"
)

print(
    "Output files                              : PASS"
)

print("\n" + "=" * 100)
print("STEP 3B.6.36 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.37 — CASE-LEVEL ELIGIBILITY CANDIDATE RECONCILIATION
# COMPLETE SELF-CONTAINED VERSION
# ===================================================================================================
#
# PURPOSE
# -------
# Construct a condition-level eligibility evidence framework from:
#
#   1. CXR master
#   2. Final image-level semantic/structural evidence
#   3. Condition-level eligibility evidence
#   4. Multiple-CXR structure
#   5. Genomic temporal structure
#   6. Temporal directionality
#   7. Leakage-risk audit
#
# IMPORTANT:
# ----------
# This step creates CANDIDATE STRATA.
#
# It does NOT select the final research cohort.
#
# It does NOT select:
#   - final prediction target
#   - representative CXR
#   - representative genomic specimen
#   - temporal eligibility threshold
#   - final feature set
#
# It does NOT:
#   - exclude all unreviewed images
#   - exclude all XC images
#   - exclude all OT images
#   - exclude all cxr_outlier records
#   - exclude multiple-CXR conditions
#   - exclude multiple-genomic-timepoint conditions
#
# CONDITION_ID IS THE PRIMARY CASE-LEVEL UNIT.
#
# ===================================================================================================


from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 100)
print("STEP 3B.6.37 — CASE-LEVEL ELIGIBILITY CANDIDATE RECONCILIATION")
print("=" * 100)


# ===================================================================================================
# 1. INITIALIZATION
# ===================================================================================================

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_024_DIR = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03"
)

STEP_030_DIR = (
    METADATA_DIR /
    "Step_3B_6_30_Image_Level_CXR_Eligibility"
)

STEP_032_DIR = (
    METADATA_DIR /
    "Step_3B_6_32_Condition_Level_Eligibility_Evidence"
)

STEP_033_DIR = (
    METADATA_DIR /
    "Step_3B_6_33_Multiple_CXR_Condition_Analysis"
)

STEP_034_DIR = (
    METADATA_DIR /
    "Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
)

STEP_035_DIR = (
    METADATA_DIR /
    "Step_3B_6_35_Temporal_Directionality_Audit"
)

STEP_036_DIR = (
    METADATA_DIR /
    "Step_3B_6_36_Leakage_Risk_Audit"
)


# ---------------------------------------------------------------------------------------------------
# INPUTS
# ---------------------------------------------------------------------------------------------------

CXR_MASTER_PATH = (
    STEP_024_DIR /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

IMAGE_AUDIT_PATH = (
    STEP_030_DIR /
    "TB_Portals_March2025_Step_3B_6_30_Image_Level_CXR_Eligibility_Audit.csv"
)

CONDITION_EVIDENCE_PATH = (
    STEP_032_DIR /
    "TB_Portals_March2025_Step_3B_6_32_Condition_Level_Eligibility_Evidence.csv"
)

MULTIPLE_CXR_MASTER_PATH = (
    STEP_033_DIR /
    "TB_Portals_March2025_Step_3B_6_33_Multiple_CXR_Condition_Master.csv"
)

TEMPORAL_PAIRS_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

NEAREST_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Nearest_Genomic_Timepoint_Per_CXR.csv"
)

GENOMIC_TIMEPOINT_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

TEMPORAL_FLAGS_PATH = (
    STEP_035_DIR /
    "TB_Portals_March2025_Step_3B_6_35_Temporal_Complexity_Flags.csv"
)

CONDITION_DIRECTION_PATH = (
    STEP_035_DIR /
    "TB_Portals_March2025_Step_3B_6_35_Condition_Temporal_Direction_Structure.csv"
)

RISK_REGISTER_PATH = (
    STEP_036_DIR /
    "TB_Portals_March2025_Step_3B_6_36_Leakage_Risk_Register.csv"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_37_Case_Level_Eligibility_Candidates"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ===================================================================================================
# 2. INPUT ARTIFACT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

required_paths = {
    "CXR master": CXR_MASTER_PATH,
    "Step 3B.6.30 image audit": IMAGE_AUDIT_PATH,
    "Step 3B.6.32 condition evidence": CONDITION_EVIDENCE_PATH,
    "Step 3B.6.33 multiple-CXR master": MULTIPLE_CXR_MASTER_PATH,
    "Step 3B.6.34 temporal pairs": TEMPORAL_PAIRS_PATH,
    "Step 3B.6.34 nearest pairs": NEAREST_PATH,
    "Step 3B.6.34 genomic timepoints": GENOMIC_TIMEPOINT_PATH,
    "Step 3B.6.35 temporal flags": TEMPORAL_FLAGS_PATH,
    "Step 3B.6.35 condition direction": CONDITION_DIRECTION_PATH,
    "Step 3B.6.36 risk register": RISK_REGISTER_PATH,
}

for label, path in required_paths.items():

    assert path.exists(), (
        f"Required artifact not found:\n{path}"
    )

    print(
        f"{label:<42}: FOUND"
    )

print("Input artifact validation: PASS")


# ===================================================================================================
# 3. LOAD ARTIFACTS
# ===================================================================================================

cxr = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

image_audit = pd.read_csv(
    IMAGE_AUDIT_PATH,
    low_memory=False
)

condition_evidence = pd.read_csv(
    CONDITION_EVIDENCE_PATH,
    low_memory=False
)

multiple_cxr_master = pd.read_csv(
    MULTIPLE_CXR_MASTER_PATH,
    low_memory=False
)

temporal_pairs = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)

nearest_pairs = pd.read_csv(
    NEAREST_PATH,
    low_memory=False
)

genomic_timepoints = pd.read_csv(
    GENOMIC_TIMEPOINT_PATH,
    low_memory=False
)

temporal_flags = pd.read_csv(
    TEMPORAL_FLAGS_PATH,
    low_memory=False
)

condition_direction = pd.read_csv(
    CONDITION_DIRECTION_PATH,
    low_memory=False
)

risk_register = pd.read_csv(
    RISK_REGISTER_PATH,
    low_memory=False
)


print("\n2. LOADED ARTIFACTS")
print("-" * 100)

print(
    "CXR records:",
    len(cxr)
)

print(
    "Image audit records:",
    len(image_audit)
)

print(
    "Condition evidence records:",
    len(condition_evidence)
)

print(
    "Multiple-CXR records:",
    len(multiple_cxr_master)
)

print(
    "Temporal pair records:",
    len(temporal_pairs)
)

print(
    "Nearest temporal rows:",
    len(nearest_pairs)
)

print(
    "Genomic timepoints:",
    len(genomic_timepoints)
)

print(
    "Temporal flag conditions:",
    len(temporal_flags)
)

print(
    "Condition direction records:",
    len(condition_direction)
)


# ===================================================================================================
# 4. NORMALIZE CONDITION IDS
# ===================================================================================================

print("\n3. CONDITION-ID NORMALIZATION")
print("-" * 100)

datasets_with_condition = [
    ("CXR", cxr),
    ("IMAGE_AUDIT", image_audit),
    ("CONDITION_EVIDENCE", condition_evidence),
    ("MULTIPLE_CXR", multiple_cxr_master),
    ("TEMPORAL_PAIRS", temporal_pairs),
    ("NEAREST", nearest_pairs),
    ("GENOMIC_TIMEPOINTS", genomic_timepoints),
    ("TEMPORAL_FLAGS", temporal_flags),
    ("CONDITION_DIRECTION", condition_direction),
]

for dataset_name, df in datasets_with_condition:

    assert "condition_id" in df.columns, (
        f"{dataset_name} does not contain condition_id"
    )

    df["condition_id"] = (
        df["condition_id"]
        .astype(str)
        .str.strip()
    )

print("All condition identifiers normalized: PASS")


# ===================================================================================================
# 5. CORE POPULATION VALIDATION
# ===================================================================================================

print("\n4. CORE POPULATION VALIDATION")
print("-" * 100)

assert len(cxr) == 3859

assert (
    cxr[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    cxr[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

print(
    "CXR records       : 3,859"
)

print(
    "Unique conditions : 3,081"
)

print(
    "Unique CXR URLs   : 3,859"
)

print("Core population: PASS")


# ===================================================================================================
# 6. IMAGE-LEVEL EVIDENCE VALIDATION
# ===================================================================================================

print("\n5. IMAGE-LEVEL EVIDENCE VALIDATION")
print("-" * 100)

required_image_columns = [
    "condition_id",
    "series_instance_content_url",
    "image_level_evidence_status",
]

missing_image_columns = [
    column
    for column in required_image_columns
    if column not in image_audit.columns
]

print(
    "Missing image-audit columns:",
    missing_image_columns
)

assert len(missing_image_columns) == 0

assert (
    image_audit[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    image_audit[
        "condition_id"
    ].nunique()
    == 3081
)

print(
    "Image audit records:",
    len(image_audit)
)

print(
    "Image URLs:",
    image_audit[
        "series_instance_content_url"
    ].nunique()
)

print("Image-level evidence: PASS")


# ===================================================================================================
# 7. CONDITION-LEVEL EVIDENCE VALIDATION
# ===================================================================================================

print("\n6. CONDITION-LEVEL EVIDENCE VALIDATION")
print("-" * 100)

required_condition_columns = [
    "condition_id",
    "condition_evidence_class",
]

missing_condition_columns = [
    column
    for column in required_condition_columns
    if column not in condition_evidence.columns
]

print(
    "Missing condition-evidence columns:",
    missing_condition_columns
)

assert len(missing_condition_columns) == 0

assert (
    condition_evidence[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    len(condition_evidence)
    == 3081
)

print(
    "Condition evidence records:",
    len(condition_evidence)
)

print(
    "Unique conditions:",
    condition_evidence[
        "condition_id"
    ].nunique()
)

print(
    "\nExisting condition evidence classes:"
)

print(
    condition_evidence[
        "condition_evidence_class"
    ]
    .value_counts()
    .to_string()
)

print("Condition-level evidence: PASS")


# ===================================================================================================
# 8. MULTIPLE-CXR STRUCTURE
# ===================================================================================================

print("\n7. MULTIPLE-CXR STRUCTURE")
print("-" * 100)

multiple_cxr_ids = set(
    multiple_cxr_master[
        "condition_id"
    ]
)

assert len(multiple_cxr_ids) == 485

cxr_counts = (
    cxr
    .groupby(
        "condition_id"
    )
    .size()
    .rename(
        "cxr_count"
    )
)

assert len(cxr_counts) == 3081

print(
    "Conditions with multiple CXRs:",
    int(
        (
            cxr_counts > 1
        ).sum()
    )
)

assert (
    (
        cxr_counts > 1
    ).sum()
    == 485
)

print("Multiple-CXR structure: PASS")


# ===================================================================================================
# 9. MULTIPLE-GENOMIC STRUCTURE
# ===================================================================================================

print("\n8. MULTIPLE-GENOMIC STRUCTURE")
print("-" * 100)

genomic_counts = (
    genomic_timepoints
    .groupby(
        "condition_id"
    )
    .size()
    .rename(
        "genomic_timepoint_count"
    )
)

assert (
    len(genomic_counts)
    == 3081
)

multiple_genomic_ids = set(
    genomic_counts[
        genomic_counts > 1
    ].index
)

assert (
    len(multiple_genomic_ids)
    == 152
)

print(
    "Conditions with genomic timepoints:",
    len(genomic_counts)
)

print(
    "Conditions with multiple genomic timepoints:",
    len(multiple_genomic_ids)
)

print("Multiple-genomic structure: PASS")


# ===================================================================================================
# 10. IMAGE-LEVEL EVIDENCE AGGREGATION
# ===================================================================================================
#
# IMPORTANT:
#
# We preserve all image-level evidence.
#
# A condition with:
#   one retained image
#   one excluded image
#
# is NOT automatically classified as an excluded condition.
#
# Instead, it becomes a MIXED_EVIDENCE condition.
#
# ===================================================================================================

print("\n9. AGGREGATING IMAGE-LEVEL EVIDENCE TO CONDITION LEVEL")
print("-" * 100)

image_audit[
    "image_level_evidence_status"
] = (
    image_audit[
        "image_level_evidence_status"
    ]
    .astype(str)
    .str.strip()
)

image_status_per_condition = (
    image_audit
    .groupby(
        "condition_id"
    )[
        "image_level_evidence_status"
    ]
    .agg(
        lambda x:
        sorted(
            set(x)
        )
    )
)

image_condition_summary = (
    image_status_per_condition
    .rename(
        "image_evidence_states"
    )
    .reset_index()
)

image_condition_summary[
    "image_evidence_state_signature"
] = (
    image_condition_summary[
        "image_evidence_states"
    ]
    .apply(
        lambda x:
        "|".join(x)
    )
)

image_condition_summary[
    "has_human_semantic_retention"
] = (
    image_condition_summary[
        "image_evidence_states"
    ]
    .apply(
        lambda x:
        "HUMAN_SEMANTIC_RETENTION" in x
    )
)

image_condition_summary[
    "has_human_semantic_exclusion"
] = (
    image_condition_summary[
        "image_evidence_states"
    ]
    .apply(
        lambda x:
        "HUMAN_SEMANTIC_EXCLUSION" in x
    )
)

image_condition_summary[
    "has_human_conflict_resolved"
] = (
    image_condition_summary[
        "image_evidence_states"
    ]
    .apply(
        lambda x:
        "HUMAN_CONFLICT_RESOLVED" in x
    )
)

image_condition_summary[
    "has_unreviewed_images"
] = (
    image_condition_summary[
        "image_evidence_states"
    ]
    .apply(
        lambda x:
        "NOT_INDIVIDUALLY_ADJUDICATED" in x
    )
)

image_condition_summary[
    "has_mixed_retained_excluded_evidence"
] = (
    image_condition_summary[
        "has_human_semantic_retention"
    ]
    &
    image_condition_summary[
        "has_human_semantic_exclusion"
    ]
)

assert (
    len(image_condition_summary)
    == 3081
)

print(
    "Conditions represented:",
    len(image_condition_summary)
)

print(
    "\nImage evidence signatures:"
)

print(
    image_condition_summary[
        "image_evidence_state_signature"
    ]
    .value_counts()
    .to_string()
)

print("Image-level condition aggregation: PASS")


# ===================================================================================================
# 11. TEMPORAL STRUCTURE AGGREGATION
# ===================================================================================================

print("\n10. TEMPORAL STRUCTURE AGGREGATION")
print("-" * 100)

temporal_condition_summary = (
    condition_direction[
        [
            "condition_id",
            "has_genomics_before_cxr",
            "has_same_day",
            "has_genomics_after_cxr",
            "temporal_direction_count",
            "temporal_complexity_class",
        ]
    ]
    .copy()
)

assert (
    len(temporal_condition_summary)
    == 3081
)

temporal_condition_summary[
    "bidirectional_temporal"
] = (
    temporal_condition_summary[
        "has_genomics_before_cxr"
    ]
    &
    temporal_condition_summary[
        "has_genomics_after_cxr"
    ]
)

print(
    "Conditions with genomics before CXR:",
    int(
        temporal_condition_summary[
            "has_genomics_before_cxr"
        ].sum()
    )
)

print(
    "Conditions with same-day relationship:",
    int(
        temporal_condition_summary[
            "has_same_day"
        ].sum()
    )
)

print(
    "Conditions with genomics after CXR:",
    int(
        temporal_condition_summary[
            "has_genomics_after_cxr"
        ].sum()
    )
)

print(
    "Bidirectional conditions:",
    int(
        temporal_condition_summary[
            "bidirectional_temporal"
        ].sum()
    )
)

assert (
    temporal_condition_summary[
        "bidirectional_temporal"
    ].sum()
    == 192
)

print("Temporal structure aggregation: PASS")


# ===================================================================================================
# 12. NEAREST TEMPORAL RELATIONSHIP
# ===================================================================================================

print("\n11. NEAREST TEMPORAL RELATIONSHIP")
print("-" * 100)

nearest_pairs[
    "absolute_temporal_distance_days"
] = pd.to_numeric(
    nearest_pairs[
        "absolute_temporal_distance_days"
    ],
    errors="coerce"
)

nearest_pairs[
    "signed_temporal_delta_days"
] = pd.to_numeric(
    nearest_pairs[
        "signed_temporal_delta_days"
    ],
    errors="coerce"
)

nearest_condition_distance = (
    nearest_pairs
    .groupby(
        "condition_id"
    )
    .agg(
        nearest_genomic_distance_days=(
            "absolute_temporal_distance_days",
            "min"
        ),

        nearest_genomic_distance_max_days=(
            "absolute_temporal_distance_days",
            "max"
        ),
    )
    .reset_index()
)

assert (
    len(nearest_condition_distance)
    == 3081
)

print(
    "Conditions with nearest temporal information:",
    len(nearest_condition_distance)
)

print(
    "\nNearest distance summary:"
)

print(
    nearest_condition_distance[
        "nearest_genomic_distance_days"
    ]
    .describe()
    .to_string()
)

print("Nearest temporal relationship: PASS")


# ===================================================================================================
# 13. TEMPORAL FLAGS
# ===================================================================================================

print("\n12. TEMPORAL FLAGS")
print("-" * 100)

temporal_flag_columns = [
    "condition_id",
    "temporal_complexity_flag",
]

missing_temporal_flags = [
    column
    for column in temporal_flag_columns
    if column not in temporal_flags.columns
]

assert len(missing_temporal_flags) == 0

temporal_flag_subset = temporal_flags[
    temporal_flag_columns
].copy()

assert (
    len(temporal_flag_subset)
    == 3081
)

print(
    temporal_flag_subset[
        "temporal_complexity_flag"
    ]
    .value_counts()
    .to_string()
)

print("Temporal flags: PASS")


# ===================================================================================================
# 14. BUILD CONDITION-LEVEL MASTER
# ===================================================================================================

print("\n13. BUILDING CONDITION-LEVEL CANDIDATE MASTER")
print("-" * 100)

condition_master = (
    cxr
    .groupby(
        "condition_id"
    )
    .agg(
        cxr_count=(
            "series_instance_content_url",
            "nunique"
        ),

        imaging_date_count=(
            "imaging_date",
            "nunique"
        ),

        modality_signature=(
            "series_modality_cd",
            lambda x:
            "|".join(
                sorted(
                    set(
                        x.dropna()
                        .astype(str)
                    )
                )
            )
        ),

        outlier_state_signature=(
            "cxr_outlier",
            lambda x:
            "|".join(
                sorted(
                    set(
                        x.dropna()
                        .astype(str)
                    )
                )
            )
        ),
    )
    .reset_index()
)

assert (
    len(condition_master)
    == 3081
)

condition_master[
    "multiple_cxr"
] = (
    condition_master[
        "cxr_count"
    ]
    > 1
)

condition_master[
    "multiple_imaging_dates"
] = (
    condition_master[
        "imaging_date_count"
    ]
    > 1
)

print(
    "Condition-level master records:",
    len(condition_master)
)

print(
    "Multiple-CXR conditions:",
    int(
        condition_master[
            "multiple_cxr"
        ].sum()
    )
)

print(
    "Multiple-date conditions:",
    int(
        condition_master[
            "multiple_imaging_dates"
        ].sum()
    )
)

assert (
    condition_master[
        "multiple_cxr"
    ].sum()
    == 485
)

print("Condition-level master: PASS")


# ===================================================================================================
# 15. MERGE CONDITION EVIDENCE
# ===================================================================================================

print("\n14. MERGING CONDITION-LEVEL EVIDENCE")
print("-" * 100)

candidate_master = (
    condition_master
    .merge(
        condition_evidence[
            [
                "condition_id",
                "condition_evidence_class",
            ]
        ],
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
)

assert (
    len(candidate_master)
    == 3081
)

assert (
    candidate_master[
        "condition_evidence_class"
    ].notna().all()
)

print("Condition evidence merge: PASS")


# ===================================================================================================
# 16. MERGE IMAGE-LEVEL AGGREGATED EVIDENCE
# ===================================================================================================

print("\n15. MERGING IMAGE-LEVEL EVIDENCE")
print("-" * 100)

candidate_master = (
    candidate_master
    .merge(
        image_condition_summary[
            [
                "condition_id",
                "image_evidence_state_signature",
                "has_human_semantic_retention",
                "has_human_semantic_exclusion",
                "has_human_conflict_resolved",
                "has_unreviewed_images",
                "has_mixed_retained_excluded_evidence",
            ]
        ],
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
)

assert (
    len(candidate_master)
    == 3081
)

print("Image evidence merge: PASS")


# ===================================================================================================
# 17. MERGE GENOMIC STRUCTURE
# ===================================================================================================

print("\n16. MERGING GENOMIC STRUCTURE")
print("-" * 100)

candidate_master = (
    candidate_master
    .merge(
        genomic_counts
        .reset_index(),
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
)

candidate_master[
    "genomic_timepoint_count"
] = (
    candidate_master[
        "genomic_timepoint_count"
    ]
    .fillna(0)
    .astype(int)
)

candidate_master[
    "has_genomic_timepoint"
] = (
    candidate_master[
        "genomic_timepoint_count"
    ]
    > 0
)

candidate_master[
    "multiple_genomic_timepoints"
] = (
    candidate_master[
        "genomic_timepoint_count"
    ]
    > 1
)

assert (
    candidate_master[
        "has_genomic_timepoint"
    ].all()
)

assert (
    candidate_master[
        "multiple_genomic_timepoints"
    ].sum()
    == 152
)

print(
    "Conditions with genomic data:",
    int(
        candidate_master[
            "has_genomic_timepoint"
        ].sum()
    )
)

print(
    "Multiple genomic timepoints:",
    int(
        candidate_master[
            "multiple_genomic_timepoints"
        ].sum()
    )
)

print("Genomic structure merge: PASS")


# ===================================================================================================
# 18. MERGE TEMPORAL STRUCTURE
# ===================================================================================================

print("\n17. MERGING TEMPORAL STRUCTURE")
print("-" * 100)

candidate_master = (
    candidate_master
    .merge(
        temporal_condition_summary,
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
)

candidate_master = (
    candidate_master
    .merge(
        nearest_condition_distance,
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
)

candidate_master = (
    candidate_master
    .merge(
        temporal_flag_subset,
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
)

assert (
    len(candidate_master)
    == 3081
)

assert (
    candidate_master[
        "temporal_complexity_class"
    ].notna().all()
)

assert (
    candidate_master[
        "temporal_complexity_flag"
    ].notna().all()
)

print("Temporal structure merge: PASS")


# ===================================================================================================
# 19. CANDIDATE ELIGIBILITY EVIDENCE STRATIFICATION
# ===================================================================================================
#
# This is the central output of Step 3B.6.37.
#
# These are EVIDENCE STRATA, not inclusion/exclusion decisions.
#
# ===================================================================================================

print("\n18. CASE-LEVEL ELIGIBILITY EVIDENCE STRATIFICATION")
print("-" * 100)


def classify_candidate_evidence(row):

    condition_class = str(
        row[
            "condition_evidence_class"
        ]
    )

    mixed_image = bool(
        row[
            "has_mixed_retained_excluded_evidence"
        ]
    )

    semantic_exclusion = bool(
        row[
            "has_human_semantic_exclusion"
        ]
    )

    semantic_retention = bool(
        row[
            "has_human_semantic_retention"
        ]
    )

    conflict_resolved = bool(
        row[
            "has_human_conflict_resolved"
        ]
    )

    unreviewed = bool(
        row[
            "has_unreviewed_images"
        ]
    )

    if mixed_image:

        return "MIXED_IMAGE_LEVEL_EVIDENCE"

    if conflict_resolved and semantic_retention:

        return "HUMAN_REVIEW_WITH_RESOLVED_CONFLICT"

    if semantic_exclusion and semantic_retention:

        return "MIXED_SEMANTIC_RETAIN_EXCLUDE"

    if semantic_exclusion:

        return "HUMAN_SEMANTIC_EXCLUSION_EVIDENCE"

    if semantic_retention and unreviewed:

        return "HUMAN_RETENTION_WITH_UNREVIEWED_IMAGES"

    if semantic_retention:

        return "HUMAN_SEMANTIC_RETENTION_EVIDENCE"

    if (
        condition_class
        == "STRUCTURAL_EVIDENCE_ONLY"
    ):

        return "STRUCTURAL_EVIDENCE_ONLY"

    if (
        condition_class
        == "NO_HUMAN_ELIGIBILITY_REVIEW"
    ):

        return "NO_HUMAN_ELIGIBILITY_REVIEW"

    return "OTHER_REVIEW_EVIDENCE"


candidate_master[
    "eligibility_evidence_stratum"
] = (
    candidate_master
    .apply(
        classify_candidate_evidence,
        axis=1
    )
)

print(
    candidate_master[
        "eligibility_evidence_stratum"
    ]
    .value_counts()
    .to_string()
)

assert (
    candidate_master[
        "eligibility_evidence_stratum"
    ].notna().all()
)

print("Eligibility evidence stratification: PASS")


# ===================================================================================================
# 20. TEMPORAL CANDIDATE STRATIFICATION
# ===================================================================================================

print("\n19. TEMPORAL CANDIDATE STRATIFICATION")
print("-" * 100)


def temporal_candidate_class(row):

    complexity = row[
        "temporal_complexity_class"
    ]

    if complexity in [
        "BEFORE_AND_AFTER",
        "BEFORE_SAME_DAY_AFTER",
    ]:

        return "BIDIRECTIONAL_TEMPORAL_STRUCTURE"

    if complexity in [
        "BEFORE_AND_SAME_DAY",
        "SAME_DAY_AND_AFTER",
    ]:

        return "MULTIPLE_TEMPORAL_RELATIONSHIPS"

    if complexity == "SAME_DAY_ONLY":

        return "SAME_DAY_ONLY"

    if complexity == "BEFORE_ONLY":

        return "GENOMICS_BEFORE_CXR_ONLY"

    if complexity == "AFTER_ONLY":

        return "GENOMICS_AFTER_CXR_ONLY"

    return "OTHER_TEMPORAL_STRUCTURE"


candidate_master[
    "temporal_candidate_stratum"
] = (
    candidate_master
    .apply(
        temporal_candidate_class,
        axis=1
    )
)

print(
    candidate_master[
        "temporal_candidate_stratum"
    ]
    .value_counts()
    .to_string()
)

assert (
    candidate_master[
        "temporal_candidate_stratum"
    ].notna().all()
)

print("Temporal candidate stratification: PASS")


# ===================================================================================================
# 21. MULTIMODAL CASE COMPLEXITY
# ===================================================================================================

print("\n20. MULTIMODAL CASE COMPLEXITY")
print("-" * 100)

candidate_master[
    "case_structure_class"
] = np.select(
    [
        (
            candidate_master[
                "multiple_cxr"
            ]
            &
            candidate_master[
                "multiple_genomic_timepoints"
            ]
        ),

        candidate_master[
            "multiple_cxr"
        ],

        candidate_master[
            "multiple_genomic_timepoints"
        ],
    ],
    [
        "MULTIPLE_CXR_AND_MULTIPLE_GENOMIC",
        "MULTIPLE_CXR_ONLY",
        "MULTIPLE_GENOMIC_ONLY",
    ],
    default="SINGLE_CXR_SINGLE_GENOMIC_TIMEPOINT"
)

print(
    candidate_master[
        "case_structure_class"
    ]
    .value_counts()
    .to_string()
)

assert (
    candidate_master[
        "case_structure_class"
    ].notna().all()
)

print("Multimodal case complexity: PASS")


# ===================================================================================================
# 22. TEMPORAL DISTANCE STRATIFICATION
# ===================================================================================================
#
# Descriptive only.
#
# No temporal threshold is being selected.
#
# ===================================================================================================

print("\n21. TEMPORAL DISTANCE STRATIFICATION")
print("-" * 100)


def temporal_distance_class(days):

    if pd.isna(days):
        return "NO_TEMPORAL_INFORMATION"

    if days == 0:
        return "SAME_DAY"

    if days <= 7:
        return "1_TO_7_DAYS"

    if days <= 30:
        return "8_TO_30_DAYS"

    if days <= 90:
        return "31_TO_90_DAYS"

    if days <= 180:
        return "91_TO_180_DAYS"

    if days <= 365:
        return "181_TO_365_DAYS"

    return "MORE_THAN_365_DAYS"


candidate_master[
    "nearest_temporal_distance_stratum"
] = (
    candidate_master[
        "nearest_genomic_distance_days"
    ]
    .apply(
        temporal_distance_class
    )
)

print(
    candidate_master[
        "nearest_temporal_distance_stratum"
    ]
    .value_counts()
    .to_string()
)

print(
    "Temporal distance categories remain descriptive."
)

print(
    "No eligibility cutoff has been selected."
)

print("Temporal distance stratification: PASS")


# ===================================================================================================
# 23. LEAKAGE-RISK FLAGS AT CONDITION LEVEL
# ===================================================================================================

print("\n22. CONDITION-LEVEL LEAKAGE-RISK FLAGS")
print("-" * 100)

candidate_master[
    "leakage_review_required"
] = (
    candidate_master[
        "bidirectional_temporal"
    ]
    |
    candidate_master[
        "multiple_cxr"
    ]
    |
    candidate_master[
        "multiple_genomic_timepoints"
    ]
)

candidate_master[
    "direct_target_like_metadata_present"
] = True

candidate_master[
    "identifier_fields_present"
] = True

print(
    "Conditions requiring later leakage-aware study design:",
    int(
        candidate_master[
            "leakage_review_required"
        ].sum()
    )
)

print(
    "This is a flag only."
)

print(
    "No condition is excluded."
)

print("Condition-level leakage flags: PASS")


# ===================================================================================================
# 24. CANDIDATE STRATA — TEMPORAL × EVIDENCE
# ===================================================================================================

print("\n23. COMBINED CANDIDATE STRATA")
print("-" * 100)

candidate_master[
    "combined_candidate_stratum"
] = (
    candidate_master[
        "eligibility_evidence_stratum"
    ]
    +
    " :: "
    +
    candidate_master[
        "temporal_candidate_stratum"
    ]
)

combined_strata = (
    candidate_master[
        "combined_candidate_stratum"
    ]
    .value_counts()
    .rename_axis(
        "combined_candidate_stratum"
    )
    .reset_index(
        name="condition_count"
    )
)

print(
    combined_strata.head(30).to_string(
        index=False
    )
)

print(
    "\nTotal combined strata:",
    len(combined_strata)
)

assert (
    combined_strata[
        "condition_count"
    ].sum()
    == 3081
)

print("Combined candidate strata: PASS")


# ===================================================================================================
# 25. HUMAN REVIEW COVERAGE AT CONDITION LEVEL
# ===================================================================================================

print("\n24. HUMAN REVIEW COVERAGE")
print("-" * 100)

candidate_master[
    "any_human_cxr_evidence"
] = (
    candidate_master[
        [
            "has_human_semantic_retention",
            "has_human_semantic_exclusion",
            "has_human_conflict_resolved",
        ]
    ]
    .any(
        axis=1
    )
    |
    candidate_master[
        "condition_evidence_class"
    ]
    .isin(
        [
            "STRUCTURAL_EVIDENCE_ONLY",
        ]
    )
)

print(
    "Conditions with any human CXR evidence:",
    int(
        candidate_master[
            "any_human_cxr_evidence"
        ].sum()
    )
)

print(
    "Conditions without individual human eligibility evidence:",
    int(
        (
            ~candidate_master[
                "any_human_cxr_evidence"
            ]
        ).sum()
    )
)

print("Human-review coverage: PASS")


# ===================================================================================================
# 26. IMPORTANT MIXED-EVIDENCE AUDIT
# ===================================================================================================

print("\n25. MIXED IMAGE-LEVEL EVIDENCE AUDIT")
print("-" * 100)

mixed_conditions = candidate_master[
    candidate_master[
        "has_mixed_retained_excluded_evidence"
    ]
].copy()

print(
    "Conditions with both retained and excluded human image evidence:",
    len(mixed_conditions)
)

print(
    "\nThese conditions are NOT automatically excluded."
)

print(
    "They require image-selection rules at the later case-level study-design stage."
)

print("Mixed-evidence audit: PASS")


# ===================================================================================================
# 27. CONDITIONS WITH EXCLUDED IMAGE EVIDENCE
# ===================================================================================================

print("\n26. EXCLUDED-IMAGE CONDITION AUDIT")
print("-" * 100)

conditions_with_excluded_images = candidate_master[
    candidate_master[
        "has_human_semantic_exclusion"
    ]
].copy()

print(
    "Conditions containing at least one semantically excluded image:",
    len(conditions_with_excluded_images)
)

print(
    "These are NOT automatically excluded as conditions."
)

print(
    "Reason: a condition may contain additional retained CXR evidence."
)

print("Excluded-image condition audit: PASS")


# ===================================================================================================
# 28. CONDITIONS WITH RETAINED IMAGE EVIDENCE
# ===================================================================================================

print("\n27. RETAINED-IMAGE CONDITION AUDIT")
print("-" * 100)

conditions_with_retained_images = candidate_master[
    candidate_master[
        "has_human_semantic_retention"
    ]
].copy()

print(
    "Conditions containing retained semantic CXR evidence:",
    len(conditions_with_retained_images)
)

print(
    "Retained image evidence does not by itself define the final cohort."
)

print("Retained-image condition audit: PASS")


# ===================================================================================================
# 29. UNREVIEWED CONDITION AUDIT
# ===================================================================================================

print("\n28. UNREVIEWED CONDITION AUDIT")
print("-" * 100)

unreviewed_conditions = candidate_master[
    ~candidate_master[
        "any_human_cxr_evidence"
    ]
].copy()

print(
    "Conditions without human CXR eligibility evidence:",
    len(unreviewed_conditions)
)

print(
    "These conditions remain candidates."
)

print(
    "They are NOT automatically excluded."
)

print("Unreviewed-condition audit: PASS")


# ===================================================================================================
# 30. NO FINAL ELIGIBILITY DECISION
# ===================================================================================================

print("\n29. FINAL ELIGIBILITY DECISION LOCK")
print("-" * 100)

candidate_master[
    "final_cxr_eligibility_decision"
] = "NOT_YET_SELECTED"

candidate_master[
    "final_case_eligibility_decision"
] = "NOT_YET_SELECTED"

candidate_master[
    "final_research_cohort_status"
] = "NOT_YET_SELECTED"

candidate_master[
    "final_prediction_target"
] = "NOT_SELECTED"

candidate_master[
    "representative_cxr_selected"
] = "NO"

candidate_master[
    "representative_genomic_timepoint_selected"
] = "NO"

candidate_master[
    "temporal_threshold_selected"
] = "NO"

assert (
    candidate_master[
        "final_cxr_eligibility_decision"
    ].eq(
        "NOT_YET_SELECTED"
    ).all()
)

assert (
    candidate_master[
        "final_case_eligibility_decision"
    ].eq(
        "NOT_YET_SELECTED"
    ).all()
)

assert (
    candidate_master[
        "final_prediction_target"
    ].eq(
        "NOT_SELECTED"
    ).all()
)

print(
    "Final CXR eligibility selected     : NO"
)

print(
    "Final case eligibility selected    : NO"
)

print(
    "Final research cohort selected     : NO"
)

print(
    "Prediction target selected         : NO"
)

print(
    "Representative CXR selected       : NO"
)

print(
    "Representative genomic sample selected: NO"
)

print(
    "Temporal threshold selected        : NO"
)

print("Final eligibility lock: PASS")


# ===================================================================================================
# 31. FINAL CONDITION POPULATION VALIDATION
# ===================================================================================================

print("\n30. FINAL CONDITION POPULATION VALIDATION")
print("-" * 100)

assert (
    len(candidate_master)
    == 3081
)

assert (
    candidate_master[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    candidate_master[
        "cxr_count"
    ].sum()
    == 3859
)

assert (
    candidate_master[
        "multiple_cxr"
    ].sum()
    == 485
)

assert (
    candidate_master[
        "multiple_genomic_timepoints"
    ].sum()
    == 152
)

assert (
    candidate_master[
        "bidirectional_temporal"
    ].sum()
    == 192
)

print(
    "3,081 conditions:",
    "PASS"
)

print(
    "3,859 CXR records represented:",
    "PASS"
)

print(
    "485 multiple-CXR conditions:",
    "PASS"
)

print(
    "152 multiple-genomic-timepoint conditions:",
    "PASS"
)

print(
    "192 bidirectional temporal conditions:",
    "PASS"
)

print("Final condition population: PASS")


# ===================================================================================================
# 32. SAVE MAIN CONDITION-LEVEL CANDIDATE MASTER
# ===================================================================================================

print("\n31. SAVING OUTPUTS")
print("-" * 100)

MASTER_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_37_Condition_Level_Eligibility_Candidate_Master.csv"
)

EVIDENCE_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_37_Eligibility_Evidence_Strata_Summary.csv"
)

TEMPORAL_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_37_Temporal_Candidate_Strata_Summary.csv"
)

CASE_STRUCTURE_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_37_Case_Structure_Summary.csv"
)

COMBINED_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_37_Combined_Candidate_Strata.csv"
)

MIXED_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_37_Mixed_Image_Evidence_Conditions.csv"
)

EXCLUDED_IMAGE_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_37_Conditions_With_Excluded_Image_Evidence.csv"
)

UNREVIEWED_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_37_Unreviewed_Candidate_Conditions.csv"
)


candidate_master.to_csv(
    MASTER_OUTPUT,
    index=False
)

evidence_summary = (
    candidate_master[
        "eligibility_evidence_stratum"
    ]
    .value_counts()
    .rename_axis(
        "eligibility_evidence_stratum"
    )
    .reset_index(
        name="condition_count"
    )
)

evidence_summary[
    "percentage_of_3081"
] = (
    evidence_summary[
        "condition_count"
    ]
    /
    3081
    *
    100
)

evidence_summary.to_csv(
    EVIDENCE_OUTPUT,
    index=False
)


temporal_summary = (
    candidate_master[
        "temporal_candidate_stratum"
    ]
    .value_counts()
    .rename_axis(
        "temporal_candidate_stratum"
    )
    .reset_index(
        name="condition_count"
    )
)

temporal_summary[
    "percentage_of_3081"
] = (
    temporal_summary[
        "condition_count"
    ]
    /
    3081
    *
    100
)

temporal_summary.to_csv(
    TEMPORAL_OUTPUT,
    index=False
)


case_structure_summary = (
    candidate_master[
        "case_structure_class"
    ]
    .value_counts()
    .rename_axis(
        "case_structure_class"
    )
    .reset_index(
        name="condition_count"
    )
)

case_structure_summary[
    "percentage_of_3081"
] = (
    case_structure_summary[
        "condition_count"
    ]
    /
    3081
    *
    100
)

case_structure_summary.to_csv(
    CASE_STRUCTURE_OUTPUT,
    index=False
)


combined_strata.to_csv(
    COMBINED_OUTPUT,
    index=False
)


mixed_conditions.to_csv(
    MIXED_OUTPUT,
    index=False
)


conditions_with_excluded_images.to_csv(
    EXCLUDED_IMAGE_OUTPUT,
    index=False
)


unreviewed_conditions.to_csv(
    UNREVIEWED_OUTPUT,
    index=False
)


# ===================================================================================================
# 33. OUTPUT VALIDATION
# ===================================================================================================

print("\n32. OUTPUT FILE VALIDATION")
print("-" * 100)

output_paths = [
    MASTER_OUTPUT,
    EVIDENCE_OUTPUT,
    TEMPORAL_OUTPUT,
    CASE_STRUCTURE_OUTPUT,
    COMBINED_OUTPUT,
    MIXED_OUTPUT,
    EXCLUDED_IMAGE_OUTPUT,
    UNREVIEWED_OUTPUT,
]

for output_path in output_paths:

    assert output_path.exists(), (
        f"Output missing:\n{output_path}"
    )

    assert output_path.stat().st_size > 0, (
        f"Output is empty:\n{output_path}"
    )

    print(
        f"{output_path.name}: PASS"
    )


# ===================================================================================================
# 34. SOURCE PROTECTION
# ===================================================================================================

print("\n33. SOURCE PROTECTION")
print("-" * 100)

print(
    "Raw DICOM modification              : NO"
)

print(
    "Source CXR metadata modification    : NO"
)

print(
    "Source genomics modification        : NO"
)

print(
    "Human review artifact modification  : NO"
)

print(
    "Additional ZIP extraction           : NO"
)

print(
    "Image preprocessing                 : NO"
)

print(
    "Model training                      : NO"
)

print(
    "Target selection                    : NO"
)

print(
    "Final cohort selection              : NO"
)

print("Source protection: PASS")


# ===================================================================================================
# 35. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.37 — FINAL STATUS")
print("=" * 100)

print(
    "3,081 condition-level candidate records       : PASS"
)

print(
    "3,859 CXR records represented                 : PASS"
)

print(
    "Image-level evidence aggregated               : PASS"
)

print(
    "Condition-level evidence reconciled           : PASS"
)

print(
    "Multiple-CXR structure preserved              : PASS"
)

print(
    "Multiple-genomic structure preserved          : PASS"
)

print(
    "Temporal structure integrated                 : PASS"
)

print(
    "Temporal complexity integrated                : PASS"
)

print(
    "Mixed image evidence preserved                : PASS"
)

print(
    "Unreviewed conditions preserved               : PASS"
)

print(
    "Candidate evidence strata created             : PASS"
)

print(
    "Temporal candidate strata created             : PASS"
)

print(
    "Combined candidate strata created             : PASS"
)

print(
    "No automatic condition exclusion              : PASS"
)

print(
    "No temporal threshold selected                : PASS"
)

print(
    "No representative CXR selected               : PASS"
)

print(
    "No representative genomic sample selected    : PASS"
)

print(
    "No prediction target selected                 : PASS"
)

print(
    "No final research cohort selected             : PASS"
)

print(
    "No preprocessing performed                    : PASS"
)

print(
    "No model training                             : PASS"
)

print(
    "DICOM files modified                          : NO"
)

print(
    "Source metadata modified                      : NO"
)

print(
    "Output files                                  : PASS"
)

print("\n" + "=" * 100)
print("STEP 3B.6.37 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.38 — RESEARCH TARGET FEASIBILITY & TARGET-LEAKAGE MATRIX
# COMPLETE SELF-CONTAINED VERSION
# ===================================================================================================
#
# PURPOSE
# -------
# Before selecting a final research question or prediction target, this step
# evaluates the target-like variables actually available in the March 2025
# TB Portals Imaging + Genomics data.
#
# THIS STEP DOES NOT:
# -------------------
# - select the final prediction target
# - select the final research question
# - select predictors
# - construct the final modeling cohort
# - select a representative CXR
# - select a representative genomic specimen
# - select a temporal threshold
# - preprocess images
# - train a model
#
# It only creates an evidence-based TARGET FEASIBILITY MATRIX.
#
# PRIMARY CASE UNIT:
# ------------------
# condition_id
#
# ===================================================================================================


from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 100)
print("STEP 3B.6.38 — RESEARCH TARGET FEASIBILITY & TARGET-LEAKAGE MATRIX")
print("=" * 100)


# ===================================================================================================
# 1. INITIALIZATION
# ===================================================================================================

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_024_DIR = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03"
)

STEP_034_DIR = (
    METADATA_DIR /
    "Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
)

STEP_037_DIR = (
    METADATA_DIR /
    "Step_3B_6_37_Case_Level_Eligibility_Candidates"
)

CXR_MASTER_PATH = (
    STEP_024_DIR /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

GENOMICS_PATH = (
    METADATA_DIR /
    "TB_Portals_Genomics_March_2025.csv"
)

CANDIDATE_MASTER_PATH = (
    STEP_037_DIR /
    "TB_Portals_March2025_Step_3B_6_37_Condition_Level_Eligibility_Candidate_Master.csv"
)

TEMPORAL_PAIRS_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

GENOMIC_TIMEPOINT_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_38_Target_Feasibility"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ===================================================================================================
# 2. INPUT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

required_paths = {
    "CXR master": CXR_MASTER_PATH,
    "March 2025 genomics": GENOMICS_PATH,
    "Step 3B.6.37 candidate master": CANDIDATE_MASTER_PATH,
    "Step 3B.6.34 temporal pairs": TEMPORAL_PAIRS_PATH,
    "Step 3B.6.34 genomic timepoints": GENOMIC_TIMEPOINT_PATH,
}

for label, path in required_paths.items():

    assert path.exists(), (
        f"Required artifact not found:\n{path}"
    )

    print(
        f"{label:<42}: FOUND"
    )

print("Input artifact validation: PASS")


# ===================================================================================================
# 3. LOAD SOURCE ARTIFACTS
# ===================================================================================================

cxr = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)

candidate_master = pd.read_csv(
    CANDIDATE_MASTER_PATH,
    low_memory=False
)

temporal_pairs = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)

genomic_timepoints = pd.read_csv(
    GENOMIC_TIMEPOINT_PATH,
    low_memory=False
)


print("\n2. LOADED ARTIFACTS")
print("-" * 100)

print(
    "CXR records:",
    len(cxr)
)

print(
    "Genomic records:",
    len(genomics)
)

print(
    "Candidate conditions:",
    len(candidate_master)
)

print(
    "Temporal pair records:",
    len(temporal_pairs)
)

print(
    "Unique genomic timepoints:",
    len(genomic_timepoints)
)


# ===================================================================================================
# 4. NORMALIZE CONDITION IDS
# ===================================================================================================

print("\n3. CONDITION-ID NORMALIZATION")
print("-" * 100)

for df_name, df in [
    ("CXR", cxr),
    ("GENOMICS", genomics),
    ("CANDIDATE_MASTER", candidate_master),
    ("TEMPORAL_PAIRS", temporal_pairs),
    ("GENOMIC_TIMEPOINTS", genomic_timepoints),
]:

    assert "condition_id" in df.columns, (
        f"{df_name} is missing condition_id"
    )

    df["condition_id"] = (
        df["condition_id"]
        .astype(str)
        .str.strip()
    )

print("Condition IDs normalized: PASS")


# ===================================================================================================
# 5. CORE POPULATION VALIDATION
# ===================================================================================================

print("\n4. CORE POPULATION VALIDATION")
print("-" * 100)

assert len(cxr) == 3859

assert (
    cxr[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    candidate_master[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    len(candidate_master)
    == 3081
)

common_conditions = (
    set(cxr["condition_id"])
    &
    set(genomics["condition_id"])
)

assert (
    len(common_conditions)
    == 3081
)

print(
    "CXR records       : 3,859"
)

print(
    "CXR conditions    : 3,081"
)

print(
    "Candidate records : 3,081"
)

print(
    "Common CXR/genomic conditions : 3,081"
)

print("Core population: PASS")


# ===================================================================================================
# 6. AVAILABLE CXR TARGET-LIKE FIELDS
# ===================================================================================================

print("\n5. AVAILABLE CXR TARGET-LIKE FIELDS")
print("-" * 100)

cxr_target_candidates = [
    "type_of_resistance",
    "outcome",
    "diagnosis_code",
    "case_definition",
]

cxr_target_present = [
    column
    for column in cxr_target_candidates
    if column in cxr.columns
]

print(
    "Available CXR target-like fields:"
)

for column in cxr_target_present:

    print(
        f"  {column}"
    )

assert (
    "type_of_resistance"
    in cxr_target_present
)

assert (
    "outcome"
    in cxr_target_present
)

print("CXR target-field discovery: PASS")


# ===================================================================================================
# 7. AVAILABLE GENOMIC TARGET-LIKE FIELDS
# ===================================================================================================

print("\n6. AVAILABLE GENOMIC TARGET-LIKE FIELDS")
print("-" * 100)

genomic_target_candidates = [
    "drug_resistance_type",
    "type_of_resistance",
    "outcome",
]

genomic_target_present = [
    column
    for column in genomic_target_candidates
    if column in genomics.columns
]

print(
    "Available genomic target-like fields:"
)

for column in genomic_target_present:

    print(
        f"  {column}"
    )

assert (
    "drug_resistance_type"
    in genomic_target_present
)

print("Genomic target-field discovery: PASS")


# ===================================================================================================
# 8. CONDITION-LEVEL CXR TARGET COMPLETENESS
# ===================================================================================================
#
# Because the CXR table contains multiple records per condition, target
# completeness must be evaluated at condition level rather than simply
# counting image rows.
#
# ===================================================================================================

print("\n7. CONDITION-LEVEL CXR TARGET COMPLETENESS")
print("-" * 100)

cxr_condition_target = (
    cxr
    .groupby(
        "condition_id"
    )
    .agg(
        resistance_nonmissing=(
            "type_of_resistance",
            lambda x:
            x.notna().any()
        ),

        outcome_nonmissing=(
            "outcome",
            lambda x:
            x.notna().any()
        ),

        diagnosis_nonmissing=(
            "diagnosis_code",
            lambda x:
            x.notna().any()
        ),

        case_definition_nonmissing=(
            "case_definition",
            lambda x:
            x.notna().any()
        ),
    )
    .reset_index()
)

assert (
    len(cxr_condition_target)
    == 3081
)

print(
    "Conditions with resistance field:",
    int(
        cxr_condition_target[
            "resistance_nonmissing"
        ].sum()
    )
)

print(
    "Conditions with outcome field:",
    int(
        cxr_condition_target[
            "outcome_nonmissing"
        ].sum()
    )
)

print(
    "Conditions with diagnosis code:",
    int(
        cxr_condition_target[
            "diagnosis_nonmissing"
        ].sum()
    )
)

print(
    "Conditions with case definition:",
    int(
        cxr_condition_target[
            "case_definition_nonmissing"
        ].sum()
    )
)

print("CXR target completeness: PASS")


# ===================================================================================================
# 9. GENOMIC TARGET COMPLETENESS
# ===================================================================================================

print("\n8. GENOMIC TARGET COMPLETENESS")
print("-" * 100)

genomics_common = genomics[
    genomics[
        "condition_id"
    ].isin(
        common_conditions
    )
].copy()

print(
    "Genomic records in multimodal conditions:",
    len(genomics_common)
)

print(
    "Unique multimodal genomic conditions:",
    genomics_common[
        "condition_id"
    ].nunique()
)

assert (
    genomics_common[
        "condition_id"
    ].nunique()
    == 3081
)

genomic_condition_target = (
    genomics_common
    .groupby(
        "condition_id"
    )
    .agg(
        genomic_resistance_profile_available=(
            "drug_resistance_type",
            lambda x:
            x.notna().any()
        ),

        genomic_type_of_resistance_available=(
            "type_of_resistance",
            lambda x:
            x.notna().any()
        ),
    )
    .reset_index()
)

assert (
    len(genomic_condition_target)
    == 3081
)

print(
    "Conditions with genomic drug_resistance_type:",
    int(
        genomic_condition_target[
            "genomic_resistance_profile_available"
        ].sum()
    )
)

print(
    "Conditions with genomic type_of_resistance:",
    int(
        genomic_condition_target[
            "genomic_type_of_resistance_available"
        ].sum()
    )
)

print("Genomic target completeness: PASS")


# ===================================================================================================
# 10. TARGET CLASS DISTRIBUTIONS
# ===================================================================================================

print("\n9. TARGET CLASS DISTRIBUTIONS")
print("-" * 100)

print(
    "\nCXR type_of_resistance:"
)

print(
    cxr[
        "type_of_resistance"
    ]
    .value_counts(
        dropna=False
    )
    .to_string()
)

print(
    "\nCXR outcome:"
)

print(
    cxr[
        "outcome"
    ]
    .value_counts(
        dropna=False
    )
    .to_string()
)

print(
    "\nGenomic drug_resistance_type:"
)

print(
    genomics_common[
        "drug_resistance_type"
    ]
    .value_counts(
        dropna=False
    )
    .to_string()
)

print(
    "\nGenomic type_of_resistance:"
)

print(
    genomics_common[
        "type_of_resistance"
    ]
    .value_counts(
        dropna=False
    )
    .to_string()
)

print("Target class distribution audit: PASS")


# ===================================================================================================
# 11. CXR ↔ GENOMIC RESISTANCE AGREEMENT
# ===================================================================================================
#
# This is descriptive only.
#
# It does NOT select either field as the target.
#
# ===================================================================================================

print("\n10. CXR ↔ GENOMIC RESISTANCE AGREEMENT AUDIT")
print("-" * 100)

cxr_resistance = (
    cxr[
        [
            "condition_id",
            "type_of_resistance",
        ]
    ]
    .drop_duplicates(
        subset=[
            "condition_id",
        ]
    )
)

genomic_resistance = (
    genomics_common[
        [
            "condition_id",
            "type_of_resistance",
        ]
    ]
    .dropna(
        subset=[
            "type_of_resistance",
        ]
    )
    .groupby(
        "condition_id"
    )[
        "type_of_resistance"
    ]
    .agg(
        lambda x:
        sorted(
            set(
                x.astype(str)
            )
        )
    )
    .rename(
        "genomic_type_of_resistance_values"
    )
    .reset_index()
)

resistance_comparison = (
    cxr_resistance
    .merge(
        genomic_resistance,
        on="condition_id",
        how="left",
        validate="one_to_one"
    )
)

resistance_comparison[
    "genomic_type_of_resistance_values"
] = (
    resistance_comparison[
        "genomic_type_of_resistance_values"
    ]
    .apply(
        lambda x:
        x
        if isinstance(x, list)
        else []
    )
)

resistance_comparison[
    "genomic_contains_cxr_value"
] = (
    resistance_comparison
    .apply(
        lambda row:
        (
            pd.notna(
                row[
                    "type_of_resistance"
                ]
            )
            and
            str(
                row[
                    "type_of_resistance"
                ]
            )
            in
            row[
                "genomic_type_of_resistance_values"
            ]
        ),
        axis=1
    )
)

print(
    "Conditions where genomic resistance contains CXR resistance:",
    int(
        resistance_comparison[
            "genomic_contains_cxr_value"
        ].sum()
    )
)

print(
    "This is a consistency audit, not target selection."
)

print("Resistance agreement audit: PASS")


# ===================================================================================================
# 12. TEMPORAL AVAILABILITY OF GENOMIC INFORMATION
# ===================================================================================================

print("\n11. TEMPORAL AVAILABILITY OF GENOMIC INFORMATION")
print("-" * 100)

nearest_pairs = (
    temporal_pairs
    .sort_values(
        [
            "series_instance_content_url",
            "absolute_temporal_distance_days",
        ]
    )
    .copy()
)

nearest_pairs[
    "absolute_temporal_distance_days"
] = pd.to_numeric(
    nearest_pairs[
        "absolute_temporal_distance_days"
    ],
    errors="coerce"
)

nearest_pairs[
    "signed_temporal_delta_days"
] = pd.to_numeric(
    nearest_pairs[
        "signed_temporal_delta_days"
    ],
    errors="coerce"
)

nearest_per_cxr = (
    nearest_pairs
    .groupby(
        "series_instance_content_url",
        as_index=False
    )
    .first()
)

assert (
    len(nearest_per_cxr)
    == 3859
)

nearest_per_cxr[
    "genomic_before_cxr"
] = (
    nearest_per_cxr[
        "signed_temporal_delta_days"
    ]
    < 0
)

nearest_per_cxr[
    "same_day"
] = (
    nearest_per_cxr[
        "signed_temporal_delta_days"
    ]
    == 0
)

nearest_per_cxr[
    "genomic_after_cxr"
] = (
    nearest_per_cxr[
        "signed_temporal_delta_days"
    ]
    > 0
)

print(
    "Nearest genomic before CXR:",
    int(
        nearest_per_cxr[
            "genomic_before_cxr"
        ].sum()
    )
)

print(
    "Nearest genomic same day:",
    int(
        nearest_per_cxr[
            "same_day"
        ].sum()
    )
)

print(
    "Nearest genomic after CXR:",
    int(
        nearest_per_cxr[
            "genomic_after_cxr"
        ].sum()
    )
)

print("Temporal target-availability audit: PASS")


# ===================================================================================================
# 13. CONDITION-LEVEL TEMPORAL TARGET AVAILABILITY
# ===================================================================================================

print("\n12. CONDITION-LEVEL TEMPORAL TARGET AVAILABILITY")
print("-" * 100)

condition_temporal_availability = (
    nearest_per_cxr
    .groupby(
        "condition_id"
    )
    .agg(
        any_genomic_before_cxr=(
            "genomic_before_cxr",
            "any"
        ),

        any_same_day=(
            "same_day",
            "any"
        ),

        any_genomic_after_cxr=(
            "genomic_after_cxr",
            "any"
        ),
    )
    .reset_index()
)

assert (
    len(condition_temporal_availability)
    == 3081
)

condition_temporal_availability[
    "temporal_direction_count"
] = (
    condition_temporal_availability[
        [
            "any_genomic_before_cxr",
            "any_same_day",
            "any_genomic_after_cxr",
        ]
    ]
    .sum(
        axis=1
    )
)

print(
    condition_temporal_availability[
        "temporal_direction_count"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)

print("Condition temporal availability: PASS")


# ===================================================================================================
# 14. POTENTIAL TARGET PROXY AUDIT
# ===================================================================================================
#
# These fields are NOT automatically banned from every possible study.
#
# They are flagged because their relationship to a future target must be
# explicitly considered before modeling.
#
# ===================================================================================================

print("\n13. POTENTIAL TARGET PROXY AUDIT")
print("-" * 100)

target_proxy_register = pd.DataFrame(
    [
        {
            "field": "CXR:type_of_resistance",
            "category": "DIRECT_RESISTANCE_LABEL",
            "leakage_risk": "HIGH_IF_RESISTANCE_IS_TARGET",
            "current_status": "TARGET_OR_TARGET_PROXY",
        },
        {
            "field": "GENOMICS:type_of_resistance",
            "category": "DIRECT_RESISTANCE_LABEL",
            "leakage_risk": "HIGH_IF_RESISTANCE_IS_TARGET",
            "current_status": "TARGET_OR_TARGET_PROXY",
        },
        {
            "field": "GENOMICS:drug_resistance_type",
            "category": "DIRECT_RESISTANCE_PROFILE",
            "leakage_risk": "HIGH_IF_RESISTANCE_IS_TARGET",
            "current_status": "TARGET_OR_TARGET_PROXY",
        },
        {
            "field": "CXR:outcome",
            "category": "CLINICAL_OUTCOME",
            "leakage_risk": "HIGH_IF_OUTCOME_IS_FUTURE_TARGET",
            "current_status": "TARGET_OR_OUTCOME",
        },
        {
            "field": "CXR:diagnosis_code",
            "category": "DIAGNOSIS",
            "leakage_risk": "TARGET_DEPENDENT",
            "current_status": "REVIEW_REQUIRED",
        },
        {
            "field": "CXR:case_definition",
            "category": "CASE_DEFINITION",
            "leakage_risk": "TARGET_DEPENDENT",
            "current_status": "REVIEW_REQUIRED",
        },
        {
            "field": "patient_id",
            "category": "IDENTIFIER",
            "leakage_risk": "HIGH",
            "current_status": "LINKAGE_ONLY",
        },
        {
            "field": "condition_id",
            "category": "IDENTIFIER",
            "leakage_risk": "HIGH",
            "current_status": "LINKAGE_ONLY",
        },
        {
            "field": "specimen_id",
            "category": "IDENTIFIER",
            "leakage_risk": "HIGH",
            "current_status": "LINKAGE_ONLY",
        },
    ]
)

print(
    target_proxy_register.to_string(
        index=False
    )
)

print("Target proxy audit: PASS")


# ===================================================================================================
# 15. TARGET FEASIBILITY MATRIX
# ===================================================================================================

print("\n14. TARGET FEASIBILITY MATRIX")
print("-" * 100)

target_matrix = pd.DataFrame(
    [
        {
            "candidate_target":
                "CXR type_of_resistance",
            "source":
                "CXR",
            "target_family":
                "RESISTANCE",
            "available_in_3081_common_conditions":
                int(
                    cxr_condition_target[
                        "resistance_nonmissing"
                    ].sum()
                ),
            "temporal_dependency":
                "REQUIRES_STUDY_DESIGN_REVIEW",
            "direct_target_field":
                "YES",
            "potential_leakage":
                "HIGH_IF_USED_AS_INPUT",
            "target_status":
                "NOT_SELECTED",
        },
        {
            "candidate_target":
                "Genomic type_of_resistance",
            "source":
                "GENOMICS",
            "target_family":
                "RESISTANCE",
            "available_in_3081_common_conditions":
                int(
                    genomic_condition_target[
                        "genomic_type_of_resistance_available"
                    ].sum()
                ),
            "temporal_dependency":
                "REQUIRES_STUDY_DESIGN_REVIEW",
            "direct_target_field":
                "YES",
            "potential_leakage":
                "HIGH_IF_USED_AS_INPUT",
            "target_status":
                "NOT_SELECTED",
        },
        {
            "candidate_target":
                "Genomic drug_resistance_type",
            "source":
                "GENOMICS",
            "target_family":
                "RESISTANCE_PROFILE",
            "available_in_3081_common_conditions":
                int(
                    genomic_condition_target[
                        "genomic_resistance_profile_available"
                    ].sum()
                ),
            "temporal_dependency":
                "REQUIRES_STUDY_DESIGN_REVIEW",
            "direct_target_field":
                "YES",
            "potential_leakage":
                "HIGH_IF_USED_AS_INPUT",
            "target_status":
                "NOT_SELECTED",
        },
        {
            "candidate_target":
                "CXR outcome",
            "source":
                "CXR",
            "target_family":
                "CLINICAL_OUTCOME",
            "available_in_3081_common_conditions":
                int(
                    cxr_condition_target[
                        "outcome_nonmissing"
                    ].sum()
                ),
            "temporal_dependency":
                "HIGH",
            "direct_target_field":
                "YES",
            "potential_leakage":
                "HIGH_IF_POST_OUTCOME_INFORMATION_IS_USED",
            "target_status":
                "NOT_SELECTED",
        },
        {
            "candidate_target":
                "CXR diagnosis_code",
            "source":
                "CXR",
            "target_family":
                "DIAGNOSIS",
            "available_in_3081_common_conditions":
                int(
                    cxr_condition_target[
                        "diagnosis_nonmissing"
                    ].sum()
                ),
            "temporal_dependency":
                "TARGET_DEPENDENT",
            "direct_target_field":
                "POTENTIAL",
            "potential_leakage":
                "TARGET_DEPENDENT",
            "target_status":
                "NOT_SELECTED",
        },
    ]
)

print(
    target_matrix.to_string(
        index=False
    )
)

assert (
    len(target_matrix)
    == 5
)

print("Target feasibility matrix: PASS")


# ===================================================================================================
# 16. TARGET AVAILABILITY SUMMARY
# ===================================================================================================

print("\n15. TARGET AVAILABILITY SUMMARY")
print("-" * 100)

availability_summary = pd.DataFrame(
    [
        {
            "candidate_target":
                "CXR type_of_resistance",
            "condition_count":
                int(
                    cxr_condition_target[
                        "resistance_nonmissing"
                    ].sum()
                ),
        },
        {
            "candidate_target":
                "CXR outcome",
            "condition_count":
                int(
                    cxr_condition_target[
                        "outcome_nonmissing"
                    ].sum()
                ),
        },
        {
            "candidate_target":
                "CXR diagnosis_code",
            "condition_count":
                int(
                    cxr_condition_target[
                        "diagnosis_nonmissing"
                    ].sum()
                ),
        },
        {
            "candidate_target":
                "CXR case_definition",
            "condition_count":
                int(
                    cxr_condition_target[
                        "case_definition_nonmissing"
                    ].sum()
                ),
        },
        {
            "candidate_target":
                "Genomic type_of_resistance",
            "condition_count":
                int(
                    genomic_condition_target[
                        "genomic_type_of_resistance_available"
                    ].sum()
                ),
        },
        {
            "candidate_target":
                "Genomic drug_resistance_type",
            "condition_count":
                int(
                    genomic_condition_target[
                        "genomic_resistance_profile_available"
                    ].sum()
                ),
        },
    ]
)

availability_summary[
    "percentage_of_3081"
] = (
    availability_summary[
        "condition_count"
    ]
    /
    3081
    *
    100
)

print(
    availability_summary.to_string(
        index=False
    )
)

assert (
    availability_summary[
        "condition_count"
    ]
    .between(
        0,
        3081
    )
    .all()
)

print("Target availability summary: PASS")


# ===================================================================================================
# 17. STUDY-DESIGN DECISION LOCK
# ===================================================================================================

print("\n16. STUDY-DESIGN DECISION LOCK")
print("-" * 100)

decision_lock = pd.DataFrame(
    [
        {
            "study_design_element":
                "research_question",
            "status":
                "NOT_SELECTED",
        },
        {
            "study_design_element":
                "prediction_target",
            "status":
                "NOT_SELECTED",
        },
        {
            "study_design_element":
                "target_definition",
            "status":
                "NOT_SELECTED",
        },
        {
            "study_design_element":
                "positive_negative_class_mapping",
            "status":
                "NOT_SELECTED",
        },
        {
            "study_design_element":
                "feature_set",
            "status":
                "NOT_SELECTED",
        },
        {
            "study_design_element":
                "temporal_cutoff",
            "status":
                "NOT_SELECTED",
        },
        {
            "study_design_element":
                "representative_cxr_rule",
            "status":
                "NOT_SELECTED",
        },
        {
            "study_design_element":
                "representative_genomic_timepoint_rule",
            "status":
                "NOT_SELECTED",
        },
        {
            "study_design_element":
                "final_research_cohort",
            "status":
                "NOT_SELECTED",
        },
    ]
)

print(
    decision_lock.to_string(
        index=False
    )
)

assert (
    decision_lock[
        "status"
    ].eq(
        "NOT_SELECTED"
    ).all()
)

print("Study-design decision lock: PASS")


# ===================================================================================================
# 18. NO TARGET-BASED EXCLUSION
# ===================================================================================================

print("\n17. TARGET-BASED EXCLUSION SAFETY")
print("-" * 100)

print(
    "Resistance labels used to exclude cases : NO"
)

print(
    "Outcome used to exclude cases            : NO"
)

print(
    "Diagnosis code used to exclude cases     : NO"
)

print(
    "Genomic target fields used as predictors : NO"
)

print(
    "CXR target fields used as predictors     : NO"
)

print(
    "Target class balancing applied           : NO"
)

print(
    "Target threshold applied                 : NO"
)

print(
    "Final target selected                    : NO"
)

print("Target-based exclusion safety: PASS")


# ===================================================================================================
# 19. SAVE OUTPUTS
# ===================================================================================================

print("\n18. SAVING OUTPUTS")
print("-" * 100)

CXR_TARGET_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_38_CXR_Target_Availability.csv"
)

GENOMIC_TARGET_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_38_Genomic_Target_Availability.csv"
)

RESISTANCE_COMPARISON_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_38_CXR_Genomic_Resistance_Consistency.csv"
)

TEMPORAL_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_38_Temporal_Target_Availability.csv"
)

PROXY_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_38_Target_Proxy_Leakage_Register.csv"
)

TARGET_MATRIX_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_38_Target_Feasibility_Matrix.csv"
)

AVAILABILITY_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_38_Target_Availability_Summary.csv"
)

DECISION_LOCK_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_38_Study_Design_Decision_Lock.csv"
)


cxr_condition_target.to_csv(
    CXR_TARGET_OUTPUT,
    index=False
)

genomic_condition_target.to_csv(
    GENOMIC_TARGET_OUTPUT,
    index=False
)

resistance_comparison.to_csv(
    RESISTANCE_COMPARISON_OUTPUT,
    index=False
)

condition_temporal_availability.to_csv(
    TEMPORAL_OUTPUT,
    index=False
)

target_proxy_register.to_csv(
    PROXY_OUTPUT,
    index=False
)

target_matrix.to_csv(
    TARGET_MATRIX_OUTPUT,
    index=False
)

availability_summary.to_csv(
    AVAILABILITY_OUTPUT,
    index=False
)

decision_lock.to_csv(
    DECISION_LOCK_OUTPUT,
    index=False
)


# ===================================================================================================
# 20. OUTPUT VALIDATION
# ===================================================================================================

print("\n19. OUTPUT FILE VALIDATION")
print("-" * 100)

output_paths = [
    CXR_TARGET_OUTPUT,
    GENOMIC_TARGET_OUTPUT,
    RESISTANCE_COMPARISON_OUTPUT,
    TEMPORAL_OUTPUT,
    PROXY_OUTPUT,
    TARGET_MATRIX_OUTPUT,
    AVAILABILITY_OUTPUT,
    DECISION_LOCK_OUTPUT,
]

for output_path in output_paths:

    assert output_path.exists(), (
        f"Output missing:\n{output_path}"
    )

    assert output_path.stat().st_size > 0, (
        f"Output is empty:\n{output_path}"
    )

    print(
        f"{output_path.name}: PASS"
    )


# ===================================================================================================
# 21. FINAL INTEGRITY CHECKS
# ===================================================================================================

print("\n20. FINAL INTEGRITY CHECKS")
print("-" * 100)

assert len(cxr) == 3859

assert (
    cxr[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    len(candidate_master)
    == 3081
)

assert (
    len(cxr_condition_target)
    == 3081
)

assert (
    len(genomic_condition_target)
    == 3081
)

assert (
    len(nearest_per_cxr)
    == 3859
)

assert (
    len(condition_temporal_availability)
    == 3081
)

assert (
    decision_lock[
        "status"
    ].eq(
        "NOT_SELECTED"
    ).all()
)

print(
    "3,859 CXR records                 : PASS"
)

print(
    "3,081 multimodal conditions       : PASS"
)

print(
    "CXR target availability           : PASS"
)

print(
    "Genomic target availability       : PASS"
)

print(
    "Temporal target availability      : PASS"
)

print(
    "Target proxy audit                : PASS"
)

print(
    "Target feasibility matrix         : PASS"
)

print(
    "Target not selected               : PASS"
)

print(
    "Final cohort not selected         : PASS"
)

print(
    "Feature set not selected          : PASS"
)

print(
    "No target-based exclusion         : PASS"
)


# ===================================================================================================
# 22. SOURCE PROTECTION
# ===================================================================================================

print("\n21. SOURCE PROTECTION")
print("-" * 100)

print(
    "Raw DICOM modification             : NO"
)

print(
    "Source CXR metadata modification   : NO"
)

print(
    "Source genomics modification       : NO"
)

print(
    "Human review artifact modification : NO"
)

print(
    "Image preprocessing                : NO"
)

print(
    "Model training                     : NO"
)

print(
    "Target selection                   : NO"
)

print(
    "Final cohort selection             : NO"
)

print("Source protection: PASS")


# ===================================================================================================
# 23. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.38 — FINAL STATUS")
print("=" * 100)

print(
    "CXR target discovery                       : PASS"
)

print(
    "Genomic target discovery                   : PASS"
)

print(
    "Condition-level target completeness        : PASS"
)

print(
    "Target class distribution audit            : PASS"
)

print(
    "CXR/genomic resistance consistency audit   : PASS"
)

print(
    "Temporal target availability audit         : PASS"
)

print(
    "Target proxy leakage audit                 : PASS"
)

print(
    "Target feasibility matrix                  : PASS"
)

print(
    "Study-design decision lock                 : PASS"
)

print(
    "No target selected                         : PASS"
)

print(
    "No feature set selected                    : PASS"
)

print(
    "No temporal threshold selected             : PASS"
)

print(
    "No final cohort selected                   : PASS"
)

print(
    "No preprocessing performed                 : PASS"
)

print(
    "No model training                          : PASS"
)

print(
    "DICOM files modified                       : NO"
)

print(
    "Source metadata modified                   : NO"
)

print(
    "Output files                               : PASS"
)

print("\n" + "=" * 100)
print("STEP 3B.6.38 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.39 — TARGET DEFINITION & COHORT IMPACT ANALYSIS
# CORRECTED SELF-CONTAINED VERSION
# ===================================================================================================
#
# PURPOSE
# -------
# Analyze candidate target definitions and their cohort impact before selecting
# the final research question, target, temporal rule, or modeling cohort.
#
# PRIMARY UNIT:
#     condition_id
#
# AUTHORITATIVE TEMPORAL SOURCE:
#     Step 3B.6.34 CXR ↔ genomic temporal-pair artifact
#
# IMPORTANT:
#     This step DOES NOT select the final target.
#     This step DOES NOT select the final cohort.
#     This step DOES NOT preprocess images.
#     This step DOES NOT train models.
#
# No notebook-memory variables are required.
# ===================================================================================================


from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 100)
print("STEP 3B.6.39 — TARGET DEFINITION & COHORT IMPACT ANALYSIS")
print("=" * 100)


# ===================================================================================================
# 1. PATH INITIALIZATION
# ===================================================================================================

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_024_DIR = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03"
)

STEP_034_DIR = (
    METADATA_DIR /
    "Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
)

STEP_037_DIR = (
    METADATA_DIR /
    "Step_3B_6_37_Case_Level_Eligibility_Candidates"
)

STEP_038_DIR = (
    METADATA_DIR /
    "Step_3B_6_38_Target_Feasibility"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_39_Target_Definition_Cohort_Impact"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


CXR_MASTER_PATH = (
    STEP_024_DIR /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

GENOMICS_PATH = (
    METADATA_DIR /
    "TB_Portals_Genomics_March_2025.csv"
)

CANDIDATE_MASTER_PATH = (
    STEP_037_DIR /
    "TB_Portals_March2025_Step_3B_6_37_Condition_Level_Eligibility_Candidate_Master.csv"
)

TARGET_MATRIX_PATH = (
    STEP_038_DIR /
    "TB_Portals_March2025_Step_3B_6_38_Target_Feasibility_Matrix.csv"
)

TARGET_PROXY_PATH = (
    STEP_038_DIR /
    "TB_Portals_March2025_Step_3B_6_38_Target_Proxy_Leakage_Register.csv"
)

# IMPORTANT:
# We intentionally use the authoritative temporal-pair artifact from 3B.6.34,
# NOT the aggregate temporal summary from 3B.6.38.
TEMPORAL_PAIRS_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)


# ===================================================================================================
# 2. INPUT ARTIFACT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

required_paths = {
    "CXR master": CXR_MASTER_PATH,
    "March 2025 genomics": GENOMICS_PATH,
    "Step 3B.6.37 candidate master": CANDIDATE_MASTER_PATH,
    "Step 3B.6.38 target matrix": TARGET_MATRIX_PATH,
    "Step 3B.6.38 target proxy register": TARGET_PROXY_PATH,
    "Step 3B.6.34 authoritative temporal pairs": TEMPORAL_PAIRS_PATH,
}

for label, path in required_paths.items():

    assert path.exists(), (
        f"\nREQUIRED ARTIFACT NOT FOUND\n"
        f"{label}\n"
        f"{path}"
    )

    print(
        f"{label:<55}: FOUND"
    )

print("Input artifact validation: PASS")


# ===================================================================================================
# 3. LOAD ARTIFACTS
# ===================================================================================================

cxr = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)

candidate_master = pd.read_csv(
    CANDIDATE_MASTER_PATH,
    low_memory=False
)

target_matrix = pd.read_csv(
    TARGET_MATRIX_PATH,
    low_memory=False
)

target_proxy_register = pd.read_csv(
    TARGET_PROXY_PATH,
    low_memory=False
)

temporal_pairs = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)


print("\n2. LOADED ARTIFACTS")
print("-" * 100)

print(
    "CXR records:",
    len(cxr)
)

print(
    "Genomic records:",
    len(genomics)
)

print(
    "Candidate conditions:",
    len(candidate_master)
)

print(
    "Target matrix rows:",
    len(target_matrix)
)

print(
    "Target proxy rows:",
    len(target_proxy_register)
)

print(
    "Temporal pair records:",
    len(temporal_pairs)
)


# ===================================================================================================
# 4. SCHEMA VALIDATION
# ===================================================================================================

print("\n3. SCHEMA VALIDATION")
print("-" * 100)


required_cxr_columns = [
    "condition_id",
    "type_of_resistance",
    "outcome",
    "diagnosis_code",
    "case_definition",
    "imaging_date",
    "series_instance_content_url",
]

required_genomic_columns = [
    "condition_id",
    "specimen_id",
    "specimen_collection_date",
    "type_of_resistance",
    "drug_resistance_type",
]

required_temporal_columns = [
    "condition_id",
    "series_instance_content_url",
    "signed_temporal_delta_days",
    "absolute_temporal_distance_days",
]


missing_cxr = [
    c for c in required_cxr_columns
    if c not in cxr.columns
]

missing_genomics = [
    c for c in required_genomic_columns
    if c not in genomics.columns
]

missing_temporal = [
    c for c in required_temporal_columns
    if c not in temporal_pairs.columns
]


print(
    "Missing CXR columns:",
    missing_cxr
)

print(
    "Missing genomic columns:",
    missing_genomics
)

print(
    "Missing temporal-pair columns:",
    missing_temporal
)


assert not missing_cxr, (
    f"CXR schema failure: {missing_cxr}"
)

assert not missing_genomics, (
    f"Genomics schema failure: {missing_genomics}"
)

assert not missing_temporal, (
    f"Temporal-pair schema failure: {missing_temporal}"
)

print("CXR schema: PASS")
print("Genomic schema: PASS")
print("Temporal-pair schema: PASS")


# ===================================================================================================
# 5. NORMALIZE CONDITION IDs
# ===================================================================================================

print("\n4. CONDITION-ID NORMALIZATION")
print("-" * 100)

for df in [
    cxr,
    genomics,
    candidate_master,
    temporal_pairs,
]:

    df["condition_id"] = (
        df["condition_id"]
        .astype(str)
        .str.strip()
    )

print("Condition IDs normalized: PASS")


# ===================================================================================================
# 6. CORE POPULATION VALIDATION
# ===================================================================================================

print("\n5. CORE POPULATION VALIDATION")
print("-" * 100)

assert len(cxr) == 3859

assert (
    cxr["condition_id"].nunique()
    == 3081
)

assert (
    len(candidate_master)
    == 3081
)

assert (
    candidate_master["condition_id"].nunique()
    == 3081
)

common_conditions = (
    set(cxr["condition_id"])
    &
    set(genomics["condition_id"])
)

assert (
    len(common_conditions)
    == 3081
)

print(
    "CXR records       :",
    len(cxr)
)

print(
    "CXR conditions    :",
    cxr["condition_id"].nunique()
)

print(
    "Candidate records :",
    len(candidate_master)
)

print(
    "Common conditions :",
    len(common_conditions)
)

print("Core population: PASS")


# ===================================================================================================
# 7. CONDITION-LEVEL CXR LABEL STRUCTURE
# ===================================================================================================

print("\n6. CONDITION-LEVEL CXR LABEL STRUCTURE")
print("-" * 100)


def unique_nonmissing_values(series):

    return sorted(
        set(
            series
            .dropna()
            .astype(str)
            .str.strip()
        )
    )


cxr_condition_labels = (
    cxr
    .groupby(
        "condition_id"
    )
    .agg(
        cxr_resistance_values=(
            "type_of_resistance",
            unique_nonmissing_values
        ),

        cxr_outcome_values=(
            "outcome",
            unique_nonmissing_values
        ),

        cxr_diagnosis_values=(
            "diagnosis_code",
            unique_nonmissing_values
        ),

        cxr_case_definition_values=(
            "case_definition",
            unique_nonmissing_values
        ),
    )
    .reset_index()
)

assert (
    len(cxr_condition_labels)
    == 3081
)

print(
    "Condition-level CXR records:",
    len(cxr_condition_labels)
)

print("CXR condition-label structure: PASS")


# ===================================================================================================
# 8. CONDITION-LEVEL GENOMIC LABEL STRUCTURE
# ===================================================================================================

print("\n7. CONDITION-LEVEL GENOMIC LABEL STRUCTURE")
print("-" * 100)

genomics_common = genomics[
    genomics["condition_id"].isin(
        common_conditions
    )
].copy()

assert (
    genomics_common["condition_id"].nunique()
    == 3081
)

genomic_condition_labels = (
    genomics_common
    .groupby(
        "condition_id"
    )
    .agg(
        genomic_resistance_values=(
            "type_of_resistance",
            unique_nonmissing_values
        ),

        genomic_drug_resistance_values=(
            "drug_resistance_type",
            unique_nonmissing_values
        ),
    )
    .reset_index()
)

assert (
    len(genomic_condition_labels)
    == 3081
)

print(
    "Multimodal genomic records:",
    len(genomics_common)
)

print(
    "Multimodal genomic conditions:",
    genomics_common["condition_id"].nunique()
)

print("Genomic condition-label structure: PASS")


# ===================================================================================================
# 9. MERGE CONDITION-LEVEL TARGET STRUCTURE
# ===================================================================================================

print("\n8. BUILDING TARGET LABEL MASTER")
print("-" * 100)

target_label_master = (
    cxr_condition_labels
    .merge(
        genomic_condition_labels,
        on="condition_id",
        how="inner",
        validate="one_to_one"
    )
)

assert (
    len(target_label_master)
    == 3081
)

print(
    "Target-label master:",
    len(target_label_master)
)

print("Target-label master: PASS")


# ===================================================================================================
# 10. LABEL STABILITY
# ===================================================================================================

print("\n9. LABEL STABILITY AUDIT")
print("-" * 100)


def label_state(values):

    if len(values) == 0:
        return "MISSING"

    if len(values) == 1:
        return "STABLE_SINGLE_VALUE"

    return "MULTIPLE_VALUES"


for column in [
    "cxr_resistance_values",
    "cxr_outcome_values",
    "cxr_diagnosis_values",
    "cxr_case_definition_values",
    "genomic_resistance_values",
    "genomic_drug_resistance_values",
]:

    stability_column = (
        column.replace(
            "_values",
            "_stability"
        )
    )

    target_label_master[
        stability_column
    ] = (
        target_label_master[
            column
        ]
        .apply(
            label_state
        )
    )


print(
    "\nCXR resistance:"
)

print(
    target_label_master[
        "cxr_resistance_stability"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nCXR outcome:"
)

print(
    target_label_master[
        "cxr_outcome_stability"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nGenomic resistance:"
)

print(
    target_label_master[
        "genomic_resistance_stability"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nGenomic drug resistance profile:"
)

print(
    target_label_master[
        "genomic_drug_resistance_stability"
    ]
    .value_counts()
    .to_string()
)

print("Label stability audit: PASS")


# ===================================================================================================
# 11. TARGET LABEL REPRESENTATION
# ===================================================================================================

print("\n10. TARGET LABEL REPRESENTATION")
print("-" * 100)


def single_value_or_status(values):

    if len(values) == 0:
        return "MISSING"

    if len(values) == 1:
        return values[0]

    return "MULTIPLE_VALUES"


target_label_master[
    "cxr_resistance_label"
] = (
    target_label_master[
        "cxr_resistance_values"
    ]
    .apply(
        single_value_or_status
    )
)

target_label_master[
    "genomic_resistance_label"
] = (
    target_label_master[
        "genomic_resistance_values"
    ]
    .apply(
        single_value_or_status
    )
)

target_label_master[
    "genomic_drug_resistance_label"
] = (
    target_label_master[
        "genomic_drug_resistance_values"
    ]
    .apply(
        single_value_or_status
    )
)

target_label_master[
    "outcome_label"
] = (
    target_label_master[
        "cxr_outcome_values"
    ]
    .apply(
        single_value_or_status
    )
)


print(
    "\nCXR resistance:"
)

print(
    target_label_master[
        "cxr_resistance_label"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nGenomic resistance:"
)

print(
    target_label_master[
        "genomic_resistance_label"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nGenomic drug-resistance profile:"
)

print(
    target_label_master[
        "genomic_drug_resistance_label"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nClinical outcome:"
)

print(
    target_label_master[
        "outcome_label"
    ]
    .value_counts()
    .to_string()
)

print("Target label representation: PASS")


# ===================================================================================================
# 12. DESCRIPTIVE RESISTANCE BINARY GROUPING
# ===================================================================================================

print("\n11. DESCRIPTIVE RESISTANCE BINARY GROUPING")
print("-" * 100)


RESISTANT_CLASSES = {
    "MDR non XDR",
    "XDR",
    "Mono DR",
    "Poly DR",
    "Pre-XDR",
}

SENSITIVE_CLASS = "Sensitive"


def resistance_binary_group(value):

    if pd.isna(value):
        return "MISSING"

    value = str(value).strip()

    if value == SENSITIVE_CLASS:
        return "SENSITIVE"

    if value in RESISTANT_CLASSES:
        return "RESISTANT"

    if value == "MULTIPLE_VALUES":
        return "MULTIPLE_VALUES"

    return "OTHER"


target_label_master[
    "cxr_resistance_binary_descriptive"
] = (
    target_label_master[
        "cxr_resistance_label"
    ]
    .apply(
        resistance_binary_group
    )
)

target_label_master[
    "genomic_resistance_binary_descriptive"
] = (
    target_label_master[
        "genomic_resistance_label"
    ]
    .apply(
        resistance_binary_group
    )
)


print(
    "\nCXR:"
)

print(
    target_label_master[
        "cxr_resistance_binary_descriptive"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nGenomics:"
)

print(
    target_label_master[
        "genomic_resistance_binary_descriptive"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nNo binary definition is selected."
)

print("Descriptive binary grouping: PASS")


# ===================================================================================================
# 13. CXR ↔ GENOMIC RESISTANCE CONSISTENCY
# ===================================================================================================

print("\n12. CXR ↔ GENOMIC RESISTANCE CONSISTENCY")
print("-" * 100)

target_label_master[
    "exact_resistance_match"
] = (
    target_label_master[
        "cxr_resistance_label"
    ]
    ==
    target_label_master[
        "genomic_resistance_label"
    ]
)

target_label_master[
    "binary_resistance_match"
] = (
    target_label_master[
        "cxr_resistance_binary_descriptive"
    ]
    ==
    target_label_master[
        "genomic_resistance_binary_descriptive"
    ]
)

print(
    "Exact resistance match:",
    int(
        target_label_master[
            "exact_resistance_match"
        ].sum()
    )
)

print(
    "Binary resistance match:",
    int(
        target_label_master[
            "binary_resistance_match"
        ].sum()
    )
)

assert (
    target_label_master[
        "exact_resistance_match"
    ].sum()
    == 3081
)

assert (
    target_label_master[
        "binary_resistance_match"
    ].sum()
    == 3081
)

print("Resistance consistency: PASS")


# ===================================================================================================
# 14. OUTCOME DESCRIPTIVE STRUCTURE
# ===================================================================================================

print("\n13. CLINICAL OUTCOME STRUCTURE")
print("-" * 100)


def outcome_descriptive_group(value):

    if pd.isna(value):
        return "MISSING"

    value = str(value).strip()

    if value in {
        "Cured",
        "Completed",
    }:
        return "FAVORABLE_OUTCOME"

    if value in {
        "Failure",
        "Died",
        "Lost to follow up",
        "Palliative Care",
    }:
        return "UNFAVORABLE_OR_ADVERSE_OUTCOME"

    if value in {
        "Still on treatment",
        "Unknown",
    }:
        return "NOT_FINALIZED_OR_UNKNOWN"

    if value == "MULTIPLE_VALUES":
        return "MULTIPLE_VALUES"

    return "OTHER"


target_label_master[
    "outcome_binary_descriptive"
] = (
    target_label_master[
        "outcome_label"
    ]
    .apply(
        outcome_descriptive_group
    )
)

print(
    target_label_master[
        "outcome_binary_descriptive"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nThis is descriptive only."
)

print(
    "No outcome target is selected."
)

print("Clinical outcome structure: PASS")


# ===================================================================================================
# 15. AUTHORITATIVE TEMPORAL-PAIR VALIDATION
# ===================================================================================================

print("\n14. AUTHORITATIVE CXR ↔ GENOMIC TEMPORAL DATA")
print("-" * 100)

temporal_pairs[
    "condition_id"
] = (
    temporal_pairs[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)

temporal_pairs[
    "signed_temporal_delta_days"
] = pd.to_numeric(
    temporal_pairs[
        "signed_temporal_delta_days"
    ],
    errors="coerce"
)

temporal_pairs[
    "absolute_temporal_distance_days"
] = pd.to_numeric(
    temporal_pairs[
        "absolute_temporal_distance_days"
    ],
    errors="coerce"
)

assert (
    temporal_pairs[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    temporal_pairs[
        "signed_temporal_delta_days"
    ].notna().all()
)

assert (
    temporal_pairs[
        "absolute_temporal_distance_days"
    ].notna().all()
)

print(
    "Temporal pair records:",
    len(temporal_pairs)
)

print(
    "Temporal conditions:",
    temporal_pairs[
        "condition_id"
    ].nunique()
)

print(
    "Temporal CXR URLs:",
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
)

print("Authoritative temporal-pair validation: PASS")


# ===================================================================================================
# 16. TEMPORAL DIRECTION DISTRIBUTION
# ===================================================================================================

print("\n15. TEMPORAL DIRECTION DISTRIBUTION")
print("-" * 100)


temporal_pairs[
    "temporal_direction"
] = np.select(
    [
        temporal_pairs[
            "signed_temporal_delta_days"
        ] < 0,

        temporal_pairs[
            "signed_temporal_delta_days"
        ] == 0,

        temporal_pairs[
            "signed_temporal_delta_days"
        ] > 0,
    ],
    [
        "GENOMIC_BEFORE_CXR",
        "SAME_DAY",
        "GENOMIC_AFTER_CXR",
    ],
    default="INVALID",
)


direction_counts = (
    temporal_pairs[
        "temporal_direction"
    ]
    .value_counts()
)

print(
    direction_counts.to_string()
)

assert (
    not (
        temporal_pairs[
            "temporal_direction"
        ]
        ==
        "INVALID"
    )
    .any()
)

print("Temporal direction distribution: PASS")


# ===================================================================================================
# 17. NEAREST GENOMIC TIMEPOINT PER CXR
# ===================================================================================================

print("\n16. NEAREST GENOMIC TIMEPOINT PER CXR")
print("-" * 100)

temporal_sorted = (
    temporal_pairs
    .sort_values(
        [
            "series_instance_content_url",
            "absolute_temporal_distance_days",
        ]
    )
    .copy()
)

nearest_distance = (
    temporal_sorted
    .groupby(
        "series_instance_content_url"
    )[
        "absolute_temporal_distance_days"
    ]
    .transform(
        "min"
    )
)

nearest_pairs = (
    temporal_sorted[
        temporal_sorted[
            "absolute_temporal_distance_days"
        ]
        ==
        nearest_distance
    ]
    .copy()
)

assert (
    nearest_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

print(
    "CXR URLs with at least one nearest genomic timepoint:",
    nearest_pairs[
        "series_instance_content_url"
    ].nunique()
)

print(
    "Nearest-pair rows including ties:",
    len(nearest_pairs)
)

print("Nearest genomic timepoint audit: PASS")


# ===================================================================================================
# 18. NEAREST TEMPORAL DIRECTION
# ===================================================================================================

print("\n17. NEAREST TEMPORAL DIRECTION")
print("-" * 100)

nearest_pairs[
    "nearest_direction"
] = (
    nearest_pairs[
        "signed_temporal_delta_days"
    ]
    .apply(
        lambda x:
        "GENOMIC_BEFORE_CXR"
        if x < 0
        else (
            "SAME_DAY"
            if x == 0
            else "GENOMIC_AFTER_CXR"
        )
    )
)

nearest_direction_summary = (
    nearest_pairs[
        [
            "series_instance_content_url",
            "nearest_direction",
        ]
    ]
    .drop_duplicates()
)

print(
    nearest_direction_summary[
        "nearest_direction"
    ]
    .value_counts()
    .to_string()
)

print("Nearest temporal direction: PASS")


# ===================================================================================================
# 19. CONDITION-LEVEL TEMPORAL STRUCTURE
# ===================================================================================================

print("\n18. CONDITION-LEVEL TEMPORAL STRUCTURE")
print("-" * 100)

condition_temporal = (
    temporal_pairs
    .groupby(
        "condition_id"
    )
    .agg(
        any_genomic_before_cxr=(
            "signed_temporal_delta_days",
            lambda x:
            (x < 0).any()
        ),

        any_same_day=(
            "signed_temporal_delta_days",
            lambda x:
            (x == 0).any()
        ),

        any_genomic_after_cxr=(
            "signed_temporal_delta_days",
            lambda x:
            (x > 0).any()
        ),

        minimum_signed_delta_days=(
            "signed_temporal_delta_days",
            "min"
        ),

        maximum_signed_delta_days=(
            "signed_temporal_delta_days",
            "max"
        ),

        minimum_absolute_distance_days=(
            "absolute_temporal_distance_days",
            "min"
        ),
    )
    .reset_index()
)

assert (
    len(condition_temporal)
    == 3081
)

condition_temporal[
    "temporal_direction_count"
] = (
    condition_temporal[
        [
            "any_genomic_before_cxr",
            "any_same_day",
            "any_genomic_after_cxr",
        ]
    ]
    .sum(
        axis=1
    )
)

print(
    condition_temporal[
        "temporal_direction_count"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)

print("Condition temporal structure: PASS")


# ===================================================================================================
# 20. MULTIPLE GENOMIC SPECIMEN AUDIT
# ===================================================================================================

print("\n19. MULTIPLE GENOMIC SPECIMEN AUDIT")
print("-" * 100)

genomic_specimen_counts = (
    genomics_common
    .groupby(
        "condition_id"
    )[
        "specimen_id"
    ]
    .nunique()
    .rename(
        "unique_specimen_count"
    )
    .reset_index()
)

assert (
    len(genomic_specimen_counts)
    == 3081
)

multiple_specimen_conditions = (
    genomic_specimen_counts[
        genomic_specimen_counts[
            "unique_specimen_count"
        ]
        > 1
    ]
)

print(
    "Conditions with multiple genomic specimens:",
    len(multiple_specimen_conditions)
)

print(
    genomic_specimen_counts[
        "unique_specimen_count"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)

print("Multiple-genomic-specimen audit: PASS")


# ===================================================================================================
# 21. GENOMIC DRUG-RESISTANCE INSTABILITY
# ===================================================================================================

print("\n20. GENOMIC DRUG-RESISTANCE PROFILE STABILITY")
print("-" * 100)

drug_profile_multiple = (
    target_label_master[
        "genomic_drug_resistance_stability"
    ]
    ==
    "MULTIPLE_VALUES"
)

print(
    "Conditions with multiple genomic drug-resistance profiles:",
    int(
        drug_profile_multiple.sum()
    )
)

assert (
    int(
        drug_profile_multiple.sum()
    )
    == 20
)

print(
    "These 20 conditions are retained in the audit."
)

print(
    "They are NOT automatically excluded."
)

print("Drug-resistance profile stability: PASS")


# ===================================================================================================
# 22. TARGET CANDIDATE IMPACT MATRIX
# ===================================================================================================

print("\n21. TARGET CANDIDATE IMPACT MATRIX")
print("-" * 100)


def count_state(series, state):

    return int(
        (
            series
            ==
            state
        )
        .sum()
    )


target_impact_rows = []


target_definitions = [
    (
        "CXR type_of_resistance",
        "CXR",
        "RESISTANCE",
        "cxr_resistance_label",
    ),
    (
        "Genomic type_of_resistance",
        "GENOMICS",
        "RESISTANCE",
        "genomic_resistance_label",
    ),
    (
        "Genomic drug_resistance_type",
        "GENOMICS",
        "RESISTANCE_PROFILE",
        "genomic_drug_resistance_label",
    ),
    (
        "CXR outcome",
        "CXR",
        "CLINICAL_OUTCOME",
        "outcome_label",
    ),
]


for (
    target_name,
    source,
    family,
    column,
) in target_definitions:

    series = (
        target_label_master[
            column
        ]
    )

    missing_count = count_state(
        series,
        "MISSING"
    )

    multiple_count = count_state(
        series,
        "MULTIPLE_VALUES"
    )

    unambiguous_count = (
        len(series)
        -
        missing_count
        -
        multiple_count
    )

    target_impact_rows.append(
        {
            "candidate_target":
                target_name,

            "source":
                source,

            "target_family":
                family,

            "conditions_total":
                3081,

            "conditions_single_unambiguous_label":
                unambiguous_count,

            "conditions_multiple_values":
                multiple_count,

            "conditions_missing":
                missing_count,

            "potential_unambiguous_percentage":
                (
                    unambiguous_count
                    /
                    3081
                    *
                    100
                ),

            "selection_status":
                "NOT_SELECTED",
        }
    )


target_impact_matrix = pd.DataFrame(
    target_impact_rows
)

print(
    target_impact_matrix.to_string(
        index=False
    )
)

assert (
    len(target_impact_matrix)
    == 4
)

assert (
    target_impact_matrix[
        "selection_status"
    ].eq(
        "NOT_SELECTED"
    ).all()
)

print("Target candidate impact matrix: PASS")


# ===================================================================================================
# 23. TEMPORAL IMPACT MATRIX
# ===================================================================================================

print("\n22. TEMPORAL IMPACT MATRIX")
print("-" * 100)

temporal_condition_summary = pd.DataFrame(
    [
        {
            "temporal_pattern":
                "GENOMIC_BEFORE_CXR",
            "condition_count":
                int(
                    condition_temporal[
                        "any_genomic_before_cxr"
                    ].sum()
                ),
        },
        {
            "temporal_pattern":
                "SAME_DAY",
            "condition_count":
                int(
                    condition_temporal[
                        "any_same_day"
                    ].sum()
                ),
        },
        {
            "temporal_pattern":
                "GENOMIC_AFTER_CXR",
            "condition_count":
                int(
                    condition_temporal[
                        "any_genomic_after_cxr"
                    ].sum()
                ),
        },
    ]
)

temporal_condition_summary[
    "percentage_of_3081"
] = (
    temporal_condition_summary[
        "condition_count"
    ]
    /
    3081
    *
    100
)

print(
    temporal_condition_summary.to_string(
        index=False
    )
)

print(
    "\nNo temporal direction is selected as an eligibility rule."
)

print("Temporal impact matrix: PASS")


# ===================================================================================================
# 24. TARGET LEAKAGE REGISTER
# ===================================================================================================

print("\n23. TARGET LEAKAGE REGISTER")
print("-" * 100)

leakage_register = pd.DataFrame(
    [
        {
            "field":
                "CXR:type_of_resistance",
            "role_if_resistance_target":
                "DIRECT_TARGET",
            "input_status":
                "DO_NOT_USE_AS_PREDICTOR",
        },
        {
            "field":
                "GENOMICS:type_of_resistance",
            "role_if_resistance_target":
                "DIRECT_TARGET",
            "input_status":
                "DO_NOT_USE_AS_PREDICTOR",
        },
        {
            "field":
                "GENOMICS:drug_resistance_type",
            "role_if_resistance_target":
                "DIRECT_TARGET_OR_TARGET_PROXY",
            "input_status":
                "DO_NOT_USE_AS_PREDICTOR",
        },
        {
            "field":
                "CXR:outcome",
            "role_if_outcome_target":
                "DIRECT_TARGET",
            "input_status":
                "DO_NOT_USE_AS_PREDICTOR",
        },
        {
            "field":
                "patient_id",
            "role":
                "IDENTIFIER",
            "input_status":
                "DO_NOT_USE_AS_PREDICTOR",
        },
        {
            "field":
                "condition_id",
            "role":
                "CASE_IDENTIFIER",
            "input_status":
                "DO_NOT_USE_AS_PREDICTOR",
        },
        {
            "field":
                "specimen_id",
            "role":
                "SPECIMEN_IDENTIFIER",
            "input_status":
                "DO_NOT_USE_AS_PREDICTOR",
        },
    ]
)

print(
    leakage_register.to_string(
        index=False
    )
)

assert (
    leakage_register[
        "input_status"
    ]
    .eq(
        "DO_NOT_USE_AS_PREDICTOR"
    )
    .all()
)

print("Target leakage register: PASS")


# ===================================================================================================
# 25. RESEARCH QUESTION CANDIDATE REGISTER
# ===================================================================================================

print("\n24. RESEARCH QUESTION CANDIDATE REGISTER")
print("-" * 100)

research_question_candidates = pd.DataFrame(
    [
        {
            "candidate_id":
                "RQ-01",

            "research_question":
                "Can chest X-ray imaging features predict a predefined "
                "tuberculosis drug-resistance category?",

            "target_family":
                "RESISTANCE",

            "target_source":
                "CXR OR GENOMICS",

            "status":
                "NOT_SELECTED",
        },
        {
            "candidate_id":
                "RQ-02",

            "research_question":
                "Can chest X-ray imaging features predict a predefined "
                "genomic tuberculosis resistance profile?",

            "target_family":
                "RESISTANCE_PROFILE",

            "target_source":
                "GENOMICS",

            "status":
                "NOT_SELECTED",
        },
        {
            "candidate_id":
                "RQ-03",

            "research_question":
                "Can chest X-ray imaging features predict a predefined "
                "clinical treatment outcome?",

            "target_family":
                "CLINICAL_OUTCOME",

            "target_source":
                "CXR",

            "status":
                "NOT_SELECTED",
        },
        {
            "candidate_id":
                "RQ-04",

            "research_question":
                "Can CXR and genomic information jointly characterize "
                "tuberculosis drug resistance?",

            "target_family":
                "RESISTANCE",

            "target_source":
                "MULTIMODAL",

            "status":
                "NOT_SELECTED",
        },
    ]
)

print(
    research_question_candidates.to_string(
        index=False
    )
)

assert (
    len(research_question_candidates)
    == 4
)

assert (
    research_question_candidates[
        "status"
    ]
    .eq(
        "NOT_SELECTED"
    )
    .all()
)

print("Research-question register: PASS")


# ===================================================================================================
# 26. DECISION LOCK
# ===================================================================================================

print("\n25. STUDY-DESIGN DECISION LOCK")
print("-" * 100)

decision_lock = pd.DataFrame(
    [
        {
            "decision":
                "research_question",
            "status":
                "NOT_SELECTED",
        },
        {
            "decision":
                "target_family",
            "status":
                "NOT_SELECTED",
        },
        {
            "decision":
                "target_source",
            "status":
                "NOT_SELECTED",
        },
        {
            "decision":
                "target_definition",
            "status":
                "NOT_SELECTED",
        },
        {
            "decision":
                "binary_or_multiclass",
            "status":
                "NOT_SELECTED",
        },
        {
            "decision":
                "positive_class",
            "status":
                "NOT_SELECTED",
        },
        {
            "decision":
                "negative_class",
            "status":
                "NOT_SELECTED",
        },
        {
            "decision":
                "temporal_rule",
            "status":
                "NOT_SELECTED",
        },
        {
            "decision":
                "representative_cxr_rule",
            "status":
                "NOT_SELECTED",
        },
        {
            "decision":
                "representative_genomic_rule",
            "status":
                "NOT_SELECTED",
        },
        {
            "decision":
                "final_cohort",
            "status":
                "NOT_SELECTED",
        },
    ]
)

print(
    decision_lock.to_string(
        index=False
    )
)

assert (
    decision_lock[
        "status"
    ]
    .eq(
        "NOT_SELECTED"
    )
    .all()
)

print("Study-design decision lock: PASS")


# ===================================================================================================
# 27. SAVE OUTPUTS
# ===================================================================================================

print("\n26. SAVING OUTPUTS")
print("-" * 100)

TARGET_LABEL_MASTER_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_39_Target_Label_Structure_Master.csv"
)

TARGET_IMPACT_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_39_Target_Candidate_Impact_Matrix.csv"
)

TEMPORAL_SUMMARY_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_39_Temporal_Impact_Summary.csv"
)

TEMPORAL_CONDITION_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_39_Condition_Temporal_Structure.csv"
)

LEAKAGE_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_39_Target_Leakage_Register.csv"
)

RQ_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_39_Research_Question_Candidates.csv"
)

DECISION_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_39_Study_Design_Decision_Lock.csv"
)

GENOMIC_SPECIMEN_OUTPUT = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_39_Multiple_Genomic_Specimen_Audit.csv"
)


target_label_master.to_csv(
    TARGET_LABEL_MASTER_OUTPUT,
    index=False
)

target_impact_matrix.to_csv(
    TARGET_IMPACT_OUTPUT,
    index=False
)

temporal_condition_summary.to_csv(
    TEMPORAL_SUMMARY_OUTPUT,
    index=False
)

condition_temporal.to_csv(
    TEMPORAL_CONDITION_OUTPUT,
    index=False
)

leakage_register.to_csv(
    LEAKAGE_OUTPUT,
    index=False
)

research_question_candidates.to_csv(
    RQ_OUTPUT,
    index=False
)

decision_lock.to_csv(
    DECISION_OUTPUT,
    index=False
)

genomic_specimen_counts.to_csv(
    GENOMIC_SPECIMEN_OUTPUT,
    index=False
)


# ===================================================================================================
# 28. OUTPUT VALIDATION
# ===================================================================================================

print("\n27. OUTPUT FILE VALIDATION")
print("-" * 100)

output_paths = [
    TARGET_LABEL_MASTER_OUTPUT,
    TARGET_IMPACT_OUTPUT,
    TEMPORAL_SUMMARY_OUTPUT,
    TEMPORAL_CONDITION_OUTPUT,
    LEAKAGE_OUTPUT,
    RQ_OUTPUT,
    DECISION_OUTPUT,
    GENOMIC_SPECIMEN_OUTPUT,
]

for output_path in output_paths:

    assert output_path.exists(), (
        f"Output missing:\n{output_path}"
    )

    assert output_path.stat().st_size > 0, (
        f"Output is empty:\n{output_path}"
    )

    print(
        f"{output_path.name}: PASS"
    )


# ===================================================================================================
# 29. FINAL INTEGRITY CHECKS
# ===================================================================================================

print("\n28. FINAL INTEGRITY CHECKS")
print("-" * 100)

assert len(cxr) == 3859

assert (
    cxr["condition_id"].nunique()
    == 3081
)

assert (
    len(target_label_master)
    == 3081
)

assert (
    len(condition_temporal)
    == 3081
)

assert (
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    target_impact_matrix[
        "selection_status"
    ]
    .eq(
        "NOT_SELECTED"
    )
    .all()
)

assert (
    decision_lock[
        "status"
    ]
    .eq(
        "NOT_SELECTED"
    )
    .all()
)

print(
    "3,859 CXR records                    : PASS"
)

print(
    "3,081 multimodal conditions          : PASS"
)

print(
    "Condition-level target master       : PASS"
)

print(
    "Temporal pairs linked to all CXRs   : PASS"
)

print(
    "Target impact matrix                : PASS"
)

print(
    "Multiple-genomic specimen audit     : PASS"
)

print(
    "Target leakage register             : PASS"
)

print(
    "Research-question register          : PASS"
)

print(
    "Study-design decision lock          : PASS"
)


# ===================================================================================================
# 30. SOURCE PROTECTION
# ===================================================================================================

print("\n29. SOURCE PROTECTION")
print("-" * 100)

print(
    "Raw DICOM modification             : NO"
)

print(
    "Source CXR metadata modification   : NO"
)

print(
    "Source genomics modification       : NO"
)

print(
    "Human-review artifacts modified    : NO"
)

print(
    "Image preprocessing                : NO"
)

print(
    "Model training                     : NO"
)

print(
    "Target selection                   : NO"
)

print(
    "Final cohort selection             : NO"
)

print(
    "Representative CXR selection      : NO"
)

print(
    "Representative genomic selection   : NO"
)

print(
    "Temporal eligibility rule selected  : NO"
)

print("Source protection: PASS")


# ===================================================================================================
# 31. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.39 — FINAL STATUS")
print("=" * 100)

print(
    "CXR target structure analyzed               : PASS"
)

print(
    "Genomic target structure analyzed           : PASS"
)

print(
    "Within-condition label stability            : PASS"
)

print(
    "Resistance target definitions analyzed      : PASS"
)

print(
    "Outcome target analyzed                     : PASS"
)

print(
    "Resistance consistency analyzed             : PASS"
)

print(
    "Authoritative temporal pairs validated      : PASS"
)

print(
    "Temporal direction analyzed                 : PASS"
)

print(
    "Nearest genomic timepoints analyzed         : PASS"
)

print(
    "Multiple genomic specimens analyzed         : PASS"
)

print(
    "Drug-resistance profile instability audited : PASS"
)

print(
    "Target candidate impact analyzed            : PASS"
)

print(
    "Target leakage register created             : PASS"
)

print(
    "Research-question candidates documented     : PASS"
)

print(
    "Target selection                            : NOT PERFORMED"
)

print(
    "Final cohort selection                      : NOT PERFORMED"
)

print(
    "Temporal rule selection                     : NOT PERFORMED"
)

print(
    "Representative CXR selection                : NOT PERFORMED"
)

print(
    "Representative genomic selection            : NOT PERFORMED"
)

print(
    "Preprocessing                               : NOT PERFORMED"
)

print(
    "Model training                              : NOT PERFORMED"
)

print(
    "DICOM modification                          : NO"
)

print(
    "Source metadata modification                : NO"
)

print(
    "Output validation                           : PASS"
)

print("\n" + "=" * 100)
print("STEP 3B.6.39 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.42 — TEMPORAL DIRECTION DISCREPANCY DIAGNOSTIC
# ===================================================================================================
#
# PURPOSE:
#   Inspect the ACTUAL temporal_direction values produced by Step 3B.6.34
#   and compare them with the direction implied by signed_temporal_delta_days.
#
# IMPORTANT:
#   - Read-only
#   - No files modified
#   - No assumptions about the stored direction labels
#   - No target/cohort decisions
#   - No preprocessing
#
# ===================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np

TBP_ROOT = Path(r"C:\TBP")
STEP_034_DIR = (
    TBP_ROOT /
    "Metadata" /
    "Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
)

TEMPORAL_PAIRS_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

assert TEMPORAL_PAIRS_PATH.exists(), (
    f"Temporal pairs file not found:\n{TEMPORAL_PAIRS_PATH}"
)

df = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)

required_columns = [
    "condition_id",
    "series_instance_content_url",
    "imaging_date_num",
    "specimen_id",
    "specimen_collection_date_num",
    "signed_temporal_delta_days",
    "absolute_temporal_distance_days",
    "temporal_direction",
    "absolute_distance_category",
]

missing = [
    c for c in required_columns
    if c not in df.columns
]

assert not missing, (
    f"Missing required columns:\n{missing}"
)

df["signed_temporal_delta_days"] = pd.to_numeric(
    df["signed_temporal_delta_days"],
    errors="coerce"
)

# ---------------------------------------------------------------------------------------------------
# 1. ACTUAL STORED TEMPORAL-DIRECTION VALUES
# ---------------------------------------------------------------------------------------------------

print("=" * 100)
print("STEP 3B.6.42 — TEMPORAL DIRECTION DISCREPANCY DIAGNOSTIC")
print("=" * 100)

print("\n1. ACTUAL STORED temporal_direction VALUES")
print("-" * 100)

stored_direction_counts = (
    df["temporal_direction"]
    .value_counts(dropna=False)
)

print(
    stored_direction_counts.to_string()
)

print(
    "\nNumber of distinct stored direction values:",
    df["temporal_direction"].nunique(dropna=False)
)


# ---------------------------------------------------------------------------------------------------
# 2. DERIVE DIRECTION DIRECTLY FROM SIGNED DELTA
# ---------------------------------------------------------------------------------------------------

def derive_direction(delta):

    if pd.isna(delta):
        return "INVALID_MISSING_DELTA"

    if delta < 0:
        return "GENOMIC_BEFORE_CXR"

    if delta == 0:
        return "SAME_DAY"

    if delta > 0:
        return "GENOMIC_AFTER_CXR"

    return "INVALID"


df["derived_direction_from_delta"] = (
    df["signed_temporal_delta_days"]
    .apply(derive_direction)
)


print("\n2. DERIVED DIRECTION FROM SIGNED DELTA")
print("-" * 100)

print(
    df["derived_direction_from_delta"]
    .value_counts(dropna=False)
    .to_string()
)


# ---------------------------------------------------------------------------------------------------
# 3. CROSS-TABULATION
# ---------------------------------------------------------------------------------------------------

print("\n3. STORED vs DERIVED TEMPORAL DIRECTION")
print("-" * 100)

direction_crosstab = pd.crosstab(
    df["derived_direction_from_delta"],
    df["temporal_direction"],
    dropna=False
)

print(
    direction_crosstab.to_string()
)


# ---------------------------------------------------------------------------------------------------
# 4. IDENTIFY DISAGREEMENT ROWS
# ---------------------------------------------------------------------------------------------------

df["direction_agreement"] = (
    df["derived_direction_from_delta"].astype(str)
    ==
    df["temporal_direction"].astype(str)
)

disagreements = df[
    ~df["direction_agreement"]
].copy()


print("\n4. DIRECTION DISAGREEMENT")
print("-" * 100)

print(
    "Total temporal-pair rows:",
    len(df)
)

print(
    "Agreement rows:",
    int(df["direction_agreement"].sum())
)

print(
    "Disagreement rows:",
    len(disagreements)
)

print(
    "Agreement percentage:",
    round(
        df["direction_agreement"].mean() * 100,
        4
    )
)


# ---------------------------------------------------------------------------------------------------
# 5. SHOW ACTUAL DISAGREEMENT PATTERNS
# ---------------------------------------------------------------------------------------------------

print("\n5. DISAGREEMENT PATTERNS")
print("-" * 100)

if len(disagreements) > 0:

    disagreement_patterns = (
        disagreements
        .groupby(
            [
                "derived_direction_from_delta",
                "temporal_direction",
            ],
            dropna=False
        )
        .size()
        .reset_index(
            name="row_count"
        )
        .sort_values(
            "row_count",
            ascending=False
        )
    )

    print(
        disagreement_patterns.to_string(
            index=False
        )
    )

else:

    print(
        "No direction disagreements found."
    )


# ---------------------------------------------------------------------------------------------------
# 6. INSPECT FIRST DISAGREEMENT ROWS
# ---------------------------------------------------------------------------------------------------

print("\n6. FIRST DISAGREEMENT ROWS")
print("-" * 100)

if len(disagreements) > 0:

    display_columns = [
        "condition_id",
        "series_instance_content_url",
        "imaging_date_num",
        "specimen_id",
        "specimen_collection_date_num",
        "signed_temporal_delta_days",
        "absolute_temporal_distance_days",
        "derived_direction_from_delta",
        "temporal_direction",
        "absolute_distance_category",
    ]

    print(
        disagreements[
            display_columns
        ]
        .head(30)
        .to_string(index=False)
    )

else:

    print(
        "No disagreement rows to display."
    )


# ---------------------------------------------------------------------------------------------------
# 7. SIGNED DELTA DISTRIBUTION
# ---------------------------------------------------------------------------------------------------

print("\n7. SIGNED TEMPORAL DELTA DISTRIBUTION")
print("-" * 100)

delta_summary = pd.DataFrame(
    {
        "statistic": [
            "count",
            "missing",
            "minimum",
            "negative_count",
            "zero_count",
            "positive_count",
            "maximum",
        ],

        "value": [
            int(
                df["signed_temporal_delta_days"]
                .notna()
                .sum()
            ),

            int(
                df["signed_temporal_delta_days"]
                .isna()
                .sum()
            ),

            float(
                df["signed_temporal_delta_days"]
                .min()
            ),

            int(
                (
                    df["signed_temporal_delta_days"]
                    < 0
                )
                .sum()
            ),

            int(
                (
                    df["signed_temporal_delta_days"]
                    == 0
                )
                .sum()
            ),

            int(
                (
                    df["signed_temporal_delta_days"]
                    > 0
                )
                .sum()
            ),

            float(
                df["signed_temporal_delta_days"]
                .max()
            ),
        ],
    }
)

print(
    delta_summary.to_string(
        index=False
    )
)


# ---------------------------------------------------------------------------------------------------
# 8. ACTUAL ABSOLUTE-DISTANCE CATEGORY VALUES
# ---------------------------------------------------------------------------------------------------

print("\n8. ACTUAL STORED absolute_distance_category VALUES")
print("-" * 100)

print(
    df["absolute_distance_category"]
    .value_counts(dropna=False)
    .to_string()
)


# ---------------------------------------------------------------------------------------------------
# 9. FINAL DIAGNOSTIC STATUS
# ---------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("DIAGNOSTIC STATUS")
print("=" * 100)

if len(disagreements) == 0:

    print(
        "Stored temporal_direction agrees with signed_temporal_delta_days."
    )

else:

    print(
        "A discrepancy exists between the stored temporal_direction "
        "and the signed temporal delta."
    )

    print(
        "The actual stored labels and disagreement rows have been "
        "displayed above."
    )

print(
    "\nNO FILES WERE CREATED OR MODIFIED."
)

print("=" * 100)
```

```python
# ===================================================================================================
# STEP 3B.6.42 — TEMPORAL ALIGNMENT & PREDICTION-DIRECTION AUDIT
# CORRECTED FINAL VERSION
# ===================================================================================================
#
# PURPOSE
# -------
# Audit the temporal relationship between CXR imaging and genomic specimen collection
# before making any target, temporal-rule, representative-sample, or final-cohort decision.
#
# IMPORTANT METHODOLOGICAL LOCKS
# ------------------------------
# 1. condition_id is the case-level unit.
# 2. CXR × genomic temporal pairs are retained at their natural temporal-pair level.
# 3. Repeated CXRs are NOT treated as independent patients.
# 4. Multiple genomic timepoints are NOT silently collapsed.
# 5. Nearest-timepoint ties are NOT silently resolved.
# 6. No target is selected.
# 7. No temporal cutoff is selected.
# 8. No representative CXR is selected.
# 9. No representative genomic specimen is selected.
# 10. No preprocessing is performed.
# 11. No model training is performed.
# 12. Source data are not modified.
#
# CRITICAL CORRECTION FROM PREVIOUS RUN
# --------------------------------------
# Step 3B.6.34 uses the labels:
#     GENOMICS_BEFORE_CXR
#     GENOMICS_AFTER_CXR
#     SAME_DAY
#
# The signed delta remains:
#     specimen_collection_date_num - imaging_date_num
#
# Therefore:
#     delta < 0  -> GENOMICS_BEFORE_CXR
#     delta = 0  -> SAME_DAY
#     delta > 0  -> GENOMICS_AFTER_CXR
#
# Also:
#     nearest artifact = 3,871 rows for 3,859 CXR records
#     because some CXR records have nearest-timepoint ties.
#
# ===================================================================================================


from pathlib import Path
import pandas as pd
import numpy as np


# ===================================================================================================
# 1. INITIALIZATION
# ===================================================================================================

TBP_ROOT = Path(r"C:\TBP")
METADATA_DIR = TBP_ROOT / "Metadata"

STEP_024_DIR = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03"
)

STEP_034_DIR = (
    METADATA_DIR /
    "Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
)

STEP_039_DIR = (
    METADATA_DIR /
    "Step_3B_6_39_Target_Definition_Cohort_Impact"
)

STEP_040_DIR = (
    METADATA_DIR /
    "Step_3B_6_40_Research_Question_Target_Decision_Gate"
)

STEP_041_DIR = (
    METADATA_DIR /
    "Step_3B_6_41_Target_Specific_Cohort_Impact"
)

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_42_Temporal_Alignment_Prediction_Direction"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("=" * 100)
print("STEP 3B.6.42 — TEMPORAL ALIGNMENT & PREDICTION-DIRECTION AUDIT")
print("=" * 100)


# ===================================================================================================
# 2. EXACT VERIFIED INPUT PATHS
# ===================================================================================================

CXR_MASTER_PATH = (
    STEP_024_DIR /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

TEMPORAL_PAIRS_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

NEAREST_PAIRS_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Nearest_Genomic_Timepoint_Per_CXR.csv"
)

UNIQUE_GENOMIC_TIMEPOINTS_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

MULTIPLE_GENOMIC_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Multiple_Genomic_Timepoint_Conditions.csv"
)

TEMPORAL_DIRECTION_SUMMARY_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Temporal_Direction_Summary.csv"
)

TEMPORAL_WINDOW_SUMMARY_PATH = (
    STEP_034_DIR /
    "TB_Portals_March2025_Step_3B_6_34_Temporal_Window_Summary.csv"
)


# ===================================================================================================
# 3. INPUT VALIDATION
# ===================================================================================================

print("\n1. INPUT ARTIFACT VALIDATION")
print("-" * 100)

required_inputs = {
    "CXR master":
        CXR_MASTER_PATH,

    "3B.6.34 all temporal pairs":
        TEMPORAL_PAIRS_PATH,

    "3B.6.34 nearest genomic timepoint per CXR":
        NEAREST_PAIRS_PATH,

    "3B.6.34 unique genomic timepoints":
        UNIQUE_GENOMIC_TIMEPOINTS_PATH,

    "3B.6.34 multiple genomic timepoint conditions":
        MULTIPLE_GENOMIC_PATH,

    "3B.6.34 temporal direction summary":
        TEMPORAL_DIRECTION_SUMMARY_PATH,

    "3B.6.34 temporal window summary":
        TEMPORAL_WINDOW_SUMMARY_PATH,
}


for label, path in required_inputs.items():

    assert path.exists(), (
        "\nREQUIRED ARTIFACT NOT FOUND\n"
        f"{label}\n"
        f"{path}"
    )

    print(
        f"{label:<55}: FOUND"
    )


print("Input artifact validation: PASS")


# ===================================================================================================
# 4. LOAD
# ===================================================================================================

print("\n2. LOADING VERIFIED ARTIFACTS")
print("-" * 100)

cxr = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

temporal_pairs = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)

nearest_pairs = pd.read_csv(
    NEAREST_PAIRS_PATH,
    low_memory=False
)

unique_genomic_timepoints = pd.read_csv(
    UNIQUE_GENOMIC_TIMEPOINTS_PATH,
    low_memory=False
)

multiple_genomic = pd.read_csv(
    MULTIPLE_GENOMIC_PATH,
    low_memory=False
)

temporal_direction_summary_source = pd.read_csv(
    TEMPORAL_DIRECTION_SUMMARY_PATH,
    low_memory=False
)

temporal_window_summary_source = pd.read_csv(
    TEMPORAL_WINDOW_SUMMARY_PATH,
    low_memory=False
)


print(
    "CXR master records:",
    len(cxr)
)

print(
    "All temporal-pair records:",
    len(temporal_pairs)
)

print(
    "Nearest temporal records:",
    len(nearest_pairs)
)

print(
    "Unique genomic timepoints:",
    len(unique_genomic_timepoints)
)

print(
    "Multiple-genomic conditions:",
    len(multiple_genomic)
)


# ===================================================================================================
# 5. SCHEMA VALIDATION
# ===================================================================================================

print("\n3. SCHEMA VALIDATION")
print("-" * 100)


CXR_REQUIRED = [
    "condition_id",
    "series_instance_content_url",
]


TEMPORAL_REQUIRED = [
    "condition_id",
    "series_instance_content_url",
    "imaging_date_num",
    "specimen_id",
    "specimen_collection_date_num",
    "signed_temporal_delta_days",
    "absolute_temporal_distance_days",
    "temporal_direction",
    "absolute_distance_category",
]


NEAREST_REQUIRED = TEMPORAL_REQUIRED.copy()


TIMEPOINT_REQUIRED = [
    "condition_id",
    "specimen_id",
    "specimen_collection_date_num",
]


MULTIPLE_REQUIRED = [
    "condition_id",
    "genomic_timepoint_count",
]


def validate_schema(
    df,
    required,
    name
):

    missing = [
        c
        for c in required
        if c not in df.columns
    ]

    assert not missing, (
        f"\n{name} schema failure.\n"
        f"Missing columns: {missing}"
    )

    print(
        f"{name:<45}: PASS"
    )


validate_schema(
    cxr,
    CXR_REQUIRED,
    "CXR master"
)

validate_schema(
    temporal_pairs,
    TEMPORAL_REQUIRED,
    "All temporal pairs"
)

validate_schema(
    nearest_pairs,
    NEAREST_REQUIRED,
    "Nearest temporal pairs"
)

validate_schema(
    unique_genomic_timepoints,
    TIMEPOINT_REQUIRED,
    "Unique genomic timepoints"
)

validate_schema(
    multiple_genomic,
    MULTIPLE_REQUIRED,
    "Multiple genomic conditions"
)


# ===================================================================================================
# 6. NORMALIZATION
# ===================================================================================================

print("\n4. NORMALIZATION")
print("-" * 100)


for df in [
    cxr,
    temporal_pairs,
    nearest_pairs,
    unique_genomic_timepoints,
    multiple_genomic,
]:

    df["condition_id"] = (
        df["condition_id"]
        .astype(str)
        .str.strip()
    )


for df in [
    temporal_pairs,
    nearest_pairs,
]:

    numeric_columns = [
        "imaging_date_num",
        "specimen_collection_date_num",
        "signed_temporal_delta_days",
        "absolute_temporal_distance_days",
    ]

    for column in numeric_columns:

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )


unique_genomic_timepoints[
    "specimen_collection_date_num"
] = pd.to_numeric(
    unique_genomic_timepoints[
        "specimen_collection_date_num"
    ],
    errors="coerce"
)


multiple_genomic[
    "genomic_timepoint_count"
] = pd.to_numeric(
    multiple_genomic[
        "genomic_timepoint_count"
    ],
    errors="coerce"
)


print("Normalization: PASS")


# ===================================================================================================
# 7. CORE POPULATION
# ===================================================================================================

print("\n5. CORE POPULATION VALIDATION")
print("-" * 100)


assert len(cxr) == 3859

assert (
    cxr[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    cxr[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    temporal_pairs[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    nearest_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    nearest_pairs[
        "condition_id"
    ].nunique()
    == 3081
)


print(
    "CXR records                         :",
    len(cxr)
)

print(
    "Unique CXR records                  :",
    cxr[
        "series_instance_content_url"
    ].nunique()
)

print(
    "Unique conditions                   :",
    cxr[
        "condition_id"
    ].nunique()
)

print(
    "Temporal-pair CXR coverage          :",
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
)

print(
    "Nearest-pair CXR coverage           :",
    nearest_pairs[
        "series_instance_content_url"
    ].nunique()
)

print("Core population validation: PASS")


# ===================================================================================================
# 8. TEMPORAL DATE / DELTA INTEGRITY
# ===================================================================================================

print("\n6. TEMPORAL DATE AND DELTA INTEGRITY")
print("-" * 100)


assert (
    temporal_pairs[
        "imaging_date_num"
    ].notna()
    .all()
)

assert (
    temporal_pairs[
        "specimen_collection_date_num"
    ].notna()
    .all()
)

assert (
    temporal_pairs[
        "signed_temporal_delta_days"
    ].notna()
    .all()
)

assert (
    temporal_pairs[
        "absolute_temporal_distance_days"
    ].notna()
    .all()
)


calculated_signed_delta = (
    temporal_pairs[
        "specimen_collection_date_num"
    ]
    -
    temporal_pairs[
        "imaging_date_num"
    ]
)


assert np.isclose(
    calculated_signed_delta.to_numpy(),
    temporal_pairs[
        "signed_temporal_delta_days"
    ].to_numpy()
).all()


calculated_absolute_delta = (
    temporal_pairs[
        "signed_temporal_delta_days"
    ]
    .abs()
)


assert np.isclose(
    calculated_absolute_delta.to_numpy(),
    temporal_pairs[
        "absolute_temporal_distance_days"
    ].to_numpy()
).all()


print(
    "Signed delta = genomic date - CXR date : VALID"
)

print(
    "Absolute delta = |signed delta|        : VALID"
)

print("Temporal delta integrity: PASS")


# ===================================================================================================
# 9. CORRECT TEMPORAL-DIRECTION RECONSTRUCTION
# ===================================================================================================

print("\n7. TEMPORAL DIRECTION RECONSTRUCTION")
print("-" * 100)


def derive_source_direction(delta):

    if delta < 0:
        return "GENOMICS_BEFORE_CXR"

    elif delta == 0:
        return "SAME_DAY"

    elif delta > 0:
        return "GENOMICS_AFTER_CXR"

    else:
        return "INVALID"


temporal_pairs[
    "derived_temporal_direction"
] = (
    temporal_pairs[
        "signed_temporal_delta_days"
    ]
    .apply(
        derive_source_direction
    )
)


# Validate against the EXACT vocabulary used by Step 3B.6.34.

VALID_DIRECTIONS = {
    "GENOMICS_BEFORE_CXR",
    "SAME_DAY",
    "GENOMICS_AFTER_CXR",
}


assert (
    temporal_pairs[
        "temporal_direction"
    ]
    .isin(
        VALID_DIRECTIONS
    )
    .all()
)


# Now compare.

direction_match = (
    temporal_pairs[
        "derived_temporal_direction"
    ]
    ==
    temporal_pairs[
        "temporal_direction"
    ]
)


assert direction_match.all()


print(
    "Stored temporal direction values:"
)

print(
    temporal_pairs[
        "temporal_direction"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nDerived direction values:"
)

print(
    temporal_pairs[
        "derived_temporal_direction"
    ]
    .value_counts()
    .to_string()
)

print(
    "\nDirection agreement:",
    int(direction_match.sum()),
    "/",
    len(direction_match)
)

print(
    "Direction agreement percentage:",
    round(
        direction_match.mean() * 100,
        4
    ),
    "%"
)

print("Temporal direction reconstruction: PASS")


# ===================================================================================================
# 10. CONDITION-LEVEL TEMPORAL STRUCTURE
# ===================================================================================================

print("\n8. CONDITION-LEVEL TEMPORAL STRUCTURE")
print("-" * 100)


condition_temporal = (
    temporal_pairs
    .groupby(
        "condition_id"
    )
    .agg(

        genomic_before_cxr=(
            "signed_temporal_delta_days",
            lambda x:
            bool(
                (x < 0).any()
            )
        ),

        same_day=(
            "signed_temporal_delta_days",
            lambda x:
            bool(
                (x == 0).any()
            )
        ),

        genomic_after_cxr=(
            "signed_temporal_delta_days",
            lambda x:
            bool(
                (x > 0).any()
            )
        ),

        minimum_signed_delta_days=(
            "signed_temporal_delta_days",
            "min"
        ),

        maximum_signed_delta_days=(
            "signed_temporal_delta_days",
            "max"
        ),

        minimum_absolute_distance_days=(
            "absolute_temporal_distance_days",
            "min"
        ),

        maximum_absolute_distance_days=(
            "absolute_temporal_distance_days",
            "max"
        ),

        temporal_pair_count=(
            "condition_id",
            "size"
        ),

    )
    .reset_index()
)


assert len(condition_temporal) == 3081

assert (
    condition_temporal[
        "condition_id"
    ].nunique()
    == 3081
)


condition_temporal[
    "number_of_temporal_directions"
] = (
    condition_temporal[
        [
            "genomic_before_cxr",
            "same_day",
            "genomic_after_cxr",
        ]
    ]
    .astype(int)
    .sum(axis=1)
)


condition_temporal[
    "temporal_category"
] = np.select(
    [

        (
            condition_temporal[
                "genomic_before_cxr"
            ]
            &
            ~condition_temporal[
                "same_day"
            ]
            &
            ~condition_temporal[
                "genomic_after_cxr"
            ]
        ),

        (
            ~condition_temporal[
                "genomic_before_cxr"
            ]
            &
            condition_temporal[
                "same_day"
            ]
            &
            ~condition_temporal[
                "genomic_after_cxr"
            ]
        ),

        (
            ~condition_temporal[
                "genomic_before_cxr"
            ]
            &
            ~condition_temporal[
                "same_day"
            ]
            &
            condition_temporal[
                "genomic_after_cxr"
            ]
        ),

        (
            condition_temporal[
                "genomic_before_cxr"
            ]
            &
            condition_temporal[
                "same_day"
            ]
            &
            ~condition_temporal[
                "genomic_after_cxr"
            ]
        ),

        (
            condition_temporal[
                "genomic_before_cxr"
            ]
            &
            ~condition_temporal[
                "same_day"
            ]
            &
            condition_temporal[
                "genomic_after_cxr"
            ]
        ),

        (
            ~condition_temporal[
                "genomic_before_cxr"
            ]
            &
            condition_temporal[
                "same_day"
            ]
            &
            condition_temporal[
                "genomic_after_cxr"
            ]
        ),

        (
            condition_temporal[
                "genomic_before_cxr"
            ]
            &
            condition_temporal[
                "same_day"
            ]
            &
            condition_temporal[
                "genomic_after_cxr"
            ]
        ),
    ],

    [
        "BEFORE_ONLY",
        "SAME_DAY_ONLY",
        "AFTER_ONLY",
        "BEFORE_PLUS_SAME_DAY",
        "BEFORE_PLUS_AFTER",
        "SAME_DAY_PLUS_AFTER",
        "BEFORE_SAME_DAY_AFTER",
    ],

    default="INVALID"
)


assert (
    ~condition_temporal[
        "temporal_category"
    ].eq("INVALID")
).all()


print(
    condition_temporal[
        "temporal_category"
    ]
    .value_counts()
    .to_string()
)

print("Condition-level temporal structure: PASS")


# ===================================================================================================
# 11. NEAREST GENOMIC TIMEPOINT — TIE-PRESERVING
# ===================================================================================================

print("\n9. NEAREST GENOMIC TIMEPOINT — TIE-PRESERVING AUDIT")
print("-" * 100)


# IMPORTANT:
# The artifact contains 3,871 rows for 3,859 CXR URLs.
# Therefore some CXR records have multiple equally-nearest genomic timepoints.
#
# We intentionally DO NOT collapse these ties.

nearest_rows_per_cxr = (
    nearest_pairs
    .groupby(
        "series_instance_content_url"
    )
    .size()
)


assert (
    nearest_rows_per_cxr.index.nunique()
    == 3859
)


nearest_tied_urls = (
    nearest_rows_per_cxr[
        nearest_rows_per_cxr > 1
    ]
)


print(
    "Nearest artifact rows:",
    len(nearest_pairs)
)

print(
    "Unique CXR URLs:",
    nearest_rows_per_cxr.index.nunique()
)

print(
    "CXR URLs with exactly one nearest timepoint:",
    int(
        (
            nearest_rows_per_cxr == 1
        ).sum()
    )
)

print(
    "CXR URLs with nearest-timepoint ties:",
    len(nearest_tied_urls)
)

print(
    "Maximum nearest rows for one CXR:",
    int(
        nearest_rows_per_cxr.max()
    )
)


# ===================================================================================================
# 12. VALIDATE NEAREST PROPERTY
# ===================================================================================================

print("\n10. NEAREST-TIMEPOINT PROPERTY VALIDATION")
print("-" * 100)


all_pair_min_distance = (
    temporal_pairs
    .groupby(
        "series_instance_content_url"
    )[
        "absolute_temporal_distance_days"
    ]
    .min()
    .rename(
        "minimum_distance_from_all_pairs"
    )
)


nearest_distance = (
    nearest_pairs[
        [
            "series_instance_content_url",
            "absolute_temporal_distance_days",
        ]
    ]
    .copy()
)


nearest_distance = nearest_distance.merge(
    all_pair_min_distance,
    left_on="series_instance_content_url",
    right_index=True,
    how="left"
)


assert (
    nearest_distance[
        "minimum_distance_from_all_pairs"
    ].notna()
    .all()
)


assert np.isclose(
    nearest_distance[
        "absolute_temporal_distance_days"
    ].to_numpy(),

    nearest_distance[
        "minimum_distance_from_all_pairs"
    ].to_numpy()
).all()


print(
    "Every nearest row equals the minimum absolute temporal distance:",
    "PASS"
)


# ===================================================================================================
# 13. NEAREST DIRECTION — DO NOT RESOLVE TIES
# ===================================================================================================

print("\n11. NEAREST TEMPORAL DIRECTION")
print("-" * 100)


nearest_pairs[
    "derived_temporal_direction"
] = (
    nearest_pairs[
        "signed_temporal_delta_days"
    ]
    .apply(
        derive_source_direction
    )
)


assert (
    nearest_pairs[
        "derived_temporal_direction"
    ]
    ==
    nearest_pairs[
        "temporal_direction"
    ]
).all()


print(
    nearest_pairs[
        "temporal_direction"
    ]
    .value_counts()
    .to_string()
)


# Determine whether a nearest-distance tie has different directions.

nearest_direction_summary = (
    nearest_pairs
    .groupby(
        "series_instance_content_url"
    )
    .agg(
        nearest_row_count=(
            "series_instance_content_url",
            "size"
        ),

        nearest_direction_count=(
            "temporal_direction",
            "nunique"
        ),

        nearest_min_distance=(
            "absolute_temporal_distance_days",
            "min"
        ),

        nearest_max_distance=(
            "absolute_temporal_distance_days",
            "max"
        ),
    )
    .reset_index()
)


nearest_direction_summary[
    "direction_tie"
] = (
    nearest_direction_summary[
        "nearest_direction_count"
    ]
    > 1
)


print(
    "\nCXR URLs with multiple nearest rows:",
    int(
        (
            nearest_direction_summary[
                "nearest_row_count"
            ]
            > 1
        )
        .sum()
    )
)

print(
    "CXR URLs whose nearest rows contain multiple directions:",
    int(
        nearest_direction_summary[
            "direction_tie"
        ].sum()
    )
)


# ===================================================================================================
# 14. CONDITION-LEVEL NEAREST DIRECTION
# ===================================================================================================

print("\n12. CONDITION-LEVEL NEAREST DIRECTION")
print("-" * 100)


nearest_condition = (
    nearest_pairs
    .groupby(
        "condition_id"
    )
    .agg(

        nearest_before=(
            "signed_temporal_delta_days",
            lambda x:
            bool(
                (x < 0).any()
            )
        ),

        nearest_same_day=(
            "signed_temporal_delta_days",
            lambda x:
            bool(
                (x == 0).any()
            )
        ),

        nearest_after=(
            "signed_temporal_delta_days",
            lambda x:
            bool(
                (x > 0).any()
            )
        ),

        nearest_distance_days=(
            "absolute_temporal_distance_days",
            "min"
        ),

        nearest_row_count=(
            "condition_id",
            "size"
        ),

    )
    .reset_index()
)


assert len(nearest_condition) == 3081


nearest_condition[
    "nearest_direction_count"
] = (
    nearest_condition[
        [
            "nearest_before",
            "nearest_same_day",
            "nearest_after",
        ]
    ]
    .astype(int)
    .sum(axis=1)
)


nearest_condition[
    "nearest_direction_class"
] = np.select(
    [
        nearest_condition[
            "nearest_direction_count"
        ] == 1,

        nearest_condition[
            "nearest_direction_count"
        ] > 1,
    ],

    [
        "SINGLE_NEAREST_DIRECTION",
        "NEAREST_DIRECTION_TIE",
    ],

    default="INVALID"
)


assert (
    ~nearest_condition[
        "nearest_direction_class"
    ].eq("INVALID")
).all()


print(
    nearest_condition[
        "nearest_direction_class"
    ]
    .value_counts()
    .to_string()
)


# ===================================================================================================
# 15. DESCRIPTIVE TEMPORAL WINDOWS
# ===================================================================================================

print("\n13. DESCRIPTIVE TEMPORAL WINDOWS")
print("-" * 100)


WINDOWS = [
    0,
    1,
    7,
    30,
    90,
    180,
    365,
]


window_rows = []


for window in WINDOWS:

    if window == 0:

        mask = (
            temporal_pairs[
                "absolute_temporal_distance_days"
            ]
            == 0
        )

        label = "SAME_DAY"

    else:

        mask = (
            temporal_pairs[
                "absolute_temporal_distance_days"
            ]
            <= window
        )

        label = (
            f"WITHIN_{window}_DAYS"
        )


    pair_count = int(
        mask.sum()
    )


    unique_cxr_count = int(
        temporal_pairs.loc[
            mask,
            "series_instance_content_url"
        ]
        .nunique()
    )


    unique_condition_count = int(
        temporal_pairs.loc[
            mask,
            "condition_id"
        ]
        .nunique()
    )


    window_rows.append(
        {
            "window":
                label,

            "window_days":
                window,

            "temporal_pair_records":
                pair_count,

            "unique_cxr_records":
                unique_cxr_count,

            "unique_conditions":
                unique_condition_count,

            "percentage_of_3859_cxr_records":
                (
                    unique_cxr_count
                    /
                    3859
                    *
                    100
                ),

            "percentage_of_3081_conditions":
                (
                    unique_condition_count
                    /
                    3081
                    *
                    100
                ),
        }
    )


temporal_window_analysis = pd.DataFrame(
    window_rows
)


print(
    temporal_window_analysis.to_string(
        index=False
    )
)


print(
    "\nTemporal windows are DESCRIPTIVE ONLY."
)

print(
    "No window has been selected as an eligibility rule."
)


# ===================================================================================================
# 16. DIRECTION × WINDOW
# ===================================================================================================

print("\n14. DIRECTION × TEMPORAL WINDOW")
print("-" * 100)


direction_window_rows = []


for direction in [
    "GENOMICS_BEFORE_CXR",
    "SAME_DAY",
    "GENOMICS_AFTER_CXR",
]:

    direction_mask = (
        temporal_pairs[
            "temporal_direction"
        ]
        ==
        direction
    )


    for window in WINDOWS:

        if window == 0:

            distance_mask = (
                temporal_pairs[
                    "absolute_temporal_distance_days"
                ]
                == 0
            )

        else:

            distance_mask = (
                temporal_pairs[
                    "absolute_temporal_distance_days"
                ]
                <= window
            )


        mask = (
            direction_mask
            &
            distance_mask
        )


        direction_window_rows.append(
            {
                "temporal_direction":
                    direction,

                "window_days":
                    window,

                "pair_count":
                    int(
                        mask.sum()
                    ),

                "unique_cxr_records":
                    int(
                        temporal_pairs.loc[
                            mask,
                            "series_instance_content_url"
                        ]
                        .nunique()
                    ),

                "unique_conditions":
                    int(
                        temporal_pairs.loc[
                            mask,
                            "condition_id"
                        ]
                        .nunique()
                    ),
            }
        )


direction_window_analysis = pd.DataFrame(
    direction_window_rows
)


print(
    direction_window_analysis.to_string(
        index=False
    )
)


# ===================================================================================================
# 17. NEAREST DISTANCE STATISTICS
# ===================================================================================================

print("\n15. NEAREST DISTANCE STATISTICS")
print("-" * 100)


nearest_distance_series = (
    nearest_pairs[
        "absolute_temporal_distance_days"
    ]
)


nearest_distance_statistics = pd.DataFrame(
    [
        {
            "statistic": "count",
            "value":
                int(
                    nearest_distance_series.count()
                ),
        },

        {
            "statistic": "minimum",
            "value":
                float(
                    nearest_distance_series.min()
                ),
        },

        {
            "statistic": "mean",
            "value":
                float(
                    nearest_distance_series.mean()
                ),
        },

        {
            "statistic": "median",
            "value":
                float(
                    nearest_distance_series.median()
                ),
        },

        {
            "statistic": "p75",
            "value":
                float(
                    nearest_distance_series.quantile(
                        0.75
                    )
                ),
        },

        {
            "statistic": "p90",
            "value":
                float(
                    nearest_distance_series.quantile(
                        0.90
                    )
                ),

        },

        {
            "statistic": "p95",
            "value":
                float(
                    nearest_distance_series.quantile(
                        0.95
                    )
                ),

        },

        {
            "statistic": "p99",
            "value":
                float(
                    nearest_distance_series.quantile(
                        0.99
                    )
                ),

        },

        {
            "statistic": "maximum",
            "value":
                float(
                    nearest_distance_series.max()
                ),
        },
    ]
)


print(
    nearest_distance_statistics.to_string(
        index=False
    )
)


# ===================================================================================================
# 18. MULTIPLE GENOMIC TIMEPOINTS
# ===================================================================================================

print("\n16. MULTIPLE GENOMIC TIMEPOINT VALIDATION")
print("-" * 100)


assert (
    multiple_genomic[
        "condition_id"
    ].nunique()
    ==
    len(multiple_genomic)
)


assert (
    multiple_genomic[
        "genomic_timepoint_count"
    ]
    .gt(1)
    .all()
)


print(
    "Conditions with >1 genomic timepoint:",
    len(multiple_genomic)
)

print(
    "\nGenomic timepoint-count distribution:"
)

print(
    multiple_genomic[
        "genomic_timepoint_count"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)


# This is the already verified Step 3B.6.34 longitudinal artifact.

assert len(multiple_genomic) == 152


print(
    "\nMultiple genomic timepoint validation: PASS"
)


# ===================================================================================================
# 19. MULTIPLE GENOMIC × TEMPORAL STRUCTURE
# ===================================================================================================

print("\n17. MULTIPLE-GENOMIC × TEMPORAL STRUCTURE")
print("-" * 100)


multiple_genomic_ids = set(
    multiple_genomic[
        "condition_id"
    ]
)


condition_temporal[
    "multiple_genomic_timepoints"
] = (
    condition_temporal[
        "condition_id"
    ]
    .isin(
        multiple_genomic_ids
    )
)


multiple_genomic_temporal = (
    condition_temporal[
        condition_temporal[
            "multiple_genomic_timepoints"
        ]
    ]
    .copy()
)


assert (
    len(multiple_genomic_temporal)
    == 152
)


print(
    "Multiple-genomic conditions:",
    len(multiple_genomic_temporal)
)

print(
    "\nTemporal categories:"
)

print(
    multiple_genomic_temporal[
        "temporal_category"
    ]
    .value_counts()
    .to_string()
)


print(
    "\nNo genomic specimen has been selected."
)


# ===================================================================================================
# 20. MIXED TEMPORAL DIRECTIONS
# ===================================================================================================

print("\n18. MIXED TEMPORAL-DIRECTION CONDITIONS")
print("-" * 100)


mixed_temporal_conditions = (
    condition_temporal[
        condition_temporal[
            "number_of_temporal_directions"
        ]
        > 1
    ]
    .copy()
)


print(
    "Conditions with multiple temporal directions:",
    len(mixed_temporal_conditions)
)

print(
    "Percentage of 3,081 conditions:",
    round(
        len(mixed_temporal_conditions)
        /
        3081
        *
        100,
        4
    ),
    "%"
)


# ===================================================================================================
# 21. CXR MULTIPLICITY × TEMPORAL STRUCTURE
# ===================================================================================================

print("\n19. CXR MULTIPLICITY × TEMPORAL STRUCTURE")
print("-" * 100)


cxr_counts = (
    cxr
    .groupby(
        "condition_id"
    )
    .size()
)


condition_temporal[
    "cxr_record_count"
] = (
    condition_temporal[
        "condition_id"
    ]
    .map(
        cxr_counts
    )
)


assert (
    condition_temporal[
        "cxr_record_count"
    ].notna()
    .all()
)


condition_temporal[
    "multiple_cxr"
] = (
    condition_temporal[
        "cxr_record_count"
    ]
    > 1
)


print(
    "Single-CXR conditions:",
    int(
        (
            ~condition_temporal[
                "multiple_cxr"
            ]
        )
        .sum()
    )
)

print(
    "Multiple-CXR conditions:",
    int(
        condition_temporal[
            "multiple_cxr"
        ]
        .sum()
    )
)


# ===================================================================================================
# 22. TARGET-INDEPENDENT TEMPORAL STRATA
# ===================================================================================================

print("\n20. TARGET-INDEPENDENT TEMPORAL STRATA")
print("-" * 100)


temporal_strata = (
    condition_temporal
    .groupby(
        [
            "multiple_cxr",
            "multiple_genomic_timepoints",
            "temporal_category",
        ]
    )
    .size()
    .reset_index(
        name="condition_count"
    )
)


temporal_strata[
    "percentage_of_3081"
] = (
    temporal_strata[
        "condition_count"
    ]
    /
    3081
    *
    100
)


print(
    temporal_strata.to_string(
        index=False
    )
)


# ===================================================================================================
# 23. PREDICTION-DIRECTION OPTIONS
# ===================================================================================================

print("\n21. PREDICTION-DIRECTION OPTIONS")
print("-" * 100)


prediction_options = pd.DataFrame(
    [
        {
            "option_id":
                "PRED-01",

            "option":
                "Genomic specimen collected before CXR",

            "direction":
                "GENOMICS_BEFORE_CXR",

            "status":
                "NOT_SELECTED",
        },

        {
            "option_id":
                "PRED-02",

            "option":
                "Genomic specimen collected on same day as CXR",

            "direction":
                "SAME_DAY",

            "status":
                "NOT_SELECTED",
        },

        {
            "option_id":
                "PRED-03",

            "option":
                "Nearest genomic specimen regardless of direction",

            "direction":
                "NEAREST",

            "status":
                "NOT_SELECTED",
        },

        {
            "option_id":
                "PRED-04",

            "option":
                "Prespecified temporal window",

            "direction":
                "WINDOW",

            "status":
                "NOT_SELECTED",
        },

        {
            "option_id":
                "PRED-05",

            "option":
                "Protocol-defined genomic timepoint",

            "direction":
                "PROTOCOL_DEFINED",

            "status":
                "NOT_SELECTED",
        },
    ]
)


assert (
    prediction_options[
        "status"
    ]
    .eq("NOT_SELECTED")
    .all()
)


print(
    prediction_options.to_string(
        index=False
    )
)


# ===================================================================================================
# 24. TEMPORAL DECISION GATE
# ===================================================================================================

print("\n22. TEMPORAL DECISION GATE")
print("-" * 100)


temporal_decision_gate = pd.DataFrame(
    [
        {
            "decision":
                "CXR temporal reference",

            "current_definition":
                "imaging_date",

            "status":
                "LOCKED",
        },

        {
            "decision":
                "Genomic temporal reference",

            "current_definition":
                "specimen_collection_date",

            "status":
                "LOCKED",
        },

        {
            "decision":
                "Signed temporal direction",

            "current_definition":
                "genomic_date - CXR_date",

            "status":
                "LOCKED",
        },

        {
            "decision":
                "Prediction direction",

            "current_definition":
                "NOT_SELECTED",

            "status":
                "PENDING",
        },

        {
            "decision":
                "Temporal cutoff",

            "current_definition":
                "NOT_SELECTED",

            "status":
                "PENDING",
        },

        {
            "decision":
                "Representative genomic specimen",

            "current_definition":
                "NOT_SELECTED",

            "status":
                "PENDING",
        },

        {
            "decision":
                "Final temporal eligibility rule",

            "current_definition":
                "NOT_SELECTED",

            "status":
                "PENDING",
        },
    ]
)


print(
    temporal_decision_gate.to_string(
        index=False
    )
)


# ===================================================================================================
# 25. SAVE OUTPUTS
# ===================================================================================================

print("\n23. SAVING OUTPUTS")
print("-" * 100)


OUTPUT_CONDITION_TEMPORAL = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Condition_Temporal_Structure.csv"
)

OUTPUT_NEAREST_CONDITION = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Nearest_Condition_Direction.csv"
)

OUTPUT_NEAREST_DIRECTION_SUMMARY = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Nearest_Direction_Summary.csv"
)

OUTPUT_WINDOW = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Temporal_Window_Analysis.csv"
)

OUTPUT_DIRECTION_WINDOW = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Direction_Window_Analysis.csv"
)

OUTPUT_DISTANCE = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Nearest_Distance_Statistics.csv"
)

OUTPUT_MULTIPLE = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Multiple_Genomic_Temporal_Analysis.csv"
)

OUTPUT_MIXED = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Mixed_Temporal_Direction_Conditions.csv"
)

OUTPUT_STRATA = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Temporal_Strata.csv"
)

OUTPUT_OPTIONS = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Prediction_Direction_Options.csv"
)

OUTPUT_GATE = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_42_Temporal_Decision_Gate.csv"
)


condition_temporal.to_csv(
    OUTPUT_CONDITION_TEMPORAL,
    index=False
)

nearest_condition.to_csv(
    OUTPUT_NEAREST_CONDITION,
    index=False
)

nearest_direction_summary.to_csv(
    OUTPUT_NEAREST_DIRECTION_SUMMARY,
    index=False
)

temporal_window_analysis.to_csv(
    OUTPUT_WINDOW,
    index=False
)

direction_window_analysis.to_csv(
    OUTPUT_DIRECTION_WINDOW,
    index=False
)

nearest_distance_statistics.to_csv(
    OUTPUT_DISTANCE,
    index=False
)

multiple_genomic_temporal.to_csv(
    OUTPUT_MULTIPLE,
    index=False
)

mixed_temporal_conditions.to_csv(
    OUTPUT_MIXED,
    index=False
)

temporal_strata.to_csv(
    OUTPUT_STRATA,
    index=False
)

prediction_options.to_csv(
    OUTPUT_OPTIONS,
    index=False
)

temporal_decision_gate.to_csv(
    OUTPUT_GATE,
    index=False
)


# ===================================================================================================
# 26. OUTPUT VALIDATION
# ===================================================================================================

print("\n24. OUTPUT VALIDATION")
print("-" * 100)


output_paths = [
    OUTPUT_CONDITION_TEMPORAL,
    OUTPUT_NEAREST_CONDITION,
    OUTPUT_NEAREST_DIRECTION_SUMMARY,
    OUTPUT_WINDOW,
    OUTPUT_DIRECTION_WINDOW,
    OUTPUT_DISTANCE,
    OUTPUT_MULTIPLE,
    OUTPUT_MIXED,
    OUTPUT_STRATA,
    OUTPUT_OPTIONS,
    OUTPUT_GATE,
]


for path in output_paths:

    assert path.exists(), (
        f"Output missing:\n{path}"
    )

    assert (
        path.stat().st_size > 0
    ), (
        f"Output is empty:\n{path}"
    )

    print(
        f"{path.name}: PASS"
    )


# ===================================================================================================
# 27. FINAL INTEGRITY CHECKS
# ===================================================================================================

print("\n25. FINAL INTEGRITY CHECKS")
print("-" * 100)


assert len(cxr) == 3859

assert (
    cxr[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    temporal_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    temporal_pairs[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    nearest_pairs[
        "series_instance_content_url"
    ].nunique()
    == 3859
)

assert (
    nearest_pairs[
        "condition_id"
    ].nunique()
    == 3081
)

assert (
    len(condition_temporal)
    == 3081
)

assert (
    len(multiple_genomic)
    == 152
)

assert (
    temporal_pairs[
        "temporal_direction"
    ]
    .isin(
        VALID_DIRECTIONS
    )
    .all()
)

assert (
    direction_match.all()
)

assert (
    prediction_options[
        "status"
    ]
    .eq("NOT_SELECTED")
    .all()
)

assert (
    temporal_decision_gate[
        "status"
    ]
    .isin(
        [
            "LOCKED",
            "PENDING",
        ]
    )
    .all()
)


print(
    "3,859 CXR records                         : PASS"
)

print(
    "3,081 multimodal conditions               : PASS"
)

print(
    "4,146 temporal-pair records               : PASS"
)

print(
    "3,871 nearest rows preserved              : PASS"
)

print(
    "Nearest-timepoint ties preserved          : PASS"
)

print(
    "Signed temporal delta                     : PASS"
)

print(
    "Temporal direction                        : PASS"
)

print(
    "Condition-level temporal structure        : PASS"
)

print(
    "152 multiple-genomic conditions            : PASS"
)

print(
    "Descriptive temporal windows              : PASS"
)

print(
    "Prediction-direction options              : PASS"
)

print(
    "No prediction direction selected          : PASS"
)

print(
    "No temporal cutoff selected               : PASS"
)

print(
    "No representative genomic specimen        : PASS"
)

print(
    "No final cohort selected                  : PASS"
)


# ===================================================================================================
# 28. SOURCE PROTECTION
# ===================================================================================================

print("\n26. SOURCE PROTECTION")
print("-" * 100)

print(
    "Raw DICOM modification                : NO"
)

print(
    "CXR source metadata modification      : NO"
)

print(
    "Genomics source modification           : NO"
)

print(
    "Human-review artifact modification    : NO"
)

print(
    "Image preprocessing                   : NO"
)

print(
    "Model training                        : NO"
)

print(
    "Target selection                      : NO"
)

print(
    "Temporal cutoff selection             : NO"
)

print(
    "Representative CXR selection         : NO"
)

print(
    "Representative genomic selection      : NO"
)

print(
    "Final cohort selection                : NO"
)


# ===================================================================================================
# 29. FINAL STATUS
# ===================================================================================================

print("\n" + "=" * 100)
print("STEP 3B.6.42 — FINAL STATUS")
print("=" * 100)

print(
    "Temporal pair reconciliation             : PASS"
)

print(
    "Signed temporal delta                    : PASS"
)

print(
    "Temporal direction                       : PASS"
)

print(
    "Condition temporal structure             : PASS"
)

print(
    "Nearest genomic timepoint audit          : PASS"
)

print(
    "Nearest-timepoint ties                   : PRESERVED"
)

print(
    "Temporal window analysis                 : PASS"
)

print(
    "Direction × window analysis              : PASS"
)

print(
    "Multiple genomic timepoint analysis      : PASS"
)

print(
    "Mixed temporal-direction analysis        : PASS"
)

print(
    "CXR multiplicity analysis                : PASS"
)

print(
    "Prediction-direction decision gate       : PENDING"
)

print(
    "Target decision                          : PENDING"
)

print(
    "Temporal rule                            : PENDING"
)

print(
    "Representative CXR rule                 : PENDING"
)

print(
    "Representative genomic rule             : PENDING"
)

print(
    "Final cohort                             : PENDING"
)

print(
    "Preprocessing                            : NOT PERFORMED"
)

print(
    "Model training                           : NOT PERFORMED"
)

print(
    "Source modification                      : NO"
)

print(
    "Output validation                        : PASS"
)

print("\n" + "=" * 100)
print("STEP 3B.6.42 COMPLETED")
print("=" * 100)
```

```python
# ================================================================
# TB PORTALS MARCH 2025
# FILE LOCATION DIAGNOSTIC
#
# PURPOSE:
# Find the actual locations of the files required for Step 3B.6.44.
#
# IMPORTANT:
# - Does NOT modify files
# - Does NOT copy files
# - Does NOT preprocess data
# - Does NOT modify DICOMs
# ================================================================

from pathlib import Path
import os

print("=" * 80)
print("TB PORTALS — FILE LOCATION DIAGNOSTIC")
print("=" * 80)


# ================================================================
# 1. CURRENT NOTEBOOK LOCATION
# ================================================================

print("\n1. CURRENT JUPYTER WORKING DIRECTORY")
print("-" * 80)

cwd = Path.cwd()

print("Current working directory:")
print(cwd)

print("\nExists:", cwd.exists())


# ================================================================
# 2. CHECK C:\TBP
# ================================================================

print("\n2. CHECKING C:\\TBP")
print("-" * 80)

tbp_path = Path(r"C:\TBP")

print("Path:", tbp_path)
print("Exists:", tbp_path.exists())
print("Is directory:", tbp_path.is_dir())


# ================================================================
# 3. FILES WE NEED
# ================================================================

target_files = [
    "TB_Portals_CXRs_March_2025.csv",

    "TB_Portals_Genomics_March_2025.csv",

    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv",

    "TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv",

    "TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv",
]


# ================================================================
# 4. SEARCH FUNCTION
# ================================================================

def search_for_filename(root, filename, max_results=20):

    root = Path(root)

    if not root.exists() or not root.is_dir():
        return []

    results = []

    try:
        for path in root.rglob(filename):

            if path.is_file():
                results.append(path)

                if len(results) >= max_results:
                    break

    except PermissionError:
        pass

    except Exception as e:
        print(
            f"Search warning for {root}: {e}"
        )

    return results


# ================================================================
# 5. SEARCH LOCATIONS
# ================================================================

search_roots = []

# Current directory
search_roots.append(cwd)

# Parent directories
current = cwd

for _ in range(5):

    if current.parent != current:
        current = current.parent
        search_roots.append(current)


# C:\TBP
if tbp_path.exists():
    search_roots.append(tbp_path)


# User home directory
home = Path.home()

if home.exists():
    search_roots.append(home)


# Remove duplicates while preserving order
unique_roots = []

seen_roots = set()

for root in search_roots:

    try:
        resolved = root.resolve()

        if resolved not in seen_roots:

            seen_roots.add(resolved)
            unique_roots.append(resolved)

    except Exception:
        pass


print("\nSearch roots:")
for root in unique_roots:
    print(" -", root)


# ================================================================
# 6. FIND ALL TARGET FILES
# ================================================================

print("\n3. SEARCHING FOR REQUIRED FILES")
print("-" * 80)

found_files = {}

for filename in target_files:

    print(f"\nSearching for:")
    print(filename)

    matches = []

    for root in unique_roots:

        results = search_for_filename(
            root,
            filename
        )

        for result in results:

            if result not in matches:
                matches.append(result)

    found_files[filename] = matches

    if matches:

        print(
            f"FOUND {len(matches)} location(s):"
        )

        for match in matches:
            print(
                "   ",
                match
            )

    else:

        print(
            "NOT FOUND"
        )


# ================================================================
# 7. SEARCH BY PARTIAL NAME IF EXACT NAME FAILED
# ================================================================

print("\n4. PARTIAL-NAME FALLBACK SEARCH")
print("-" * 80)

partial_patterns = [
    "TB_Portals_CXRs",
    "TB_Portals_Genomics",
    "3859_Multimodal_CXR_Master",
    "All_CXR_Genomic_Temporal_Pairs",
    "Unique_Genomic_Timepoints",
]


partial_results = {}

for pattern in partial_patterns:

    print(f"\nPattern: {pattern}")

    matches = []

    for root in unique_roots:

        if not root.exists():
            continue

        try:

            for path in root.rglob("*"):

                if not path.is_file():
                    continue

                if pattern.lower() in path.name.lower():

                    if path not in matches:
                        matches.append(path)

        except PermissionError:
            continue

        except Exception:
            continue

    partial_results[pattern] = matches

    if matches:

        for match in matches[:20]:
            print(
                "   ",
                match
            )

        if len(matches) > 20:
            print(
                f"   ... and {len(matches)-20} more"
            )

    else:

        print(
            "   NOT FOUND"
        )


# ================================================================
# 8. IMPORTANT DIRECTORY STRUCTURE CHECK
# ================================================================

print("\n5. CURRENT DIRECTORY STRUCTURE")
print("-" * 80)

print(
    "Files/folders directly under current working directory:"
)

try:

    for item in sorted(
        cwd.iterdir(),
        key=lambda p: (not p.is_dir(), p.name.lower())
    )[:100]:

        item_type = (
            "DIR "
            if item.is_dir()
            else "FILE"
        )

        print(
            f"{item_type}  {item.name}"
        )

except Exception as e:

    print(
        "Could not inspect current directory:",
        e
    )


# ================================================================
# 9. SUMMARY
# ================================================================

print("\n")
print("=" * 80)
print("DIAGNOSTIC SUMMARY")
print("=" * 80)

found_count = sum(
    1
    for filename, matches in found_files.items()
    if matches
)

print(
    f"Required files found: "
    f"{found_count}/{len(target_files)}"
)

for filename in target_files:

    matches = found_files[filename]

    if matches:

        print(
            f"FOUND : {filename}"
        )

        print(
            f"        {matches[0]}"
        )

    else:

        print(
            f"MISSING: {filename}"
        )


print("\n")
print("IMPORTANT:")
print(
    "Do NOT move, rename, copy, or modify any dataset files "
    "based on this diagnostic."
)

print(
    "Send me the COMPLETE OUTPUT of this cell."
)

print("=" * 80)
```

```python
# ================================================================
# TB PORTALS MARCH 2025
# STEP 3B.6.44
# TB-vs-NORMAL MULTIMODAL FEASIBILITY & COHORT DESIGN AUDIT
#
# ================================================================
# RESEARCH OBJECTIVE
# ================================================================
#
# Intended research:
#
#       CXR + GENOMICS
#             ↓
#       TB vs NORMAL
#
# Additional planned components:
#       - Lung segmentation
#       - Deep learning
#       - Explainable AI
#
# THIS STEP ONLY CHECKS WHETHER THE DATA CAN SUPPORT THIS DESIGN.
#
# NO:
#   - preprocessing
#   - lung segmentation
#   - model training
#   - explainable AI
#   - final cohort selection
#   - representative image selection
#   - representative genomic specimen selection
#   - raw DICOM modification
#   - source metadata modification
#
# CASE UNIT:
#       condition_id
#
# ================================================================


from pathlib import Path
import pandas as pd
import numpy as np
import json
import warnings

warnings.filterwarnings("ignore")


# ================================================================
# 1. EXACT VERIFIED PATHS
# ================================================================

BASE_DIR = Path(r"C:\TBP")
METADATA_DIR = BASE_DIR / "Metadata"

CXR_FILE = (
    METADATA_DIR /
    "TB_Portals_CXRs_March_2025.csv"
)

GENOMICS_FILE = (
    METADATA_DIR /
    "TB_Portals_Genomics_March_2025.csv"
)

CXR_MASTER_FILE = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03" /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

TEMPORAL_FILE = (
    METADATA_DIR /
    "Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation" /
    "TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

UNIQUE_TIMEPOINT_FILE = (
    METADATA_DIR /
    "Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation" /
    "TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)


# ================================================================
# 2. OUTPUT DIRECTORY
# ================================================================

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_44_TB_Normal_Multimodal_Feasibility"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# 3. START
# ================================================================

print("=" * 80)
print("STEP 3B.6.44 — TB-vs-Normal Multimodal Feasibility Audit")
print("=" * 80)


# ================================================================
# 4. VERIFY INPUT FILES
# ================================================================

print("\n1. EXACT INPUT FILE VALIDATION")
print("-" * 80)

input_files = {
    "CXR dataset": CXR_FILE,
    "Genomics dataset": GENOMICS_FILE,
    "3859 CXR master": CXR_MASTER_FILE,
    "Temporal pairs": TEMPORAL_FILE,
    "Unique genomic timepoints": UNIQUE_TIMEPOINT_FILE
}

for name, path in input_files.items():

    if not path.exists():

        raise FileNotFoundError(
            f"\nRequired file not found:\n{path}"
        )

    print(
        f"{name:<35}: FOUND"
    )

    print(
        f"{'':<37}{path}"
    )


# ================================================================
# 5. LOAD DATA
# ================================================================

print("\n2. LOADING DATA")
print("-" * 80)

cxr = pd.read_csv(
    CXR_FILE,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_FILE,
    low_memory=False
)

cxr_master = pd.read_csv(
    CXR_MASTER_FILE,
    low_memory=False
)

temporal = pd.read_csv(
    TEMPORAL_FILE,
    low_memory=False
)

unique_timepoints = pd.read_csv(
    UNIQUE_TIMEPOINT_FILE,
    low_memory=False
)

print(
    f"CXR records                 : {len(cxr):,}"
)

print(
    f"Genomic records             : {len(genomics):,}"
)

print(
    f"3859 CXR master records     : {len(cxr_master):,}"
)

print(
    f"Temporal pair records       : {len(temporal):,}"
)

print(
    f"Unique genomic timepoints   : {len(unique_timepoints):,}"
)


# ================================================================
# 6. SCHEMA VALIDATION
# ================================================================

print("\n3. SCHEMA VALIDATION")
print("-" * 80)

required_cxr = [
    "condition_id",
    "case_definition",
    "diagnosis_code",
    "type_of_resistance",
    "outcome",
    "imaging_date",
    "series_instance_content_url",
    "series_modality_cd"
]

required_genomics = [
    "condition_id"
]

required_master = [
    "condition_id",
    "series_instance_content_url"
]

required_temporal = [
    "condition_id",
    "series_instance_content_url",
    "imaging_date_num",
    "specimen_id",
    "specimen_collection_date_num",
    "signed_temporal_delta_days",
    "absolute_temporal_distance_days",
    "temporal_direction"
]

required_timepoints = [
    "condition_id",
    "specimen_id",
    "specimen_collection_date_num"
]


def validate_schema(
    dataframe,
    required_columns,
    dataset_name
):

    missing_columns = [
        column
        for column in required_columns
        if column not in dataframe.columns
    ]

    if missing_columns:

        raise ValueError(
            f"{dataset_name} is missing columns: "
            f"{missing_columns}"
        )

    print(
        f"{dataset_name:<35}: PASS"
    )


validate_schema(
    cxr,
    required_cxr,
    "CXR dataset"
)

validate_schema(
    genomics,
    required_genomics,
    "Genomics dataset"
)

validate_schema(
    cxr_master,
    required_master,
    "CXR master"
)

validate_schema(
    temporal,
    required_temporal,
    "Temporal pairs"
)

validate_schema(
    unique_timepoints,
    required_timepoints,
    "Unique genomic timepoints"
)


# ================================================================
# 7. CONDITION_ID STANDARDIZATION
# ================================================================

print("\n4. CONDITION_ID INTEGRITY")
print("-" * 80)

for dataframe in [
    cxr,
    genomics,
    cxr_master,
    temporal,
    unique_timepoints
]:

    dataframe["condition_id"] = (
        dataframe["condition_id"]
        .astype("string")
        .str.strip()
    )


cxr_missing = int(
    cxr["condition_id"].isna().sum()
)

genomics_missing = int(
    genomics["condition_id"].isna().sum()
)

master_missing = int(
    cxr_master["condition_id"].isna().sum()
)

print(
    f"CXR missing condition_id        : {cxr_missing:,}"
)

print(
    f"Genomics missing condition_id   : {genomics_missing:,}"
)

print(
    f"CXR master missing condition_id : {master_missing:,}"
)


# ================================================================
# 8. BASIC CONDITION COUNTS
# ================================================================

print("\n5. CONDITION-LEVEL MODALITY AVAILABILITY")
print("-" * 80)

cxr_conditions = set(
    cxr["condition_id"]
    .dropna()
    .unique()
)

genomic_conditions = set(
    genomics["condition_id"]
    .dropna()
    .unique()
)

common_conditions = (
    cxr_conditions &
    genomic_conditions
)

print(
    f"Unique CXR conditions        : "
    f"{len(cxr_conditions):,}"
)

print(
    f"Unique genomic conditions    : "
    f"{len(genomic_conditions):,}"
)

print(
    f"Common CXR + genomic cases   : "
    f"{len(common_conditions):,}"
)


# ================================================================
# 9. VERIFY 3859 CXR MASTER
# ================================================================

print("\n6. 3859 MULTIMODAL CXR MASTER VALIDATION")
print("-" * 80)

master_conditions = set(
    cxr_master["condition_id"]
    .dropna()
    .unique()
)

master_urls = set(
    cxr_master[
        "series_instance_content_url"
    ]
    .dropna()
    .astype(str)
)

print(
    f"Master CXR rows              : "
    f"{len(cxr_master):,}"
)

print(
    f"Master unique conditions     : "
    f"{len(master_conditions):,}"
)

print(
    f"Master unique CXR URLs       : "
    f"{len(master_urls):,}"
)

print(
    f"Master conditions also in "
    f"CXR + genomics              : "
    f"{len(master_conditions & common_conditions):,}"
)


# ================================================================
# 10. COMMON MULTIMODAL RECORDS
# ================================================================

common_cxr = cxr[
    cxr["condition_id"]
    .isin(common_conditions)
].copy()

common_genomics = genomics[
    genomics["condition_id"]
    .isin(common_conditions)
].copy()

print("\nCOMMON MULTIMODAL RECORDS")
print(
    f"CXR records                  : "
    f"{len(common_cxr):,}"
)

print(
    f"Genomic records              : "
    f"{len(common_genomics):,}"
)


# ================================================================
# 11. CXR LABEL SUMMARY
# ================================================================

print("\n7. CXR LABEL FIELD INSPECTION")
print("-" * 80)

cxr_label_fields = [
    "case_definition",
    "diagnosis_code",
    "type_of_resistance",
    "outcome"
]

cxr_label_summary_rows = []

for field in cxr_label_fields:

    counts = (
        common_cxr[field]
        .fillna("<MISSING>")
        .astype(str)
        .value_counts(
            dropna=False
        )
    )

    print(f"\nFIELD: {field}")
    print(
        counts.to_string()
    )

    cxr_label_summary_rows.append({
        "field": field,
        "unique_values":
            int(
                common_cxr[field]
                .nunique(dropna=True)
            ),
        "missing_records":
            int(
                common_cxr[field]
                .isna()
                .sum()
            ),
        "top_value":
            (
                str(counts.index[0])
                if len(counts) > 0
                else ""
            ),
        "top_value_count":
            (
                int(counts.iloc[0])
                if len(counts) > 0
                else 0
            )
    })


cxr_label_summary = pd.DataFrame(
    cxr_label_summary_rows
)


# ================================================================
# 12. CONDITION-LEVEL LABEL CONSISTENCY
# ================================================================

print("\n8. CONDITION-LEVEL CXR LABEL CONSISTENCY")
print("-" * 80)

consistency_rows = []

for field in cxr_label_fields:

    condition_counts = (
        common_cxr
        .groupby("condition_id")[field]
        .nunique(dropna=True)
    )

    total = len(condition_counts)

    multiple = int(
        (condition_counts > 1).sum()
    )

    stable = total - multiple

    consistency_rows.append({
        "field": field,
        "conditions_evaluated": total,
        "stable_or_single_value": stable,
        "multiple_values": multiple
    })

    print(
        f"{field:<25}: "
        f"stable={stable:,}, "
        f"multiple_values={multiple:,}"
    )


cxr_condition_label_consistency = pd.DataFrame(
    consistency_rows
)


# ================================================================
# 13. EXPLORATORY CXR TB/NORMAL VOCABULARY
#
# IMPORTANT:
# This is ONLY an exploratory search.
#
# It does NOT create the final TB/Normal labels.
# ================================================================

print("\n9. EXPLORATORY NORMAL/TB CXR LABEL SEARCH")
print("-" * 80)

normal_patterns = [
    "normal",
    "healthy",
    "control",
    "no tb",
    "no_tb",
    "non-tb",
    "non tb",
    "negative"
]

tb_patterns = [
    "tuberculosis",
    "tuberculous",
    "tb",
    "ptb"
]


def exploratory_label(value):

    if pd.isna(value):
        return "MISSING"

    text_value = (
        str(value)
        .strip()
        .lower()
    )

    if text_value == "":
        return "MISSING"

    normal_match = any(
        pattern in text_value
        for pattern in normal_patterns
    )

    tb_match = any(
        pattern in text_value
        for pattern in tb_patterns
    )

    if normal_match and not tb_match:
        return "NORMAL_CANDIDATE"

    if tb_match and not normal_match:
        return "TB_CANDIDATE"

    if normal_match and tb_match:
        return "CONFLICTING_TERMS"

    return "OTHER_UNCLEAR"


exploratory_rows = []

for field in [
    "case_definition",
    "diagnosis_code"
]:

    temp = common_cxr[
        [
            "condition_id",
            field
        ]
    ].copy()

    temp["source_field"] = field

    temp["exploratory_label"] = (
        temp[field]
        .apply(exploratory_label)
    )

    exploratory_rows.append(
        temp
    )


if len(exploratory_rows) > 0:

    exploratory_cxr_labels = pd.concat(
        exploratory_rows,
        ignore_index=True
    )

else:

    exploratory_cxr_labels = pd.DataFrame()


if len(exploratory_cxr_labels) > 0:

    print(
        exploratory_cxr_labels[
            "exploratory_label"
        ]
        .value_counts()
        .to_string()
    )


# ================================================================
# 14. SEARCH FULL CXR DATASET FOR NORMAL CANDIDATES
# ================================================================

print("\n10. FULL CXR DATASET — NORMAL CANDIDATE SEARCH")
print("-" * 80)

normal_condition_sets = []

for field in [
    "case_definition",
    "diagnosis_code"
]:

    series = (
        cxr[field]
        .fillna("")
        .astype(str)
        .str.lower()
    )

    mask = series.str.contains(
        r"\bnormal\b|\bhealthy\b|\bcontrol\b|\bno[\s_-]?tb\b|\bnon[\s-]?tb\b|\bnegative\b",
        regex=True,
        na=False
    )

    ids = set(
        cxr.loc[
            mask,
            "condition_id"
        ]
        .dropna()
        .astype(str)
    )

    normal_condition_sets.append(
        ids
    )

    print(
        f"{field:<25}: "
        f"{len(ids):,} candidate conditions"
    )


if len(normal_condition_sets) > 0:

    full_normal_conditions = set.union(
        *normal_condition_sets
    )

else:

    full_normal_conditions = set()


print(
    f"\nUnique potential Normal CXR conditions: "
    f"{len(full_normal_conditions):,}"
)


# ================================================================
# 15. NORMAL CXR + GENOMICS OVERLAP
# ================================================================

print("\n11. POTENTIAL NORMAL CXR + GENOMICS OVERLAP")
print("-" * 80)

potential_normal_multimodal = (
    full_normal_conditions &
    genomic_conditions
)

print(
    f"Potential Normal CXR conditions "
    f"with genomic records          : "
    f"{len(potential_normal_multimodal):,}"
)


# ================================================================
# 16. GENOMIC COLUMN INVENTORY
# ================================================================

print("\n12. GENOMIC DATASET COLUMN INVENTORY")
print("-" * 80)

for index, column in enumerate(
    genomics.columns,
    start=1
):

    print(
        f"{index:03d}. {column}"
    )


# ================================================================
# 17. POSSIBLE GENOMIC LABEL FIELDS
# ================================================================

print("\n13. POSSIBLE GENOMIC LABEL-RELATED FIELDS")
print("-" * 80)

genomic_search_terms = [
    "diagnos",
    "case",
    "tb",
    "tuberc",
    "normal",
    "healthy",
    "control",
    "disease",
    "condition",
    "status",
    "resistance",
    "outcome"
]

possible_genomic_label_fields = []

for column in genomics.columns:

    column_lower = column.lower()

    if any(
        term in column_lower
        for term in genomic_search_terms
    ):

        possible_genomic_label_fields.append(
            column
        )


for column in possible_genomic_label_fields:

    print(column)


# ================================================================
# 18. GENOMIC LOW-CARDINALITY FIELD AUDIT
# ================================================================

print("\n14. GENOMIC LOW-CARDINALITY FIELD AUDIT")
print("-" * 80)

genomic_low_cardinality_rows = []

for column in genomics.columns:

    series = genomics[column].dropna()

    if len(series) == 0:
        continue

    unique_count = series.nunique()

    if unique_count <= 30:

        counts = (
            series
            .astype(str)
            .value_counts()
        )

        top_values = " | ".join(
            [
                f"{value}={count}"
                for value, count
                in counts.head(10).items()
            ]
        )

        genomic_low_cardinality_rows.append({
            "field": column,
            "non_missing": len(series),
            "unique_values": unique_count,
            "top_values": top_values
        })


genomic_low_cardinality = pd.DataFrame(
    genomic_low_cardinality_rows
)


if len(genomic_low_cardinality) > 0:

    print(
        genomic_low_cardinality
        .to_string(index=False)
    )


# ================================================================
# 19. EXPLICIT NORMAL GENOMIC SEARCH
# ================================================================

print("\n15. EXPLICIT NORMAL/HEALTHY/CONTROL GENOMIC EVIDENCE")
print("-" * 80)

normal_genomic_hits = []

for column in genomics.columns:

    series = (
        genomics[column]
        .fillna("")
        .astype(str)
        .str.lower()
    )

    count = int(
        series.str.contains(
            r"\bnormal\b|\bhealthy\b|\bcontrol\b",
            regex=True,
            na=False
        ).sum()
    )

    if count > 0:

        normal_genomic_hits.append({
            "field": column,
            "normal_healthy_control_mentions":
                count
        })


normal_genomic_evidence = pd.DataFrame(
    normal_genomic_hits
)


if len(normal_genomic_evidence) > 0:

    print(
        normal_genomic_evidence
        .sort_values(
            "normal_healthy_control_mentions",
            ascending=False
        )
        .to_string(index=False)
    )

else:

    print(
        "No explicit Normal/Healthy/Control "
        "genomic values found."
    )


# ================================================================
# 20. EXPLICIT TB GENOMIC SEARCH
# ================================================================

print("\n16. TB VOCABULARY IN GENOMIC DATA")
print("-" * 80)

tb_genomic_hits = []

for column in genomics.columns:

    series = (
        genomics[column]
        .fillna("")
        .astype(str)
        .str.lower()
    )

    count = int(
        series.str.contains(
            r"\btb\b|tuberculosis",
            regex=True,
            na=False
        ).sum()
    )

    if count > 0:

        tb_genomic_hits.append({
            "field": column,
            "tb_tuberculosis_mentions":
                count
        })


tb_genomic_evidence = pd.DataFrame(
    tb_genomic_hits
)


if len(tb_genomic_evidence) > 0:

    print(
        tb_genomic_evidence
        .sort_values(
            "tb_tuberculosis_mentions",
            ascending=False
        )
        .to_string(index=False)
    )

else:

    print(
        "No explicit TB/Tuberculosis "
        "genomic values found."
    )


# ================================================================
# 21. EXPLICIT NORMAL GENOMIC CONDITION IDS
# ================================================================

print("\n17. EXPLICIT NORMAL GENOMIC CONDITION IDS")
print("-" * 80)

normal_genomic_condition_ids = set()

if len(normal_genomic_evidence) > 0:

    for column in normal_genomic_evidence["field"]:

        series = (
            genomics[column]
            .fillna("")
            .astype(str)
            .str.lower()
        )

        mask = series.str.contains(
            r"\bnormal\b|\bhealthy\b|\bcontrol\b",
            regex=True,
            na=False
        )

        ids = set(
            genomics.loc[
                mask,
                "condition_id"
            ]
            .dropna()
            .astype(str)
        )

        normal_genomic_condition_ids.update(
            ids
        )


print(
    f"Explicit Normal/Healthy/Control "
    f"genomic conditions             : "
    f"{len(normal_genomic_condition_ids):,}"
)


# ================================================================
# 22. EXPLICIT TB GENOMIC CONDITION IDS
# ================================================================

tb_genomic_condition_ids = set()

if len(tb_genomic_evidence) > 0:

    for column in tb_genomic_evidence["field"]:

        series = (
            genomics[column]
            .fillna("")
            .astype(str)
            .str.lower()
        )

        mask = series.str.contains(
            r"\btb\b|tuberculosis",
            regex=True,
            na=False
        )

        ids = set(
            genomics.loc[
                mask,
                "condition_id"
            ]
            .dropna()
            .astype(str)
        )

        tb_genomic_condition_ids.update(
            ids
        )


print(
    f"Explicit TB/Tuberculosis "
    f"genomic conditions             : "
    f"{len(tb_genomic_condition_ids):,}"
)


# ================================================================
# 23. EXPLICIT NORMAL CXR + GENOMIC OVERLAP
# ================================================================

print("\n18. EXPLICIT NORMAL CXR + GENOMIC OVERLAP")
print("-" * 80)

explicit_normal_multimodal = (
    normal_genomic_condition_ids
    &
    full_normal_conditions
    &
    cxr_conditions
)

print(
    f"Explicit Normal CXR + genomic "
    f"conditions                    : "
    f"{len(explicit_normal_multimodal):,}"
)


# ================================================================
# 24. TEMPORAL COVERAGE
# ================================================================

print("\n19. TEMPORAL COVERAGE")
print("-" * 80)

temporal_common = temporal[
    temporal["condition_id"]
    .isin(common_conditions)
].copy()

temporal_conditions = set(
    temporal_common[
        "condition_id"
    ]
    .dropna()
    .astype(str)
)

print(
    f"Temporal pair records          : "
    f"{len(temporal_common):,}"
)

print(
    f"Temporal-linked conditions     : "
    f"{len(temporal_conditions):,}"
)

print(
    f"Common conditions without "
    f"temporal evidence             : "
    f"{len(common_conditions - temporal_conditions):,}"
)

print("\nTemporal direction:")

print(
    temporal_common[
        "temporal_direction"
    ]
    .value_counts()
    .to_string()
)


# ================================================================
# 25. GENOMIC LONGITUDINAL STRUCTURE
# ================================================================

print("\n20. GENOMIC LONGITUDINAL STRUCTURE")
print("-" * 80)

timepoint_counts = (
    unique_timepoints[
        unique_timepoints["condition_id"]
        .isin(common_conditions)
    ]
    .drop_duplicates(
        [
            "condition_id",
            "specimen_id",
            "specimen_collection_date_num"
        ]
    )
    .groupby("condition_id")
    .size()
    .rename(
        "genomic_timepoint_count"
    )
    .reset_index()
)

multiple_timepoint_conditions = (
    timepoint_counts[
        timepoint_counts[
            "genomic_timepoint_count"
        ] > 1
    ]
)

print(
    f"Conditions with >1 genomic "
    f"timepoint                     : "
    f"{len(multiple_timepoint_conditions):,}"
)

print("\nTimepoint distribution:")

print(
    timepoint_counts[
        "genomic_timepoint_count"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)


# ================================================================
# 26. TARGET LEAKAGE INVENTORY
# ================================================================

print("\n21. TARGET LEAKAGE INVENTORY")
print("-" * 80)

leakage_fields = [
    "type_of_resistance",
    "drug_resistance_type",
    "outcome"
]

leakage_rows = []

for dataset_name, dataframe in [
    ("CXR", cxr),
    ("Genomics", genomics)
]:

    for field in leakage_fields:

        if field in dataframe.columns:

            leakage_rows.append({
                "dataset":
                    dataset_name,

                "field":
                    field,

                "status":
                    "PROHIBITED_AS_PREDICTOR"
            })

            print(
                f"{dataset_name:<12}"
                f"{field:<30}"
                f"PROHIBITED_AS_PREDICTOR"
            )


leakage_inventory = pd.DataFrame(
    leakage_rows
)


# ================================================================
# 27. CASE-LEVEL MULTIMODAL AVAILABILITY
# ================================================================

print("\n22. CONDITION-LEVEL MULTIMODAL AVAILABILITY")
print("-" * 80)

cxr_counts = (
    cxr
    .groupby("condition_id")
    .size()
    .rename("cxr_record_count")
)

genomic_counts = (
    genomics
    .groupby("condition_id")
    .size()
    .rename("genomic_record_count")
)

condition_table = pd.concat(
    [
        cxr_counts,
        genomic_counts
    ],
    axis=1
).fillna(0)

condition_table[
    "cxr_available"
] = (
    condition_table[
        "cxr_record_count"
    ] > 0
)

condition_table[
    "genomics_available"
] = (
    condition_table[
        "genomic_record_count"
    ] > 0
)

condition_table[
    "paired_multimodal"
] = (
    condition_table[
        "cxr_available"
    ]
    &
    condition_table[
        "genomics_available"
    ]
)

condition_table = (
    condition_table
    .reset_index()
)


print(
    f"Total condition records      : "
    f"{len(condition_table):,}"
)

print(
    f"Paired multimodal conditions  : "
    f"{int(condition_table['paired_multimodal'].sum()):,}"
)


# ================================================================
# 28. TB-vs-NORMAL CORE FEASIBILITY TEST
# ================================================================

print("\n23. TB-vs-NORMAL MULTIMODAL COHORT TEST")
print("-" * 80)

print(
    f"Paired CXR + genomic conditions"
    f"                         : "
    f"{len(common_conditions):,}"
)

print(
    f"Potential Normal CXR conditions"
    f"                         : "
    f"{len(full_normal_conditions):,}"
)

print(
    f"Potential Normal CXR + genomic"
    f"                         : "
    f"{len(potential_normal_multimodal):,}"
)

print(
    f"Explicit Normal genomic cases"
    f"                         : "
    f"{len(normal_genomic_condition_ids):,}"
)

print(
    f"Explicit Normal CXR + genomic"
    f"                         : "
    f"{len(explicit_normal_multimodal):,}"
)


# ================================================================
# 29. FEASIBILITY STATUS
# ================================================================

if len(explicit_normal_multimodal) > 0:

    feasibility_status = (
        "CONDITIONAL — explicit Normal CXR+genomic "
        "cases identified; label provenance requires "
        "formal validation before cohort construction."
    )

elif len(potential_normal_multimodal) > 0:

    feasibility_status = (
        "CONDITIONAL — potential Normal CXR+genomic "
        "cases identified; Normal labels require "
        "formal validation before cohort construction."
    )

else:

    feasibility_status = (
        "NOT_DEMONSTRATED — a verified Normal "
        "CXR+genomic class has not been established "
        "from the currently identified data."
    )


print("\n24. FEASIBILITY INTERPRETATION")
print("-" * 80)

print(
    feasibility_status
)


# ================================================================
# 30. DESIGN OPTIONS
# ================================================================

print("\n25. MULTIMODAL DESIGN OPTIONS")
print("-" * 80)

design_options = pd.DataFrame([
    {
        "design_id":
            "MM-01",

        "design":
            "Paired CXR + genomic TB-vs-Normal classification",

        "requirement":
            "Both TB and Normal classes must have "
            "paired CXR and genomic information.",

        "current_status":
            (
                "CONDITIONAL"
                if len(potential_normal_multimodal) > 0
                else "NOT_DEMONSTRATED"
            )
    },

    {
        "design_id":
            "MM-02",

        "design":
            "CXR-only TB-vs-Normal classification",

        "requirement":
            "Valid TB and Normal CXR labels.",

        "current_status":
            "NOT_SELECTED"
    },

    {
        "design_id":
            "MM-03",

        "design":
            "TB-only CXR + genomic characterization",

        "requirement":
            "Paired TB CXR and genomic information.",

        "current_status":
            "NOT_SELECTED"
    }
])

print(
    design_options.to_string(
        index=False
    )
)


# ================================================================
# 31. COHORT REQUIREMENTS
# ================================================================

cohort_requirements = pd.DataFrame([
    {
        "requirement_id":
            "REQ-01",

        "requirement":
            "TB class",

        "evidence":
            f"{len(common_conditions):,} "
            f"paired CXR+genomic conditions",

        "status":
            "SUPPORTED"
    },

    {
        "requirement_id":
            "REQ-02",

        "requirement":
            "Normal CXR class",

        "evidence":
            f"{len(full_normal_conditions):,} "
            f"potential Normal CXR conditions",

        "status":
            (
                "POTENTIAL"
                if len(full_normal_conditions) > 0
                else "NOT_DEMONSTRATED"
            )
    },

    {
        "requirement_id":
            "REQ-03",

        "requirement":
            "Normal CXR + genomic pairing",

        "evidence":
            f"{len(potential_normal_multimodal):,} "
            f"potential paired cases",

        "status":
            (
                "POTENTIAL"
                if len(potential_normal_multimodal) > 0
                else "NOT_DEMONSTRATED"
            )
    },

    {
        "requirement_id":
            "REQ-04",

        "requirement":
            "Temporal information",

        "evidence":
            f"{len(temporal_conditions):,} "
            f"common conditions represented",

        "status":
            (
                "SUPPORTED"
                if len(temporal_conditions) > 0
                else "NOT_SUPPORTED"
            )
    },

    {
        "requirement_id":
            "REQ-05",

        "requirement":
            "Case-level independence",

        "evidence":
            "condition_id",

        "status":
            "LOCKED"
    },

    {
        "requirement_id":
            "REQ-06",

        "requirement":
            "Target leakage protection",

        "evidence":
            "Direct target fields prohibited as predictors",

        "status":
            "LOCKED"
    }
])


# ================================================================
# 32. AUDIT SUMMARY
# ================================================================

audit_summary = pd.DataFrame([
    {
        "metric":
            "CXR records",
        "value":
            len(cxr)
    },

    {
        "metric":
            "Genomic records",
        "value":
            len(genomics)
    },

    {
        "metric":
            "Unique CXR conditions",
        "value":
            len(cxr_conditions)
    },

    {
        "metric":
            "Unique genomic conditions",
        "value":
            len(genomic_conditions)
    },

    {
        "metric":
            "Common CXR + genomic conditions",
        "value":
            len(common_conditions)
    },

    {
        "metric":
            "Common CXR records",
        "value":
            len(common_cxr)
    },

    {
        "metric":
            "Common genomic records",
        "value":
            len(common_genomics)
    },

    {
        "metric":
            "Temporal pair records",
        "value":
            len(temporal_common)
    },

    {
        "metric":
            "Temporal-linked conditions",
        "value":
            len(temporal_conditions)
    },

    {
        "metric":
            "Conditions with multiple genomic timepoints",
        "value":
            len(multiple_timepoint_conditions)
    },

    {
        "metric":
            "Potential Normal CXR conditions",
        "value":
            len(full_normal_conditions)
    },

    {
        "metric":
            "Potential Normal CXR + genomic conditions",
        "value":
            len(potential_normal_multimodal)
    },

    {
        "metric":
            "Explicit Normal genomic conditions",
        "value":
            len(normal_genomic_condition_ids)
    },

    {
        "metric":
            "Explicit Normal CXR + genomic conditions",
        "value":
            len(explicit_normal_multimodal)
    },

    {
        "metric":
            "TB-vs-Normal multimodal feasibility",
        "value":
            feasibility_status
    }
])


# ================================================================
# 33. SAVE OUTPUT FUNCTION
# ================================================================

print("\n26. SAVING OUTPUTS")
print("-" * 80)


def save_csv(
    dataframe,
    filename
):

    if dataframe is None:

        raise ValueError(
            f"Cannot save {filename}: dataframe is None."
        )

    if not isinstance(
        dataframe,
        pd.DataFrame
    ):

        raise TypeError(
            f"Cannot save {filename}: "
            f"object is not a pandas DataFrame."
        )

    output_path = (
        OUTPUT_DIR /
        filename
    )

    dataframe.to_csv(
        output_path,
        index=False,
        encoding="utf-8-sig"
    )

    print(
        f"{filename:<75}: PASS"
    )


# ================================================================
# 34. SAVE ALL DATAFRAME OUTPUTS
# ================================================================

save_csv(
    audit_summary,
    "TB_Portals_March2025_Step_3B_6_44_Audit_Summary.csv"
)

save_csv(
    condition_table,
    "TB_Portals_March2025_Step_3B_6_44_Condition_Level_Multimodal_Availability.csv"
)

save_csv(
    cxr_label_summary,
    "TB_Portals_March2025_Step_3B_6_44_CXR_Label_Summary.csv"
)

save_csv(
    cxr_condition_label_consistency,
    "TB_Portals_March2025_Step_3B_6_44_CXR_Label_Consistency.csv"
)

save_csv(
    exploratory_cxr_labels,
    "TB_Portals_March2025_Step_3B_6_44_Exploratory_CXR_Label_Evidence.csv"
)

save_csv(
    genomic_low_cardinality,
    "TB_Portals_March2025_Step_3B_6_44_Genomic_Low_Cardinality_Fields.csv"
)

save_csv(
    normal_genomic_evidence,
    "TB_Portals_March2025_Step_3B_6_44_Normal_Genomic_Evidence.csv"
)

save_csv(
    tb_genomic_evidence,
    "TB_Portals_March2025_Step_3B_6_44_TB_Genomic_Evidence.csv"
)

save_csv(
    temporal_common,
    "TB_Portals_March2025_Step_3B_6_44_Common_Cohort_Temporal_Evidence.csv"
)

save_csv(
    timepoint_counts,
    "TB_Portals_March2025_Step_3B_6_44_Genomic_Timepoint_Structure.csv"
)

save_csv(
    leakage_inventory,
    "TB_Portals_March2025_Step_3B_6_44_Target_Leakage_Inventory.csv"
)

save_csv(
    design_options,
    "TB_Portals_March2025_Step_3B_6_44_Design_Options.csv"
)

save_csv(
    cohort_requirements,
    "TB_Portals_March2025_Step_3B_6_44_Cohort_Requirements.csv"
)


# ================================================================
# 35. SAVE DECISION RECORD
# ================================================================

decision_record = {

    "step":
        "3B.6.44",

    "research_objective":
        "Investigate feasibility of TB-vs-Normal "
        "classification using paired CXR + genomic information",

    "case_unit":
        "condition_id",

    "cxr_records":
        int(len(cxr)),

    "genomic_records":
        int(len(genomics)),

    "unique_cxr_conditions":
        int(len(cxr_conditions)),

    "unique_genomic_conditions":
        int(len(genomic_conditions)),

    "common_multimodal_conditions":
        int(len(common_conditions)),

    "potential_normal_cxr_conditions":
        int(len(full_normal_conditions)),

    "potential_normal_cxr_genomic_conditions":
        int(len(potential_normal_multimodal)),

    "explicit_normal_genomic_conditions":
        int(len(normal_genomic_condition_ids)),

    "explicit_normal_cxr_genomic_conditions":
        int(len(explicit_normal_multimodal)),

    "multiple_genomic_timepoint_conditions":
        int(len(multiple_timepoint_conditions)),

    "feasibility_status":
        feasibility_status,

    "final_target_selected":
        False,

    "final_cohort_locked":
        False,

    "representative_cxr_selected":
        False,

    "representative_genomic_specimen_selected":
        False,

    "preprocessing_performed":
        False,

    "lung_segmentation_performed":
        False,

    "model_training_performed":
        False,

    "explainable_ai_performed":
        False,

    "raw_dicom_modified":
        False,

    "source_metadata_modified":
        False
}


decision_path = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_44_Decision_Record.json"
)

with open(
    decision_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        decision_record,
        file,
        indent=4
    )

print(
    f"{decision_path.name:<75}: PASS"
)


# ================================================================
# 36. FINAL VALIDATION
# ================================================================

print("\n27. FINAL VALIDATION")
print("-" * 80)

validation_checks = {

    "CXR dataset loaded":
        len(cxr) > 0,

    "Genomics dataset loaded":
        len(genomics) > 0,

    "3859 CXR master loaded":
        len(cxr_master) == 3859,

    "Temporal data loaded":
        len(temporal) > 0,

    "Unique genomic timepoints loaded":
        len(unique_timepoints) > 0,

    "condition_id in CXR":
        "condition_id" in cxr.columns,

    "condition_id in genomics":
        "condition_id" in genomics.columns,

    "3859 master conditions":
        len(master_conditions) == 3081,

    "3859 master URLs":
        len(master_urls) == 3859,

    "No preprocessing":
        True,

    "No lung segmentation":
        True,

    "No model training":
        True,

    "No explainable AI":
        True,

    "Raw DICOM unchanged":
        True,

    "Source metadata unchanged":
        True
}


all_validation_passed = True

for check, status in validation_checks.items():

    print(
        f"{check:<50}: "
        f"{'PASS' if status else 'FAIL'}"
    )

    if not status:
        all_validation_passed = False


# ================================================================
# 37. FINAL STATUS
# ================================================================

print("\n")
print("=" * 80)
print("STEP 3B.6.44 — FINAL STATUS")
print("=" * 80)

print(
    f"Unique CXR conditions                 : "
    f"{len(cxr_conditions):,}"
)

print(
    f"Unique genomic conditions             : "
    f"{len(genomic_conditions):,}"
)

print(
    f"Common CXR + genomic conditions       : "
    f"{len(common_conditions):,}"
)

print(
    f"Potential Normal CXR conditions       : "
    f"{len(full_normal_conditions):,}"
)

print(
    f"Potential Normal CXR + genomic cases : "
    f"{len(potential_normal_multimodal):,}"
)

print(
    f"Explicit Normal genomic conditions   : "
    f"{len(normal_genomic_condition_ids):,}"
)

print(
    f"Explicit Normal CXR + genomic cases  : "
    f"{len(explicit_normal_multimodal):,}"
)

print(
    f"Multiple genomic timepoint conditions: "
    f"{len(multiple_timepoint_conditions):,}"
)

print()
print(
    "TB-vs-Normal multimodal feasibility:"
)

print(
    feasibility_status
)

print()

print(
    f"Technical validation: "
    f"{'PASS' if all_validation_passed else 'FAIL'}"
)

print()
print("FINAL TARGET                  : NOT LOCKED")
print("FINAL COHORT                  : NOT LOCKED")
print("REPRESENTATIVE CXR           : NOT SELECTED")
print("REPRESENTATIVE GENOMIC SAMPLE: NOT SELECTED")
print("PREPROCESSING                 : NOT PERFORMED")
print("LUNG SEGMENTATION             : NOT PERFORMED")
print("MODEL TRAINING                : NOT PERFORMED")
print("EXPLAINABLE AI                : NOT PERFORMED")
print("RAW DICOM MODIFICATION       : NO")
print("SOURCE METADATA MODIFICATION : NO")

print("=" * 80)
print("STEP 3B.6.44 COMPLETE")
print("=" * 80)
```

```python
# ================================================================
# TB PORTALS MARCH 2025
# STEP 3B.6.48
# EXTERNAL NORMAL SOURCE EVIDENCE &
# RESEARCH-DESIGN DECISION AUDIT
# ================================================================
#
# PURPOSE
# ================================================================
#
# 3B.6.47 established the requirements for an external Normal
# multimodal source.
#
# 3B.6.48 now performs the formal evidence audit of the candidate
# sources and determines whether each source can satisfy the
# ORIGINAL research objective:
#
#       CXR + GENOMICS
#             |
#             v
#       TB vs NORMAL
#
# IMPORTANT:
#
# This step does NOT:
#
#   - create a Normal class artificially
#   - pair unrelated patients
#   - attach external Normal CXR to TB genomics
#   - download datasets
#   - preprocess images
#   - perform segmentation
#   - train models
#   - perform XAI
#
# It only records verified source characteristics and the
# scientific implications.
#
# ================================================================


from pathlib import Path
import pandas as pd
import json
import warnings

warnings.filterwarnings("ignore")


# ================================================================
# 1. BASE DIRECTORIES
# ================================================================

BASE_DIR = Path(r"C:\TBP")

METADATA_DIR = (
    BASE_DIR /
    "Metadata"
)


# ================================================================
# 2. STEP 3B.6.47 INPUTS
# ================================================================

STEP_47_DIR = (
    METADATA_DIR /
    "Step_3B_6_47_External_Normal_Source_Audit"
)


STEP_47_DECISION = (
    STEP_47_DIR /
    "TB_Portals_March2025_Step_3B_6_47_Decision_Record.json"
)

STEP_47_PENDING = (
    STEP_47_DIR /
    "TB_Portals_March2025_Step_3B_6_47_Pending_Items.csv"
)

STEP_47_CANDIDATES = (
    STEP_47_DIR /
    "TB_Portals_March2025_Step_3B_6_47_External_Candidate_Source_Register.csv"
)

STEP_47_COMPATIBILITY = (
    STEP_47_DIR /
    "TB_Portals_March2025_Step_3B_6_47_Multimodal_Compatibility_Matrix.csv"
)


# ================================================================
# 3. STEP 3B.6.46 CLOSURE
# ================================================================

STEP_46_DIR = (
    METADATA_DIR /
    "Step_3B_6_46_Notebook_03_Closure_Audit"
)

STEP_46_CLOSURE = (
    STEP_46_DIR /
    "TB_Portals_March2025_Step_3B_6_46_Notebook_03_Closure_Record.json"
)


# ================================================================
# 4. CXR MASTER
# ================================================================

CXR_MASTER = (
    METADATA_DIR /
    "Step_3B_6_24_Notebook_03" /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)


# ================================================================
# 5. OUTPUT DIRECTORY
# ================================================================

OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_48_External_Source_Evidence_Decision_Audit"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# 6. START
# ================================================================

print("=" * 80)
print("STEP 3B.6.48 — EXTERNAL NORMAL SOURCE EVIDENCE")
print("                         & RESEARCH-DESIGN DECISION AUDIT")
print("=" * 80)


# ================================================================
# 7. INPUT FILE VALIDATION
# ================================================================

print("\n1. INPUT FILE VALIDATION")
print("-" * 80)

required_files = {

    "3B.6.47 decision record":
        STEP_47_DECISION,

    "3B.6.47 pending items":
        STEP_47_PENDING,

    "3B.6.47 candidate register":
        STEP_47_CANDIDATES,

    "3B.6.47 compatibility matrix":
        STEP_47_COMPATIBILITY,

    "3B.6.46 closure":
        STEP_46_CLOSURE,

    "3859 CXR master":
        CXR_MASTER
}


for name, path in required_files.items():

    if not path.exists():

        raise FileNotFoundError(
            f"\nRequired file missing:\n"
            f"{name}\n"
            f"{path}"
        )

    print(
        f"{name:<40}: FOUND"
    )


# ================================================================
# 8. LOAD INPUTS
# ================================================================

print("\n2. LOADING PREVIOUS EVIDENCE")
print("-" * 80)

with open(
    STEP_47_DECISION,
    "r",
    encoding="utf-8"
) as file:

    step_47_decision = json.load(
        file
    )


with open(
    STEP_46_CLOSURE,
    "r",
    encoding="utf-8"
) as file:

    step_46_closure = json.load(
        file
    )


step_47_pending = pd.read_csv(
    STEP_47_PENDING,
    low_memory=False
)

step_47_candidates = pd.read_csv(
    STEP_47_CANDIDATES,
    low_memory=False
)

step_47_compatibility = pd.read_csv(
    STEP_47_COMPATIBILITY,
    low_memory=False
)

cxr_master = pd.read_csv(
    CXR_MASTER,
    low_memory=False
)


print(
    f"CXR master rows              : "
    f"{len(cxr_master):,}"
)

print(
    f"External candidates          : "
    f"{len(step_47_candidates):,}"
)

print(
    f"Compatibility records        : "
    f"{len(step_47_compatibility):,}"
)


# ================================================================
# 9. PREVIOUS STATE VALIDATION
# ================================================================

print("\n3. PREVIOUS STATE VALIDATION")
print("-" * 80)

if (
    step_46_closure[
        "paired_multimodal_conditions"
    ]
    != 3081
):

    raise ValueError(
        "Expected 3,081 paired TB multimodal conditions "
        "from Notebook 03."
    )


if (
    step_46_closure[
        "verified_normal_paired_conditions"
    ]
    != 0
):

    raise ValueError(
        "Previous closure unexpectedly contains verified "
        "Normal paired cases."
    )


print(
    "Paired TB multimodal cases   : "
    f"{step_46_closure['paired_multimodal_conditions']:,}"
)

print(
    "Verified Normal paired cases : "
    f"{step_46_closure['verified_normal_paired_conditions']:,}"
)

print(
    "Previous final cohort locked : "
    f"{step_46_closure['final_cohort_locked']}"
)


# ================================================================
# 10. CXR MASTER INTEGRITY
# ================================================================

print("\n4. CXR MASTER INTEGRITY")
print("-" * 80)

required_cxr_columns = [
    "condition_id",
    "series_instance_content_url"
]

missing_columns = [
    c
    for c in required_cxr_columns
    if c not in cxr_master.columns
]

if missing_columns:

    raise ValueError(
        f"CXR master missing columns: {missing_columns}"
    )


cxr_master[
    "condition_id"
] = (
    cxr_master[
        "condition_id"
    ]
    .astype("string")
    .str.strip()
)

cxr_master[
    "series_instance_content_url"
] = (
    cxr_master[
        "series_instance_content_url"
    ]
    .astype("string")
    .str.strip()
)


cxr_rows = len(
    cxr_master
)

cxr_conditions = (
    cxr_master[
        "condition_id"
    ]
    .nunique()
)

cxr_urls = (
    cxr_master[
        "series_instance_content_url"
    ]
    .nunique()
)


print(
    f"CXR records                 : "
    f"{cxr_rows:,}"
)

print(
    f"Unique conditions            : "
    f"{cxr_conditions:,}"
)

print(
    f"Unique CXR URLs              : "
    f"{cxr_urls:,}"
)


if cxr_rows != 3859:
    raise ValueError(
        "Unexpected CXR master row count."
    )

if cxr_conditions != 3081:
    raise ValueError(
        "Unexpected CXR condition count."
    )

if cxr_urls != 3859:
    raise ValueError(
        "Unexpected unique CXR URL count."
    )


print(
    "CXR master integrity         : PASS"
)


# ================================================================
# 11. EXTERNAL SOURCE EVIDENCE REGISTER
# ================================================================
#
# IMPORTANT:
#
# The statuses below are evidence classifications.
# They are NOT recommendations.
#
# ================================================================

print("\n5. VERIFIED EXTERNAL SOURCE EVIDENCE")
print("-" * 80)

external_evidence = pd.DataFrame([

    {
        "source_id":
            "EXT-01",

        "source":
            "NIH ChestX-ray14",

        "CXR_available":
            "YES",

        "Normal_CXR":
            "POTENTIAL/NO-FINDING",

        "genomics_available":
            "NOT_ESTABLISHED",

        "same_subject_CXR_genomic_link":
            "NOT_ESTABLISHED",

        "paired_Normal_multimodal":
            "NO",

        "evidence_basis":
            "Public CXR dataset with no-finding/CXR information; "
            "no paired Normal genomic component established.",

        "scientific_role":
            "CXR-only candidate"
    },

    {
        "source_id":
            "EXT-02",

        "source":
            "Shenzhen Hospital CXR Set",

        "CXR_available":
            "YES",

        "Normal_CXR":
            "YES",

        "genomics_available":
            "NO_EVIDENCE",

        "same_subject_CXR_genomic_link":
            "NO_EVIDENCE",

        "paired_Normal_multimodal":
            "NO",

        "evidence_basis":
            "662 PA CXR images; 326 normal and 336 TB-consistent "
            "abnormal cases. No paired genomic component.",

        "scientific_role":
            "CXR-only candidate"
    },

    {
        "source_id":
            "EXT-03",

        "source":
            "Montgomery County CXR Set",

        "CXR_available":
            "YES",

        "Normal_CXR":
            "YES",

        "genomics_available":
            "NO_EVIDENCE",

        "same_subject_CXR_genomic_link":
            "NO_EVIDENCE",

        "paired_Normal_multimodal":
            "NO",

        "evidence_basis":
            "138 PA CXR images; 80 normal and 58 TB-consistent "
            "abnormal cases. No paired genomic component.",

        "scientific_role":
            "CXR-only candidate"
    },

    {
        "source_id":
            "EXT-04",

        "source":
            "TBX11K",

        "CXR_available":
            "YES",

        "Normal_CXR":
            "YES",

        "genomics_available":
            "NO_EVIDENCE",

        "same_subject_CXR_genomic_link":
            "NO_EVIDENCE",

        "paired_Normal_multimodal":
            "NO",

        "evidence_basis":
            "CXR dataset containing Healthy and other CXR classes; "
            "no paired Normal genomic component established.",

        "scientific_role":
            "CXR-only candidate"
    }
])


print(
    external_evidence.to_string(
        index=False
    )
)


# ================================================================
# 12. SOURCE REQUIREMENT TEST
# ================================================================

print("\n6. EXTERNAL SOURCE REQUIREMENT TEST")
print("-" * 80)

requirements = [

    (
        "Normal CXR",
        "Required"
    ),

    (
        "Normal genomic data",
        "Required"
    ),

    (
        "Same-subject CXR/genomic linkage",
        "Required"
    ),

    (
        "Documented source provenance",
        "Required"
    ),

    (
        "Research-use permission",
        "Required"
    ),

    (
        "Leakage-safe subject identifiers",
        "Required"
    ),

    (
        "Imaging compatibility",
        "Required"
    ),

    (
        "Genomic compatibility",
        "Required"
    )
]


requirement_rows = []


for candidate_id in external_evidence[
    "source_id"
].unique():

    row = external_evidence[
        external_evidence[
            "source_id"
        ]
        == candidate_id
    ].iloc[0]

    source_name = row[
        "source"
    ]

    tests = {

        "Normal CXR":
            row[
                "Normal_CXR"
            ]
            in [
                "YES",
                "POTENTIAL/NO-FINDING"
            ],

        "Normal genomic data":
            row[
                "genomics_available"
            ]
            == "YES",

        "Same-subject CXR/genomic linkage":
            row[
                "same_subject_CXR_genomic_link"
            ]
            == "YES",

        "Documented source provenance":
            True,

        "Research-use permission":
            "NOT_AUDITED",

        "Leakage-safe subject identifiers":
            "NOT_AUDITED",

        "Imaging compatibility":
            "NOT_AUDITED",

        "Genomic compatibility":
            False
            if row[
                "genomics_available"
            ]
            != "YES"
            else "NOT_AUDITED"
    }

    for requirement, result in tests.items():

        if result is True:

            status = "SUPPORTED"

        elif result is False:

            status = "NOT_SUPPORTED"

        else:

            status = str(
                result
            )

        requirement_rows.append({

            "source_id":
                candidate_id,

            "source":
                source_name,

            "requirement":
                requirement,

            "status":
                status
        })


source_requirement_tests = pd.DataFrame(
    requirement_rows
)


print(
    source_requirement_tests.to_string(
        index=False
    )
)


# ================================================================
# 13. FULLY PAIRED NORMAL TEST
# ================================================================

print("\n7. FULLY PAIRED NORMAL SOURCE TEST")
print("-" * 80)

fully_paired_sources = []

for candidate_id in external_evidence[
    "source_id"
]:

    row = external_evidence[
        external_evidence[
            "source_id"
        ]
        == candidate_id
    ].iloc[0]

    fully_paired = (
        row[
            "Normal_CXR"
        ]
        == "YES"
        and
        row[
            "genomics_available"
        ]
        == "YES"
        and
        row[
            "same_subject_CXR_genomic_link"
        ]
        == "YES"
    )

    fully_paired_sources.append({

        "source_id":
            candidate_id,

        "source":
            row[
                "source"
            ],

        "fully_paired_Normal":
            fully_paired,

        "status":
            (
                "ELIGIBLE_FOR_DETAILED_AUDIT"
                if fully_paired
                else "NOT_ELIGIBLE"
            )
    })


fully_paired_test = pd.DataFrame(
    fully_paired_sources
)


print(
    fully_paired_test.to_string(
        index=False
    )
)


# ================================================================
# 14. FALSE-PAIRING PROTECTION
# ================================================================

print("\n8. FALSE-PAIRING PROTECTION")
print("-" * 80)

pairing_protection = pd.DataFrame([

    {
        "rule_id":
            "PAIR-48-01",

        "rule":
            "Do not attach external Normal CXR to TB Portals "
            "genomic records.",

        "status":
            "LOCKED"
    },

    {
        "rule_id":
            "PAIR-48-02",

        "rule":
            "Do not attach external Normal genomics to TB Portals "
            "CXR records.",

        "status":
            "LOCKED"
    },

    {
        "rule_id":
            "PAIR-48-03",

        "rule":
            "Do not treat a CXR-only Normal dataset as a fully "
            "paired multimodal Normal dataset.",

        "status":
            "LOCKED"
    },

    {
        "rule_id":
            "PAIR-48-04",

        "rule":
            "Do not infer Normal genomic features from missing "
            "genomic information.",

        "status":
            "LOCKED"
    },

    {
        "rule_id":
            "PAIR-48-05",

        "rule":
            "Do not create synthetic genomic records merely to "
            "balance the Normal class.",

        "status":
            "LOCKED"
    }
])


print(
    pairing_protection.to_string(
        index=False
    )
)


# ================================================================
# 15. RESEARCH DESIGN DECISION MATRIX
# ================================================================

print("\n9. RESEARCH DESIGN DECISION MATRIX")
print("-" * 80)

design_matrix = pd.DataFrame([

    {
        "design_id":
            "DESIGN-A",

        "research_question":
            "Can CXR + genomics distinguish TB from Normal?",

        "Normal_CXR":
            "YES",

        "Normal_genomics":
            "YES",

        "same_subject_pair":
            "YES",

        "current_status":
            "BLOCKED",

        "reason":
            "No verified external paired Normal source."
    },

    {
        "design_id":
            "DESIGN-B",

        "research_question":
            "Can a multimodal model classify TB vs Normal "
            "when Normal cases lack genomics?",

        "Normal_CXR":
            "YES",

        "Normal_genomics":
            "NO",

        "same_subject_pair":
            "NO",

        "current_status":
            "NOT_SELECTED",

        "reason":
            "Requires an explicit missing-modality architecture "
            "and changes the original fully paired design."
    },

    {
        "design_id":
            "DESIGN-C",

        "research_question":
            "Can CXR distinguish TB from Normal?",

        "Normal_CXR":
            "YES",

        "Normal_genomics":
            "NOT_REQUIRED",

        "same_subject_pair":
            "NOT_APPLICABLE",

        "current_status":
            "ALTERNATIVE_ONLY",

        "reason":
            "Changes the original multimodal research objective."
    },

    {
        "design_id":
            "DESIGN-D",

        "research_question":
            "Can multimodal CXR + genomics characterize TB "
            "without a Normal class?",

        "Normal_CXR":
            "NOT_REQUIRED",

        "Normal_genomics":
            "NOT_REQUIRED",

        "same_subject_pair":
            "YES",

        "current_status":
            "ALTERNATIVE_ONLY",

        "reason":
            "Changes the original TB-vs-Normal objective."
    }
])


print(
    design_matrix.to_string(
        index=False
    )
)


# ================================================================
# 16. CURRENT SCIENTIFIC DECISION
# ================================================================

print("\n10. CURRENT SCIENTIFIC DECISION")
print("-" * 80)

paired_source_count = int(
    (
        fully_paired_test[
            "fully_paired_Normal"
        ]
        == True
    ).sum()
)


if paired_source_count > 0:

    current_decision = (
        "PAIRED_NORMAL_SOURCE_AVAILABLE_FOR_DETAILED_AUDIT"
    )

    final_cohort_status = (
        "NOT_YET_LOCKED"
    )

else:

    current_decision = (
        "NO_VERIFIED_PAIRED_NORMAL_SOURCE"
    )

    final_cohort_status = (
        "NOT_LOCKED"
    )


print(
    f"Fully paired Normal sources  : "
    f"{paired_source_count}"
)

print(
    f"Current decision              : "
    f"{current_decision}"
)

print(
    f"Final cohort status           : "
    f"{final_cohort_status}"
)


# ================================================================
# 17. WHAT REMAINS PENDING
# ================================================================

print("\n11. PENDING SCIENTIFIC DECISIONS")
print("-" * 80)

pending = pd.DataFrame([

    {
        "decision_id":
            "DEC-48-01",

        "decision":
            "Identify a verified Normal CXR+genomic source",

        "status":
            (
                "RESOLVED"
                if paired_source_count > 0
                else "PENDING"
            ),

        "required_before":
            "Fully paired multimodal cohort"
    },

    {
        "decision_id":
            "DEC-48-02",

        "decision":
            "Verify same-subject linkage in the selected source",

        "status":
            (
                "PENDING"
                if paired_source_count > 0
                else "NOT_REACHED"
            ),

        "required_before":
            "External source integration"
    },

    {
        "decision_id":
            "DEC-48-03",

        "decision":
            "Finalize TB-vs-Normal research design",

        "status":
            "PENDING",

        "required_before":
            "Final cohort lock"
    },

    {
        "decision_id":
            "DEC-48-04",

        "decision":
            "Define representative CXR rule",

        "status":
            "PENDING",

        "required_before":
            "Image preprocessing"
    },

    {
        "decision_id":
            "DEC-48-05",

        "decision":
            "Define representative genomic specimen rule",

        "status":
            "PENDING",

        "required_before":
            "Genomic preprocessing"
    }
])


print(
    pending.to_string(
        index=False
    )
)


# ================================================================
# 18. NOTEBOOK 04 LOCK
# ================================================================

notebook_04_lock = pd.DataFrame([

    {
        "operation":
            "External Normal source integration",

        "status":
            (
                "ALLOWED_FOR_DETAILED_AUDIT"
                if paired_source_count > 0
                else "BLOCKED"
            )
    },

    {
        "operation":
            "TB Portals CXR preprocessing",

        "status":
            "BLOCKED"
    },

    {
        "operation":
            "Normal CXR preprocessing",

        "status":
            "BLOCKED"
    },

    {
        "operation":
            "Lung segmentation",

        "status":
            "BLOCKED"
    },

    {
        "operation":
            "Multimodal model training",

        "status":
            "BLOCKED"
    },

    {
        "operation":
            "Explainable AI",

        "status":
            "BLOCKED"
    }
])


# ================================================================
# 19. SAVE OUTPUTS
# ================================================================

print("\n12. SAVING OUTPUTS")
print("-" * 80)


def save_csv(
    dataframe,
    filename
):

    path = (
        OUTPUT_DIR /
        filename
    )

    dataframe.to_csv(
        path,
        index=False,
        encoding="utf-8-sig"
    )

    print(
        f"{filename:<75}: PASS"
    )


save_csv(
    external_evidence,
    "TB_Portals_March2025_Step_3B_6_48_External_Source_Evidence.csv"
)

save_csv(
    source_requirement_tests,
    "TB_Portals_March2025_Step_3B_6_48_Source_Requirement_Tests.csv"
)

save_csv(
    fully_paired_test,
    "TB_Portals_March2025_Step_3B_6_48_Fully_Paired_Normal_Test.csv"
)

save_csv(
    pairing_protection,
    "TB_Portals_March2025_Step_3B_6_48_False_Pairing_Protection.csv"
)

save_csv(
    design_matrix,
    "TB_Portals_March2025_Step_3B_6_48_Research_Design_Decision_Matrix.csv"
)

save_csv(
    pending,
    "TB_Portals_March2025_Step_3B_6_48_Pending_Decisions.csv"
)

save_csv(
    notebook_04_lock,
    "TB_Portals_March2025_Step_3B_6_48_Notebook_04_Lock.csv"
)


# ================================================================
# 20. DECISION RECORD
# ================================================================

decision_record = {

    "step":
        "3B.6.48",

    "research_objective":
        "TB vs Normal using CXR + genomic information",

    "case_unit":
        "condition_id",

    "tb_portals_paired_tb_conditions":
        3081,

    "tb_portals_verified_normal_paired_conditions":
        0,

    "verified_external_fully_paired_normal_sources":
        paired_source_count,

    "current_decision":
        current_decision,

    "final_cohort_status":
        final_cohort_status,

    "false_pairing_prohibited":
        True,

    "synthetic_normal_genomics_prohibited":
        True,

    "preprocessing_started":
        False,

    "lung_segmentation_started":
        False,

    "model_training_started":
        False,

    "xai_started":
        False,

    "raw_dicom_modified":
        False,

    "source_metadata_modified":
        False,

    "next_step":
        (
            "Detailed audit of a verified paired Normal source"
            if paired_source_count > 0
            else
            "Scientific decision required: identify a legitimate "
            "paired Normal CXR+genomic source or formally redefine "
            "the research design."
        )
}


json_output = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_48_Decision_Record.json"
)

with open(
    json_output,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        decision_record,
        file,
        indent=4
    )


print(
    f"{json_output.name:<75}: PASS"
)


# ================================================================
# 21. FINAL VALIDATION
# ================================================================

print("\n13. FINAL VALIDATION")
print("-" * 80)

validation = {

    "CXR master = 3859":
        cxr_rows == 3859,

    "CXR conditions = 3081":
        cxr_conditions == 3081,

    "CXR URLs = 3859":
        cxr_urls == 3859,

    "TB paired cases = 3081":
        step_46_closure[
            "paired_multimodal_conditions"
        ] == 3081,

    "TB Normal paired cases = 0":
        step_46_closure[
            "verified_normal_paired_conditions"
        ] == 0,

    "No false Normal pairing":
        True,

    "No synthetic Normal genomics":
        True,

    "Final cohort remains unlocked":
        final_cohort_status == "NOT_LOCKED",

    "No preprocessing":
        True,

    "No lung segmentation":
        True,

    "No model training":
        True,

    "No XAI":
        True,

    "Raw DICOM unchanged":
        True,

    "Source metadata unchanged":
        True
}


all_pass = True

for check, result in validation.items():

    print(
        f"{check:<55}: "
        f"{'PASS' if result else 'FAIL'}"
    )

    if not result:

        all_pass = False


# ================================================================
# 22. FINAL STATUS
# ================================================================

print()
print("=" * 80)
print("STEP 3B.6.48 — FINAL STATUS")
print("=" * 80)

print(
    f"TB Portals paired TB cases       : "
    f"3,081"
)

print(
    f"TB Portals paired Normal cases   : "
    f"0"
)

print(
    f"Verified external paired Normal  : "
    f"{paired_source_count}"
)

print(
    f"Research-design status           : "
    f"{current_decision}"
)

print(
    f"Final cohort                     : "
    f"{final_cohort_status}"
)

print(
    f"Technical validation             : "
    f"{'PASS' if all_pass else 'FAIL'}"
)

print()
print("DICOM PREPROCESSING              : NOT STARTED")
print("LUNG SEGMENTATION                : NOT STARTED")
print("MULTIMODAL MODEL                 : NOT STARTED")
print("EXPLAINABLE AI                   : NOT STARTED")
print("RAW DICOM MODIFIED               : NO")
print("SOURCE METADATA MODIFIED         : NO")

print("=" * 80)
print("STEP 3B.6.48 COMPLETE")
print("=" * 80)
```

```python
# ================================================================
# TB PORTALS MARCH 2025
# STEP 3B.6.49
# INTERNAL NORMAL-CASE DISCOVERY AUDIT
# ================================================================
#
# RESEARCH OBJECTIVE
# ================================================================
#
# Original objective:
#
#       CXR IMAGE + GENOMIC INFORMATION
#                    ↓
#              TB vs NORMAL
#
# This step searches ONLY inside the TB Portals March 2025
# dataset already available locally.
#
# NO external datasets are used.
#
# Specifically excluded from this audit:
#   - Montgomery
#   - Shenzhen
#   - TBX11K
#   - NIH ChestX-ray14
#   - any other external dataset
#
# ================================================================
#
# SCIENTIFIC RULES
# ================================================================
#
# 1. Missing data ≠ Normal
# 2. Missing genomics ≠ Normal
# 3. No resistance information ≠ Normal
# 4. Unknown outcome ≠ Normal
# 5. A keyword match is ONLY a candidate
# 6. Candidate status must be reviewed from the source field
# 7. CXR + genomic pairing requires the same condition_id
# 8. No artificial Normal labels will be created
# 9. No DICOM files will be modified
# 10. No preprocessing/modeling will be performed
#
# ================================================================


from pathlib import Path
import pandas as pd
import numpy as np
import json
import re
import warnings

warnings.filterwarnings("ignore")


# ================================================================
# 1. PATHS
# ================================================================

BASE_DIR = Path(r"C:\TBP")
METADATA_DIR = BASE_DIR / "Metadata"


CXR_PATH = (
    METADATA_DIR /
    "TB_Portals_CXRs_March_2025.csv"
)

GENOMICS_PATH = (
    METADATA_DIR /
    "TB_Portals_Genomics_March_2025.csv"
)


PREVIOUS_STEP_47_DIR = (
    METADATA_DIR /
    "Step_3B_6_47_External_Normal_Source_Audit"
)


PREVIOUS_STEP_48_DIR = (
    METADATA_DIR /
    "Step_3B_6_48_External_Source_Evidence_Decision_Audit"
)


OUTPUT_DIR = (
    METADATA_DIR /
    "Step_3B_6_49_TB_Portals_Internal_Normal_Discovery"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# 2. START
# ================================================================

print("=" * 80)
print("STEP 3B.6.49 — TB PORTALS INTERNAL NORMAL-CASE DISCOVERY")
print("=" * 80)

print()
print("IMPORTANT:")
print("This audit uses ONLY TB Portals March 2025 local metadata.")
print("No external datasets are used.")
print()


# ================================================================
# 3. INPUT VALIDATION
# ================================================================

print("1. INPUT FILE VALIDATION")
print("-" * 80)


required_inputs = {
    "TB Portals CXR metadata":
        CXR_PATH,

    "TB Portals genomic metadata":
        GENOMICS_PATH
}


for name, path in required_inputs.items():

    if not path.exists():

        raise FileNotFoundError(
            f"\nRequired input file not found:\n"
            f"{name}\n"
            f"{path}"
        )

    print(
        f"{name:<35}: FOUND"
    )


# ================================================================
# 4. LOAD DATA
# ================================================================

print()
print("2. LOADING TB PORTALS DATA")
print("-" * 80)


cxr = pd.read_csv(
    CXR_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)


print(
    f"CXR records                  : "
    f"{len(cxr):,}"
)

print(
    f"Genomic records              : "
    f"{len(genomics):,}"
)


# ================================================================
# 5. BASIC SCHEMA VALIDATION
# ================================================================

print()
print("3. SCHEMA VALIDATION")
print("-" * 80)


if "condition_id" not in cxr.columns:

    raise ValueError(
        "TB Portals CXR metadata does not contain condition_id."
    )


if "condition_id" not in genomics.columns:

    raise ValueError(
        "TB Portals genomic metadata does not contain condition_id."
    )


cxr["condition_id"] = (
    cxr["condition_id"]
    .astype("string")
    .str.strip()
)

genomics["condition_id"] = (
    genomics["condition_id"]
    .astype("string")
    .str.strip()
)


print(
    f"CXR columns                  : "
    f"{len(cxr.columns)}"
)

print(
    f"Genomic columns              : "
    f"{len(genomics.columns)}"
)

print(
    "condition_id in CXR          : PASS"
)

print(
    "condition_id in genomics     : PASS"
)


# ================================================================
# 6. RECORD/CASE COUNTS
# ================================================================

print()
print("4. TB PORTALS DATASET STRUCTURE")
print("-" * 80)


cxr_unique_conditions = (
    cxr["condition_id"]
    .dropna()
    .nunique()
)

genomic_unique_conditions = (
    genomics["condition_id"]
    .dropna()
    .nunique()
)

common_conditions = (
    set(
        cxr["condition_id"]
        .dropna()
        .unique()
    )
    &
    set(
        genomics["condition_id"]
        .dropna()
        .unique()
    )
)


print(
    f"CXR records                  : "
    f"{len(cxr):,}"
)

print(
    f"CXR unique conditions        : "
    f"{cxr_unique_conditions:,}"
)

print(
    f"Genomic records              : "
    f"{len(genomics):,}"
)

print(
    f"Genomic unique conditions    : "
    f"{genomic_unique_conditions:,}"
)

print(
    f"Common CXR/genomic cases     : "
    f"{len(common_conditions):,}"
)


# ================================================================
# 7. IDENTIFY TEXT-LIKE COLUMNS
# ================================================================

print()
print("5. IDENTIFYING SEARCHABLE METADATA COLUMNS")
print("-" * 80)


def searchable_columns(
    dataframe
):

    cols = []

    for column in dataframe.columns:

        dtype = dataframe[column].dtype

        if (
            pd.api.types.is_object_dtype(dtype)
            or
            pd.api.types.is_string_dtype(dtype)
            or
            pd.api.types.is_categorical_dtype(dtype)
        ):

            cols.append(column)

    return cols


cxr_text_columns = searchable_columns(
    cxr
)

genomic_text_columns = searchable_columns(
    genomics
)


print(
    f"CXR text-like columns       : "
    f"{len(cxr_text_columns)}"
)

print(
    f"Genomic text-like columns   : "
    f"{len(genomic_text_columns)}"
)


print()
print("CXR searchable columns:")
print(
    cxr_text_columns
)

print()
print("Genomic searchable columns:")
print(
    genomic_text_columns
)


# ================================================================
# 8. DEFINE EXPLICIT NORMAL/CONTROL VOCABULARY
# ================================================================
#
# IMPORTANT:
#
# These are candidate terms only.
#
# We do NOT automatically assign the Normal class from a match.
#
# ================================================================

normal_patterns = {

    "NORMAL":
        r"\bnormal\b",

    "HEALTHY":
        r"\bhealthy\b",

    "CONTROL":
        r"\bcontrols?\b",

    "TB_NEGATIVE":
        r"\btb[\s_-]*negative\b",

    "TB_NEG":
        r"\btb[\s_-]*neg\b",

    "NON_TB":
        r"\bnon[\s_-]*tb\b",

    "NO_TB":
        r"\bno[\s_-]*tb\b",

    "NO_EVIDENCE_TB":
        r"\bno[\s_-]+evidence[\s_-]+of[\s_-]+tb\b",

    "TB_ABSENT":
        r"\btb[\s_-]*absent\b",

    "NEGATIVE_FOR_TB":
        r"\bnegative[\s_-]+for[\s_-]+tb\b",

    "TB_FREE":
        r"\btb[\s_-]*free\b",

    "NON_TUBERCULOUS":
        r"\bnon[\s_-]*tuberculous\b"
}


compiled_patterns = {
    label: re.compile(
        pattern,
        flags=re.IGNORECASE
    )
    for label, pattern in normal_patterns.items()
}


# ================================================================
# 9. SEARCH ALL CXR TEXT METADATA
# ================================================================

print()
print("6. SEARCHING ALL CXR METADATA FOR NORMAL/CONTROL TERMS")
print("-" * 80)


cxr_candidates = []


for column in cxr_text_columns:

    series = (
        cxr[column]
        .fillna("")
        .astype(str)
    )

    for term_label, pattern in compiled_patterns.items():

        mask = series.str.contains(
            pattern,
            regex=True,
            na=False
        )

        if not mask.any():

            continue

        matched_rows = cxr.loc[
            mask,
            [
                "condition_id"
            ]
        ].copy()

        matched_rows[
            "source_dataset"
        ] = "CXR"

        matched_rows[
            "source_column"
        ] = column

        matched_rows[
            "matched_term"
        ] = term_label

        matched_rows[
            "source_value"
        ] = (
            cxr.loc[
                mask,
                column
            ]
            .astype(str)
            .values
        )

        cxr_candidates.append(
            matched_rows
        )


if cxr_candidates:

    cxr_normal_candidates = pd.concat(
        cxr_candidates,
        ignore_index=True
    )

else:

    cxr_normal_candidates = pd.DataFrame(
        columns=[
            "condition_id",
            "source_dataset",
            "source_column",
            "matched_term",
            "source_value"
        ]
    )


print(
    f"Raw CXR keyword matches      : "
    f"{len(cxr_normal_candidates):,}"
)

print(
    f"Unique candidate conditions  : "
    f"{cxr_normal_candidates['condition_id'].nunique():,}"
)


# ================================================================
# 10. SEARCH ALL GENOMIC METADATA
# ================================================================

print()
print("7. SEARCHING ALL GENOMIC METADATA FOR NORMAL/CONTROL TERMS")
print("-" * 80)


genomic_candidates = []


for column in genomic_text_columns:

    series = (
        genomics[column]
        .fillna("")
        .astype(str)
    )

    for term_label, pattern in compiled_patterns.items():

        mask = series.str.contains(
            pattern,
            regex=True,
            na=False
        )

        if not mask.any():

            continue

        matched_rows = genomics.loc[
            mask,
            [
                "condition_id"
            ]
        ].copy()

        matched_rows[
            "source_dataset"
        ] = "GENOMICS"

        matched_rows[
            "source_column"
        ] = column

        matched_rows[
            "matched_term"
        ] = term_label

        matched_rows[
            "source_value"
        ] = (
            genomics.loc[
                mask,
                column
            ]
            .astype(str)
            .values
        )

        genomic_candidates.append(
            matched_rows
        )


if genomic_candidates:

    genomic_normal_candidates = pd.concat(
        genomic_candidates,
        ignore_index=True
    )

else:

    genomic_normal_candidates = pd.DataFrame(
        columns=[
            "condition_id",
            "source_dataset",
            "source_column",
            "matched_term",
            "source_value"
        ]
    )


print(
    f"Raw genomic keyword matches  : "
    f"{len(genomic_normal_candidates):,}"
)

print(
    f"Unique candidate conditions  : "
    f"{genomic_normal_candidates['condition_id'].nunique():,}"
)


# ================================================================
# 11. SAVE RAW CANDIDATE EVIDENCE
# ================================================================

print()
print("8. SAVING RAW CANDIDATE EVIDENCE")
print("-" * 80)


cxr_raw_path = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_49_CXR_Normal_Keyword_Candidates.csv"
)

genomic_raw_path = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_49_Genomics_Normal_Keyword_Candidates.csv"
)


cxr_normal_candidates.to_csv(
    cxr_raw_path,
    index=False,
    encoding="utf-8-sig"
)

genomic_normal_candidates.to_csv(
    genomic_raw_path,
    index=False,
    encoding="utf-8-sig"
)


print(
    f"{cxr_raw_path.name:<75}: PASS"
)

print(
    f"{genomic_raw_path.name:<75}: PASS"
)


# ================================================================
# 12. SHOW UNIQUE CXR CANDIDATE CONDITIONS
# ================================================================

print()
print("9. UNIQUE CXR NORMAL/CONTROL CANDIDATES")
print("-" * 80)


if len(cxr_normal_candidates) > 0:

    cxr_candidate_summary = (
        cxr_normal_candidates
        .groupby(
            "condition_id",
            dropna=False
        )
        .agg(
            matched_terms=(
                "matched_term",
                lambda x:
                "|".join(
                    sorted(
                        set(
                            x.astype(str)
                        )
                    )
                )
            ),

            source_columns=(
                "source_column",
                lambda x:
                "|".join(
                    sorted(
                        set(
                            x.astype(str)
                        )
                    )
                )
            ),

            source_values=(
                "source_value",
                lambda x:
                " || ".join(
                    sorted(
                        set(
                            x.astype(str)
                        )
                    )
                )
            ),

            keyword_match_count=(
                "matched_term",
                "size"
            )
        )
        .reset_index()
    )

else:

    cxr_candidate_summary = pd.DataFrame(
        columns=[
            "condition_id",
            "matched_terms",
            "source_columns",
            "source_values",
            "keyword_match_count"
        ]
    )


print(
    f"Candidate CXR conditions    : "
    f"{len(cxr_candidate_summary):,}"
)


if len(cxr_candidate_summary) > 0:

    print(
        cxr_candidate_summary.to_string(
            index=False
        )
    )

else:

    print(
        "NO EXPLICIT NORMAL/CONTROL CXR "
        "KEYWORD CANDIDATES FOUND."
    )


# ================================================================
# 13. UNIQUE GENOMIC CANDIDATE CONDITIONS
# ================================================================

print()
print("10. UNIQUE GENOMIC NORMAL/CONTROL CANDIDATES")
print("-" * 80)


if len(genomic_normal_candidates) > 0:

    genomic_candidate_summary = (
        genomic_normal_candidates
        .groupby(
            "condition_id",
            dropna=False
        )
        .agg(
            matched_terms=(
                "matched_term",
                lambda x:
                "|".join(
                    sorted(
                        set(
                            x.astype(str)
                        )
                    )
                )
            ),

            source_columns=(
                "source_column",
                lambda x:
                "|".join(
                    sorted(
                        set(
                            x.astype(str)
                        )
                    )
                )
            ),

            source_values=(
                "source_value",
                lambda x:
                " || ".join(
                    sorted(
                        set(
                            x.astype(str)
                        )
                    )
                )
            ),

            keyword_match_count=(
                "matched_term",
                "size"
            )
        )
        .reset_index()
    )

else:

    genomic_candidate_summary = pd.DataFrame(
        columns=[
            "condition_id",
            "matched_terms",
            "source_columns",
            "source_values",
            "keyword_match_count"
        ]
    )


print(
    f"Candidate genomic conditions: "
    f"{len(genomic_candidate_summary):,}"
)


if len(genomic_candidate_summary) > 0:

    print(
        genomic_candidate_summary.to_string(
            index=False
        )
    )

else:

    print(
        "NO EXPLICIT NORMAL/CONTROL GENOMIC "
        "KEYWORD CANDIDATES FOUND."
    )


# ================================================================
# 14. CXR + GENOMIC OVERLAP
# ================================================================

print()
print("11. SAME-CONDITION CXR + GENOMIC NORMAL CANDIDATES")
print("-" * 80)


cxr_candidate_ids = set(
    cxr_candidate_summary[
        "condition_id"
    ]
    .dropna()
    .astype(str)
)

genomic_candidate_ids = set(
    genomic_candidate_summary[
        "condition_id"
    ]
    .dropna()
    .astype(str)
)


paired_normal_candidate_ids = (
    cxr_candidate_ids
    &
    genomic_candidate_ids
)


print(
    f"CXR candidate conditions    : "
    f"{len(cxr_candidate_ids):,}"
)

print(
    f"Genomic candidate conditions: "
    f"{len(genomic_candidate_ids):,}"
)

print(
    f"Same-condition overlap      : "
    f"{len(paired_normal_candidate_ids):,}"
)


# ================================================================
# 15. BUILD PAIRED CANDIDATE TABLE
# ================================================================

paired_candidate_rows = []


for condition_id in sorted(
    paired_normal_candidate_ids
):

    cxr_rows = cxr_normal_candidates[
        cxr_normal_candidates[
            "condition_id"
        ]
        .astype(str)
        == condition_id
    ]

    genomic_rows = genomic_normal_candidates[
        genomic_normal_candidates[
            "condition_id"
        ]
        .astype(str)
        == condition_id
    ]

    paired_candidate_rows.append({

        "condition_id":
            condition_id,

        "CXR_keyword_evidence":
            " || ".join(
                sorted(
                    set(
                        cxr_rows[
                            "source_value"
                        ]
                        .astype(str)
                    )
                )
            ),

        "CXR_source_columns":
            "|".join(
                sorted(
                    set(
                        cxr_rows[
                            "source_column"
                        ]
                        .astype(str)
                    )
                )
            ),

        "CXR_matched_terms":
            "|".join(
                sorted(
                    set(
                        cxr_rows[
                            "matched_term"
                        ]
                        .astype(str)
                    )
                )
            ),

        "genomic_keyword_evidence":
            " || ".join(
                sorted(
                    set(
                        genomic_rows[
                            "source_value"
                        ]
                        .astype(str)
                    )
                )
            ),

        "genomic_source_columns":
            "|".join(
                sorted(
                    set(
                        genomic_rows[
                            "source_column"
                        ]
                        .astype(str)
                    )
                )
            ),

        "genomic_matched_terms":
            "|".join(
                sorted(
                    set(
                        genomic_rows[
                            "matched_term"
                        ]
                        .astype(str)
                    )
                )
            ),

        "candidate_status":
            "KEYWORD_CANDIDATE_REQUIRES_SOURCE_REVIEW"
    })


paired_candidate_summary = pd.DataFrame(
    paired_candidate_rows
)


if len(paired_candidate_summary) > 0:

    print(
        paired_candidate_summary.to_string(
            index=False
        )
    )

else:

    print(
        "NO SAME-CONDITION CXR + GENOMIC "
        "NORMAL KEYWORD CANDIDATES FOUND."
    )


# ================================================================
# 16. IMPORTANT NEGATIVE-CONTROL TEST
# ================================================================
#
# We explicitly test whether common "absence" conditions are
# accidentally being interpreted as Normal.
#
# ================================================================

print()
print("12. NEGATIVE-CONTROL / NON-INFERENCE CHECK")
print("-" * 80)


inference_rules = pd.DataFrame([

    {
        "rule_id":
            "NEG-01",

        "rule":
            "Missing genomic record means Normal",

        "allowed":
            False,

        "status":
            "PROHIBITED"
    },

    {
        "rule_id":
            "NEG-02",

        "rule":
            "Missing resistance information means Normal",

        "allowed":
            False,

        "status":
            "PROHIBITED"
    },

    {
        "rule_id":
            "NEG-03",

        "rule":
            "Unknown outcome means Normal",

        "allowed":
            False,

        "status":
            "PROHIBITED"
    },

    {
        "rule_id":
            "NEG-04",

        "rule":
            "No keyword match means Normal",

        "allowed":
            False,

        "status":
            "PROHIBITED"
    },

    {
        "rule_id":
            "NEG-05",

        "rule":
            "Keyword match alone proves Normal diagnosis",

        "allowed":
            False,

        "status":
            "PROHIBITED"
    },

    {
        "rule_id":
            "NEG-06",

        "rule":
            "Normal CXR from another dataset can be paired "
            "with TB Portals genomics",

        "allowed":
            False,

        "status":
            "PROHIBITED"
    }
])


print(
    inference_rules.to_string(
        index=False
    )
)


# ================================================================
# 17. SEARCH RESULT CLASSIFICATION
# ================================================================

print()
print("13. INTERNAL TB PORTALS NORMAL-CASE RESULT")
print("-" * 80)


if len(cxr_candidate_ids) == 0:

    cxr_result = (
        "NO_EXPLICIT_NORMAL_KEYWORD_CANDIDATES"
    )

else:

    cxr_result = (
        "EXPLICIT_NORMAL_KEYWORD_CANDIDATES_FOUND"
    )


if len(genomic_candidate_ids) == 0:

    genomic_result = (
        "NO_EXPLICIT_NORMAL_KEYWORD_CANDIDATES"
    )

else:

    genomic_result = (
        "EXPLICIT_NORMAL_KEYWORD_CANDIDATES_FOUND"
    )


if len(paired_normal_candidate_ids) == 0:

    paired_result = (
        "NO_SAME_CONDITION_NORMAL_CXR_GENOMIC_CANDIDATES"
    )

else:

    paired_result = (
        "SAME_CONDITION_NORMAL_CXR_GENOMIC_CANDIDATES_FOUND"
    )


print(
    f"CXR result                  : "
    f"{cxr_result}"
)

print(
    f"Genomic result              : "
    f"{genomic_result}"
)

print(
    f"CXR + genomic result        : "
    f"{paired_result}"
)


# ================================================================
# 18. FORMAL DECISION LOGIC
# ================================================================
#
# IMPORTANT:
#
# A keyword candidate is NOT automatically a verified Normal case.
#
# Therefore:
#
# If candidates exist:
#     REVIEW_REQUIRED
#
# If no candidates exist:
#     NOT_FOUND_IN_SEARCHABLE_METADATA
#
# ================================================================


if len(paired_normal_candidate_ids) > 0:

    final_internal_status = (
        "REVIEW_REQUIRED_SAME_CONDITION_NORMAL_CANDIDATES_FOUND"
    )

elif (
    len(cxr_candidate_ids) > 0
    or
    len(genomic_candidate_ids) > 0
):

    final_internal_status = (
        "NORMAL_KEYWORD_CANDIDATES_FOUND_BUT_NO_PAIRED_NORMAL_COHORT"
    )

else:

    final_internal_status = (
        "NO_EXPLICIT_NORMAL_CANDIDATES_FOUND_IN_TB_PORTALS_METADATA"
    )


print()
print(
    "FORMAL INTERNAL STATUS:"
)

print(
    final_internal_status
)


# ================================================================
# 19. COHORT STATUS
# ================================================================

if len(paired_normal_candidate_ids) > 0:

    final_cohort_status = (
        "NOT_LOCKED_REQUIRES_HUMAN_SOURCE_REVIEW"
    )

else:

    final_cohort_status = (
        "NOT_LOCKED"
    )


# ================================================================
# 20. SAVE CANDIDATE SUMMARY
# ================================================================

print()
print("14. SAVING OUTPUTS")
print("-" * 80)


cxr_summary_path = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_49_CXR_Normal_Candidate_Summary.csv"
)

genomic_summary_path = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_49_Genomics_Normal_Candidate_Summary.csv"
)

paired_summary_path = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_49_Paired_Normal_Candidate_Summary.csv"
)

inference_path = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_49_Non_Inference_Rules.csv"
)


cxr_candidate_summary.to_csv(
    cxr_summary_path,
    index=False,
    encoding="utf-8-sig"
)

genomic_candidate_summary.to_csv(
    genomic_summary_path,
    index=False,
    encoding="utf-8-sig"
)

paired_candidate_summary.to_csv(
    paired_summary_path,
    index=False,
    encoding="utf-8-sig"
)

inference_rules.to_csv(
    inference_path,
    index=False,
    encoding="utf-8-sig"
)


print(
    f"{cxr_summary_path.name:<75}: PASS"
)

print(
    f"{genomic_summary_path.name:<75}: PASS"
)

print(
    f"{paired_summary_path.name:<75}: PASS"
)

print(
    f"{inference_path.name:<75}: PASS"
)


# ================================================================
# 21. CREATE COMPLETE DATASET AUDIT
# ================================================================

dataset_audit = pd.DataFrame([

    {
        "dataset":
            "TB Portals CXR March 2025",

        "records":
            len(cxr),

        "unique_conditions":
            cxr_unique_conditions,

        "explicit_normal_keyword_candidates":
            len(cxr_candidate_ids),

        "status":
            cxr_result
    },

    {
        "dataset":
            "TB Portals Genomics March 2025",

        "records":
            len(genomics),

        "unique_conditions":
            genomic_unique_conditions,

        "explicit_normal_keyword_candidates":
            len(genomic_candidate_ids),

        "status":
            genomic_result
    },

    {
        "dataset":
            "TB Portals CXR + Genomics",

        "records":
            np.nan,

        "unique_conditions":
            len(common_conditions),

        "explicit_normal_keyword_candidates":
            len(paired_normal_candidate_ids),

        "status":
            paired_result
    }
])


dataset_audit_path = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_49_Dataset_Audit_Summary.csv"
)


dataset_audit.to_csv(
    dataset_audit_path,
    index=False,
    encoding="utf-8-sig"
)


print(
    f"{dataset_audit_path.name:<75}: PASS"
)


# ================================================================
# 22. DECISION RECORD
# ================================================================

decision_record = {

    "step":
        "3B.6.49",

    "title":
        "TB Portals Internal Normal-Case Discovery Audit",

    "dataset_scope":
        "TB Portals March 2025 local data only",

    "external_datasets_used":
        False,

    "external_dataset_names":
        [],

    "cxr_records":
        int(len(cxr)),

    "cxr_unique_conditions":
        int(cxr_unique_conditions),

    "genomic_records":
        int(len(genomics)),

    "genomic_unique_conditions":
        int(genomic_unique_conditions),

    "common_cxr_genomic_conditions":
        int(len(common_conditions)),

    "cxr_keyword_candidate_conditions":
        int(len(cxr_candidate_ids)),

    "genomic_keyword_candidate_conditions":
        int(len(genomic_candidate_ids)),

    "same_condition_cxr_genomic_keyword_candidates":
        int(len(paired_normal_candidate_ids)),

    "internal_cxr_result":
        cxr_result,

    "internal_genomic_result":
        genomic_result,

    "internal_paired_result":
        paired_result,

    "scientific_rule":
        "Keyword matches are candidates only and require source review.",

    "normal_from_missing_data":
        False,

    "normal_from_missing_genomics":
        False,

    "artificial_normal_labels":
        False,

    "cross_dataset_pairing":
        False,

    "final_internal_status":
        final_internal_status,

    "final_cohort_status":
        final_cohort_status,

    "preprocessing_started":
        False,

    "lung_segmentation_started":
        False,

    "model_training_started":
        False,

    "xai_started":
        False,

    "raw_dicom_modified":
        False,

    "source_metadata_modified":
        False
}


decision_path = (
    OUTPUT_DIR /
    "TB_Portals_March2025_Step_3B_6_49_Decision_Record.json"
)


with open(
    decision_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        decision_record,
        file,
        indent=4
    )


print(
    f"{decision_path.name:<75}: PASS"
)


# ================================================================
# 23. FINAL VALIDATION
# ================================================================

print()
print("15. FINAL VALIDATION")
print("-" * 80)


validation = {

    "CXR records = 16,723":
        len(cxr) == 16723,

    "CXR unique conditions = 13,777":
        cxr_unique_conditions == 13777,

    "Genomic records = 4,467":
        len(genomics) == 4467,

    "Genomic unique conditions = 4,155":
        genomic_unique_conditions == 4155,

    "condition_id available":
        (
            "condition_id" in cxr.columns
            and
            "condition_id" in genomics.columns
        ),

    "No external dataset used":
        True,

    "No artificial Normal labels":
        True,

    "Missing genomics not treated as Normal":
        True,

    "No cross-dataset pairing":
        True,

    "No preprocessing":
        True,

    "No segmentation":
        True,

    "No model training":
        True,

    "No XAI":
        True,

    "Raw DICOM unchanged":
        True,

    "Source metadata unchanged":
        True
}


all_pass = True


for check, result in validation.items():

    print(
        f"{check:<55}: "
        f"{'PASS' if result else 'FAIL'}"
    )

    if not result:

        all_pass = False


# ================================================================
# 24. FINAL STATUS
# ================================================================

print()
print("=" * 80)
print("STEP 3B.6.49 — FINAL STATUS")
print("=" * 80)

print(
    f"TB Portals CXR records          : "
    f"{len(cxr):,}"
)

print(
    f"TB Portals CXR conditions       : "
    f"{cxr_unique_conditions:,}"
)

print(
    f"TB Portals genomic records      : "
    f"{len(genomics):,}"
)

print(
    f"TB Portals genomic conditions   : "
    f"{genomic_unique_conditions:,}"
)

print(
    f"CXR + genomic common conditions : "
    f"{len(common_conditions):,}"
)

print(
    f"CXR Normal candidates           : "
    f"{len(cxr_candidate_ids):,}"
)

print(
    f"Genomic Normal candidates       : "
    f"{len(genomic_candidate_ids):,}"
)

print(
    f"Paired Normal candidates        : "
    f"{len(paired_normal_candidate_ids):,}"
)

print(
    f"Final internal status            : "
    f"{final_internal_status}"
)

print(
    f"Final cohort                     : "
    f"{final_cohort_status}"
)

print(
    f"Technical validation             : "
    f"{'PASS' if all_pass else 'FAIL'}"
)

print()
print("EXTERNAL DATASETS USED           : NO")
print("DICOM PREPROCESSING              : NOT STARTED")
print("LUNG SEGMENTATION                : NOT STARTED")
print("MULTIMODAL MODEL                 : NOT STARTED")
print("EXPLAINABLE AI                   : NOT STARTED")
print("RAW DICOM MODIFIED               : NO")
print("SOURCE METADATA MODIFIED         : NO")

print("=" * 80)
print("STEP 3B.6.49 COMPLETE")
print("=" * 80)
```

```python
# =============================================================================
# STEP 3B.6.50 — CXR NORMAL CANDIDATE PROVENANCE AUDIT
# =============================================================================
#
# PURPOSE
# -------
# Investigate the 10,842 CXR condition-level Normal/Healthy/Control keyword
# candidates identified in Step 3B.6.49.
#
# IMPORTANT SCIENTIFIC RULES
# --------------------------
# 1. TB Portals March 2025 data ONLY.
# 2. No Montgomery dataset.
# 3. No Shenzhen dataset.
# 4. No external dataset.
# 5. Do NOT create Normal labels automatically.
# 6. A keyword match is ONLY a candidate.
# 7. Do NOT treat missing genomics as Normal.
# 8. Do NOT treat absence of TB as Normal.
# 9. Do NOT pair unrelated subjects.
# 10. Do NOT modify source metadata.
#
# MAIN QUESTION
# -------------
# Do the 10,842 CXR keyword candidates contain a documented patient-level
# Normal/Healthy/Control diagnosis or case definition that could support
# a valid Normal class?
#
# OUTPUTS
# -------
# 01_Match_Level_Detail.csv
# 02_Candidate_Condition_Summary.csv
# 03_Source_Column_Summary.csv
# 04_Matched_Term_Summary.csv
# 05_Candidate_Overlap_With_Genomics.csv
# 06_Candidate_Case_Definition_Summary.csv
# 07_Candidate_Diagnosis_Summary.csv
# 08_Candidate_Outcome_Summary.csv
# 09_Paired_Normal_Candidate_Audit.csv
# 10_Final_Decision_Evidence.csv
# 11_Step_3B_6_50_Decision_Record.json
# 12_Step_3B_6_50_Audit_Summary.csv
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re
import json
from datetime import datetime

# =============================================================================
# 1. CONFIGURATION
# =============================================================================

CXR_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_CXRs_March_2025.csv"
)

GENOMICS_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_Genomics_March_2025.csv"
)

PREVIOUS_OUTPUT_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_49_TB_Normal_Internal_Audit"
)

OUTPUT_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_50_CXR_Normal_Candidate_Provenance_Audit"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# =============================================================================
# 2. REQUIRED COLUMN
# =============================================================================

REQUIRED_CXR_COLUMNS = {
    "condition_id"
}

REQUIRED_GENOMIC_COLUMNS = {
    "condition_id"
}

# =============================================================================
# 3. NORMAL / CONTROL VOCABULARY
# =============================================================================
#
# These are candidate-search terms only.
# They DO NOT constitute a diagnosis.
#
# =============================================================================

NORMAL_TERMS = [
    "NORMAL",
    "HEALTHY",
    "CONTROL",
    "TB_NEGATIVE",
    "TB_NEG",
    "NON_TB",
    "NO_TB",
    "NO_EVIDENCE_TB",
    "TB_ABSENT",
    "NEGATIVE_FOR_TB",
    "TB_FREE",
    "NON_TUBERCULOUS"
]

# Longer/specific phrases should be checked before shorter terms.
NORMAL_TERMS_SORTED = sorted(
    NORMAL_TERMS,
    key=len,
    reverse=True
)

# =============================================================================
# 4. HELPER FUNCTIONS
# =============================================================================

def normalize_text(value):
    """
    Convert a value to normalized uppercase text for candidate searching.
    Does NOT modify the original dataframe.
    """
    if pd.isna(value):
        return ""

    text = str(value).strip().upper()

    # Normalize separators for robust matching.
    text = text.replace("-", "_")
    text = text.replace("/", "_")

    # Collapse repeated whitespace.
    text = re.sub(r"\s+", " ", text)

    return text


def find_matching_terms(value):
    """
    Return all candidate Normal/Healthy/Control terms found in a value.
    Keyword detection only; no diagnostic interpretation.
    """
    text = normalize_text(value)

    if not text:
        return []

    matches = []

    for term in NORMAL_TERMS_SORTED:

        # Convert underscore terminology into flexible separators.
        escaped_parts = [
            re.escape(part)
            for part in term.split("_")
        ]

        pattern = r"[\s_\-/]+".join(escaped_parts)

        try:
            if re.search(pattern, text):
                matches.append(term)
        except re.error:
            continue

    return matches


def is_text_like_series(series):
    """
    Determine whether a column is reasonable to inspect as text.

    Object/string/category columns are included.
    Numeric columns are excluded from the broad semantic keyword search.
    """
    return (
        pd.api.types.is_object_dtype(series)
        or pd.api.types.is_string_dtype(series)
        or isinstance(series.dtype, pd.CategoricalDtype)
    )


# =============================================================================
# 5. LOAD SOURCE DATA
# =============================================================================

print("=" * 80)
print("STEP 3B.6.50 — CXR NORMAL CANDIDATE PROVENANCE AUDIT")
print("=" * 80)

print("\n1. LOADING SOURCE DATA")
print("-" * 80)

if not CXR_PATH.exists():
    raise FileNotFoundError(
        f"CXR source file not found:\n{CXR_PATH}"
    )

if not GENOMICS_PATH.exists():
    raise FileNotFoundError(
        f"Genomics source file not found:\n{GENOMICS_PATH}"
    )

cxr = pd.read_csv(CXR_PATH, low_memory=False)
genomics = pd.read_csv(GENOMICS_PATH, low_memory=False)

print(f"CXR records      : {len(cxr):,}")
print(f"CXR columns      : {len(cxr.columns):,}")
print(f"Genomic records  : {len(genomics):,}")
print(f"Genomic columns  : {len(genomics.columns):,}")

# =============================================================================
# 6. SCHEMA VALIDATION
# =============================================================================

print("\n2. SCHEMA VALIDATION")
print("-" * 80)

missing_cxr = REQUIRED_CXR_COLUMNS - set(cxr.columns)
missing_genomic = REQUIRED_GENOMIC_COLUMNS - set(genomics.columns)

if missing_cxr:
    raise ValueError(
        f"Missing required CXR columns: {sorted(missing_cxr)}"
    )

if missing_genomic:
    raise ValueError(
        f"Missing required genomic columns: {sorted(missing_genomic)}"
    )

print("Required CXR columns       : PASS")
print("Required genomic columns   : PASS")

# =============================================================================
# 7. BASIC IDENTIFIER VALIDATION
# =============================================================================

cxr["condition_id"] = cxr["condition_id"].astype(str).str.strip()
genomics["condition_id"] = genomics["condition_id"].astype(str).str.strip()

cxr_unique_conditions = set(
    cxr.loc[cxr["condition_id"] != "", "condition_id"].unique()
)

genomic_unique_conditions = set(
    genomics.loc[
        genomics["condition_id"] != "",
        "condition_id"
    ].unique()
)

common_conditions = (
    cxr_unique_conditions &
    genomic_unique_conditions
)

print("\n3. CONDITION STRUCTURE")
print("-" * 80)

print(f"CXR unique conditions       : {len(cxr_unique_conditions):,}")
print(f"Genomic unique conditions   : {len(genomic_unique_conditions):,}")
print(f"Common conditions           : {len(common_conditions):,}")

# =============================================================================
# 8. IDENTIFY TEXT-LIKE CXR COLUMNS
# =============================================================================

print("\n4. CXR TEXT-LIKE COLUMNS")
print("-" * 80)

cxr_text_columns = [
    col for col in cxr.columns
    if is_text_like_series(cxr[col])
]

print(f"Text-like CXR columns: {len(cxr_text_columns):,}")

for col in cxr_text_columns:
    print(f"  - {col}")

# =============================================================================
# 9. CANDIDATE MATCH EXTRACTION
# =============================================================================
#
# IMPORTANT:
# We retain:
#   condition_id
#   source_column
#   actual_value
#   matched_term
#   CXR row index
#
# This allows us to determine EXACTLY why each condition was identified.
#
# =============================================================================

print("\n5. EXTRACTING CXR CANDIDATE MATCHES")
print("-" * 80)

match_rows = []

for col in cxr_text_columns:

    series = cxr[col]

    for row_index, value in series.items():

        matches = find_matching_terms(value)

        if not matches:
            continue

        condition_id = cxr.at[row_index, "condition_id"]

        if not condition_id:
            continue

        for matched_term in matches:

            match_rows.append({
                "cxr_row_index": int(row_index),
                "condition_id": condition_id,
                "source_column": col,
                "matched_term": matched_term,
                "actual_value": value
            })

matches_df = pd.DataFrame(match_rows)

if matches_df.empty:
    raise RuntimeError(
        "No CXR keyword matches were found in Step 3B.6.50."
    )

# Remove exact duplicate evidence rows.
matches_df = matches_df.drop_duplicates(
    subset=[
        "cxr_row_index",
        "condition_id",
        "source_column",
        "matched_term",
        "actual_value"
    ]
).reset_index(drop=True)

candidate_conditions = set(
    matches_df["condition_id"].unique()
)

print(f"Keyword evidence rows       : {len(matches_df):,}")
print(f"Candidate CXR conditions    : {len(candidate_conditions):,}")

# =============================================================================
# 10. MATCH SOURCE COLUMN SUMMARY
# =============================================================================

print("\n6. SOURCE COLUMN SUMMARY")
print("-" * 80)

source_column_summary = (
    matches_df
    .groupby("source_column", dropna=False)
    .agg(
        evidence_rows=("condition_id", "size"),
        unique_conditions=("condition_id", "nunique"),
        unique_terms=("matched_term", "nunique")
    )
    .reset_index()
    .sort_values(
        ["unique_conditions", "evidence_rows"],
        ascending=False
    )
)

print(source_column_summary.to_string(index=False))

# =============================================================================
# 11. MATCHED TERM SUMMARY
# =============================================================================

print("\n7. MATCHED TERM SUMMARY")
print("-" * 80)

matched_term_summary = (
    matches_df
    .groupby("matched_term", dropna=False)
    .agg(
        evidence_rows=("condition_id", "size"),
        unique_conditions=("condition_id", "nunique"),
        source_columns=("source_column", "nunique")
    )
    .reset_index()
    .sort_values(
        ["unique_conditions", "evidence_rows"],
        ascending=False
    )
)

print(matched_term_summary.to_string(index=False))

# =============================================================================
# 12. BUILD CONDITION-LEVEL CANDIDATE SUMMARY
# =============================================================================

print("\n8. BUILDING CONDITION-LEVEL CANDIDATE SUMMARY")
print("-" * 80)

candidate_condition_summary = (
    matches_df
    .groupby("condition_id", dropna=False)
    .agg(
        evidence_rows=("condition_id", "size"),
        matched_terms=(
            "matched_term",
            lambda x: "|".join(sorted(set(x)))
        ),
        source_columns=(
            "source_column",
            lambda x: "|".join(sorted(set(x)))
        )
    )
    .reset_index()
)

candidate_condition_summary[
    "candidate_in_common_cxr_genomic_cohort"
] = candidate_condition_summary["condition_id"].isin(
    common_conditions
)

candidate_condition_summary[
    "candidate_has_genomic_record"
] = candidate_condition_summary[
    "candidate_in_common_cxr_genomic_cohort"
]

# =============================================================================
# 13. JOIN CXR METADATA TO CANDIDATES
# =============================================================================

print("\n9. ATTACHING CXR METADATA")
print("-" * 80)

# Keep condition-level metadata without pretending that every field is
# diagnostic. For repeated CXR records, use a deterministic first observed
# value only for audit display.

candidate_cxr_rows = cxr[
    cxr["condition_id"].isin(candidate_conditions)
].copy()

candidate_cxr_rows_count = len(candidate_cxr_rows)

print(
    f"CXR records belonging to candidate conditions: "
    f"{candidate_cxr_rows_count:,}"
)

# =============================================================================
# 14. LIKELY DIAGNOSTIC / CASE-DEFINITION FIELDS
# =============================================================================
#
# We inspect columns whose names indicate that they may contain diagnosis,
# case-definition, findings, outcome, resistance, etc.
#
# This is NOT an automatic labeling step.
#
# =============================================================================

diagnostic_name_patterns = [
    "case",
    "definition",
    "diagnos",
    "finding",
    "condition",
    "disease",
    "status",
    "outcome",
    "result",
    "clinical",
    "impression",
    "interpret",
    "tb"
]

diagnostic_candidate_columns = []

for col in cxr.columns:

    col_lower = str(col).lower()

    if any(
        pattern in col_lower
        for pattern in diagnostic_name_patterns
    ):
        diagnostic_candidate_columns.append(col)

# Ensure condition_id remains available.
if "condition_id" not in diagnostic_candidate_columns:
    diagnostic_candidate_columns.insert(0, "condition_id")

diagnostic_candidate_columns = list(
    dict.fromkeys(diagnostic_candidate_columns)
)

print("\nPotentially informative metadata columns:")
for col in diagnostic_candidate_columns:
    print(f"  - {col}")

# =============================================================================
# 15. CASE-DEFINITION AUDIT
# =============================================================================

print("\n10. CASE-DEFINITION AUDIT")
print("-" * 80)

if "case_definition" in candidate_cxr_rows.columns:

    case_definition_summary = (
        candidate_cxr_rows[
            ["condition_id", "case_definition"]
        ]
        .drop_duplicates()
        .sort_values(
            ["case_definition", "condition_id"],
            na_position="last"
        )
    )

    print(
        case_definition_summary[
            "case_definition"
        ]
        .fillna("<MISSING>")
        .astype(str)
        .value_counts(dropna=False)
        .to_string()
    )

else:

    case_definition_summary = pd.DataFrame(
        columns=["condition_id", "case_definition"]
    )

    print("case_definition column not present.")

# =============================================================================
# 16. DIAGNOSIS CODE AUDIT
# =============================================================================

print("\n11. DIAGNOSIS CODE AUDIT")
print("-" * 80)

if "diagnosis_code" in candidate_cxr_rows.columns:

    diagnosis_summary = (
        candidate_cxr_rows[
            ["condition_id", "diagnosis_code"]
        ]
        .drop_duplicates()
    )

    print(
        diagnosis_summary[
            "diagnosis_code"
        ]
        .fillna("<MISSING>")
        .astype(str)
        .value_counts(dropna=False)
        .head(100)
        .to_string()
    )

else:

    diagnosis_summary = pd.DataFrame(
        columns=["condition_id", "diagnosis_code"]
    )

    print("diagnosis_code column not present.")

# =============================================================================
# 17. OUTCOME AUDIT
# =============================================================================

print("\n12. OUTCOME AUDIT")
print("-" * 80)

if "outcome" in candidate_cxr_rows.columns:

    outcome_summary = (
        candidate_cxr_rows[
            ["condition_id", "outcome"]
        ]
        .drop_duplicates()
    )

    print(
        outcome_summary[
            "outcome"
        ]
        .fillna("<MISSING>")
        .astype(str)
        .value_counts(dropna=False)
        .to_string()
    )

else:

    outcome_summary = pd.DataFrame(
        columns=["condition_id", "outcome"]
    )

    print("outcome column not present.")

# =============================================================================
# 18. RESISTANCE AUDIT
# =============================================================================

print("\n13. RESISTANCE AUDIT")
print("-" * 80)

if "type_of_resistance" in candidate_cxr_rows.columns:

    resistance_summary = (
        candidate_cxr_rows[
            ["condition_id", "type_of_resistance"]
        ]
        .drop_duplicates()
    )

    print(
        resistance_summary[
            "type_of_resistance"
        ]
        .fillna("<MISSING>")
        .astype(str)
        .value_counts(dropna=False)
        .to_string()
    )

else:

    resistance_summary = pd.DataFrame(
        columns=["condition_id", "type_of_resistance"]
    )

    print("type_of_resistance column not present.")

# =============================================================================
# 19. GENOMIC OVERLAP
# =============================================================================

print("\n14. CANDIDATE OVERLAP WITH GENOMICS")
print("-" * 80)

candidate_overlap = candidate_condition_summary[
    [
        "condition_id",
        "evidence_rows",
        "matched_terms",
        "source_columns",
        "candidate_in_common_cxr_genomic_cohort",
        "candidate_has_genomic_record"
    ]
].copy()

overlap_counts = (
    candidate_overlap[
        "candidate_has_genomic_record"
    ]
    .value_counts()
)

print(
    f"CXR candidate conditions : "
    f"{len(candidate_overlap):,}"
)

print(
    f"With genomic record       : "
    f"{int(overlap_counts.get(True, 0)):,}"
)

print(
    f"Without genomic record    : "
    f"{int(overlap_counts.get(False, 0)):,}"
)

# =============================================================================
# 20. GENOMIC DETAILS FOR CANDIDATE CONDITIONS
# =============================================================================

candidate_genomic_rows = genomics[
    genomics["condition_id"].isin(candidate_conditions)
].copy()

print(
    f"Genomic records belonging to CXR candidate conditions: "
    f"{len(candidate_genomic_rows):,}"
)

# =============================================================================
# 21. SEARCH GENOMIC RECORDS FOR DIAGNOSTIC CONTEXT
# =============================================================================
#
# We do NOT search genomic rows for Normal to create a label.
# We only determine whether candidate CXR conditions have genomic records
# and what broad metadata may accompany them.
#
# =============================================================================

genomic_text_columns = [
    col for col in genomics.columns
    if is_text_like_series(genomics[col])
]

print(
    f"Genomic text-like columns: {len(genomic_text_columns):,}"
)

# =============================================================================
# 22. BUILD CANDIDATE-LEVEL EVIDENCE TABLE
# =============================================================================

print("\n15. BUILDING FINAL EVIDENCE TABLE")
print("-" * 80)

candidate_metadata = (
    candidate_cxr_rows
    .sort_index()
    .groupby("condition_id", as_index=False)
    .first()
)

# Keep useful columns if present.
useful_metadata_columns = [
    "condition_id",
    "patient_id",
    "age_of_onset",
    "sex",
    "case_definition",
    "diagnosis_code",
    "type_of_resistance",
    "comorbidity",
    "country",
    "bmi",
    "outcome",
    "imagingstudy_id",
    "imaging_date",
    "series_modality_cd",
    "cxr_outlier"
]

useful_metadata_columns = [
    col for col in useful_metadata_columns
    if col in candidate_metadata.columns
]

candidate_metadata = candidate_metadata[
    useful_metadata_columns
].copy()

final_candidate_evidence = candidate_condition_summary.merge(
    candidate_metadata,
    on="condition_id",
    how="left"
)

# =============================================================================
# 23. DETERMINE WHETHER MATCH IS IN A POTENTIALLY DIAGNOSTIC FIELD
# =============================================================================

diagnostic_field_names = {
    "case_definition",
    "diagnosis_code"
}

final_candidate_evidence[
    "keyword_match_in_potential_diagnostic_field"
] = False

for col in ["case_definition", "diagnosis_code"]:

    if col in matches_df["source_column"].unique():

        affected_conditions = set(
            matches_df.loc[
                matches_df["source_column"] == col,
                "condition_id"
            ]
        )

        final_candidate_evidence[
            "keyword_match_in_potential_diagnostic_field"
        ] = (
            final_candidate_evidence["condition_id"]
            .isin(affected_conditions)
        )

# =============================================================================
# 24. EXACT DIAGNOSTIC-FIELD MATCH EVIDENCE
# =============================================================================

diagnostic_matches = matches_df[
    matches_df["source_column"].isin(
        diagnostic_field_names
    )
].copy()

if not diagnostic_matches.empty:

    diagnostic_evidence = (
        diagnostic_matches
        .groupby("condition_id")
        .agg(
            diagnostic_match_columns=(
                "source_column",
                lambda x: "|".join(sorted(set(x)))
            ),
            diagnostic_matched_terms=(
                "matched_term",
                lambda x: "|".join(sorted(set(x)))
            ),
            diagnostic_actual_values=(
                "actual_value",
                lambda x: " || ".join(
                    sorted(
                        set(
                            str(v)
                            for v in x
                            if not pd.isna(v)
                        )
                    )
                )
            )
        )
        .reset_index()
    )

else:

    diagnostic_evidence = pd.DataFrame(
        columns=[
            "condition_id",
            "diagnostic_match_columns",
            "diagnostic_matched_terms",
            "diagnostic_actual_values"
        ]
    )

final_candidate_evidence = final_candidate_evidence.merge(
    diagnostic_evidence,
    on="condition_id",
    how="left"
)

# =============================================================================
# 25. MULTIMODAL CANDIDATE AUDIT
# =============================================================================

print("\n16. MULTIMODAL CANDIDATE AUDIT")
print("-" * 80)

paired_candidate_conditions = set(
    candidate_conditions &
    common_conditions
)

print(
    f"CXR Normal keyword candidate conditions : "
    f"{len(candidate_conditions):,}"
)

print(
    f"Candidate conditions with genomics       : "
    f"{len(paired_candidate_conditions):,}"
)

paired_candidate_matches = matches_df[
    matches_df["condition_id"].isin(
        paired_candidate_conditions
    )
].copy()

# =============================================================================
# 26. PAIRED CANDIDATE DETAIL
# =============================================================================

if not paired_candidate_matches.empty:

    paired_normal_candidate_audit = (
        paired_candidate_matches
        .merge(
            candidate_metadata,
            on="condition_id",
            how="left"
        )
        .sort_values(
            [
                "condition_id",
                "source_column",
                "matched_term"
            ]
        )
    )

else:

    paired_normal_candidate_audit = pd.DataFrame(
        columns=list(matches_df.columns)
    )

print(
    f"Paired candidate evidence rows          : "
    f"{len(paired_normal_candidate_audit):,}"
)

# =============================================================================
# 27. CASE-LEVEL SUMMARY
# =============================================================================

print("\n17. CANDIDATE CONDITION SUMMARY")
print("-" * 80)

total_candidates = len(candidate_conditions)

diagnostic_field_candidate_count = int(
    final_candidate_evidence[
        "keyword_match_in_potential_diagnostic_field"
    ].sum()
)

paired_candidate_count = len(
    paired_candidate_conditions
)

print(
    f"Total CXR keyword candidate conditions       : "
    f"{total_candidates:,}"
)

print(
    f"Candidates with keyword in diagnostic field  : "
    f"{diagnostic_field_candidate_count:,}"
)

print(
    f"Candidates with CXR + genomic data           : "
    f"{paired_candidate_count:,}"
)

# =============================================================================
# 28. IMPORTANT: DO NOT AUTO-LABEL
# =============================================================================
#
# Every candidate remains UNVERIFIED unless the source metadata explicitly
# supports a Normal/Healthy/Control patient-level interpretation.
#
# This step does NOT decide that a keyword candidate is Normal.
#
# =============================================================================

final_candidate_evidence[
    "scientific_classification"
] = "UNVERIFIED_NORMAL_CANDIDATE"

final_candidate_evidence[
    "automatic_normal_label_allowed"
] = False

final_candidate_evidence[
    "reason"
] = (
    "Keyword evidence requires source-level clinical interpretation; "
    "keyword match alone does not establish a Normal diagnosis."
)

# =============================================================================
# 29. DECISION EVIDENCE
# =============================================================================

decision_rows = [
    {
        "evidence_id":
            "DEC-50-01",
        "question":
            "Are CXR Normal/Healthy/Control keyword candidates present?",
        "result":
            "YES",
        "count":
            total_candidates,
        "interpretation":
            "Keyword candidates exist but are not automatically Normal labels."
    },
    {
        "evidence_id":
            "DEC-50-02",
        "question":
            "Are genomic Normal/Healthy/Control keyword candidates present?",
        "result":
            "NO",
        "count":
            0,
        "interpretation":
            "No explicit genomic Normal/Healthy/Control keyword candidates were found in Step 3B.6.49."
    },
    {
        "evidence_id":
            "DEC-50-03",
        "question":
            "Are CXR Normal keyword candidates paired with genomic records?",
        "result":
            "YES" if paired_candidate_count > 0 else "NO",
        "count":
            paired_candidate_count,
        "interpretation":
            (
                "Some candidate CXR conditions have genomic records, "
                "but this does not establish that they are Normal."
                if paired_candidate_count > 0
                else
                "No candidate CXR conditions have genomic records."
            )
    },
    {
        "evidence_id":
            "DEC-50-04",
        "question":
            "Can keyword matches alone be converted to Normal labels?",
        "result":
            "NO",
        "count":
            None,
        "interpretation":
            "Scientific labeling requires documented patient-level evidence."
    },
    {
        "evidence_id":
            "DEC-50-05",
        "question":
            "Can missing genomic information define Normal?",
        "result":
            "NO",
        "count":
            None,
        "interpretation":
            "Missing modality is not a diagnostic class."
    },
    {
        "evidence_id":
            "DEC-50-06",
        "question":
            "Can absence of TB keyword define Normal?",
        "result":
            "NO",
        "count":
            None,
        "interpretation":
            "Absence of a keyword is not equivalent to a Normal diagnosis."
    }
]

decision_evidence_df = pd.DataFrame(decision_rows)

# =============================================================================
# 30. FINAL STATUS
# =============================================================================
#
# IMPORTANT:
# This status is intentionally conservative.
#
# Even if diagnostic-field candidates exist, they still require inspection
# of the actual values before being accepted as a Normal class.
#
# =============================================================================

if total_candidates == 0:

    final_internal_status = (
        "NO_CXR_NORMAL_KEYWORD_CANDIDATES_FOUND"
    )

elif paired_candidate_count == 0:

    final_internal_status = (
        "CXR_NORMAL_CANDIDATES_EXIST_BUT_NO_CXR_NORMAL_GENOMIC_OVERLAP"
    )

else:

    final_internal_status = (
        "CXR_NORMAL_CANDIDATES_WITH_GENOMIC_OVERLAP_REQUIRE_MANUAL_SOURCE_VALIDATION"
    )

final_cohort_status = "NOT_LOCKED"

# =============================================================================
# 31. PRINT DECISION
# =============================================================================

print("\n" + "=" * 80)
print("STEP 3B.6.50 — PROVISIONAL DECISION")
print("=" * 80)

print(
    f"CXR Normal keyword candidates          : "
    f"{total_candidates:,}"
)

print(
    f"Candidates with diagnostic-field match: "
    f"{diagnostic_field_candidate_count:,}"
)

print(
    f"Candidates with genomic records        : "
    f"{paired_candidate_count:,}"
)

print(
    f"Genomic Normal keyword candidates      : "
    f"0"
)

print(
    f"Paired verified Normal cohort          : "
    f"NOT ESTABLISHED"
)

print(
    f"Final cohort                           : "
    f"{final_cohort_status}"
)

print(
    f"Provisional internal status            : "
    f"{final_internal_status}"
)

# =============================================================================
# 32. SAVE OUTPUTS
# =============================================================================

print("\n18. SAVING OUTPUTS")
print("-" * 80)

outputs = {}

# 01
path = OUTPUT_DIR / (
    "01_Match_Level_Detail.csv"
)
matches_df.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# 02
path = OUTPUT_DIR / (
    "02_Candidate_Condition_Summary.csv"
)
final_candidate_evidence.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# 03
path = OUTPUT_DIR / (
    "03_Source_Column_Summary.csv"
)
source_column_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# 04
path = OUTPUT_DIR / (
    "04_Matched_Term_Summary.csv"
)
matched_term_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# 05
path = OUTPUT_DIR / (
    "05_Candidate_Overlap_With_Genomics.csv"
)
candidate_overlap.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# 06
path = OUTPUT_DIR / (
    "06_Candidate_Case_Definition_Summary.csv"
)
case_definition_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# 07
path = OUTPUT_DIR / (
    "07_Candidate_Diagnosis_Summary.csv"
)
diagnosis_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# 08
path = OUTPUT_DIR / (
    "08_Candidate_Outcome_Summary.csv"
)
outcome_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# 09
path = OUTPUT_DIR / (
    "09_Paired_Normal_Candidate_Audit.csv"
)
paired_normal_candidate_audit.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# 10
path = OUTPUT_DIR / (
    "10_Final_Decision_Evidence.csv"
)
decision_evidence_df.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# =============================================================================
# 33. AUDIT SUMMARY
# =============================================================================

audit_summary = pd.DataFrame([
    {
        "metric":
            "CXR records",
        "value":
            len(cxr),
        "status":
            "PASS"
    },
    {
        "metric":
            "CXR unique conditions",
        "value":
            len(cxr_unique_conditions),
        "status":
            "PASS"
    },
    {
        "metric":
            "Genomic records",
        "value":
            len(genomics),
        "status":
            "PASS"
    },
    {
        "metric":
            "Genomic unique conditions",
        "value":
            len(genomic_unique_conditions),
        "status":
            "PASS"
    },
    {
        "metric":
            "CXR + genomic common conditions",
        "value":
            len(common_conditions),
        "status":
            "PASS"
    },
    {
        "metric":
            "CXR Normal keyword candidates",
        "value":
            total_candidates,
        "status":
            "CANDIDATE_ONLY"
    },
    {
        "metric":
            "Candidates with diagnostic-field keyword evidence",
        "value":
            diagnostic_field_candidate_count,
        "status":
            "REQUIRES_SOURCE_REVIEW"
    },
    {
        "metric":
            "Candidates with genomic records",
        "value":
            paired_candidate_count,
        "status":
            "CANDIDATE_OVERLAP_ONLY"
    },
    {
        "metric":
            "Genomic Normal keyword candidates",
        "value":
            0,
        "status":
            "NONE_FOUND"
    },
    {
        "metric":
            "Verified paired Normal cohort",
        "value":
            0,
        "status":
            "NOT_ESTABLISHED"
    },
    {
        "metric":
            "Artificial Normal labels",
        "value":
            0,
        "status":
            "PROHIBITED"
    },
    {
        "metric":
            "External datasets",
        "value":
            0,
        "status":
            "NOT_USED"
    },
    {
        "metric":
            "Preprocessing",
        "value":
            0,
        "status":
            "NOT_STARTED"
    },
    {
        "metric":
            "Segmentation",
        "value":
            0,
        "status":
            "NOT_STARTED"
    },
    {
        "metric":
            "Model training",
        "value":
            0,
        "status":
            "NOT_STARTED"
    },
    {
        "metric":
            "Explainable AI",
        "value":
            0,
        "status":
            "NOT_STARTED"
    }
])

path = OUTPUT_DIR / (
    "12_Step_3B_6_50_Audit_Summary.csv"
)

audit_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)
outputs[path.name] = "PASS"

# =============================================================================
# 34. DECISION RECORD
# =============================================================================

decision_record = {
    "step": "3B.6.50",
    "title":
        "CXR Normal Candidate Provenance Audit",

    "execution_timestamp_utc":
        datetime.utcnow().isoformat() + "Z",

    "scope": {
        "dataset":
            "TB Portals March 2025",
        "external_datasets_used":
            False,
        "cxr_source":
            str(CXR_PATH),
        "genomics_source":
            str(GENOMICS_PATH)
    },

    "input_counts": {
        "cxr_records":
            int(len(cxr)),
        "cxr_unique_conditions":
            int(len(cxr_unique_conditions)),
        "genomic_records":
            int(len(genomics)),
        "genomic_unique_conditions":
            int(len(genomic_unique_conditions)),
        "common_conditions":
            int(len(common_conditions))
    },

    "candidate_results": {
        "cxr_normal_keyword_candidates":
            int(total_candidates),
        "keyword_evidence_rows":
            int(len(matches_df)),
        "diagnostic_field_candidate_conditions":
            int(diagnostic_field_candidate_count),
        "candidate_conditions_with_genomic_records":
            int(paired_candidate_count),
        "genomic_normal_keyword_candidates":
            0,
        "verified_paired_normal_cohort":
            0
    },

    "scientific_rules": {
        "keyword_alone_is_normal":
            False,
        "missing_genomics_is_normal":
            False,
        "absence_of_tb_is_normal":
            False,
        "cross_dataset_pairing_allowed":
            False,
        "artificial_normal_labels_allowed":
            False
    },

    "decision": {
        "provisional_internal_status":
            final_internal_status,
        "final_cohort":
            final_cohort_status,
        "research_question_locked":
            False,
        "normal_class_verified":
            False,
        "paired_normal_multimodal_class_verified":
            False
    },

    "next_required_action":
        (
            "Inspect source-column and actual-value evidence for the "
            "CXR Normal keyword candidates before deciding whether any "
            "candidate represents a documented Normal/Healthy/Control case."
        ),

    "outputs":
        outputs
}

json_path = OUTPUT_DIR / (
    "11_Step_3B_6_50_Decision_Record.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        decision_record,
        f,
        indent=4,
        ensure_ascii=False
    )

outputs[json_path.name] = "PASS"

# =============================================================================
# 35. FINAL VALIDATION
# =============================================================================

print("\n19. FINAL VALIDATION")
print("-" * 80)

validation = {
    "CXR source unchanged": True,
    "Genomics source unchanged": True,
    "External dataset used": False,
    "Artificial Normal labels created": False,
    "Missing genomics treated as Normal": False,
    "Absence of TB treated as Normal": False,
    "Cross-dataset pairing performed": False,
    "Preprocessing performed": False,
    "Segmentation performed": False,
    "Model training performed": False,
    "XAI performed": False
}

for name, status in validation.items():
    print(
        f"{name:<55}: "
        f"{'PASS' if status else 'FAIL'}"
    )

# =============================================================================
# 36. FINAL OUTPUT
# =============================================================================

print("\n" + "=" * 80)
print("STEP 3B.6.50 — COMPLETE")
print("=" * 80)

print(
    f"CXR Normal keyword candidates          : "
    f"{total_candidates:,}"
)

print(
    f"Diagnostic-field candidate conditions  : "
    f"{diagnostic_field_candidate_count:,}"
)

print(
    f"Candidates with genomic records        : "
    f"{paired_candidate_count:,}"
)

print(
    f"Genomic Normal keyword candidates      : "
    f"0"
)

print(
    f"Verified paired Normal cohort          : "
    f"NOT ESTABLISHED"
)

print(
    f"Final cohort                           : "
    f"{final_cohort_status}"
)

print(
    f"Provisional status                     : "
    f"{final_internal_status}"
)

print("\nOUTPUT DIRECTORY:")
print(OUTPUT_DIR)

print("\nIMPORTANT FILES:")
print("01_Match_Level_Detail.csv")
print("02_Candidate_Condition_Summary.csv")
print("03_Source_Column_Summary.csv")
print("04_Matched_Term_Summary.csv")
print("05_Candidate_Overlap_With_Genomics.csv")
print("06_Candidate_Case_Definition_Summary.csv")
print("07_Candidate_Diagnosis_Summary.csv")
print("08_Candidate_Outcome_Summary.csv")
print("09_Paired_Normal_Candidate_Audit.csv")
print("10_Final_Decision_Evidence.csv")
print("11_Step_3B_6_50_Decision_Record.json")
print("12_Step_3B_6_50_Audit_Summary.csv")

print("=" * 80)
```

```python
# =============================================================================
# STEP 3B.6.51 — CXR_OUTLIER SEMANTIC / FIELD AUDIT
# =============================================================================
#
# PURPOSE
# -------
# Establish exactly what the CXR metadata field `cxr_outlier` represents
# and determine whether:
#
#       cxr_outlier = NORMAL
#
# can scientifically represent a Normal patient/control diagnosis.
#
# IMPORTANT:
# -----------
# This step does NOT create Normal labels.
# This step does NOT train a model.
# This step does NOT preprocess images.
# This step does NOT perform lung segmentation.
# This step does NOT perform XAI.
# This step does NOT use any external dataset.
#
# ONLY:
#       TB Portals March 2025 CXR metadata
#
# MAIN QUESTIONS
# --------------
# Q1. What are all values present in cxr_outlier?
# Q2. Is "normal" exclusively an outlier/QC-style value?
# Q3. Does cxr_outlier=normal correspond to a clinical diagnosis?
# Q4. Do those records have TB diagnostic codes?
# Q5. Do those records have TB case definitions?
# Q6. Do those records have resistance categories?
# Q7. Do those records have treatment outcomes?
# Q8. Are there any explicit Normal/Healthy/Control values in clinical
#     diagnostic fields?
# Q9. Do any clinically documented Normal candidates have genomics?
# Q10. Can cxr_outlier=normal legitimately define the Normal class?
#
# OUTPUT DIRECTORY
# ----------------
# C:\TBP\Metadata\Step_3B_6_51_CXR_Outlier_Semantic_Audit
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import re
from datetime import datetime


# =============================================================================
# 1. CONFIGURATION
# =============================================================================

CXR_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_CXRs_March_2025.csv"
)

GENOMICS_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_Genomics_March_2025.csv"
)

OUTPUT_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_51_CXR_Outlier_Semantic_Audit"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# =============================================================================
# 2. REQUIRED COLUMNS
# =============================================================================

REQUIRED_CXR_COLUMNS = {
    "condition_id",
    "patient_id",
    "case_definition",
    "diagnosis_code",
    "type_of_resistance",
    "outcome",
    "cxr_outlier"
}

REQUIRED_GENOMIC_COLUMNS = {
    "condition_id"
}


# =============================================================================
# 3. SCIENTIFIC VOCABULARY
# =============================================================================

NORMAL_TERMS = [
    "NORMAL",
    "HEALTHY",
    "CONTROL",
    "TB_NEGATIVE",
    "TB_NEG",
    "NON_TB",
    "NO_TB",
    "NO_EVIDENCE_TB",
    "TB_ABSENT",
    "NEGATIVE_FOR_TB",
    "TB_FREE",
    "NON_TUBERCULOUS"
]

NORMAL_TERMS_SORTED = sorted(
    NORMAL_TERMS,
    key=len,
    reverse=True
)


# =============================================================================
# 4. HELPER FUNCTIONS
# =============================================================================

def clean_value(value):
    """
    Convert metadata value to a clean string for audit display.
    """
    if pd.isna(value):
        return "<MISSING>"

    value = str(value).strip()

    if value == "":
        return "<EMPTY>"

    return value


def normalize_text(value):
    """
    Normalize text for keyword detection only.
    """
    if pd.isna(value):
        return ""

    text = str(value).strip().upper()

    text = text.replace("-", "_")
    text = text.replace("/", "_")

    text = re.sub(r"\s+", " ", text)

    return text


def find_normal_terms(value):
    """
    Find Normal/Healthy/Control candidate terms.

    IMPORTANT:
    This function creates CANDIDATE evidence only.
    It never creates a Normal diagnosis.
    """

    text = normalize_text(value)

    if not text:
        return []

    matches = []

    for term in NORMAL_TERMS_SORTED:

        parts = [
            re.escape(part)
            for part in term.split("_")
        ]

        pattern = r"[\s_\-/]+".join(parts)

        try:
            if re.search(pattern, text):
                matches.append(term)
        except re.error:
            pass

    return matches


def print_section(title):
    print("\n" + "=" * 80)
    print(title)
    print("=" * 80)


# =============================================================================
# 5. LOAD DATA
# =============================================================================

print_section(
    "STEP 3B.6.51 — CXR_OUTLIER SEMANTIC / FIELD AUDIT"
)

print("\n1. LOADING SOURCE DATA")
print("-" * 80)

if not CXR_PATH.exists():
    raise FileNotFoundError(
        f"CXR source file not found:\n{CXR_PATH}"
    )

if not GENOMICS_PATH.exists():
    raise FileNotFoundError(
        f"Genomics source file not found:\n{GENOMICS_PATH}"
    )

cxr = pd.read_csv(
    CXR_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)

print(
    f"CXR records     : {len(cxr):,}"
)

print(
    f"CXR columns     : {len(cxr.columns):,}"
)

print(
    f"Genomic records : {len(genomics):,}"
)

print(
    f"Genomic columns : {len(genomics.columns):,}"
)


# =============================================================================
# 6. SCHEMA VALIDATION
# =============================================================================

print("\n2. SCHEMA VALIDATION")
print("-" * 80)

missing_cxr = REQUIRED_CXR_COLUMNS - set(cxr.columns)
missing_genomics = REQUIRED_GENOMIC_COLUMNS - set(genomics.columns)

if missing_cxr:
    raise ValueError(
        "Missing required CXR columns:\n"
        + "\n".join(sorted(missing_cxr))
    )

if missing_genomics:
    raise ValueError(
        "Missing required genomic columns:\n"
        + "\n".join(sorted(missing_genomics))
    )

print("Required CXR columns     : PASS")
print("Required genomic columns : PASS")


# =============================================================================
# 7. IDENTIFIER NORMALIZATION FOR AUDIT
# =============================================================================

cxr["condition_id"] = (
    cxr["condition_id"]
    .astype(str)
    .str.strip()
)

cxr["patient_id"] = (
    cxr["patient_id"]
    .astype(str)
    .str.strip()
)

genomics["condition_id"] = (
    genomics["condition_id"]
    .astype(str)
    .str.strip()
)


# =============================================================================
# 8. BASIC COUNTS
# =============================================================================

print("\n3. BASIC DATASET STRUCTURE")
print("-" * 80)

cxr_conditions = set(
    cxr.loc[
        cxr["condition_id"] != "",
        "condition_id"
    ].unique()
)

genomic_conditions = set(
    genomics.loc[
        genomics["condition_id"] != "",
        "condition_id"
    ].unique()
)

common_conditions = (
    cxr_conditions &
    genomic_conditions
)

print(
    f"CXR records              : {len(cxr):,}"
)

print(
    f"CXR unique conditions    : {len(cxr_conditions):,}"
)

print(
    f"Genomic records          : {len(genomics):,}"
)

print(
    f"Genomic unique conditions: {len(genomic_conditions):,}"
)

print(
    f"Common conditions        : {len(common_conditions):,}"
)


# =============================================================================
# 9. COMPLETE cxr_outlier VALUE AUDIT
# =============================================================================

print("\n4. COMPLETE cxr_outlier VALUE AUDIT")
print("-" * 80)

outlier_value_counts = (
    cxr["cxr_outlier"]
    .map(clean_value)
    .value_counts(dropna=False)
    .rename_axis("cxr_outlier_value")
    .reset_index(name="record_count")
)

outlier_value_counts["percentage"] = (
    outlier_value_counts["record_count"]
    / len(cxr)
    * 100
)

print(
    outlier_value_counts.to_string(index=False)
)


# =============================================================================
# 10. UNIQUE cxr_outlier VALUES
# =============================================================================

print("\n5. UNIQUE cxr_outlier VALUES")
print("-" * 80)

unique_outlier_values = (
    cxr["cxr_outlier"]
    .map(clean_value)
    .drop_duplicates()
    .tolist()
)

for value in unique_outlier_values:
    print(
        f"  - {value}"
    )


# =============================================================================
# 11. NORMAL cxr_outlier RECORDS
# =============================================================================

normal_outlier_mask = (
    cxr["cxr_outlier"]
    .map(normalize_text)
    .eq("NORMAL")
)

normal_outlier_cxr = cxr[
    normal_outlier_mask
].copy()

normal_outlier_conditions = set(
    normal_outlier_cxr["condition_id"].unique()
)

print("\n6. cxr_outlier = NORMAL")
print("-" * 80)

print(
    f"CXR records with cxr_outlier=normal : "
    f"{len(normal_outlier_cxr):,}"
)

print(
    f"Unique conditions                   : "
    f"{len(normal_outlier_conditions):,}"
)

print(
    f"Unique patients                     : "
    f"{normal_outlier_cxr['patient_id'].nunique():,}"
)


# =============================================================================
# 12. COMPARE cxr_outlier WITH CLINICAL DIAGNOSTIC FIELDS
# =============================================================================

print("\n7. cxr_outlier=normal — CLINICAL FIELD AUDIT")
print("-" * 80)

clinical_columns = [
    "case_definition",
    "diagnosis_code",
    "type_of_resistance",
    "outcome"
]

clinical_columns = [
    col for col in clinical_columns
    if col in normal_outlier_cxr.columns
]

for col in clinical_columns:

    print(
        f"\n--- {col} ---"
    )

    counts = (
        normal_outlier_cxr[col]
        .map(clean_value)
        .value_counts(dropna=False)
    )

    print(
        counts.to_string()
    )


# =============================================================================
# 13. DIAGNOSIS CODE CLASSIFICATION
# =============================================================================
#
# ICD-10 pulmonary/extrapulmonary TB codes in the observed data begin with
# A15-A19.
#
# We do NOT infer diagnosis from absence of code.
# We only describe the observed coding.
#
# =============================================================================

def classify_observed_diagnosis_code(value):

    value = clean_value(value)

    if value in {"<MISSING>", "<EMPTY>", "Not Reported"}:
        return "MISSING_OR_NOT_REPORTED"

    normalized = value.upper()

    if re.match(r"^A1[5-9](?:\.|$)", normalized):
        return "TB_CODE_A15_A19"

    return "OTHER_CODE"


normal_outlier_cxr[
    "observed_diagnosis_code_class"
] = normal_outlier_cxr[
    "diagnosis_code"
].apply(
    classify_observed_diagnosis_code
)

diagnosis_class_summary = (
    normal_outlier_cxr[
        "observed_diagnosis_code_class"
    ]
    .value_counts(dropna=False)
    .rename_axis("diagnosis_code_class")
    .reset_index(name="record_count")
)

print("\n8. DIAGNOSIS CODE CLASSIFICATION")
print("-" * 80)

print(
    diagnosis_class_summary.to_string(index=False)
)


# =============================================================================
# 14. CASE DEFINITION AUDIT
# =============================================================================

print("\n9. CASE DEFINITION AUDIT FOR cxr_outlier=normal")
print("-" * 80)

case_definition_summary = (
    normal_outlier_cxr[
        "case_definition"
    ]
    .map(clean_value)
    .value_counts(dropna=False)
    .rename_axis("case_definition")
    .reset_index(name="record_count")
)

print(
    case_definition_summary.to_string(index=False)
)


# =============================================================================
# 15. RESISTANCE AUDIT
# =============================================================================

print("\n10. RESISTANCE AUDIT FOR cxr_outlier=normal")
print("-" * 80)

resistance_summary = (
    normal_outlier_cxr[
        "type_of_resistance"
    ]
    .map(clean_value)
    .value_counts(dropna=False)
    .rename_axis("type_of_resistance")
    .reset_index(name="record_count")
)

print(
    resistance_summary.to_string(index=False)
)


# =============================================================================
# 16. OUTCOME AUDIT
# =============================================================================

print("\n11. OUTCOME AUDIT FOR cxr_outlier=normal")
print("-" * 80)

outcome_summary = (
    normal_outlier_cxr[
        "outcome"
    ]
    .map(clean_value)
    .value_counts(dropna=False)
    .rename_axis("outcome")
    .reset_index(name="record_count")
)

print(
    outcome_summary.to_string(index=False)
)


# =============================================================================
# 17. COUNTRY AUDIT
# =============================================================================

print("\n12. COUNTRY AUDIT FOR cxr_outlier=normal")
print("-" * 80)

if "country" in normal_outlier_cxr.columns:

    country_summary = (
        normal_outlier_cxr[
            "country"
        ]
        .map(clean_value)
        .value_counts(dropna=False)
        .rename_axis("country")
        .reset_index(name="record_count")
    )

    print(
        country_summary.to_string(index=False)
    )

else:

    country_summary = pd.DataFrame(
        columns=[
            "country",
            "record_count"
        ]
    )

    print(
        "country column not available."
    )


# =============================================================================
# 18. MODALITY AUDIT
# =============================================================================

print("\n13. SERIES MODALITY AUDIT FOR cxr_outlier=normal")
print("-" * 80)

if "series_modality_cd" in normal_outlier_cxr.columns:

    modality_summary = (
        normal_outlier_cxr[
            "series_modality_cd"
        ]
        .map(clean_value)
        .value_counts(dropna=False)
        .rename_axis("series_modality_cd")
        .reset_index(name="record_count")
    )

    print(
        modality_summary.to_string(index=False)
    )

else:

    modality_summary = pd.DataFrame(
        columns=[
            "series_modality_cd",
            "record_count"
        ]
    )


# =============================================================================
# 19. DOES cxr_outlier=normal HAVE GENOMICS?
# =============================================================================

print("\n14. GENOMIC OVERLAP FOR cxr_outlier=normal")
print("-" * 80)

normal_outlier_with_genomics = (
    normal_outlier_conditions &
    genomic_conditions
)

normal_outlier_without_genomics = (
    normal_outlier_conditions -
    genomic_conditions
)

print(
    f"cxr_outlier=normal conditions : "
    f"{len(normal_outlier_conditions):,}"
)

print(
    f"With genomic records           : "
    f"{len(normal_outlier_with_genomics):,}"
)

print(
    f"Without genomic records        : "
    f"{len(normal_outlier_without_genomics):,}"
)


# =============================================================================
# 20. CLINICAL NORMAL KEYWORD SEARCH
# =============================================================================
#
# CRITICAL:
#
# We now search ONLY the potentially clinical fields.
#
# We deliberately exclude:
#
#   cxr_outlier
#
# because we already established that it is the source of the previous
# 10,842 keyword candidates.
#
# =============================================================================

print("\n15. CLINICAL NORMAL/HEALTHY/CONTROL SEARCH")
print("-" * 80)

clinical_search_columns = [
    "case_definition",
    "diagnosis_code",
    "outcome",
    "type_of_resistance",
    "comorbidity"
]

clinical_search_columns = [
    col
    for col in clinical_search_columns
    if col in cxr.columns
]

print(
    "Clinical fields searched:"
)

for col in clinical_search_columns:
    print(
        f"  - {col}"
    )


clinical_match_rows = []

for col in clinical_search_columns:

    for row_index, value in cxr[col].items():

        matched_terms = find_normal_terms(value)

        if not matched_terms:
            continue

        condition_id = cxr.at[
            row_index,
            "condition_id"
        ]

        for term in matched_terms:

            clinical_match_rows.append({
                "cxr_row_index":
                    int(row_index),
                "condition_id":
                    condition_id,
                "source_column":
                    col,
                "matched_term":
                    term,
                "actual_value":
                    value
            })


clinical_matches_df = pd.DataFrame(
    clinical_match_rows
)

if clinical_matches_df.empty:

    print(
        "NO NORMAL/HEALTHY/CONTROL KEYWORD MATCHES "
        "FOUND IN THE SEARCHED CLINICAL FIELDS."
    )

else:

    clinical_matches_df = (
        clinical_matches_df
        .drop_duplicates()
        .reset_index(drop=True)
    )

    print(
        f"Clinical evidence rows : "
        f"{len(clinical_matches_df):,}"
    )

    print(
        f"Clinical candidate conditions : "
        f"{clinical_matches_df['condition_id'].nunique():,}"
    )

    print("\nClinical source-column distribution:")

    print(
        clinical_matches_df[
            "source_column"
        ]
        .value_counts()
        .to_string()
    )

    print("\nClinical matched-term distribution:")

    print(
        clinical_matches_df[
            "matched_term"
        ]
        .value_counts()
        .to_string()
    )


# =============================================================================
# 21. BUILD CLINICAL NORMAL CANDIDATE CONDITIONS
# =============================================================================

if clinical_matches_df.empty:

    clinical_normal_conditions = set()

else:

    clinical_normal_conditions = set(
        clinical_matches_df[
            "condition_id"
        ].unique()
    )


# =============================================================================
# 22. CLINICAL NORMAL CANDIDATE GENOMIC OVERLAP
# =============================================================================

clinical_normal_genomic_overlap = (
    clinical_normal_conditions &
    genomic_conditions
)

print("\n16. CLINICAL NORMAL CANDIDATE + GENOMIC OVERLAP")
print("-" * 80)

print(
    f"Clinical Normal candidate conditions : "
    f"{len(clinical_normal_conditions):,}"
)

print(
    f"Same-condition genomic overlap        : "
    f"{len(clinical_normal_genomic_overlap):,}"
)


# =============================================================================
# 23. CHECK WHETHER cxr_outlier=normal CASES ARE ACTUALLY TB CASES
# =============================================================================

print("\n17. cxr_outlier=normal VS TB DIAGNOSTIC EVIDENCE")
print("-" * 80)

tb_code_mask = (
    normal_outlier_cxr[
        "diagnosis_code"
    ]
    .astype(str)
    .str.upper()
    .str.match(
        r"^A1[5-9](?:\.|$)",
        na=False
    )
)

tb_code_records = int(
    tb_code_mask.sum()
)

tb_code_conditions = (
    normal_outlier_cxr.loc[
        tb_code_mask,
        "condition_id"
    ]
    .nunique()
)

print(
    f"Records with observed A15-A19 TB code : "
    f"{tb_code_records:,}"
)

print(
    f"Conditions with observed A15-A19 TB code : "
    f"{tb_code_conditions:,}"
)


# =============================================================================
# 24. BUILD CONDITION-LEVEL INTERPRETATION TABLE
# =============================================================================

print("\n18. BUILDING CONDITION-LEVEL AUDIT")
print("-" * 80)

condition_audit = (
    normal_outlier_cxr
    .groupby("condition_id")
    .agg(
        patient_id=(
            "patient_id",
            lambda x: "|".join(
                sorted(
                    set(
                        str(v)
                        for v in x
                        if not pd.isna(v)
                    )
                )
            )
        ),
        cxr_record_count=(
            "condition_id",
            "size"
        ),
        diagnosis_codes=(
            "diagnosis_code",
            lambda x: "|".join(
                sorted(
                    set(
                        clean_value(v)
                        for v in x
                    )
                )
            )
        ),
        case_definitions=(
            "case_definition",
            lambda x: "|".join(
                sorted(
                    set(
                        clean_value(v)
                        for v in x
                    )
                )
            )
        ),
        resistance_categories=(
            "type_of_resistance",
            lambda x: "|".join(
                sorted(
                    set(
                        clean_value(v)
                        for v in x
                    )
                )
            )
        ),
        outcomes=(
            "outcome",
            lambda x: "|".join(
                sorted(
                    set(
                        clean_value(v)
                        for v in x
                    )
                )
            )
        )
    )
    .reset_index()
)

condition_audit[
    "has_genomic_record"
] = condition_audit[
    "condition_id"
].isin(genomic_conditions)

condition_audit[
    "has_tb_diagnosis_code"
] = (
    condition_audit[
        "diagnosis_codes"
    ]
    .str.upper()
    .str.contains(
        r"A1[5-9]",
        regex=True,
        na=False
    )
)

# =============================================================================
# 25. POTENTIAL CLINICAL NORMAL CANDIDATE CONDITION AUDIT
# =============================================================================

if clinical_normal_conditions:

    clinical_normal_condition_audit = (
        cxr[
            cxr["condition_id"].isin(
                clinical_normal_conditions
            )
        ]
        .copy()
    )

    clinical_normal_condition_audit = (
        clinical_normal_condition_audit
        .sort_values(
            [
                "condition_id"
            ]
        )
    )

else:

    clinical_normal_condition_audit = pd.DataFrame(
        columns=cxr.columns
    )


# =============================================================================
# 26. FINAL SCIENTIFIC INTERPRETATION
# =============================================================================
#
# The rules below are deliberately conservative.
#
# We NEVER declare a Normal cohort merely because:
#
#   cxr_outlier = normal
#
# A verified Normal class requires explicit clinical/patient-level evidence.
#
# =============================================================================

cxr_outlier_is_diagnostic_label = False

if clinical_normal_conditions:

    clinical_normal_status = (
        "CLINICAL_NORMAL_KEYWORD_CANDIDATES_FOUND"
    )

else:

    clinical_normal_status = (
        "NO_CLINICAL_NORMAL_KEYWORD_CANDIDATES_FOUND"
    )


if clinical_normal_genomic_overlap:

    paired_clinical_normal_status = (
        "CLINICAL_NORMAL_CANDIDATES_HAVE_GENOMIC_OVERLAP_REQUIRES_REVIEW"
    )

else:

    paired_clinical_normal_status = (
        "NO_CLINICAL_NORMAL_GENOMIC_OVERLAP"
    )


# =============================================================================
# 27. FINAL DECISION
# =============================================================================

if clinical_normal_conditions:

    final_status = (
        "CLINICAL_NORMAL_CANDIDATES_REQUIRE_SOURCE_LEVEL_VALIDATION"
    )

else:

    final_status = (
        "CXR_OUTLIER_NORMAL_IS_NOT_A_VERIFIED_NORMAL_DIAGNOSTIC_CLASS"
    )

final_cohort = (
    "NOT_LOCKED"
)

# =============================================================================
# 28. DECISION EVIDENCE TABLE
# =============================================================================

decision_evidence = pd.DataFrame([
    {
        "decision_id":
            "DEC-51-01",
        "question":
            "Does cxr_outlier contain NORMAL?",
        "result":
            "YES",
        "count":
            int(normal_outlier_cxr.shape[0]),
        "scientific_interpretation":
            "The field contains the value NORMAL."
    },
    {
        "decision_id":
            "DEC-51-02",
        "question":
            "Does cxr_outlier=normal itself establish a Normal patient?",
        "result":
            "NO",
        "count":
            None,
        "scientific_interpretation":
            "The field is not treated as a clinical diagnostic label."
    },
    {
        "decision_id":
            "DEC-51-03",
        "question":
            "Are explicit Normal/Healthy/Control candidates present in clinical fields?",
        "result":
            "YES" if clinical_normal_conditions else "NO",
        "count":
            int(len(clinical_normal_conditions)),
        "scientific_interpretation":
            (
                "Clinical-field candidates require source-level validation."
                if clinical_normal_conditions
                else
                "No Normal/Healthy/Control keyword candidates were found "
                "in the searched clinical fields."
            )
    },
    {
        "decision_id":
            "DEC-51-04",
        "question":
            "Do clinical Normal candidates have same-condition genomics?",
        "result":
            "YES" if clinical_normal_genomic_overlap else "NO",
        "count":
            int(len(clinical_normal_genomic_overlap)),
        "scientific_interpretation":
            (
                "Overlap exists but does not by itself prove Normal status."
                if clinical_normal_genomic_overlap
                else
                "No same-condition genomic overlap was identified."
            )
    },
    {
        "decision_id":
            "DEC-51-05",
        "question":
            "Can missing genomics define Normal?",
        "result":
            "NO",
        "count":
            None,
        "scientific_interpretation":
            "Missing modality is not a diagnostic class."
    },
    {
        "decision_id":
            "DEC-51-06",
        "question":
            "Can absence of TB define Normal?",
        "result":
            "NO",
        "count":
            None,
        "scientific_interpretation":
            "Absence of TB coding/keywords is not equivalent to a Normal diagnosis."
    },
    {
        "decision_id":
            "DEC-51-07",
        "question":
            "Can cxr_outlier=normal be used as the Normal target?",
        "result":
            "NO",
        "count":
            int(len(normal_outlier_conditions)),
        "scientific_interpretation":
            "It is not accepted as a patient-level Normal diagnostic label."
    }
])


# =============================================================================
# 29. CORRECT VALIDATION LOGIC
# =============================================================================
#
# IMPORTANT:
# True = prohibited action occurred -> FAIL
# False = prohibited action did NOT occur -> PASS
#
# This corrects the misleading output in Step 3B.6.50.
#
# =============================================================================

validation_results = [
    {
        "validation":
            "External dataset used",
        "observed":
            False,
        "status":
            "PASS"
    },
    {
        "validation":
            "Artificial Normal labels created",
        "observed":
            False,
        "status":
            "PASS"
    },
    {
        "validation":
            "Missing genomics treated as Normal",
        "observed":
            False,
        "status":
            "PASS"
    },
    {
        "validation":
            "Absence of TB treated as Normal",
        "observed":
            False,
        "status":
            "PASS"
    },
    {
        "validation":
            "Cross-dataset pairing performed",
        "observed":
            False,
        "status":
            "PASS"
    },
    {
        "validation":
            "Preprocessing performed",
        "observed":
            False,
        "status":
            "PASS"
    },
    {
        "validation":
            "Segmentation performed",
        "observed":
            False,
        "status":
            "PASS"
    },
    {
        "validation":
            "Model training performed",
        "observed":
            False,
        "status":
            "PASS"
    },
    {
        "validation":
            "XAI performed",
        "observed":
            False,
        "status":
            "PASS"
    },
    {
        "validation":
            "Source metadata modified",
        "observed":
            False,
        "status":
            "PASS"
    },
    {
        "validation":
            "Raw DICOM modified",
        "observed":
            False,
        "status":
            "PASS"
    }
]

validation_df = pd.DataFrame(
    validation_results
)


# =============================================================================
# 30. SAVE OUTPUTS
# =============================================================================

print("\n19. SAVING OUTPUTS")
print("-" * 80)

output_files = {}


# -----------------------------------------------------------------------------
# 01. COMPLETE cxr_outlier VALUE DISTRIBUTION
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "01_cxr_outlier_Value_Distribution.csv"
)

outlier_value_counts.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 02. ALL cxr_outlier=NORMAL RECORDS
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "02_cxr_outlier_Normal_All_Records.csv"
)

normal_outlier_cxr.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 03. CONDITION LEVEL AUDIT
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "03_cxr_outlier_Normal_Condition_Audit.csv"
)

condition_audit.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 04. DIAGNOSIS CODE SUMMARY
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "04_cxr_outlier_Normal_Diagnosis_Code_Summary.csv"
)

diagnosis_class_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 05. CASE DEFINITION SUMMARY
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "05_cxr_outlier_Normal_Case_Definition_Summary.csv"
)

case_definition_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 06. RESISTANCE SUMMARY
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "06_cxr_outlier_Normal_Resistance_Summary.csv"
)

resistance_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 07. OUTCOME SUMMARY
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "07_cxr_outlier_Normal_Outcome_Summary.csv"
)

outcome_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 08. CLINICAL NORMAL MATCHES
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "08_Clinical_Normal_Keyword_Matches.csv"
)

clinical_matches_df.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 09. CLINICAL NORMAL CONDITION AUDIT
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "09_Clinical_Normal_Candidate_Condition_Audit.csv"
)

clinical_normal_condition_audit.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 10. PAIRED CLINICAL NORMAL CANDIDATE AUDIT
# -----------------------------------------------------------------------------

if clinical_normal_conditions:

    paired_audit = (
        cxr[
            cxr["condition_id"].isin(
                clinical_normal_genomic_overlap
            )
        ]
        .copy()
    )

else:

    paired_audit = pd.DataFrame(
        columns=cxr.columns
    )

path = OUTPUT_DIR / (
    "10_Paired_Clinical_Normal_Candidate_Audit.csv"
)

paired_audit.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 11. DECISION EVIDENCE
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "11_Step_3B_6_51_Decision_Evidence.csv"
)

decision_evidence.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# -----------------------------------------------------------------------------
# 12. VALIDATION
# -----------------------------------------------------------------------------

path = OUTPUT_DIR / (
    "12_Step_3B_6_51_Validation.csv"
)

validation_df.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# =============================================================================
# 31. FINAL AUDIT SUMMARY
# =============================================================================

audit_summary = pd.DataFrame([
    {
        "metric":
            "CXR records",
        "value":
            len(cxr),
        "status":
            "PASS"
    },
    {
        "metric":
            "CXR unique conditions",
        "value":
            len(cxr_conditions),
        "status":
            "PASS"
    },
    {
        "metric":
            "Genomic records",
        "value":
            len(genomics),
        "status":
            "PASS"
    },
    {
        "metric":
            "Genomic unique conditions",
        "value":
            len(genomic_conditions),
        "status":
            "PASS"
    },
    {
        "metric":
            "CXR + genomic common conditions",
        "value":
            len(common_conditions),
        "status":
            "PASS"
    },
    {
        "metric":
            "cxr_outlier=normal CXR records",
        "value":
            len(normal_outlier_cxr),
        "status":
            "OBSERVED_FIELD_VALUE"
    },
    {
        "metric":
            "cxr_outlier=normal conditions",
        "value":
            len(normal_outlier_conditions),
        "status":
            "OBSERVED_FIELD_VALUE"
    },
    {
        "metric":
            "cxr_outlier=normal conditions with genomics",
        "value":
            len(normal_outlier_with_genomics),
        "status":
            "OVERLAP_ONLY"
    },
    {
        "metric":
            "Clinical Normal keyword candidates",
        "value":
            len(clinical_normal_conditions),
        "status":
            clinical_normal_status
    },
    {
        "metric":
            "Clinical Normal candidates with genomics",
        "value":
            len(clinical_normal_genomic_overlap),
        "status":
            paired_clinical_normal_status
    },
    {
        "metric":
            "Verified Normal patient class",
        "value":
            0,
        "status":
            "NOT_ESTABLISHED"
    },
    {
        "metric":
            "Verified paired Normal CXR + genomic class",
        "value":
            0,
        "status":
            "NOT_ESTABLISHED"
    }
])

path = OUTPUT_DIR / (
    "13_Step_3B_6_51_Audit_Summary.csv"
)

audit_summary.to_csv(
    path,
    index=False,
    encoding="utf-8-sig"
)

output_files[path.name] = "PASS"


# =============================================================================
# 32. DECISION RECORD JSON
# =============================================================================

decision_record = {

    "step":
        "3B.6.51",

    "title":
        "CXR_outlier Semantic / Field Audit",

    "execution_timestamp_utc":
        datetime.utcnow().isoformat() + "Z",

    "scope":
        {
            "dataset":
                "TB Portals March 2025",
            "external_dataset_used":
                False,
            "cxr_source":
                str(CXR_PATH),
            "genomics_source":
                str(GENOMICS_PATH)
        },

    "dataset_counts":
        {
            "cxr_records":
                int(len(cxr)),
            "cxr_unique_conditions":
                int(len(cxr_conditions)),
            "genomic_records":
                int(len(genomics)),
            "genomic_unique_conditions":
                int(len(genomic_conditions)),
            "common_conditions":
                int(len(common_conditions))
        },

    "cxr_outlier":
        {
            "normal_record_count":
                int(len(normal_outlier_cxr)),
            "normal_condition_count":
                int(len(normal_outlier_conditions)),
            "normal_conditions_with_genomics":
                int(len(normal_outlier_with_genomics)),
            "treated_as_diagnostic_normal":
                False
        },

    "clinical_search":
        {
            "fields_searched":
                clinical_search_columns,
            "candidate_condition_count":
                int(len(clinical_normal_conditions)),
            "candidate_conditions_with_genomics":
                int(len(clinical_normal_genomic_overlap))
        },

    "scientific_rules":
        {
            "cxr_outlier_normal_is_diagnostic_label":
                False,
            "missing_genomics_is_normal":
                False,
            "absence_of_tb_is_normal":
                False,
            "keyword_alone_is_normal":
                False,
            "cross_dataset_pairing":
                False,
            "artificial_normal_labels":
                False
        },

    "decision":
        {
            "final_status":
                final_status,
            "final_cohort":
                final_cohort,
            "normal_class_verified":
                False,
            "paired_normal_multimodal_class_verified":
                False
        },

    "next_step":
        (
            "Use the clinical-field evidence to determine whether a "
            "documented patient-level Normal/Healthy/Control class exists. "
            "Do not use cxr_outlier=normal as a Normal target."
        ),

    "outputs":
        output_files
}


json_path = OUTPUT_DIR / (
    "14_Step_3B_6_51_Decision_Record.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        decision_record,
        f,
        indent=4,
        ensure_ascii=False
    )

output_files[json_path.name] = "PASS"


# =============================================================================
# 33. FINAL CONSOLE REPORT
# =============================================================================

print_section(
    "STEP 3B.6.51 — FINAL RESULT"
)

print(
    f"CXR records                         : "
    f"{len(cxr):,}"
)

print(
    f"CXR unique conditions               : "
    f"{len(cxr_conditions):,}"
)

print(
    f"Genomic records                     : "
    f"{len(genomics):,}"
)

print(
    f"Genomic unique conditions           : "
    f"{len(genomic_conditions):,}"
)

print(
    f"CXR + genomic common conditions     : "
    f"{len(common_conditions):,}"
)

print(
    f"\ncxr_outlier=normal records          : "
    f"{len(normal_outlier_cxr):,}"
)

print(
    f"cxr_outlier=normal conditions       : "
    f"{len(normal_outlier_conditions):,}"
)

print(
    f"Those conditions with genomics      : "
    f"{len(normal_outlier_with_genomics):,}"
)

print(
    f"\nClinical Normal keyword candidates  : "
    f"{len(clinical_normal_conditions):,}"
)

print(
    f"Clinical Normal + genomic overlap   : "
    f"{len(clinical_normal_genomic_overlap):,}"
)

print(
    "\nCan cxr_outlier=normal define Normal?"
)

print(
    "NO — NOT ACCEPTED AS A PATIENT-LEVEL "
    "NORMAL DIAGNOSTIC LABEL."
)

print(
    f"\nFinal internal status               : "
    f"{final_status}"
)

print(
    f"Final cohort                        : "
    f"{final_cohort}"
)

print(
    "\nScientific validation:"
)

for row in validation_results:
    print(
        f"  {row['validation']:<50} : "
        f"{row['status']}"
    )

print(
    "\nOUTPUT DIRECTORY:"
)

print(
    OUTPUT_DIR
)

print_section(
    "STEP 3B.6.51 COMPLETE"
)
```

```python
# =============================================================================
# STEP 3B.6.54
# PAIRED NORMAL MULTIMODAL DATA SOURCE INVESTIGATION
# =============================================================================
#
# PURPOSE
# -------
# Determine whether the original research objective can be supported by a
# scientifically valid Normal multimodal cohort.
#
# ORIGINAL RESEARCH QUESTION
# ---------------------------
# Can Tuberculosis versus Normal be predicted using paired Chest X-ray
# imaging and genomic information from the same patient-level case,
# with lung segmentation and explainable AI?
#
# IMPORTANT SCIENTIFIC RULES
# --------------------------
# 1. TB and Normal must be explicitly defined.
# 2. Normal CXR must have a documented Normal/Healthy/Control interpretation.
# 3. Normal genomic information must belong to the SAME subject.
# 4. Missing genomic data cannot be interpreted as Normal.
# 5. A Normal CXR from subject A cannot be paired with genomics from subject B.
# 6. Absence of a TB diagnosis cannot automatically define Normal.
# 7. cxr_outlier=normal cannot be used as a Normal diagnostic label.
# 8. No artificial Normal labels.
# 9. No preprocessing.
# 10. No lung segmentation.
# 11. No model training.
# 12. No XAI.
#
# THIS STEP DOES NOT SELECT A FINAL RESEARCH DESIGN.
#
# =============================================================================


from pathlib import Path
import pandas as pd
import json
from datetime import datetime


# =============================================================================
# 1. CONFIGURATION
# =============================================================================

BASE_DIR = Path(
    r"C:\TBP\Metadata"
)

STEP_53_DIR = (
    BASE_DIR /
    "Step_3B_6_53_Research_Design_Options_Decision_Gate"
)

OUTPUT_DIR = (
    BASE_DIR /
    "Step_3B_6_54_Paired_Normal_Multimodal_Data_Source_Investigation"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# =============================================================================
# 2. SOURCE FILES
# =============================================================================

CXR_PATH = (
    BASE_DIR /
    "TB_Portals_CXRs_March_2025.csv"
)

GENOMICS_PATH = (
    BASE_DIR /
    "TB_Portals_Genomics_March_2025.csv"
)

CXR_MASTER_PATH = (
    BASE_DIR /
    "Step_3B_6_24_Notebook_03" /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

STEP_53_EVIDENCE_PATH = (
    STEP_53_DIR /
    "01_Verified_Evidence.csv"
)

STEP_53_STATE_PATH = (
    STEP_53_DIR /
    "07_Current_Research_State.csv"
)


# =============================================================================
# 3. VERIFY REQUIRED FILES
# =============================================================================

print("=" * 80)
print("STEP 3B.6.54 — PAIRED NORMAL MULTIMODAL DATA SOURCE INVESTIGATION")
print("=" * 80)

required_paths = {
    "CXR source":
        CXR_PATH,

    "Genomics source":
        GENOMICS_PATH,

    "CXR multimodal master":
        CXR_MASTER_PATH,

    "Step 3B.6.53 evidence":
        STEP_53_EVIDENCE_PATH,

    "Step 3B.6.53 state":
        STEP_53_STATE_PATH
}


missing = []

for label, path in required_paths.items():

    if path.exists():

        print(
            f"{label:<35}: FOUND"
        )

    else:

        print(
            f"{label:<35}: MISSING"
        )

        missing.append(
            f"{label}: {path}"
        )


if missing:

    raise FileNotFoundError(
        "Required input file(s) missing:\n"
        + "\n".join(missing)
    )


# =============================================================================
# 4. LOAD TB PORTALS DATA
# =============================================================================

print("\n1. LOADING TB PORTALS DATA")
print("-" * 80)

cxr = pd.read_csv(
    CXR_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)

cxr_master = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

step53_evidence = pd.read_csv(
    STEP_53_EVIDENCE_PATH,
    dtype=str
)

step53_state = pd.read_csv(
    STEP_53_STATE_PATH,
    dtype=str
)


print(
    f"CXR records      : {len(cxr):,}"
)

print(
    f"Genomic records   : {len(genomics):,}"
)

print(
    f"CXR master rows   : {len(cxr_master):,}"
)


# =============================================================================
# 5. REQUIRED COLUMN VALIDATION
# =============================================================================

print("\n2. REQUIRED COLUMN VALIDATION")
print("-" * 80)


required_cxr_columns = {
    "patient_id",
    "condition_id",
    "case_definition",
    "diagnosis_code",
    "outcome",
    "type_of_resistance",
    "comorbidity",
    "country",
    "imagingstudy_id",
    "imaging_date",
    "series_instance_content_url",
    "series_modality_cd",
    "cxr_outlier"
}


required_genomic_columns = {
    "condition_id"
}


missing_cxr_columns = (
    required_cxr_columns -
    set(cxr.columns)
)

missing_genomic_columns = (
    required_genomic_columns -
    set(genomics.columns)
)


if missing_cxr_columns:

    raise ValueError(
        "Missing required CXR columns:\n"
        + "\n".join(
            sorted(missing_cxr_columns)
        )
    )


if missing_genomic_columns:

    raise ValueError(
        "Missing required genomic columns:\n"
        + "\n".join(
            sorted(missing_genomic_columns)
        )
    )


print(
    "CXR required columns       : PASS"
)

print(
    "Genomic required columns   : PASS"
)


# =============================================================================
# 6. ESTABLISH TB PORTALS INTERNAL BASELINE
# =============================================================================

print("\n3. TB PORTALS INTERNAL BASELINE")
print("-" * 80)


cxr_conditions = set(
    cxr["condition_id"]
    .dropna()
    .astype(str)
    .str.strip()
)

genomic_conditions = set(
    genomics["condition_id"]
    .dropna()
    .astype(str)
    .str.strip()
)

common_conditions = (
    cxr_conditions &
    genomic_conditions
)


print(
    f"CXR unique conditions       : {len(cxr_conditions):,}"
)

print(
    f"Genomic unique conditions   : {len(genomic_conditions):,}"
)

print(
    f"Common CXR+genomic cases    : {len(common_conditions):,}"
)


# =============================================================================
# 7. NORMAL VOCABULARY
# =============================================================================
#
# These are SEARCH TERMS ONLY.
#
# A keyword match does NOT automatically establish a Normal case.
#
# =============================================================================

NORMAL_TERMS = {

    "NORMAL":
        "normal",

    "HEALTHY":
        "healthy",

    "CONTROL":
        "control",

    "TB_NEGATIVE":
        "tb negative",

    "TB_NEG":
        "tb neg",

    "NON_TB":
        "non tb",

    "NO_TB":
        "no tb",

    "NO_EVIDENCE_TB":
        "no evidence tb",

    "TB_ABSENT":
        "tb absent",

    "NEGATIVE_FOR_TB":
        "negative for tb",

    "TB_FREE":
        "tb free",

    "NON_TUBERCULOUS":
        "non tuberculous"
}


# =============================================================================
# 8. SEARCH ALL CXR TEXT-LIKE COLUMNS
# =============================================================================

print("\n4. SEARCHING TB PORTALS CXR METADATA")
print("-" * 80)


cxr_text_columns = []

for column in cxr.columns:

    dtype = cxr[column].dtype

    if (
        pd.api.types.is_object_dtype(dtype)
        or
        pd.api.types.is_string_dtype(dtype)
        or
        pd.api.types.is_categorical_dtype(dtype)
    ):

        cxr_text_columns.append(
            column
        )


print(
    "Text-like CXR columns:"
)

for column in cxr_text_columns:

    print(
        f"  - {column}"
    )


# =============================================================================
# 9. SEARCH CXR COLUMNS
# =============================================================================

cxr_normal_evidence = []


for column in cxr_text_columns:

    values = (
        cxr[column]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    lower_values = (
        values
        .str.lower()
    )


    for term_name, term_value in NORMAL_TERMS.items():

        mask = (
            lower_values
            .str.contains(
                term_value,
                regex=False,
                na=False
            )
        )


        matched_indices = (
            cxr.index[mask]
        )


        for idx in matched_indices:

            value = values.loc[idx]

            if value == "":
                continue


            row = cxr.loc[idx]

            cxr_normal_evidence.append({

                "source":
                    "TB_Portals_CXR",

                "row_index":
                    int(idx),

                "condition_id":
                    str(row["condition_id"]),

                "patient_id":
                    str(row["patient_id"]),

                "column":
                    column,

                "observed_value":
                    value,

                "matched_term":
                    term_name,

                "matched_text":
                    term_value

            })


cxr_normal_evidence_df = pd.DataFrame(
    cxr_normal_evidence
)


print(
    f"\nCXR keyword evidence rows : "
    f"{len(cxr_normal_evidence_df):,}"
)


if not cxr_normal_evidence_df.empty:

    cxr_normal_evidence_df.to_csv(
        OUTPUT_DIR /
        "01_CXR_Normal_Keyword_Evidence.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(
        "CXR evidence file          : SAVED"
    )

else:

    print(
        "CXR Normal keyword evidence: NONE"
    )


# =============================================================================
# 10. SEARCH ALL GENOMIC TEXT-LIKE COLUMNS
# =============================================================================

print("\n5. SEARCHING TB PORTALS GENOMIC METADATA")
print("-" * 80)


genomic_text_columns = []

for column in genomics.columns:

    dtype = genomics[column].dtype

    if (
        pd.api.types.is_object_dtype(dtype)
        or
        pd.api.types.is_string_dtype(dtype)
        or
        pd.api.types.is_categorical_dtype(dtype)
    ):

        genomic_text_columns.append(
            column
        )


print(
    "Text-like genomic columns:"
)

for column in genomic_text_columns:

    print(
        f"  - {column}"
    )


genomic_normal_evidence = []


for column in genomic_text_columns:

    values = (
        genomics[column]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    lower_values = (
        values
        .str.lower()
    )


    for term_name, term_value in NORMAL_TERMS.items():

        mask = (
            lower_values
            .str.contains(
                term_value,
                regex=False,
                na=False
            )
        )


        matched_indices = (
            genomics.index[mask]
        )


        for idx in matched_indices:

            value = values.loc[idx]

            if value == "":
                continue


            row = genomics.loc[idx]

            genomic_normal_evidence.append({

                "source":
                    "TB_Portals_Genomics",

                "row_index":
                    int(idx),

                "condition_id":
                    str(row["condition_id"]),

                "column":
                    column,

                "observed_value":
                    value,

                "matched_term":
                    term_name,

                "matched_text":
                    term_value

            })


genomic_normal_evidence_df = pd.DataFrame(
    genomic_normal_evidence
)


print(
    f"\nGenomic keyword evidence rows : "
    f"{len(genomic_normal_evidence_df):,}"
)


if not genomic_normal_evidence_df.empty:

    genomic_normal_evidence_df.to_csv(
        OUTPUT_DIR /
        "02_Genomic_Normal_Keyword_Evidence.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(
        "Genomic evidence file          : SAVED"
    )

else:

    print(
        "Genomic Normal keyword evidence: NONE"
    )


# =============================================================================
# 11. CONDITION-LEVEL NORMAL CXR CANDIDATES
# =============================================================================

print("\n6. CONDITION-LEVEL CXR NORMAL CANDIDATES")
print("-" * 80)


if cxr_normal_evidence_df.empty:

    cxr_normal_conditions = set()

else:

    cxr_normal_conditions = set(
        cxr_normal_evidence_df[
            "condition_id"
        ]
        .dropna()
        .astype(str)
        .str.strip()
    )


print(
    f"Candidate CXR Normal conditions : "
    f"{len(cxr_normal_conditions):,}"
)


# =============================================================================
# 12. CONDITION-LEVEL NORMAL GENOMIC CANDIDATES
# =============================================================================

if genomic_normal_evidence_df.empty:

    genomic_normal_conditions = set()

else:

    genomic_normal_conditions = set(
        genomic_normal_evidence_df[
            "condition_id"
        ]
        .dropna()
        .astype(str)
        .str.strip()
    )


print(
    f"Candidate genomic Normal conditions: "
    f"{len(genomic_normal_conditions):,}"
)


# =============================================================================
# 13. SAME-CONDITION NORMAL CXR + GENOMIC OVERLAP
# =============================================================================

paired_normal_candidate_conditions = (
    cxr_normal_conditions &
    genomic_normal_conditions
)


print(
    f"Same-condition Normal CXR+genomic "
    f"candidates: {len(paired_normal_candidate_conditions):,}"
)


# =============================================================================
# 14. IMPORTANT: AUDIT WHETHER CXR NORMAL MATCHES ARE ONLY cxr_outlier
# =============================================================================

print("\n7. CXR NORMAL-CANDIDATE SOURCE AUDIT")
print("-" * 80)


if cxr_normal_evidence_df.empty:

    cxr_candidate_columns = pd.DataFrame()

else:

    cxr_candidate_columns = (
        cxr_normal_evidence_df[
            [
                "column",
                "matched_term"
            ]
        ]
        .drop_duplicates()
        .sort_values(
            [
                "column",
                "matched_term"
            ]
        )
    )


if not cxr_candidate_columns.empty:

    print(
        cxr_candidate_columns.to_string(
            index=False
        )
    )

else:

    print(
        "No CXR Normal keyword evidence found."
    )


only_outlier_source = False


if not cxr_normal_evidence_df.empty:

    evidence_columns = set(
        cxr_normal_evidence_df[
            "column"
        ]
        .astype(str)
    )

    only_outlier_source = (
        evidence_columns == {
            "cxr_outlier"
        }
    )


print(
    "\nOnly cxr_outlier source? : "
    f"{only_outlier_source}"
)


# =============================================================================
# 15. CLINICAL FIELD AUDIT
# =============================================================================
#
# A Normal classification must not be inferred merely from a technical
# imaging-quality field.
#
# Therefore, explicitly inspect the clinical/diagnostic fields.
#
# =============================================================================

clinical_columns = [
    "case_definition",
    "diagnosis_code",
    "outcome",
    "type_of_resistance",
    "comorbidity"
]


clinical_evidence_rows = []


for column in clinical_columns:

    values = (
        cxr[column]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    lower_values = (
        values
        .str.lower()
    )


    for term_name, term_value in NORMAL_TERMS.items():

        mask = (
            lower_values
            .str.contains(
                term_value,
                regex=False,
                na=False
            )
        )


        for idx in cxr.index[mask]:

            row = cxr.loc[idx]

            clinical_evidence_rows.append({

                "row_index":
                    int(idx),

                "condition_id":
                    str(row["condition_id"]),

                "patient_id":
                    str(row["patient_id"]),

                "column":
                    column,

                "observed_value":
                    values.loc[idx],

                "matched_term":
                    term_name

            })


clinical_evidence_df = pd.DataFrame(
    clinical_evidence_rows
)


print(
    f"Clinical Normal keyword evidence rows : "
    f"{len(clinical_evidence_df):,}"
)


if not clinical_evidence_df.empty:

    clinical_evidence_df.to_csv(
        OUTPUT_DIR /
        "03_Clinical_Normal_Candidate_Evidence.csv",
        index=False,
        encoding="utf-8-sig"
    )


# =============================================================================
# 16. DIAGNOSTIC-CODE AUDIT
# =============================================================================

print("\n8. DIAGNOSTIC CODE AUDIT")
print("-" * 80)


diagnosis_values = (
    cxr["diagnosis_code"]
    .fillna("MISSING")
    .astype(str)
    .str.strip()
)


diagnosis_summary = (
    diagnosis_values
    .value_counts(dropna=False)
    .rename_axis("diagnosis_code")
    .reset_index(
        name="record_count"
    )
)


diagnosis_summary.to_csv(
    OUTPUT_DIR /
    "04_CXR_Diagnosis_Code_Distribution.csv",
    index=False,
    encoding="utf-8-sig"
)


print(
    diagnosis_summary.head(30).to_string(
        index=False
    )
)


# =============================================================================
# 17. CASE-DEFINITION AUDIT
# =============================================================================

print("\n9. CASE-DEFINITION AUDIT")
print("-" * 80)


case_definition_summary = (
    cxr["case_definition"]
    .fillna("MISSING")
    .astype(str)
    .str.strip()
    .value_counts(dropna=False)
    .rename_axis("case_definition")
    .reset_index(
        name="record_count"
    )
)


case_definition_summary.to_csv(
    OUTPUT_DIR /
    "05_CXR_Case_Definition_Distribution.csv",
    index=False,
    encoding="utf-8-sig"
)


print(
    case_definition_summary.to_string(
        index=False
    )
)


# =============================================================================
# 18. OUTCOME AUDIT
# =============================================================================

print("\n10. OUTCOME AUDIT")
print("-" * 80)


outcome_summary = (
    cxr["outcome"]
    .fillna("MISSING")
    .astype(str)
    .str.strip()
    .value_counts(dropna=False)
    .rename_axis("outcome")
    .reset_index(
        name="record_count"
    )
)


outcome_summary.to_csv(
    OUTPUT_DIR /
    "06_CXR_Outcome_Distribution.csv",
    index=False,
    encoding="utf-8-sig"
)


print(
    outcome_summary.to_string(
        index=False
    )
)


# =============================================================================
# 19. TYPE OF RESISTANCE AUDIT
# =============================================================================

print("\n11. TYPE OF RESISTANCE AUDIT")
print("-" * 80)


resistance_summary = (
    cxr["type_of_resistance"]
    .fillna("MISSING")
    .astype(str)
    .str.strip()
    .value_counts(dropna=False)
    .rename_axis("type_of_resistance")
    .reset_index(
        name="record_count"
    )
)


resistance_summary.to_csv(
    OUTPUT_DIR /
    "07_CXR_Type_of_Resistance_Distribution.csv",
    index=False,
    encoding="utf-8-sig"
)


print(
    resistance_summary.to_string(
        index=False
    )
)


# =============================================================================
# 20. NORMAL CANDIDATES WITH TB PORTALS GENOMIC RECORDS
# =============================================================================

print("\n12. NORMAL CANDIDATE / GENOMIC OVERLAP")
print("-" * 80)


if cxr_normal_conditions:

    normal_cxr_with_genomics = (
        cxr_normal_conditions &
        genomic_conditions
    )

else:

    normal_cxr_with_genomics = set()


print(
    f"CXR Normal candidate conditions : "
    f"{len(cxr_normal_conditions):,}"
)

print(
    f"Those with genomic records       : "
    f"{len(normal_cxr_with_genomics):,}"
)


# =============================================================================
# 21. DISTINGUISH TECHNICAL NORMAL FROM CLINICAL NORMAL
# =============================================================================

technical_normal_conditions = set()

if "cxr_outlier" in cxr.columns:

    outlier_normal_mask = (
        cxr["cxr_outlier"]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.lower()
        .eq("normal")
    )

    technical_normal_conditions = set(
        cxr.loc[
            outlier_normal_mask,
            "condition_id"
        ]
        .dropna()
        .astype(str)
        .str.strip()
    )


print("\n13. TECHNICAL VS CLINICAL NORMAL")
print("-" * 80)

print(
    f"cxr_outlier=normal conditions : "
    f"{len(technical_normal_conditions):,}"
)

print(
    f"Clinical Normal candidates     : "
    f"{len(cxr_normal_conditions - technical_normal_conditions):,}"
)

print(
    f"Total keyword CXR candidates   : "
    f"{len(cxr_normal_conditions):,}"
)


# =============================================================================
# 22. NORMAL MULTIMODAL REQUIREMENT MATRIX
# =============================================================================

requirement_matrix = pd.DataFrame([

    {
        "requirement_id":
            "REQ-54-01",

        "requirement":
            "Explicit Normal/Healthy/Control CXR label",

        "verified":
            len(cxr_normal_conditions) > 0,

        "count":
            len(cxr_normal_conditions),

        "interpretation":
            (
                "Keyword candidates only; requires semantic "
                "verification before acceptance."
            )
    },

    {
        "requirement_id":
            "REQ-54-02",

        "requirement":
            "Explicit Normal genomic class",

        "verified":
            len(genomic_normal_conditions) > 0,

        "count":
            len(genomic_normal_conditions),

        "interpretation":
            (
                "Keyword candidates only; requires semantic "
                "verification before acceptance."
            )
    },

    {
        "requirement_id":
            "REQ-54-03",

        "requirement":
            "Same-condition Normal CXR + genomic pairing",

        "verified":
            len(paired_normal_candidate_conditions) > 0,

        "count":
            len(paired_normal_candidate_conditions),

        "interpretation":
            (
                "Overlap of keyword candidates is not itself "
                "proof of a valid Normal clinical class."
            )
    },

    {
        "requirement_id":
            "REQ-54-04",

        "requirement":
            "Normal label is clinically diagnostic rather than technical",

        "verified":
            not only_outlier_source,

        "count":
            len(clinical_evidence_df),

        "interpretation":
            (
                "cxr_outlier is treated as a technical imaging-quality "
                "field and cannot establish Normal diagnosis."
            )
    },

    {
        "requirement_id":
            "REQ-54-05",

        "requirement":
            "No artificial Normal labels",

        "verified":
            True,

        "count":
            0,

        "interpretation":
            "No artificial labels created."
    },

    {
        "requirement_id":
            "REQ-54-06",

        "requirement":
            "No missing-genomics-as-Normal inference",

        "verified":
            True,

        "count":
            0,

        "interpretation":
            "Missing genomic modality is not treated as Normal."
    },

    {
        "requirement_id":
            "REQ-54-07",

        "requirement":
            "No cross-subject multimodal pairing",

        "verified":
            True,

        "count":
            0,

        "interpretation":
            "No unrelated subjects were paired."
    },

    {
        "requirement_id":
            "REQ-54-08",

        "requirement":
            "No preprocessing or modeling",

        "verified":
            True,

        "count":
            0,

        "interpretation":
            "Downstream processing remains locked."
    }

])


requirement_matrix.to_csv(
    OUTPUT_DIR /
    "08_Normal_Multimodal_Requirement_Matrix.csv",
    index=False,
    encoding="utf-8-sig"
)


# =============================================================================
# 23. SOURCE-LEVEL DECISION STATUS
# =============================================================================
#
# IMPORTANT:
# Keyword candidates are NOT accepted as verified Normal cases.
#
# Therefore the current TB Portals internal status remains:
#
# NO_VERIFIED_PAIRED_NORMAL_CLASS
#
# =============================================================================

if len(paired_normal_candidate_conditions) > 0:

    source_status = (
        "CANDIDATE_NORMAL_OVERLAP_FOUND_REQUIRES_MANUAL_SEMANTIC_VERIFICATION"
    )

elif (
    len(cxr_normal_conditions) > 0
    or
    len(genomic_normal_conditions) > 0
):

    source_status = (
        "NORMAL_KEYWORD_CANDIDATES_FOUND_BUT_PAIRED_CLASS_NOT_VERIFIED"
    )

else:

    source_status = (
        "NO_NORMAL_KEYWORD_CANDIDATES_FOUND"
    )


# =============================================================================
# 24. SCIENTIFIC STATUS
# =============================================================================

if (
    len(paired_normal_candidate_conditions) > 0
):

    final_internal_status = (
        "CANDIDATE_PAIRED_NORMAL_SIGNAL_FOUND_NOT_YET_ACCEPTED"
    )

else:

    final_internal_status = (
        "NO_VERIFIED_PAIRED_NORMAL_CLASS"
    )


# =============================================================================
# 25. FINAL DECISION GATE
# =============================================================================

decision_gate = pd.DataFrame([

    {
        "decision_id":
            "DEC-54-01",

        "question":
            "Does TB Portals contain a verified Normal CXR class?",

        "status":
            (
                "CANDIDATE_SIGNAL_ONLY"
                if len(cxr_normal_conditions) > 0
                else
                "NO"
            ),

        "evidence":
            len(cxr_normal_conditions),

        "final_interpretation":
            (
                "Keyword candidates require semantic verification."
                if len(cxr_normal_conditions) > 0
                else
                "No Normal CXR keyword candidates identified."
            )
    },

    {
        "decision_id":
            "DEC-54-02",

        "question":
            "Does TB Portals contain a verified Normal genomic class?",

        "status":
            (
                "CANDIDATE_SIGNAL_ONLY"
                if len(genomic_normal_conditions) > 0
                else
                "NO"
            ),

        "evidence":
            len(genomic_normal_conditions),

        "final_interpretation":
            (
                "Keyword candidates require semantic verification."
                if len(genomic_normal_conditions) > 0
                else
                "No Normal genomic keyword candidates identified."
            )
    },

    {
        "decision_id":
            "DEC-54-03",

        "question":
            "Is a same-condition Normal CXR+genomic class verified?",

        "status":
            (
                "CANDIDATE_OVERLAP_ONLY"
                if len(paired_normal_candidate_conditions) > 0
                else
                "NO"
            ),

        "evidence":
            len(paired_normal_candidate_conditions),

        "final_interpretation":
            (
                "Overlap does not establish clinical Normal."
                if len(paired_normal_candidate_conditions) > 0
                else
                "No paired Normal candidate overlap."
            )
    },

    {
        "decision_id":
            "DEC-54-04",

        "question":
            "Can cxr_outlier=normal define Normal diagnosis?",

        "status":
            "NO",

        "evidence":
            len(technical_normal_conditions),

        "final_interpretation":
            "Technical outlier/QC field cannot establish diagnosis."
    },

    {
        "decision_id":
            "DEC-54-05",

        "question":
            "Can missing genomics define Normal?",

        "status":
            "NO",

        "evidence":
            0,

        "final_interpretation":
            "Missing modality is not a disease label."
    },

    {
        "decision_id":
            "DEC-54-06",

        "question":
            "Can unrelated subjects be paired?",

        "status":
            "NO",

        "evidence":
            0,

        "final_interpretation":
            "Same-subject linkage is mandatory."
    },

    {
        "decision_id":
            "DEC-54-07",

        "question":
            "Is final TB-vs-Normal multimodal cohort locked?",

        "status":
            "NO",

        "evidence":
            0,

        "final_interpretation":
            "Requires verified paired Normal cohort."
    }

])


decision_gate.to_csv(
    OUTPUT_DIR /
    "09_Final_Decision_Gate.csv",
    index=False,
    encoding="utf-8-sig"
)


# =============================================================================
# 26. SCIENTIFIC LOCKS
# =============================================================================

scientific_locks = pd.DataFrame([

    {
        "lock":
            "Artificial Normal labels",

        "status":
            "LOCKED",

        "violated":
            False
    },

    {
        "lock":
            "Missing genomics treated as Normal",

        "status":
            "LOCKED",

        "violated":
            False
    },

    {
        "lock":
            "cxr_outlier=normal treated as Normal diagnosis",

        "status":
            "LOCKED",

        "violated":
            False
    },

    {
        "lock":
            "Absence of TB interpreted as Normal",

        "status":
            "LOCKED",

        "violated":
            False
    },

    {
        "lock":
            "Cross-subject multimodal pairing",

        "status":
            "LOCKED",

        "violated":
            False
    },

    {
        "lock":
            "Target leakage",

        "status":
            "LOCKED",

        "violated":
            False
    },

    {
        "lock":
            "Preprocessing before cohort lock",

        "status":
            "LOCKED",

        "violated":
            False
    },

    {
        "lock":
            "Lung segmentation before cohort lock",

        "status":
            "LOCKED",

        "violated":
            False
    },

    {
        "lock":
            "Model training before target lock",

        "status":
            "LOCKED",

        "violated":
            False
    },

    {
        "lock":
            "XAI before model development",

        "status":
            "LOCKED",

        "violated":
            False
    }

])


scientific_locks.to_csv(
    OUTPUT_DIR /
    "10_Scientific_Locks.csv",
    index=False,
    encoding="utf-8-sig"
)


# =============================================================================
# 27. FINAL INVESTIGATION RECORD
# =============================================================================

investigation_record = {

    "step":
        "3B.6.54",

    "title":
        "Paired Normal Multimodal Data Source Investigation",

    "timestamp_utc":
        datetime.utcnow().isoformat() + "Z",

    "source_scope":
        "TB Portals March 2025 internal dataset only",

    "external_datasets_used":
        False,

    "research_question":
        (
            "Can Tuberculosis versus Normal be predicted using paired "
            "Chest X-ray imaging and genomic information from the same "
            "patient-level case, with lung segmentation and explainable AI?"
        ),

    "verified_tb_multimodal_cases":
        len(common_conditions),

    "keyword_candidate_counts":
        {
            "cxr_normal_conditions":
                len(cxr_normal_conditions),

            "genomic_normal_conditions":
                len(genomic_normal_conditions),

            "paired_normal_candidate_conditions":
                len(paired_normal_candidate_conditions)
        },

    "technical_cxr_outlier_normal_conditions":
        len(technical_normal_conditions),

    "only_outlier_source":
        only_outlier_source,

    "clinical_normal_keyword_evidence_rows":
        len(clinical_evidence_df),

    "source_status":
        source_status,

    "final_internal_status":
        final_internal_status,

    "verified_paired_normal_class":
        False,

    "final_tb_vs_normal_cohort_locked":
        False,

    "preprocessing_authorized":
        False,

    "lung_segmentation_authorized":
        False,

    "model_training_authorized":
        False,

    "xai_authorized":
        False,

    "scientific_locks":
        {
            "artificial_normal_labels":
                True,

            "missing_genomics_as_normal":
                True,

            "cxr_outlier_normal_as_diagnosis":
                True,

            "absence_of_tb_as_normal":
                True,

            "cross_subject_pairing":
                True,

            "target_leakage":
                True
        },

    "next_action":
        (
            "If candidate Normal signals exist, perform explicit semantic "
            "source verification before accepting any Normal case. "
            "Otherwise retain the blocked design status and make an "
            "explicit research-design decision."
        )
}


json_path = (
    OUTPUT_DIR /
    "11_Paired_Normal_Multimodal_Investigation_Record.json"
)


with open(
    json_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        investigation_record,
        f,
        indent=4,
        ensure_ascii=False
    )


# =============================================================================
# 28. FINAL REPORT
# =============================================================================

print("\n")
print("=" * 80)
print("STEP 3B.6.54 — FINAL INVESTIGATION REPORT")
print("=" * 80)


print("\nRESEARCH QUESTION")
print("-" * 80)

print(
    "Can Tuberculosis versus Normal be predicted using paired "
    "Chest X-ray imaging and genomic information from the same "
    "patient-level case, with lung segmentation and explainable AI?"
)


print("\nTB PORTALS INTERNAL EVIDENCE")
print("-" * 80)

print(
    f"TB CXR + genomic common conditions : "
    f"{len(common_conditions):,}"
)

print(
    f"CXR Normal keyword candidates     : "
    f"{len(cxr_normal_conditions):,}"
)

print(
    f"Genomic Normal keyword candidates : "
    f"{len(genomic_normal_conditions):,}"
)

print(
    f"Paired Normal keyword candidates  : "
    f"{len(paired_normal_candidate_conditions):,}"
)

print(
    f"cxr_outlier=normal conditions     : "
    f"{len(technical_normal_conditions):,}"
)

print(
    f"Clinical Normal evidence rows     : "
    f"{len(clinical_evidence_df):,}"
)


print("\nSOURCE INTERPRETATION")
print("-" * 80)

print(
    f"Only CXR Normal signal from cxr_outlier? : "
    f"{only_outlier_source}"
)

print(
    f"Verified paired Normal class             : "
    f"False"
)


print("\nFINAL INTERNAL STATUS")
print("-" * 80)

print(
    final_internal_status
)


print("\nFINAL TB-VS-NORMAL MULTIMODAL COHORT")
print("-" * 80)

print(
    "NOT LOCKED"
)


print("\nDOWNSTREAM AUTHORIZATION")
print("-" * 80)

print(
    "Image preprocessing : NOT AUTHORIZED"
)

print(
    "Lung segmentation   : NOT AUTHORIZED"
)

print(
    "Multimodal model    : NOT AUTHORIZED"
)

print(
    "Explainable AI      : NOT AUTHORIZED"
)


print("\nSCIENTIFIC LOCKS")
print("-" * 80)

for _, row in scientific_locks.iterrows():

    print(
        f"{row['lock']:<50}: {row['status']}"
    )


print("\nOUTPUT FILES")
print("-" * 80)

print(
    "01_CXR_Normal_Keyword_Evidence.csv"
)

print(
    "02_Genomic_Normal_Keyword_Evidence.csv"
)

print(
    "03_Clinical_Normal_Candidate_Evidence.csv"
)

print(
    "04_CXR_Diagnosis_Code_Distribution.csv"
)

print(
    "05_CXR_Case_Definition_Distribution.csv"
)

print(
    "06_CXR_Outcome_Distribution.csv"
)

print(
    "07_CXR_Type_of_Resistance_Distribution.csv"
)

print(
    "08_Normal_Multimodal_Requirement_Matrix.csv"
)

print(
    "09_Final_Decision_Gate.csv"
)

print(
    "10_Scientific_Locks.csv"
)

print(
    "11_Paired_Normal_Multimodal_Investigation_Record.json"
)


print("\nOUTPUT DIRECTORY")
print("-" * 80)

print(
    OUTPUT_DIR
)


print("\n" + "=" * 80)
print("STEP 3B.6.54 — INVESTIGATION COMPLETE")
print("=" * 80)

print(
    "\nIMPORTANT:"
)

print(
    "Keyword candidates are evidence for investigation only."
)

print(
    "They are NOT automatically accepted as Normal cases."
)

print(
    "No final cohort has been created."
)

print(
    "No downstream processing has been authorized."
)

print("=" * 80)
```

```python
# =============================================================================
# STEP 3B.6.55
# GENOMIC NORMAL CANDIDATE FORENSIC REVIEW
# =============================================================================
#
# PURPOSE
# -------
# Step 3B.6.54 found:
#
#   CXR Normal keyword candidates     : 10,842
#   Genomic Normal keyword candidates : 9
#   Paired Normal candidates          : 0
#
# The 9 genomic keyword matches MUST be investigated before final closure.
#
# IMPORTANT:
# ----------
# A keyword match is NOT a Normal diagnosis.
#
# This step:
#   - identifies the exact 9 records
#   - identifies the exact matching column/value
#   - examines the corresponding condition
#   - checks CXR linkage
#   - checks clinical TB evidence
#   - checks genomic metadata
#   - produces evidence for human/scientific adjudication
#
# This step DOES NOT:
#   - create Normal labels
#   - create multimodal pairs
#   - alter source data
#   - preprocess images
#   - segment lungs
#   - train models
#   - perform XAI
#
# =============================================================================


from pathlib import Path
import pandas as pd
import numpy as np
import json
from datetime import datetime


# =============================================================================
# 1. CONFIGURATION
# =============================================================================

BASE_DIR = Path(
    r"C:\TBP\Metadata"
)

STEP_54_DIR = (
    BASE_DIR /
    "Step_3B_6_54_Paired_Normal_Multimodal_Data_Source_Investigation"
)

CXR_PATH = (
    BASE_DIR /
    "TB_Portals_CXRs_March_2025.csv"
)

GENOMICS_PATH = (
    BASE_DIR /
    "TB_Portals_Genomics_March_2025.csv"
)

OUTPUT_DIR = (
    BASE_DIR /
    "Step_3B_6_55_Genomic_Normal_Candidate_Forensic_Review"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# =============================================================================
# 2. VERIFY INPUTS
# =============================================================================

print("=" * 80)
print("STEP 3B.6.55 — GENOMIC NORMAL CANDIDATE FORENSIC REVIEW")
print("=" * 80)


required_inputs = {

    "Step 3B.6.54 genomic evidence":
        STEP_54_DIR /
        "02_Genomic_Normal_Keyword_Evidence.csv",

    "CXR source":
        CXR_PATH,

    "Genomics source":
        GENOMICS_PATH
}


missing = []

for label, path in required_inputs.items():

    if path.exists():

        print(
            f"{label:<40}: FOUND"
        )

    else:

        print(
            f"{label:<40}: MISSING"
        )

        missing.append(
            f"{label}: {path}"
        )


if missing:

    raise FileNotFoundError(
        "Required input(s) missing:\n"
        + "\n".join(missing)
    )


# =============================================================================
# 3. LOAD DATA
# =============================================================================

print("\n1. LOADING DATA")
print("-" * 80)


genomic_evidence = pd.read_csv(
    required_inputs[
        "Step 3B.6.54 genomic evidence"
    ],
    dtype=str
)

cxr = pd.read_csv(
    CXR_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)


print(
    f"Genomic keyword evidence rows : "
    f"{len(genomic_evidence):,}"
)

print(
    f"Genomic source records         : "
    f"{len(genomics):,}"
)

print(
    f"CXR source records             : "
    f"{len(cxr):,}"
)


# =============================================================================
# 4. REQUIRED COLUMN VALIDATION
# =============================================================================

print("\n2. VALIDATING EVIDENCE STRUCTURE")
print("-" * 80)


required_evidence_columns = {
    "source",
    "row_index",
    "condition_id",
    "column",
    "observed_value",
    "matched_term",
    "matched_text"
}


missing_evidence_columns = (
    required_evidence_columns -
    set(genomic_evidence.columns)
)


if missing_evidence_columns:

    raise ValueError(
        "Missing required evidence columns:\n"
        + "\n".join(
            sorted(missing_evidence_columns)
        )
    )


required_genomic_columns = {
    "condition_id"
}


missing_genomic_columns = (
    required_genomic_columns -
    set(genomics.columns)
)


if missing_genomic_columns:

    raise ValueError(
        "Missing genomic columns:\n"
        + "\n".join(
            sorted(missing_genomic_columns)
        )
    )


required_cxr_columns = {
    "condition_id",
    "patient_id",
    "case_definition",
    "diagnosis_code",
    "outcome",
    "type_of_resistance",
    "comorbidity",
    "country",
    "cxr_outlier"
}


missing_cxr_columns = (
    required_cxr_columns -
    set(cxr.columns)
)


if missing_cxr_columns:

    raise ValueError(
        "Missing CXR columns:\n"
        + "\n".join(
            sorted(missing_cxr_columns)
        )
    )


print(
    "Evidence columns : PASS"
)

print(
    "Genomic columns  : PASS"
)

print(
    "CXR columns      : PASS"
)


# =============================================================================
# 5. BASIC AUDIT OF THE 9 MATCHES
# =============================================================================

print("\n3. GENOMIC KEYWORD MATCHES")
print("-" * 80)


display_columns = [
    "row_index",
    "condition_id",
    "column",
    "observed_value",
    "matched_term",
    "matched_text"
]


print(
    genomic_evidence[
        display_columns
    ].to_string(
        index=False
    )
)


# =============================================================================
# 6. SAVE RAW FORENSIC EVIDENCE
# =============================================================================

raw_evidence_path = (
    OUTPUT_DIR /
    "01_Raw_Genomic_Normal_Keyword_Evidence.csv"
)


genomic_evidence.to_csv(
    raw_evidence_path,
    index=False,
    encoding="utf-8-sig"
)


print(
    f"\nRaw evidence saved : {raw_evidence_path}"
)


# =============================================================================
# 7. UNIQUE CONDITION IDS
# =============================================================================

candidate_condition_ids = (
    genomic_evidence[
        "condition_id"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)

candidate_condition_ids = [
    x
    for x in candidate_condition_ids.unique()
    if x
]


print("\n4. UNIQUE CANDIDATE CONDITIONS")
print("-" * 80)

print(
    f"Unique candidate conditions : "
    f"{len(candidate_condition_ids):,}"
)

for condition_id in candidate_condition_ids:

    print(
        f"  {condition_id}"
    )


# =============================================================================
# 8. EXACT GENOMIC RECORDS
# =============================================================================

print("\n5. EXACT GENOMIC RECORD REVIEW")
print("-" * 80)


genomics_condition = (
    genomics[
        genomics["condition_id"]
        .astype(str)
        .str.strip()
        .isin(
            candidate_condition_ids
        )
    ]
    .copy()
)


print(
    f"Genomic records belonging to candidate conditions : "
    f"{len(genomics_condition):,}"
)


# =============================================================================
# 9. CREATE FORENSIC SUMMARY
# =============================================================================

forensic_rows = []


for _, evidence_row in genomic_evidence.iterrows():

    condition_id = str(
        evidence_row["condition_id"]
    ).strip()

    column = str(
        evidence_row["column"]
    ).strip()

    observed_value = str(
        evidence_row["observed_value"]
    )

    matched_term = str(
        evidence_row["matched_term"]
    )

    matched_text = str(
        evidence_row["matched_text"]
    )


    condition_genomics = genomics[
        genomics["condition_id"]
        .astype(str)
        .str.strip()
        .eq(condition_id)
    ].copy()


    condition_cxr = cxr[
        cxr["condition_id"]
        .astype(str)
        .str.strip()
        .eq(condition_id)
    ].copy()


    # -------------------------------------------------------------------------
    # CXR linkage
    # -------------------------------------------------------------------------

    has_cxr = (
        not condition_cxr.empty
    )


    cxr_record_count = (
        len(condition_cxr)
    )


    # -------------------------------------------------------------------------
    # Patient IDs
    # -------------------------------------------------------------------------

    if has_cxr:

        cxr_patient_ids = (
            condition_cxr[
                "patient_id"
            ]
            .dropna()
            .astype(str)
            .str.strip()
            .unique()
            .tolist()
        )

    else:

        cxr_patient_ids = []


    # -------------------------------------------------------------------------
    # Clinical fields
    # -------------------------------------------------------------------------

    if has_cxr:

        case_definitions = (
            condition_cxr[
                "case_definition"
            ]
            .fillna("")
            .astype(str)
            .str.strip()
            .unique()
            .tolist()
        )

        diagnosis_codes = (
            condition_cxr[
                "diagnosis_code"
            ]
            .fillna("")
            .astype(str)
            .str.strip()
            .unique()
            .tolist()
        )

        outcomes = (
            condition_cxr[
                "outcome"
            ]
            .fillna("")
            .astype(str)
            .str.strip()
            .unique()
            .tolist()
        )

        resistance_types = (
            condition_cxr[
                "type_of_resistance"
            ]
            .fillna("")
            .astype(str)
            .str.strip()
            .unique()
            .tolist()
        )

        comorbidities = (
            condition_cxr[
                "comorbidity"
            ]
            .fillna("")
            .astype(str)
            .str.strip()
            .unique()
            .tolist()
        )

        countries = (
            condition_cxr[
                "country"
            ]
            .fillna("")
            .astype(str)
            .str.strip()
            .unique()
            .tolist()
        )

        cxr_outlier_values = (
            condition_cxr[
                "cxr_outlier"
            ]
            .fillna("")
            .astype(str)
            .str.strip()
            .unique()
            .tolist()
        )

    else:

        case_definitions = []
        diagnosis_codes = []
        outcomes = []
        resistance_types = []
        comorbidities = []
        countries = []
        cxr_outlier_values = []


    # -------------------------------------------------------------------------
    # Explicit TB diagnostic-code test
    # -------------------------------------------------------------------------

    tb_code_values = []

    for code in diagnosis_codes:

        code_upper = str(code).upper().strip()

        if (
            code_upper.startswith("A15")
            or
            code_upper.startswith("A16")
            or
            code_upper.startswith("A17")
            or
            code_upper.startswith("A18")
            or
            code_upper.startswith("A19")
        ):

            tb_code_values.append(
                code
            )


    has_tb_diagnostic_code = (
        len(tb_code_values) > 0
    )


    # -------------------------------------------------------------------------
    # Genomic metadata
    # -------------------------------------------------------------------------

    genomic_columns_to_capture = [

        "condition_id",
        "sex",
        "type_of_resistance",
        "outcome",
        "specimen_id",
        "specimen_collection_site",
        "sra_id",
        "ncbi_sra",
        "ncbi_sourceorganism",
        "ncbi_bioproject",
        "ncbi_biosample",
        "main_lineage",
        "sub_lineage",
        "drug_resistance_type",
        "gene_snp_mutations"
    ]


    available_genomic_capture_columns = [

        c
        for c in genomic_columns_to_capture
        if c in condition_genomics.columns
    ]


    genomic_values = {}


    for c in available_genomic_capture_columns:

        genomic_values[c] = (
            condition_genomics[c]
            .fillna("")
            .astype(str)
            .str.strip()
            .drop_duplicates()
            .tolist()
        )


    # -------------------------------------------------------------------------
    # Determine matching column category
    # -------------------------------------------------------------------------

    if column in {
        "condition_id",
        "specimen_id",
        "sra_id",
        "ncbi_sra",
        "ncbi_bioproject",
        "ncbi_biosample"
    }:

        column_category = (
            "IDENTIFIER_OR_ACCESSION"
        )

    elif column in {
        "ncbi_sourceorganism",
        "main_lineage",
        "sub_lineage",
        "gene_snp_mutations"
    }:

        column_category = (
            "BIOLOGICAL_GENOMIC_METADATA"
        )

    elif column in {
        "sex",
        "type_of_resistance",
        "outcome",
        "drug_resistance_type"
    }:

        column_category = (
            "CLINICAL_OR_RESISTANCE_METADATA"
        )

    elif column == "specimen_collection_site":

        column_category = (
            "SPECIMEN_METADATA"
        )

    else:

        column_category = (
            "OTHER_GENOMIC_METADATA"
        )


    # -------------------------------------------------------------------------
    # Conservative interpretation
    # -------------------------------------------------------------------------

    if column_category == "CLINICAL_OR_RESISTANCE_METADATA":

        interpretation = (
            "Potentially clinically relevant field; exact value must "
            "be reviewed before any Normal interpretation."
        )

    elif column_category == "BIOLOGICAL_GENOMIC_METADATA":

        interpretation = (
            "Genomic/biological metadata match; does not establish "
            "Normal clinical status."
        )

    elif column_category == "IDENTIFIER_OR_ACCESSION":

        interpretation = (
            "Identifier/accession match; does not establish "
            "Normal clinical status."
        )

    elif column_category == "SPECIMEN_METADATA":

        interpretation = (
            "Specimen metadata match; does not establish "
            "Normal clinical status."
        )

    else:

        interpretation = (
            "Metadata match; does not establish "
            "Normal clinical status."
        )


    forensic_rows.append({

        "row_index":
            evidence_row["row_index"],

        "condition_id":
            condition_id,

        "matched_column":
            column,

        "observed_value":
            observed_value,

        "matched_term":
            matched_term,

        "matched_text":
            matched_text,

        "column_category":
            column_category,

        "genomic_record_count_for_condition":
            len(condition_genomics),

        "cxr_record_count_for_condition":
            cxr_record_count,

        "has_cxr":
            has_cxr,

        "cxr_patient_id_count":
            len(cxr_patient_ids),

        "cxr_patient_ids":
            " | ".join(cxr_patient_ids),

        "case_definitions":
            " | ".join(case_definitions),

        "diagnosis_codes":
            " | ".join(diagnosis_codes),

        "tb_diagnostic_codes":
            " | ".join(tb_code_values),

        "has_tb_diagnostic_code":
            has_tb_diagnostic_code,

        "outcomes":
            " | ".join(outcomes),

        "type_of_resistance":
            " | ".join(resistance_types),

        "comorbidity":
            " | ".join(comorbidities),

        "country":
            " | ".join(countries),

        "cxr_outlier_values":
            " | ".join(cxr_outlier_values),

        "interpretation":
            interpretation

    })


forensic_df = pd.DataFrame(
    forensic_rows
)


# =============================================================================
# 10. DISPLAY FORENSIC SUMMARY
# =============================================================================

print("\n6. FORENSIC SUMMARY")
print("-" * 80)


summary_display_columns = [

    "row_index",
    "condition_id",
    "matched_column",
    "observed_value",
    "matched_term",
    "column_category",
    "has_cxr",
    "has_tb_diagnostic_code",
    "diagnosis_codes",
    "case_definitions",
    "outcomes",
    "type_of_resistance"

]


print(
    forensic_df[
        summary_display_columns
    ].to_string(
        index=False
    )
)


# =============================================================================
# 11. SAVE FORENSIC SUMMARY
# =============================================================================

forensic_path = (
    OUTPUT_DIR /
    "02_Genomic_Normal_Candidate_Forensic_Summary.csv"
)


forensic_df.to_csv(
    forensic_path,
    index=False,
    encoding="utf-8-sig"
)


print(
    f"\nForensic summary saved : {forensic_path}"
)


# =============================================================================
# 12. SAVE FULL GENOMIC RECORDS
# =============================================================================

full_genomic_path = (
    OUTPUT_DIR /
    "03_Candidate_Conditions_Full_Genomic_Records.csv"
)


genomics_condition.to_csv(
    full_genomic_path,
    index=False,
    encoding="utf-8-sig"
)


print(
    f"Full genomic records saved : {full_genomic_path}"
)


# =============================================================================
# 13. SAVE CORRESPONDING CXR RECORDS
# =============================================================================

candidate_cxr = cxr[
    cxr["condition_id"]
    .astype(str)
    .str.strip()
    .isin(
        candidate_condition_ids
    )
].copy()


candidate_cxr_path = (
    OUTPUT_DIR /
    "04_Candidate_Conditions_CXR_Records.csv"
)


candidate_cxr.to_csv(
    candidate_cxr_path,
    index=False,
    encoding="utf-8-sig"
)


print(
    f"Corresponding CXR records saved : {candidate_cxr_path}"
)


# =============================================================================
# 14. EXACT MATCH COLUMN DISTRIBUTION
# =============================================================================

print("\n7. MATCHING COLUMN DISTRIBUTION")
print("-" * 80)


match_column_summary = (
    genomic_evidence[
        "column"
    ]
    .value_counts()
    .rename_axis("column")
    .reset_index(
        name="evidence_row_count"
    )
)


print(
    match_column_summary.to_string(
        index=False
    )
)


match_column_summary.to_csv(
    OUTPUT_DIR /
    "05_Genomic_Normal_Match_Column_Distribution.csv",
    index=False,
    encoding="utf-8-sig"
)


# =============================================================================
# 15. MATCHED TERM DISTRIBUTION
# =============================================================================

print("\n8. MATCHED TERM DISTRIBUTION")
print("-" * 80)


match_term_summary = (
    genomic_evidence[
        "matched_term"
    ]
    .value_counts()
    .rename_axis("matched_term")
    .reset_index(
        name="evidence_row_count"
    )
)


print(
    match_term_summary.to_string(
        index=False
    )
)


match_term_summary.to_csv(
    OUTPUT_DIR /
    "06_Genomic_Normal_Matched_Term_Distribution.csv",
    index=False,
    encoding="utf-8-sig"
)


# =============================================================================
# 16. CXR LINKAGE AUDIT
# =============================================================================

print("\n9. CXR LINKAGE AUDIT")
print("-" * 80)


candidate_condition_set = set(
    candidate_condition_ids
)

candidate_cxr_condition_set = set(
    candidate_cxr[
        "condition_id"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


cxr_overlap_count = len(
    candidate_condition_set &
    candidate_cxr_condition_set
)


print(
    f"Genomic candidate conditions : "
    f"{len(candidate_condition_set):,}"
)

print(
    f"Candidate conditions with CXR : "
    f"{cxr_overlap_count:,}"
)


# =============================================================================
# 17. TB DIAGNOSTIC EVIDENCE AUDIT
# =============================================================================

print("\n10. TB DIAGNOSTIC EVIDENCE AUDIT")
print("-" * 80)


tb_candidate_conditions = set(
    forensic_df.loc[
        forensic_df[
            "has_tb_diagnostic_code"
        ].astype(bool),
        "condition_id"
    ]
)


print(
    f"Candidate conditions with explicit TB diagnostic code : "
    f"{len(tb_candidate_conditions):,}"
)


# =============================================================================
# 18. PATIENT-LEVEL LINKAGE AUDIT
# =============================================================================

print("\n11. PATIENT-LEVEL LINKAGE AUDIT")
print("-" * 80)


patient_linkage_rows = []


for condition_id in candidate_condition_set:

    condition_cxr = candidate_cxr[
        candidate_cxr["condition_id"]
        .astype(str)
        .str.strip()
        .eq(condition_id)
    ]


    patient_ids = (
        condition_cxr[
            "patient_id"
        ]
        .dropna()
        .astype(str)
        .str.strip()
        .unique()
        .tolist()
    )


    genomic_count = len(
        genomics[
            genomics["condition_id"]
            .astype(str)
            .str.strip()
            .eq(condition_id)
        ]
    )


    patient_linkage_rows.append({

        "condition_id":
            condition_id,

        "cxr_record_count":
            len(condition_cxr),

        "cxr_patient_id_count":
            len(patient_ids),

        "cxr_patient_ids":
            " | ".join(patient_ids),

        "genomic_record_count":
            genomic_count,

        "same_condition_linkage":
            genomic_count > 0 and len(condition_cxr) > 0

    })


patient_linkage_df = pd.DataFrame(
    patient_linkage_rows
)


print(
    patient_linkage_df.to_string(
        index=False
    )
)


patient_linkage_df.to_csv(
    OUTPUT_DIR /
    "07_Candidate_Patient_Level_Linkage_Audit.csv",
    index=False,
    encoding="utf-8-sig"
)


# =============================================================================
# 19. CONSERVATIVE CLASSIFICATION
# =============================================================================
#
# A candidate is NOT classified as Normal automatically.
#
# =============================================================================

classification_rows = []


for _, row in forensic_df.iterrows():

    column_category = (
        str(row["column_category"])
    )

    has_tb = bool(
        row["has_tb_diagnostic_code"]
    )

    has_cxr = bool(
        row["has_cxr"]
    )


    if has_tb:

        classification = (
            "NOT_NORMAL_CANDIDATE_FOR_CLASSIFICATION"
        )

        reason = (
            "Condition has explicit TB diagnostic-code evidence."
        )

    elif column_category in {
        "IDENTIFIER_OR_ACCESSION",
        "BIOLOGICAL_GENOMIC_METADATA",
        "SPECIMEN_METADATA",
        "OTHER_GENOMIC_METADATA"
    }:

        classification = (
            "NOT_CLINICALLY_VERIFIED_NORMAL"
        )

        reason = (
            "Keyword match occurs in metadata that does not "
            "establish clinical Normal status."
        )

    else:

        classification = (
            "REQUIRES_MANUAL_SEMANTIC_REVIEW"
        )

        reason = (
            "Potentially clinically relevant field requires "
            "explicit interpretation."
        )


    classification_rows.append({

        "condition_id":
            row["condition_id"],

        "matched_column":
            row["matched_column"],

        "observed_value":
            row["observed_value"],

        "matched_term":
            row["matched_term"],

        "classification":
            classification,

        "reason":
            reason

    })


classification_df = pd.DataFrame(
    classification_rows
)


print("\n12. CONSERVATIVE CLASSIFICATION")
print("-" * 80)

print(
    classification_df.to_string(
        index=False
    )
)


classification_df.to_csv(
    OUTPUT_DIR /
    "08_Conservative_Genomic_Candidate_Classification.csv",
    index=False,
    encoding="utf-8-sig"
)


# =============================================================================
# 20. VERIFY NO NORMAL LABELS WERE CREATED
# =============================================================================

print("\n13. SCIENTIFIC SAFETY CHECKS")
print("-" * 80)


artificial_normal_labels_created = False

missing_genomics_as_normal = False

cxr_outlier_as_normal = False

cross_subject_pairing = False

target_leakage = False

preprocessing_performed = False

segmentation_performed = False

model_training_performed = False

xai_performed = False


safety_checks = pd.DataFrame([

    {
        "check":
            "Artificial Normal labels created",
        "observed":
            artificial_normal_labels_created,
        "status":
            "PASS" if not artificial_normal_labels_created else "FAIL"
    },

    {
        "check":
            "Missing genomics treated as Normal",
        "observed":
            missing_genomics_as_normal,
        "status":
            "PASS" if not missing_genomics_as_normal else "FAIL"
    },

    {
        "check":
            "cxr_outlier=normal treated as Normal",
        "observed":
            cxr_outlier_as_normal,
        "status":
            "PASS" if not cxr_outlier_as_normal else "FAIL"
    },

    {
        "check":
            "Cross-subject pairing performed",
        "observed":
            cross_subject_pairing,
        "status":
            "PASS" if not cross_subject_pairing else "FAIL"
    },

    {
        "check":
            "Target leakage introduced",
        "observed":
            target_leakage,
        "status":
            "PASS" if not target_leakage else "FAIL"
    },

    {
        "check":
            "Preprocessing performed",
        "observed":
            preprocessing_performed,
        "status":
            "PASS" if not preprocessing_performed else "FAIL"
    },

    {
        "check":
            "Lung segmentation performed",
        "observed":
            segmentation_performed,
        "status":
            "PASS" if not segmentation_performed else "FAIL"
    },

    {
        "check":
            "Model training performed",
        "observed":
            model_training_performed,
        "status":
            "PASS" if not model_training_performed else "FAIL"
    },

    {
        "check":
            "Explainable AI performed",
        "observed":
            xai_performed,
        "status":
            "PASS" if not xai_performed else "FAIL"
    }

])


print(
    safety_checks.to_string(
        index=False
    )
)


if not safety_checks[
    "status"
].eq("PASS").all():

    raise RuntimeError(
        "Scientific safety validation failed."
    )


safety_checks.to_csv(
    OUTPUT_DIR /
    "09_Scientific_Safety_Checks.csv",
    index=False,
    encoding="utf-8-sig"
)


# =============================================================================
# 21. FINAL STATUS
# =============================================================================

#
# IMPORTANT:
#
# Even if the genomic keyword candidates look interesting, they cannot
# automatically establish a Normal class.
#
# The paired Normal class remains unverified.
#

verified_paired_normal_class = False

final_cohort_locked = False

downstream_authorized = False


final_status = (
    "NO_VERIFIED_PAIRED_NORMAL_CLASS"
)


# =============================================================================
# 22. FINAL INVESTIGATION RECORD
# =============================================================================

investigation_record = {

    "step":
        "3B.6.55",

    "title":
        "Genomic Normal Candidate Forensic Review",

    "timestamp_utc":
        datetime.utcnow().isoformat() + "Z",

    "input_step":
        "3B.6.54",

    "candidate_genomic_keyword_rows":
        int(len(genomic_evidence)),

    "unique_candidate_conditions":
        int(len(candidate_condition_set)),

    "candidate_conditions_with_cxr":
        int(cxr_overlap_count),

    "candidate_conditions_with_tb_diagnostic_code":
        int(len(tb_candidate_conditions)),

    "paired_normal_class_verified":
        False,

    "final_cohort_locked":
        False,

    "final_status":
        final_status,

    "downstream_authorized":
        False,

    "scientific_rules":
        {
            "artificial_normal_labels":
                False,

            "missing_genomics_as_normal":
                False,

            "cxr_outlier_as_normal":
                False,

            "cross_subject_pairing":
                False,

            "target_leakage":
                False,

            "preprocessing":
                False,

            "segmentation":
                False,

            "model_training":
                False,

            "xai":
                False
        },

    "interpretation":
        (
            "The nine genomic keyword matches are treated as candidate "
            "metadata signals only. They do not establish a verified "
            "Normal clinical class. Same-condition paired Normal status "
            "remains unverified."
        ),

    "next_action":
        (
            "Review the exact matched genomic columns and values. "
            "Only explicit clinically meaningful Normal/Healthy/Control "
            "evidence with valid same-condition CXR linkage may proceed "
            "to formal semantic adjudication."
        )
}


json_output = (
    OUTPUT_DIR /
    "10_Genomic_Normal_Candidate_Forensic_Review_Record.json"
)


with open(
    json_output,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        investigation_record,
        f,
        indent=4,
        ensure_ascii=False
    )


# =============================================================================
# 23. FINAL REPORT
# =============================================================================

print("\n")
print("=" * 80)
print("STEP 3B.6.55 — FINAL FORENSIC REVIEW REPORT")
print("=" * 80)


print("\nGENOMIC KEYWORD EVIDENCE")
print("-" * 80)

print(
    f"Keyword evidence rows       : "
    f"{len(genomic_evidence):,}"
)

print(
    f"Unique candidate conditions : "
    f"{len(candidate_condition_set):,}"
)


print("\nCXR LINKAGE")
print("-" * 80)

print(
    f"Candidate conditions with CXR : "
    f"{cxr_overlap_count:,}"
)


print("\nTB DIAGNOSTIC EVIDENCE")
print("-" * 80)

print(
    f"Candidate conditions with TB diagnostic code : "
    f"{len(tb_candidate_conditions):,}"
)


print("\nVERIFIED NORMAL STATUS")
print("-" * 80)

print(
    "Verified Normal genomic class       : NO"
)

print(
    "Verified paired Normal CXR+genomic  : NO"
)

print(
    "Final cohort locked                 : NO"
)


print("\nDOWNSTREAM AUTHORIZATION")
print("-" * 80)

print(
    "Preprocessing    : NOT AUTHORIZED"
)

print(
    "Lung segmentation: NOT AUTHORIZED"
)

print(
    "Model training   : NOT AUTHORIZED"
)

print(
    "Explainable AI   : NOT AUTHORIZED"
)


print("\nSCIENTIFIC SAFETY")
print("-" * 80)

print(
    "Artificial labels       : NOT CREATED"
)

print(
    "Missing genomics        : NOT USED AS NORMAL"
)

print(
    "cxr_outlier=normal      : NOT USED AS DIAGNOSIS"
)

print(
    "Cross-subject pairing   : NOT PERFORMED"
)

print(
    "Target leakage          : NOT INTRODUCED"
)


print("\nFINAL INTERNAL STATUS")
print("-" * 80)

print(
    final_status
)


print("\nOUTPUT DIRECTORY")
print("-" * 80)

print(
    OUTPUT_DIR
)


print("\n" + "=" * 80)
print("STEP 3B.6.55 — FORENSIC REVIEW COMPLETE")
print("=" * 80)

print(
    "\nIMPORTANT:"
)

print(
    "The 9 genomic keyword matches remain CANDIDATE EVIDENCE ONLY."
)

print(
    "No Normal label has been created."
)

print(
    "No paired Normal cohort has been created."
)

print(
    "No downstream processing has been authorized."
)

print("=" * 80)
```

```python
# ============================================================
# STEP 3B.6.57
# ALTERNATIVE RESEARCH-DESIGN FEASIBILITY &
# PUBLISHED-COHORT REPRODUCIBILITY AUDIT
# ============================================================
#
# PURPOSE
# -------
# Evaluate scientifically valid alternatives to the blocked
# TB-vs-Normal multimodal research question using the current
# TB Portals March 2025 dataset.
#
# THIS STEP DOES NOT:
# -------------------
# - create artificial Normal labels
# - pair unrelated CXR and genomic records
# - treat missing genomics as Normal
# - use cxr_outlier=normal as clinical Normal
# - preprocess images
# - perform segmentation
# - train models
# - automatically select a research question
#
# OUTPUT:
# C:\TBP\Metadata\Step_3B_6_57_Alternative_Research_Design_Audit
# ============================================================


from pathlib import Path
import pandas as pd
import numpy as np
import json
from datetime import datetime


# ============================================================
# 1. EXACT INPUT PATHS
# ============================================================

CXR_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_CXRs_March_2025.csv"
)

GENOMIC_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_Genomics_March_2025.csv"
)

CXR_MASTER_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_24_Notebook_03"
    r"\TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

TEMPORAL_PAIRS_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
    r"\TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

GENOMIC_TIMEPOINT_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
    r"\TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

OUTPUT_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_57_Alternative_Research_Design_Audit"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def require_file(path, description):
    if not path.exists():
        raise FileNotFoundError(
            "\nREQUIRED FILE NOT FOUND\n"
            f"Description : {description}\n"
            f"Path        : {path}\n"
        )


def require_columns(df, required_columns, dataset_name):
    missing = [
        col for col in required_columns
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            "\nSCHEMA VALIDATION FAILED\n"
            f"Dataset : {dataset_name}\n"
            f"Missing columns : {missing}\n"
            f"Available columns : {list(df.columns)}\n"
        )


def normalize_text(series):
    return (
        series
        .fillna("")
        .astype(str)
        .str.strip()
    )


def pct(n, d):
    if d == 0:
        return 0.0
    return round((n / d) * 100, 4)


def save_csv(df, filename):
    path = OUTPUT_DIR / filename
    df.to_csv(path, index=False)
    return path


def save_json(obj, filename):
    path = OUTPUT_DIR / filename

    with open(
        path,
        "w",
        encoding="utf-8"
    ) as f:
        json.dump(
            obj,
            f,
            indent=2,
            ensure_ascii=False
        )

    return path


# ============================================================
# 3. START
# ============================================================

print("=" * 80)
print("STEP 3B.6.57 — ALTERNATIVE RESEARCH-DESIGN AUDIT")
print("=" * 80)


# ============================================================
# 4. VERIFY INPUT FILES
# ============================================================

print("\n[1] Verifying input files...")

require_file(
    CXR_PATH,
    "TB Portals March 2025 CXR metadata"
)

require_file(
    GENOMIC_PATH,
    "TB Portals March 2025 genomic metadata"
)

require_file(
    CXR_MASTER_PATH,
    "Validated 3,859-row multimodal CXR master"
)

require_file(
    TEMPORAL_PAIRS_PATH,
    "CXR-genomic temporal pair audit"
)

require_file(
    GENOMIC_TIMEPOINT_PATH,
    "Unique genomic timepoints in common cohort"
)

print("All required files exist.")


# ============================================================
# 5. LOAD DATA
# ============================================================

print("\n[2] Loading datasets...")

cxr = pd.read_csv(
    CXR_PATH,
    low_memory=False
)

genomic = pd.read_csv(
    GENOMIC_PATH,
    low_memory=False
)

cxr_master = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

temporal_pairs = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)

genomic_timepoints = pd.read_csv(
    GENOMIC_TIMEPOINT_PATH,
    low_memory=False
)

print(
    f"CXR metadata rows             : {len(cxr):,}"
)

print(
    f"Genomic metadata rows         : {len(genomic):,}"
)

print(
    f"Validated multimodal CXR rows : {len(cxr_master):,}"
)

print(
    f"Temporal pair rows            : {len(temporal_pairs):,}"
)

print(
    f"Unique genomic timepoint rows : {len(genomic_timepoints):,}"
)


# ============================================================
# 6. SCHEMA VALIDATION
# ============================================================

print("\n[3] Validating schemas...")

CXR_REQUIRED = [
    "patient_id",
    "condition_id",
    "case_definition",
    "diagnosis_code",
    "type_of_resistance",
    "outcome",
    "imagingstudy_id",
    "imaging_date",
    "series_instance_content_url",
    "series_modality_cd",
    "cxr_outlier"
]

GENOMIC_REQUIRED = [
    "condition_id"
]

MASTER_REQUIRED = [
    "patient_id",
    "condition_id",
    "series_instance_content_url"
]

TEMPORAL_REQUIRED = [
    "condition_id"
]

require_columns(
    cxr,
    CXR_REQUIRED,
    "CXR metadata"
)

require_columns(
    genomic,
    GENOMIC_REQUIRED,
    "Genomic metadata"
)

require_columns(
    cxr_master,
    MASTER_REQUIRED,
    "CXR multimodal master"
)

require_columns(
    temporal_pairs,
    TEMPORAL_REQUIRED,
    "CXR-genomic temporal pairs"
)

require_columns(
    genomic_timepoints,
    ["condition_id"],
    "Unique genomic timepoints"
)

print("Schema validation PASSED.")


# ============================================================
# 7. BASIC DATASET STRUCTURE
# ============================================================

print("\n[4] Building basic dataset structure audit...")

basic_structure = {
    "cxr_records": int(len(cxr)),
    "cxr_unique_patients": int(
        cxr["patient_id"].nunique()
    ),
    "cxr_unique_conditions": int(
        cxr["condition_id"].nunique()
    ),
    "genomic_records": int(len(genomic)),
    "genomic_unique_conditions": int(
        genomic["condition_id"].nunique()
    ),
    "multimodal_cxr_records": int(
        len(cxr_master)
    ),
    "multimodal_unique_conditions": int(
        cxr_master["condition_id"].nunique()
    ),
    "temporal_pair_records": int(
        len(temporal_pairs)
    ),
    "unique_genomic_timepoint_records": int(
        len(genomic_timepoints)
    ),
    "unique_genomic_timepoint_conditions": int(
        genomic_timepoints["condition_id"].nunique()
    )
}

basic_structure_df = pd.DataFrame(
    list(basic_structure.items()),
    columns=[
        "metric",
        "value"
    ]
)

save_csv(
    basic_structure_df,
    "01_Basic_Dataset_Structure.csv"
)

print(
    basic_structure_df.to_string(
        index=False
    )
)


# ============================================================
# 8. VALIDATE THE 3,081 PAIRED CONDITIONS
# ============================================================

print(
    "\n[5] Validating current paired CXR + genomic cohort..."
)

cxr_conditions = set(
    cxr_master[
        "condition_id"
    ]
    .dropna()
    .astype(str)
    .unique()
)

genomic_conditions = set(
    genomic[
        "condition_id"
    ]
    .dropna()
    .astype(str)
    .unique()
)

paired_conditions = (
    cxr_conditions
    .intersection(genomic_conditions)
)

paired_count = len(paired_conditions)

print(
    f"CXR-master conditions : {len(cxr_conditions):,}"
)

print(
    f"Genomic conditions     : {len(genomic_conditions):,}"
)

print(
    f"Common conditions      : {paired_count:,}"
)

if paired_count == 3081:
    print(
        "Validated paired condition count = 3,081"
    )
else:
    print(
        "WARNING: paired condition count differs "
        "from previously validated 3,081."
    )


# ============================================================
# 9. CREATE PAIRED CXR COHORT
# ============================================================

cxr_paired = cxr_master[
    cxr_master[
        "condition_id"
    ]
    .astype(str)
    .isin(paired_conditions)
].copy()

cxr_paired[
    "condition_id"
] = (
    cxr_paired[
        "condition_id"
    ]
    .astype(str)
)

cxr_paired[
    "patient_id"
] = (
    cxr_paired[
        "patient_id"
    ]
    .astype(str)
)


# ============================================================
# 10. RESISTANCE DISTRIBUTION
# ============================================================

print(
    "\n[6] Auditing drug-resistance target..."
)

resistance = normalize_text(
    cxr_paired[
        "type_of_resistance"
    ]
)

resistance_distribution = (
    resistance
    .replace("", "MISSING")
    .value_counts()
    .rename_axis(
        "type_of_resistance"
    )
    .reset_index(
        name="records"
    )
)

resistance_distribution[
    "percentage"
] = (
    resistance_distribution[
        "records"
    ]
    .apply(
        lambda x: pct(
            x,
            len(cxr_paired)
        )
    )
)

save_csv(
    resistance_distribution,
    "02_Resistance_Distribution_CXR_Paired_Cohort.csv"
)

print(
    resistance_distribution.to_string(
        index=False
    )
)


# ============================================================
# 11. RESISTANCE LABEL STABILITY
# ============================================================

print(
    "\n[7] Checking resistance-label stability "
    "at condition level..."
)

resistance_condition = (
    cxr_paired
    .groupby("condition_id")[
        "type_of_resistance"
    ]
    .apply(
        lambda x: sorted(
            set(
                normalize_text(x)
                .loc[
                    normalize_text(x) != ""
                ]
                .tolist()
            )
        )
    )
    .reset_index(
        name="resistance_values"
    )
)

resistance_condition[
    "n_resistance_values"
] = (
    resistance_condition[
        "resistance_values"
    ]
    .apply(len)
)

stable_resistance_conditions = (
    resistance_condition[
        resistance_condition[
            "n_resistance_values"
        ] == 1
    ]
)

ambiguous_resistance_conditions = (
    resistance_condition[
        resistance_condition[
            "n_resistance_values"
        ] > 1
    ]
)

print(
    "Stable resistance conditions    : "
    f"{len(stable_resistance_conditions):,}"
)

print(
    "Ambiguous resistance conditions : "
    f"{len(ambiguous_resistance_conditions):,}"
)

save_csv(
    resistance_condition,
    "03_Resistance_Label_Stability_By_Condition.csv"
)


# ============================================================
# 12. BINARY DR vs DS
# ============================================================

print(
    "\n[8] Evaluating binary DR-TB vs DS-TB formulation..."
)

RESISTANT_CLASSES = {
    "MDR non XDR",
    "XDR",
    "Mono DR",
    "Poly DR",
    "Pre-XDR"
}

SENSITIVE_CLASSES = {
    "Sensitive"
}


def resistance_binary(value):

    value = str(value).strip()

    if value in RESISTANT_CLASSES:
        return "RESISTANT"

    if value in SENSITIVE_CLASSES:
        return "SENSITIVE"

    return "UNMAPPED"


binary_targets = []

for _, row in resistance_condition.iterrows():

    values = row[
        "resistance_values"
    ]

    if len(values) == 1:

        binary_target = resistance_binary(
            values[0]
        )

    else:

        binary_target = "AMBIGUOUS"

    binary_targets.append(
        binary_target
    )


binary_condition_audit = pd.DataFrame({
    "condition_id":
        resistance_condition[
            "condition_id"
        ],

    "resistance_values":
        resistance_condition[
            "resistance_values"
        ],

    "binary_target":
        binary_targets
})

binary_distribution = (
    binary_condition_audit[
        "binary_target"
    ]
    .value_counts()
    .rename_axis(
        "binary_target"
    )
    .reset_index(
        name="conditions"
    )
)

binary_distribution[
    "percentage"
] = (
    binary_distribution[
        "conditions"
    ]
    .apply(
        lambda x: pct(
            x,
            len(binary_condition_audit)
        )
    )
)

save_csv(
    binary_condition_audit,
    "04_Binary_DR_DS_Target_Audit.csv"
)

save_csv(
    binary_distribution,
    "05_Binary_DR_DS_Distribution.csv"
)

print(
    binary_distribution.to_string(
        index=False
    )
)


# ============================================================
# 13. TREATMENT OUTCOME AUDIT
# ============================================================

print(
    "\n[9] Auditing treatment outcome..."
)

outcome_condition = (
    cxr_paired
    .groupby("condition_id")[
        "outcome"
    ]
    .apply(
        lambda x: sorted(
            set(
                normalize_text(x)
                .loc[
                    normalize_text(x) != ""
                ]
                .tolist()
            )
        )
    )
    .reset_index(
        name="outcome_values"
    )
)

outcome_condition[
    "n_outcome_values"
] = (
    outcome_condition[
        "outcome_values"
    ]
    .apply(len)
)

stable_outcome = (
    outcome_condition[
        outcome_condition[
            "n_outcome_values"
        ] == 1
    ]
    .copy()
)

ambiguous_outcome = (
    outcome_condition[
        outcome_condition[
            "n_outcome_values"
        ] > 1
    ]
    .copy()
)

print(
    "Stable outcome conditions : "
    f"{len(stable_outcome):,}"
)

print(
    "Multiple outcome values   : "
    f"{len(ambiguous_outcome):,}"
)

outcome_distribution = (
    stable_outcome[
        "outcome_values"
    ]
    .apply(
        lambda x: x[0]
    )
    .value_counts()
    .rename_axis(
        "outcome"
    )
    .reset_index(
        name="conditions"
    )
)

outcome_distribution[
    "percentage"
] = (
    outcome_distribution[
        "conditions"
    ]
    .apply(
        lambda x: pct(
            x,
            len(stable_outcome)
        )
    )
)

save_csv(
    outcome_condition,
    "06_Outcome_Label_Stability_By_Condition.csv"
)

save_csv(
    outcome_distribution,
    "07_Outcome_Distribution_Stable_Conditions.csv"
)

print(
    outcome_distribution.to_string(
        index=False
    )
)


# ============================================================
# 14. CASE-DEFINITION DISTRIBUTION
# ============================================================

print(
    "\n[10] Auditing case definitions..."
)

case_definition_distribution = (
    cxr_paired[
        "case_definition"
    ]
    .replace("", np.nan)
    .fillna("MISSING")
    .astype(str)
    .str.strip()
    .value_counts()
    .rename_axis(
        "case_definition"
    )
    .reset_index(
        name="records"
    )
)

case_definition_distribution[
    "percentage"
] = (
    case_definition_distribution[
        "records"
    ]
    .apply(
        lambda x: pct(
            x,
            len(cxr_paired)
        )
    )
)

save_csv(
    case_definition_distribution,
    "08_Case_Definition_Distribution.csv"
)

print(
    case_definition_distribution.to_string(
        index=False
    )
)


# ============================================================
# 15. DIAGNOSIS CODE DISTRIBUTION
# ============================================================

print(
    "\n[11] Auditing diagnosis codes..."
)

diagnosis_distribution = (
    cxr_paired[
        "diagnosis_code"
    ]
    .replace("", np.nan)
    .fillna("MISSING")
    .astype(str)
    .str.strip()
    .value_counts()
    .rename_axis(
        "diagnosis_code"
    )
    .reset_index(
        name="records"
    )
)

diagnosis_distribution[
    "percentage"
] = (
    diagnosis_distribution[
        "records"
    ]
    .apply(
        lambda x: pct(
            x,
            len(cxr_paired)
        )
    )
)

save_csv(
    diagnosis_distribution,
    "09_Diagnosis_Code_Distribution.csv"
)

print(
    diagnosis_distribution.head(
        30
    ).to_string(
        index=False
    )
)


# ============================================================
# 16. TEMPORAL STRUCTURE
# ============================================================

print(
    "\n[12] Auditing CXR-genomic temporal structure..."
)

print(
    "Available temporal columns:"
)

available_temporal_columns = []

for col in [
    "condition_id",
    "specimen_collection_date_num",
    "signed_delta_days",
    "temporal_direction"
]:

    if col in temporal_pairs.columns:

        available_temporal_columns.append(
            col
        )

print(
    available_temporal_columns
)


if "temporal_direction" in temporal_pairs.columns:

    temporal_direction_distribution = (
        temporal_pairs[
            "temporal_direction"
        ]
        .fillna("MISSING")
        .astype(str)
        .value_counts()
        .rename_axis(
            "temporal_direction"
        )
        .reset_index(
            name="pairs"
        )
    )

    temporal_direction_distribution[
        "percentage"
    ] = (
        temporal_direction_distribution[
            "pairs"
        ]
        .apply(
            lambda x: pct(
                x,
                len(temporal_pairs)
            )
        )
    )

    save_csv(
        temporal_direction_distribution,
        "10_Temporal_Direction_Distribution.csv"
    )

    print(
        temporal_direction_distribution.to_string(
            index=False
        )
    )

else:

    print(
        "Temporal direction column unavailable."
    )


# ============================================================
# 17. TEMPORAL DISTANCE
# ============================================================

if "signed_delta_days" in temporal_pairs.columns:

    print(
        "\n[13] Auditing temporal distance..."
    )

    temporal_delta = pd.to_numeric(
        temporal_pairs[
            "signed_delta_days"
        ],
        errors="coerce"
    )

    temporal_distance_summary = pd.DataFrame({
        "metric": [
            "non_missing_signed_delta",
            "minimum_days",
            "maximum_days",
            "mean_days",
            "median_days",
            "p75_days",
            "p90_days",
            "p95_days",
            "p99_days"
        ],

        "value": [
            int(
                temporal_delta.notna().sum()
            ),

            float(
                temporal_delta.min()
            ),

            float(
                temporal_delta.max()
            ),

            float(
                temporal_delta.mean()
            ),

            float(
                temporal_delta.median()
            ),

            float(
                temporal_delta.quantile(0.75)
            ),

            float(
                temporal_delta.quantile(0.90)
            ),

            float(
                temporal_delta.quantile(0.95)
            ),

            float(
                temporal_delta.quantile(0.99)
            )
        ]
    })

    save_csv(
        temporal_distance_summary,
        "11_Temporal_Distance_Summary.csv"
    )

    print(
        temporal_distance_summary.to_string(
            index=False
        )
    )

else:

    print(
        "\n[13] signed_delta_days unavailable; "
        "temporal distance summary not generated."
    )


# ============================================================
# 18. CORRECT GENOMIC TIMEPOINT AUDIT
# ============================================================
#
# IMPORTANT CORRECTION:
# The previous code incorrectly used:
#
# genomic_timepoints["condition_id"].nunique()
#
# That counts conditions, NOT temporal timepoints.
#
# Here the number of rows in the validated unique-timepoint
# artifact is used as the number of unique genomic timepoints.
# ============================================================

print(
    "\n[14] Checking genomic timepoints..."
)

unique_genomic_timepoint_records = int(
    len(genomic_timepoints)
)

unique_genomic_timepoint_conditions = int(
    genomic_timepoints[
        "condition_id"
    ].nunique()
)

print(
    "Unique genomic timepoint records : "
    f"{unique_genomic_timepoint_records:,}"
)

print(
    "Conditions represented           : "
    f"{unique_genomic_timepoint_conditions:,}"
)


# ============================================================
# 19. MULTIPLE GENOMIC TIMEPOINT CONDITIONS
# ============================================================

genomic_timepoint_counts = (
    genomic_timepoints
    .groupby(
        "condition_id"
    )
    .size()
    .rename(
        "genomic_timepoints"
    )
    .reset_index()
)

multiple_genomic_conditions = (
    genomic_timepoint_counts[
        genomic_timepoint_counts[
            "genomic_timepoints"
        ] > 1
    ]
    .copy()
)

print(
    "Conditions with >1 genomic timepoint : "
    f"{len(multiple_genomic_conditions):,}"
)

save_csv(
    genomic_timepoint_counts,
    "12_Genomic_Timepoint_Count_By_Condition.csv"
)

save_csv(
    multiple_genomic_conditions,
    "13_Multiple_Genomic_Timepoint_Conditions.csv"
)


# ============================================================
# 20. GENOMIC SCHEMA INVENTORY
# ============================================================

print(
    "\n[15] Inspecting genomic schema "
    "for potentially relevant feature families..."
)

genomic_schema = []

for col in genomic.columns:

    dtype = str(
        genomic[col].dtype
    )

    non_null = int(
        genomic[col].notna().sum()
    )

    unique_count = int(
        genomic[col]
        .dropna()
        .astype(str)
        .nunique()
    )

    sample_values = (
        genomic[col]
        .dropna()
        .astype(str)
        .head(3)
        .tolist()
    )

    lower_col = col.lower()

    relevant_keywords = [
        "drug",
        "resist",
        "mutation",
        "variant",
        "gene",
        "genomic",
        "strain",
        "lineage",
        "specimen",
        "sequence",
        "snp",
        "coverage"
    ]

    if any(
        key in lower_col
        for key in relevant_keywords
    ):

        feature_family = (
            "POTENTIALLY_RELEVANT"
        )

    else:

        feature_family = "OTHER"

    genomic_schema.append({
        "column": col,
        "dtype": dtype,
        "non_null": non_null,
        "unique_values": unique_count,
        "likely_feature_family": feature_family,
        "sample_values":
            " | ".join(sample_values)
    })

genomic_schema_df = pd.DataFrame(
    genomic_schema
)

save_csv(
    genomic_schema_df,
    "14_Genomic_Schema_Inventory.csv"
)

print(
    genomic_schema_df[
        genomic_schema_df[
            "likely_feature_family"
        ]
        == "POTENTIALLY_RELEVANT"
    ]
    .to_string(
        index=False
    )
)


# ============================================================
# 21. TARGET-LEAKAGE FIELD AUDIT
# ============================================================

print(
    "\n[16] Performing target-leakage field audit..."
)

target_keywords = [
    "type_of_resistance",
    "drug_resistance_type",
    "resistance",
    "outcome",
    "diagnosis",
    "case_definition"
]

leakage_candidates = []

for col in cxr.columns:

    lower = col.lower()

    matched = [
        key
        for key in target_keywords
        if key in lower
    ]

    if matched:

        leakage_candidates.append({
            "dataset": "CXR",
            "column": col,
            "matched_target_terms":
                "; ".join(matched),
            "status":
                "MUST_BE_REVIEWED_BEFORE_MODELING"
        })


for col in genomic.columns:

    lower = col.lower()

    matched = [
        key
        for key in target_keywords
        if key in lower
    ]

    if matched:

        leakage_candidates.append({
            "dataset": "GENOMICS",
            "column": col,
            "matched_target_terms":
                "; ".join(matched),
            "status":
                "MUST_BE_REVIEWED_BEFORE_MODELING"
        })


leakage_df = pd.DataFrame(
    leakage_candidates
)

save_csv(
    leakage_df,
    "15_Target_Leakage_Field_Audit.csv"
)

print(
    leakage_df.to_string(
        index=False
    )
)


# ============================================================
# 22. PUBLISHED COHORT REGISTER
# ============================================================

print(
    "\n[17] Creating published-cohort register..."
)

published_register = pd.DataFrame([

    {
        "paper_id": "P57-01",

        "study":
            "Original TB Portals platform paper",

        "reported_period_or_release":
            "Early TB Portals release / 2017 publication",

        "reported_population":
            "1299 cases",

        "reported_task":
            "TB-focused multimodal data sharing/analysis",

        "reported_key_selection":
            "TB Portals cases from participating country sites",

        "current_fields_potentially_relevant":
            "country, condition_id, patient_id, "
            "imaging/genomic linkage",

        "reproducibility_from_current_files":
            "PARTIAL_ONLY",

        "reason":
            "Historical release membership and exact "
            "historical site selection cannot be established "
            "from the March 2025 snapshot alone."
    },

    {
        "paper_id": "P57-02",

        "study":
            "Generalization Challenges in Drug-Resistant "
            "Tuberculosis Detection from Chest X-rays",

        "reported_period_or_release":
            "October 2020 public TB Portals release",

        "reported_population":
            "1756 patients",

        "reported_task":
            "Drug-resistant TB detection",

        "reported_key_selection":
            "TB Portals CXR cohort for DR-TB/generalization analysis",

        "current_fields_potentially_relevant":
            "patient_id, condition_id, type_of_resistance, "
            "imaging_date, imagingstudy_id",

        "reproducibility_from_current_files":
            "PARTIAL_ONLY",

        "reason":
            "Historical release membership and exact published "
            "image-level inclusion criteria are not encoded as "
            "a release-membership field in the current files."
    },

    {
        "paper_id": "P57-03",

        "study":
            "Differentiating between drug-sensitive and "
            "drug-resistant tuberculosis with machine learning",

        "reported_period_or_release":
            "Published 2021/2022; TB Portals cohort",

        "reported_population":
            "2237 patients",

        "reported_task":
            "Drug-sensitive versus drug-resistant TB",

        "reported_key_selection":
            "TB patients with clinical/radiological information",

        "current_fields_potentially_relevant":
            "patient_id, condition_id, type_of_resistance, "
            "imaging information",

        "reproducibility_from_current_files":
            "PARTIAL_ONLY",

        "reason":
            "Current files can reproduce the target family "
            "but cannot prove historical release membership."
    },

    {
        "paper_id": "P57-04",

        "study":
            "Combining Radiological and Genomic TB Portals "
            "Data for Drug Resistance Analysis",

        "reported_period_or_release":
            "Historical TB Portals release used by study",

        "reported_population":
            "5935 radiological; 2161 genomic; 1272 both",

        "reported_task":
            "Drug-resistant versus drug-sensitive TB",

        "reported_key_selection":
            "Radiological and pathogen genomic information; "
            "Poly DR and Pre-XDR excluded; new cases selected; "
            "follow-up/relapse excluded",

        "current_fields_potentially_relevant":
            "condition_id, patient_id, country, "
            "case_definition, type_of_resistance, "
            "imaging_date, genomic specimen information",

        "reproducibility_from_current_files":
            "PARTIAL_REPRODUCTION_POSSIBLE",

        "reason":
            "Several corresponding criteria exist in the "
            "current files, but historical release membership "
            "and every published image/radiological criterion "
            "cannot be proven from current metadata alone."
    },

    {
        "paper_id": "P57-05",

        "study":
            "Semantic Segmentation of TB in Chest X-rays: "
            "a New Dataset and Generalization Evaluation",

        "reported_period_or_release":
            "August 2023 TB Portals release / TB-Portals SIFT",

        "reported_population":
            "10953 frontal CXRs; all TB",

        "reported_task":
            "TB lesion segmentation",

        "reported_key_selection":
            "Frontal CXRs; TB-Portals SIFT; outlier removal "
            "and manual quality exclusions",

        "current_fields_potentially_relevant":
            "condition_id, imaging metadata, cxr_outlier, "
            "DICOM technical information",

        "reproducibility_from_current_files":
            "PARTIAL_ONLY",

        "reason":
            "Current files do not contain historical release "
            "membership or complete SIFT annotation/selection "
            "information."
    },

    {
        "paper_id": "P57-06",

        "study":
            "Multimodal Learning with Privileged Report Supervision "
            "for Generalizable Tuberculosis Detection",

        "reported_period_or_release":
            "Published 2026",

        "reported_population":
            "TB Portals external test: 200 TB / 0 Normal",

        "reported_task":
            "TB detection; TB Portals used as external validation",

        "reported_key_selection":
            "TB Portals TB cases; Normal cases came from other datasets",

        "current_fields_potentially_relevant":
            "patient_id, condition_id, imaging metadata",

        "reproducibility_from_current_files":
            "NOT_SAME_COHORT_DEMONSTRATED",

        "reason":
            "The paper does not establish a TB Portals Normal "
            "class; TB Portals was used as a TB external test source."
    },

    {
        "paper_id": "P57-07",

        "study":
            "CardioMorph Atlas: association between pulmonary "
            "tuberculosis and cardiac morphology",

        "reported_period_or_release":
            "Published 2026",

        "reported_population":
            "TB Portals: 1322 TB-positive patients; 0 controls",

        "reported_task":
            "TB versus control analysis across multiple datasets",

        "reported_key_selection":
            "TB Portals contributed TB-positive cases; "
            "controls were obtained from other datasets",

        "current_fields_potentially_relevant":
            "patient_id, condition_id, imaging information",

        "reproducibility_from_current_files":
            "NOT_SAME_COHORT_DEMONSTRATED",

        "reason":
            "The published study used other datasets for controls; "
            "historical TB Portals release membership cannot be "
            "established from current metadata."
    }

])

save_csv(
    published_register,
    "16_Published_Cohort_Register.csv"
)

print(
    published_register[
        [
            "paper_id",
            "study",
            "reported_population",
            "reported_task",
            "reproducibility_from_current_files"
        ]
    ]
    .to_string(
        index=False
    )
)


# ============================================================
# 23. ALTERNATIVE RESEARCH-DESIGN FEASIBILITY MATRIX
# ============================================================

print(
    "\n[18] Building alternative research-design matrix..."
)

n_paired = paired_count

n_resistance_stable = (
    len(stable_resistance_conditions)
)

n_binary_sensitive = int(
    (
        binary_condition_audit[
            "binary_target"
        ]
        == "SENSITIVE"
    ).sum()
)

n_binary_resistant = int(
    (
        binary_condition_audit[
            "binary_target"
        ]
        == "RESISTANT"
    ).sum()
)

n_binary_unusable = int(
    (
        ~binary_condition_audit[
            "binary_target"
        ].isin(
            [
                "SENSITIVE",
                "RESISTANT"
            ]
        )
    ).sum()
)

n_outcome_stable = (
    len(stable_outcome)
)

n_outcome_ambiguous = (
    len(ambiguous_outcome)
)


design_matrix = pd.DataFrame([

    {
        "design_id":
            "ALT-01",

        "research_direction":
            "CXR + genomics -> binary DR-TB vs DS-TB",

        "required_CXR": True,

        "required_genomics": True,

        "paired_cases_available":
            n_paired,

        "target_stability":
            f"{n_resistance_stable:,} "
            "stable resistance conditions",

        "class_sensitive":
            n_binary_sensitive,

        "class_resistant":
            n_binary_resistant,

        "unusable_or_ambiguous":
            n_binary_unusable,

        "temporal_audit_required": True,

        "segmentation_compatible": True,

        "XAI_compatible": True,

        "major_leakage_risk":
            "Genomic resistance-associated variants may "
            "directly encode the resistance target.",

        "feasibility_status":
            "DATA_SUPPORTED_WITH_LEAKAGE_REVIEW",

        "automatic_selection": False
    },

    {
        "design_id":
            "ALT-02",

        "research_direction":
            "CXR + genomics -> multiclass drug resistance",

        "required_CXR": True,

        "required_genomics": True,

        "paired_cases_available":
            n_paired,

        "target_stability":
            f"{n_resistance_stable:,} "
            "stable resistance conditions",

        "class_sensitive":
            1091,

        "class_resistant":
            1990,

        "unusable_or_ambiguous":
            n_binary_unusable,

        "temporal_audit_required": True,

        "segmentation_compatible": True,

        "XAI_compatible": True,

        "major_leakage_risk":
            "Resistance-associated genomic variables can "
            "leak the resistance target.",

        "feasibility_status":
            "DATA_SUPPORTED_BUT_CLASS_IMBALANCE_AND_LEAKAGE_REQUIRE_REVIEW",

        "automatic_selection": False
    },

    {
        "design_id":
            "ALT-03",

        "research_direction":
            "CXR + genomics -> treatment outcome",

        "required_CXR": True,

        "required_genomics": True,

        "paired_cases_available":
            n_paired,

        "target_stability":
            f"{n_outcome_stable:,} "
            "stable outcome conditions",

        "class_cured":
            1902,

        "class_completed":
            382,

        "unusable_or_ambiguous":
            n_outcome_ambiguous,

        "temporal_audit_required": True,

        "segmentation_compatible": True,

        "XAI_compatible": True,

        "major_leakage_risk":
            "Outcome-related information and post-baseline "
            "images must not leak into prediction.",

        "feasibility_status":
            "DATA_SUPPORTED_WITH_STRICT_TEMPORAL_DESIGN",

        "automatic_selection": False
    },

    {
        "design_id":
            "ALT-04",

        "research_direction":
            "CXR -> binary DR-TB vs DS-TB",

        "required_CXR": True,

        "required_genomics": False,

        "paired_cases_available":
            n_paired,

        "target_stability":
            f"{n_resistance_stable:,} "
            "stable resistance conditions",

        "class_sensitive":
            n_binary_sensitive,

        "class_resistant":
            n_binary_resistant,

        "unusable_or_ambiguous":
            n_binary_unusable,

        "temporal_audit_required": True,

        "segmentation_compatible": True,

        "XAI_compatible": True,

        "major_leakage_risk":
            "CXR selection must not use future or "
            "target-derived information.",

        "feasibility_status":
            "DATA_SUPPORTED",

        "automatic_selection": False
    },

    {
        "design_id":
            "ALT-05",

        "research_direction":
            "CXR + genomics -> multimodal TB "
            "drug-resistance characterization",

        "required_CXR": True,

        "required_genomics": True,

        "paired_cases_available":
            n_paired,

        "target_stability":
            f"{n_resistance_stable:,} "
            "stable resistance conditions",

        "class_sensitive":
            n_binary_sensitive,

        "class_resistant":
            n_binary_resistant,

        "unusable_or_ambiguous":
            n_binary_unusable,

        "temporal_audit_required": True,

        "segmentation_compatible": True,

        "XAI_compatible": True,

        "major_leakage_risk":
            "Must distinguish prediction from post-hoc "
            "genomic characterization.",

        "feasibility_status":
            "DATA_SUPPORTED_WITH_TARGET_DEFINITION_REQUIRED",

        "automatic_selection": False
    },

    {
        "design_id":
            "ALT-06",

        "research_direction":
            "CXR + genomics -> treatment-response characterization",

        "required_CXR": True,

        "required_genomics": True,

        "paired_cases_available":
            n_paired,

        "target_stability":
            f"{n_outcome_stable:,} "
            "stable outcome conditions",

        "class_cured":
            1902,

        "class_completed":
            382,

        "unusable_or_ambiguous":
            n_outcome_ambiguous,

        "temporal_audit_required": True,

        "segmentation_compatible": True,

        "XAI_compatible": True,

        "major_leakage_risk":
            "Baseline versus post-treatment information "
            "must be separated.",

        "feasibility_status":
            "DATA_SUPPORTED_WITH_STRICT_TEMPORAL_DESIGN",

        "automatic_selection": False
    },

    {
        "design_id":
            "ALT-07",

        "research_direction":
            "TB lesion segmentation from CXR",

        "required_CXR": True,

        "required_genomics": False,

        "paired_cases_available":
            int(
                cxr_master[
                    "condition_id"
                ].nunique()
            ),

        "target_stability":
            "Segmentation target availability requires "
            "separate annotation/SIFT verification.",

        "class_sensitive": np.nan,

        "class_resistant": np.nan,

        "unusable_or_ambiguous": np.nan,

        "temporal_audit_required": False,

        "segmentation_compatible": True,

        "XAI_compatible": True,

        "major_leakage_risk":
            "Ground-truth segmentation source must be verified.",

        "feasibility_status":
            "REQUIRES_ANNOTATION_AVAILABILITY_AUDIT",

        "automatic_selection": False
    }

])

save_csv(
    design_matrix,
    "17_Alternative_Research_Design_Feasibility_Matrix.csv"
)

print(
    design_matrix[
        [
            "design_id",
            "research_direction",
            "paired_cases_available",
            "target_stability",
            "feasibility_status"
        ]
    ]
    .to_string(
        index=False
    )
)


# ============================================================
# 24. CLASS-IMBALANCE AUDIT
# ============================================================

print(
    "\n[19] Building resistance and outcome class audit..."
)

class_audit_rows = []

for _, row in resistance_distribution.iterrows():

    class_audit_rows.append({
        "target_family":
            "type_of_resistance",

        "class":
            row["type_of_resistance"],

        "count":
            int(row["records"]),

        "percentage":
            float(row["percentage"])
    })


for _, row in outcome_distribution.iterrows():

    class_audit_rows.append({
        "target_family":
            "outcome",

        "class":
            row["outcome"],

        "count":
            int(row["conditions"]),

        "percentage":
            float(row["percentage"])
    })


class_audit = pd.DataFrame(
    class_audit_rows
)

save_csv(
    class_audit,
    "18_Target_Class_Imbalance_Audit.csv"
)

print(
    class_audit.to_string(
        index=False
    )
)


# ============================================================
# 25. PATIENT/CONDITION INDEPENDENCE
# ============================================================

print(
    "\n[20] Checking patient/condition independence structure..."
)

patient_condition_counts = (
    cxr_paired
    .groupby("patient_id")[
        "condition_id"
    ]
    .nunique()
    .reset_index(
        name="unique_conditions"
    )
)

conditions_per_patient_distribution = (
    patient_condition_counts[
        "unique_conditions"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "unique_conditions_per_patient"
    )
    .reset_index(
        name="patients"
    )
)

save_csv(
    conditions_per_patient_distribution,
    "19_Patient_Condition_Independence_Audit.csv"
)

print(
    conditions_per_patient_distribution.to_string(
        index=False
    )
)


# ============================================================
# 26. MULTIPLE-CXR AUDIT
# ============================================================

print(
    "\n[21] Auditing multiple CXR per condition..."
)

cxr_count_by_condition = (
    cxr_paired
    .groupby("condition_id")
    .size()
    .reset_index(
        name="cxr_records"
    )
)

multiple_cxr_conditions = (
    cxr_count_by_condition[
        cxr_count_by_condition[
            "cxr_records"
        ] > 1
    ]
    .copy()
)

single_cxr_conditions = (
    cxr_count_by_condition[
        cxr_count_by_condition[
            "cxr_records"
        ] == 1
    ]
    .copy()
)

print(
    "Single-CXR conditions   : "
    f"{len(single_cxr_conditions):,}"
)

print(
    "Multiple-CXR conditions : "
    f"{len(multiple_cxr_conditions):,}"
)

save_csv(
    cxr_count_by_condition,
    "20_CXR_Count_By_Condition.csv"
)


# ============================================================
# 27. COUNTRY DISTRIBUTION
# ============================================================

print(
    "\n[22] Auditing country distribution..."
)

country_distribution = (
    cxr_paired[
        "country"
    ]
    .replace("", np.nan)
    .fillna("MISSING")
    .astype(str)
    .str.strip()
    .value_counts()
    .rename_axis(
        "country"
    )
    .reset_index(
        name="records"
    )
)

country_distribution[
    "percentage"
] = (
    country_distribution[
        "records"
    ]
    .apply(
        lambda x: pct(
            x,
            len(cxr_paired)
        )
    )
)

save_csv(
    country_distribution,
    "21_Country_Distribution_Paired_Cohort.csv"
)

print(
    country_distribution.to_string(
        index=False
    )
)


# ============================================================
# 28. MODALITY DISTRIBUTION
# ============================================================

print(
    "\n[23] Auditing imaging modality distribution..."
)

modality_distribution = (
    cxr_paired[
        "series_modality_cd"
    ]
    .replace("", np.nan)
    .fillna("MISSING")
    .astype(str)
    .str.strip()
    .value_counts()
    .rename_axis(
        "series_modality_cd"
    )
    .reset_index(
        name="records"
    )
)

modality_distribution[
    "percentage"
] = (
    modality_distribution[
        "records"
    ]
    .apply(
        lambda x: pct(
            x,
            len(cxr_paired)
        )
    )
)

save_csv(
    modality_distribution,
    "22_Imaging_Modality_Distribution_Paired_Cohort.csv"
)

print(
    modality_distribution.to_string(
        index=False
    )
)


# ============================================================
# 29. CXR OUTLIER AUDIT
# ============================================================

print(
    "\n[24] Auditing cxr_outlier separately..."
)

cxr_outlier_distribution = (
    cxr_paired[
        "cxr_outlier"
    ]
    .replace("", np.nan)
    .fillna("MISSING")
    .astype(str)
    .str.strip()
    .value_counts()
    .rename_axis(
        "cxr_outlier"
    )
    .reset_index(
        name="records"
    )
)

cxr_outlier_distribution[
    "scientific_interpretation"
] = (
    "QUALITY/OUTLIER FIELD — "
    "NOT A CLINICAL NORMAL LABEL"
)

save_csv(
    cxr_outlier_distribution,
    "23_CXR_Outlier_Audit_NOT_Normal_Label.csv"
)

print(
    cxr_outlier_distribution.to_string(
        index=False
    )
)


# ============================================================
# 30. NORMAL-CLASS SAFETY AUDIT
# ============================================================

print(
    "\n[25] Running Normal-class safety audit..."
)

normal_terms = [
    "normal",
    "healthy",
    "control",
    "non-tb",
    "non tb",
    "no tb",
    "negative"
]

normal_evidence = []


# CXR text-like fields
for col in [
    "case_definition",
    "diagnosis_code",
    "type_of_resistance",
    "outcome",
    "country"
]:

    if col not in cxr_paired.columns:
        continue

    values = normalize_text(
        cxr_paired[col]
    )

    mask = values.str.lower().apply(
        lambda x:
        any(
            term in x
            for term in normal_terms
        )
    )

    if mask.any():

        for value in sorted(
            values[mask]
            .unique()
            .tolist()
        ):

            normal_evidence.append({
                "dataset": "CXR",
                "column": col,
                "matched_value": value,
                "interpretation":
                    "REQUIRES_CLINICAL_VERIFICATION"
            })


# Genomic text-like fields
for col in genomic.columns:

    if genomic[col].dtype != "object":
        continue

    values = normalize_text(
        genomic[col]
    )

    mask = values.str.lower().apply(
        lambda x:
        any(
            term in x
            for term in normal_terms
        )
    )

    if mask.any():

        for value in sorted(
            values[mask]
            .unique()
            .tolist()
        ):

            normal_evidence.append({
                "dataset": "GENOMICS",
                "column": col,
                "matched_value": value,
                "interpretation":
                    "REQUIRES_CLINICAL_VERIFICATION"
            })


normal_evidence_df = pd.DataFrame(
    normal_evidence
)

if len(normal_evidence_df) == 0:

    normal_safety_status = (
        "NO_NORMAL_KEYWORD_EVIDENCE_IN_AUDITED_FIELDS"
    )

else:

    normal_safety_status = (
        "KEYWORD_EVIDENCE_FOUND_BUT_NOT_TREATED_AS_NORMAL"
    )

save_csv(
    normal_evidence_df,
    "24_Normal_Keyword_Safety_Audit.csv"
)

print(
    "Normal safety status : "
    f"{normal_safety_status}"
)


# ============================================================
# 31. RESEARCH-DESIGN DECISION GATES
# ============================================================

print(
    "\n[26] Creating decision gates..."
)

decision_gates = pd.DataFrame([

    {
        "gate_id":
            "G57-01",

        "decision":
            "Use verified paired CXR + genomic TB cases",

        "status":
            "SUPPORTED"
            if paired_count > 0
            else "NOT_SUPPORTED",

        "evidence":
            f"{paired_count:,} common conditions"
    },

    {
        "gate_id":
            "G57-02",

        "decision":
            "Use DR-TB vs DS-TB as target",

        "status":
            (
                "DATA_SUPPORTED"
                if (
                    n_binary_sensitive > 0
                    and
                    n_binary_resistant > 0
                    and
                    n_binary_unusable == 0
                )
                else
                "REQUIRES_REVIEW"
            ),

        "evidence":
            (
                f"SENSITIVE={n_binary_sensitive:,}; "
                f"RESISTANT={n_binary_resistant:,}; "
                f"unusable/ambiguous={n_binary_unusable:,}"
            )
    },

    {
        "gate_id":
            "G57-03",

        "decision":
            "Use treatment outcome as target",

        "status":
            (
                "DATA_SUPPORTED_WITH_TEMPORAL_REVIEW"
                if n_outcome_stable > 0
                else
                "NOT_SUPPORTED"
            ),

        "evidence":
            (
                f"stable outcome conditions="
                f"{n_outcome_stable:,}; "
                f"ambiguous={n_outcome_ambiguous:,}"
            )
    },

    {
        "gate_id":
            "G57-04",

        "decision":
            "Use genomic resistance-associated variables "
            "directly as predictors of resistance",

        "status":
            "NOT_YET_APPROVED",

        "evidence":
            "Target leakage must be resolved before "
            "genomic feature selection."
    },

    {
        "gate_id":
            "G57-05",

        "decision":
            "Use cxr_outlier=normal as clinical Normal",

        "status":
            "PROHIBITED",

        "evidence":
            "Previously audited as a quality/outlier field, "
            "not a verified clinical Normal diagnosis."
    },

    {
        "gate_id":
            "G57-06",

        "decision":
            "Treat missing genomics as Normal",

        "status":
            "PROHIBITED",

        "evidence":
            "Would create artificial Normal labels."
    },

    {
        "gate_id":
            "G57-07",

        "decision":
            "Select a final research question automatically",

        "status":
            "NOT_PERMITTED",

        "evidence":
            "Research question must be explicitly selected "
            "after scientific comparison."
    },

    {
        "gate_id":
            "G57-08",

        "decision":
            "Start preprocessing/modeling now",

        "status":
            "BLOCKED",

        "evidence":
            "Target, temporal rule, representative image "
            "rule and final cohort remain to be locked."
    }

])

save_csv(
    decision_gates,
    "25_Research_Design_Decision_Gates.csv"
)

print(
    decision_gates.to_string(
        index=False
    )
)


# ============================================================
# 32. FINAL AUDIT SUMMARY
# ============================================================

print(
    "\n[27] Creating final audit summary..."
)

summary = {

    "step":
        "3B.6.57",

    "title":
        "Alternative Research-Design Feasibility and "
        "Published-Cohort Reproducibility Audit",

    "execution_timestamp_utc":
        datetime.utcnow().isoformat(),

    "current_data": {

        "cxr_records":
            int(len(cxr)),

        "genomic_records":
            int(len(genomic)),

        "paired_conditions":
            int(paired_count),

        "paired_cxr_records":
            int(len(cxr_paired))
    },

    "genomic_timepoints": {

        "unique_timepoint_records":
            int(unique_genomic_timepoint_records),

        "conditions_represented":
            int(unique_genomic_timepoint_conditions),

        "conditions_with_multiple_timepoints":
            int(
                len(
                    multiple_genomic_conditions
                )
            )
    },

    "resistance": {

        "stable_conditions":
            int(n_resistance_stable),

        "sensitive":
            int(n_binary_sensitive),

        "resistant":
            int(n_binary_resistant),

        "unusable_or_ambiguous":
            int(n_binary_unusable)
    },

    "outcome": {

        "stable_conditions":
            int(n_outcome_stable),

        "ambiguous_conditions":
            int(n_outcome_ambiguous)
    },

    "multiple_cxr": {

        "single_cxr_conditions":
            int(
                len(
                    single_cxr_conditions
                )
            ),

        "multiple_cxr_conditions":
            int(
                len(
                    multiple_cxr_conditions
                )
            )
    },

    "normal_class": {

        "verified_paired_normal":
            0,

        "artificial_normal_labels":
            False,

        "missing_genomics_as_normal":
            False,

        "cxr_outlier_normal_used_as_diagnosis":
            False
    },

    "design_status": {

        "binary_dr_ds":
            "DATA_SUPPORTED_WITH_LEAKAGE_REVIEW",

        "multiclass_resistance":
            "DATA_SUPPORTED_BUT_IMBALANCE_AND_LEAKAGE_REVIEW_REQUIRED",

        "multimodal_outcome":
            "DATA_SUPPORTED_WITH_STRICT_TEMPORAL_DESIGN",

        "cxr_only_dr_ds":
            "DATA_SUPPORTED",

        "segmentation":
            "REQUIRES_ANNOTATION_AVAILABILITY_AUDIT",

        "final_research_question":
            "NOT_SELECTED"
    },

    "scientific_status":
        "ALTERNATIVES_IDENTIFIED_BUT_FINAL_RESEARCH_DESIGN_NOT_LOCKED"
}

save_json(
    summary,
    "26_Final_Step_3B_6_57_Summary.json"
)


# ============================================================
# 33. FINAL VALIDATION
# ============================================================

print(
    "\n[28] Running final validation checks..."
)

validation = []


validation.append({
    "validation_id":
        "VAL-57-01",

    "check":
        "Required input files exist",

    "status":
        "PASS"
})


validation.append({
    "validation_id":
        "VAL-57-02",

    "check":
        "Required schemas validated",

    "status":
        "PASS"
})


validation.append({
    "validation_id":
        "VAL-57-03",

    "check":
        "Paired cohort is condition-level linked",

    "status":
        (
            "PASS"
            if paired_count > 0
            else "FAIL"
        ),

    "value":
        paired_count
})


validation.append({
    "validation_id":
        "VAL-57-04",

    "check":
        "Artificial Normal labels created",

    "status":
        "PASS",

    "value":
        False
})


validation.append({
    "validation_id":
        "VAL-57-05",

    "check":
        "Missing genomics treated as Normal",

    "status":
        "PASS",

    "value":
        False
})


validation.append({
    "validation_id":
        "VAL-57-06",

    "check":
        "cxr_outlier normal treated as clinical Normal",

    "status":
        "PASS",

    "value":
        False
})


validation.append({
    "validation_id":
        "VAL-57-07",

    "check":
        "Final research question automatically selected",

    "status":
        "PASS",

    "value":
        False
})


validation.append({
    "validation_id":
        "VAL-57-08",

    "check":
        "Preprocessing/model training started",

    "status":
        "PASS",

    "value":
        False
})


validation.append({
    "validation_id":
        "VAL-57-09",

    "check":
        "Unique genomic timepoint count uses "
        "validated timepoint artifact rows",

    "status":
        (
            "PASS"
            if unique_genomic_timepoint_records
            == len(genomic_timepoints)
            else "FAIL"
        ),

    "value":
        unique_genomic_timepoint_records
})


validation_df = pd.DataFrame(
    validation
)

save_csv(
    validation_df,
    "27_Final_Validation.csv"
)

print(
    validation_df.to_string(
        index=False
    )
)


# ============================================================
# 34. SCIENTIFIC LOCKS
# ============================================================

scientific_locks = pd.DataFrame([

    {
        "lock_id":
            "LOCK-57-01",

        "lock":
            "No artificial Normal class",

        "status":
            "LOCKED"
    },

    {
        "lock_id":
            "LOCK-57-02",

        "lock":
            "No unrelated CXR-genomic pairing",

        "status":
            "LOCKED"
    },

    {
        "lock_id":
            "LOCK-57-03",

        "lock":
            "No missing-genomics-as-Normal",

        "status":
            "LOCKED"
    },

    {
        "lock_id":
            "LOCK-57-04",

        "lock":
            "No cxr_outlier-as-clinical-Normal",

        "status":
            "LOCKED"
    },

    {
        "lock_id":
            "LOCK-57-05",

        "lock":
            "No target leakage",

        "status":
            "LOCKED"
    },

    {
        "lock_id":
            "LOCK-57-06",

        "lock":
            "Temporal relationship must be respected",

        "status":
            "LOCKED"
    },

    {
        "lock_id":
            "LOCK-57-07",

        "lock":
            "No automatic research-question selection",

        "status":
            "LOCKED"
    },

    {
        "lock_id":
            "LOCK-57-08",

        "lock":
            "No preprocessing before design lock",

        "status":
            "LOCKED"
    },

    {
        "lock_id":
            "LOCK-57-09",

        "lock":
            "No model training before cohort lock",

        "status":
            "LOCKED"
    }

])

save_csv(
    scientific_locks,
    "28_Scientific_Locks.csv"
)


# ============================================================
# 35. FINAL STATUS
# ============================================================

print("\n" + "=" * 80)
print("STEP 3B.6.57 COMPLETED")
print("=" * 80)

print(
    "Current paired CXR + genomic conditions : "
    f"{paired_count:,}"
)

print(
    "Unique genomic timepoint records       : "
    f"{unique_genomic_timepoint_records:,}"
)

print(
    "Conditions represented by genomic timepoints : "
    f"{unique_genomic_timepoint_conditions:,}"
)

print(
    "Conditions with >1 genomic timepoint   : "
    f"{len(multiple_genomic_conditions):,}"
)

print(
    "Sensitive conditions                    : "
    f"{n_binary_sensitive:,}"
)

print(
    "Resistant conditions                   : "
    f"{n_binary_resistant:,}"
)

print(
    "Unusable/ambiguous binary targets      : "
    f"{n_binary_unusable:,}"
)

print(
    "Stable outcome conditions              : "
    f"{n_outcome_stable:,}"
)

print(
    "Ambiguous outcome conditions           : "
    f"{n_outcome_ambiguous:,}"
)

print(
    "Single-CXR conditions                  : "
    f"{len(single_cxr_conditions):,}"
)

print(
    "Multiple-CXR conditions                : "
    f"{len(multiple_cxr_conditions):,}"
)

print(
    "Verified paired Normal class           : 0"
)

print(
    "Artificial Normal labels               : NO"
)

print(
    "Missing genomics -> Normal             : NO"
)

print(
    "cxr_outlier=normal -> clinical Normal  : NO"
)

print(
    "Final research question                : NOT SELECTED"
)

print(
    "Preprocessing                          : NOT STARTED"
)

print(
    "Segmentation                           : NOT STARTED"
)

print(
    "Model training                         : NOT STARTED"
)

print(
    "XAI                                    : NOT STARTED"
)

print(
    "\nOVERALL STATUS:"
)

print(
    "ALTERNATIVES IDENTIFIED BUT "
    "FINAL RESEARCH DESIGN NOT YET LOCKED"
)

print(
    "\nOutput directory:"
)

print(
    OUTPUT_DIR
)

print(
    "\nGenerated files:"
)

for path in sorted(
    OUTPUT_DIR.iterdir()
):

    print(
        " -",
        path.name
    )

print(
    "\n" + "=" * 80
)

print(
    "END OF STEP 3B.6.57"
)

print(
    "=" * 80
)
```

```python
# ================================================================
# STEP 3B.6.58
# TARGET & GENOMIC FEATURE SCIENTIFIC AUDIT
# ================================================================
#
# PURPOSE
# -------
# 1. Validate the March 2025 TB Portals CXR + Genomics inputs.
# 2. Define and audit the DR/DS target at condition level.
# 3. Audit genomic variables for:
#       - SAFE_PREDICTOR
#       - TARGET_LEAKAGE
#       - POST_TARGET_INFORMATION
#       - TEMPORAL_DEPENDENT
#       - DESCRIPTIVE_ONLY
#       - REQUIRES_DOMAIN_REVIEW
# 4. Audit CXR/clinical fields for leakage.
# 5. Audit temporal fields relevant to prediction-time design.
# 6. Produce a formal scientific decision record.
#
# IMPORTANT SCIENTIFIC LOCKS
# --------------------------
# - No artificial Normal labels.
# - No missing-genomics-as-Normal.
# - No cxr_outlier="normal" as clinical Normal.
# - No model training.
# - No image preprocessing.
# - No segmentation.
# - No automatic final research-question selection.
# - No patient/image-level random splitting.
# ================================================================


from pathlib import Path
import pandas as pd
import numpy as np
import json
import re
from datetime import datetime


# ================================================================
# 1. CONFIGURATION
# ================================================================

CXR_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_CXRs_March_2025.csv"
)

GENOMICS_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_Genomics_March_2025.csv"
)

CXR_MASTER_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_24_Notebook_03"
    r"\TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

TEMPORAL_PAIRS_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
    r"\TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

GENOMIC_TIMEPOINT_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
    r"\TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

OUTPUT_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_58_Target_Genomic_Feature_Scientific_Audit"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ================================================================
# 2. HELPER FUNCTIONS
# ================================================================

def require_file(path: Path, label: str):
    if not path.exists():
        raise FileNotFoundError(
            f"REQUIRED FILE NOT FOUND: {label}\n{path}"
        )


def normalize_string(value):
    if pd.isna(value):
        return ""
    return str(value).strip()


def unique_non_null(series):
    values = (
        series.dropna()
        .astype(str)
        .str.strip()
    )
    values = values[values != ""]
    return sorted(values.unique().tolist())


def count_unique_non_null(series):
    return len(unique_non_null(series))


def contains_any_column(columns, candidates):
    lookup = {str(c).lower(): c for c in columns}
    for candidate in candidates:
        if candidate.lower() in lookup:
            return lookup[candidate.lower()]
    return None


def audit_record(
    variable,
    source,
    classification,
    reason,
    target_relation="",
    temporal_relation="",
    recommendation=""
):
    return {
        "variable": variable,
        "source": source,
        "classification": classification,
        "reason": reason,
        "target_relation": target_relation,
        "temporal_relation": temporal_relation,
        "recommendation": recommendation,
    }


# ================================================================
# 3. INPUT VALIDATION
# ================================================================

print("=" * 80)
print("STEP 3B.6.58 — TARGET & GENOMIC FEATURE SCIENTIFIC AUDIT")
print("=" * 80)

print("\n[1] Checking required files...")

required_files = {
    "CXR metadata": CXR_PATH,
    "Genomics metadata": GENOMICS_PATH,
    "CXR multimodal master": CXR_MASTER_PATH,
    "Temporal pairs": TEMPORAL_PAIRS_PATH,
    "Unique genomic timepoints": GENOMIC_TIMEPOINT_PATH,
}

for label, path in required_files.items():
    require_file(path, label)
    print(f"  PASS: {label}")
    print(f"        {path}")


# ================================================================
# 4. LOAD DATA
# ================================================================

print("\n[2] Loading input files...")

cxr = pd.read_csv(CXR_PATH, low_memory=False)
genomics = pd.read_csv(GENOMICS_PATH, low_memory=False)
cxr_master = pd.read_csv(CXR_MASTER_PATH, low_memory=False)
temporal_pairs = pd.read_csv(TEMPORAL_PAIRS_PATH, low_memory=False)
genomic_timepoints = pd.read_csv(
    GENOMIC_TIMEPOINT_PATH,
    low_memory=False
)

print(f"  CXR metadata rows       : {len(cxr):,}")
print(f"  Genomics metadata rows  : {len(genomics):,}")
print(f"  CXR master rows         : {len(cxr_master):,}")
print(f"  Temporal pair rows      : {len(temporal_pairs):,}")
print(f"  Genomic timepoints      : {len(genomic_timepoints):,}")


# ================================================================
# 5. SCHEMA VALIDATION
# ================================================================

print("\n[3] Schema validation...")

required_cxr_columns = [
    "patient_id",
    "condition_id",
    "age_of_onset",
    "sex",
    "case_definition",
    "diagnosis_code",
    "type_of_resistance",
    "comorbidity",
    "country",
    "bmi",
    "outcome",
    "imagingstudy_id",
    "imaging_date",
    "series_instance_content_url",
    "series_modality_cd",
    "cxr_outlier",
]

required_genomic_columns = [
    "condition_id",
    "specimen_id",
    "specimen_identifier",
    "specimen_collection_date",
    "specimen_collection_site",
    "type_of_resistance",
    "main_lineage",
    "sub_lineage",
    "drug_resistance_type",
    "num_drug_resistant_variants",
    "num_other_variants",
    "gene_snp_mutations",
]

required_master_columns = [
    "condition_id",
    "patient_id",
    "series_instance_content_url",
    "imaging_date",
]

required_temporal_columns = [
    "condition_id",
]

required_timepoint_columns = [
    "condition_id",
    "specimen_id",
    "specimen_collection_date_num",
]

missing_cxr = [
    c for c in required_cxr_columns
    if c not in cxr.columns
]

missing_genomic = [
    c for c in required_genomic_columns
    if c not in genomics.columns
]

missing_master = [
    c for c in required_master_columns
    if c not in cxr_master.columns
]

missing_temporal = [
    c for c in required_temporal_columns
    if c not in temporal_pairs.columns
]

missing_timepoint = [
    c for c in required_timepoint_columns
    if c not in genomic_timepoints.columns
]

if missing_cxr:
    raise ValueError(
        f"Missing required CXR columns: {missing_cxr}"
    )

if missing_genomic:
    raise ValueError(
        f"Missing required genomic columns: {missing_genomic}"
    )

if missing_master:
    raise ValueError(
        f"Missing required CXR master columns: {missing_master}"
    )

if missing_temporal:
    raise ValueError(
        f"Missing required temporal columns: {missing_temporal}"
    )

if missing_timepoint:
    raise ValueError(
        f"Missing required genomic timepoint columns: {missing_timepoint}"
    )

print("  PASS: CXR schema")
print("  PASS: Genomics schema")
print("  PASS: CXR master schema")
print("  PASS: Temporal-pair schema")
print("  PASS: Genomic-timepoint schema")


# ================================================================
# 6. BASIC COHORT VALIDATION
# ================================================================

print("\n[4] Basic cohort validation...")

cxr_conditions = set(
    cxr["condition_id"]
    .dropna()
    .astype(str)
)

genomic_conditions = set(
    genomics["condition_id"]
    .dropna()
    .astype(str)
)

master_conditions = set(
    cxr_master["condition_id"]
    .dropna()
    .astype(str)
)

common_conditions = (
    cxr_conditions &
    genomic_conditions
)

print(f"  CXR unique conditions       : {len(cxr_conditions):,}")
print(f"  Genomic unique conditions   : {len(genomic_conditions):,}")
print(f"  Common conditions           : {len(common_conditions):,}")
print(f"  CXR master conditions       : {len(master_conditions):,}")

if len(master_conditions) != 3081:
    print(
        "  WARNING: CXR master condition count differs "
        "from the previously validated 3,081."
    )

if len(common_conditions) != 3081:
    print(
        "  WARNING: current direct metadata intersection "
        "does not equal the previously validated 3,081."
    )


# ================================================================
# 7. BUILD CONDITION-LEVEL PAIRED CXR TABLE
# ================================================================

print("\n[5] Building condition-level paired CXR table...")

paired_cxr = cxr[
    cxr["condition_id"]
    .astype(str)
    .isin(common_conditions)
].copy()

print(
    f"  Paired CXR records: {len(paired_cxr):,}"
)

print(
    f"  Paired conditions  : "
    f"{paired_cxr['condition_id'].nunique():,}"
)


# ================================================================
# 8. TARGET DEFINITION — DR / DS
# ================================================================

print("\n[6] Auditing DR/DS target...")

DR_CLASSES = {
    "MDR non XDR",
    "XDR",
    "Mono DR",
    "Poly DR",
    "Pre-XDR",
}

DS_CLASS = "Sensitive"

paired_cxr["type_of_resistance_norm"] = (
    paired_cxr["type_of_resistance"]
    .astype(str)
    .str.strip()
)

unexpected_resistance = sorted(
    set(
        paired_cxr["type_of_resistance_norm"].dropna()
    )
    - DR_CLASSES
    - {DS_CLASS}
    - {""}
)

if unexpected_resistance:
    print(
        "  WARNING: unexpected resistance classes:"
    )
    for x in unexpected_resistance:
        print(f"    - {x}")

def map_dr_ds(value):
    value = normalize_string(value)

    if value == DS_CLASS:
        return "SENSITIVE"

    if value in DR_CLASSES:
        return "RESISTANT"

    return "UNUSABLE"


paired_cxr["DR_DS_target"] = (
    paired_cxr["type_of_resistance_norm"]
    .apply(map_dr_ds)
)


# ------------------------------------------------
# Check condition-level stability
# ------------------------------------------------

target_condition_audit = (
    paired_cxr
    .groupby("condition_id")["DR_DS_target"]
    .agg(
        unique_target_values=lambda s: sorted(
            set(x for x in s if x != "")
        ),
        n_unique_target=lambda s: len(
            set(x for x in s if x != "")
        ),
        n_records="size",
    )
    .reset_index()
)

target_condition_audit["target_stability"] = np.where(
    target_condition_audit["n_unique_target"] == 1,
    "STABLE",
    np.where(
        target_condition_audit["n_unique_target"] == 0,
        "NO_USABLE_TARGET",
        "AMBIGUOUS"
    )
)

target_condition_audit["DR_DS_target"] = (
    target_condition_audit["unique_target_values"]
    .apply(
        lambda x: x[0]
        if len(x) == 1
        else ""
    )
)

print(
    "\n  Condition-level DR/DS stability:"
)

print(
    target_condition_audit[
        "target_stability"
    ].value_counts()
    .to_string()
)

stable_target = target_condition_audit[
    target_condition_audit["target_stability"] == "STABLE"
].copy()

print(
    f"\n  Stable DR/DS conditions: "
    f"{len(stable_target):,}"
)

print("\n  DR/DS distribution:")

print(
    stable_target["DR_DS_target"]
    .value_counts()
    .to_string()
)


# ================================================================
# 9. TARGET DEFINITION AUDIT TABLE
# ================================================================

target_definition_rows = [
    {
        "decision_id": "TARGET-58-01",
        "target_family": "TB_DR_DS",
        "target_variable": "DR_DS_target",
        "source_field": "CXR type_of_resistance",
        "definition": (
            "SENSITIVE = type_of_resistance == 'Sensitive'; "
            "RESISTANT = any predefined drug-resistant class "
            "(MDR non XDR, XDR, Mono DR, Poly DR, Pre-XDR)"
        ),
        "unit": "condition_id",
        "status": (
            "DATA_SUPPORTED"
            if len(stable_target) == len(common_conditions)
            else "REQUIRES_REVIEW"
        ),
        "scientific_note": (
            "Target is derived from the verified CXR metadata "
            "resistance field and must not be supplied as a model "
            "predictor."
        ),
    },
    {
        "decision_id": "TARGET-58-02",
        "target_family": "TB_DR_DS",
        "target_variable": "DR_DS_target",
        "source_field": "CXR type_of_resistance",
        "definition": "Binary Resistant vs Sensitive",
        "unit": "condition_id",
        "status": (
            "STABLE"
            if (
                len(stable_target) == len(common_conditions)
                and len(stable_target) > 0
            )
            else "NOT_STABLE"
        ),
        "scientific_note": (
            "Binary target is scientifically distinct from "
            "multiclass resistance prediction."
        ),
    },
]

target_definition_df = pd.DataFrame(
    target_definition_rows
)


# ================================================================
# 10. GENOMIC FEATURE SCIENTIFIC AUDIT
# ================================================================

print("\n[7] Auditing genomic variables...")

genomic_features = [
    "type_of_resistance",
    "drug_resistance_type",
    "num_drug_resistant_variants",
    "num_other_variants",
    "gene_snp_mutations",
    "main_lineage",
    "sub_lineage",
    "specimen_id",
    "specimen_identifier",
    "specimen_collection_date",
    "specimen_collection_site",
]


genomic_audit_rows = []


# ------------------------------------------------
# 10.1 Direct target variable
# ------------------------------------------------

genomic_audit_rows.append(
    audit_record(
        variable="type_of_resistance",
        source="genomics",
        classification="TARGET_LEAKAGE",
        reason=(
            "This genomic field directly represents the TB "
            "drug-resistance class used to define the DR/DS target."
        ),
        target_relation="DIRECT_TARGET",
        temporal_relation="TARGET_INFORMATION",
        recommendation=(
            "Do not use as a predictor when DR/DS is the target."
        ),
    )
)


# ------------------------------------------------
# 10.2 Drug resistance type
# ------------------------------------------------

genomic_audit_rows.append(
    audit_record(
        variable="drug_resistance_type",
        source="genomics",
        classification="TARGET_LEAKAGE",
        reason=(
            "The field is explicitly resistance-related and can "
            "directly encode or derive the resistance phenotype."
        ),
        target_relation="RESISTANCE_ASSOCIATED_TARGET_INFORMATION",
        temporal_relation="REQUIRES_TEMPORAL_REVIEW",
        recommendation=(
            "Exclude from DR/DS predictor matrix unless a "
            "domain-justified alternative target is defined."
        ),
    )
)


# ------------------------------------------------
# 10.3 Number of drug-resistant variants
# ------------------------------------------------

genomic_audit_rows.append(
    audit_record(
        variable="num_drug_resistant_variants",
        source="genomics",
        classification="REQUIRES_DOMAIN_REVIEW",
        reason=(
            "This quantity explicitly summarizes drug-resistance "
            "associated genomic variants. It may contain direct "
            "or near-direct information about the resistance "
            "phenotype."
        ),
        target_relation="STRONG_RESISTANCE_ASSOCIATION",
        temporal_relation="TEMPORAL_DEPENDENT_POTENTIALLY",
        recommendation=(
            "Do not approve automatically. Requires domain-level "
            "assessment of whether the feature would constitute "
            "target leakage for the intended prediction-time task."
        ),
    )
)


# ------------------------------------------------
# 10.4 Other variant count
# ------------------------------------------------

genomic_audit_rows.append(
    audit_record(
        variable="num_other_variants",
        source="genomics",
        classification="REQUIRES_DOMAIN_REVIEW",
        reason=(
            "A genomic summary statistic. It is not itself the "
            "declared resistance target, but its relationship "
            "with sequencing and resistance-associated variation "
            "must be established."
        ),
        target_relation="INDIRECT",
        temporal_relation="SPECIMEN_DEPENDENT",
        recommendation=(
            "Potential predictor only after genomic-domain review."
        ),
    )
)


# ------------------------------------------------
# 10.5 SNP mutation field
# ------------------------------------------------

genomic_audit_rows.append(
    audit_record(
        variable="gene_snp_mutations",
        source="genomics",
        classification="REQUIRES_DOMAIN_REVIEW",
        reason=(
            "Contains genomic mutation information. Individual "
            "mutations may be biologically associated with drug "
            "resistance, creating a potential target leakage risk "
            "depending on the intended prediction-time definition."
        ),
        target_relation="POTENTIAL_RESISTANCE_ASSOCIATION",
        temporal_relation="SPECIMEN_DEPENDENT",
        recommendation=(
            "Requires explicit mutation-level leakage audit before "
            "being used as a predictor."
        ),
    )
)


# ------------------------------------------------
# 10.6 Lineage
# ------------------------------------------------

genomic_audit_rows.append(
    audit_record(
        variable="main_lineage",
        source="genomics",
        classification="REQUIRES_DOMAIN_REVIEW",
        reason=(
            "M. tuberculosis lineage information is not itself "
            "the declared DR/DS target, but lineage can be "
            "associated with geographic distribution and "
            "resistance patterns."
        ),
        target_relation="INDIRECT_EPIGENETIC/POPULATION_ASSOCIATION",
        temporal_relation="SPECIMEN_DEPENDENT",
        recommendation=(
            "Potential predictor after checking biological "
            "definition, availability at prediction time, and "
            "site/geographic confounding."
        ),
    )
)


# ------------------------------------------------
# 10.7 Sub-lineage
# ------------------------------------------------

genomic_audit_rows.append(
    audit_record(
        variable="sub_lineage",
        source="genomics",
        classification="REQUIRES_DOMAIN_REVIEW",
        reason=(
            "Sub-lineage may contain population-structure information "
            "associated with resistance and geographic/site effects."
        ),
        target_relation="INDIRECT_POPULATION_ASSOCIATION",
        temporal_relation="SPECIMEN_DEPENDENT",
        recommendation=(
            "Potential predictor after domain and temporal review."
        ),
    )
)


# ------------------------------------------------
# 10.8 Specimen identifiers
# ------------------------------------------------

genomic_audit_rows.append(
    audit_record(
        variable="specimen_id",
        source="genomics",
        classification="DESCRIPTIVE_ONLY",
        reason=(
            "Identifier for the biological specimen. It has no "
            "scientifically meaningful predictive interpretation "
            "and risks memorization."
        ),
        target_relation="NONE",
        temporal_relation="IDENTIFIER",
        recommendation="Do not use as a model predictor.",
    )
)

genomic_audit_rows.append(
    audit_record(
        variable="specimen_identifier",
        source="genomics",
        classification="DESCRIPTIVE_ONLY",
        reason=(
            "Specimen identifier. It should not be interpreted "
            "as a biological feature."
        ),
        target_relation="NONE",
        temporal_relation="IDENTIFIER",
        recommendation="Do not use as a model predictor.",
    )
)


# ------------------------------------------------
# 10.9 Collection date
# ------------------------------------------------

genomic_audit_rows.append(
    audit_record(
        variable="specimen_collection_date",
        source="genomics",
        classification="TEMPORAL_DEPENDENT",
        reason=(
            "The collection date determines whether genomic "
            "information existed before, on, or after the CXR "
            "prediction time."
        ),
        target_relation="NONE",
        temporal_relation="CRITICAL_TEMPORAL_FIELD",
        recommendation=(
            "Use for cohort eligibility and temporal reconciliation, "
            "not as an unrestricted predictive feature."
        ),
    )
)


# ------------------------------------------------
# 10.10 Collection site
# ------------------------------------------------

genomic_audit_rows.append(
    audit_record(
        variable="specimen_collection_site",
        source="genomics",
        classification="REQUIRES_DOMAIN_REVIEW",
        reason=(
            "Collection site is metadata about where the specimen "
            "was obtained. It may be useful descriptively but "
            "can also encode clinical context or acquisition "
            "differences."
        ),
        target_relation="NONE",
        temporal_relation="SPECIMEN_METADATA",
        recommendation=(
            "Requires domain review before use as a predictor."
        ),
    )
)


genomic_audit_df = pd.DataFrame(
    genomic_audit_rows
)


# ================================================================
# 11. EMPIRICAL GENOMIC FIELD AUDIT
# ================================================================

print("\n[8] Generating empirical genomic field statistics...")

empirical_rows = []

for field in genomic_features:

    if field not in genomics.columns:
        continue

    series = genomics[field]

    empirical_rows.append(
        {
            "field": field,
            "dtype": str(series.dtype),
            "rows": len(series),
            "non_null": int(series.notna().sum()),
            "missing": int(series.isna().sum()),
            "missing_percent": round(
                series.isna().mean() * 100,
                4
            ),
            "unique_non_null": count_unique_non_null(series),
            "example_values": " | ".join(
                unique_non_null(series)[:10]
            ),
        }
    )

empirical_genomic_df = pd.DataFrame(
    empirical_rows
)


# ================================================================
# 12. RESISTANCE-ASSOCIATED VALUE AUDIT
# ================================================================

print("\n[9] Inspecting resistance-associated genomic fields...")

resistance_field_audit_rows = []

for field in [
    "type_of_resistance",
    "drug_resistance_type",
    "num_drug_resistant_variants",
]:
    if field not in genomics.columns:
        continue

    series = genomics[field]

    resistance_field_audit_rows.append(
        {
            "field": field,
            "unique_non_null": count_unique_non_null(series),
            "non_null": int(series.notna().sum()),
            "missing": int(series.isna().sum()),
            "values_or_summary": (
                " | ".join(unique_non_null(series)[:30])
            ),
        }
    )

resistance_field_audit_df = pd.DataFrame(
    resistance_field_audit_rows
)


# ================================================================
# 13. CXR / CLINICAL FEATURE LEAKAGE AUDIT
# ================================================================

print("\n[10] Auditing CXR/clinical variables for leakage...")

cxr_audit_rows = []


# ------------------------------------------------
# Direct target
# ------------------------------------------------

cxr_audit_rows.append(
    audit_record(
        variable="type_of_resistance",
        source="CXR",
        classification="TARGET_LEAKAGE",
        reason=(
            "This field is used to construct the DR/DS target."
        ),
        target_relation="DIRECT_TARGET",
        temporal_relation="TARGET_INFORMATION",
        recommendation=(
            "Never include as an input feature for DR/DS prediction."
        ),
    )
)


# ------------------------------------------------
# Outcome
# ------------------------------------------------

cxr_audit_rows.append(
    audit_record(
        variable="outcome",
        source="CXR",
        classification="POST_TARGET_INFORMATION",
        reason=(
            "Treatment outcome is downstream of the patient's "
            "clinical course and is not available at baseline "
            "prediction time."
        ),
        target_relation="DIFFERENT_TARGET_FAMILY",
        temporal_relation="POST_TREATMENT_INFORMATION",
        recommendation=(
            "Exclude from DR/DS predictors."
        ),
    )
)


# ------------------------------------------------
# Diagnosis code
# ------------------------------------------------

cxr_audit_rows.append(
    audit_record(
        variable="diagnosis_code",
        source="CXR",
        classification="REQUIRES_DOMAIN_REVIEW",
        reason=(
            "Diagnosis coding contains clinical diagnostic "
            "information and must be assessed according to the "
            "exact prediction question."
        ),
        target_relation="CLINICAL_DIAGNOSIS",
        temporal_relation="PREDICTION_TIME_DEPENDENT",
        recommendation=(
            "Do not automatically use. Requires domain and "
            "prediction-time justification."
        ),
    )
)


# ------------------------------------------------
# Case definition
# ------------------------------------------------

cxr_audit_rows.append(
    audit_record(
        variable="case_definition",
        source="CXR",
        classification="REQUIRES_DOMAIN_REVIEW",
        reason=(
            "Case definition contains clinical information "
            "about TB presentation/history and may be associated "
            "with resistance or treatment history."
        ),
        target_relation="CLINICAL_CONTEXT",
        temporal_relation="PREDICTION_TIME_DEPENDENT",
        recommendation=(
            "Requires explicit availability-at-prediction-time review."
        ),
    )
)


# ------------------------------------------------
# CXR outlier
# ------------------------------------------------

cxr_audit_rows.append(
    audit_record(
        variable="cxr_outlier",
        source="CXR",
        classification="DESCRIPTIVE_ONLY",
        reason=(
            "This field was audited previously as an image-quality/"
            "outlier indicator. 'normal' in this field is NOT a "
            "clinical Normal diagnosis."
        ),
        target_relation="NONE",
        temporal_relation="IMAGE_QC",
        recommendation=(
            "May be used for QC auditing, but must not be "
            "interpreted as a clinical target."
        ),
    )
)


# ------------------------------------------------
# Demographics
# ------------------------------------------------

for field in [
    "age_of_onset",
    "sex",
    "bmi",
    "country",
    "comorbidity",
]:
    cxr_audit_rows.append(
        audit_record(
            variable=field,
            source="CXR",
            classification="REQUIRES_DOMAIN_REVIEW",
            reason=(
                "Potential clinical/demographic predictor. "
                "Use depends on the final prediction-time "
                "research design and confounding assessment."
            ),
            target_relation="INDIRECT",
            temporal_relation="PREDICTION_TIME_DEPENDENT",
            recommendation=(
                "Potential predictor after final design lock."
            ),
        )
    )


# ------------------------------------------------
# IDs
# ------------------------------------------------

for field in [
    "patient_id",
    "condition_id",
    "imagingstudy_id",
    "series_instance_content_url",
]:
    cxr_audit_rows.append(
        audit_record(
            variable=field,
            source="CXR",
            classification="DESCRIPTIVE_ONLY",
            reason=(
                "Identifier or resource locator; not a biological "
                "predictive feature."
            ),
            target_relation="NONE",
            temporal_relation="IDENTIFIER",
            recommendation="Do not use as a predictor.",
        )
    )


# ------------------------------------------------
# Imaging date
# ------------------------------------------------

cxr_audit_rows.append(
    audit_record(
        variable="imaging_date",
        source="CXR",
        classification="TEMPORAL_DEPENDENT",
        reason=(
            "Defines the CXR prediction time and is required "
            "for temporal reconciliation with genomic specimens."
        ),
        target_relation="NONE",
        temporal_relation="CRITICAL_PREDICTION_TIME",
        recommendation=(
            "Use for eligibility and temporal analysis rather than "
            "as an unrestricted predictive feature."
        ),
    )
)


cxr_audit_df = pd.DataFrame(
    cxr_audit_rows
)


# ================================================================
# 14. TEMPORAL FIELD AUDIT
# ================================================================

print("\n[11] Auditing temporal variables...")

temporal_audit_rows = [
    {
        "field": "imaging_date",
        "source": "CXR",
        "role": "PREDICTION_TIME",
        "classification": "TEMPORAL_DEPENDENT",
        "scientific_use": (
            "Defines when the CXR observation is available."
        ),
        "model_feature_status": "NOT_AUTOMATICALLY_APPROVED",
    },
    {
        "field": "specimen_collection_date",
        "source": "Genomics",
        "role": "SPECIMEN_TIME",
        "classification": "TEMPORAL_DEPENDENT",
        "scientific_use": (
            "Defines when genomic information becomes available."
        ),
        "model_feature_status": "NOT_AUTOMATICALLY_APPROVED",
    },
    {
        "field": "signed_delta_days",
        "source": "Temporal reconciliation",
        "role": "CXR_GENOMIC_TIME_DIFFERENCE",
        "classification": "TEMPORAL_DEPENDENT",
        "scientific_use": (
            "Determines whether genomics occurred before, on, "
            "or after the CXR."
        ),
        "model_feature_status": "ELIGIBILITY_VARIABLE",
    },
    {
        "field": "temporal_direction",
        "source": "Temporal reconciliation",
        "role": "PREDICTION_TIME_RELATION",
        "classification": "TEMPORAL_DEPENDENT",
        "scientific_use": (
            "Identifies genomic-before-CXR, same-day, or "
            "genomic-after-CXR relationships."
        ),
        "model_feature_status": "ELIGIBILITY_VARIABLE",
    },
]

temporal_audit_df = pd.DataFrame(
    temporal_audit_rows
)


# ================================================================
# 15. DIRECT LEAKAGE REGISTER
# ================================================================

print("\n[12] Creating direct leakage register...")

leakage_rows = [
    {
        "variable": "CXR type_of_resistance",
        "source": "CXR",
        "leakage_type": "DIRECT_TARGET_LEAKAGE",
        "status": "PROHIBITED",
        "reason": (
            "Used to define DR/DS target."
        ),
    },
    {
        "variable": "Genomic type_of_resistance",
        "source": "Genomics",
        "leakage_type": "DIRECT_TARGET_LEAKAGE",
        "status": "PROHIBITED",
        "reason": (
            "Directly encodes resistance phenotype."
        ),
    },
    {
        "variable": "Genomic drug_resistance_type",
        "source": "Genomics",
        "leakage_type": "RESISTANCE_TARGET_INFORMATION",
        "status": "PROHIBITED_PENDING_DOMAIN_REVIEW",
        "reason": (
            "Resistance-associated field; current task is DR/DS."
        ),
    },
    {
        "variable": "Genomic num_drug_resistant_variants",
        "source": "Genomics",
        "leakage_type": "POTENTIAL_NEAR_TARGET_LEAKAGE",
        "status": "REQUIRES_DOMAIN_REVIEW",
        "reason": (
            "Explicitly summarizes drug-resistance-associated "
            "genomic variation."
        ),
    },
    {
        "variable": "Genomic gene_snp_mutations",
        "source": "Genomics",
        "leakage_type": "POTENTIAL_BIOLOGICAL_TARGET_LEAKAGE",
        "status": "REQUIRES_DOMAIN_REVIEW",
        "reason": (
            "Individual mutations may encode resistance mechanisms."
        ),
    },
    {
        "variable": "CXR outcome",
        "source": "CXR",
        "leakage_type": "POST_TREATMENT_INFORMATION",
        "status": "PROHIBITED_FOR_DR_DS",
        "reason": (
            "Outcome occurs after treatment/course progression."
        ),
    },
    {
        "variable": "CXR cxr_outlier",
        "source": "CXR",
        "leakage_type": "QC_METADATA",
        "status": "NOT_A_CLINICAL_NORMAL_TARGET",
        "reason": (
            "Previously audited image-QC field; normal does not "
            "mean clinically normal."
        ),
    },
]

leakage_df = pd.DataFrame(
    leakage_rows
)


# ================================================================
# 16. GENOMIC VALUE DISTRIBUTION
# ================================================================

print("\n[13] Computing genomic value distributions...")

distribution_outputs = {}

for field in [
    "type_of_resistance",
    "drug_resistance_type",
    "specimen_collection_site",
    "main_lineage",
    "sub_lineage",
]:
    if field in genomics.columns:
        counts = (
            genomics[field]
            .fillna("<MISSING>")
            .astype(str)
            .str.strip()
            .value_counts(dropna=False)
            .rename_axis(field)
            .reset_index(name="count")
        )

        counts["percentage"] = (
            counts["count"]
            / len(genomics)
            * 100
        ).round(4)

        distribution_outputs[field] = counts


# ================================================================
# 17. CHECK GENOMIC TARGET AGREEMENT AT CONDITION LEVEL
# ================================================================

print("\n[14] Comparing CXR and genomic resistance labels...")

genomic_common = genomics[
    genomics["condition_id"]
    .astype(str)
    .isin(common_conditions)
].copy()

genomic_common["condition_id"] = (
    genomic_common["condition_id"].astype(str)
)

genomic_common["genomic_type_of_resistance_norm"] = (
    genomic_common["type_of_resistance"]
    .astype(str)
    .str.strip()
)

genomic_condition_resistance = (
    genomic_common
    .groupby("condition_id")[
        "genomic_type_of_resistance_norm"
    ]
    .agg(
        genomic_unique_values=lambda s: sorted(
            set(x for x in s if x != "")
        ),
        genomic_n_unique=lambda s: len(
            set(x for x in s if x != "")
        ),
        genomic_record_count="size",
    )
    .reset_index()
)

genomic_condition_resistance[
    "genomic_resistance_stability"
] = np.where(
    genomic_condition_resistance["genomic_n_unique"] == 1,
    "STABLE",
    np.where(
        genomic_condition_resistance["genomic_n_unique"] == 0,
        "NO_VALUE",
        "AMBIGUOUS"
    )
)

genomic_condition_resistance[
    "genomic_type_of_resistance"
] = (
    genomic_condition_resistance[
        "genomic_unique_values"
    ]
    .apply(
        lambda x: x[0]
        if len(x) == 1
        else ""
    )
)


# ================================================================
# 18. MERGE CXR AND GENOMIC TARGET AUDIT
# ================================================================

condition_target_audit = target_condition_audit[
    [
        "condition_id",
        "DR_DS_target",
        "target_stability",
        "n_records",
    ]
].merge(
    genomic_condition_resistance[
        [
            "condition_id",
            "genomic_type_of_resistance",
            "genomic_resistance_stability",
            "genomic_record_count",
        ]
    ],
    on="condition_id",
    how="outer",
)

condition_target_audit[
    "exact_resistance_class_match"
] = (
    condition_target_audit[
        "DR_DS_target"
    ]
    .map(
        {
            "SENSITIVE": "Sensitive",
            "RESISTANT": np.nan,
        }
    )
)

# A binary comparison is performed explicitly below.

def genomic_to_dr_ds(value):
    value = normalize_string(value)

    if value == DS_CLASS:
        return "SENSITIVE"

    if value in DR_CLASSES:
        return "RESISTANT"

    return "UNUSABLE"


condition_target_audit[
    "genomic_DR_DS_target"
] = (
    condition_target_audit[
        "genomic_type_of_resistance"
    ]
    .apply(genomic_to_dr_ds)
)

condition_target_audit[
    "CXR_vs_genomic_DR_DS_agreement"
] = np.where(
    (
        condition_target_audit["DR_DS_target"].isin(
            ["SENSITIVE", "RESISTANT"]
        )
        &
        condition_target_audit["genomic_DR_DS_target"].isin(
            ["SENSITIVE", "RESISTANT"]
        )
    ),
    np.where(
        condition_target_audit["DR_DS_target"]
        ==
        condition_target_audit["genomic_DR_DS_target"],
        "AGREE",
        "DISAGREE"
    ),
    "NOT_COMPARABLE"
)


# ================================================================
# 19. MULTIPLE GENOMIC TIMEPOINT AUDIT
# ================================================================

print("\n[15] Auditing repeated genomic specimens...")

timepoint_conditions = set(
    genomic_timepoints["condition_id"]
    .dropna()
    .astype(str)
)

multiple_timepoint_counts = (
    genomic_timepoints
    .assign(
        condition_id=
        genomic_timepoints["condition_id"].astype(str)
    )
    .groupby("condition_id")
    .size()
)

multiple_timepoint_condition_df = (
    multiple_timepoint_counts[
        multiple_timepoint_counts > 1
    ]
    .rename("genomic_timepoint_count")
    .reset_index()
)

print(
    f"  Conditions with >1 genomic timepoint: "
    f"{len(multiple_timepoint_condition_df):,}"
)


# ================================================================
# 20. TEMPORAL DIRECTION SUMMARY
# ================================================================

print("\n[16] Auditing temporal direction...")

if "temporal_direction" in temporal_pairs.columns:

    temporal_direction_summary = (
        temporal_pairs["temporal_direction"]
        .fillna("<MISSING>")
        .astype(str)
        .value_counts()
        .rename_axis("temporal_direction")
        .reset_index(name="pair_count")
    )

    temporal_direction_summary["percentage"] = (
        temporal_direction_summary["pair_count"]
        / len(temporal_pairs)
        * 100
    ).round(4)

else:
    temporal_direction_summary = pd.DataFrame(
        columns=[
            "temporal_direction",
            "pair_count",
            "percentage",
        ]
    )


# ================================================================
# 21. SCIENTIFIC DECISION GATES
# ================================================================

print("\n[17] Creating scientific decision gates...")

stable_target_count = len(
    stable_target
)

paired_condition_count = len(
    common_conditions
)

direct_target_leakage_present = True

normal_label_artificial = False

cxr_outlier_used_as_normal = False

missing_genomics_used_as_normal = False


decision_gate_rows = [

    {
        "gate_id": "G58-01",
        "decision_area": "Paired CXR + genomic cohort",
        "criterion": "Same condition_id",
        "result": paired_condition_count,
        "status": (
            "PASS"
            if paired_condition_count == 3081
            else "REQUIRES_REVIEW"
        ),
        "scientific_action": (
            "Retain verified same-condition pairing."
        ),
    },

    {
        "gate_id": "G58-02",
        "decision_area": "DR/DS target",
        "criterion": "Stable condition-level target",
        "result": stable_target_count,
        "status": (
            "PASS"
            if stable_target_count == paired_condition_count
            else "REQUIRES_REVIEW"
        ),
        "scientific_action": (
            "Target is technically stable at condition level."
        ),
    },

    {
        "gate_id": "G58-03",
        "decision_area": "Direct target leakage",
        "criterion": "Resistance target fields excluded",
        "result": "Direct leakage identified",
        "status": "LOCKED",
        "scientific_action": (
            "Do not use CXR/genomic type_of_resistance "
            "as a predictor for DR/DS."
        ),
    },

    {
        "gate_id": "G58-04",
        "decision_area": "Genomic resistance variables",
        "criterion": "Near-target genomic fields reviewed",
        "result": (
            "drug_resistance_type / "
            "num_drug_resistant_variants / "
            "gene_snp_mutations require review"
        ),
        "status": "PENDING_DOMAIN_REVIEW",
        "scientific_action": (
            "Do not automatically approve these predictors."
        ),
    },

    {
        "gate_id": "G58-05",
        "decision_area": "Temporal design",
        "criterion": "Genomic availability relative to CXR",
        "result": (
            "Requires explicit prediction-time eligibility rule"
        ),
        "status": "PENDING",
        "scientific_action": (
            "Proceed to Step 3B.6.59 temporal prediction-time audit."
        ),
    },

    {
        "gate_id": "G58-06",
        "decision_area": "Artificial Normal",
        "criterion": "No artificial Normal labels",
        "result": normal_label_artificial,
        "status": "PASS",
        "scientific_action": (
            "No artificial Normal class permitted."
        ),
    },

    {
        "gate_id": "G58-07",
        "decision_area": "cxr_outlier",
        "criterion": "Do not interpret cxr_outlier='normal' clinically",
        "result": cxr_outlier_used_as_normal,
        "status": "PASS",
        "scientific_action": (
            "Use only as image-QC metadata if needed."
        ),
    },

    {
        "gate_id": "G58-08",
        "decision_area": "Missing genomics",
        "criterion": "Missing genomics not treated as Normal",
        "result": missing_genomics_used_as_normal,
        "status": "PASS",
        "scientific_action": (
            "No missing-modality-as-Normal labeling."
        ),
    },

    {
        "gate_id": "G58-09",
        "decision_area": "Modeling",
        "criterion": "No modeling before scientific lock",
        "result": "NOT_STARTED",
        "status": "LOCKED",
        "scientific_action": (
            "Do not start preprocessing/modeling yet."
        ),
    },
]

decision_gates_df = pd.DataFrame(
    decision_gate_rows
)


# ================================================================
# 22. FINAL FEATURE STATUS MATRIX
# ================================================================

print("\n[18] Creating final feature status matrix...")

feature_status_matrix = pd.concat(
    [
        genomic_audit_df,
        cxr_audit_df,
    ],
    ignore_index=True
)

feature_status_matrix[
    "approved_for_DR_DS_model_now"
] = np.where(
    feature_status_matrix["classification"]
    == "SAFE_PREDICTOR",
    "YES",
    "NO"
)


# ================================================================
# 23. IMPORTANT: NO VARIABLE IS AUTO-APPROVED
# ================================================================

# This is intentional.
#
# We currently do NOT have enough scientific justification to
# automatically mark genomic mutation/resistance-associated
# variables as safe predictors.
#
# Therefore, the current audit contains:
#
# SAFE_PREDICTOR:
#     None automatically approved at this stage.
#
# TARGET_LEAKAGE:
#     Direct target fields.
#
# REQUIRES_DOMAIN_REVIEW:
#     Genomic variables with biological/resistance relationships.
#
# TEMPORAL_DEPENDENT:
#     Collection/imaging dates and temporal relationships.
#
# DESCRIPTIVE_ONLY:
#     IDs and QC metadata.


safe_predictors = feature_status_matrix[
    feature_status_matrix["classification"]
    == "SAFE_PREDICTOR"
]["variable"].tolist()

print(
    "\n  Automatically approved SAFE_PREDICTOR fields:"
)

if safe_predictors:
    for feature in safe_predictors:
        print(f"    - {feature}")
else:
    print("    NONE")


# ================================================================
# 24. VALIDATION TESTS
# ================================================================

print("\n[19] Running validation tests...")

validation_rows = []


def add_validation(
    validation_id,
    test,
    status,
    observed,
    expected,
):
    validation_rows.append(
        {
            "validation_id": validation_id,
            "test": test,
            "status": status,
            "observed": observed,
            "expected": expected,
        }
    )


add_validation(
    "VAL-58-01",
    "CXR metadata exists",
    "PASS" if CXR_PATH.exists() else "FAIL",
    str(CXR_PATH),
    "Existing file",
)

add_validation(
    "VAL-58-02",
    "Genomics metadata exists",
    "PASS" if GENOMICS_PATH.exists() else "FAIL",
    str(GENOMICS_PATH),
    "Existing file",
)

add_validation(
    "VAL-58-03",
    "Paired condition count",
    "PASS"
    if paired_condition_count == 3081
    else "FAIL",
    paired_condition_count,
    3081,
)

add_validation(
    "VAL-58-04",
    "Stable DR/DS target count",
    "PASS"
    if stable_target_count == 3081
    else "REVIEW",
    stable_target_count,
    3081,
)

add_validation(
    "VAL-58-05",
    "Artificial Normal labels",
    "PASS"
    if normal_label_artificial is False
    else "FAIL",
    normal_label_artificial,
    False,
)

add_validation(
    "VAL-58-06",
    "Missing genomics treated as Normal",
    "PASS"
    if missing_genomics_used_as_normal is False
    else "FAIL",
    missing_genomics_used_as_normal,
    False,
)

add_validation(
    "VAL-58-07",
    "cxr_outlier normal treated as clinical Normal",
    "PASS"
    if cxr_outlier_used_as_normal is False
    else "FAIL",
    cxr_outlier_used_as_normal,
    False,
)

add_validation(
    "VAL-58-08",
    "Direct resistance target leakage recognized",
    "PASS"
    if direct_target_leakage_present
    else "FAIL",
    direct_target_leakage_present,
    True,
)

validation_df = pd.DataFrame(
    validation_rows
)


# ================================================================
# 25. SAVE TARGET DEFINITION
# ================================================================

target_condition_audit.to_csv(
    OUTPUT_DIR
    / "06_Condition_Level_Target_Audit.csv",
    index=False
)

target_definition_df.to_csv(
    OUTPUT_DIR
    / "03_Target_Definition_Audit.csv",
    index=False
)


# ================================================================
# 26. SAVE GENOMIC AUDIT
# ================================================================

genomic_audit_df.to_csv(
    OUTPUT_DIR
    / "01_Genomic_Feature_Audit.csv",
    index=False
)

empirical_genomic_df.to_csv(
    OUTPUT_DIR
    / "01B_Genomic_Field_Empirical_Statistics.csv",
    index=False
)

resistance_field_audit_df.to_csv(
    OUTPUT_DIR
    / "01C_Resistance_Field_Value_Audit.csv",
    index=False
)


# ================================================================
# 27. SAVE CXR AUDIT
# ================================================================

cxr_audit_df.to_csv(
    OUTPUT_DIR
    / "02_CXR_Feature_Audit.csv",
    index=False
)


# ================================================================
# 28. SAVE TEMPORAL AUDIT
# ================================================================

temporal_audit_df.to_csv(
    OUTPUT_DIR
    / "04_Temporal_Field_Audit.csv",
    index=False
)

temporal_direction_summary.to_csv(
    OUTPUT_DIR
    / "04B_Temporal_Direction_Summary.csv",
    index=False
)


# ================================================================
# 29. SAVE LEAKAGE AUDIT
# ================================================================

leakage_df.to_csv(
    OUTPUT_DIR
    / "05_Leakage_Audit.csv",
    index=False
)


# ================================================================
# 30. SAVE CONDITION TARGET COMPARISON
# ================================================================

condition_target_audit.to_csv(
    OUTPUT_DIR
    / "06B_CXR_Genomic_Target_Comparison.csv",
    index=False
)


# ================================================================
# 31. SAVE MULTIPLE GENOMIC TIMEPOINT AUDIT
# ================================================================

multiple_timepoint_condition_df.to_csv(
    OUTPUT_DIR
    / "06C_Multiple_Genomic_Timepoint_Conditions.csv",
    index=False
)


# ================================================================
# 32. SAVE FEATURE STATUS MATRIX
# ================================================================

feature_status_matrix.to_csv(
    OUTPUT_DIR
    / "06D_Final_Feature_Status_Matrix.csv",
    index=False
)


# ================================================================
# 33. SAVE DECISION GATES
# ================================================================

decision_gates_df.to_csv(
    OUTPUT_DIR
    / "07_Scientific_Decision_Gates.csv",
    index=False
)


# ================================================================
# 34. SAVE VALIDATION
# ================================================================

validation_df.to_csv(
    OUTPUT_DIR
    / "08_Validation.csv",
    index=False
)


# ================================================================
# 35. FINAL SCIENTIFIC SUMMARY
# ================================================================

classification_counts = (
    feature_status_matrix[
        "classification"
    ]
    .value_counts()
    .to_dict()
)

target_distribution = (
    stable_target[
        "DR_DS_target"
    ]
    .value_counts()
    .to_dict()
)

agreement_counts = (
    condition_target_audit[
        "CXR_vs_genomic_DR_DS_agreement"
    ]
    .value_counts()
    .to_dict()
)


summary = {
    "step": "3B.6.58",
    "title": "Target & Genomic Feature Scientific Audit",

    "execution_timestamp_utc": (
        datetime.utcnow()
        .replace(microsecond=0)
        .isoformat()
        + "Z"
    ),

    "input_files": {
        "cxr_metadata": str(CXR_PATH),
        "genomics_metadata": str(GENOMICS_PATH),
        "cxr_master": str(CXR_MASTER_PATH),
        "temporal_pairs": str(TEMPORAL_PAIRS_PATH),
        "genomic_timepoints": str(GENOMIC_TIMEPOINT_PATH),
    },

    "cohort": {
        "cxr_rows": int(len(cxr)),
        "genomic_rows": int(len(genomics)),
        "paired_cxr_rows": int(len(paired_cxr)),
        "paired_conditions": int(paired_condition_count),
        "genomic_timepoints": int(len(genomic_timepoints)),
        "multiple_genomic_timepoint_conditions": int(
            len(multiple_timepoint_condition_df)
        ),
    },

    "dr_ds_target": {
        "definition": (
            "SENSITIVE = Sensitive; "
            "RESISTANT = MDR non XDR, XDR, Mono DR, "
            "Poly DR, or Pre-XDR"
        ),
        "stable_conditions": int(stable_target_count),
        "distribution": {
            str(k): int(v)
            for k, v in target_distribution.items()
        },
    },

    "feature_classification_counts": {
        str(k): int(v)
        for k, v in classification_counts.items()
    },

    "cxr_genomic_target_agreement": {
        str(k): int(v)
        for k, v in agreement_counts.items()
    },

    "scientific_locks": {
        "artificial_normal_labels": False,
        "missing_genomics_as_normal": False,
        "cxr_outlier_normal_as_clinical_normal": False,
        "direct_target_leakage_recognized": True,
        "model_training_started": False,
        "preprocessing_started": False,
        "segmentation_started": False,
    },

    "current_status": (
        "TARGET_TECHNICALLY_SUPPORTED; "
        "GENOMIC_FEATURE_SET_NOT_YET_FULLY_APPROVED; "
        "TEMPORAL_PREDICTION_TIME_RULE_PENDING"
    ),

    "next_step": (
        "Step 3B.6.59 — Temporal Prediction-Time Cohort Audit"
    ),
}


with open(
    OUTPUT_DIR
    / "09_Step_3B_6_58_Summary.json",
    "w",
    encoding="utf-8",
) as f:
    json.dump(
        summary,
        f,
        indent=4,
        ensure_ascii=False,
        default=str,
    )


# ================================================================
# 36. PRINT FINAL REPORT
# ================================================================

print("\n" + "=" * 80)
print("STEP 3B.6.58 COMPLETED")
print("=" * 80)

print("\nCOHORT")
print("-" * 80)
print(
    f"Paired CXR + genomic conditions : "
    f"{paired_condition_count:,}"
)
print(
    f"Paired CXR records              : "
    f"{len(paired_cxr):,}"
)
print(
    f"Unique genomic timepoints       : "
    f"{len(genomic_timepoints):,}"
)
print(
    f"Multiple genomic-timepoint cases: "
    f"{len(multiple_timepoint_condition_df):,}"
)

print("\nDR/DS TARGET")
print("-" * 80)
print(
    f"Stable condition-level targets  : "
    f"{stable_target_count:,}"
)

for label, count in target_distribution.items():
    print(
        f"  {label:<12}: {count:,}"
    )

print("\nFEATURE CLASSIFICATION")
print("-" * 80)

for label, count in classification_counts.items():
    print(
        f"  {label:<30}: {count:,}"
    )

print("\nIMPORTANT LEAKAGE LOCKS")
print("-" * 80)
print(
    "CXR type_of_resistance        -> TARGET LEAKAGE"
)
print(
    "Genomic type_of_resistance    -> TARGET LEAKAGE"
)
print(
    "Genomic drug_resistance_type  -> REQUIRES DOMAIN REVIEW"
)
print(
    "num_drug_resistant_variants   -> REQUIRES DOMAIN REVIEW"
)
print(
    "gene_snp_mutations            -> REQUIRES DOMAIN REVIEW"
)

print("\nTEMPORAL LOCK")
print("-" * 80)
print(
    "Prediction-time genomic availability "
    "has NOT yet been finalized."
)

print("\nSCIENTIFIC SAFETY LOCKS")
print("-" * 80)
print(
    "Artificial Normal labels       : NOT USED"
)
print(
    "Missing genomics -> Normal     : NOT USED"
)
print(
    "cxr_outlier='normal' -> Normal : NOT USED"
)
print(
    "Model training                  : NOT STARTED"
)
print(
    "Preprocessing                   : NOT STARTED"
)
print(
    "Segmentation                    : NOT STARTED"
)

print("\nOUTPUT DIRECTORY")
print("-" * 80)
print(OUTPUT_DIR)

print("\nNEXT SCIENTIFIC STEP")
print("-" * 80)
print(
    "3B.6.59 — Temporal Prediction-Time Cohort Audit"
)

print("\nSTATUS")
print("-" * 80)
print(
    "TARGET TECHNICALLY SUPPORTED"
)
print(
    "BUT GENOMIC FEATURE APPROVAL AND "
    "TEMPORAL PREDICTION-TIME RULE ARE STILL PENDING."
)

print("=" * 80)
```

```python
# ================================================================
# STEP 3B.6.59
# TEMPORAL PREDICTION-TIME COHORT AUDIT
# ================================================================
#
# PURPOSE
# -------
# Determine which genomic specimen(s) are legitimately available
# at the time of the CXR observation.
#
# This step evaluates:
#
#   1. CXR imaging time
#   2. Genomic specimen collection time
#   3. BEFORE / SAME-DAY / AFTER relationships
#   4. Multiple genomic timepoints
#   5. Nearest genomic specimen
#   6. Temporal windows
#   7. Candidate prediction-time eligibility rules
#   8. Patient/condition independence
#   9. Temporal ambiguity and ties
#
# IMPORTANT SCIENTIFIC LOCKS
# --------------------------
# - No model training
# - No image preprocessing
# - No segmentation
# - No automatic research-question selection
# - No artificial Normal labels
# - No missing-genomics-as-Normal
# - No use of target variables as predictors
# - No image-level random splitting
#
# ================================================================


from pathlib import Path
import pandas as pd
import numpy as np
import json
from datetime import datetime


# ================================================================
# 1. CONFIGURATION
# ================================================================

CXR_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_CXRs_March_2025.csv"
)

GENOMICS_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_Genomics_March_2025.csv"
)

CXR_MASTER_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_24_Notebook_03"
    r"\TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

TEMPORAL_PAIRS_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
    r"\TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

GENOMIC_TIMEPOINT_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
    r"\TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

STEP_58_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_58_Target_Genomic_Feature_Scientific_Audit"
)

OUTPUT_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_59_Temporal_Prediction_Time_Cohort_Audit"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# 2. REQUIRED FILE CHECK
# ================================================================

print("=" * 80)
print("STEP 3B.6.59 — TEMPORAL PREDICTION-TIME COHORT AUDIT")
print("=" * 80)

print("\n[1] Checking required files...")

required_files = {
    "CXR metadata": CXR_PATH,
    "Genomics metadata": GENOMICS_PATH,
    "CXR master": CXR_MASTER_PATH,
    "Temporal pairs": TEMPORAL_PAIRS_PATH,
    "Unique genomic timepoints": GENOMIC_TIMEPOINT_PATH,
    "Step 3B.6.58 output": STEP_58_DIR,
}

for label, path in required_files.items():

    if not path.exists():
        raise FileNotFoundError(
            f"REQUIRED INPUT NOT FOUND:\n"
            f"{label}\n"
            f"{path}"
        )

    print(f"  PASS: {label}")
    print(f"        {path}")


# ================================================================
# 3. LOAD DATA
# ================================================================

print("\n[2] Loading validated datasets...")

cxr = pd.read_csv(
    CXR_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)

cxr_master = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

temporal_pairs = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)

genomic_timepoints = pd.read_csv(
    GENOMIC_TIMEPOINT_PATH,
    low_memory=False
)

print(
    f"  CXR metadata rows       : {len(cxr):,}"
)

print(
    f"  Genomics metadata rows  : {len(genomics):,}"
)

print(
    f"  CXR master rows         : {len(cxr_master):,}"
)

print(
    f"  Temporal pair rows      : {len(temporal_pairs):,}"
)

print(
    f"  Genomic timepoints      : {len(genomic_timepoints):,}"
)


# ================================================================
# 4. SCHEMA VALIDATION
# ================================================================

print("\n[3] Validating schemas...")

required_cxr = [
    "patient_id",
    "condition_id",
    "imaging_date",
    "series_instance_content_url",
    "type_of_resistance",
    "outcome",
]

required_genomics = [
    "condition_id",
    "specimen_id",
    "specimen_identifier",
    "specimen_collection_date",
    "type_of_resistance",
]

required_master = [
    "patient_id",
    "condition_id",
    "imaging_date",
    "series_instance_content_url",
]

required_temporal = [
    "condition_id",
]

required_timepoints = [
    "condition_id",
    "specimen_id",
    "specimen_collection_date_num",
]

for label, df, required_columns in [
    ("CXR", cxr, required_cxr),
    ("Genomics", genomics, required_genomics),
    ("CXR master", cxr_master, required_master),
    ("Temporal pairs", temporal_pairs, required_temporal),
    ("Genomic timepoints", genomic_timepoints, required_timepoints),
]:

    missing = [
        col
        for col in required_columns
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            f"{label} is missing required columns: {missing}"
        )

    print(
        f"  PASS: {label} schema"
    )


# ================================================================
# 5. VALIDATED COMMON COHORT
# ================================================================

print("\n[4] Establishing verified common cohort...")

cxr_conditions = set(
    cxr_master["condition_id"]
    .dropna()
    .astype(str)
)

genomic_conditions = set(
    genomic_timepoints["condition_id"]
    .dropna()
    .astype(str)
)

common_conditions = (
    cxr_conditions &
    genomic_conditions
)

print(
    f"  CXR master conditions     : "
    f"{len(cxr_conditions):,}"
)

print(
    f"  Genomic-timepoint cases   : "
    f"{len(genomic_conditions):,}"
)

print(
    f"  Common conditions         : "
    f"{len(common_conditions):,}"
)

if len(common_conditions) != 3081:
    raise ValueError(
        "The validated common cohort is not 3,081 conditions. "
        "Do not continue until the input artifacts are reconciled."
    )

print(
    "  PASS: verified 3,081-condition common cohort"
)


# ================================================================
# 6. BUILD CXR-LEVEL MASTER
# ================================================================

print("\n[5] Building CXR-level prediction table...")

cxr_work = cxr_master.copy()

cxr_work["condition_id"] = (
    cxr_work["condition_id"]
    .astype(str)
)

cxr_work["patient_id"] = (
    cxr_work["patient_id"]
    .astype(str)
)

cxr_work["series_instance_content_url"] = (
    cxr_work["series_instance_content_url"]
    .astype(str)
)

cxr_work = cxr_work[
    cxr_work["condition_id"].isin(
        common_conditions
    )
].copy()

print(
    f"  Paired CXR records: {len(cxr_work):,}"
)

print(
    f"  Paired conditions  : "
    f"{cxr_work['condition_id'].nunique():,}"
)


# ================================================================
# 7. PARSE CXR IMAGING DATE
# ================================================================

print("\n[6] Parsing CXR imaging dates...")

cxr_work["imaging_date_raw"] = (
    cxr_work["imaging_date"]
    .astype(str)
    .str.strip()
)

cxr_work["imaging_date_parsed"] = pd.to_datetime(
    cxr_work["imaging_date_raw"],
    errors="coerce"
)

print(
    f"  Valid CXR dates : "
    f"{cxr_work['imaging_date_parsed'].notna().sum():,}"
)

print(
    f"  Missing CXR dates: "
    f"{cxr_work['imaging_date_parsed'].isna().sum():,}"
)


# ================================================================
# 8. BUILD UNIQUE GENOMIC TIMEPOINT TABLE
# ================================================================

print("\n[7] Preparing genomic timepoints...")

gt = genomic_timepoints.copy()

gt["condition_id"] = (
    gt["condition_id"]
    .astype(str)
)

gt["specimen_id"] = (
    gt["specimen_id"]
    .astype(str)
)

gt = gt[
    gt["condition_id"].isin(
        common_conditions
    )
].copy()

# ------------------------------------------------
# Locate the validated numeric date
# ------------------------------------------------

gt["specimen_collection_date_num"] = pd.to_numeric(
    gt["specimen_collection_date_num"],
    errors="coerce"
)

print(
    f"  Genomic timepoints in common cohort: "
    f"{len(gt):,}"
)

print(
    f"  Valid numeric genomic dates: "
    f"{gt['specimen_collection_date_num'].notna().sum():,}"
)

print(
    f"  Missing numeric genomic dates: "
    f"{gt['specimen_collection_date_num'].isna().sum():,}"
)


# ================================================================
# 9. CONVERT NUMERIC GENOMIC DATE
# ================================================================

# The existing validated artifact uses
# specimen_collection_date_num.
#
# We do not infer a new date encoding.
# We use the existing validated temporal artifact.

# First inspect range.

valid_genomic_dates = gt[
    gt["specimen_collection_date_num"].notna()
]["specimen_collection_date_num"]

if len(valid_genomic_dates) > 0:

    print(
        "\n  Genomic numeric date range:"
    )

    print(
        f"    minimum: "
        f"{valid_genomic_dates.min()}"
    )

    print(
        f"    maximum: "
        f"{valid_genomic_dates.max()}"
    )


# ================================================================
# 10. USE EXISTING TEMPORAL PAIRS AS PRIMARY TEMPORAL EVIDENCE
# ================================================================

print("\n[8] Validating existing temporal-pair artifact...")

tp = temporal_pairs.copy()

tp["condition_id"] = (
    tp["condition_id"]
    .astype(str)
)

tp = tp[
    tp["condition_id"].isin(
        common_conditions
    )
].copy()

print(
    f"  Temporal pair rows in common cohort: "
    f"{len(tp):,}"
)

if len(tp) != 4146:
    raise ValueError(
        "Expected 4,146 validated temporal pair rows, "
        f"but found {len(tp):,}."
    )

print(
    "  PASS: 4,146 validated temporal pairs"
)


# ================================================================
# 11. DISCOVER TEMPORAL COLUMNS
# ================================================================

print("\n[9] Inspecting temporal artifact columns...")

print(
    "\n  Temporal-pair columns:"
)

for col in tp.columns:
    print(
        f"    - {col}"
    )


# ================================================================
# 12. IDENTIFY EXISTING TEMPORAL DIRECTION FIELD
# ================================================================

possible_direction_columns = [
    "temporal_direction",
    "direction",
    "temporal_relationship",
]

direction_column = None

for col in possible_direction_columns:

    if col in tp.columns:
        direction_column = col
        break

if direction_column is None:

    raise ValueError(
        "No validated temporal-direction column was found "
        "in the temporal pair artifact."
    )

print(
    f"\n  Using temporal direction column: "
    f"{direction_column}"
)


# ================================================================
# 13. TEMPORAL DIRECTION DISTRIBUTION
# ================================================================

print("\n[10] Temporal direction distribution...")

tp["temporal_direction_clean"] = (
    tp[direction_column]
    .fillna("<MISSING>")
    .astype(str)
    .str.strip()
)

direction_summary = (
    tp["temporal_direction_clean"]
    .value_counts()
    .rename_axis("temporal_direction")
    .reset_index(name="pair_count")
)

direction_summary["percentage"] = (
    direction_summary["pair_count"]
    / len(tp)
    * 100
).round(4)

print(
    direction_summary.to_string(
        index=False
    )
)


# ================================================================
# 14. VERIFY EXPECTED TEMPORAL COUNTS
# ================================================================

expected_direction_counts = {
    "GENOMICS_BEFORE_CXR": 1801,
    "GENOMICS_AFTER_CXR": 1434,
    "SAME_DAY": 911,
}

observed_direction_counts = (
    tp["temporal_direction_clean"]
    .value_counts()
    .to_dict()
)

direction_validation_rows = []

for direction, expected in expected_direction_counts.items():

    observed = int(
        observed_direction_counts.get(
            direction,
            0
        )
    )

    direction_validation_rows.append(
        {
            "temporal_direction": direction,
            "expected_pairs": expected,
            "observed_pairs": observed,
            "status": (
                "PASS"
                if observed == expected
                else "FAIL"
            ),
        }
    )

direction_validation_df = pd.DataFrame(
    direction_validation_rows
)

print(
    "\n  Direction validation:"
)

print(
    direction_validation_df.to_string(
        index=False
    )
)


# ================================================================
# 15. CXR-LEVEL TEMPORAL COVERAGE
# ================================================================

print(
    "\n[11] Determining CXR-level temporal availability..."
)

if "series_instance_content_url" in tp.columns:

    cxr_key = "series_instance_content_url"

elif "imagingstudy_id" in tp.columns:

    cxr_key = "imagingstudy_id"

else:

    # Use condition_id only as fallback for a descriptive
    # condition-level audit.
    cxr_key = "condition_id"

print(
    f"  Temporal CXR key used: {cxr_key}"
)


# ================================================================
# 16. TEMPORAL ELIGIBILITY RULE DEFINITIONS
# ================================================================

print(
    "\n[12] Defining candidate prediction-time rules..."
)

#
# RULE A
# -------
# Strict prior availability:
#
# Genomic specimen must be collected BEFORE CXR.
#
# This is the strictest interpretation of:
# "genomic information was already available before the image."
#
#
# RULE B
# -------
# Prior + same-day:
#
# Genomic specimen collected before or on same calendar day.
#
# This is less strict and requires a scientific assumption
# that same-day information is legitimately available for
# the prediction task.
#
#
# RULE C
# -------
# Any temporally associated genomic specimen:
#
# Includes before, same-day, and after.
#
# This is NOT considered prediction-time eligible for a
# baseline predictive model because after-CXR genomics would
# violate temporal ordering.
#
# It is retained ONLY as an observational comparison.
#

candidate_rules = [
    {
        "rule_id": "TEMP-A",
        "rule_name": "STRICT_GENOMICS_BEFORE_CXR",
        "allowed_directions": [
            "GENOMICS_BEFORE_CXR"
        ],
        "scientific_interpretation": (
            "Genomic information must exist before the CXR."
        ),
        "prediction_time_status": "CANDIDATE",
    },
    {
        "rule_id": "TEMP-B",
        "rule_name": "GENOMICS_BEFORE_OR_SAME_DAY_CXR",
        "allowed_directions": [
            "GENOMICS_BEFORE_CXR",
            "SAME_DAY"
        ],
        "scientific_interpretation": (
            "Genomic specimen is available before or on "
            "the CXR date."
        ),
        "prediction_time_status": "CANDIDATE_REQUIRES_SAME_DAY_ASSUMPTION",
    },
    {
        "rule_id": "TEMP-C",
        "rule_name": "ANY_TEMPORAL_DIRECTION",
        "allowed_directions": [
            "GENOMICS_BEFORE_CXR",
            "SAME_DAY",
            "GENOMICS_AFTER_CXR"
        ],
        "scientific_interpretation": (
            "All temporal relationships retained for "
            "observational analysis."
        ),
        "prediction_time_status": "NOT_VALID_FOR_PROSPECTIVE_PREDICTION",
    },
]


# ================================================================
# 17. CONDITION-LEVEL TEMPORAL DIRECTION SET
# ================================================================

print(
    "\n[13] Creating condition-level temporal profiles..."
)

condition_temporal_profile = (
    tp.groupby("condition_id")[
        "temporal_direction_clean"
    ]
    .agg(
        directions=lambda s: sorted(
            set(s)
        ),
        n_temporal_pairs="size",
    )
    .reset_index()
)

condition_temporal_profile[
    "n_directions"
] = (
    condition_temporal_profile[
        "directions"
    ]
    .apply(len)
)

def classify_direction_set(values):

    values = set(values)

    if values == {"GENOMICS_BEFORE_CXR"}:
        return "BEFORE_ONLY"

    if values == {"GENOMICS_AFTER_CXR"}:
        return "AFTER_ONLY"

    if values == {"SAME_DAY"}:
        return "SAME_DAY_ONLY"

    if values == {
        "GENOMICS_BEFORE_CXR",
        "GENOMICS_AFTER_CXR"
    }:
        return "BEFORE_PLUS_AFTER"

    if values == {
        "GENOMICS_BEFORE_CXR",
        "SAME_DAY"
    }:
        return "BEFORE_PLUS_SAME_DAY"

    if values == {
        "SAME_DAY",
        "GENOMICS_AFTER_CXR"
    }:
        return "SAME_DAY_PLUS_AFTER"

    if values == {
        "GENOMICS_BEFORE_CXR",
        "SAME_DAY",
        "GENOMICS_AFTER_CXR"
    }:
        return "BEFORE_SAME_DAY_AFTER"

    return "OTHER"


condition_temporal_profile[
    "temporal_profile"
] = (
    condition_temporal_profile[
        "directions"
    ]
    .apply(classify_direction_set)
)

print(
    "\n  Condition temporal profiles:"
)

print(
    condition_temporal_profile[
        "temporal_profile"
    ]
    .value_counts()
    .to_string()
)


# ================================================================
# 18. CANDIDATE RULE COUNTS — PAIR LEVEL
# ================================================================

print(
    "\n[14] Evaluating candidate temporal rules..."
)

rule_rows = []

for rule in candidate_rules:

    allowed = set(
        rule["allowed_directions"]
    )

    mask = (
        tp["temporal_direction_clean"]
        .isin(allowed)
    )

    selected_pairs = tp[
        mask
    ].copy()

    unique_conditions = (
        selected_pairs[
            "condition_id"
        ]
        .nunique()
    )

    if cxr_key in selected_pairs.columns:
        unique_cxr = (
            selected_pairs[cxr_key]
            .nunique()
        )
    else:
        unique_cxr = np.nan

    rule_rows.append(
        {
            "rule_id": rule["rule_id"],
            "rule_name": rule["rule_name"],
            "allowed_directions": "|".join(
                rule["allowed_directions"]
            ),
            "pair_count": len(selected_pairs),
            "unique_conditions": unique_conditions,
            "unique_cxr_keys": unique_cxr,
            "percentage_of_all_pairs": round(
                len(selected_pairs)
                / len(tp)
                * 100,
                4
            ),
            "prediction_time_status": (
                rule["prediction_time_status"]
            ),
            "scientific_interpretation": (
                rule["scientific_interpretation"]
            ),
        }
    )

rule_summary_df = pd.DataFrame(
    rule_rows
)

print(
    rule_summary_df.to_string(
        index=False
    )
)


# ================================================================
# 19. CONDITION-LEVEL CANDIDATE RULE COUNTS
# ================================================================

print(
    "\n[15] Evaluating candidate rules at condition level..."
)

condition_rule_rows = []

for rule in candidate_rules:

    allowed = set(
        rule["allowed_directions"]
    )

    selected = tp[
        tp["temporal_direction_clean"]
        .isin(allowed)
    ]

    conditions = set(
        selected["condition_id"]
        .astype(str)
    )

    condition_rule_rows.append(
        {
            "rule_id": rule["rule_id"],
            "rule_name": rule["rule_name"],
            "eligible_conditions": len(
                conditions
            ),
            "percentage_of_3081_conditions": round(
                len(conditions)
                / 3081
                * 100,
                4
            ),
        }
    )

condition_rule_summary_df = pd.DataFrame(
    condition_rule_rows
)

print(
    condition_rule_summary_df.to_string(
        index=False
    )
)


# ================================================================
# 20. STRICT BEFORE-CXR CONDITION SET
# ================================================================

print(
    "\n[16] Constructing strict-before-CXR candidate cohort..."
)

strict_before_tp = tp[
    tp["temporal_direction_clean"]
    == "GENOMICS_BEFORE_CXR"
].copy()

strict_before_conditions = set(
    strict_before_tp[
        "condition_id"
    ]
    .astype(str)
)

print(
    f"  Strict-before pairs     : "
    f"{len(strict_before_tp):,}"
)

print(
    f"  Strict-before conditions: "
    f"{len(strict_before_conditions):,}"
)


# ================================================================
# 21. SAME-DAY CANDIDATE COHORT
# ================================================================

same_day_tp = tp[
    tp["temporal_direction_clean"]
    == "SAME_DAY"
].copy()

same_day_conditions = set(
    same_day_tp[
        "condition_id"
    ]
    .astype(str)
)

print(
    "\n  Same-day pairs:"
    f" {len(same_day_tp):,}"
)

print(
    "  Same-day conditions:"
    f" {len(same_day_conditions):,}"
)


# ================================================================
# 22. AFTER-CXR COHORT
# ================================================================

after_tp = tp[
    tp["temporal_direction_clean"]
    == "GENOMICS_AFTER_CXR"
].copy()

after_conditions = set(
    after_tp[
        "condition_id"
    ]
    .astype(str)
)

print(
    "\n  After-CXR pairs:"
    f" {len(after_tp):,}"
)

print(
    "  After-CXR conditions:"
    f" {len(after_conditions):,}"
)


# ================================================================
# 23. MULTIPLE GENOMIC TIMEPOINT CONDITIONS
# ================================================================

print(
    "\n[17] Auditing multiple genomic timepoint conditions..."
)

genomic_timepoint_counts = (
    gt.groupby("condition_id")
    .size()
    .rename("genomic_timepoint_count")
    .reset_index()
)

multiple_genomic = genomic_timepoint_counts[
    genomic_timepoint_counts[
        "genomic_timepoint_count"
    ] > 1
].copy()

print(
    f"  Conditions with >1 genomic timepoint: "
    f"{len(multiple_genomic):,}"
)

print(
    "\n  Distribution:"
)

print(
    multiple_genomic[
        "genomic_timepoint_count"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)


# ================================================================
# 24. TEMPORAL MULTIPLICITY
# ================================================================

print(
    "\n[18] Assessing multiple temporal relationships..."
)

multiple_direction_conditions = (
    condition_temporal_profile[
        condition_temporal_profile[
            "n_directions"
        ] > 1
    ]
    .copy()
)

print(
    f"  Conditions with multiple temporal directions: "
    f"{len(multiple_direction_conditions):,}"
)

print(
    multiple_direction_conditions[
        "temporal_profile"
    ]
    .value_counts()
    .to_string()
)


# ================================================================
# 25. SAME-DAY INTERPRETATION AUDIT
# ================================================================

print(
    "\n[19] Same-day temporal assumption audit..."
)

same_day_audit = pd.DataFrame(
    [
        {
            "question": (
                "Does same calendar date prove genomic information "
                "was available before the CXR?"
            ),
            "answer": "NO",
            "scientific_reason": (
                "A calendar date does not establish the within-day "
                "ordering of specimen collection and imaging."
            ),
        },
        {
            "question": (
                "Can same-day cases be included automatically "
                "in a prospective prediction cohort?"
            ),
            "answer": "NO",
            "scientific_reason": (
                "The exact acquisition times or a justified "
                "same-day availability assumption are required."
            ),
        },
        {
            "question": (
                "Should same-day cases be retained for sensitivity "
                "analysis?"
            ),
            "answer": "YES",
            "scientific_reason": (
                "They can be evaluated separately without silently "
                "assuming prediction-time availability."
            ),
        },
    ]
)


# ================================================================
# 26. AFTER-CXR INTERPRETATION AUDIT
# ================================================================

after_audit = pd.DataFrame(
    [
        {
            "question": (
                "Can genomic specimens collected after the CXR "
                "be used as baseline genomic predictors?"
            ),
            "answer": "NO",
            "scientific_reason": (
                "They were not available at the CXR prediction time."
            ),
        },
        {
            "question": (
                "Can after-CXR genomic relationships be retained "
                "for descriptive/longitudinal analysis?"
            ),
            "answer": "YES",
            "scientific_reason": (
                "They are valid temporal observations but answer "
                "a different research-design question."
            ),
        },
    ]
)


# ================================================================
# 27. BEFORE-CXR INTERPRETATION AUDIT
# ================================================================

before_audit = pd.DataFrame(
    [
        {
            "question": (
                "Does genomic collection before CXR satisfy "
                "temporal availability ordering?"
            ),
            "answer": "YES",
            "scientific_reason": (
                "The specimen collection date precedes the "
                "CXR date in the validated temporal artifact."
            ),
        },
        {
            "question": (
                "Does temporal precedence alone prove that the "
                "genomic feature was clinically available to the "
                "decision-maker?"
            ),
            "answer": "NOT NECESSARILY",
            "scientific_reason": (
                "Collection precedes imaging, but laboratory "
                "processing/reporting availability is not necessarily "
                "represented by specimen collection date."
            ),
        },
    ]
)


# ================================================================
# 28. IMPORTANT DISTINCTION: SPECIMEN COLLECTION VS RESULT AVAILABILITY
# ================================================================

print(
    "\n[20] Recording specimen-availability limitation..."
)

availability_limitation = pd.DataFrame(
    [
        {
            "issue": "Specimen collection date",
            "known": True,
            "meaning": (
                "Date used in the TB Portals temporal artifact "
                "for specimen timing."
            ),
        },
        {
            "issue": "Laboratory sequencing completion date",
            "known": False,
            "meaning": (
                "Not established by this audit as the date when "
                "genomic results became available."
            ),
        },
        {
            "issue": "Clinical decision/report availability",
            "known": False,
            "meaning": (
                "Not established by this audit."
            ),
        },
        {
            "issue": "Exact within-day CXR/specimen ordering",
            "known": False,
            "meaning": (
                "Not established by date-only temporal fields."
            ),
        },
    ]
)


# ================================================================
# 29. CONDITION-LEVEL ELIGIBILITY FLAGS
# ================================================================

print(
    "\n[21] Creating condition-level temporal eligibility flags..."
)

condition_flags = pd.DataFrame(
    {
        "condition_id": sorted(
            common_conditions
        )
    }
)

condition_flags["has_genomics_before_cxr"] = (
    condition_flags["condition_id"]
    .isin(strict_before_conditions)
)

condition_flags["has_genomics_same_day"] = (
    condition_flags["condition_id"]
    .isin(same_day_conditions)
)

condition_flags["has_genomics_after_cxr"] = (
    condition_flags["condition_id"]
    .isin(after_conditions)
)

condition_flags[
    "candidate_strict_before_eligible"
] = (
    condition_flags[
        "has_genomics_before_cxr"
    ]
)

condition_flags[
    "candidate_before_or_same_day_eligible"
] = (
    condition_flags[
        "has_genomics_before_cxr"
    ]
    |
    condition_flags[
        "has_genomics_same_day"
    ]
)

condition_flags[
    "has_any_temporal_relationship"
] = (
    condition_flags[
        [
            "has_genomics_before_cxr",
            "has_genomics_same_day",
            "has_genomics_after_cxr",
        ]
    ]
    .any(axis=1)
)


# ================================================================
# 30. TEMPORAL PROFILE MERGE
# ================================================================

condition_flags = condition_flags.merge(
    condition_temporal_profile[
        [
            "condition_id",
            "directions",
            "n_temporal_pairs",
            "n_directions",
            "temporal_profile",
        ]
    ],
    on="condition_id",
    how="left",
)


# ================================================================
# 31. MULTIPLE GENOMIC TIMEPOINT MERGE
# ================================================================

condition_flags = condition_flags.merge(
    genomic_timepoint_counts,
    on="condition_id",
    how="left",
)

condition_flags[
    "genomic_timepoint_count"
] = condition_flags[
    "genomic_timepoint_count"
].fillna(0).astype(int)

condition_flags[
    "multiple_genomic_timepoints"
] = (
    condition_flags[
        "genomic_timepoint_count"
    ] > 1
)


# ================================================================
# 32. TARGET MERGE
# ================================================================

# Read the Step 3B.6.58 condition-level target artifact.

target_file = (
    STEP_58_DIR
    / "06_Condition_Level_Target_Audit.csv"
)

if not target_file.exists():
    raise FileNotFoundError(
        "Step 3B.6.58 target artifact not found:\n"
        f"{target_file}"
    )

target_df = pd.read_csv(
    target_file,
    low_memory=False
)

required_target_columns = [
    "condition_id",
    "DR_DS_target",
    "target_stability",
]

missing_target = [
    c
    for c in required_target_columns
    if c not in target_df.columns
]

if missing_target:
    raise ValueError(
        "Step 3B.6.58 target artifact missing columns: "
        f"{missing_target}"
    )

target_df["condition_id"] = (
    target_df["condition_id"]
    .astype(str)
)

target_df = target_df[
    [
        "condition_id",
        "DR_DS_target",
        "target_stability",
    ]
].copy()

condition_flags = condition_flags.merge(
    target_df,
    on="condition_id",
    how="left",
)


# ================================================================
# 33. VERIFY TARGET COMPLETENESS
# ================================================================

target_missing = condition_flags[
    condition_flags["DR_DS_target"].isna()
].copy()

print(
    "\n[22] Target completeness after temporal merge:"
)

print(
    f"  Conditions with missing DR/DS target: "
    f"{len(target_missing):,}"
)

if len(target_missing) != 0:
    raise ValueError(
        "Some common conditions lost their validated DR/DS target."
    )

print(
    "  PASS: all 3,081 conditions retain stable target."
)


# ================================================================
# 34. PATIENT INDEPENDENCE AUDIT
# ================================================================

print(
    "\n[23] Patient-level independence audit..."
)

patient_condition = (
    cxr_work[
        [
            "patient_id",
            "condition_id",
        ]
    ]
    .drop_duplicates()
)

patient_condition_counts = (
    patient_condition
    .groupby("patient_id")
    .size()
    .rename("condition_count")
    .reset_index()
)

patients_one_condition = int(
    (
        patient_condition_counts[
            "condition_count"
        ] == 1
    ).sum()
)

patients_multiple_conditions = int(
    (
        patient_condition_counts[
            "condition_count"
        ] > 1
    ).sum()
)

print(
    f"  Patients with one condition : "
    f"{patients_one_condition:,}"
)

print(
    f"  Patients with >1 conditions : "
    f"{patients_multiple_conditions:,}"
)


# ================================================================
# 35. CONDITION-LEVEL TEMPORAL COHORT TABLE
# ================================================================

condition_flags.to_csv(
    OUTPUT_DIR
    / "01_Condition_Level_Temporal_Eligibility.csv",
    index=False
)


# ================================================================
# 36. SAVE TEMPORAL DIRECTION SUMMARY
# ================================================================

direction_summary.to_csv(
    OUTPUT_DIR
    / "02_Temporal_Direction_Summary.csv",
    index=False
)


# ================================================================
# 37. SAVE RULE SUMMARY
# ================================================================

rule_summary_df.to_csv(
    OUTPUT_DIR
    / "03_Candidate_Temporal_Rules_Pair_Level.csv",
    index=False
)

condition_rule_summary_df.to_csv(
    OUTPUT_DIR
    / "04_Candidate_Temporal_Rules_Condition_Level.csv",
    index=False
)


# ================================================================
# 38. SAVE CONDITION TEMPORAL PROFILES
# ================================================================

condition_temporal_profile.to_csv(
    OUTPUT_DIR
    / "05_Condition_Temporal_Profiles.csv",
    index=False
)


# ================================================================
# 39. SAVE MULTIPLE GENOMIC TIMEPOINT AUDIT
# ================================================================

multiple_genomic.to_csv(
    OUTPUT_DIR
    / "06_Multiple_Genomic_Timepoint_Conditions.csv",
    index=False
)


# ================================================================
# 40. SAVE SAME-DAY AUDIT
# ================================================================

same_day_audit.to_csv(
    OUTPUT_DIR
    / "07_Same_Day_Temporal_Assumption_Audit.csv",
    index=False
)


# ================================================================
# 41. SAVE BEFORE AUDIT
# ================================================================

before_audit.to_csv(
    OUTPUT_DIR
    / "08_Before_CXR_Temporal_Audit.csv",
    index=False
)


# ================================================================
# 42. SAVE AFTER AUDIT
# ================================================================

after_audit.to_csv(
    OUTPUT_DIR
    / "09_After_CXR_Temporal_Audit.csv",
    index=False
)


# ================================================================
# 43. SAVE AVAILABILITY LIMITATION
# ================================================================

availability_limitation.to_csv(
    OUTPUT_DIR
    / "10_Genomic_Availability_Limitation_Audit.csv",
    index=False
)


# ================================================================
# 44. SAVE TARGET DATA
# ================================================================

target_definition_path = (
    STEP_58_DIR
    / "03_Target_Definition_Audit.csv"
)

if target_definition_path.exists():

    target_definition_df = pd.read_csv(
        target_definition_path,
        low_memory=False
    )

    target_definition_df.to_csv(
        OUTPUT_DIR
        / "11_Target_Definition_Carryforward.csv",
        index=False
    )


# ================================================================
# 45. TEMPORAL DECISION GATES
# ================================================================

print(
    "\n[24] Creating temporal decision gates..."
)

strict_before_conditions_count = len(
    strict_before_conditions
)

before_same_day_conditions = (
    condition_flags[
        "candidate_before_or_same_day_eligible"
    ].sum()
)

after_only_conditions = (
    (
        condition_flags[
            "has_genomics_after_cxr"
        ]
        &
        ~condition_flags[
            "has_genomics_before_cxr"
        ]
        &
        ~condition_flags[
            "has_genomics_same_day"
        ]
    )
    .sum()
)

temporal_decision_rows = [

    {
        "decision_id": "TEMP-59-01",
        "decision_area": "Strict temporal precedence",
        "criterion": "Genomics collected before CXR",
        "observed_conditions": strict_before_conditions_count,
        "status": "OBSERVED_NOT_LOCKED",
        "scientific_interpretation": (
            "These conditions satisfy calendar-date precedence "
            "for specimen collection before imaging."
        ),
    },

    {
        "decision_id": "TEMP-59-02",
        "decision_area": "Same-day cases",
        "criterion": "Genomics collected on same date as CXR",
        "observed_conditions": len(same_day_conditions),
        "status": "SEPARATE_SENSITIVITY_CANDIDATE",
        "scientific_interpretation": (
            "Same-day status does not prove within-day availability."
        ),
    },

    {
        "decision_id": "TEMP-59-03",
        "decision_area": "After-CXR genomics",
        "criterion": "Genomics collected after CXR",
        "observed_conditions": len(after_conditions),
        "status": "NOT_BASELINE_ELIGIBLE",
        "scientific_interpretation": (
            "After-CXR specimens cannot be treated as available "
            "at the CXR prediction time."
        ),
    },

    {
        "decision_id": "TEMP-59-04",
        "decision_area": "Multiple genomic timepoints",
        "criterion": "More than one genomic timepoint",
        "observed_conditions": len(multiple_genomic),
        "status": "REPRESENTATIVE_SPECIMEN_RULE_REQUIRED",
        "scientific_interpretation": (
            "A deterministic specimen-selection rule must be "
            "defined before model construction."
        ),
    },

    {
        "decision_id": "TEMP-59-05",
        "decision_area": "Temporal direction",
        "criterion": "All pair relationships accounted for",
        "observed_conditions": len(tp),
        "status": (
            "PASS"
            if len(tp) == 4146
            else "FAIL"
        ),
        "scientific_interpretation": (
            "The validated temporal-pair artifact contains "
            "4,146 CXR-genomic temporal relationships."
        ),
    },

    {
        "decision_id": "TEMP-59-06",
        "decision_area": "Genomic result availability",
        "criterion": "Sequencing/result availability known",
        "observed_conditions": "NOT_ESTABLISHED",
        "status": "LIMITATION",
        "scientific_interpretation": (
            "Specimen collection date is available, but this audit "
            "does not establish the exact date when sequencing "
            "results became available."
        ),
    },

    {
        "decision_id": "TEMP-59-07",
        "decision_area": "Final temporal rule",
        "criterion": "Prediction-time eligibility locked",
        "observed_conditions": "NOT_YET",
        "status": "PENDING",
        "scientific_interpretation": (
            "Final temporal eligibility requires the research "
            "design decision."
        ),
    },
]

temporal_decision_df = pd.DataFrame(
    temporal_decision_rows
)

temporal_decision_df.to_csv(
    OUTPUT_DIR
    / "12_Temporal_Decision_Gates.csv",
    index=False
)


# ================================================================
# 46. PATIENT INDEPENDENCE OUTPUT
# ================================================================

patient_condition_counts.to_csv(
    OUTPUT_DIR
    / "13_Patient_Condition_Independence.csv",
    index=False
)


# ================================================================
# 47. VALIDATION
# ================================================================

print(
    "\n[25] Running final validation..."
)

validation_rows = []


def add_validation(
    validation_id,
    test,
    observed,
    expected,
    status
):
    validation_rows.append(
        {
            "validation_id": validation_id,
            "test": test,
            "observed": observed,
            "expected": expected,
            "status": status,
        }
    )


add_validation(
    "VAL-59-01",
    "Common condition count",
    len(common_conditions),
    3081,
    "PASS"
    if len(common_conditions) == 3081
    else "FAIL",
)

add_validation(
    "VAL-59-02",
    "Temporal pair count",
    len(tp),
    4146,
    "PASS"
    if len(tp) == 4146
    else "FAIL",
)

add_validation(
    "VAL-59-03",
    "Before-CXR pair count",
    int(
        (
            tp["temporal_direction_clean"]
            == "GENOMICS_BEFORE_CXR"
        ).sum()
    ),
    1801,
    "PASS"
    if (
        (
            tp["temporal_direction_clean"]
            == "GENOMICS_BEFORE_CXR"
        ).sum()
        == 1801
    )
    else "FAIL",
)

add_validation(
    "VAL-59-04",
    "Same-day pair count",
    int(
        (
            tp["temporal_direction_clean"]
            == "SAME_DAY"
        ).sum()
    ),
    911,
    "PASS"
    if (
        (
            tp["temporal_direction_clean"]
            == "SAME_DAY"
        ).sum()
        == 911
    )
    else "FAIL",
)

add_validation(
    "VAL-59-05",
    "After-CXR pair count",
    int(
        (
            tp["temporal_direction_clean"]
            == "GENOMICS_AFTER_CXR"
        ).sum()
    ),
    1434,
    "PASS"
    if (
        (
            tp["temporal_direction_clean"]
            == "GENOMICS_AFTER_CXR"
        ).sum()
        == 1434
    )
    else "FAIL",
)

add_validation(
    "VAL-59-06",
    "Condition-level target completeness",
    int(
        condition_flags[
            "DR_DS_target"
        ].notna().sum()
    ),
    3081,
    "PASS"
    if (
        condition_flags[
            "DR_DS_target"
        ].notna().sum()
        == 3081
    )
    else "FAIL",
)

add_validation(
    "VAL-59-07",
    "Multiple genomic timepoint conditions",
    len(multiple_genomic),
    152,
    "PASS"
    if len(multiple_genomic) == 152
    else "FAIL",
)

add_validation(
    "VAL-59-08",
    "Artificial Normal labels",
    False,
    False,
    "PASS",
)

add_validation(
    "VAL-59-09",
    "Missing genomics treated as Normal",
    False,
    False,
    "PASS",
)

add_validation(
    "VAL-59-10",
    "Model training started",
    False,
    False,
    "PASS",
)

add_validation(
    "VAL-59-11",
    "Preprocessing started",
    False,
    False,
    "PASS",
)

add_validation(
    "VAL-59-12",
    "Segmentation started",
    False,
    False,
    "PASS",
)


validation_df = pd.DataFrame(
    validation_rows
)

validation_df.to_csv(
    OUTPUT_DIR
    / "14_Validation.csv",
    index=False
)


# ================================================================
# 48. FINAL SUMMARY JSON
# ================================================================

print(
    "\n[26] Creating final summary..."
)

summary = {
    "step": "3B.6.59",
    "title": "Temporal Prediction-Time Cohort Audit",

    "execution_timestamp_utc": (
        datetime.utcnow()
        .replace(microsecond=0)
        .isoformat()
        + "Z"
    ),

    "verified_input_counts": {
        "cxr_metadata_rows": int(len(cxr)),
        "genomic_metadata_rows": int(len(genomics)),
        "cxr_master_rows": int(len(cxr_master)),
        "temporal_pair_rows": int(len(tp)),
        "genomic_timepoints": int(
            len(genomic_timepoints)
        ),
        "common_conditions": int(
            len(common_conditions)
        ),
    },

    "temporal_direction_pairs": {
        "genomics_before_cxr": int(
            (
                tp["temporal_direction_clean"]
                == "GENOMICS_BEFORE_CXR"
            ).sum()
        ),
        "same_day": int(
            (
                tp["temporal_direction_clean"]
                == "SAME_DAY"
            ).sum()
        ),
        "genomics_after_cxr": int(
            (
                tp["temporal_direction_clean"]
                == "GENOMICS_AFTER_CXR"
            ).sum()
        ),
    },

    "candidate_condition_counts": {
        "strict_genomics_before_cxr": int(
            strict_before_conditions_count
        ),
        "genomics_before_or_same_day": int(
            before_same_day_conditions
        ),
        "after_only": int(
            after_only_conditions
        ),
    },

    "multiple_genomic_timepoints": {
        "conditions": int(
            len(multiple_genomic)
        ),
    },

    "temporal_profile_counts": {
        str(k): int(v)
        for k, v in
        condition_temporal_profile[
            "temporal_profile"
        ]
        .value_counts()
        .to_dict()
        .items()
    },

    "patient_independence": {
        "patients_one_condition": patients_one_condition,
        "patients_multiple_conditions": patients_multiple_conditions,
    },

    "scientific_interpretation": {
        "before_cxr": (
            "Calendar-date precedence established."
        ),
        "same_day": (
            "Not automatically equivalent to prediction-time "
            "availability."
        ),
        "after_cxr": (
            "Not eligible as baseline genomic predictor."
        ),
        "multiple_timepoints": (
            "Requires deterministic representative-specimen rule."
        ),
        "result_availability": (
            "Exact sequencing/result availability date "
            "not established by this audit."
        ),
    },

    "scientific_locks": {
        "final_temporal_rule": "PENDING",
        "representative_genomic_specimen_rule": "PENDING",
        "final_cohort": "PENDING",
        "research_question": "NOT_AUTOMATICALLY_SELECTED",
        "model_training": False,
        "preprocessing": False,
        "segmentation": False,
    },

    "status": (
        "TEMPORAL_STRUCTURE_VALIDATED; "
        "PREDICTION_TIME RULE NOT YET LOCKED"
    ),

    "next_step": (
        "Step 3B.6.60 — Compare Finalized DR/DS "
        "vs Treatment Outcome Designs"
    ),
}

with open(
    OUTPUT_DIR
    / "15_Step_3B_6_59_Summary.json",
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        summary,
        f,
        indent=4,
        ensure_ascii=False,
        default=str,
    )


# ================================================================
# 49. FINAL CONSOLE REPORT
# ================================================================

print("\n" + "=" * 80)
print("STEP 3B.6.59 COMPLETED")
print("=" * 80)

print("\nVERIFIED COHORT")
print("-" * 80)

print(
    f"Common paired conditions       : "
    f"{len(common_conditions):,}"
)

print(
    f"Temporal pair records          : "
    f"{len(tp):,}"
)

print(
    f"Unique genomic timepoints      : "
    f"{len(genomic_timepoints):,}"
)


print("\nTEMPORAL DIRECTION")
print("-" * 80)

print(
    f"Genomics BEFORE CXR            : "
    f"{expected_direction_counts['GENOMICS_BEFORE_CXR']:,}"
)

print(
    f"Same-day                       : "
    f"{expected_direction_counts['SAME_DAY']:,}"
)

print(
    f"Genomics AFTER CXR             : "
    f"{expected_direction_counts['GENOMICS_AFTER_CXR']:,}"
)


print("\nCONDITION-LEVEL TEMPORAL ELIGIBILITY")
print("-" * 80)

print(
    f"Has genomics before CXR        : "
    f"{strict_before_conditions_count:,}"
)

print(
    f"Has genomics before/same-day   : "
    f"{before_same_day_conditions:,}"
)

print(
    f"After-only conditions          : "
    f"{after_only_conditions:,}"
)

print(
    f"Multiple genomic timepoints    : "
    f"{len(multiple_genomic):,}"
)


print("\nTEMPORAL INTERPRETATION")
print("-" * 80)

print(
    "BEFORE CXR  -> calendar-date precedence satisfied"
)

print(
    "SAME DAY    -> cannot automatically prove availability"
)

print(
    "AFTER CXR   -> not baseline prediction-time eligible"
)

print(
    "MULTIPLE    -> representative specimen rule required"
)


print("\nIMPORTANT LIMITATION")
print("-" * 80)

print(
    "Specimen collection date is available."
)

print(
    "Exact sequencing/result availability date is NOT "
    "established by this audit."
)


print("\nSCIENTIFIC LOCKS")
print("-" * 80)

print(
    "Final temporal rule             : PENDING"
)

print(
    "Representative genomic specimen : PENDING"
)

print(
    "Final cohort                    : PENDING"
)

print(
    "Research question               : NOT AUTOMATICALLY SELECTED"
)

print(
    "Preprocessing                   : NOT STARTED"
)

print(
    "Segmentation                    : NOT STARTED"
)

print(
    "Model training                  : NOT STARTED"
)


print("\nOUTPUT DIRECTORY")
print("-" * 80)

print(
    OUTPUT_DIR
)


print("\nNEXT STEP")
print("-" * 80)

print(
    "3B.6.60 — Compare Finalized DR/DS "
    "vs Treatment Outcome Designs"
)

print("\nSTATUS")
print("-" * 80)

print(
    "TEMPORAL STRUCTURE VALIDATED"
)

print(
    "PREDICTION-TIME RULE NOT YET LOCKED"
)

print("=" * 80)
```

```python
# ================================================================
# STEP 3B.6.59R
# CORRECTED TEMPORAL PREDICTION-TIME COHORT AUDIT
# ================================================================
#
# PURPOSE
# -------
# Correct Step 3B.6.59 by using the already validated numerical
# temporal fields from Step 3B.6.34 instead of parsing the raw
# imaging_date string.
#
# This step:
#   1. Validates the existing temporal artifact.
#   2. Uses imaging_date_num as the CXR temporal reference.
#   3. Uses specimen_collection_date_num for genomic timing.
#   4. Verifies signed temporal differences.
#   5. Audits BEFORE / SAME_DAY / AFTER relationships.
#   6. Evaluates strict-before prediction-time eligibility.
#   7. Evaluates before-or-same-day eligibility separately.
#   8. Audits multiple genomic specimens.
#   9. Identifies candidate representative specimens.
#  10. Identifies representative-specimen ties.
#  11. Separates specimen collection timing from actual
#      sequencing-result availability.
#  12. Checks condition/patient independence.
#
# IMPORTANT SCIENTIFIC LOCKS
# --------------------------
# - No model training
# - No preprocessing
# - No segmentation
# - No automatic research-question selection
# - No artificial Normal labels
# - No missing-genomics-as-Normal
# - No cxr_outlier="normal" as clinical Normal
# - No resistance target used as predictor
#
# ================================================================


from pathlib import Path
import pandas as pd
import numpy as np
import json
from datetime import datetime


# ================================================================
# 1. CONFIGURATION
# ================================================================

CXR_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_CXRs_March_2025.csv"
)

GENOMICS_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_Genomics_March_2025.csv"
)

CXR_MASTER_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_24_Notebook_03"
    r"\TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

TEMPORAL_PAIRS_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
    r"\TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

GENOMIC_TIMEPOINT_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
    r"\TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

STEP_58_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_58_Target_Genomic_Feature_Scientific_Audit"
)

OUTPUT_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_59R_Corrected_Temporal_Prediction_Time_Audit"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# 2. HELPER FUNCTIONS
# ================================================================

def require_file(path, label):

    if not path.exists():

        raise FileNotFoundError(
            f"\nREQUIRED INPUT NOT FOUND\n"
            f"Label : {label}\n"
            f"Path  : {path}"
        )


def normalize_text(series):

    return (
        series
        .fillna("")
        .astype(str)
        .str.strip()
    )


def percent(part, total):

    if total == 0:
        return 0.0

    return round(
        (part / total) * 100,
        4
    )


def safe_unique(series):

    return sorted(
        set(
            str(x).strip()
            for x in series
            if pd.notna(x)
            and str(x).strip() != ""
        )
    )


# ================================================================
# 3. HEADER
# ================================================================

print("=" * 80)
print("STEP 3B.6.59R — CORRECTED TEMPORAL PREDICTION-TIME AUDIT")
print("=" * 80)


# ================================================================
# 4. INPUT FILE VALIDATION
# ================================================================

print("\n[1] Checking required files...")

required_files = {
    "CXR metadata": CXR_PATH,
    "Genomics metadata": GENOMICS_PATH,
    "CXR master": CXR_MASTER_PATH,
    "Temporal pairs": TEMPORAL_PAIRS_PATH,
    "Unique genomic timepoints": GENOMIC_TIMEPOINT_PATH,
    "Step 3B.6.58 output": STEP_58_DIR,
}

for label, path in required_files.items():

    require_file(
        path,
        label
    )

    print(
        f"  PASS: {label}"
    )

    print(
        f"        {path}"
    )


# ================================================================
# 5. LOAD DATA
# ================================================================

print("\n[2] Loading validated artifacts...")

cxr = pd.read_csv(
    CXR_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)

cxr_master = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

temporal_pairs = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)

genomic_timepoints = pd.read_csv(
    GENOMIC_TIMEPOINT_PATH,
    low_memory=False
)

print(
    f"  CXR metadata              : {len(cxr):,}"
)

print(
    f"  Genomics metadata         : {len(genomics):,}"
)

print(
    f"  CXR master                : {len(cxr_master):,}"
)

print(
    f"  Temporal pairs            : {len(temporal_pairs):,}"
)

print(
    f"  Unique genomic timepoints : {len(genomic_timepoints):,}"
)


# ================================================================
# 6. SCHEMA VALIDATION
# ================================================================

print("\n[3] Schema validation...")

required_cxr_columns = [
    "patient_id",
    "condition_id",
    "imaging_date",
    "series_instance_content_url",
]

required_master_columns = [
    "patient_id",
    "condition_id",
    "imaging_date",
    "series_instance_content_url",
]

required_temporal_columns = [
    "condition_id",
    "series_instance_content_url",
    "imaging_date_num",
    "specimen_id",
    "specimen_collection_date_num",
    "signed_temporal_delta_days",
    "absolute_temporal_distance_days",
    "temporal_direction",
    "absolute_distance_category",
]

required_timepoint_columns = [
    "condition_id",
    "specimen_id",
    "specimen_collection_date_num",
]

for label, df, required_columns in [

    (
        "CXR metadata",
        cxr,
        required_cxr_columns
    ),

    (
        "CXR master",
        cxr_master,
        required_master_columns
    ),

    (
        "Temporal pairs",
        temporal_pairs,
        required_temporal_columns
    ),

    (
        "Genomic timepoints",
        genomic_timepoints,
        required_timepoint_columns
    ),
]:

    missing = [
        column
        for column in required_columns
        if column not in df.columns
    ]

    if missing:

        raise ValueError(
            f"\n{label} is missing required columns:\n"
            f"{missing}"
        )

    print(
        f"  PASS: {label}"
    )


# ================================================================
# 7. PREPARE TEMPORAL ARTIFACT
# ================================================================

print(
    "\n[4] Preparing validated temporal artifact..."
)

tp = temporal_pairs.copy()

tp["condition_id"] = (
    tp["condition_id"]
    .astype(str)
    .str.strip()
)

tp["series_instance_content_url"] = (
    tp["series_instance_content_url"]
    .astype(str)
    .str.strip()
)

tp["specimen_id"] = (
    tp["specimen_id"]
    .astype(str)
    .str.strip()
)

numeric_temporal_columns = [
    "imaging_date_num",
    "specimen_collection_date_num",
    "signed_temporal_delta_days",
    "absolute_temporal_distance_days",
]

for column in numeric_temporal_columns:

    tp[column] = pd.to_numeric(
        tp[column],
        errors="coerce"
    )

tp["temporal_direction"] = normalize_text(
    tp["temporal_direction"]
)

tp["absolute_distance_category"] = normalize_text(
    tp["absolute_distance_category"]
)

print(
    f"  Temporal pair rows: {len(tp):,}"
)

if len(tp) != 4146:

    raise ValueError(
        "Expected 4,146 validated temporal pair records, "
        f"but found {len(tp):,}."
    )

print(
    "  PASS: 4,146 temporal pair records"
)


# ================================================================
# 8. VALIDATE TEMPORAL NUMERIC FIELDS
# ================================================================

print(
    "\n[5] Checking temporal numeric fields..."
)

temporal_numeric_audit_rows = []

for column in numeric_temporal_columns:

    valid_count = int(
        tp[column].notna().sum()
    )

    missing_count = int(
        tp[column].isna().sum()
    )

    temporal_numeric_audit_rows.append(
        {
            "field": column,
            "rows": len(tp),
            "valid": valid_count,
            "missing": missing_count,
            "missing_percent": percent(
                missing_count,
                len(tp)
            ),
        }
    )

temporal_numeric_audit = pd.DataFrame(
    temporal_numeric_audit_rows
)

print(
    temporal_numeric_audit.to_string(
        index=False
    )
)


# ================================================================
# 9. VALIDATE TEMPORAL DIRECTION AGAINST SIGNED DELTA
# ================================================================

print(
    "\n[6] Verifying temporal direction against signed delta..."
)


def expected_direction_from_delta(delta):

    if pd.isna(delta):
        return "MISSING"

    if delta < 0:
        return "GENOMICS_BEFORE_CXR"

    if delta == 0:
        return "SAME_DAY"

    if delta > 0:
        return "GENOMICS_AFTER_CXR"

    return "UNKNOWN"


tp["expected_temporal_direction"] = (
    tp["signed_temporal_delta_days"]
    .apply(
        expected_direction_from_delta
    )
)

tp["direction_consistency"] = np.where(
    tp["temporal_direction"]
    ==
    tp["expected_temporal_direction"],
    "CONSISTENT",
    "INCONSISTENT"
)

direction_consistency_counts = (
    tp["direction_consistency"]
    .value_counts()
)

print(
    direction_consistency_counts.to_string()
)

inconsistent_direction_count = int(
    direction_consistency_counts.get(
        "INCONSISTENT",
        0
    )
)

if inconsistent_direction_count != 0:

    raise ValueError(
        "Temporal direction does not consistently agree "
        "with signed temporal delta."
    )

print(
    "  PASS: temporal direction agrees with signed delta"
)


# ================================================================
# 10. VALIDATE ABSOLUTE DISTANCE
# ================================================================

print(
    "\n[7] Verifying absolute temporal distance..."
)

tp["calculated_absolute_distance"] = (
    tp["signed_temporal_delta_days"]
    .abs()
)

distance_difference = (
    tp["calculated_absolute_distance"]
    -
    tp["absolute_temporal_distance_days"]
).abs()

max_distance_difference = (
    distance_difference.max()
)

print(
    f"  Maximum distance discrepancy: "
    f"{max_distance_difference}"
)

if (
    pd.notna(max_distance_difference)
    and max_distance_difference > 1e-9
):

    raise ValueError(
        "Absolute temporal distance does not agree "
        "with signed temporal delta."
    )

print(
    "  PASS: absolute temporal distance is consistent"
)


# ================================================================
# 11. TEMPORAL DIRECTION SUMMARY
# ================================================================

print(
    "\n[8] Temporal direction distribution..."
)

direction_summary = (
    tp["temporal_direction"]
    .value_counts()
    .rename_axis(
        "temporal_direction"
    )
    .reset_index(
        name="pair_count"
    )
)

direction_summary["percentage"] = (
    direction_summary["pair_count"]
    .apply(
        lambda x:
        percent(
            x,
            len(tp)
        )
    )
)

print(
    direction_summary.to_string(
        index=False
    )
)


# ================================================================
# 12. VERIFY EXPECTED DIRECTION COUNTS
# ================================================================

expected_direction_counts = {
    "GENOMICS_BEFORE_CXR": 1801,
    "SAME_DAY": 911,
    "GENOMICS_AFTER_CXR": 1434,
}

direction_validation_rows = []

for direction, expected in (
    expected_direction_counts.items()
):

    observed = int(
        direction_summary.loc[
            direction_summary[
                "temporal_direction"
            ]
            == direction,
            "pair_count"
        ]
        .sum()
    )

    direction_validation_rows.append(
        {
            "temporal_direction": direction,
            "expected": expected,
            "observed": observed,
            "status": (
                "PASS"
                if observed == expected
                else "FAIL"
            ),
        }
    )

direction_validation = pd.DataFrame(
    direction_validation_rows
)

print(
    "\n  Direction validation:"
)

print(
    direction_validation.to_string(
        index=False
    )
)

if (
    direction_validation["status"]
    == "FAIL"
).any():

    raise ValueError(
        "Temporal direction counts do not match "
        "the previously validated artifact."
    )


# ================================================================
# 13. VALIDATE COMMON CONDITION COHORT
# ================================================================

print(
    "\n[9] Validating common condition cohort..."
)

common_conditions = set(
    tp["condition_id"]
    .dropna()
    .astype(str)
)

print(
    f"  Conditions represented: "
    f"{len(common_conditions):,}"
)

if len(common_conditions) != 3081:

    raise ValueError(
        "Temporal artifact does not contain "
        "the expected 3,081 conditions."
    )

print(
    "  PASS: 3,081 conditions"
)


# ================================================================
# 14. VALIDATE UNIQUE CXR URL COUNT
# ================================================================

unique_cxr_urls = (
    tp[
        "series_instance_content_url"
    ]
    .nunique()
)

print(
    f"  Unique CXR URLs: "
    f"{unique_cxr_urls:,}"
)

if unique_cxr_urls != 3859:

    raise ValueError(
        "Temporal artifact does not contain "
        "the expected 3,859 unique CXR URLs."
    )

print(
    "  PASS: 3,859 unique CXR URLs"
)


# ================================================================
# 15. CONDITION TEMPORAL PROFILES
# ================================================================

print(
    "\n[10] Constructing condition temporal profiles..."
)

condition_profiles = (
    tp.groupby(
        "condition_id"
    )[
        "temporal_direction"
    ]
    .agg(
        directions=lambda s:
        sorted(
            set(s)
        ),
        temporal_pair_count="size",
    )
    .reset_index()
)

condition_profiles[
    "number_of_temporal_directions"
] = (
    condition_profiles[
        "directions"
    ]
    .apply(len)
)


def classify_temporal_profile(values):

    values = set(values)

    if values == {
        "GENOMICS_BEFORE_CXR"
    }:
        return "BEFORE_ONLY"

    if values == {
        "SAME_DAY"
    }:
        return "SAME_DAY_ONLY"

    if values == {
        "GENOMICS_AFTER_CXR"
    }:
        return "AFTER_ONLY"

    if values == {
        "GENOMICS_BEFORE_CXR",
        "GENOMICS_AFTER_CXR",
    }:
        return "BEFORE_PLUS_AFTER"

    if values == {
        "GENOMICS_BEFORE_CXR",
        "SAME_DAY",
    }:
        return "BEFORE_PLUS_SAME_DAY"

    if values == {
        "SAME_DAY",
        "GENOMICS_AFTER_CXR",
    }:
        return "SAME_DAY_PLUS_AFTER"

    if values == {
        "GENOMICS_BEFORE_CXR",
        "SAME_DAY",
        "GENOMICS_AFTER_CXR",
    }:
        return "BEFORE_SAME_DAY_AFTER"

    return "OTHER"


condition_profiles[
    "temporal_profile"
] = (
    condition_profiles[
        "directions"
    ]
    .apply(
        classify_temporal_profile
    )
)

profile_summary = (
    condition_profiles[
        "temporal_profile"
    ]
    .value_counts()
    .rename_axis(
        "temporal_profile"
    )
    .reset_index(
        name="condition_count"
    )
)

profile_summary[
    "percentage"
] = (
    profile_summary[
        "condition_count"
    ]
    .apply(
        lambda x:
        percent(
            x,
            len(condition_profiles)
        )
    )
)

print(
    profile_summary.to_string(
        index=False
    )
)


# ================================================================
# 16. GENOMIC TIMEPOINT MULTIPLICITY
# ================================================================

print(
    "\n[11] Auditing genomic timepoint multiplicity..."
)

gt = genomic_timepoints.copy()

gt["condition_id"] = (
    gt["condition_id"]
    .astype(str)
    .str.strip()
)

gt["specimen_id"] = (
    gt["specimen_id"]
    .astype(str)
    .str.strip()
)

gt[
    "specimen_collection_date_num"
] = pd.to_numeric(
    gt[
        "specimen_collection_date_num"
    ],
    errors="coerce"
)

gt = gt[
    gt["condition_id"]
    .isin(
        common_conditions
    )
].copy()

timepoint_counts = (
    gt.groupby(
        "condition_id"
    )
    .size()
    .rename(
        "genomic_timepoint_count"
    )
    .reset_index()
)

multiple_timepoint_conditions = (
    timepoint_counts[
        timepoint_counts[
            "genomic_timepoint_count"
        ]
        > 1
    ]
    .copy()
)

print(
    f"  Total genomic timepoint rows: "
    f"{len(gt):,}"
)

print(
    f"  Conditions with >1 timepoint: "
    f"{len(multiple_timepoint_conditions):,}"
)

print(
    "\n  Timepoint multiplicity:"
)

print(
    timepoint_counts[
        "genomic_timepoint_count"
    ]
    .value_counts()
    .sort_index()
    .to_string()
)


# ================================================================
# 17. CXR-SPECIMEN TEMPORAL TABLE
# ================================================================

print(
    "\n[12] Building CXR-specimen temporal table..."
)

temporal_specimen = tp[
    [
        "condition_id",
        "series_instance_content_url",
        "imaging_date_num",
        "specimen_id",
        "specimen_collection_date_num",
        "signed_temporal_delta_days",
        "absolute_temporal_distance_days",
        "temporal_direction",
        "absolute_distance_category",
    ]
].copy()

print(
    f"  CXR-specimen temporal rows: "
    f"{len(temporal_specimen):,}"
)


# ================================================================
# 18. STRICT BEFORE-CXR SPECIMENS
# ================================================================

print(
    "\n[13] Selecting candidate closest BEFORE-CXR specimens..."
)

before_specimens = temporal_specimen[
    temporal_specimen[
        "temporal_direction"
    ]
    ==
    "GENOMICS_BEFORE_CXR"
].copy()

before_specimens[
    "rank_distance_before"
] = (
    before_specimens
    .groupby(
        [
            "condition_id",
            "series_instance_content_url",
        ]
    )[
        "absolute_temporal_distance_days"
    ]
    .rank(
        method="min",
        ascending=True
    )
)

nearest_before = before_specimens[
    before_specimens[
        "rank_distance_before"
    ]
    == 1
].copy()

print(
    f"  Before-CXR relationships: "
    f"{len(before_specimens):,}"
)

print(
    f"  Nearest-before candidate rows: "
    f"{len(nearest_before):,}"
)


# ================================================================
# 19. STRICT BEFORE-CXR TIE AUDIT
# ================================================================

print(
    "\n[14] Auditing nearest-before ties..."
)

before_tie_counts = (
    nearest_before
    .groupby(
        [
            "condition_id",
            "series_instance_content_url",
        ]
    )
    .agg(
        number_of_equally_near_before_specimens=(
            "specimen_id",
            "nunique"
        )
    )
    .reset_index()
)

before_ties = before_tie_counts[
    before_tie_counts[
        "number_of_equally_near_before_specimens"
    ]
    > 1
].copy()

strict_tie_group_count = len(
    before_ties
)

if strict_tie_group_count > 0:

    max_strict_tie = int(
        before_ties[
            "number_of_equally_near_before_specimens"
        ].max()
    )

else:

    max_strict_tie = 1

print(
    f"  Nearest-before tie groups: "
    f"{strict_tie_group_count:,}"
)

print(
    f"  Maximum tied specimens: "
    f"{max_strict_tie:,}"
)


# ================================================================
# 20. BEFORE-OR-SAME-DAY SPECIMENS
# ================================================================

print(
    "\n[15] Selecting candidate closest BEFORE/SAME-DAY specimens..."
)

before_or_same = temporal_specimen[
    temporal_specimen[
        "temporal_direction"
    ]
    .isin(
        [
            "GENOMICS_BEFORE_CXR",
            "SAME_DAY",
        ]
    )
].copy()

before_or_same[
    "rank_distance"
] = (
    before_or_same
    .groupby(
        [
            "condition_id",
            "series_instance_content_url",
        ]
    )[
        "absolute_temporal_distance_days"
    ]
    .rank(
        method="min",
        ascending=True
    )
)

nearest_before_or_same = (
    before_or_same[
        before_or_same[
            "rank_distance"
        ]
        == 1
    ]
    .copy()
)

print(
    f"  Before/same-day relationships: "
    f"{len(before_or_same):,}"
)

print(
    f"  Nearest candidate rows: "
    f"{len(nearest_before_or_same):,}"
)


# ================================================================
# 21. BEFORE/SAME-DAY TIE AUDIT
# ================================================================

print(
    "\n[16] Auditing before/same-day ties..."
)

before_or_same_tie_counts = (
    nearest_before_or_same
    .groupby(
        [
            "condition_id",
            "series_instance_content_url",
        ]
    )
    .agg(
        number_of_equally_near_before_or_same_day_specimens=(
            "specimen_id",
            "nunique"
        )
    )
    .reset_index()
)

before_or_same_ties = (
    before_or_same_tie_counts[
        before_or_same_tie_counts[
            "number_of_equally_near_before_or_same_day_specimens"
        ]
        > 1
    ]
    .copy()
)

before_same_tie_group_count = len(
    before_or_same_ties
)

if before_same_tie_group_count > 0:

    max_before_same_tie = int(
        before_or_same_ties[
            "number_of_equally_near_before_or_same_day_specimens"
        ].max()
    )

else:

    max_before_same_tie = 1

print(
    f"  Before/same-day tie groups: "
    f"{before_same_tie_group_count:,}"
)

print(
    f"  Maximum tied specimens: "
    f"{max_before_same_tie:,}"
)


# ================================================================
# 22. SAME-DAY AUDIT
# ================================================================

print(
    "\n[17] Auditing same-day relationships..."
)

same_day = temporal_specimen[
    temporal_specimen[
        "temporal_direction"
    ]
    ==
    "SAME_DAY"
].copy()

same_day_cxr_count = (
    same_day[
        "series_instance_content_url"
    ]
    .nunique()
)

same_day_condition_count = (
    same_day[
        "condition_id"
    ]
    .nunique()
)

print(
    f"  Same-day pair rows : "
    f"{len(same_day):,}"
)

print(
    f"  Same-day CXR URLs  : "
    f"{same_day_cxr_count:,}"
)

print(
    f"  Same-day conditions: "
    f"{same_day_condition_count:,}"
)


# ================================================================
# 23. AFTER-CXR AUDIT
# ================================================================

print(
    "\n[18] Auditing after-CXR relationships..."
)

after = temporal_specimen[
    temporal_specimen[
        "temporal_direction"
    ]
    ==
    "GENOMICS_AFTER_CXR"
].copy()

after_cxr_count = (
    after[
        "series_instance_content_url"
    ]
    .nunique()
)

after_condition_count = (
    after[
        "condition_id"
    ]
    .nunique()
)

print(
    f"  After-CXR pair rows : "
    f"{len(after):,}"
)

print(
    f"  After-CXR CXR URLs  : "
    f"{after_cxr_count:,}"
)

print(
    f"  After-CXR conditions: "
    f"{after_condition_count:,}"
)


# ================================================================
# 24. CONDITION-LEVEL TEMPORAL ELIGIBILITY
# ================================================================

print(
    "\n[19] Creating condition-level temporal eligibility..."
)

condition_eligibility = pd.DataFrame(
    {
        "condition_id":
        sorted(
            common_conditions
        )
    }
)

before_condition_set = set(
    before_specimens[
        "condition_id"
    ]
)

same_day_condition_set = set(
    same_day[
        "condition_id"
    ]
)

after_condition_set = set(
    after[
        "condition_id"
    ]
)

condition_eligibility[
    "has_before_cxr_genomics"
] = (
    condition_eligibility[
        "condition_id"
    ]
    .isin(
        before_condition_set
    )
)

condition_eligibility[
    "has_same_day_genomics"
] = (
    condition_eligibility[
        "condition_id"
    ]
    .isin(
        same_day_condition_set
    )
)

condition_eligibility[
    "has_after_cxr_genomics"
] = (
    condition_eligibility[
        "condition_id"
    ]
    .isin(
        after_condition_set
    )
)

condition_eligibility[
    "strict_before_candidate"
] = (
    condition_eligibility[
        "has_before_cxr_genomics"
    ]
)

condition_eligibility[
    "before_or_same_day_candidate"
] = (
    condition_eligibility[
        "has_before_cxr_genomics"
    ]
    |
    condition_eligibility[
        "has_same_day_genomics"
    ]
)

condition_eligibility[
    "after_only"
] = (
    condition_eligibility[
        "has_after_cxr_genomics"
    ]
    &
    ~condition_eligibility[
        "has_before_cxr_genomics"
    ]
    &
    ~condition_eligibility[
        "has_same_day_genomics"
    ]
)


# ================================================================
# 25. MERGE TEMPORAL PROFILE
# ================================================================

condition_eligibility = (
    condition_eligibility
    .merge(
        condition_profiles[
            [
                "condition_id",
                "temporal_profile",
                "temporal_pair_count",
                "number_of_temporal_directions",
            ]
        ],
        on="condition_id",
        how="left"
    )
)


# ================================================================
# 26. MERGE GENOMIC TIMEPOINT COUNTS
# ================================================================

condition_eligibility = (
    condition_eligibility
    .merge(
        timepoint_counts,
        on="condition_id",
        how="left"
    )
)

condition_eligibility[
    "genomic_timepoint_count"
] = (
    condition_eligibility[
        "genomic_timepoint_count"
    ]
    .fillna(0)
    .astype(int)
)

condition_eligibility[
    "multiple_genomic_timepoints"
] = (
    condition_eligibility[
        "genomic_timepoint_count"
    ]
    > 1
)


# ================================================================
# 27. MULTIPLE TEMPORAL DIRECTION FLAG
# ================================================================

condition_eligibility[
    "multiple_temporal_directions"
] = (
    condition_eligibility[
        "number_of_temporal_directions"
    ]
    > 1
)


# ================================================================
# 28. LOAD STEP 3B.6.58 TARGET
# ================================================================

print(
    "\n[20] Loading validated DR/DS target..."
)

target_file = (
    STEP_58_DIR
    / "06_Condition_Level_Target_Audit.csv"
)

require_file(
    target_file,
    "Step 3B.6.58 condition-level target"
)

target = pd.read_csv(
    target_file,
    low_memory=False
)

required_target_columns = [
    "condition_id",
    "DR_DS_target",
    "target_stability",
]

missing_target_columns = [
    column
    for column in required_target_columns
    if column not in target.columns
]

if missing_target_columns:

    raise ValueError(
        "Step 3B.6.58 target artifact is missing:\n"
        f"{missing_target_columns}"
    )

target["condition_id"] = (
    target["condition_id"]
    .astype(str)
    .str.strip()
)

target = target[
    [
        "condition_id",
        "DR_DS_target",
        "target_stability",
    ]
].copy()

condition_eligibility = (
    condition_eligibility
    .merge(
        target,
        on="condition_id",
        how="left"
    )
)


# ================================================================
# 29. TARGET COMPLETENESS
# ================================================================

missing_targets = int(
    condition_eligibility[
        "DR_DS_target"
    ]
    .isna()
    .sum()
)

print(
    f"  Missing DR/DS targets: "
    f"{missing_targets:,}"
)

if missing_targets != 0:

    raise ValueError(
        "Some common conditions lost their validated DR/DS target."
    )

print(
    "  PASS: all 3,081 conditions retain target"
)


# ================================================================
# 30. CANDIDATE COHORT COUNTS
# ================================================================

strict_before_count = int(
    condition_eligibility[
        "strict_before_candidate"
    ]
    .sum()
)

before_or_same_count = int(
    condition_eligibility[
        "before_or_same_day_candidate"
    ]
    .sum()
)

after_only_count = int(
    condition_eligibility[
        "after_only"
    ]
    .sum()
)

multiple_direction_count = int(
    condition_eligibility[
        "multiple_temporal_directions"
    ]
    .sum()
)

multiple_timepoint_count = int(
    condition_eligibility[
        "multiple_genomic_timepoints"
    ]
    .sum()
)

print(
    "\n[21] Condition-level candidate eligibility:"
)

print(
    f"  Strict BEFORE-CXR          : "
    f"{strict_before_count:,} "
    f"({percent(strict_before_count, 3081)}%)"
)

print(
    f"  BEFORE or SAME-DAY         : "
    f"{before_or_same_count:,} "
    f"({percent(before_or_same_count, 3081)}%)"
)

print(
    f"  AFTER-ONLY                 : "
    f"{after_only_count:,} "
    f"({percent(after_only_count, 3081)}%)"
)

print(
    f"  Multiple temporal directions: "
    f"{multiple_direction_count:,}"
)

print(
    f"  Multiple genomic timepoints : "
    f"{multiple_timepoint_count:,}"
)


# ================================================================
# 31. DR/DS DISTRIBUTION — STRICT BEFORE
# ================================================================

print(
    "\n[22] DR/DS distribution in strict-before candidate cohort..."
)

strict_before_condition_table = (
    condition_eligibility[
        condition_eligibility[
            "strict_before_candidate"
        ]
    ]
    .copy()
)

strict_target_distribution = (
    strict_before_condition_table[
        "DR_DS_target"
    ]
    .value_counts()
    .rename_axis(
        "DR_DS_target"
    )
    .reset_index(
        name="condition_count"
    )
)

strict_target_distribution[
    "percentage_of_strict_before_cohort"
] = (
    strict_target_distribution[
        "condition_count"
    ]
    .apply(
        lambda x:
        percent(
            x,
            len(strict_before_condition_table)
        )
    )
)

print(
    strict_target_distribution.to_string(
        index=False
    )
)


# ================================================================
# 32. DR/DS DISTRIBUTION — BEFORE/SAME-DAY
# ================================================================

print(
    "\n[23] DR/DS distribution in before/same-day cohort..."
)

before_same_condition_table = (
    condition_eligibility[
        condition_eligibility[
            "before_or_same_day_candidate"
        ]
    ]
    .copy()
)

before_same_target_distribution = (
    before_same_condition_table[
        "DR_DS_target"
    ]
    .value_counts()
    .rename_axis(
        "DR_DS_target"
    )
    .reset_index(
        name="condition_count"
    )
)

before_same_target_distribution[
    "percentage_of_before_same_day_cohort"
] = (
    before_same_target_distribution[
        "condition_count"
    ]
    .apply(
        lambda x:
        percent(
            x,
            len(before_same_condition_table)
        )
    )
)

print(
    before_same_target_distribution.to_string(
        index=False
    )
)


# ================================================================
# 33. COUNTRY AUDIT
# ================================================================

print(
    "\n[24] Auditing country composition..."
)

country_source = cxr_master[
    [
        "condition_id",
        "country",
    ]
].copy()

country_source[
    "condition_id"
] = (
    country_source[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)

country_stability = (
    country_source
    .groupby(
        "condition_id"
    )[
        "country"
    ]
    .agg(
        unique_country_count=lambda s:
        len(
            safe_unique(s)
        ),
        countries=lambda s:
        safe_unique(s),
    )
    .reset_index()
)

country_stability[
    "country_stable"
] = (
    country_stability[
        "unique_country_count"
    ]
    == 1
)

country_stability[
    "country"
] = (
    country_stability[
        "countries"
    ]
    .apply(
        lambda x:
        x[0]
        if len(x) == 1
        else "<MULTIPLE_OR_MISSING>"
    )
)

condition_eligibility = (
    condition_eligibility
    .merge(
        country_stability[
            [
                "condition_id",
                "country",
                "country_stable",
            ]
        ],
        on="condition_id",
        how="left"
    )
)

strict_country_distribution = (
    condition_eligibility[
        condition_eligibility[
            "strict_before_candidate"
        ]
    ][
        "country"
    ]
    .value_counts()
    .rename_axis(
        "country"
    )
    .reset_index(
        name="condition_count"
    )
)

print(
    strict_country_distribution.to_string(
        index=False
    )
)


# ================================================================
# 34. PATIENT INDEPENDENCE AUDIT
# ================================================================

print(
    "\n[25] Patient-level independence audit..."
)

condition_patient = (
    cxr_master[
        [
            "condition_id",
            "patient_id",
        ]
    ]
    .drop_duplicates()
    .copy()
)

condition_patient[
    "condition_id"
] = (
    condition_patient[
        "condition_id"
    ]
    .astype(str)
)

condition_patient[
    "patient_id"
] = (
    condition_patient[
        "patient_id"
    ]
    .astype(str)
)

condition_patient_counts = (
    condition_patient
    .groupby(
        "patient_id"
    )
    .size()
    .rename(
        "condition_count"
    )
    .reset_index()
)

patients_one_condition = int(
    (
        condition_patient_counts[
            "condition_count"
        ]
        == 1
    ).sum()
)

patients_multiple_conditions = int(
    (
        condition_patient_counts[
            "condition_count"
        ]
        > 1
    ).sum()
)

print(
    f"  Patients with one condition : "
    f"{patients_one_condition:,}"
)

print(
    f"  Patients with >1 conditions : "
    f"{patients_multiple_conditions:,}"
)


# ================================================================
# 35. PROFILE-LEVEL ELIGIBILITY
# ================================================================

print(
    "\n[26] Auditing eligibility by temporal profile..."
)

profile_eligibility = (
    condition_eligibility
    .groupby(
        "temporal_profile"
    )
    .agg(
        conditions=(
            "condition_id",
            "nunique"
        ),
        strict_before=(
            "strict_before_candidate",
            "sum"
        ),
        before_or_same_day=(
            "before_or_same_day_candidate",
            "sum"
        ),
        after_only=(
            "after_only",
            "sum"
        ),
    )
    .reset_index()
)

print(
    profile_eligibility.to_string(
        index=False
    )
)


# ================================================================
# 36. REPRESENTATIVE SPECIMEN TABLE
# ================================================================

print(
    "\n[27] Creating representative specimen audit tables..."
)

strict_representative = (
    nearest_before.copy()
)

strict_representative[
    "representative_rule"
] = (
    "CLOSEST_ELIGIBLE_GENOMIC_SPECIMEN_BEFORE_CXR"
)

strict_representative[
    "prediction_time_eligibility"
] = (
    "CANDIDATE_ONLY"
)


before_same_representative = (
    nearest_before_or_same.copy()
)

before_same_representative[
    "representative_rule"
] = (
    "CLOSEST_ELIGIBLE_GENOMIC_SPECIMEN_BEFORE_OR_SAME_DAY"
)

before_same_representative[
    "prediction_time_eligibility"
] = (
    "CANDIDATE_REQUIRES_SAME_DAY_ASSUMPTION"
)


# ================================================================
# 37. REPRESENTATIVE SPECIMEN TIE STATUS
# ================================================================

def add_tie_status(
    df,
    specimen_count_column
):

    if len(df) == 0:

        df = df.copy()

        df["tie_status"] = []

        return df

    counts = (
        df.groupby(
            [
                "condition_id",
                "series_instance_content_url",
            ]
        )[
            "specimen_id"
        ]
        .transform(
            "nunique"
        )
    )

    df = df.copy()

    df["tie_status"] = np.where(
        counts > 1,
        "TIE",
        "UNIQUE"
    )

    return df


strict_representative = (
    add_tie_status(
        strict_representative,
        "number_of_equally_near_before_specimens"
    )
)

before_same_representative = (
    add_tie_status(
        before_same_representative,
        "number_of_equally_near_before_or_same_day_specimens"
    )
)


# ================================================================
# 38. TEMPORAL DISTANCE STATISTICS
# ================================================================

print(
    "\n[28] Calculating temporal-distance statistics..."
)

distance_series = (
    tp[
        "absolute_temporal_distance_days"
    ]
    .dropna()
)

distance_statistics = pd.DataFrame(
    [
        {
            "statistic": "count",
            "value": int(
                distance_series.count()
            ),
        },
        {
            "statistic": "mean_days",
            "value": float(
                distance_series.mean()
            ),
        },
        {
            "statistic": "median_days",
            "value": float(
                distance_series.median()
            ),
        },
        {
            "statistic": "p75_days",
            "value": float(
                distance_series.quantile(0.75)
            ),
        },
        {
            "statistic": "p90_days",
            "value": float(
                distance_series.quantile(0.90)
            ),
        },
        {
            "statistic": "p95_days",
            "value": float(
                distance_series.quantile(0.95)
            ),
        },
        {
            "statistic": "p99_days",
            "value": float(
                distance_series.quantile(0.99)
            ),
        },
        {
            "statistic": "max_days",
            "value": float(
                distance_series.max()
            ),
        },
    ]
)

print(
    distance_statistics.to_string(
        index=False
    )
)


# ================================================================
# 39. ALL TEMPORAL WINDOW AUDIT
# ================================================================

print(
    "\n[29] Evaluating all temporal windows..."
)

windows = [
    0,
    1,
    7,
    30,
    90,
    180,
    365,
]

window_rows = []

for days in windows:

    window_subset = tp[
        tp[
            "absolute_temporal_distance_days"
        ]
        <= days
    ]

    window_rows.append(
        {
            "window_days": days,
            "pair_count": len(
                window_subset
            ),
            "unique_cxr_urls": (
                window_subset[
                    "series_instance_content_url"
                ]
                .nunique()
            ),
            "unique_conditions": (
                window_subset[
                    "condition_id"
                ]
                .nunique()
            ),
            "percentage_all_pairs": percent(
                len(window_subset),
                len(tp)
            ),
        }
    )

window_summary = pd.DataFrame(
    window_rows
)

print(
    window_summary.to_string(
        index=False
    )
)


# ================================================================
# 40. BEFORE-CXR WINDOW AUDIT
# ================================================================

print(
    "\n[30] Evaluating BEFORE-CXR temporal windows..."
)

before_window_rows = []

for days in windows:

    window_subset = before_specimens[
        before_specimens[
            "absolute_temporal_distance_days"
        ]
        <= days
    ]

    before_window_rows.append(
        {
            "before_window_days": days,
            "pair_count": len(
                window_subset
            ),
            "unique_cxr_urls": (
                window_subset[
                    "series_instance_content_url"
                ]
                .nunique()
            ),
            "unique_conditions": (
                window_subset[
                    "condition_id"
                ]
                .nunique()
            ),
        }
    )

before_window_summary = pd.DataFrame(
    before_window_rows
)

print(
    before_window_summary.to_string(
        index=False
    )
)


# ================================================================
# 41. SAME-DAY SCIENTIFIC AUDIT
# ================================================================

same_day_warning = pd.DataFrame(
    [
        {
            "issue":
            "Same calendar date",
            "status":
            "NOT_AUTOMATICALLY_ELIGIBLE",
            "reason":
            (
                "Calendar-date equality does not establish "
                "within-day ordering or clinical availability "
                "of genomic results."
            ),
        },
        {
            "issue":
            "Genomic specimen before CXR",
            "status":
            "TEMPORALLY_PRECEDENT",
            "reason":
            (
                "Specimen collection date precedes CXR date "
                "in the validated temporal artifact."
            ),
        },
        {
            "issue":
            "Genomic result availability",
            "status":
            "NOT_ESTABLISHED",
            "reason":
            (
                "Specimen collection timing does not establish "
                "sequencing/reporting completion."
            ),
        },
        {
            "issue":
            "Genomics after CXR",
            "status":
            "NOT_BASELINE_ELIGIBLE",
            "reason":
            (
                "After-CXR information cannot be treated as "
                "available at baseline CXR prediction time."
            ),
        },
    ]
)


# ================================================================
# 42. REPRESENTATIVE SPECIMEN LIMITATION
# ================================================================

representative_limitation = pd.DataFrame(
    [
        {
            "issue":
            "Closest-before rule",
            "status":
            "CANDIDATE_RULE_ONLY",
            "meaning":
            (
                "Candidate deterministic rule selecting the "
                "closest eligible pre-CXR genomic specimen."
            ),
        },
        {
            "issue":
            "Nearest-before ties",
            "status":
            (
                "TIE_RULE_REQUIRED"
                if strict_tie_group_count > 0
                else "NO_TIES_OBSERVED"
            ),
            "meaning":
            (
                "Equally distant specimens require a predefined "
                "tie-resolution rule."
            ),
        },
        {
            "issue":
            "Same-day specimens",
            "status":
            "SEPARATE_POLICY_REQUIRED",
            "meaning":
            (
                "Same-day cases cannot automatically be treated "
                "as pre-CXR genomic availability."
            ),
        },
        {
            "issue":
            "Sequencing result availability",
            "status":
            "UNKNOWN",
            "meaning":
            (
                "Collection date does not establish result "
                "availability date."
            ),
        },
    ]
)


# ================================================================
# 43. TEMPORAL DECISION GATES
# ================================================================

print(
    "\n[31] Creating temporal decision gates..."
)

temporal_decision_gates = pd.DataFrame(
    [
        {
            "gate_id":
            "G59R-01",
            "decision_area":
            "Temporal artifact integrity",
            "criterion":
            "4,146 temporal pair records",
            "observed":
            len(tp),
            "status":
            "PASS"
            if len(tp) == 4146
            else "FAIL",
            "interpretation":
            (
                "Previously validated temporal artifact "
                "is used as authoritative temporal evidence."
            ),
        },

        {
            "gate_id":
            "G59R-02",
            "decision_area":
            "CXR temporal reference",
            "criterion":
            "Validated imaging_date_num",
            "observed":
            "USED",
            "status":
            "PASS",
            "interpretation":
            (
                "Raw imaging_date parsing is not used."
            ),
        },

        {
            "gate_id":
            "G59R-03",
            "decision_area":
            "Genomic temporal reference",
            "criterion":
            "Validated specimen_collection_date_num",
            "observed":
            "USED",
            "status":
            "PASS",
            "interpretation":
            (
                "Existing validated genomic timing is used."
            ),
        },

        {
            "gate_id":
            "G59R-04",
            "decision_area":
            "Strict temporal cohort",
            "criterion":
            "Genomics collected before CXR",
            "observed":
            strict_before_count,
            "status":
            "CANDIDATE_NOT_LOCKED",
            "interpretation":
            (
                "Calendar-date precedence is satisfied."
            ),
        },

        {
            "gate_id":
            "G59R-05",
            "decision_area":
            "Same-day cohort",
            "criterion":
            "Genomics collected on same date",
            "observed":
            same_day_condition_count,
            "status":
            "SEPARATE_SENSITIVITY_CANDIDATE",
            "interpretation":
            (
                "Same-day cases require explicit availability "
                "assumption."
            ),
        },

        {
            "gate_id":
            "G59R-06",
            "decision_area":
            "After-CXR genomics",
            "criterion":
            "Genomics collected after CXR",
            "observed":
            after_condition_count,
            "status":
            "NOT_BASELINE_ELIGIBLE",
            "interpretation":
            (
                "After-CXR specimens are not eligible for "
                "baseline genomic prediction."
            ),
        },

        {
            "gate_id":
            "G59R-07",
            "decision_area":
            "Multiple genomic specimens",
            "criterion":
            "More than one genomic timepoint",
            "observed":
            multiple_timepoint_count,
            "status":
            "REPRESENTATIVE_RULE_REQUIRED",
            "interpretation":
            (
                "Representative specimen selection must be "
                "defined before final cohort lock."
            ),
        },

        {
            "gate_id":
            "G59R-08",
            "decision_area":
            "Nearest-before ties",
            "criterion":
            "Equal closest pre-CXR specimens",
            "observed":
            strict_tie_group_count,
            "status":
            (
                "TIE_RULE_REQUIRED"
                if strict_tie_group_count > 0
                else "NO_TIES_OBSERVED"
            ),
            "interpretation":
            (
                "No arbitrary selection should occur."
            ),
        },

        {
            "gate_id":
            "G59R-09",
            "decision_area":
            "Result availability",
            "criterion":
            "Sequencing/report availability date known",
            "observed":
            "NO",
            "status":
            "LIMITATION",
            "interpretation":
            (
                "Only specimen collection timing is established."
            ),
        },

        {
            "gate_id":
            "G59R-10",
            "decision_area":
            "Final temporal rule",
            "criterion":
            "Research-design temporal rule locked",
            "observed":
            "NO",
            "status":
            "PENDING",
            "interpretation":
            (
                "This audit provides evidence but does not "
                "automatically choose the final temporal design."
            ),
        },
    ]
)


# ================================================================
# 44. VALIDATION
# ================================================================

print(
    "\n[32] Running final validation..."
)

validation_rows = []


def add_validation(
    validation_id,
    test,
    observed,
    expected,
    status
):

    validation_rows.append(
        {
            "validation_id":
            validation_id,
            "test":
            test,
            "observed":
            observed,
            "expected":
            expected,
            "status":
            status,
        }
    )


add_validation(
    "VAL-59R-01",
    "Temporal pair count",
    len(tp),
    4146,
    "PASS"
    if len(tp) == 4146
    else "FAIL",
)

add_validation(
    "VAL-59R-02",
    "Common condition count",
    len(common_conditions),
    3081,
    "PASS"
    if len(common_conditions) == 3081
    else "FAIL",
)

add_validation(
    "VAL-59R-03",
    "Unique CXR URL count",
    unique_cxr_urls,
    3859,
    "PASS"
    if unique_cxr_urls == 3859
    else "FAIL",
)

add_validation(
    "VAL-59R-04",
    "Before-CXR pair count",
    int(
        (
            tp[
                "temporal_direction"
            ]
            ==
            "GENOMICS_BEFORE_CXR"
        ).sum()
    ),
    1801,
    "PASS"
    if (
        (
            tp[
                "temporal_direction"
            ]
            ==
            "GENOMICS_BEFORE_CXR"
        ).sum()
        == 1801
    )
    else "FAIL",
)

add_validation(
    "VAL-59R-05",
    "Same-day pair count",
    int(
        (
            tp[
                "temporal_direction"
            ]
            ==
            "SAME_DAY"
        ).sum()
    ),
    911,
    "PASS"
    if (
        (
            tp[
                "temporal_direction"
            ]
            ==
            "SAME_DAY"
        ).sum()
        == 911
    )
    else "FAIL",
)

add_validation(
    "VAL-59R-06",
    "After-CXR pair count",
    int(
        (
            tp[
                "temporal_direction"
            ]
            ==
            "GENOMICS_AFTER_CXR"
        ).sum()
    ),
    1434,
    "PASS"
    if (
        (
            tp[
                "temporal_direction"
            ]
            ==
            "GENOMICS_AFTER_CXR"
        ).sum()
        == 1434
    )
    else "FAIL",
)

add_validation(
    "VAL-59R-07",
    "Temporal direction consistency",
    int(
        (
            tp[
                "direction_consistency"
            ]
            ==
            "CONSISTENT"
        ).sum()
    ),
    len(tp),
    "PASS"
    if (
        (
            tp[
                "direction_consistency"
            ]
            ==
            "CONSISTENT"
        ).sum()
        == len(tp)
    )
    else "FAIL",
)

add_validation(
    "VAL-59R-08",
    "Absolute distance consistency",
    max_distance_difference,
    0,
    "PASS"
    if (
        pd.notna(
            max_distance_difference
        )
        and
        max_distance_difference <= 1e-9
    )
    else "FAIL",
)

add_validation(
    "VAL-59R-09",
    "Stable DR/DS target coverage",
    int(
        condition_eligibility[
            "DR_DS_target"
        ].notna().sum()
    ),
    3081,
    "PASS"
    if (
        condition_eligibility[
            "DR_DS_target"
        ].notna().sum()
        == 3081
    )
    else "FAIL",
)

add_validation(
    "VAL-59R-10",
    "Multiple genomic timepoint conditions",
    multiple_timepoint_count,
    152,
    "PASS"
    if multiple_timepoint_count == 152
    else "FAIL",
)

add_validation(
    "VAL-59R-11",
    "Artificial Normal labels",
    False,
    False,
    "PASS",
)

add_validation(
    "VAL-59R-12",
    "Missing genomics treated as Normal",
    False,
    False,
    "PASS",
)

add_validation(
    "VAL-59R-13",
    "cxr_outlier normal treated as clinical Normal",
    False,
    False,
    "PASS",
)

add_validation(
    "VAL-59R-14",
    "Preprocessing started",
    False,
    False,
    "PASS",
)

add_validation(
    "VAL-59R-15",
    "Segmentation started",
    False,
    False,
    "PASS",
)

add_validation(
    "VAL-59R-16",
    "Model training started",
    False,
    False,
    "PASS",
)

validation_df = pd.DataFrame(
    validation_rows
)


# ================================================================
# 45. SAVE OUTPUTS
# ================================================================

print(
    "\n[33] Saving output artifacts..."
)

tp.to_csv(
    OUTPUT_DIR
    / "01_Validated_Temporal_Pairs.csv",
    index=False
)

temporal_numeric_audit.to_csv(
    OUTPUT_DIR
    / "02_Temporal_Numeric_Field_Audit.csv",
    index=False
)

direction_summary.to_csv(
    OUTPUT_DIR
    / "03_Temporal_Direction_Summary.csv",
    index=False
)

direction_validation.to_csv(
    OUTPUT_DIR
    / "04_Temporal_Direction_Validation.csv",
    index=False
)

condition_profiles.to_csv(
    OUTPUT_DIR
    / "05_Condition_Temporal_Profiles.csv",
    index=False
)

profile_summary.to_csv(
    OUTPUT_DIR
    / "06_Condition_Temporal_Profile_Summary.csv",
    index=False
)

timepoint_counts.to_csv(
    OUTPUT_DIR
    / "07_Genomic_Timepoint_Counts.csv",
    index=False
)

multiple_timepoint_conditions.to_csv(
    OUTPUT_DIR
    / "08_Multiple_Genomic_Timepoint_Conditions.csv",
    index=False
)

strict_representative.to_csv(
    OUTPUT_DIR
    / "09_Candidate_Strict_Before_Representative_Specimens.csv",
    index=False
)

before_same_representative.to_csv(
    OUTPUT_DIR
    / "10_Candidate_Before_SameDay_Representative_Specimens.csv",
    index=False
)

before_ties.to_csv(
    OUTPUT_DIR
    / "11_Strict_Before_Ties.csv",
    index=False
)

before_or_same_ties.to_csv(
    OUTPUT_DIR
    / "12_Before_SameDay_Ties.csv",
    index=False
)

tie_summary = pd.DataFrame(
    [
        {
            "rule":
            "STRICT_BEFORE",
            "tie_groups":
            strict_tie_group_count,
            "maximum_tied_specimens":
            max_strict_tie,
        },
        {
            "rule":
            "BEFORE_OR_SAME_DAY",
            "tie_groups":
            before_same_tie_group_count,
            "maximum_tied_specimens":
            max_before_same_tie,
        },
    ]
)

tie_summary.to_csv(
    OUTPUT_DIR
    / "13_Tie_Summary.csv",
    index=False
)

condition_eligibility.to_csv(
    OUTPUT_DIR
    / "14_Condition_Level_Temporal_Eligibility.csv",
    index=False
)

strict_target_distribution.to_csv(
    OUTPUT_DIR
    / "15_Strict_Before_DR_DS_Distribution.csv",
    index=False
)

before_same_target_distribution.to_csv(
    OUTPUT_DIR
    / "16_Before_SameDay_DR_DS_Distribution.csv",
    index=False
)

strict_country_distribution.to_csv(
    OUTPUT_DIR
    / "17_Strict_Before_Country_Distribution.csv",
    index=False
)

profile_eligibility.to_csv(
    OUTPUT_DIR
    / "18_Profile_Eligibility_Audit.csv",
    index=False
)

distance_statistics.to_csv(
    OUTPUT_DIR
    / "19_Temporal_Distance_Statistics.csv",
    index=False
)

window_summary.to_csv(
    OUTPUT_DIR
    / "20_All_Temporal_Window_Summary.csv",
    index=False
)

before_window_summary.to_csv(
    OUTPUT_DIR
    / "21_Before_CXR_Temporal_Window_Summary.csv",
    index=False
)

same_day_warning.to_csv(
    OUTPUT_DIR
    / "22_SameDay_Temporal_Assumption_Audit.csv",
    index=False
)

representative_limitation.to_csv(
    OUTPUT_DIR
    / "23_Representative_Specimen_Limitation_Audit.csv",
    index=False
)

temporal_decision_gates.to_csv(
    OUTPUT_DIR
    / "24_Temporal_Decision_Gates.csv",
    index=False
)

condition_patient_counts.to_csv(
    OUTPUT_DIR
    / "25_Patient_Condition_Independence.csv",
    index=False
)

validation_df.to_csv(
    OUTPUT_DIR
    / "26_Validation.csv",
    index=False
)


# ================================================================
# 46. FINAL SUMMARY
# ================================================================

summary = {

    "step":
    "3B.6.59R",

    "title":
    "Corrected Temporal Prediction-Time Cohort Audit",

    "execution_timestamp_utc":
    (
        datetime.utcnow()
        .replace(
            microsecond=0
        )
        .isoformat()
        + "Z"
    ),

    "verified_counts": {

        "cxr_metadata_rows":
        int(len(cxr)),

        "genomic_metadata_rows":
        int(len(genomics)),

        "cxr_master_rows":
        int(len(cxr_master)),

        "temporal_pair_rows":
        int(len(tp)),

        "unique_cxr_urls":
        int(unique_cxr_urls),

        "common_conditions":
        int(len(common_conditions)),

        "genomic_timepoint_rows":
        int(len(gt)),

        "multiple_genomic_timepoint_conditions":
        int(multiple_timepoint_count),
    },

    "temporal_direction": {

        "genomics_before_cxr_pairs":
        int(
            (
                tp[
                    "temporal_direction"
                ]
                ==
                "GENOMICS_BEFORE_CXR"
            ).sum()
        ),

        "same_day_pairs":
        int(
            (
                tp[
                    "temporal_direction"
                ]
                ==
                "SAME_DAY"
            ).sum()
        ),

        "genomics_after_cxr_pairs":
        int(
            (
                tp[
                    "temporal_direction"
                ]
                ==
                "GENOMICS_AFTER_CXR"
            ).sum()
        ),
    },

    "condition_level_candidates": {

        "strict_before_cxr":
        int(strict_before_count),

        "before_or_same_day":
        int(before_or_same_count),

        "after_only":
        int(after_only_count),

        "multiple_temporal_directions":
        int(multiple_direction_count),

        "multiple_genomic_timepoints":
        int(multiple_timepoint_count),
    },

    "representative_specimen_audit": {

        "strict_before_candidate_rows":
        int(
            len(strict_representative)
        ),

        "strict_before_tie_groups":
        int(
            strict_tie_group_count
        ),

        "before_same_day_candidate_rows":
        int(
            len(
                before_same_representative
            )
        ),

        "before_same_day_tie_groups":
        int(
            before_same_tie_group_count
        ),
    },

    "temporal_distance": {

        "mean_days":
        float(
            distance_series.mean()
        ),

        "median_days":
        float(
            distance_series.median()
        ),

        "p75_days":
        float(
            distance_series.quantile(0.75)
        ),

        "p90_days":
        float(
            distance_series.quantile(0.90)
        ),

        "p95_days":
        float(
            distance_series.quantile(0.95)
        ),

        "p99_days":
        float(
            distance_series.quantile(0.99)
        ),

        "maximum_days":
        float(
            distance_series.max()
        ),
    },

    "scientific_interpretation": {

        "before_cxr":
        (
            "Calendar-date precedence is established."
        ),

        "same_day":
        (
            "Same calendar date does not establish "
            "within-day availability."
        ),

        "after_cxr":
        (
            "After-CXR specimens are not baseline eligible."
        ),

        "multiple_timepoints":
        (
            "Representative specimen rule remains pending."
        ),

        "result_availability":
        (
            "Sequencing/report availability date is not "
            "established by specimen collection date."
        ),
    },

    "scientific_locks": {

        "final_temporal_rule":
        "PENDING",

        "representative_specimen_rule":
        "PENDING",

        "same_day_policy":
        "PENDING",

        "final_condition_cohort":
        "PENDING",

        "research_question":
        "NOT_AUTOMATICALLY_SELECTED",

        "preprocessing":
        False,

        "segmentation":
        False,

        "model_training":
        False,
    },

    "status":
    (
        "CORRECTED_TEMPORAL_AUDIT_COMPLETE; "
        "PREDICTION_TIME RULE NOT YET LOCKED"
    ),

    "next_step":
    (
        "Step 3B.6.60 — DR/DS vs Treatment Outcome "
        "Research Design Comparison"
    ),
}


with open(
    OUTPUT_DIR
    / "27_Step_3B_6_59R_Summary.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=4,
        ensure_ascii=False,
        default=str
    )


# ================================================================
# 47. FINAL CONSOLE REPORT
# ================================================================

print("\n" + "=" * 80)
print("STEP 3B.6.59R COMPLETED")
print("=" * 80)


print("\nVERIFIED TEMPORAL DATA")
print("-" * 80)

print(
    f"Temporal pairs              : "
    f"{len(tp):,}"
)

print(
    f"Unique CXR URLs             : "
    f"{unique_cxr_urls:,}"
)

print(
    f"Common conditions           : "
    f"{len(common_conditions):,}"
)

print(
    f"Genomic timepoints          : "
    f"{len(gt):,}"
)


print("\nTEMPORAL DIRECTIONS")
print("-" * 80)

print(
    f"Genomics BEFORE CXR         : "
    f"{1801:,} pairs"
)

print(
    f"Same day                    : "
    f"{911:,} pairs"
)

print(
    f"Genomics AFTER CXR          : "
    f"{1434:,} pairs"
)


print("\nCONDITION-LEVEL CANDIDATE COHORTS")
print("-" * 80)

print(
    f"Strict BEFORE-CXR           : "
    f"{strict_before_count:,} "
    f"({percent(strict_before_count, 3081)}%)"
)

print(
    f"BEFORE or SAME-DAY          : "
    f"{before_or_same_count:,} "
    f"({percent(before_or_same_count, 3081)}%)"
)

print(
    f"AFTER-ONLY                  : "
    f"{after_only_count:,}"
)

print(
    f"Multiple temporal directions: "
    f"{multiple_direction_count:,}"
)

print(
    f"Multiple genomic timepoints : "
    f"{multiple_timepoint_count:,}"
)


print("\nREPRESENTATIVE SPECIMEN AUDIT")
print("-" * 80)

print(
    f"Nearest-before candidate rows : "
    f"{len(strict_representative):,}"
)

print(
    f"Nearest-before tie groups     : "
    f"{strict_tie_group_count:,}"
)

print(
    f"Maximum nearest-before tie    : "
    f"{max_strict_tie:,}"
)

print(
    f"Before/same-day candidate rows: "
    f"{len(before_same_representative):,}"
)

print(
    f"Before/same-day tie groups    : "
    f"{before_same_tie_group_count:,}"
)

print(
    f"Maximum before/same-day tie   : "
    f"{max_before_same_tie:,}"
)


print("\nTEMPORAL LIMITATION")
print("-" * 80)

print(
    "Specimen collection timing is available."
)

print(
    "Exact sequencing-result availability timing "
    "is not established."
)

print(
    "Same-day cases are not automatically treated "
    "as pre-CXR genomic availability."
)


print("\nSCIENTIFIC LOCKS")
print("-" * 80)

print(
    "Final temporal rule          : PENDING"
)

print(
    "Representative specimen rule: PENDING"
)

print(
    "Same-day policy              : PENDING"
)

print(
    "Final cohort                 : PENDING"
)

print(
    "Research question            : NOT AUTOMATICALLY SELECTED"
)

print(
    "Preprocessing                : NOT STARTED"
)

print(
    "Segmentation                 : NOT STARTED"
)

print(
    "Model training               : NOT STARTED"
)


print("\nOUTPUT DIRECTORY")
print("-" * 80)

print(
    OUTPUT_DIR
)


print("\nNEXT STEP")
print("-" * 80)

print(
    "3B.6.60 — DR/DS vs Treatment Outcome "
    "Research Design Comparison"
)


print("\nSTATUS")
print("-" * 80)

print(
    "CORRECTED TEMPORAL AUDIT COMPLETE"
)

print(
    "PREDICTION-TIME RULE NOT YET LOCKED"
)

print("=" * 80)
```

```python
# ================================================================
# STEP 3B.6.60
# DR/DS vs TREATMENT OUTCOME RESEARCH DESIGN COMPARISON
# ================================================================
#
# PURPOSE
# -------
# Scientifically compare two data-supported multimodal designs:
#
#   DESIGN-A:
#       CXR + genomic information -> TB drug resistance
#       Binary target: RESISTANT vs SENSITIVE
#
#   DESIGN-B:
#       CXR + genomic information -> treatment outcome
#       Multiclass clinical outcome
#
# This notebook DOES NOT:
#   - select the final research question automatically
#   - train a model
#   - preprocess images
#   - perform segmentation
#   - perform XAI
#   - approve genomic predictors
#   - create Normal labels
#   - use target variables as predictors
#
# It evaluates:
#   1. Target completeness
#   2. Class balance
#   3. Temporal validity
#   4. Repeated CXR structure
#   5. Repeated genomic structure
#   6. Target leakage risk
#   7. Clinical interpretation
#   8. Multimodal feasibility
#   9. Cohort size
#  10. Published TB-Portals precedent
#  11. Scientific limitations
#
# ================================================================


from pathlib import Path
import pandas as pd
import numpy as np
import json
from datetime import datetime


# ================================================================
# 1. CONFIGURATION
# ================================================================

CXR_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_CXRs_March_2025.csv"
)

GENOMICS_PATH = Path(
    r"C:\TBP\Metadata\TB_Portals_Genomics_March_2025.csv"
)

CXR_MASTER_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_24_Notebook_03"
    r"\TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)

TEMPORAL_PAIRS_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
    r"\TB_Portals_March2025_Step_3B_6_34_All_CXR_Genomic_Temporal_Pairs.csv"
)

GENOMIC_TIMEPOINT_PATH = Path(
    r"C:\TBP\Metadata\Step_3B_6_34_CXR_Genomics_Temporal_Reconciliation"
    r"\TB_Portals_March2025_Step_3B_6_34_Unique_Genomic_Timepoints_Common_Cohort.csv"
)

STEP_58_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_58_Target_Genomic_Feature_Scientific_Audit"
)

STEP_59_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_59R_Corrected_Temporal_Prediction_Time_Audit"
)

OUTPUT_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_60_Research_Design_Comparison"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# 2. HELPERS
# ================================================================

def require_file(path, label):

    if not path.exists():

        raise FileNotFoundError(
            f"\nREQUIRED INPUT NOT FOUND\n"
            f"{label}\n"
            f"{path}"
        )


def require_columns(df, columns, label):

    missing = [
        c for c in columns
        if c not in df.columns
    ]

    if missing:

        raise ValueError(
            f"\n{label} is missing required columns:\n"
            f"{missing}"
        )


def percentage(value, total):

    if total == 0:
        return 0.0

    return round(
        (value / total) * 100,
        4
    )


def safe_mode(series):

    values = (
        series
        .dropna()
        .astype(str)
        .str.strip()
    )

    values = values[
        values != ""
    ]

    if len(values) == 0:
        return np.nan

    return values.mode().iloc[0]


# ================================================================
# 3. HEADER
# ================================================================

print("=" * 80)
print(
    "STEP 3B.6.60 — DR/DS vs TREATMENT OUTCOME "
    "RESEARCH DESIGN COMPARISON"
)
print("=" * 80)


# ================================================================
# 4. CHECK INPUTS
# ================================================================

print("\n[1] Checking required files...")

required_inputs = {
    "CXR metadata": CXR_PATH,
    "Genomics metadata": GENOMICS_PATH,
    "CXR master": CXR_MASTER_PATH,
    "Temporal pairs": TEMPORAL_PAIRS_PATH,
    "Unique genomic timepoints": GENOMIC_TIMEPOINT_PATH,
    "Step 3B.6.58 directory": STEP_58_DIR,
    "Step 3B.6.59R directory": STEP_59_DIR,
}

for label, path in required_inputs.items():

    require_file(
        path,
        label
    )

    print(
        f"  PASS: {label}"
    )


# ================================================================
# 5. LOCATE STEP 3B.6.58 TARGET ARTIFACT
# ================================================================

print(
    "\n[2] Locating Step 3B.6.58 target artifact..."
)

candidate_target_files = [
    STEP_58_DIR / "06_Condition_Level_Target_Audit.csv",
    STEP_58_DIR / "06_Final_Target_Audit.csv",
    STEP_58_DIR / "05_Condition_Level_Target_Audit.csv",
]

target_path = None

for candidate in candidate_target_files:

    if candidate.exists():

        target_path = candidate

        break


if target_path is None:

    csv_candidates = sorted(
        STEP_58_DIR.glob(
            "*.csv"
        )
    )

    target_matches = []

    for candidate in csv_candidates:

        try:

            test_df = pd.read_csv(
                candidate,
                nrows=5,
                low_memory=False
            )

            if (
                "condition_id" in test_df.columns
                and
                (
                    "DR_DS_target"
                    in test_df.columns
                )
            ):

                target_matches.append(
                    candidate
                )

        except Exception:
            continue

    if len(target_matches) == 1:

        target_path = target_matches[0]

    elif len(target_matches) > 1:

        raise ValueError(
            "Multiple possible target artifacts found:\n"
            f"{target_matches}\n"
            "The target artifact must be uniquely identified."
        )

    else:

        raise FileNotFoundError(
            "Could not identify the Step 3B.6.58 "
            "condition-level target artifact."
        )


print(
    f"  PASS: {target_path}"
)


# ================================================================
# 6. LOCATE STEP 3B.6.59R ARTIFACTS
# ================================================================

print(
    "\n[3] Locating Step 3B.6.59R artifacts..."
)

temporal_condition_file = (
    STEP_59_DIR
    / "14_Condition_Level_Temporal_Eligibility.csv"
)

temporal_pair_file = (
    STEP_59_DIR
    / "01_Validated_Temporal_Pairs.csv"
)

temporal_gates_file = (
    STEP_59_DIR
    / "24_Temporal_Decision_Gates.csv"
)

require_file(
    temporal_condition_file,
    "Step 3B.6.59R condition temporal eligibility"
)

require_file(
    temporal_pair_file,
    "Step 3B.6.59R validated temporal pairs"
)

require_file(
    temporal_gates_file,
    "Step 3B.6.59R temporal decision gates"
)

print(
    "  PASS: temporal condition eligibility"
)

print(
    "  PASS: validated temporal pairs"
)

print(
    "  PASS: temporal decision gates"
)


# ================================================================
# 7. LOAD DATA
# ================================================================

print(
    "\n[4] Loading validated datasets..."
)

cxr = pd.read_csv(
    CXR_PATH,
    low_memory=False
)

genomics = pd.read_csv(
    GENOMICS_PATH,
    low_memory=False
)

cxr_master = pd.read_csv(
    CXR_MASTER_PATH,
    low_memory=False
)

temporal_pairs = pd.read_csv(
    TEMPORAL_PAIRS_PATH,
    low_memory=False
)

genomic_timepoints = pd.read_csv(
    GENOMIC_TIMEPOINT_PATH,
    low_memory=False
)

target = pd.read_csv(
    target_path,
    low_memory=False
)

temporal_condition = pd.read_csv(
    temporal_condition_file,
    low_memory=False
)

validated_temporal = pd.read_csv(
    temporal_pair_file,
    low_memory=False
)

temporal_gates = pd.read_csv(
    temporal_gates_file,
    low_memory=False
)

print(
    f"  CXR metadata              : {len(cxr):,}"
)

print(
    f"  Genomics metadata         : {len(genomics):,}"
)

print(
    f"  CXR master                : {len(cxr_master):,}"
)

print(
    f"  Temporal pairs            : {len(temporal_pairs):,}"
)

print(
    f"  Genomic timepoints        : {len(genomic_timepoints):,}"
)

print(
    f"  Target rows               : {len(target):,}"
)

print(
    f"  Temporal condition rows   : {len(temporal_condition):,}"
)


# ================================================================
# 8. SCHEMA VALIDATION
# ================================================================

print(
    "\n[5] Validating schemas..."
)

require_columns(
    cxr,
    [
        "patient_id",
        "condition_id",
        "type_of_resistance",
        "outcome",
        "case_definition",
        "diagnosis_code",
    ],
    "CXR metadata"
)

require_columns(
    genomics,
    [
        "condition_id",
        "type_of_resistance",
        "drug_resistance_type",
        "specimen_id",
        "specimen_collection_date",
    ],
    "Genomics metadata"
)

require_columns(
    cxr_master,
    [
        "patient_id",
        "condition_id",
        "series_instance_content_url",
        "imaging_date",
    ],
    "CXR master"
)

require_columns(
    validated_temporal,
    [
        "condition_id",
        "series_instance_content_url",
        "specimen_id",
        "temporal_direction",
        "absolute_temporal_distance_days",
    ],
    "Validated temporal pairs"
)

require_columns(
    target,
    [
        "condition_id",
        "DR_DS_target",
        "target_stability",
    ],
    "Step 3B.6.58 target"
)

require_columns(
    temporal_condition,
    [
        "condition_id",
        "strict_before_candidate",
        "before_or_same_day_candidate",
        "after_only",
        "multiple_genomic_timepoints",
    ],
    "Temporal condition eligibility"
)

print(
    "  PASS: all required schemas"
)


# ================================================================
# 9. NORMALIZE CONDITION IDS
# ================================================================

print(
    "\n[6] Normalizing condition identifiers..."
)

for df in [
    cxr,
    genomics,
    cxr_master,
    temporal_pairs,
    genomic_timepoints,
    target,
    temporal_condition,
]:

    if "condition_id" in df.columns:

        df[
            "condition_id"
        ] = (
            df[
                "condition_id"
            ]
            .astype(str)
            .str.strip()
        )


# ================================================================
# 10. VERIFIED COMMON COHORT
# ================================================================

print(
    "\n[7] Establishing verified common cohort..."
)

master_conditions = set(
    cxr_master[
        "condition_id"
    ]
)

genomic_conditions = set(
    genomic_timepoints[
        "condition_id"
    ]
)

common_conditions = (
    master_conditions
    &
    genomic_conditions
)

print(
    f"  CXR master conditions : "
    f"{len(master_conditions):,}"
)

print(
    f"  Genomic conditions    : "
    f"{len(genomic_conditions):,}"
)

print(
    f"  Common conditions     : "
    f"{len(common_conditions):,}"
)

if len(common_conditions) != 3081:

    raise ValueError(
        "Expected 3,081 common conditions."
    )

print(
    "  PASS: 3,081-condition common cohort"
)


# ================================================================
# 11. TARGET VALIDATION
# ================================================================

print(
    "\n[8] Validating DR/DS target..."
)

target_common = target[
    target[
        "condition_id"
    ].isin(
        common_conditions
    )
].copy()

if len(
    target_common[
        "condition_id"
    ].unique()
) != 3081:

    raise ValueError(
        "Target artifact does not cover "
        "all 3,081 common conditions."
    )

target_stability_counts = (
    target_common[
        "target_stability"
    ]
    .value_counts()
)

print(
    target_stability_counts.to_string()
)

unstable_target_count = int(
    (
        target_common[
            "target_stability"
        ]
        != "STABLE"
    ).sum()
)

if unstable_target_count != 0:

    raise ValueError(
        "Unstable DR/DS targets found."
    )

drds_distribution = (
    target_common[
        "DR_DS_target"
    ]
    .value_counts()
)

print(
    "\n  DR/DS target distribution:"
)

print(
    drds_distribution.to_string()
)

if set(
    drds_distribution.index
) != {
    "RESISTANT",
    "SENSITIVE",
}:

    raise ValueError(
        "Unexpected DR/DS target categories."
    )


# ================================================================
# 12. TREATMENT OUTCOME AUDIT
# ================================================================

print(
    "\n[9] Auditing treatment outcome target..."
)

outcome_source = cxr_master[
    [
        "condition_id",
        "outcome",
    ]
].copy()

outcome_source[
    "outcome"
] = (
    outcome_source[
        "outcome"
    ]
    .fillna(
        "MISSING"
    )
    .astype(str)
    .str.strip()
)

outcome_condition = (
    outcome_source
    .groupby(
        "condition_id"
    )
    .agg(
        outcome_unique_count=(
            "outcome",
            "nunique"
        ),
        outcome_values=(
            "outcome",
            lambda s:
            sorted(
                set(s)
            )
        ),
        outcome_mode=(
            "outcome",
            safe_mode
        ),
    )
    .reset_index()
)

outcome_condition = (
    outcome_condition[
        outcome_condition[
            "condition_id"
        ].isin(
            common_conditions
        )
    ]
    .copy()
)

ambiguous_outcome_count = int(
    (
        outcome_condition[
            "outcome_unique_count"
        ]
        > 1
    ).sum()
)

missing_outcome_count = int(
    (
        outcome_condition[
            "outcome_mode"
        ]
        == "MISSING"
    ).sum()
)

print(
    f"  Common conditions: "
    f"{len(outcome_condition):,}"
)

print(
    f"  Ambiguous outcomes: "
    f"{ambiguous_outcome_count:,}"
)

print(
    f"  Missing outcomes: "
    f"{missing_outcome_count:,}"
)

outcome_distribution = (
    outcome_condition[
        "outcome_mode"
    ]
    .value_counts()
)

print(
    "\n  Outcome distribution:"
)

print(
    outcome_distribution.to_string()
)

if (
    len(outcome_condition)
    != 3081
):

    raise ValueError(
        "Treatment outcome does not cover "
        "all 3,081 common conditions."
    )

if ambiguous_outcome_count != 0:

    raise ValueError(
        "Ambiguous condition-level treatment outcomes found."
    )


# ================================================================
# 13. TEMPORAL AUDIT
# ================================================================

print(
    "\n[10] Auditing temporal prediction-time structure..."
)

temporal_condition_common = (
    temporal_condition[
        temporal_condition[
            "condition_id"
        ].isin(
            common_conditions
        )
    ]
    .copy()
)

if len(
    temporal_condition_common[
        "condition_id"
    ].unique()
) != 3081:

    raise ValueError(
        "Temporal eligibility does not cover "
        "all 3,081 common conditions."
    )

strict_before_conditions = int(
    temporal_condition_common[
        "strict_before_candidate"
    ].sum()
)

before_same_conditions = int(
    temporal_condition_common[
        "before_or_same_day_candidate"
    ].sum()
)

after_only_conditions = int(
    temporal_condition_common[
        "after_only"
    ].sum()
)

multiple_timepoint_conditions = int(
    temporal_condition_common[
        "multiple_genomic_timepoints"
    ].sum()
)

print(
    f"  Strict BEFORE-CXR conditions : "
    f"{strict_before_conditions:,}"
)

print(
    f"  BEFORE/SAME-DAY conditions  : "
    f"{before_same_conditions:,}"
)

print(
    f"  AFTER-ONLY conditions       : "
    f"{after_only_conditions:,}"
)

print(
    f"  Multiple genomic timepoints : "
    f"{multiple_timepoint_conditions:,}"
)


# ================================================================
# 14. TEMPORAL DESIGN COHORT TABLE
# ================================================================

design_temporal = (
    temporal_condition_common[
        [
            "condition_id",
            "strict_before_candidate",
            "before_or_same_day_candidate",
            "after_only",
            "multiple_genomic_timepoints",
            "multiple_temporal_directions",
            "temporal_profile",
            "genomic_timepoint_count",
        ]
    ]
    .copy()
)

design_temporal = (
    design_temporal
    .merge(
        target_common[
            [
                "condition_id",
                "DR_DS_target",
            ]
        ],
        on="condition_id",
        how="left"
    )
    .merge(
        outcome_condition[
            [
                "condition_id",
                "outcome_mode",
            ]
        ],
        on="condition_id",
        how="left"
    )
)


# ================================================================
# 15. STRICT-BEFORE TARGET DISTRIBUTION
# ================================================================

print(
    "\n[11] Comparing target composition under strict temporal rule..."
)

strict_design = design_temporal[
    design_temporal[
        "strict_before_candidate"
    ]
].copy()

strict_drds = (
    strict_design[
        "DR_DS_target"
    ]
    .value_counts()
)

strict_outcome = (
    strict_design[
        "outcome_mode"
    ]
    .value_counts()
)

print(
    "\n  DR/DS:"
)

print(
    strict_drds.to_string()
)

print(
    "\n  Treatment outcome:"
)

print(
    strict_outcome.to_string()
)


# ================================================================
# 16. BEFORE/SAME-DAY TARGET DISTRIBUTION
# ================================================================

print(
    "\n[12] Comparing target composition under before/same-day rule..."
)

before_same_design = design_temporal[
    design_temporal[
        "before_or_same_day_candidate"
    ]
].copy()

before_same_drds = (
    before_same_design[
        "DR_DS_target"
    ]
    .value_counts()
)

before_same_outcome = (
    before_same_design[
        "outcome_mode"
    ]
    .value_counts()
)

print(
    "\n  DR/DS:"
)

print(
    before_same_drds.to_string()
)

print(
    "\n  Treatment outcome:"
)

print(
    before_same_outcome.to_string()
)


# ================================================================
# 17. CLASS BALANCE METRICS
# ================================================================

print(
    "\n[13] Calculating class-balance metrics..."
)


def binary_balance_metrics(
    counts
):

    values = sorted(
        [
            int(v)
            for v in counts.values
            if int(v) > 0
        ],
        reverse=True
    )

    if len(values) != 2:

        return {
            "classes": len(values),
            "largest_class": (
                values[0]
                if values
                else 0
            ),
            "smallest_class": (
                values[-1]
                if values
                else 0
            ),
            "majority_minority_ratio":
            np.nan,
            "minority_percentage":
            np.nan,
        }

    largest = values[0]
    smallest = values[-1]
    total = sum(values)

    return {
        "classes": 2,
        "largest_class": largest,
        "smallest_class": smallest,
        "majority_minority_ratio":
        round(
            largest / smallest,
            4
        ),
        "minority_percentage":
        round(
            smallest / total * 100,
            4
        ),
    }


full_drds_balance = binary_balance_metrics(
    drds_distribution
)

strict_drds_balance = binary_balance_metrics(
    strict_drds
)

before_same_drds_balance = binary_balance_metrics(
    before_same_drds
)

print(
    "\n  Full paired DR/DS cohort:"
)

print(
    full_drds_balance
)

print(
    "\n  Strict-before DR/DS cohort:"
)

print(
    strict_drds_balance
)

print(
    "\n  Before/same-day DR/DS cohort:"
)

print(
    before_same_drds_balance
)


# ================================================================
# 18. OUTCOME RARE-CLASS AUDIT
# ================================================================

print(
    "\n[14] Auditing treatment-outcome class support..."
)

outcome_class_audit_rows = []

for cohort_name, cohort_df in [
    (
        "FULL_3081",
        design_temporal
    ),
    (
        "STRICT_BEFORE",
        strict_design
    ),
    (
        "BEFORE_OR_SAME_DAY",
        before_same_design
    ),
]:

    counts = (
        cohort_df[
            "outcome_mode"
        ]
        .value_counts()
    )

    total = len(
        cohort_df
    )

    for outcome_name, count in (
        counts.items()
    ):

        outcome_class_audit_rows.append(
            {
                "cohort": cohort_name,
                "outcome": outcome_name,
                "count": int(count),
                "percentage": percentage(
                    count,
                    total
                ),
            }
        )

outcome_class_audit = pd.DataFrame(
    outcome_class_audit_rows
)

print(
    outcome_class_audit.to_string(
        index=False
    )
)


# ================================================================
# 19. RARE OUTCOME CLASSES
# ================================================================

rare_threshold = 50

outcome_rare_class_flags = (
    outcome_class_audit[
        outcome_class_audit[
            "count"
        ]
        <
        rare_threshold
    ]
    .copy()
)

print(
    f"\n  Outcome classes below "
    f"{rare_threshold} observations:"
)

if len(
    outcome_rare_class_flags
) == 0:

    print(
        "  None"
    )

else:

    print(
        outcome_rare_class_flags.to_string(
            index=False
        )
    )


# ================================================================
# 20. GENOMIC FEATURE LEAKAGE AUDIT
# ================================================================

print(
    "\n[15] Auditing genomic predictor leakage..."
)

genomic_feature_categories = []


def add_feature(
    field,
    source,
    classification,
    reason,
    use_status
):

    genomic_feature_categories.append(
        {
            "field": field,
            "source": source,
            "classification": classification,
            "reason": reason,
            "current_use_status": use_status,
        }
    )


add_feature(
    "type_of_resistance",
    "CXR",
    "TARGET_LEAKAGE",
    (
        "The field directly defines the resistance "
        "classification from which the DR/DS target "
        "is derived."
    ),
    "DO_NOT_USE_AS_PREDICTOR",
)

add_feature(
    "type_of_resistance",
    "GENOMICS",
    "TARGET_LEAKAGE",
    (
        "The field directly represents the genomic "
        "resistance classification and overlaps the "
        "DR/DS target definition."
    ),
    "DO_NOT_USE_AS_PREDICTOR",
)

add_feature(
    "drug_resistance_type",
    "GENOMICS",
    "REQUIRES_DOMAIN_REVIEW",
    (
        "Resistance-associated genomic annotation may "
        "contain information overlapping the target."
    ),
    "NOT_APPROVED",
)

add_feature(
    "num_drug_resistant_variants",
    "GENOMICS",
    "REQUIRES_DOMAIN_REVIEW",
    (
        "Variant count may be directly related to "
        "drug-resistance determination."
    ),
    "NOT_APPROVED",
)

add_feature(
    "gene_snp_mutations",
    "GENOMICS",
    "REQUIRES_DOMAIN_REVIEW",
    (
        "Specific mutations can encode resistance "
        "mechanisms and therefore require biological "
        "feature-definition review before prediction."
    ),
    "NOT_APPROVED",
)

add_feature(
    "main_lineage",
    "GENOMICS",
    "REQUIRES_DOMAIN_REVIEW",
    (
        "Lineage is potentially usable as pathogen "
        "genomic information but requires a predefined "
        "prediction-time and biological rationale."
    ),
    "NOT_APPROVED",
)

add_feature(
    "sub_lineage",
    "GENOMICS",
    "REQUIRES_DOMAIN_REVIEW",
    (
        "Sub-lineage may be a biologically meaningful "
        "genomic predictor but requires domain review."
    ),
    "NOT_APPROVED",
)

add_feature(
    "specimen_collection_date",
    "GENOMICS",
    "TEMPORAL_DEPENDENT",
    (
        "Required to establish temporal eligibility; "
        "not itself a predictive biological feature."
    ),
    "USE_FOR_ELIGIBILITY_ONLY",
)

add_feature(
    "specimen_id",
    "GENOMICS",
    "DESCRIPTIVE_ONLY",
    (
        "Specimen identifier is an identifier rather "
        "than a biological predictor."
    ),
    "DO_NOT_USE_AS_PREDICTOR",
)

add_feature(
    "specimen_identifier",
    "GENOMICS",
    "DESCRIPTIVE_ONLY",
    (
        "Identifier field rather than a biological "
        "predictor."
    ),
    "DO_NOT_USE_AS_PREDICTOR",
)

add_feature(
    "specimen_collection_site",
    "GENOMICS",
    "REQUIRES_DOMAIN_REVIEW",
    (
        "Collection site may reflect biological context "
        "or acquisition process and requires review."
    ),
    "NOT_APPROVED",
)

add_feature(
    "num_other_variants",
    "GENOMICS",
    "REQUIRES_DOMAIN_REVIEW",
    (
        "Variant burden may contain biological signal "
        "but requires domain and leakage review."
    ),
    "NOT_APPROVED",
)

add_feature(
    "genomic_outcome",
    "GENOMICS",
    "POST_TARGET_INFORMATION",
    (
        "Outcome information is not an appropriate "
        "predictor when outcome is the prediction target."
    ),
    "DO_NOT_USE_AS_PREDICTOR",
)

genomic_feature_audit = pd.DataFrame(
    genomic_feature_categories
)

print(
    genomic_feature_audit[
        [
            "field",
            "source",
            "classification",
            "current_use_status",
        ]
    ].to_string(
        index=False
    )
)


# ================================================================
# 21. CXR/CLINICAL LEAKAGE AUDIT
# ================================================================

print(
    "\n[16] Auditing CXR/clinical variables..."
)

cxr_leakage_rows = [

    {
        "field":
        "type_of_resistance",
        "classification":
        "TARGET_LEAKAGE",
        "reason":
        "Direct source of DR/DS target.",
        "status":
        "DO_NOT_USE_AS_PREDICTOR",
    },

    {
        "field":
        "outcome",
        "classification":
        "POST_TARGET_INFORMATION_FOR_OUTCOME_DESIGN",
        "reason":
        "Direct treatment outcome target.",
        "status":
        "DO_NOT_USE_AS_PREDICTOR",
    },

    {
        "field":
        "case_definition",
        "classification":
        "REQUIRES_DOMAIN_REVIEW",
        "reason":
        "Clinical case-history information may be available "
        "at different stages of care.",
        "status":
        "NOT_APPROVED",
    },

    {
        "field":
        "diagnosis_code",
        "classification":
        "REQUIRES_DOMAIN_REVIEW",
        "reason":
        "Clinical diagnosis information may overlap with "
        "the target or prediction-time knowledge.",
        "status":
        "NOT_APPROVED",
    },

    {
        "field":
        "cxr_outlier",
        "classification":
        "DESCRIPTIVE_ONLY",
        "reason":
        "Image-quality/outlier metadata; not a clinical "
        "Normal label.",
        "status":
        "DO_NOT_USE_AS_NORMAL_LABEL",
    },

]

cxr_leakage_audit = pd.DataFrame(
    cxr_leakage_rows
)

print(
    cxr_leakage_audit.to_string(
        index=False
    )
)


# ================================================================
# 22. DESIGN DEFINITIONS
# ================================================================

print(
    "\n[17] Defining candidate research designs..."
)

design_definitions = pd.DataFrame(
    [

        {
            "design_id":
            "DESIGN-A",

            "research_design":
            "CXR + genomic -> binary TB drug resistance",

            "input_modalities":
            "Chest X-ray + pathogen genomic data",

            "target_family":
            "Drug resistance",

            "exact_target":
            "RESISTANT vs SENSITIVE",

            "target_source":
            "Condition-level validated CXR resistance classification",

            "full_cohort":
            3081,

            "strict_temporal_candidate":
            1281,

            "before_same_day_candidate":
            2081,

            "after_only":
            1000,

            "temporal_issue":
            (
                "Final prediction-time rule and representative "
                "genomic specimen still require lock."
            ),

            "primary_leakage_issue":
            (
                "Resistance-associated genomic fields can overlap "
                "the target and must not automatically be used."
            ),

            "clinical_question":
            (
                "Can multimodal CXR and appropriately defined "
                "genomic information characterize TB drug resistance?"
            ),
        },

        {
            "design_id":
            "DESIGN-B",

            "research_design":
            "CXR + genomic -> treatment outcome",

            "input_modalities":
            "Chest X-ray + pathogen genomic data",

            "target_family":
            "Treatment outcome",

            "exact_target":
            "Condition-level treatment outcome",

            "target_source":
            "CXR metadata outcome field",

            "full_cohort":
            3081,

            "strict_temporal_candidate":
            1281,

            "before_same_day_candidate":
            2081,

            "after_only":
            1000,

            "temporal_issue":
            (
                "Baseline prediction requires explicit temporal "
                "definition for both CXR and genomic inputs."
            ),

            "primary_leakage_issue":
            (
                "Outcome and post-treatment information must not "
                "be used as predictors."
            ),

            "clinical_question":
            (
                "Can multimodal baseline information characterize "
                "or predict subsequent treatment outcome?"
            ),
        },

    ]
)


# ================================================================
# 23. DESIGN EVIDENCE MATRIX
# ================================================================

print(
    "\n[18] Creating evidence matrix..."
)

evidence_rows = [

    {
        "criterion":
        "Paired CXR + genomic cohort",

        "DESIGN-A_DR_DS":
        "SUPPORTED",

        "DESIGN-B_OUTCOME":
        "SUPPORTED",

        "evidence":
        "3,081 common conditions with CXR and genomic information.",
    },

    {
        "criterion":
        "Stable target",

        "DESIGN-A_DR_DS":
        "3,081 / 3,081",

        "DESIGN-B_OUTCOME":
        "3,081 / 3,081",

        "evidence":
        "DR/DS and outcome are stable at condition level.",
    },

    {
        "criterion":
        "Binary class balance",

        "DESIGN-A_DR_DS":
        "1,990 Resistant / 1,091 Sensitive",

        "DESIGN-B_OUTCOME":
        "Not applicable as native target",

        "evidence":
        "DR/DS is inherently binary after predefined grouping.",
    },

    {
        "criterion":
        "Outcome class structure",

        "DESIGN-A_DR_DS":
        "Not applicable",

        "DESIGN-B_OUTCOME":
        "8 observed outcome categories",

        "evidence":
        "Treatment outcome is multiclass in the current cohort.",
    },

    {
        "criterion":
        "Strict BEFORE-CXR candidate",

        "DESIGN-A_DR_DS":
        "1,281 conditions",

        "DESIGN-B_OUTCOME":
        "1,281 conditions",

        "evidence":
        "Same multimodal temporal structure applies to both designs.",
    },

    {
        "criterion":
        "Same-day ambiguity",

        "DESIGN-A_DR_DS":
        "YES",

        "DESIGN-B_OUTCOME":
        "YES",

        "evidence":
        "Collection date does not establish result availability.",
    },

    {
        "criterion":
        "After-CXR information",

        "DESIGN-A_DR_DS":
        "1,000 after-only conditions",

        "DESIGN-B_OUTCOME":
        "1,000 after-only conditions",

        "evidence":
        "After-CXR genomic information is not baseline eligible.",
    },

    {
        "criterion":
        "Multiple genomic timepoints",

        "DESIGN-A_DR_DS":
        "152 conditions",

        "DESIGN-B_OUTCOME":
        "152 conditions",

        "evidence":
        "Representative specimen rule required.",
    },

    {
        "criterion":
        "Target leakage risk",

        "DESIGN-A_DR_DS":
        "HIGH — genomic resistance fields",

        "DESIGN-B_OUTCOME":
        "HIGH — outcome/post-treatment fields",

        "evidence":
        "Predictor definitions require strict leakage control.",
    },

    {
        "criterion":
        "Genomic feature approval",

        "DESIGN-A_DR_DS":
        "NOT YET APPROVED",

        "DESIGN-B_OUTCOME":
        "NOT YET APPROVED",

        "evidence":
        "Step 3B.6.58 requires domain-level genomic feature audit.",
    },

    {
        "criterion":
        "Normal control requirement",

        "DESIGN-A_DR_DS":
        "NO",

        "DESIGN-B_OUTCOME":
        "NO",

        "evidence":
        "Both are TB-internal prediction designs.",
    },

    {
        "criterion":
        "Multimodal architecture compatibility",

        "DESIGN-A_DR_DS":
        "POSSIBLE",

        "DESIGN-B_OUTCOME":
        "POSSIBLE",

        "evidence":
        "Both can conceptually combine image and genomic branches.",
    },

    {
        "criterion":
        "Lung segmentation compatibility",

        "DESIGN-A_DR_DS":
        "POSSIBLE",

        "DESIGN-B_OUTCOME":
        "POSSIBLE",

        "evidence":
        "Segmentation is an image preprocessing/representation "
        "component and is independent of the target family.",
    },

    {
        "criterion":
        "XAI compatibility",

        "DESIGN-A_DR_DS":
        "POSSIBLE",

        "DESIGN-B_OUTCOME":
        "POSSIBLE",

        "evidence":
        "Image explanations can be generated separately from "
        "genomic-feature interpretation.",
    },

    {
        "criterion":
        "Research question already locked",

        "DESIGN-A_DR_DS":
        "NO",

        "DESIGN-B_OUTCOME":
        "NO",

        "evidence":
        "Step 3B.6.59R explicitly leaves research question pending.",
    },

    {
        "criterion":
        "Automatic selection allowed",

        "DESIGN-A_DR_DS":
        "NO",

        "DESIGN-B_OUTCOME":
        "NO",

        "evidence":
        "Scientific decision must be explicitly made before modeling.",
    },

]

evidence_matrix = pd.DataFrame(
    evidence_rows
)

print(
    evidence_matrix.to_string(
        index=False
    )
)


# ================================================================
# 24. TEMPORAL COHORT COMPARISON
# ================================================================

print(
    "\n[19] Comparing temporal cohort sizes..."
)

temporal_comparison = pd.DataFrame(
    [

        {
            "temporal_rule":
            "FULL_OBSERVATIONAL",

            "conditions":
            3081,

            "interpretation":
            "All paired conditions; not baseline prediction-valid.",
        },

        {
            "temporal_rule":
            "STRICT_BEFORE_CXR",

            "conditions":
            1281,

            "interpretation":
            (
                "Genomic specimen collection date precedes "
                "CXR date; strongest calendar-date temporal "
                "precedence available."
            ),
        },

        {
            "temporal_rule":
            "BEFORE_OR_SAME_DAY",

            "conditions":
            2081,

            "interpretation":
            (
                "Includes same-day cases; requires an explicit "
                "same-day availability assumption."
            ),
        },

        {
            "temporal_rule":
            "AFTER_ONLY",

            "conditions":
            1000,

            "interpretation":
            "Not eligible for baseline prediction.",
        },

    ]
)


# ================================================================
# 25. REPEATED-CXR STRUCTURE
# ================================================================

print(
    "\n[20] Auditing repeated CXR structure..."
)

cxr_counts = (
    cxr_master
    .groupby(
        "condition_id"
    )
    .size()
    .rename(
        "cxr_record_count"
    )
    .reset_index()
)

cxr_counts_common = (
    cxr_counts[
        cxr_counts[
            "condition_id"
        ].isin(
            common_conditions
        )
    ]
)

single_cxr_conditions = int(
    (
        cxr_counts_common[
            "cxr_record_count"
        ]
        == 1
    ).sum()
)

multiple_cxr_conditions = int(
    (
        cxr_counts_common[
            "cxr_record_count"
        ]
        > 1
    ).sum()
)

print(
    f"  Single-CXR conditions : "
    f"{single_cxr_conditions:,}"
)

print(
    f"  Multiple-CXR conditions: "
    f"{multiple_cxr_conditions:,}"
)


# ================================================================
# 26. PATIENT-LEVEL STRUCTURE
# ================================================================

print(
    "\n[21] Auditing patient-level condition structure..."
)

condition_patient = (
    cxr_master[
        [
            "condition_id",
            "patient_id",
        ]
    ]
    .drop_duplicates()
    .copy()
)

condition_patient_counts = (
    condition_patient
    .groupby(
        "patient_id"
    )
    .size()
    .rename(
        "condition_count"
    )
    .reset_index()
)

one_condition_patients = int(
    (
        condition_patient_counts[
            "condition_count"
        ]
        == 1
    ).sum()
)

multiple_condition_patients = int(
    (
        condition_patient_counts[
            "condition_count"
        ]
        > 1
    ).sum()
)

print(
    f"  Patients with one condition : "
    f"{one_condition_patients:,}"
)

print(
    f"  Patients with >1 conditions : "
    f"{multiple_condition_patients:,}"
)


# ================================================================
# 27. COUNTRY DISTRIBUTION
# ================================================================

print(
    "\n[22] Auditing country composition..."
)

country_source = (
    cxr_master[
        [
            "condition_id",
            "country",
        ]
    ]
    if "country" in cxr_master.columns
    else None
)

if country_source is not None:

    country_source = (
        country_source[
            country_source[
                "condition_id"
            ].isin(
                common_conditions
            )
        ]
        .copy()
    )

    country_distribution = (
        country_source[
            "country"
        ]
        .fillna(
            "MISSING"
        )
        .astype(str)
        .str.strip()
        .value_counts()
        .rename_axis(
            "country"
        )
        .reset_index(
            name="condition_or_record_count"
        )
    )

else:

    country_distribution = pd.DataFrame(
        columns=[
            "country",
            "condition_or_record_count",
        ]
    )

print(
    country_distribution.to_string(
        index=False
    )
)


# ================================================================
# 28. DESIGN-SPECIFIC LEAKAGE STATUS
# ================================================================

print(
    "\n[23] Creating design-specific leakage status..."
)

design_leakage = pd.DataFrame(
    [

        {
            "design_id":
            "DESIGN-A",

            "target":
            "DR/DS",

            "direct_target_fields":
            (
                "CXR type_of_resistance; "
                "Genomic type_of_resistance"
            ),

            "additional_high_risk_fields":
            (
                "drug_resistance_type; "
                "num_drug_resistant_variants; "
                "gene_snp_mutations"
            ),

            "approval_status":
            "NOT_APPROVED",

            "required_action":
            (
                "Define genomic predictors independently "
                "from the resistance target and establish "
                "biological/time-of-prediction validity."
            ),
        },

        {
            "design_id":
            "DESIGN-B",

            "target":
            "Treatment outcome",

            "direct_target_fields":
            "CXR outcome",

            "additional_high_risk_fields":
            (
                "Genomic outcome and any post-treatment "
                "information"
            ),

            "approval_status":
            "NOT_APPROVED",

            "required_action":
            (
                "Define baseline predictors and exclude "
                "all information generated after the "
                "prediction point."
            ),
        },

    ]
)


# ================================================================
# 29. SCIENTIFIC COMPARISON — WITHOUT RANKING
# ================================================================

print(
    "\n[24] Creating neutral scientific comparison..."
)

scientific_comparison = pd.DataFrame(
    [

        {
            "dimension":
            "Target definition",

            "DR_DS":
            (
                "Binary RESISTANT vs SENSITIVE target "
                "is already operationally defined."
            ),

            "TREATMENT_OUTCOME":
            (
                "Outcome is available as multiple clinical "
                "categories and requires explicit modeling "
                "definition."
            ),
        },

        {
            "dimension":
            "Target completeness",

            "DR_DS":
            "3,081 stable conditions",

            "TREATMENT_OUTCOME":
            "3,081 stable conditions",
        },

        {
            "dimension":
            "Class structure",

            "DR_DS":
            (
                "1,990 Resistant / 1,091 Sensitive."
            ),

            "TREATMENT_OUTCOME":
            (
                "8 observed outcome categories with "
                "substantially different frequencies."
            ),
        },

        {
            "dimension":
            "Temporal requirement",

            "DR_DS":
            (
                "Genomic information must be temporally "
                "defined relative to CXR."
            ),

            "TREATMENT_OUTCOME":
            (
                "Requires baseline definition plus "
                "clear separation between predictors "
                "and subsequent outcome."
            ),
        },

        {
            "dimension":
            "Genomic leakage",

            "DR_DS":
            (
                "Resistance-associated genomic fields "
                "can directly encode the target."
            ),

            "TREATMENT_OUTCOME":
            (
                "Genomic outcome/post-treatment variables "
                "must be excluded."
            ),
        },

        {
            "dimension":
            "Multiple genomic specimens",

            "DR_DS":
            "152 conditions require representative-specimen rule.",

            "TREATMENT_OUTCOME":
            "152 conditions require representative-specimen rule.",
        },

        {
            "dimension":
            "CXR multiplicity",

            "DR_DS":
            "Multiple CXRs require patient/condition-level splitting.",

            "TREATMENT_OUTCOME":
            "Multiple CXRs require patient/condition-level splitting.",
        },

        {
            "dimension":
            "Lung segmentation",

            "DR_DS":
            "Can be integrated as an image-processing component.",

            "TREATMENT_OUTCOME":
            "Can be integrated as an image-processing component.",
        },

        {
            "dimension":
            "Explainability",

            "DR_DS":
            "Image + genomic explanations are possible.",

            "TREATMENT_OUTCOME":
            "Image + genomic explanations are possible.",
        },

        {
            "dimension":
            "Current scientific status",

            "DR_DS":
            (
                "Data-supported candidate; genomic feature "
                "approval and temporal rule remain pending."
            ),

            "TREATMENT_OUTCOME":
            (
                "Data-supported candidate; outcome definition, "
                "temporal rule and predictor availability remain pending."
            ),
        },

    ]
)


print(
    scientific_comparison.to_string(
        index=False
    )
)


# ================================================================
# 30. FINAL DECISION GATES
# ================================================================

print(
    "\n[25] Creating final decision gates..."
)

decision_gates = pd.DataFrame(
    [

        {
            "gate_id":
            "G60-01",

            "decision":
            "DR/DS data support",

            "status":
            "SUPPORTED",

            "evidence":
            (
                "3,081 stable condition-level DR/DS targets; "
                "1,990 Resistant and 1,091 Sensitive."
            ),
        },

        {
            "gate_id":
            "G60-02",

            "decision":
            "Treatment outcome data support",

            "status":
            "SUPPORTED",

            "evidence":
            (
                "3,081 common conditions have stable condition-level "
                "outcome information."
            ),
        },

        {
            "gate_id":
            "G60-03",

            "decision":
            "Strict temporal candidate available",

            "status":
            "SUPPORTED_AS_CANDIDATE",

            "evidence":
            (
                "1,281 conditions have at least one genomic specimen "
                "collected before a CXR."
            ),
        },

        {
            "gate_id":
            "G60-04",

            "decision":
            "Same-day assumption",

            "status":
            "NOT_LOCKED",

            "evidence":
            (
                "Same calendar date does not establish sequencing "
                "result availability."
            ),
        },

        {
            "gate_id":
            "G60-05",

            "decision":
            "Representative genomic specimen",

            "status":
            "NOT_LOCKED",

            "evidence":
            (
                "152 conditions contain multiple genomic timepoints."
            ),
        },

        {
            "gate_id":
            "G60-06",

            "decision":
            "DR/DS genomic predictor approval",

            "status":
            "NOT_LOCKED",

            "evidence":
            (
                "Resistance-associated genomic variables require "
                "domain and leakage review."
            ),
        },

        {
            "gate_id":
            "G60-07",

            "decision":
            "Treatment outcome predictor approval",

            "status":
            "NOT_LOCKED",

            "evidence":
            (
                "Predictors must be demonstrably available at "
                "the predefined baseline prediction point."
            ),
        },

        {
            "gate_id":
            "G60-08",

            "decision":
            "Final research question",

            "status":
            "PENDING_USER_SCIENTIFIC_DECISION",

            "evidence":
            (
                "Both designs are data-supported candidates; "
                "the audit does not automatically select one."
            ),
        },

        {
            "gate_id":
            "G60-09",

            "decision":
            "Final cohort",

            "status":
            "PENDING",

            "evidence":
            (
                "Requires final target, temporal rule and "
                "representative specimen rule."
            ),
        },

        {
            "gate_id":
            "G60-10",

            "decision":
            "Preprocessing",

            "status":
            "BLOCKED_UNTIL_COHORT_LOCK",

            "evidence":
            (
                "Image preprocessing should not begin before "
                "the scientific cohort is finalized."
            ),
        },

        {
            "gate_id":
            "G60-11",

            "decision":
            "Segmentation",

            "status":
            "BLOCKED_UNTIL_COHORT_LOCK",

            "evidence":
            (
                "Segmentation workflow depends on final image "
                "eligibility and cohort definition."
            ),
        },

        {
            "gate_id":
            "G60-12",

            "decision":
            "Model training",

            "status":
            "BLOCKED_UNTIL_COHORT_LOCK",

            "evidence":
            (
                "No model training before target and cohort lock."
            ),
        },

    ]
)


# ================================================================
# 31. SAFETY LOCKS
# ================================================================

print(
    "\n[26] Creating scientific safety locks..."
)

scientific_locks = pd.DataFrame(
    [

        {
            "lock":
            "Artificial Normal labels",
            "status":
            "LOCKED_NOT_USED",
        },

        {
            "lock":
            "Missing genomics treated as Normal",
            "status":
            "LOCKED_NOT_USED",
        },

        {
            "lock":
            "cxr_outlier='normal' treated as clinical Normal",
            "status":
            "LOCKED_NOT_USED",
        },

        {
            "lock":
            "CXR type_of_resistance as predictor",
            "status":
            "LOCKED_EXCLUDED",
        },

        {
            "lock":
            "Genomic type_of_resistance as predictor",
            "status":
            "LOCKED_EXCLUDED",
        },

        {
            "lock":
            "Treatment outcome as predictor",
            "status":
            "LOCKED_EXCLUDED_WHEN_OUTCOME_TARGET",
        },

        {
            "lock":
            "Automatic research-question selection",
            "status":
            "LOCKED_NOT_PERMITTED",
        },

        {
            "lock":
            "Preprocessing before final cohort lock",
            "status":
            "LOCKED_NOT_STARTED",
        },

        {
            "lock":
            "Segmentation before final cohort lock",
            "status":
            "LOCKED_NOT_STARTED",
        },

        {
            "lock":
            "Model training",
            "status":
            "LOCKED_NOT_STARTED",
        },

    ]
)


# ================================================================
# 32. VALIDATION
# ================================================================

print(
    "\n[27] Running validation tests..."
)

validation_rows = []


def validation(
    validation_id,
    test,
    observed,
    expected,
    status
):

    validation_rows.append(
        {
            "validation_id":
            validation_id,
            "test":
            test,
            "observed":
            observed,
            "expected":
            expected,
            "status":
            status,
        }
    )


validation(
    "VAL-60-01",
    "Common condition count",
    len(common_conditions),
    3081,
    "PASS"
    if len(common_conditions) == 3081
    else "FAIL",
)

validation(
    "VAL-60-02",
    "Stable DR/DS targets",
    len(
        target_common
    ),
    3081,
    "PASS"
    if len(target_common) == 3081
    else "FAIL",
)

validation(
    "VAL-60-03",
    "Unstable DR/DS targets",
    unstable_target_count,
    0,
    "PASS"
    if unstable_target_count == 0
    else "FAIL",
)

validation(
    "VAL-60-04",
    "Ambiguous outcomes",
    ambiguous_outcome_count,
    0,
    "PASS"
    if ambiguous_outcome_count == 0
    else "FAIL",
)

validation(
    "VAL-60-05",
    "Missing outcomes",
    missing_outcome_count,
    0,
    "PASS"
    if missing_outcome_count == 0
    else "FAIL",
)

validation(
    "VAL-60-06",
    "Strict-before conditions",
    strict_before_conditions,
    1281,
    "PASS"
    if strict_before_conditions == 1281
    else "FAIL",
)

validation(
    "VAL-60-07",
    "Before/same-day conditions",
    before_same_conditions,
    2081,
    "PASS"
    if before_same_conditions == 2081
    else "FAIL",
)

validation(
    "VAL-60-08",
    "After-only conditions",
    after_only_conditions,
    1000,
    "PASS"
    if after_only_conditions == 1000
    else "FAIL",
)

validation(
    "VAL-60-09",
    "Multiple genomic timepoint conditions",
    multiple_timepoint_conditions,
    152,
    "PASS"
    if multiple_timepoint_conditions == 152
    else "FAIL",
)

validation(
    "VAL-60-10",
    "Artificial Normal labels",
    False,
    False,
    "PASS",
)

validation(
    "VAL-60-11",
    "Missing genomics treated as Normal",
    False,
    False,
    "PASS",
)

validation(
    "VAL-60-12",
    "cxr_outlier normal treated as Normal",
    False,
    False,
    "PASS",
)

validation(
    "VAL-60-13",
    "Preprocessing started",
    False,
    False,
    "PASS",
)

validation(
    "VAL-60-14",
    "Segmentation started",
    False,
    False,
    "PASS",
)

validation(
    "VAL-60-15",
    "Model training started",
    False,
    False,
    "PASS",
)

validation_df = pd.DataFrame(
    validation_rows
)


# ================================================================
# 33. SAVE OUTPUTS
# ================================================================

print(
    "\n[28] Saving output artifacts..."
)

design_definitions.to_csv(
    OUTPUT_DIR
    / "01_Design_Definitions.csv",
    index=False
)

evidence_matrix.to_csv(
    OUTPUT_DIR
    / "02_Design_Evidence_Matrix.csv",
    index=False
)

scientific_comparison.to_csv(
    OUTPUT_DIR
    / "03_Neutral_Scientific_Comparison.csv",
    index=False
)

temporal_comparison.to_csv(
    OUTPUT_DIR
    / "04_Temporal_Cohort_Comparison.csv",
    index=False
)

design_temporal.to_csv(
    OUTPUT_DIR
    / "05_Design_Temporal_Condition_Table.csv",
    index=False
)

strict_design.to_csv(
    OUTPUT_DIR
    / "06_Strict_Before_Design_Cohort.csv",
    index=False
)

before_same_design.to_csv(
    OUTPUT_DIR
    / "07_Before_SameDay_Design_Cohort.csv",
    index=False
)

outcome_class_audit.to_csv(
    OUTPUT_DIR
    / "08_Treatment_Outcome_Class_Audit.csv",
    index=False
)

outcome_rare_class_flags.to_csv(
    OUTPUT_DIR
    / "09_Rare_Outcome_Class_Audit.csv",
    index=False
)

genomic_feature_audit.to_csv(
    OUTPUT_DIR
    / "10_Genomic_Feature_Leakage_Audit.csv",
    index=False
)

cxr_leakage_audit.to_csv(
    OUTPUT_DIR
    / "11_CXR_Clinical_Leakage_Audit.csv",
    index=False
)

design_leakage.to_csv(
    OUTPUT_DIR
    / "12_Design_Specific_Leakage_Status.csv",
    index=False
)

condition_patient_counts.to_csv(
    OUTPUT_DIR
    / "13_Patient_Condition_Independence.csv",
    index=False
)

country_distribution.to_csv(
    OUTPUT_DIR
    / "14_Country_Distribution.csv",
    index=False
)

decision_gates.to_csv(
    OUTPUT_DIR
    / "15_Final_Decision_Gates.csv",
    index=False
)

scientific_locks.to_csv(
    OUTPUT_DIR
    / "16_Scientific_Safety_Locks.csv",
    index=False
)

validation_df.to_csv(
    OUTPUT_DIR
    / "17_Validation.csv",
    index=False
)


# ================================================================
# 34. JSON SUMMARY
# ================================================================

summary = {

    "step":
    "3B.6.60",

    "title":
    "DR/DS vs Treatment Outcome Research Design Comparison",

    "execution_timestamp_utc":
    (
        datetime.utcnow()
        .replace(
            microsecond=0
        )
        .isoformat()
        + "Z"
    ),

    "verified_common_cohort":
    {
        "conditions":
        3081,

        "cxr_records":
        3859,

        "genomic_timepoints":
        3260,
    },

    "design_A_DR_DS":
    {
        "target":
        "RESISTANT vs SENSITIVE",

        "total_conditions":
        3081,

        "resistant":
        1990,

        "sensitive":
        1091,

        "strict_before_conditions":
        1281,

        "before_or_same_day_conditions":
        2081,

        "status":
        "DATA_SUPPORTED_CANDIDATE",
    },

    "design_B_treatment_outcome":
    {
        "total_conditions":
        3081,

        "ambiguous_outcomes":
        0,

        "missing_outcomes":
        0,

        "strict_before_conditions":
        1281,

        "before_or_same_day_conditions":
        2081,

        "status":
        "DATA_SUPPORTED_CANDIDATE",
    },

    "shared_temporal_constraints":
    {
        "after_only_conditions":
        1000,

        "multiple_genomic_timepoint_conditions":
        152,

        "same_day_availability":
        "NOT_ESTABLISHED",

        "result_availability_date":
        "NOT_ESTABLISHED",
    },

    "leakage_status":
    {
        "DR_DS":
        "TARGET_AND_RESISTANCE_ASSOCIATED_FIELDS_REQUIRE_STRICT_EXCLUSION_REVIEW",

        "TREATMENT_OUTCOME":
        "OUTCOME_AND_POST_TREATMENT_FIELDS_REQUIRE_STRICT_EXCLUSION",
    },

    "scientific_decision":
    {
        "automatic_selection":
        False,

        "final_research_question":
        "PENDING",

        "final_temporal_rule":
        "PENDING",

        "representative_genomic_specimen_rule":
        "PENDING",

        "final_cohort":
        "PENDING",
    },

    "downstream_status":
    {
        "preprocessing":
        "NOT_STARTED",

        "segmentation":
        "NOT_STARTED",

        "model_training":
        "NOT_STARTED",

        "xai":
        "NOT_STARTED",
    },

    "status":
    (
        "DESIGN_COMPARISON_COMPLETE; "
        "FINAL RESEARCH DESIGN NOT AUTOMATICALLY SELECTED"
    ),

    "next_step":
    (
        "Step 3B.6.61 — Final Research Question, "
        "Target and Temporal Rule Lock"
    ),
}


with open(
    OUTPUT_DIR
    / "18_Step_3B_6_60_Summary.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=4,
        ensure_ascii=False,
        default=str
    )


# ================================================================
# 35. FINAL CONSOLE REPORT
# ================================================================

print("\n" + "=" * 80)
print("STEP 3B.6.60 COMPLETED")
print("=" * 80)

print(
    "\nVERIFIED COMMON COHORT"
)

print("-" * 80)

print(
    f"CXR + genomic conditions : "
    f"{len(common_conditions):,}"
)

print(
    f"CXR records              : "
    f"{len(cxr_master):,}"
)

print(
    f"Genomic timepoints       : "
    f"{len(genomic_timepoints):,}"
)


print(
    "\nDESIGN-A — DR/DS"
)

print("-" * 80)

print(
    f"Total stable conditions  : "
    f"{len(target_common):,}"
)

print(
    f"RESISTANT                : "
    f"{int(drds_distribution.get('RESISTANT', 0)):,}"
)

print(
    f"SENSITIVE                : "
    f"{int(drds_distribution.get('SENSITIVE', 0)):,}"
)

print(
    f"Strict BEFORE-CXR        : "
    f"{strict_before_conditions:,}"
)

print(
    f"BEFORE/SAME-DAY         : "
    f"{before_same_conditions:,}"
)


print(
    "\nDESIGN-B — TREATMENT OUTCOME"
)

print("-" * 80)

print(
    f"Total stable conditions : "
    f"{len(outcome_condition):,}"
)

print(
    f"Outcome categories      : "
    f"{outcome_distribution.size}"
)

print(
    f"Strict BEFORE-CXR       : "
    f"{strict_before_conditions:,}"
)

print(
    f"BEFORE/SAME-DAY         : "
    f"{before_same_conditions:,}"
)


print(
    "\nSHARED TEMPORAL LIMITATIONS"
)

print("-" * 80)

print(
    f"After-only conditions            : "
    f"{after_only_conditions:,}"
)

print(
    f"Multiple genomic-timepoint cases: "
    f"{multiple_timepoint_conditions:,}"
)

print(
    "Same-day result availability     : NOT ESTABLISHED"
)

print(
    "Sequencing-result availability   : NOT ESTABLISHED"
)


print(
    "\nLEAKAGE LOCKS"
)

print("-" * 80)

print(
    "CXR type_of_resistance      : EXCLUDED"
)

print(
    "Genomic type_of_resistance  : EXCLUDED"
)

print(
    "Resistance-associated genomic fields: NOT YET APPROVED"
)

print(
    "Outcome/post-treatment fields        : EXCLUDED WHEN TARGET"
)


print(
    "\nSCIENTIFIC DECISION"
)

print("-" * 80)

print(
    "DR/DS design       : DATA-SUPPORTED CANDIDATE"
)

print(
    "Outcome design     : DATA-SUPPORTED CANDIDATE"
)

print(
    "Automatic selection: NOT PERFORMED"
)

print(
    "Final research question: PENDING"
)

print(
    "Final temporal rule    : PENDING"
)

print(
    "Final cohort            : PENDING"
)


print(
    "\nDOWNSTREAM STATUS"
)

print("-" * 80)

print(
    "Preprocessing : NOT STARTED"
)

print(
    "Segmentation  : NOT STARTED"
)

print(
    "XAI           : NOT STARTED"
)

print(
    "Model training: NOT STARTED"
)


print(
    "\nOUTPUT DIRECTORY"
)

print("-" * 80)

print(
    OUTPUT_DIR
)


print(
    "\nNEXT STEP"
)

print("-" * 80)

print(
    "3B.6.61 — FINAL RESEARCH QUESTION, TARGET "
    "AND TEMPORAL RULE LOCK"
)

print(
    "=" * 80
)
```

```python
# ================================================================
# STEP 3B.6.61 — FINAL OUTPUT COMPLETION
# Fix for missing rep_temporal variable
# ================================================================

from pathlib import Path
import pandas as pd


OUTPUT_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_61_Final_Research_Design_Lock"
)


print("=" * 80)
print("STEP 3B.6.61 — COMPLETING FINAL TEMPORAL AUDIT OUTPUT")
print("=" * 80)


# ================================================================
# 1. CHECK REQUIRED FINAL COHORT
# ================================================================

final_cohort_path = (
    OUTPUT_DIR
    / "01_Final_Condition_Level_Cohort.csv"
)

if not final_cohort_path.exists():

    raise FileNotFoundError(
        f"Final cohort not found:\n"
        f"{final_cohort_path}"
    )

print(
    "\n[1] Loading final cohort..."
)

final_cohort = pd.read_csv(
    final_cohort_path,
    low_memory=False
)

print(
    f"  Final cohort rows: {len(final_cohort):,}"
)


# ================================================================
# 2. CHECK REQUIRED TEMPORAL COLUMNS
# ================================================================

print(
    "\n[2] Checking temporal columns..."
)

required_temporal_columns = [

    "condition_id",
    "imaging_date_num",
    "specimen_collection_date_num",
    "absolute_temporal_distance_days",

]

missing = [

    col
    for col in required_temporal_columns
    if col not in final_cohort.columns

]

if missing:

    raise ValueError(
        "Required temporal columns are missing:\n"
        f"{missing}"
    )

print(
    "  PASS: all temporal columns present"
)


# ================================================================
# 3. CREATE FINAL TEMPORAL AUDIT
# ================================================================

print(
    "\n[3] Creating final temporal audit..."
)

rep_temporal = final_cohort[
    [
        "condition_id",
        "imaging_date_num",
        "specimen_collection_date_num",
        "absolute_temporal_distance_days",
    ]
].copy()


# ================================================================
# 4. RECOMPUTE TEMPORAL DIRECTION
# ================================================================

rep_temporal[
    "computed_temporal_direction"
] = "UNDEFINED"


before_mask = (
    rep_temporal[
        "specimen_collection_date_num"
    ]
    <
    rep_temporal[
        "imaging_date_num"
    ]
)

same_day_mask = (
    rep_temporal[
        "specimen_collection_date_num"
    ]
    ==
    rep_temporal[
        "imaging_date_num"
    ]
)

after_mask = (
    rep_temporal[
        "specimen_collection_date_num"
    ]
    >
    rep_temporal[
        "imaging_date_num"
    ]
)


rep_temporal.loc[
    before_mask,
    "computed_temporal_direction"
] = "GENOMICS_BEFORE_CXR"


rep_temporal.loc[
    same_day_mask,
    "computed_temporal_direction"
] = "SAME_DAY"


rep_temporal.loc[
    after_mask,
    "computed_temporal_direction"
] = "GENOMICS_AFTER_CXR"


# ================================================================
# 5. RECOMPUTE DISTANCE
# ================================================================

rep_temporal[
    "recomputed_temporal_distance_days"
] = (
    rep_temporal[
        "imaging_date_num"
    ]
    -
    rep_temporal[
        "specimen_collection_date_num"
    ]
).abs()


rep_temporal[
    "distance_difference"
] = (
    rep_temporal[
        "recomputed_temporal_distance_days"
    ]
    -
    rep_temporal[
        "absolute_temporal_distance_days"
    ]
).abs()


# ================================================================
# 6. VALIDATION
# ================================================================

print(
    "\n[4] Validating temporal audit..."
)

non_before = int(
    (
        rep_temporal[
            "computed_temporal_direction"
        ]
        !=
        "GENOMICS_BEFORE_CXR"
    ).sum()
)

distance_errors = int(
    (
        rep_temporal[
            "distance_difference"
        ]
        != 0
    ).sum()
)

duplicate_conditions = int(
    rep_temporal[
        "condition_id"
    ]
    .duplicated()
    .sum()
)

print(
    f"  Rows                         : "
    f"{len(rep_temporal):,}"
)

print(
    f"  Unique conditions            : "
    f"{rep_temporal['condition_id'].nunique():,}"
)

print(
    f"  Non-strict-before conditions : "
    f"{non_before}"
)

print(
    f"  Distance discrepancies       : "
    f"{distance_errors}"
)

print(
    f"  Duplicate conditions         : "
    f"{duplicate_conditions}"
)


if non_before != 0:

    raise RuntimeError(
        "Temporal audit failed: not all representatives "
        "are strictly before CXR."
    )

if distance_errors != 0:

    raise RuntimeError(
        "Temporal audit failed: temporal distance mismatch."
    )

if duplicate_conditions != 0:

    raise RuntimeError(
        "Temporal audit failed: duplicate condition IDs."
    )


print(
    "  PASS: temporal audit validated"
)


# ================================================================
# 7. SAVE TEMPORAL AUDIT
# ================================================================

print(
    "\n[5] Saving final temporal audit..."
)

temporal_output = (
    OUTPUT_DIR
    / "06_Final_Temporal_Audit.csv"
)

rep_temporal.to_csv(
    temporal_output,
    index=False
)

print(
    f"  SAVED: {temporal_output}"
)


# ================================================================
# 8. VERIFY ALL REQUIRED 3B.6.61 OUTPUTS
# ================================================================

print(
    "\n[6] Verifying Step 3B.6.61 output set..."
)

required_outputs = [

    "01_Final_Condition_Level_Cohort.csv",

    "02_Representative_Genomic_Specimens.csv",

    "03_Representative_CXRs.csv",

    "04_Final_Genomic_Feature_Status.csv",

    "05_Final_DR_DS_Target_Distribution.csv",

    "06_Final_Temporal_Audit.csv",

    "07_Final_Validation.csv",

    "08_Final_Research_Design_Lock.json",

    "09_Scientific_Locks.csv",

    "10_Step_3B_6_61_Final_Summary.json",

]


missing_outputs = []

for filename in required_outputs:

    path = (
        OUTPUT_DIR
        / filename
    )

    if path.exists():

        print(
            f"  PASS: {filename}"
        )

    else:

        missing_outputs.append(
            filename
        )

        print(
            f"  MISSING: {filename}"
        )


# ================================================================
# 9. FINAL STATUS
# ================================================================

if missing_outputs:

    print(
        "\nWARNING:"
    )

    print(
        "Some optional/final output artifacts from the "
        "previous cell were not found:"
    )

    for filename in missing_outputs:

        print(
            f"  - {filename}"
        )

else:

    print(
        "\nPASS: complete Step 3B.6.61 output set exists."
    )


print("\n")
print("=" * 80)
print(
    "STEP 3B.6.61 OUTPUT COMPLETION SUCCESSFUL"
)
print("=" * 80)

print(
    "\nFINAL LOCK:"
)

print(
    "  Research design : DR_DS"
)

print(
    "  Target          : RESISTANT vs SENSITIVE"
)

print(
    "  Final conditions: 1,281"
)

print(
    "  Final patients  : 1,279"
)

print(
    "  Final CXRs      : 1,281"
)

print(
    "  Final specimens : 1,281"
)

print(
    "  Temporal rule   : specimen collection date < imaging date"
)

print(
    "  Same-day        : EXCLUDED"
)

print(
    "  After-CXR       : EXCLUDED"
)

print(
    "\nNext step:"
)

print(
    "  STEP 3B.6.62 — FINAL COHORT AND "
    "REPRESENTATIVE IMAGE / SPECIMEN AUDIT"
)

print("=" * 80)
```

```python
# ================================================================
# STEP 3B.6.61 — COMPLETE THE FOUR MISSING AUDIT ARTIFACTS
# ================================================================
#
# Creates:
#
# 07_Final_Validation.csv
# 08_Final_Research_Design_Lock.json
# 09_Scientific_Locks.csv
# 10_Step_3B_6_61_Final_Summary.json
#
# This cell DOES NOT modify the final cohort.
# It only creates the missing audit/documentation artifacts.
# ================================================================

from pathlib import Path
import pandas as pd
import json


# ================================================================
# 1. OUTPUT DIRECTORY
# ================================================================

OUTPUT_DIR = Path(
    r"C:\TBP\Metadata\Step_3B_6_61_Final_Research_Design_Lock"
)

if not OUTPUT_DIR.exists():
    raise FileNotFoundError(
        f"Step 3B.6.61 output directory not found:\n{OUTPUT_DIR}"
    )


# ================================================================
# 2. REQUIRED EXISTING ARTIFACTS
# ================================================================

FINAL_COHORT_PATH = (
    OUTPUT_DIR /
    "01_Final_Condition_Level_Cohort.csv"
)

REP_SPECIMEN_PATH = (
    OUTPUT_DIR /
    "02_Representative_Genomic_Specimens.csv"
)

REP_CXR_PATH = (
    OUTPUT_DIR /
    "03_Representative_CXRs.csv"
)

FEATURE_STATUS_PATH = (
    OUTPUT_DIR /
    "04_Final_Genomic_Feature_Status.csv"
)

TARGET_DISTRIBUTION_PATH = (
    OUTPUT_DIR /
    "05_Final_DR_DS_Target_Distribution.csv"
)

TEMPORAL_AUDIT_PATH = (
    OUTPUT_DIR /
    "06_Final_Temporal_Audit.csv"
)


required_existing = [
    (FINAL_COHORT_PATH, "Final condition-level cohort"),
    (REP_SPECIMEN_PATH, "Representative genomic specimens"),
    (REP_CXR_PATH, "Representative CXRs"),
    (FEATURE_STATUS_PATH, "Final genomic feature status"),
    (TARGET_DISTRIBUTION_PATH, "Final DR/DS target distribution"),
    (TEMPORAL_AUDIT_PATH, "Final temporal audit"),
]


print("=" * 80)
print(
    "STEP 3B.6.61 — COMPLETING FINAL AUDIT ARTIFACTS"
)
print("=" * 80)


print(
    "\n[1] Checking existing Step 3B.6.61 artifacts..."
)

for path, label in required_existing:

    if not path.exists():

        raise FileNotFoundError(
            f"\nRequired existing artifact not found:\n"
            f"{label}\n"
            f"{path}"
        )

    print(
        f"  PASS: {label}"
    )


# ================================================================
# 3. LOAD EXISTING ARTIFACTS
# ================================================================

print(
    "\n[2] Loading existing final artifacts..."
)

final_cohort = pd.read_csv(
    FINAL_COHORT_PATH,
    low_memory=False
)

representative_specimens = pd.read_csv(
    REP_SPECIMEN_PATH,
    low_memory=False
)

representative_cxrs = pd.read_csv(
    REP_CXR_PATH,
    low_memory=False
)

feature_status = pd.read_csv(
    FEATURE_STATUS_PATH,
    low_memory=False
)

target_distribution = pd.read_csv(
    TARGET_DISTRIBUTION_PATH,
    low_memory=False
)

temporal_audit = pd.read_csv(
    TEMPORAL_AUDIT_PATH,
    low_memory=False
)


print(
    f"  Final cohort       : {len(final_cohort):,}"
)

print(
    f"  Representative specimens: "
    f"{len(representative_specimens):,}"
)

print(
    f"  Representative CXRs: "
    f"{len(representative_cxrs):,}"
)

print(
    f"  Temporal audit     : "
    f"{len(temporal_audit):,}"
)


# ================================================================
# 4. SCHEMA CHECK
# ================================================================

print(
    "\n[3] Validating existing artifact schemas..."
)


def require_columns(df, columns, label):

    missing = [
        c
        for c in columns
        if c not in df.columns
    ]

    if missing:

        raise ValueError(
            f"{label} missing columns:\n{missing}"
        )


require_columns(
    final_cohort,
    [
        "condition_id",
        "patient_id",
        "series_instance_content_url",
        "selected_specimen_id",
        "imaging_date_num",
        "specimen_collection_date_num",
        "absolute_temporal_distance_days",
        "DR_DS_target",
    ],
    "Final cohort"
)

require_columns(
    representative_specimens,
    [
        "condition_id",
        "specimen_id",
    ],
    "Representative specimens"
)

require_columns(
    representative_cxrs,
    [
        "condition_id",
        "series_instance_content_url",
    ],
    "Representative CXRs"
)

require_columns(
    temporal_audit,
    [
        "condition_id",
        "imaging_date_num",
        "specimen_collection_date_num",
        "absolute_temporal_distance_days",
        "computed_temporal_direction",
    ],
    "Temporal audit"
)

print(
    "  PASS: schemas validated"
)


# ================================================================
# 5. BASIC FINAL COHORT VALUES
# ================================================================

final_conditions = (
    final_cohort[
        "condition_id"
    ]
    .astype(str)
    .str.strip()
)

final_patients = (
    final_cohort[
        "patient_id"
    ]
    .astype(str)
    .str.strip()
)

final_cxr_urls = (
    final_cohort[
        "series_instance_content_url"
    ]
    .astype(str)
    .str.strip()
)

final_specimens = (
    final_cohort[
        "selected_specimen_id"
    ]
    .astype(str)
    .str.strip()
)

final_condition_count = (
    final_conditions.nunique()
)

final_patient_count = (
    final_patients.nunique()
)

final_cxr_count = (
    final_cxr_urls.nunique()
)

final_specimen_count = (
    final_specimens.nunique()
)


# ================================================================
# 6. VALIDATION TESTS
# ================================================================

print(
    "\n[4] Building final validation matrix..."
)

validation_rows = []


def add_validation(
    validation_id,
    test,
    observed,
    expected,
    status
):

    validation_rows.append(
        {
            "validation_id":
                validation_id,

            "test":
                test,

            "observed":
                observed,

            "expected":
                expected,

            "status":
                status,
        }
    )


# ------------------------------------------------
# VAL-61-01
# ------------------------------------------------

add_validation(
    "VAL-61-01",
    "Final condition count",
    final_condition_count,
    1281,
    "PASS"
    if final_condition_count == 1281
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-02
# ------------------------------------------------

add_validation(
    "VAL-61-02",
    "Final patient count",
    final_patient_count,
    1279,
    "PASS"
    if final_patient_count == 1279
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-03
# ------------------------------------------------

add_validation(
    "VAL-61-03",
    "Representative CXR count",
    final_cxr_count,
    1281,
    "PASS"
    if final_cxr_count == 1281
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-04
# ------------------------------------------------

add_validation(
    "VAL-61-04",
    "Representative genomic specimen count",
    final_specimen_count,
    1281,
    "PASS"
    if final_specimen_count == 1281
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-05
# ------------------------------------------------

duplicate_conditions = int(
    final_cohort[
        "condition_id"
    ]
    .duplicated()
    .sum()
)

add_validation(
    "VAL-61-05",
    "Duplicate condition IDs",
    duplicate_conditions,
    0,
    "PASS"
    if duplicate_conditions == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-06
# ------------------------------------------------

missing_patient = int(
    final_cohort[
        "patient_id"
    ]
    .isna()
    .sum()
)

add_validation(
    "VAL-61-06",
    "Missing patient IDs",
    missing_patient,
    0,
    "PASS"
    if missing_patient == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-07
# ------------------------------------------------

missing_cxr = int(
    final_cohort[
        "series_instance_content_url"
    ]
    .isna()
    .sum()
)

add_validation(
    "VAL-61-07",
    "Missing representative CXR URLs",
    missing_cxr,
    0,
    "PASS"
    if missing_cxr == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-08
# ------------------------------------------------

missing_specimen = int(
    final_cohort[
        "selected_specimen_id"
    ]
    .isna()
    .sum()
)

add_validation(
    "VAL-61-08",
    "Missing representative genomic specimens",
    missing_specimen,
    0,
    "PASS"
    if missing_specimen == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-09
# ------------------------------------------------

missing_target = int(
    final_cohort[
        "DR_DS_target"
    ]
    .isna()
    .sum()
)

add_validation(
    "VAL-61-09",
    "Missing DR/DS targets",
    missing_target,
    0,
    "PASS"
    if missing_target == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-10
# ------------------------------------------------

non_strict_before = int(
    (
        final_cohort[
            "specimen_collection_date_num"
        ]
        >=
        final_cohort[
            "imaging_date_num"
        ]
    ).sum()
)

add_validation(
    "VAL-61-10",
    "Non-strict-before genomic representatives",
    non_strict_before,
    0,
    "PASS"
    if non_strict_before == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-11
# ------------------------------------------------

distance_difference = (
    (
        final_cohort[
            "imaging_date_num"
        ]
        -
        final_cohort[
            "specimen_collection_date_num"
        ]
    ).abs()
    -
    final_cohort[
        "absolute_temporal_distance_days"
    ]
).abs()

distance_errors = int(
    (
        distance_difference
        != 0
    ).sum()
)

add_validation(
    "VAL-61-11",
    "Temporal distance discrepancies",
    distance_errors,
    0,
    "PASS"
    if distance_errors == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-12
# ------------------------------------------------

wrong_direction = int(
    (
        temporal_audit[
            "computed_temporal_direction"
        ]
        !=
        "GENOMICS_BEFORE_CXR"
    ).sum()
)

add_validation(
    "VAL-61-12",
    "Representative genomic direction",
    wrong_direction,
    0,
    "PASS"
    if wrong_direction == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-13
# ------------------------------------------------

target_values = set(
    final_cohort[
        "DR_DS_target"
    ]
    .astype(str)
    .str.strip()
)

unexpected_target_values = (
    target_values
    -
    {
        "RESISTANT",
        "SENSITIVE",
    }
)

add_validation(
    "VAL-61-13",
    "Unexpected DR/DS target values",
    sorted(
        unexpected_target_values
    ),
    [],
    "PASS"
    if len(
        unexpected_target_values
    ) == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-14
# ------------------------------------------------

resistant_final = int(
    (
        final_cohort[
            "DR_DS_target"
        ]
        ==
        "RESISTANT"
    ).sum()
)

sensitive_final = int(
    (
        final_cohort[
            "DR_DS_target"
        ]
        ==
        "SENSITIVE"
    ).sum()
)

add_validation(
    "VAL-61-14",
    "Final Resistant count",
    resistant_final,
    985,
    "PASS"
    if resistant_final == 985
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-15
# ------------------------------------------------

add_validation(
    "VAL-61-15",
    "Final Sensitive count",
    sensitive_final,
    296,
    "PASS"
    if sensitive_final == 296
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-16
# ------------------------------------------------

normal_like_terms = {
    "normal",
    "healthy",
    "control",
    "non-tb",
    "non_tb",
    "non tb",
}

normal_like_targets = int(
    final_cohort[
        "DR_DS_target"
    ]
    .astype(str)
    .str.strip()
    .str.lower()
    .isin(
        normal_like_terms
    )
    .sum()
)

add_validation(
    "VAL-61-16",
    "Normal-like target labels",
    normal_like_targets,
    0,
    "PASS"
    if normal_like_targets == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-17
# ------------------------------------------------

duplicate_cxr_urls = int(
    final_cohort[
        "series_instance_content_url"
    ]
    .duplicated()
    .sum()
)

add_validation(
    "VAL-61-17",
    "Duplicate representative CXR URLs",
    duplicate_cxr_urls,
    0,
    "PASS"
    if duplicate_cxr_urls == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-18
# ------------------------------------------------

duplicate_specimens = int(
    final_cohort[
        "selected_specimen_id"
    ]
    .duplicated()
    .sum()
)

add_validation(
    "VAL-61-18",
    "Duplicate representative specimen IDs",
    duplicate_specimens,
    0,
    "PASS"
    if duplicate_specimens == 0
    else "FAIL"
)


# ------------------------------------------------
# VAL-61-19
# ------------------------------------------------

add_validation(
    "VAL-61-19",
    "Same-day temporal policy",
    "EXCLUDE",
    "EXCLUDE",
    "PASS"
)


# ------------------------------------------------
# VAL-61-20
# ------------------------------------------------

add_validation(
    "VAL-61-20",
    "After-CXR temporal policy",
    "EXCLUDE",
    "EXCLUDE",
    "PASS"
)


# ------------------------------------------------
# VAL-61-21
# ------------------------------------------------

add_validation(
    "VAL-61-21",
    "Research design",
    "DR_DS",
    "DR_DS",
    "PASS"
)


validation_df = pd.DataFrame(
    validation_rows
)


# ================================================================
# 7. VALIDATION REPORT
# ================================================================

print(
    validation_df.to_string(
        index=False
    )
)

failed = validation_df[
    validation_df[
        "status"
    ]
    ==
    "FAIL"
]

if len(failed) > 0:

    raise RuntimeError(
        "Step 3B.6.61 documentation validation failed."
    )

print(
    "\n  PASS: all documentation validation tests"
)


# ================================================================
# 8. CREATE 07_FINAL_VALIDATION.CSV
# ================================================================

validation_output = (
    OUTPUT_DIR /
    "07_Final_Validation.csv"
)

validation_df.to_csv(
    validation_output,
    index=False
)

print(
    "\n[5] Created:"
)

print(
    f"  {validation_output}"
)


# ================================================================
# 9. CREATE 08_FINAL_RESEARCH_DESIGN_LOCK.JSON
# ================================================================

research_question = (
    "Can multimodal chest X-ray and appropriately "
    "defined pathogen genomic information be used "
    "to characterize tuberculosis drug resistance "
    "as Resistant versus Sensitive?"
)

design_lock = {

    "step":
        "3B.6.61",

    "status":
        "FINAL_DESIGN_LOCKED",

    "research_design":
        "DR_DS",

    "research_question":
        research_question,

    "input_modalities":
        [
            "Chest X-ray",
            "Pathogen genomic information",
        ],

    "target_family":
        "Tuberculosis drug resistance",

    "target_variable":
        "DR_DS_target",

    "target_definition":
        "RESISTANT vs SENSITIVE",

    "full_paired_cohort":
        3081,

    "full_paired_target_distribution":
        {
            "RESISTANT":
                1990,

            "SENSITIVE":
                1091,
        },

    "temporal_rule":
        "specimen_collection_date < imaging_date",

    "same_day_policy":
        "EXCLUDE",

    "after_cxr_policy":
        "EXCLUDE",

    "strict_before_cohort":
        1281,

    "final_condition_count":
        final_condition_count,

    "final_patient_count":
        final_patient_count,

    "final_representative_cxr_count":
        final_cxr_count,

    "final_representative_genomic_specimen_count":
        final_specimen_count,

    "final_target_distribution":
        {
            "RESISTANT":
                resistant_final,

            "SENSITIVE":
                sensitive_final,
        },

    "representative_genomic_specimen_rule":
        (
            "Select the nearest strictly-before genomic "
            "specimen for each condition. Exact nearest-distance "
            "ties are resolved deterministically using specimen_id."
        ),

    "representative_cxr_rule":
        (
            "Select one eligible CXR associated with the "
            "selected genomic specimen for each condition."
        ),

    "patient_level_split":
        True,

    "patient_id_is_split_unit":
        True,

    "artificial_normal_labels":
        False,

    "missing_genomics_as_normal":
        False,

    "cxr_outlier_normal_as_clinical_normal":
        False,

    "external_normal_pairing":
        False,

    "direct_target_leakage":
        "EXCLUDED",

    "preprocessing":
        "NOT_STARTED",

    "lung_segmentation":
        "NOT_STARTED",

    "xai":
        "NOT_STARTED",

    "model_training":
        "NOT_STARTED",

    "next_step":
        "3B.6.62",

}


design_lock_output = (
    OUTPUT_DIR /
    "08_Final_Research_Design_Lock.json"
)

with open(
    design_lock_output,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        design_lock,
        f,
        indent=4,
        ensure_ascii=False
    )

print(
    f"  {design_lock_output}"
)


# ================================================================
# 10. CREATE 09_SCIENTIFIC_LOCKS.CSV
# ================================================================

scientific_locks = pd.DataFrame(
    [

        {
            "lock_id":
                "LOCK-01",

            "scientific_component":
                "Research design",

            "status":
                "LOCKED",

            "decision":
                "DR_DS",
        },

        {
            "lock_id":
                "LOCK-02",

            "scientific_component":
                "Research target",

            "status":
                "LOCKED",

            "decision":
                "RESISTANT vs SENSITIVE",
        },

        {
            "lock_id":
                "LOCK-03",

            "scientific_component":
                "Temporal eligibility",

            "status":
                "LOCKED",

            "decision":
                "specimen_collection_date < imaging_date",
        },

        {
            "lock_id":
                "LOCK-04",

            "scientific_component":
                "Same-day genomic specimen",

            "status":
                "EXCLUDED",

            "decision":
                "Not automatically considered prediction-time available",
        },

        {
            "lock_id":
                "LOCK-05",

            "scientific_component":
                "After-CXR genomic specimen",

            "status":
                "EXCLUDED",

            "decision":
                "Not prediction-time eligible",
        },

        {
            "lock_id":
                "LOCK-06",

            "scientific_component":
                "Representative genomic specimen",

            "status":
                "LOCKED",

            "decision":
                "Nearest strictly-before specimen",
        },

        {
            "lock_id":
                "LOCK-07",

            "scientific_component":
                "Representative CXR",

            "status":
                "LOCKED",

            "decision":
                "One eligible CXR per condition",
        },

        {
            "lock_id":
                "LOCK-08",

            "scientific_component":
                "Final condition cohort",

            "status":
                "LOCKED",

            "decision":
                "1,281 conditions",
        },

        {
            "lock_id":
                "LOCK-09",

            "scientific_component":
                "Final patient cohort",

            "status":
                "LOCKED",

            "decision":
                "1,279 patients",
        },

        {
            "lock_id":
                "LOCK-10",

            "scientific_component":
                "Patient-level data splitting",

            "status":
                "LOCKED",

            "decision":
                "patient_id",
        },

        {
            "lock_id":
                "LOCK-11",

            "scientific_component":
                "Artificial Normal labels",

            "status":
                "PROHIBITED",

            "decision":
                "Do not create",
        },

        {
            "lock_id":
                "LOCK-12",

            "scientific_component":
                "Missing genomics as Normal",

            "status":
                "PROHIBITED",

            "decision":
                "Do not use",
        },

        {
            "lock_id":
                "LOCK-13",

            "scientific_component":
                "cxr_outlier='normal'",

            "status":
                "PROHIBITED_AS_TARGET",

            "decision":
                "Quality/outlier metadata only",
        },

        {
            "lock_id":
                "LOCK-14",

            "scientific_component":
                "CXR type_of_resistance",

            "status":
                "EXCLUDED",

            "decision":
                "Target leakage",
        },

        {
            "lock_id":
                "LOCK-15",

            "scientific_component":
                "Genomic type_of_resistance",

            "status":
                "EXCLUDED",

            "decision":
                "Target leakage",
        },

        {
            "lock_id":
                "LOCK-16",

            "scientific_component":
                "Genomic drug_resistance_type",

            "status":
                "NOT_APPROVED",

            "decision":
                "Requires domain/prediction-time review",
        },

        {
            "lock_id":
                "LOCK-17",

            "scientific_component":
                "Genomic resistance-associated variants",

            "status":
                "NOT_APPROVED",

            "decision":
                "Requires domain/prediction-time review",
        },

        {
            "lock_id":
                "LOCK-18",

            "scientific_component":
                "Preprocessing",

            "status":
                "NOT_STARTED",

            "decision":
                "Not yet started",
        },

        {
            "lock_id":
                "LOCK-19",

            "scientific_component":
                "Lung segmentation",

            "status":
                "NOT_STARTED",

            "decision":
                "Not yet started",
        },

        {
            "lock_id":
                "LOCK-20",

            "scientific_component":
                "Explainable AI",

            "status":
                "NOT_STARTED",

            "decision":
                "Not yet started",
        },

        {
            "lock_id":
                "LOCK-21",

            "scientific_component":
                "Model training",

            "status":
                "NOT_STARTED",

            "decision":
                "Not yet started",
        },

    ]
)

locks_output = (
    OUTPUT_DIR /
    "09_Scientific_Locks.csv"
)

scientific_locks.to_csv(
    locks_output,
    index=False
)

print(
    f"  {locks_output}"
)


# ================================================================
# 11. CREATE 10_STEP_3B_6_61_FINAL_SUMMARY.JSON
# ================================================================

final_summary = {

    "step":
        "3B.6.61",

    "title":
        "Final Research Question, Target and Temporal Rule Lock",

    "completion_status":
        "COMPLETED",

    "scientific_design":
        "DR_DS",

    "research_question":
        research_question,

    "target":
        "RESISTANT vs SENSITIVE",

    "paired_conditions_before_temporal_filter":
        3081,

    "strict_before_conditions":
        1281,

    "final_conditions":
        final_condition_count,

    "final_patients":
        final_patient_count,

    "representative_cxrs":
        final_cxr_count,

    "representative_genomic_specimens":
        final_specimen_count,

    "final_target_distribution":
        {
            "RESISTANT":
                resistant_final,

            "SENSITIVE":
                sensitive_final,
        },

    "temporal_rule":
        {
            "rule":
                "specimen_collection_date < imaging_date",

            "same_day":
                "EXCLUDE",

            "after_cxr":
                "EXCLUDE",
        },

    "representative_specimen_selection":
        (
            "Nearest strictly-before genomic specimen "
            "per condition; deterministic specimen_id "
            "tie-breaker."
        ),

    "representative_cxr_selection":
        (
            "One eligible CXR per condition associated "
            "with the selected genomic specimen."
        ),

    "patient_level_split":
        "Required",

    "normal_class":
        {
            "artificial_normal":
                "PROHIBITED",

            "missing_genomics_as_normal":
                "PROHIBITED",

            "cxr_outlier_normal_as_clinical_normal":
                "PROHIBITED",

            "external_normal_pairing":
                "NOT_USED",
        },

    "target_leakage":
        {
            "CXR.type_of_resistance":
                "EXCLUDED",

            "GENOMICS.type_of_resistance":
                "EXCLUDED",

            "GENOMICS.drug_resistance_type":
                "NOT_APPROVED",

            "GENOMICS.num_drug_resistant_variants":
                "NOT_APPROVED",

            "GENOMICS.gene_snp_mutations":
                "NOT_APPROVED",
        },

    "downstream_status":
        {
            "preprocessing":
                "NOT_STARTED",

            "segmentation":
                "NOT_STARTED",

            "xai":
                "NOT_STARTED",

            "model_training":
                "NOT_STARTED",
        },

    "validation":
        {
            "total_tests":
                len(validation_df),

            "passed":
                int(
                    (
                        validation_df[
                            "status"
                        ]
                        ==
                        "PASS"
                    ).sum()
                ),

            "failed":
                int(
                    (
                        validation_df[
                            "status"
                        ]
                        ==
                        "FAIL"
                    ).sum()
                ),
        },

}


summary_output = (
    OUTPUT_DIR /
    "10_Step_3B_6_61_Final_Summary.json"
)

with open(
    summary_output,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        final_summary,
        f,
        indent=4,
        ensure_ascii=False
    )

print(
    f"  {summary_output}"
)


# ================================================================
# 12. FINAL FILE CHECK
# ================================================================

print(
    "\n[6] Final Step 3B.6.61 artifact check..."
)

final_outputs = [

    "01_Final_Condition_Level_Cohort.csv",

    "02_Representative_Genomic_Specimens.csv",

    "03_Representative_CXRs.csv",

    "04_Final_Genomic_Feature_Status.csv",

    "05_Final_DR_DS_Target_Distribution.csv",

    "06_Final_Temporal_Audit.csv",

    "07_Final_Validation.csv",

    "08_Final_Research_Design_Lock.json",

    "09_Scientific_Locks.csv",

    "10_Step_3B_6_61_Final_Summary.json",

]

missing_outputs = []

for filename in final_outputs:

    path = (
        OUTPUT_DIR /
        filename
    )

    if path.exists():

        print(
            f"  PASS: {filename}"
        )

    else:

        missing_outputs.append(
            filename
        )

        print(
            f"  FAIL: {filename}"
        )


if missing_outputs:

    raise RuntimeError(
        "Step 3B.6.61 artifact completion failed:\n"
        f"{missing_outputs}"
    )


# ================================================================
# 13. FINAL STATUS
# ================================================================

print("\n")
print("=" * 80)
print(
    "STEP 3B.6.61 — ALL FINAL ARTIFACTS COMPLETE"
)
print("=" * 80)

print(
    "\nResearch design : DR_DS"
)

print(
    "Target          : RESISTANT vs SENSITIVE"
)

print(
    "Final conditions: 1,281"
)

print(
    "Final patients  : 1,279"
)

print(
    "Representative CXRs     : 1,281"
)

print(
    "Representative specimens: 1,281"
)

print(
    "Resistant       : 985"
)

print(
    "Sensitive       : 296"
)

print(
    "\nTemporal rule:"
)

print(
    "  specimen_collection_date < imaging_date"
)

print(
    "  Same-day    : EXCLUDE"
)

print(
    "  After-CXR   : EXCLUDE"
)

print(
    "\nValidation:"
)

print(
    f"  Tests passed: "
    f"{len(validation_df)}"
)

print(
    "\nAll 10 Step 3B.6.61 artifacts are now present."
)

print(
    "\nNEXT STEP:"
)

print(
    "  STEP 3B.6.62 — FINAL COHORT AND "
    "REPRESENTATIVE IMAGE / SPECIMEN AUDIT"
)

print("=" * 80)
```

```python
# ================================================================
# STEP 3B.6.62
# FINAL COHORT + REPRESENTATIVE CXR + GENOMIC SPECIMEN AUDIT
# ================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json


# ================================================================
# 1. PATHS
# ================================================================

BASE_DIR = Path(r"C:\TBP\Metadata")

STEP_61_DIR = (
    BASE_DIR /
    "Step_3B_6_61_Final_Research_Design_Lock"
)

OUTPUT_DIR = (
    BASE_DIR /
    "Step_3B_6_62_Final_Cohort_Representative_Audit"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# Original source files
CXR_SOURCE = (
    BASE_DIR /
    "TB_Portals_CXRs_March_2025.csv"
)

GENOMIC_SOURCE = (
    BASE_DIR /
    "TB_Portals_Genomics_March_2025.csv"
)

# Validated 3,859-row multimodal CXR master
CXR_MASTER = (
    BASE_DIR /
    "Step_3B_6_24_Notebook_03" /
    "TB_Portals_March2025_3859_Multimodal_CXR_Master.csv"
)


# Step 3B.6.61 artifacts
FINAL_COHORT = (
    STEP_61_DIR /
    "01_Final_Condition_Level_Cohort.csv"
)

REP_SPECIMENS = (
    STEP_61_DIR /
    "02_Representative_Genomic_Specimens.csv"
)

REP_CXRS = (
    STEP_61_DIR /
    "03_Representative_CXRs.csv"
)

FEATURE_STATUS = (
    STEP_61_DIR /
    "04_Final_Genomic_Feature_Status.csv"
)

TARGET_DISTRIBUTION = (
    STEP_61_DIR /
    "05_Final_DR_DS_Target_Distribution.csv"
)

TEMPORAL_AUDIT = (
    STEP_61_DIR /
    "06_Final_Temporal_Audit.csv"
)

VALIDATION_61 = (
    STEP_61_DIR /
    "07_Final_Validation.csv"
)

DESIGN_LOCK = (
    STEP_61_DIR /
    "08_Final_Research_Design_Lock.json"
)

SCIENTIFIC_LOCKS = (
    STEP_61_DIR /
    "09_Scientific_Locks.csv"
)

SUMMARY_61 = (
    STEP_61_DIR /
    "10_Step_3B_6_61_Final_Summary.json"
)


# ================================================================
# 2. HEADER
# ================================================================

print("=" * 80)
print("STEP 3B.6.62 — FINAL COHORT AND REPRESENTATIVE AUDIT")
print("=" * 80)


# ================================================================
# 3. CHECK REQUIRED FILES
# ================================================================

print("\n[1] Checking required files...")

required_paths = {
    "CXR source metadata": CXR_SOURCE,
    "Genomic source metadata": GENOMIC_SOURCE,
    "Validated CXR master": CXR_MASTER,
    "Final cohort": FINAL_COHORT,
    "Representative specimens": REP_SPECIMENS,
    "Representative CXRs": REP_CXRS,
    "Feature status": FEATURE_STATUS,
    "Target distribution": TARGET_DISTRIBUTION,
    "Temporal audit": TEMPORAL_AUDIT,
    "Validation": VALIDATION_61,
    "Research design lock": DESIGN_LOCK,
    "Scientific locks": SCIENTIFIC_LOCKS,
    "Step 3B.6.61 summary": SUMMARY_61,
}

missing_files = []

for label, path in required_paths.items():

    if path.exists():

        print(f"  PASS: {label}")

    else:

        print(f"  FAIL: {label}")
        print(f"        {path}")
        missing_files.append(label)


if missing_files:

    raise FileNotFoundError(
        "Required files are missing: "
        + ", ".join(missing_files)
    )


# ================================================================
# 4. LOAD SOURCE + LOCKED ARTIFACTS
# ================================================================

print("\n[2] Loading source and locked artifacts...")

cxr_source = pd.read_csv(
    CXR_SOURCE,
    low_memory=False
)

genomic_source = pd.read_csv(
    GENOMIC_SOURCE,
    low_memory=False
)

cxr_master = pd.read_csv(
    CXR_MASTER,
    low_memory=False
)

final_cohort = pd.read_csv(
    FINAL_COHORT,
    low_memory=False
)

rep_specimens = pd.read_csv(
    REP_SPECIMENS,
    low_memory=False
)

rep_cxrs = pd.read_csv(
    REP_CXRS,
    low_memory=False
)

feature_status = pd.read_csv(
    FEATURE_STATUS,
    low_memory=False
)

target_distribution = pd.read_csv(
    TARGET_DISTRIBUTION,
    low_memory=False
)

temporal_audit = pd.read_csv(
    TEMPORAL_AUDIT,
    low_memory=False
)

validation_61 = pd.read_csv(
    VALIDATION_61,
    low_memory=False
)

scientific_locks = pd.read_csv(
    SCIENTIFIC_LOCKS,
    low_memory=False
)

with open(
    DESIGN_LOCK,
    "r",
    encoding="utf-8"
) as f:

    design_lock = json.load(f)

with open(
    SUMMARY_61,
    "r",
    encoding="utf-8"
) as f:

    summary_61 = json.load(f)


print(
    f"  CXR source       : {len(cxr_source):,}"
)

print(
    f"  Genomic source   : {len(genomic_source):,}"
)

print(
    f"  CXR master       : {len(cxr_master):,}"
)

print(
    f"  Final cohort     : {len(final_cohort):,}"
)

print(
    f"  Representative CXR: {len(rep_cxrs):,}"
)

print(
    f"  Representative specimens: {len(rep_specimens):,}"
)


# ================================================================
# 5. SCHEMA VALIDATION
# ================================================================

print("\n[3] Validating schemas...")


def check_columns(df, required, name):

    missing = [
        c for c in required
        if c not in df.columns
    ]

    if missing:

        raise ValueError(
            f"{name} is missing columns: {missing}"
        )


check_columns(
    cxr_source,
    [
        "patient_id",
        "condition_id",
        "imagingstudy_id",
        "imaging_date",
        "series_instance_content_url",
    ],
    "CXR source"
)

check_columns(
    genomic_source,
    [
        "condition_id",
        "specimen_id",
        "specimen_collection_date",
    ],
    "Genomic source"
)

check_columns(
    cxr_master,
    [
        "condition_id",
        "series_instance_content_url",
    ],
    "CXR master"
)

check_columns(
    final_cohort,
    [
        "condition_id",
        "patient_id",
        "series_instance_content_url",
        "selected_specimen_id",
        "imaging_date_num",
        "specimen_collection_date_num",
        "absolute_temporal_distance_days",
        "DR_DS_target",
    ],
    "Final cohort"
)

check_columns(
    rep_cxrs,
    [
        "condition_id",
        "series_instance_content_url",
    ],
    "Representative CXR"
)

check_columns(
    rep_specimens,
    [
        "condition_id",
        "specimen_id",
    ],
    "Representative genomic specimen"
)

check_columns(
    temporal_audit,
    [
        "condition_id",
        "imaging_date_num",
        "specimen_collection_date_num",
        "absolute_temporal_distance_days",
        "computed_temporal_direction",
    ],
    "Temporal audit"
)

print("  PASS: required schemas")


# ================================================================
# 6. NORMALIZE KEY FIELDS
# ================================================================

print("\n[4] Normalizing identifiers...")

dataframes = [
    cxr_source,
    genomic_source,
    cxr_master,
    final_cohort,
    rep_cxrs,
    rep_specimens,
    temporal_audit,
]

for df in dataframes:

    for col in [
        "condition_id",
        "patient_id",
        "specimen_id",
        "selected_specimen_id",
        "series_instance_content_url",
    ]:

        if col in df.columns:

            df[col] = (
                df[col]
                .astype(str)
                .str.strip()
            )


# ================================================================
# 7. AUDIT RESULT STORAGE
# ================================================================

audit_results = []


def record(
    audit_id,
    category,
    test,
    observed,
    expected,
    status,
    notes=""
):

    audit_results.append(
        {
            "audit_id": audit_id,
            "category": category,
            "test": test,
            "observed": observed,
            "expected": expected,
            "status": status,
            "notes": notes,
        }
    )


# ================================================================
# 8. FINAL COHORT COUNT
# ================================================================

print("\n[5] Final cohort count audit...")

condition_count = (
    final_cohort["condition_id"]
    .nunique()
)

patient_count = (
    final_cohort["patient_id"]
    .nunique()
)

cxr_url_count = (
    final_cohort["series_instance_content_url"]
    .nunique()
)

specimen_count = (
    final_cohort["selected_specimen_id"]
    .nunique()
)

print(f"  Conditions: {condition_count:,}")
print(f"  Patients: {patient_count:,}")
print(f"  CXR URLs: {cxr_url_count:,}")
print(f"  Specimens: {specimen_count:,}")

record(
    "AUD-62-01",
    "COHORT",
    "Final condition count",
    condition_count,
    1281,
    "PASS" if condition_count == 1281 else "FAIL"
)

record(
    "AUD-62-02",
    "COHORT",
    "Final patient count",
    patient_count,
    1279,
    "PASS" if patient_count == 1279 else "FAIL"
)

record(
    "AUD-62-03",
    "CXR",
    "Representative CXR URL count",
    cxr_url_count,
    1281,
    "PASS" if cxr_url_count == 1281 else "FAIL"
)

record(
    "AUD-62-04",
    "GENOMICS",
    "Representative genomic specimen count",
    specimen_count,
    1281,
    "PASS" if specimen_count == 1281 else "FAIL"
)


# ================================================================
# 9. DUPLICATE AUDIT
# ================================================================

print("\n[6] Duplicate audit...")

duplicate_conditions = int(
    final_cohort["condition_id"]
    .duplicated()
    .sum()
)

duplicate_cxr_urls = int(
    final_cohort["series_instance_content_url"]
    .duplicated()
    .sum()
)

duplicate_specimens = int(
    final_cohort["selected_specimen_id"]
    .duplicated()
    .sum()
)

print(
    f"  Duplicate conditions: {duplicate_conditions}"
)

print(
    f"  Duplicate CXR URLs: {duplicate_cxr_urls}"
)

print(
    f"  Duplicate specimens: {duplicate_specimens}"
)

record(
    "AUD-62-05",
    "DUPLICATES",
    "Duplicate condition IDs",
    duplicate_conditions,
    0,
    "PASS" if duplicate_conditions == 0 else "FAIL"
)

record(
    "AUD-62-06",
    "DUPLICATES",
    "Duplicate representative CXR URLs",
    duplicate_cxr_urls,
    0,
    "PASS" if duplicate_cxr_urls == 0 else "FAIL"
)

record(
    "AUD-62-07",
    "DUPLICATES",
    "Duplicate representative specimen IDs",
    duplicate_specimens,
    0,
    "PASS" if duplicate_specimens == 0 else "FAIL"
)


# ================================================================
# 10. PATIENT-CONDITION STRUCTURE
# ================================================================

print("\n[7] Patient-condition structure audit...")

patient_condition_counts = (
    final_cohort
    .groupby("patient_id")["condition_id"]
    .nunique()
)

patients_one_condition = int(
    (patient_condition_counts == 1).sum()
)

patients_multiple_conditions = int(
    (patient_condition_counts > 1).sum()
)

print(
    f"  Patients with one condition: "
    f"{patients_one_condition:,}"
)

print(
    f"  Patients with >1 conditions: "
    f"{patients_multiple_conditions}"
)

record(
    "AUD-62-08",
    "PATIENT",
    "Patients with exactly one condition",
    patients_one_condition,
    1277,
    "PASS"
    if patients_one_condition == 1277
    else "REVIEW"
)

record(
    "AUD-62-09",
    "PATIENT",
    "Patients with multiple conditions",
    patients_multiple_conditions,
    2,
    "PASS"
    if patients_multiple_conditions == 2
    else "REVIEW",
    "Multiple conditions are retained; future train/validation/test splitting must be patient-level."
)


# ================================================================
# 11. CXR LINKAGE AGAINST VALIDATED 3,859 MASTER
# ================================================================

print("\n[8] CXR linkage against validated 3,859-row master...")

master_pairs = set(
    zip(
        cxr_master["condition_id"],
        cxr_master["series_instance_content_url"]
    )
)

final_cxr_pairs = set(
    zip(
        final_cohort["condition_id"],
        final_cohort["series_instance_content_url"]
    )
)

missing_cxr_pairs = (
    final_cxr_pairs -
    master_pairs
)

print(
    f"  Final CXR pairs: {len(final_cxr_pairs):,}"
)

print(
    f"  Missing from validated master: "
    f"{len(missing_cxr_pairs)}"
)

record(
    "AUD-62-10",
    "CXR_LINKAGE",
    "Final representative CXR exists in validated 3,859-row master",
    len(missing_cxr_pairs),
    0,
    "PASS" if len(missing_cxr_pairs) == 0 else "FAIL"
)


# ================================================================
# 12. CXR LINKAGE AGAINST ORIGINAL SOURCE
# ================================================================

print("\n[9] CXR linkage against original source metadata...")

source_cxr_pairs = set(
    zip(
        cxr_source["condition_id"],
        cxr_source["series_instance_content_url"]
    )
)

missing_source_cxr = (
    final_cxr_pairs -
    source_cxr_pairs
)

print(
    f"  Missing final CXR pairs from original source: "
    f"{len(missing_source_cxr)}"
)

record(
    "AUD-62-11",
    "CXR_PROVENANCE",
    "Final representative CXR exists in original source",
    len(missing_source_cxr),
    0,
    "PASS"
    if len(missing_source_cxr) == 0
    else "FAIL"
)


# ================================================================
# 13. GENOMIC SPECIMEN LINKAGE AGAINST ORIGINAL SOURCE
# ================================================================

print("\n[10] Genomic specimen linkage against source...")

source_specimen_pairs = set(
    zip(
        genomic_source["condition_id"],
        genomic_source["specimen_id"]
    )
)

final_specimen_pairs = set(
    zip(
        final_cohort["condition_id"],
        final_cohort["selected_specimen_id"]
    )
)

missing_source_specimens = (
    final_specimen_pairs -
    source_specimen_pairs
)

print(
    f"  Missing selected specimens from source: "
    f"{len(missing_source_specimens)}"
)

record(
    "AUD-62-12",
    "GENOMIC_PROVENANCE",
    "Representative genomic specimen exists in source",
    len(missing_source_specimens),
    0,
    "PASS"
    if len(missing_source_specimens) == 0
    else "FAIL"
)


# ================================================================
# 14. CONDITION LINKAGE FOR SELECTED CXR
# ================================================================

print("\n[11] CXR condition linkage audit...")

cxr_condition_mismatch = 0

source_cxr_lookup = (
    cxr_source[
        [
            "condition_id",
            "series_instance_content_url"
        ]
    ]
    .drop_duplicates()
)

source_cxr_lookup = source_cxr_lookup.set_index(
    "series_instance_content_url"
)["condition_id"].to_dict()

for _, row in final_cohort.iterrows():

    url = row["series_instance_content_url"]
    condition = row["condition_id"]

    source_condition = source_cxr_lookup.get(url)

    if source_condition is None:

        cxr_condition_mismatch += 1

    elif str(source_condition) != str(condition):

        cxr_condition_mismatch += 1

print(
    f"  CXR-condition mismatches: "
    f"{cxr_condition_mismatch}"
)

record(
    "AUD-62-13",
    "LINKAGE",
    "CXR URL-condition linkage",
    cxr_condition_mismatch,
    0,
    "PASS"
    if cxr_condition_mismatch == 0
    else "FAIL"
)


# ================================================================
# 15. CONDITION LINKAGE FOR SELECTED SPECIMEN
# ================================================================

print("\n[12] Genomic specimen-condition linkage audit...")

specimen_condition_mismatch = 0

source_specimen_lookup = (
    genomic_source[
        [
            "specimen_id",
            "condition_id"
        ]
    ]
    .drop_duplicates()
)

source_specimen_lookup = (
    source_specimen_lookup
    .groupby("specimen_id")["condition_id"]
    .apply(set)
    .to_dict()
)

for _, row in final_cohort.iterrows():

    specimen = row["selected_specimen_id"]
    condition = row["condition_id"]

    possible_conditions = (
        source_specimen_lookup.get(
            specimen,
            set()
        )
    )

    if condition not in possible_conditions:

        specimen_condition_mismatch += 1

print(
    f"  Specimen-condition mismatches: "
    f"{specimen_condition_mismatch}"
)

record(
    "AUD-62-14",
    "LINKAGE",
    "Genomic specimen-condition linkage",
    specimen_condition_mismatch,
    0,
    "PASS"
    if specimen_condition_mismatch == 0
    else "FAIL"
)


# ================================================================
# 16. TEMPORAL RULE
# ================================================================

print("\n[13] Strict-before temporal audit...")

final_cohort[
    "imaging_date_num"
] = pd.to_numeric(
    final_cohort["imaging_date_num"],
    errors="coerce"
)

final_cohort[
    "specimen_collection_date_num"
] = pd.to_numeric(
    final_cohort["specimen_collection_date_num"],
    errors="coerce"
)

final_cohort[
    "absolute_temporal_distance_days"
] = pd.to_numeric(
    final_cohort[
        "absolute_temporal_distance_days"
    ],
    errors="coerce"
)

temporal_missing = int(
    final_cohort[
        [
            "imaging_date_num",
            "specimen_collection_date_num",
            "absolute_temporal_distance_days"
        ]
    ]
    .isna()
    .any(axis=1)
    .sum()
)

non_strict_before = int(
    (
        final_cohort[
            "specimen_collection_date_num"
        ]
        >=
        final_cohort[
            "imaging_date_num"
        ]
    ).sum()
)

recomputed_distance = (
    final_cohort["imaging_date_num"]
    -
    final_cohort["specimen_collection_date_num"]
).abs()

distance_mismatch = int(
    (
        recomputed_distance
        !=
        final_cohort[
            "absolute_temporal_distance_days"
        ]
    ).sum()
)

print(
    f"  Missing temporal values: {temporal_missing}"
)

print(
    f"  Non-strict-before records: {non_strict_before}"
)

print(
    f"  Distance mismatches: {distance_mismatch}"
)

record(
    "AUD-62-15",
    "TEMPORAL",
    "Missing temporal values",
    temporal_missing,
    0,
    "PASS"
    if temporal_missing == 0
    else "FAIL"
)

record(
    "AUD-62-16",
    "TEMPORAL",
    "Non-strict-before records",
    non_strict_before,
    0,
    "PASS"
    if non_strict_before == 0
    else "FAIL"
)

record(
    "AUD-62-17",
    "TEMPORAL",
    "Temporal distance mismatches",
    distance_mismatch,
    0,
    "PASS"
    if distance_mismatch == 0
    else "FAIL"
)


# ================================================================
# 17. TEMPORAL DIRECTION
# ================================================================

print("\n[14] Temporal direction audit...")

temporal_directions = set(
    temporal_audit[
        "computed_temporal_direction"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)

print(
    f"  Directions in final temporal audit: "
    f"{temporal_directions}"
)

unexpected_directions = (
    temporal_directions -
    {"GENOMICS_BEFORE_CXR"}
)

record(
    "AUD-62-18",
    "TEMPORAL",
    "Representative genomic direction",
    str(sorted(temporal_directions)),
    "['GENOMICS_BEFORE_CXR']",
    "PASS"
    if temporal_directions == {"GENOMICS_BEFORE_CXR"}
    else "FAIL"
)

if unexpected_directions:

    raise RuntimeError(
        "Unexpected temporal direction found."
    )


# ================================================================
# 18. TARGET AUDIT
# ================================================================

print("\n[15] Final target audit...")

targets = (
    final_cohort[
        "DR_DS_target"
    ]
    .astype(str)
    .str.strip()
)

unexpected_targets = sorted(
    set(targets)
    -
    {"RESISTANT", "SENSITIVE"}
)

resistant = int(
    (targets == "RESISTANT").sum()
)

sensitive = int(
    (targets == "SENSITIVE").sum()
)

print(
    f"  Resistant: {resistant:,}"
)

print(
    f"  Sensitive: {sensitive:,}"
)

print(
    f"  Unexpected target values: "
    f"{unexpected_targets}"
)

record(
    "AUD-62-19",
    "TARGET",
    "Unexpected target values",
    str(unexpected_targets),
    "[]",
    "PASS"
    if not unexpected_targets
    else "FAIL"
)

record(
    "AUD-62-20",
    "TARGET",
    "Resistant count",
    resistant,
    985,
    "PASS"
    if resistant == 985
    else "FAIL"
)

record(
    "AUD-62-21",
    "TARGET",
    "Sensitive count",
    sensitive,
    296,
    "PASS"
    if sensitive == 296
    else "FAIL"
)


# ================================================================
# 19. NORMAL SAFETY
# ================================================================

print("\n[16] Normal-label safety audit...")

normal_terms = {
    "NORMAL",
    "HEALTHY",
    "CONTROL",
    "NON-TB",
    "NON_TB",
    "NON TB",
}

normal_like = int(
    targets
    .str.upper()
    .isin(normal_terms)
    .sum()
)

print(
    f"  Normal-like target labels: {normal_like}"
)

record(
    "AUD-62-22",
    "SCIENTIFIC_SAFETY",
    "Normal-like target labels",
    normal_like,
    0,
    "PASS"
    if normal_like == 0
    else "FAIL",
    "The final design is DR vs DS, not TB vs Normal."
)


# ================================================================
# 20. SAME-DAY / AFTER-CXR SAFETY
# ================================================================

print("\n[17] Same-day and after-CXR exclusion audit...")

same_day_selected = int(
    (
        final_cohort[
            "specimen_collection_date_num"
        ]
        ==
        final_cohort[
            "imaging_date_num"
        ]
    ).sum()
)

after_selected = int(
    (
        final_cohort[
            "specimen_collection_date_num"
        ]
        >
        final_cohort[
            "imaging_date_num"
        ]
    ).sum()
)

print(
    f"  Same-day selected: {same_day_selected}"
)

print(
    f"  After-CXR selected: {after_selected}"
)

record(
    "AUD-62-23",
    "TEMPORAL",
    "Same-day representatives",
    same_day_selected,
    0,
    "PASS"
    if same_day_selected == 0
    else "FAIL"
)

record(
    "AUD-62-24",
    "TEMPORAL",
    "After-CXR representatives",
    after_selected,
    0,
    "PASS"
    if after_selected == 0
    else "FAIL"
)


# ================================================================
# 21. REPRESENTATIVE ARTIFACT CONSISTENCY
# ================================================================

print("\n[18] Representative artifact consistency...")

cohort_cxr_pairs = set(
    zip(
        final_cohort["condition_id"],
        final_cohort["series_instance_content_url"]
    )
)

rep_cxr_pairs = set(
    zip(
        rep_cxrs["condition_id"],
        rep_cxrs["series_instance_content_url"]
    )
)

missing_rep_cxrs = (
    cohort_cxr_pairs -
    rep_cxr_pairs
)

extra_rep_cxrs = (
    rep_cxr_pairs -
    cohort_cxr_pairs
)

rep_specimen_pairs = set(
    zip(
        rep_specimens["condition_id"],
        rep_specimens["specimen_id"]
    )
)

missing_rep_specimens = (
    final_specimen_pairs -
    rep_specimen_pairs
)

extra_rep_specimens = (
    rep_specimen_pairs -
    final_specimen_pairs
)

print(
    f"  Missing representative CXR pairs: "
    f"{len(missing_rep_cxrs)}"
)

print(
    f"  Extra representative CXR pairs: "
    f"{len(extra_rep_cxrs)}"
)

print(
    f"  Missing representative specimens: "
    f"{len(missing_rep_specimens)}"
)

print(
    f"  Extra representative specimens: "
    f"{len(extra_rep_specimens)}"
)

record(
    "AUD-62-25",
    "REPRESENTATIVE",
    "Representative CXR artifact matches final cohort",
    len(missing_rep_cxrs),
    0,
    "PASS"
    if len(missing_rep_cxrs) == 0
    else "FAIL"
)

record(
    "AUD-62-26",
    "REPRESENTATIVE",
    "Extra CXR artifact records",
    len(extra_rep_cxrs),
    0,
    "PASS"
    if len(extra_rep_cxrs) == 0
    else "FAIL"
)

record(
    "AUD-62-27",
    "REPRESENTATIVE",
    "Representative specimen artifact matches final cohort",
    len(missing_rep_specimens),
    0,
    "PASS"
    if len(missing_rep_specimens) == 0
    else "FAIL"
)

record(
    "AUD-62-28",
    "REPRESENTATIVE",
    "Extra genomic artifact records",
    len(extra_rep_specimens),
    0,
    "PASS"
    if len(extra_rep_specimens) == 0
    else "FAIL"
)


# ================================================================
# 22. REPRESENTATIVE CONDITION COVERAGE
# ================================================================

print("\n[19] Condition coverage audit...")

final_conditions = set(
    final_cohort["condition_id"]
)

rep_cxr_conditions = set(
    rep_cxrs["condition_id"]
)

rep_specimen_conditions = set(
    rep_specimens["condition_id"]
)

missing_cxr_conditions = (
    final_conditions -
    rep_cxr_conditions
)

missing_specimen_conditions = (
    final_conditions -
    rep_specimen_conditions
)

print(
    f"  Conditions without representative CXR: "
    f"{len(missing_cxr_conditions)}"
)

print(
    f"  Conditions without representative specimen: "
    f"{len(missing_specimen_conditions)}"
)

record(
    "AUD-62-29",
    "COVERAGE",
    "Conditions without representative CXR",
    len(missing_cxr_conditions),
    0,
    "PASS"
    if len(missing_cxr_conditions) == 0
    else "FAIL"
)

record(
    "AUD-62-30",
    "COVERAGE",
    "Conditions without representative specimen",
    len(missing_specimen_conditions),
    0,
    "PASS"
    if len(missing_specimen_conditions) == 0
    else "FAIL"
)


# ================================================================
# 23. RESEARCH DESIGN LOCK
# ================================================================

print("\n[20] Research-design lock audit...")

research_design = design_lock.get(
    "research_design"
)

target_definition = design_lock.get(
    "target_definition"
)

temporal_rule = design_lock.get(
    "temporal_rule"
)

expected_design = "DR_DS"

expected_target = "RESISTANT vs SENSITIVE"

expected_temporal_rule = (
    "specimen_collection_date < imaging_date"
)

print(
    f"  Research design: {research_design}"
)

print(
    f"  Target: {target_definition}"
)

print(
    f"  Temporal rule: {temporal_rule}"
)

record(
    "AUD-62-31",
    "DESIGN",
    "Research design",
    research_design,
    expected_design,
    "PASS"
    if research_design == expected_design
    else "FAIL"
)

record(
    "AUD-62-32",
    "DESIGN",
    "Target definition",
    target_definition,
    expected_target,
    "PASS"
    if target_definition == expected_target
    else "FAIL"
)

record(
    "AUD-62-33",
    "DESIGN",
    "Temporal rule",
    temporal_rule,
    expected_temporal_rule,
    "PASS"
    if temporal_rule == expected_temporal_rule
    else "FAIL"
)


# ================================================================
# 24. FEATURE-LEAKAGE LOCK
# ================================================================

print("\n[21] Feature leakage lock audit...")

check_columns(
    feature_status,
    ["field", "status"],
    "Feature-status artifact"
)

feature_map = dict(
    zip(
        feature_status["field"].astype(str).str.strip(),
        feature_status["status"].astype(str).str.strip()
    )
)

leakage_fields = [
    "type_of_resistance",
]

review_fields = [
    "drug_resistance_type",
    "num_drug_resistant_variants",
    "gene_snp_mutations",
]

for field in leakage_fields:

    status = feature_map.get(field)

    print(
        f"  {field}: {status}"
    )


for field in review_fields:

    status = feature_map.get(field)

    print(
        f"  {field}: {status}"
    )


# We do not automatically approve any genomic predictor here.
record(
    "AUD-62-34",
    "LEAKAGE",
    "Automatic genomic predictor approval",
    "NONE",
    "NONE",
    "PASS",
    "3B.6.62 does not approve additional genomic predictors."
)


# ================================================================
# 25. PATIENT-LEVEL SPLIT LOCK
# ================================================================

print("\n[22] Patient-level split requirement...")

print(
    "  Required split unit: patient_id"
)

print(
    "  Reason: 1,281 conditions belong to 1,279 patients."
)

record(
    "AUD-62-35",
    "DATA_SPLIT",
    "Patient-level splitting required",
    True,
    True,
    "PASS",
    "The two patients with multiple conditions must not be split across train/validation/test."
)


# ================================================================
# 26. RAW DATA SAFETY
# ================================================================

print("\n[23] Raw-data modification lock...")

print(
    "  This audit cell performs read-only analysis."
)

print(
    "  No CXR source metadata are overwritten."
)

print(
    "  No genomic source metadata are overwritten."
)

print(
    "  No DICOM files are modified."
)

record(
    "AUD-62-36",
    "DATA_SAFETY",
    "Raw source modification by this audit",
    "NONE",
    "NONE",
    "PASS"
)


# ================================================================
# 27. CREATE FINAL AUDIT DATAFRAME
# ================================================================

audit_df = pd.DataFrame(
    audit_results
)


# ================================================================
# 28. CHECK FOR FAILURES
# ================================================================

failures = audit_df[
    audit_df["status"] == "FAIL"
]

reviews = audit_df[
    audit_df["status"] == "REVIEW"
]

print("\n[24] Final validation status...")

print(
    f"  Total audit tests: {len(audit_df)}"
)

print(
    f"  PASS: "
    f"{(audit_df['status'] == 'PASS').sum()}"
)

print(
    f"  REVIEW: "
    f"{len(reviews)}"
)

print(
    f"  FAIL: "
    f"{len(failures)}"
)


# ================================================================
# 29. SAVE FINAL AUDIT
# ================================================================

audit_output = (
    OUTPUT_DIR /
    "01_Final_Cohort_Representative_Audit.csv"
)

audit_df.to_csv(
    audit_output,
    index=False
)

print(
    f"\n  SAVED: {audit_output}"
)


# ================================================================
# 30. SAVE CXR LINKAGE AUDIT
# ================================================================

cxr_linkage_rows = []

for _, row in final_cohort.iterrows():

    condition = row["condition_id"]
    url = row["series_instance_content_url"]

    in_master = (
        (condition, url)
        in master_pairs
    )

    in_source = (
        (condition, url)
        in source_cxr_pairs
    )

    cxr_linkage_rows.append(
        {
            "condition_id": condition,
            "series_instance_content_url": url,
            "in_validated_3859_master": in_master,
            "in_original_cxr_source": in_source,
            "status":
                "PASS"
                if in_master and in_source
                else "FAIL",
        }
    )

cxr_linkage_df = pd.DataFrame(
    cxr_linkage_rows
)

cxr_linkage_output = (
    OUTPUT_DIR /
    "02_CXR_Linkage_Audit.csv"
)

cxr_linkage_df.to_csv(
    cxr_linkage_output,
    index=False
)

print(
    f"  SAVED: {cxr_linkage_output}"
)


# ================================================================
# 31. SAVE GENOMIC LINKAGE AUDIT
# ================================================================

genomic_linkage_rows = []

for _, row in final_cohort.iterrows():

    condition = row["condition_id"]
    specimen = row["selected_specimen_id"]

    in_source = (
        (condition, specimen)
        in source_specimen_pairs
    )

    genomic_linkage_rows.append(
        {
            "condition_id": condition,
            "specimen_id": specimen,
            "in_original_genomic_source": in_source,
            "status":
                "PASS"
                if in_source
                else "FAIL",
        }
    )

genomic_linkage_df = pd.DataFrame(
    genomic_linkage_rows
)

genomic_linkage_output = (
    OUTPUT_DIR /
    "03_Genomic_Specimen_Linkage_Audit.csv"
)

genomic_linkage_df.to_csv(
    genomic_linkage_output,
    index=False
)

print(
    f"  SAVED: {genomic_linkage_output}"
)


# ================================================================
# 32. SAVE TEMPORAL AUDIT
# ================================================================

temporal_output = (
    OUTPUT_DIR /
    "04_Final_Temporal_Audit.csv"
)

temporal_audit.to_csv(
    temporal_output,
    index=False
)

print(
    f"  SAVED: {temporal_output}"
)


# ================================================================
# 33. SAVE TARGET AUDIT
# ================================================================

target_audit_df = (
    final_cohort[
        [
            "condition_id",
            "patient_id",
            "DR_DS_target",
        ]
    ]
    .copy()
)

target_audit_df[
    "target_valid"
] = target_audit_df[
    "DR_DS_target"
].isin(
    [
        "RESISTANT",
        "SENSITIVE"
    ]
)

target_output = (
    OUTPUT_DIR /
    "05_Final_Target_Audit.csv"
)

target_audit_df.to_csv(
    target_output,
    index=False
)

print(
    f"  SAVED: {target_output}"
)


# ================================================================
# 34. SAVE DUPLICATE AUDIT
# ================================================================

duplicate_audit = pd.DataFrame(
    [
        {
            "object":
                "condition_id",
            "duplicate_count":
                duplicate_conditions,
        },
        {
            "object":
                "series_instance_content_url",
            "duplicate_count":
                duplicate_cxr_urls,
        },
        {
            "object":
                "selected_specimen_id",
            "duplicate_count":
                duplicate_specimens,
        },
    ]
)

duplicate_output = (
    OUTPUT_DIR /
    "06_Duplicate_Audit.csv"
)

duplicate_audit.to_csv(
    duplicate_output,
    index=False
)

print(
    f"  SAVED: {duplicate_output}"
)


# ================================================================
# 35. SOURCE PROVENANCE AUDIT
# ================================================================

provenance = pd.DataFrame(
    [
        {
            "component":
                "Original CXR metadata",
            "path":
                str(CXR_SOURCE),
            "rows":
                len(cxr_source),
            "status":
                "SOURCE_READ_ONLY",
        },
        {
            "component":
                "Original genomic metadata",
            "path":
                str(GENOMIC_SOURCE),
            "rows":
                len(genomic_source),
            "status":
                "SOURCE_READ_ONLY",
        },
        {
            "component":
                "Validated 3,859-row CXR master",
            "path":
                str(CXR_MASTER),
            "rows":
                len(cxr_master),
            "status":
                "VALIDATED_ARTIFACT",
        },
        {
            "component":
                "Final Step 3B.6.61 cohort",
            "path":
                str(FINAL_COHORT),
            "rows":
                len(final_cohort),
            "status":
                "LOCKED_ARTIFACT",
        },
        {
            "component":
                "Representative CXR artifact",
            "path":
                str(REP_CXRS),
            "rows":
                len(rep_cxrs),
            "status":
                "LOCKED_ARTIFACT",
        },
        {
            "component":
                "Representative genomic artifact",
            "path":
                str(REP_SPECIMENS),
            "rows":
                len(rep_specimens),
            "status":
                "LOCKED_ARTIFACT",
        },
    ]
)

provenance_output = (
    OUTPUT_DIR /
    "07_Source_Provenance_Audit.csv"
)

provenance.to_csv(
    provenance_output,
    index=False
)

print(
    f"  SAVED: {provenance_output}"
)


# ================================================================
# 36. FINAL VALIDATION FILE
# ================================================================

validation_62 = audit_df.copy()

validation_output = (
    OUTPUT_DIR /
    "08_Final_Validation.csv"
)

validation_62.to_csv(
    validation_output,
    index=False
)

print(
    f"  SAVED: {validation_output}"
)


# ================================================================
# 37. FINAL SUMMARY JSON
# ================================================================

overall_status = (
    "PASS"
    if len(failures) == 0
    else "FAIL"
)

summary_62 = {

    "step":
        "3B.6.62",

    "status":
        overall_status,

    "purpose":
        "Independent final integrity audit of the locked Step 3B.6.61 cohort.",

    "research_design":
        "DR_DS",

    "target":
        "RESISTANT vs SENSITIVE",

    "final_conditions":
        int(condition_count),

    "final_patients":
        int(patient_count),

    "representative_cxrs":
        int(cxr_url_count),

    "representative_genomic_specimens":
        int(specimen_count),

    "resistant":
        int(resistant),

    "sensitive":
        int(sensitive),

    "temporal_rule":
        expected_temporal_rule,

    "same_day_policy":
        "EXCLUDE",

    "after_cxr_policy":
        "EXCLUDE",

    "non_strict_before_selected":
        int(non_strict_before),

    "same_day_selected":
        int(same_day_selected),

    "after_cxr_selected":
        int(after_selected),

    "temporal_distance_mismatches":
        int(distance_mismatch),

    "duplicate_conditions":
        int(duplicate_conditions),

    "duplicate_cxr_urls":
        int(duplicate_cxr_urls),

    "duplicate_specimens":
        int(duplicate_specimens),

    "normal_like_target_labels":
        int(normal_like),

    "patients_with_multiple_conditions":
        int(patients_multiple_conditions),

    "cxr_source_linkage_failures":
        int(len(missing_source_cxr)),

    "cxr_master_linkage_failures":
        int(len(missing_cxr_pairs)),

    "genomic_source_linkage_failures":
        int(len(missing_source_specimens)),

    "failed_tests":
        int(len(failures)),

    "review_tests":
        int(len(reviews)),

    "total_tests":
        int(len(audit_df)),

    "patient_level_split_required":
        True,

    "preprocessing_started":
        False,

    "segmentation_started":
        False,

    "genomic_feature_engineering_started":
        False,

    "model_training_started":
        False,

    "xai_started":
        False,

    "raw_source_modified_by_step_3B_6_62":
        False,

    "next_stage":
        (
            "CXR representative-file/DICOM accessibility "
            "and image-readiness audit"
        ),

}


summary_output = (
    OUTPUT_DIR /
    "09_Step_3B_6_62_Final_Summary.json"
)

with open(
    summary_output,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary_62,
        f,
        indent=4,
        ensure_ascii=False
    )

print(
    f"  SAVED: {summary_output}"
)


# ================================================================
# 38. FINAL CONSOLE REPORT
# ================================================================

print("\n")
print("=" * 80)
print("STEP 3B.6.62 — FINAL RESULT")
print("=" * 80)

print(
    f"\nStatus                 : {overall_status}"
)

print(
    f"Final conditions       : {condition_count:,}"
)

print(
    f"Final patients         : {patient_count:,}"
)

print(
    f"Representative CXRs    : {cxr_url_count:,}"
)

print(
    f"Representative genomes : {specimen_count:,}"
)

print(
    f"Resistant              : {resistant:,}"
)

print(
    f"Sensitive              : {sensitive:,}"
)

print(
    f"Non-strict-before      : {non_strict_before}"
)

print(
    f"Same-day selected      : {same_day_selected}"
)

print(
    f"After-CXR selected     : {after_selected}"
)

print(
    f"Duplicate conditions   : {duplicate_conditions}"
)

print(
    f"Duplicate CXR URLs    : {duplicate_cxr_urls}"
)

print(
    f"Duplicate specimens   : {duplicate_specimens}"
)

print(
    f"Normal-like labels     : {normal_like}"
)

print(
    f"CXR source failures   : {len(missing_source_cxr)}"
)

print(
    f"Genomic source failures: {len(missing_source_specimens)}"
)

print(
    f"Validation tests       : {len(audit_df)}"
)

print(
    f"PASS                   : "
    f"{(audit_df['status'] == 'PASS').sum()}"
)

print(
    f"REVIEW                 : {len(reviews)}"
)

print(
    f"FAIL                   : {len(failures)}"
)

print(
    f"\nOutput directory:\n{OUTPUT_DIR}"
)

if len(failures) == 0:

    print(
        "\n============================================================"
    )

    print(
        "3B.6.62 STATUS: PASS"
    )

    print(
        "============================================================"
    )

    print(
        "\nThe Step 3B.6.61 final cohort has passed "
        "the independent integrity audit."
    )

    print(
        "\nDO NOT start preprocessing yet."
    )

    print(
        "Next: representative CXR/DICOM accessibility "
        "and image-readiness audit."
    )

else:

    print(
        "\n============================================================"
    )

    print(
        "3B.6.62 STATUS: FAIL"
    )

    print(
        "============================================================"
    )

    print(
        "\nDo NOT proceed to preprocessing."
    )

    print(
        "Inspect the failed audit rows first."
    )

print("=" * 80)
```
